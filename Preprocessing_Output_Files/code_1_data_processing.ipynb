```python
# Import required libraries

# os library is used for folder and file handling
import os

# pandas library is used for reading, modifying, and saving CSV files
import pandas as pd



# -------------------------------------------------------
# Define the dataset folder path
# -------------------------------------------------------

# Your notebook is inside the RAW_DATA folder.
# Therefore, we use a relative path.

folder_path = "./class_lecture/clms"



# -------------------------------------------------------
# Check whether the folder exists
# -------------------------------------------------------

if not os.path.exists(folder_path):

    # Display message if the folder path is incorrect
    print("Folder does not exist:")
    print(folder_path)


else:

    # -------------------------------------------------------
    # Get all CSV files from the folder
    # -------------------------------------------------------

    # os.listdir() returns all files and folders
    # sorted() keeps the file order:
    # clms1.csv, clms2.csv, ..., clms60.csv

    csv_files = sorted(
        [
            file for file in os.listdir(folder_path)
            if file.endswith(".csv")
        ]
    )


    # Display total number of files found

    print("Total CSV files found:", len(csv_files))


    # -------------------------------------------------------
    # Process each CSV file one by one
    # -------------------------------------------------------

    for file_name in csv_files:


        # Create complete file path

        file_path = os.path.join(
            folder_path,
            file_name
        )


        try:

            # -------------------------------------------------------
            # Read CSV file
            # -------------------------------------------------------

            df = pd.read_csv(file_path)



            # -------------------------------------------------------
            # Check whether TimeStamp column exists
            # -------------------------------------------------------

            if "TimeStamp" in df.columns:


                # -------------------------------------------------------
                # Convert TimeStamp into time format
                # -------------------------------------------------------

                # pd.to_datetime() converts the timestamp values
                # into datetime objects.
                #
                # strftime("%H:%M:%S.%f") keeps:
                # Hour:Minute:Second.Microsecond
                #
                # Example:
                # 10:25:30.123456

                df["TimeStamp"] = pd.to_datetime(
                    df["TimeStamp"]
                ).dt.strftime(
                    "%H:%M:%S.%f"
                )



                # -------------------------------------------------------
                # Save the modified CSV file
                # -------------------------------------------------------

                # index=False prevents pandas from creating
                # an additional index column

                df.to_csv(
                    file_path,
                    index=False
                )


                # Show successful update message

                print(
                    "Updated:",
                    file_name
                )


            else:

                # Display message if TimeStamp column is missing

                print(
                    "TimeStamp column not found:",
                    file_name
                )



        except Exception as error:

            # Display error information if any file fails

            print(
                "Error processing",
                file_name,
                ":",
                error
            )



    # -------------------------------------------------------
    # Completion message
    # -------------------------------------------------------

    print("\nAll files processed successfully.")
```

```python
# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

# os is used for file and folder operations
import os

# pandas is used for reading, modifying, and saving CSV files
import pandas as pd



# -------------------------------------------------------
# Define folders to process
# -------------------------------------------------------

# These are the four folders containing Motion Watch CSV files

folders_to_process = [

    "./class_lecture/clmw",
    "./music/mumw",
    "./normal/normw",
    "./reading/rdmw"

]



# -------------------------------------------------------
# Process each folder
# -------------------------------------------------------

for folder_path in folders_to_process:


    # Display current folder

    print("\nProcessing folder:", folder_path)



    # -------------------------------------------------------
    # Check folder exists
    # -------------------------------------------------------

    if not os.path.exists(folder_path):

        print("Folder not found:", folder_path)

        continue



    # -------------------------------------------------------
    # Get all CSV files
    # -------------------------------------------------------

    csv_files = sorted(
        [
            file
            for file in os.listdir(folder_path)
            if file.endswith(".csv")
        ]
    )


    print(
        "Total CSV files found:",
        len(csv_files)
    )



    # -------------------------------------------------------
    # Process each CSV file
    # -------------------------------------------------------

    for file_name in csv_files:


        # Create complete file path

        file_path = os.path.join(
            folder_path,
            file_name
        )


        try:


            # -------------------------------------------------------
            # Read CSV file
            # -------------------------------------------------------

            df = pd.read_csv(file_path)



            # -------------------------------------------------------
            # Convert Date column format
            # -------------------------------------------------------

            # Convert Date values into datetime format
            # and then convert into short date format
            #
            # Example:
            # 2024-10-28  -> 10/28/2024

            if "Date" in df.columns:

                df["Date"] = pd.to_datetime(
                    df["Date"]
                ).dt.strftime(
                    "%m/%d/%Y"
                )



            # -------------------------------------------------------
            # Convert Time column format
            # -------------------------------------------------------

            # Convert Time values into time format
            #
            # Example:
            # 18:08:00 -> 06:08:00 PM

            if "Time" in df.columns:

                df["Time"] = pd.to_datetime(
                    df["Time"]
                ).dt.strftime(
                    "%I:%M:%S %p"
                )



            # -------------------------------------------------------
            # Rename third column as Activity
            # -------------------------------------------------------

            # Check if the CSV has at least 3 columns

            if len(df.columns) >= 3:

                # Get the third column name

                third_column = df.columns[2]


                # Rename third column

                df.rename(
                    columns={
                        third_column: "Activity"
                    },
                    inplace=True
                )



            # -------------------------------------------------------
            # Save modified CSV file
            # -------------------------------------------------------

            # index=False prevents adding extra index column

            df.to_csv(
                file_path,
                index=False
            )


            print(
                "Updated:",
                file_name
            )



        except Exception as error:


            # Display error message if any file fails

            print(
                "Error processing",
                file_name,
                ":",
                error
            )



# -------------------------------------------------------
# Completion message
# -------------------------------------------------------

print("\nAll Motion Watch folders processed successfully.")
```

```python
# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

# os is used for folder and file operations
import os

# pandas is used for CSV reading and modification
import pandas as pd



# -------------------------------------------------------
# Define folders to process
# -------------------------------------------------------

folders_to_process = [

    "./class_lecture/clmw",
    "./music/mumw",
    "./normal/normw",
    "./reading/rdmw"

]



# -------------------------------------------------------
# Function to read CSV with different encodings
# -------------------------------------------------------

def read_csv_with_encoding(file_path):

    # List of possible encodings
    encodings = [
        "utf-8",
        "utf-8-sig",
        "latin1",
        "cp1252",
        "utf-16"
    ]


    # Try each encoding one by one

    for encoding in encodings:

        try:

            # Read CSV file

            df = pd.read_csv(
                file_path,
                encoding=encoding,
                engine="python"
            )


            # Return dataframe if successful

            return df


        except Exception:

            # Try next encoding

            continue



    # Return None if all encodings fail

    return None



# -------------------------------------------------------
# Process folders
# -------------------------------------------------------

for folder_path in folders_to_process:


    print("\nProcessing:", folder_path)



    # Check folder exists

    if not os.path.exists(folder_path):

        print("Folder not found:", folder_path)

        continue



    # Get CSV files

    csv_files = sorted(
        [
            file
            for file in os.listdir(folder_path)
            if file.endswith(".csv")
        ]
    )


    print(
        "Files found:",
        len(csv_files)
    )



    # Process each CSV file

    for file_name in csv_files:


        file_path = os.path.join(
            folder_path,
            file_name
        )


        try:


            # Read CSV using multiple encoding support

            df = read_csv_with_encoding(
                file_path
            )



            # Check if file could be read

            if df is None:

                print(
                    "Cannot read:",
                    file_name
                )

                continue



            # Remove unwanted spaces from column names

            df.columns = df.columns.str.strip()



            # Rename third column

            if len(df.columns) >= 3:


                # Replace third column name

                df.columns.values[2] = "Activity"



                # Save updated CSV

                df.to_csv(
                    file_path,
                    index=False,
                    encoding="utf-8"
                )


                print(
                    "Updated:",
                    file_name
                )


            else:

                print(
                    "Less than 3 columns:",
                    file_name
                )



        except Exception as error:


            print(
                "Error:",
                file_name,
                error
            )



print("\nActivity column update completed.")
```

```python
# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

# os is used for file and folder operations
import os

# pandas is used for reading and modifying CSV files
import pandas as pd



# -------------------------------------------------------
# Define folders to process
# -------------------------------------------------------

folders_to_process = [

    "./class_lecture/clmw",
    "./music/mumw",
    "./normal/normw",
    "./reading/rdmw"

]



# -------------------------------------------------------
# Function to read difficult CSV files
# -------------------------------------------------------

def read_problem_csv(file_path):

    # Different possible encodings
    encodings = [
        "utf-8",
        "utf-8-sig",
        "latin1",
        "cp1252",
        "utf-16"
    ]


    # Try different encodings

    for encoding in encodings:

        try:

            # Read CSV
            df = pd.read_csv(
                file_path,
                encoding=encoding,
                engine="python",
                on_bad_lines="skip"
            )


            # Return dataframe if successful

            return df


        except Exception:

            continue



    # If all methods fail

    return None




# -------------------------------------------------------
# Process folders
# -------------------------------------------------------

for folder_path in folders_to_process:


    print("\nProcessing folder:", folder_path)


    # Check folder exists

    if not os.path.exists(folder_path):

        print("Folder not found")
        continue



    # Get CSV files

    csv_files = sorted(
        [
            file
            for file in os.listdir(folder_path)
            if file.endswith(".csv")
        ]
    )


    print(
        "Total files:",
        len(csv_files)
    )



    # -------------------------------------------------------
    # Process each CSV file
    # -------------------------------------------------------

    for file_name in csv_files:


        file_path = os.path.join(
            folder_path,
            file_name
        )


        # Read CSV

        df = read_problem_csv(
            file_path
        )


        if df is None:


            print(
                "Failed:",
                file_name
            )

            continue



        try:


            # Remove hidden spaces from column names

            df.columns = df.columns.str.strip()



            # -------------------------------------------------------
            # Rename third column
            # -------------------------------------------------------

            if len(df.columns) >= 3:


                old_name = df.columns[2]


                # Force third column name

                df.columns.values[2] = "Activity"



                # Save file

                df.to_csv(
                    file_path,
                    index=False,
                    encoding="utf-8"
                )


                print(
                    "Updated:",
                    file_name,
                    "|",
                    old_name,
                    "-> Activity"
                )


            else:


                print(
                    "Less than 3 columns:",
                    file_name
                )



        except Exception as error:


            print(
                "Error:",
                file_name,
                error
            )



print("\nAll files completed.")
```

```python
# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

# os is used for file and folder operations
import os



# -------------------------------------------------------
# Define folders to process
# -------------------------------------------------------

folders_to_process = [

    "./class_lecture/clmw",
    "./music/mumw",
    "./normal/normw",
    "./reading/rdmw"

]



# -------------------------------------------------------
# Process folders
# -------------------------------------------------------

for folder_path in folders_to_process:


    print("\nProcessing folder:", folder_path)


    # Check folder exists

    if not os.path.exists(folder_path):

        print("Folder not found:", folder_path)

        continue



    # Get CSV files

    csv_files = sorted(
        [
            file
            for file in os.listdir(folder_path)
            if file.endswith(".csv")
        ]
    )


    print(
        "Total files:",
        len(csv_files)
    )



    # -------------------------------------------------------
    # Process each CSV file
    # -------------------------------------------------------

    for file_name in csv_files:


        file_path = os.path.join(
            folder_path,
            file_name
        )


        try:


            # -------------------------------------------------------
            # Read only the first line (header)
            # -------------------------------------------------------

            with open(
                file_path,
                "r",
                encoding="latin1"
            ) as file:

                header = file.readline()



            # -------------------------------------------------------
            # Split header by comma
            # -------------------------------------------------------

            columns = header.strip().split(",")



            # -------------------------------------------------------
            # Check at least 3 columns exist
            # -------------------------------------------------------

            if len(columns) >= 3:


                # Store old third column name

                old_name = columns[2]


                # Replace third column name

                columns[2] = "Activity"



                # Create new header line

                new_header = ",".join(columns)



                # -------------------------------------------------------
                # Rewrite file with new header
                # -------------------------------------------------------

                with open(
                    file_path,
                    "r",
                    encoding="latin1"
                ) as file:

                    lines = file.readlines()



                # Replace first line only

                lines[0] = new_header + "\n"



                # Save file again

                with open(
                    file_path,
                    "w",
                    encoding="utf-8"
                ) as file:

                    file.writelines(lines)



                print(
                    "Updated:",
                    file_name,
                    "|",
                    old_name,
                    "-> Activity"
                )


            else:

                print(
                    "Header has less than 3 columns:",
                    file_name
                )



        except Exception as error:


            print(
                "Error:",
                file_name,
                error
            )



print("\nHeader renaming completed.")
```

```python
# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

import os
import pandas as pd



# -------------------------------------------------------
# Define folders to process
# -------------------------------------------------------

folders_to_process = [

    "./class_lecture/clmw",
    "./music/mumw",
    "./normal/normw",
    "./reading/rdmw"

]



# -------------------------------------------------------
# Function to read CSV or Excel disguised as CSV
# -------------------------------------------------------

def read_file(file_path):


    # Check first four bytes

    with open(file_path, "rb") as file:

        signature = file.read(4)



    # If file starts with PK, it is an Excel file

    if signature == b"PK\x03\x04":


        print("Excel format detected:", file_path)


        # Read as Excel

        df = pd.read_excel(
            file_path,
            engine="openpyxl"
        )


    else:


        # Read as normal CSV

        df = pd.read_csv(
            file_path,
            encoding="latin1"
        )


    return df




# -------------------------------------------------------
# Process folders
# -------------------------------------------------------

for folder_path in folders_to_process:


    print("\nProcessing:", folder_path)



    csv_files = sorted(
        [
            f
            for f in os.listdir(folder_path)
            if f.endswith(".csv")
        ]
    )


    print(
        "Total files:",
        len(csv_files)
    )



    for file_name in csv_files:


        file_path = os.path.join(
            folder_path,
            file_name
        )


        try:


            # Read file

            df = read_file(file_path)



            # Remove spaces from column names

            df.columns = df.columns.str.strip()



            # Rename third column

            if len(df.columns) >= 3:


                old_name = df.columns[2]


                df.columns.values[2] = "Activity"



                # Save as CSV

                df.to_csv(
                    file_path,
                    index=False,
                    encoding="utf-8"
                )


                print(
                    "Updated:",
                    file_name,
                    "|",
                    old_name,
                    "-> Activity"
                )


            else:


                print(
                    "Less than 3 columns:",
                    file_name
                )



        except Exception as error:


            print(
                "Failed:",
                file_name,
                error
            )



print("\nCompleted.")
```

```python
# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

# os is used for folder and file operations
import os

# pandas is used for reading and modifying CSV files
import pandas as pd



# -------------------------------------------------------
# Define Motion Watch folders
# -------------------------------------------------------

# These folders contain Motion Watch CSV files

folders_to_process = [

    "./class_lecture/clmw",
    "./music/mumw",
    "./normal/normw",
    "./reading/rdmw"

]



# -------------------------------------------------------
# Function to read CSV files with different encodings
# -------------------------------------------------------

def read_csv_file(file_path):

    # Possible encodings of your dataset files

    encodings = [
        "utf-8",
        "utf-8-sig",
        "latin1",
        "cp1252"
    ]


    # Try different encodings

    for encoding in encodings:

        try:

            # Read CSV file

            df = pd.read_csv(
                file_path,
                encoding=encoding
            )

            return df


        except Exception:

            continue



    # Return None if file cannot be read

    return None



# -------------------------------------------------------
# Process all Motion Watch folders
# -------------------------------------------------------

for folder_path in folders_to_process:


    print("\nProcessing folder:", folder_path)



    # Check folder existence

    if not os.path.exists(folder_path):

        print(
            "Folder not found:",
            folder_path
        )

        continue



    # Get all CSV files

    csv_files = sorted(
        [
            file
            for file in os.listdir(folder_path)
            if file.endswith(".csv")
        ]
    )


    print(
        "Total CSV files:",
        len(csv_files)
    )



    # -------------------------------------------------------
    # Process each CSV file
    # -------------------------------------------------------

    for file_name in csv_files:


        file_path = os.path.join(
            folder_path,
            file_name
        )


        try:


            # Read CSV file

            df = read_csv_file(
                file_path
            )



            # Check successful reading

            if df is None:

                print(
                    "Cannot read:",
                    file_name
                )

                continue



            # Remove extra spaces from column names

            df.columns = df.columns.str.strip()



            # -------------------------------------------------------
            # Convert Date column format
            # -------------------------------------------------------

            if "Date" in df.columns:


                # Convert date to short date format
                #
                # Example:
                # 2024-11-01
                # becomes
                # 11/01/2024

                df["Date"] = pd.to_datetime(
                    df["Date"],
                    errors="coerce"
                ).dt.strftime(
                    "%m/%d/%Y"
                )



            # -------------------------------------------------------
            # Convert Time column format
            # -------------------------------------------------------

            if "Time" in df.columns:


                # Convert time into 12 hour format
                #
                # Example:
                # 14:26:00
                # becomes
                # 02:26:00 PM

                df["Time"] = pd.to_datetime(
                    df["Time"],
                    errors="coerce"
                ).dt.strftime(
                    "%I:%M:%S %p"
                )



            # -------------------------------------------------------
            # Save updated CSV file
            # -------------------------------------------------------

            df.to_csv(
                file_path,
                index=False,
                encoding="utf-8"
            )


            print(
                "Updated:",
                file_name
            )



        except Exception as error:


            print(
                "Error processing",
                file_name,
                ":",
                error
            )



print("\nDate and Time conversion completed successfully.")
```

```python
# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

# os is used for folder and file operations
import os

# pandas is used for reading, modifying, and saving CSV files
import pandas as pd



# -------------------------------------------------------
# Define Motion Watch folders
# -------------------------------------------------------

# These folders contain Motion Watch CSV files

folders_to_process = [

    "./class_lecture/clmw",
    "./music/mumw",
    "./normal/normw",
    "./reading/rdmw"

]



# -------------------------------------------------------
# Function to read CSV files with different encodings
# -------------------------------------------------------

def read_csv_file(file_path):

    # Different possible encodings in your dataset

    encodings = [
        "utf-8",
        "utf-8-sig",
        "latin1",
        "cp1252"
    ]


    # Try each encoding

    for encoding in encodings:

        try:

            # Read CSV file

            df = pd.read_csv(
                file_path,
                encoding=encoding
            )

            return df


        except Exception:

            continue


    # Return None if file cannot be read

    return None



# -------------------------------------------------------
# Process all Motion Watch folders
# -------------------------------------------------------

for folder_path in folders_to_process:


    print("\nProcessing folder:", folder_path)



    # Check folder exists

    if not os.path.exists(folder_path):

        print("Folder not found:", folder_path)

        continue



    # Get all CSV files

    csv_files = sorted(
        [
            file
            for file in os.listdir(folder_path)
            if file.endswith(".csv")
        ]
    )


    print(
        "Total files:",
        len(csv_files)
    )



    # -------------------------------------------------------
    # Process each CSV file
    # -------------------------------------------------------

    for file_name in csv_files:


        file_path = os.path.join(
            folder_path,
            file_name
        )


        try:


            # Read CSV file

            df = read_csv_file(
                file_path
            )


            # Skip if file cannot be read

            if df is None:

                print(
                    "Cannot read:",
                    file_name
                )

                continue



            # Remove unwanted spaces from column names

            df.columns = df.columns.str.strip()



            # -------------------------------------------------------
            # Convert Date column format
            # -------------------------------------------------------

            if "Date" in df.columns:


                # Convert Date format:
                #
                # Before:
                # 2024-11-01
                #
                # After:
                # 01/11/2024

                df["Date"] = pd.to_datetime(
                    df["Date"],
                    errors="coerce"
                ).dt.strftime(
                    "%d/%m/%Y"
                )



            # -------------------------------------------------------
            # Convert Time column format
            # -------------------------------------------------------

            if "Time" in df.columns:


                # Convert Time format:
                #
                # Before:
                # 14:26:00
                #
                # After:
                # 2:26:00 PM
                #
                # %I gives 12 hour format
                # %p gives AM/PM
                # %-I removes leading zero

                df["Time"] = pd.to_datetime(
                    df["Time"],
                    errors="coerce"
                ).dt.strftime(
                    "%I:%M:%S %p"
                )


                # Remove leading zero from hour
                # Example:
                # 02:26:00 PM
                # becomes
                # 2:26:00 PM

                df["Time"] = df["Time"].str.lstrip("0")



            # -------------------------------------------------------
            # Save updated CSV file
            # -------------------------------------------------------

            df.to_csv(
                file_path,
                index=False,
                encoding="utf-8"
            )


            print(
                "Updated:",
                file_name
            )



        except Exception as error:


            print(
                "Error processing",
                file_name,
                ":",
                error
            )



print("\nDate and Time formatting completed successfully.")
```

```python
# =======================================================
# ARIMA ANALYSIS FOR MOTION WATCH ACTIVITY DATA
# =======================================================


# -------------------------------------------------------
# Import required libraries
# -------------------------------------------------------

# os is used for file and folder operations
import os

# numpy is used for numerical calculations
import numpy as np

# pandas is used for reading and processing CSV files
import pandas as pd

# matplotlib is used for visualization
import matplotlib.pyplot as plt

# ARIMA model
from statsmodels.tsa.arima.model import ARIMA

# Augmented Dickey Fuller test for stationarity
from statsmodels.tsa.stattools import adfuller

# Error metrics
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)



# =======================================================
# SELECT THE MOTION WATCH FILE
# =======================================================

# Change this path when you want to analyze another file

FILE_PATH = "./class_lecture/clmw/clmw1.csv"



# =======================================================
# READ THE CSV FILE
# =======================================================

# Try several encodings because your dataset contains
# files with different encodings

encodings = [
    "utf-8",
    "utf-8-sig",
    "latin1",
    "cp1252"
]


df = None


# Try reading the file using different encodings

for encoding in encodings:

    try:

        df = pd.read_csv(
            FILE_PATH,
            encoding=encoding
        )

        print(
            "File successfully read using:",
            encoding
        )

        break

    except Exception:

        continue


# Stop if the file could not be read

if df is None:

    raise ValueError(
        "The selected CSV file could not be read."
    )



# =======================================================
# CLEAN COLUMN NAMES
# =======================================================

# Remove unnecessary spaces from column names

df.columns = df.columns.str.strip()


print("\nColumns in the file:")

print(
    df.columns.tolist()
)



# =======================================================
# CHECK REQUIRED COLUMNS
# =======================================================

# The ARIMA analysis requires Time and Activity

if "Time" not in df.columns:

    raise ValueError(
        "Time column was not found."
    )


if "Activity" not in df.columns:

    raise ValueError(
        "Activity column was not found."
    )



# =======================================================
# CONVERT TIME COLUMN
# =======================================================

# Convert Time into datetime

df["Time"] = pd.to_datetime(
    df["Time"],
    errors="coerce"
)



# =======================================================
# CONVERT ACTIVITY COLUMN TO NUMERIC
# =======================================================

# Convert Activity values into numeric values

df["Activity"] = pd.to_numeric(
    df["Activity"],
    errors="coerce"
)



# =======================================================
# REMOVE INVALID VALUES
# =======================================================

# Remove rows where Time or Activity is missing

df = df.dropna(
    subset=[
        "Time",
        "Activity"
    ]
)



# =======================================================
# SORT DATA BY TIME
# =======================================================

# Make sure observations are in chronological order

df = df.sort_values(
    "Time"
).reset_index(
    drop=True
)



# =======================================================
# DISPLAY BASIC INFORMATION
# =======================================================

print("\n============================================")
print("DATASET INFORMATION")
print("============================================")

print(
    "File:",
    FILE_PATH
)

print(
    "Number of observations:",
    len(df)
)

print(
    "Start time:",
    df["Time"].iloc[0]
)

print(
    "End time:",
    df["Time"].iloc[-1]
)

print(
    "Minimum Activity:",
    df["Activity"].min()
)

print(
    "Maximum Activity:",
    df["Activity"].max()
)

print(
    "Mean Activity:",
    df["Activity"].mean()
)

print(
    "Standard deviation:",
    df["Activity"].std()
)



# =======================================================
# CREATE TIME SERIES
# =======================================================

# Set Time as the index

activity_series = df.set_index(
    "Time"
)["Activity"]


# Remove duplicated timestamps if any exist

activity_series = activity_series[
    ~activity_series.index.duplicated(
        keep="first"
    )
]



# =======================================================
# CHECK SAMPLING INTERVAL
# =======================================================

# Calculate difference between consecutive timestamps

time_difference = activity_series.index.to_series().diff()


print("\n============================================")
print("SAMPLING INTERVAL")
print("============================================")

print(
    time_difference.value_counts().head()
)



# =======================================================
# AUGMENTED DICKEY FULLER TEST
# =======================================================

def check_stationarity(series):

    # Perform Augmented Dickey Fuller test

    result = adfuller(
        series.dropna()
    )


    # Extract ADF statistic

    adf_statistic = result[0]


    # Extract p value

    p_value = result[1]


    print(
        "ADF Statistic:",
        adf_statistic
    )

    print(
        "ADF p value:",
        p_value
    )


    # Interpret the test

    if p_value < 0.05:

        print(
            "Result: The series is stationary."
        )

        return True

    else:

        print(
            "Result: The series is not stationary."
        )

        return False



print("\n============================================")
print("STATIONARITY TEST")
print("============================================")


is_stationary = check_stationarity(
    activity_series
)



# =======================================================
# DETERMINE DIFFERENCING ORDER
# =======================================================

# Start with d = 0

d = 0


# Test first difference if original series is not stationary

if not is_stationary:


    differenced_series = activity_series.diff().dropna()


    print(
        "\nTesting first difference..."
    )


    first_difference_stationary = check_stationarity(
        differenced_series
    )


    if first_difference_stationary:

        d = 1

    else:

        d = 2


print(
    "\nSelected differencing order d =",
    d
)



# =======================================================
# TRAIN AND TEST SPLIT
# =======================================================

# Use 80 percent for training

train_size = int(
    len(activity_series) * 0.80
)


# Training data

train = activity_series.iloc[
    :train_size
]


# Testing data

test = activity_series.iloc[
    train_size:
]


print("\n============================================")
print("TRAIN TEST SPLIT")
print("============================================")

print(
    "Training observations:",
    len(train)
)

print(
    "Testing observations:",
    len(test)
)



# =======================================================
# SEARCH FOR BEST ARIMA MODEL
# =======================================================

# Store model results

model_results = []


# Search p values from 0 to 3

p_values = range(0, 4)


# Search q values from 0 to 3

q_values = range(0, 4)


print("\n============================================")
print("SEARCHING ARIMA MODELS")
print("============================================")


for p in p_values:

    for q in q_values:


        try:


            # Create ARIMA model

            model = ARIMA(
                train,
                order=(p, d, q)
            )


            # Fit model

            fitted_model = model.fit()


            # Store AIC

            model_results.append(
                {
                    "p": p,
                    "d": d,
                    "q": q,
                    "AIC": fitted_model.aic
                }
            )


            print(
                f"ARIMA({p},{d},{q}) "
                f"AIC = {fitted_model.aic:.2f}"
            )


        except Exception as error:


            print(
                f"ARIMA({p},{d},{q}) failed"
            )



# =======================================================
# FIND BEST MODEL
# =======================================================

results_df = pd.DataFrame(
    model_results
)


# Sort models by AIC

results_df = results_df.sort_values(
    "AIC"
).reset_index(
    drop=True
)


# Select model with lowest AIC

best_model_order = (
    int(results_df.loc[0, "p"]),
    int(results_df.loc[0, "d"]),
    int(results_df.loc[0, "q"])
)


print("\n============================================")
print("BEST ARIMA MODEL")
print("============================================")

print(
    "Selected model:",
    f"ARIMA{best_model_order}"
)

print(
    "Best AIC:",
    results_df.loc[0, "AIC"]
)



# =======================================================
# FIT BEST ARIMA MODEL
# =======================================================

best_model = ARIMA(
    train,
    order=best_model_order
)


best_fitted_model = best_model.fit()



# =======================================================
# FORECAST TEST DATA
# =======================================================

# Forecast exactly the same number of observations
# as the test dataset

forecast = best_fitted_model.forecast(
    steps=len(test)
)


# Make sure forecast has the same index as test

forecast.index = test.index



# =======================================================
# CALCULATE PERFORMANCE METRICS
# =======================================================

# Mean Absolute Error

mae = mean_absolute_error(
    test,
    forecast
)


# Root Mean Squared Error

rmse = np.sqrt(
    mean_squared_error(
        test,
        forecast
    )
)


# Mean Absolute Percentage Error
# Ignore zero actual values because percentage error
# cannot be calculated for zero

non_zero = test != 0


if non_zero.sum() > 0:

    mape = (
        np.mean(
            np.abs(
                (
                    test[non_zero]
                    -
                    forecast[non_zero]
                )
                /
                test[non_zero]
            )
        )
        * 100
    )

else:

    mape = np.nan



# =======================================================
# DISPLAY PERFORMANCE
# =======================================================

print("\n============================================")
print("ARIMA FORECAST PERFORMANCE")
print("============================================")

print(
    "MAE:",
    mae
)

print(
    "RMSE:",
    rmse
)

print(
    "MAPE:",
    mape,
    "%"
)



# =======================================================
# PLOT ACTUAL VS FORECAST
# =======================================================

plt.figure(
    figsize=(15, 7)
)


# Plot training data

plt.plot(
    train.index,
    train.values,
    label="Training Data"
)


# Plot actual test data

plt.plot(
    test.index,
    test.values,
    label="Actual Test Data"
)


# Plot ARIMA forecast

plt.plot(
    forecast.index,
    forecast.values,
    label="ARIMA Forecast"
)


# Graph title

plt.title(
    "ARIMA Forecast: Motion Watch Activity"
)


# X axis

plt.xlabel(
    "Time"
)


# Y axis

plt.ylabel(
    "Activity"
)


# Show legend

plt.legend()


# Show grid

plt.grid(
    True,
    alpha=0.3
)


# Adjust layout

plt.tight_layout()


# Display graph

plt.show()



# =======================================================
# DISPLAY MODEL SUMMARY
# =======================================================

print("\n============================================")
print("ARIMA MODEL SUMMARY")
print("============================================")

print(
    best_fitted_model.summary()
)
```

```python
# ============================================================
# ARIMA MODEL FITTING AND FINAL ANALYSIS
# Motion Watch Activity Dataset
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

# os is used for folder and file operations
import os

# warnings is used to hide unnecessary warning messages
import warnings

# numpy is used for numerical calculations
import numpy as np

# pandas is used for reading and processing the data
import pandas as pd

# matplotlib is used for creating graphs
import matplotlib.pyplot as plt

# ARIMA model from statsmodels
from statsmodels.tsa.arima.model import ARIMA

# Ljung Box test is used for residual analysis
from statsmodels.stats.diagnostic import acorr_ljungbox


# Ignore unnecessary warning messages
warnings.filterwarnings("ignore")


# ============================================================
# 2. DEFINE DATA FOLDERS
# ============================================================

# Main dataset directory
BASE_PATH = "."

# Motion Watch folders
FOLDERS = {
    "Class Lecture": "./class_lecture/clmw",
    "Music": "./music/mumw",
    "Normal": "./normal/normw",
    "Reading": "./reading/rdmw"
}


# ============================================================
# 3. ARIMA CONFIGURATION
# ============================================================

# ARIMA order
# p = 3
# d = 0
# q = 3
ARIMA_ORDER = (3, 0, 3)

# Percentage of data used for testing
TEST_RATIO = 0.20

# Minimum number of observations required
MINIMUM_ROWS = 100

# Number of observations used for residual Ljung Box test
LJUNG_BOX_LAGS = 10

# Folder where analysis results will be saved
OUTPUT_FOLDER = "./arima_results"

# Create output folder if it does not already exist
os.makedirs(OUTPUT_FOLDER, exist_ok=True)


# ============================================================
# 4. FUNCTION TO READ MOTION WATCH FILE
# ============================================================

def read_motion_watch_file(file_path):
    """
    Read a Motion Watch file.

    The function first checks whether the file is a normal CSV.
    If the file is actually an Excel file with a CSV extension,
    it attempts to read it as an Excel file.
    """

    # --------------------------------------------------------
    # Check whether the file is actually an Excel file
    # --------------------------------------------------------

    try:

        # Open the file in binary mode
        with open(file_path, "rb") as file:

            # Read the first four bytes
            file_signature = file.read(4)

    except Exception:

        # Return None if the file cannot be opened
        return None


    # --------------------------------------------------------
    # Excel files commonly begin with PK
    # --------------------------------------------------------

    if file_signature == b"PK\x03\x04":

        try:

            # Read the Excel file
            dataframe = pd.read_excel(file_path)

            return dataframe

        except Exception:

            # Return None if Excel reading fails
            return None


    # --------------------------------------------------------
    # Otherwise read as CSV
    # --------------------------------------------------------

    try:

        # Read standard CSV
        dataframe = pd.read_csv(
            file_path,
            low_memory=False
        )

        return dataframe

    except Exception:

        # Try Python CSV engine for irregular CSV files
        try:

            dataframe = pd.read_csv(
                file_path,
                engine="python"
            )

            return dataframe

        except Exception:

            return None


# ============================================================
# 5. FUNCTION TO PREPARE THE DATA
# ============================================================

def prepare_motion_data(dataframe):
    """
    Prepare Motion Watch data for ARIMA analysis.
    """

    # Make a copy so that the original dataframe is not changed
    df = dataframe.copy()


    # --------------------------------------------------------
    # Remove unnecessary spaces from column names
    # --------------------------------------------------------

    df.columns = [
        str(column).strip()
        for column in df.columns
    ]


    # --------------------------------------------------------
    # Check Date column
    # --------------------------------------------------------

    if "Date" not in df.columns:

        return None


    # --------------------------------------------------------
    # Check Time column
    # --------------------------------------------------------

    if "Time" not in df.columns:

        return None


    # --------------------------------------------------------
    # Identify Activity column
    # --------------------------------------------------------

    # If Activity already exists, use it
    if "Activity" in df.columns:

        activity_column = "Activity"

    else:

        # If Activity does not exist, use the third column
        # because your Motion Watch files contain three columns
        if len(df.columns) < 3:

            return None

        activity_column = df.columns[2]


    # --------------------------------------------------------
    # Rename activity column
    # --------------------------------------------------------

    df = df.rename(
        columns={
            activity_column: "Activity"
        }
    )


    # --------------------------------------------------------
    # Convert Date and Time into one datetime column
    # --------------------------------------------------------

    df["DateTime"] = pd.to_datetime(
        df["Date"].astype(str) + " " +
        df["Time"].astype(str),
        errors="coerce"
    )


    # --------------------------------------------------------
    # Convert Activity to numeric
    # --------------------------------------------------------

    df["Activity"] = pd.to_numeric(
        df["Activity"],
        errors="coerce"
    )


    # --------------------------------------------------------
    # Remove invalid observations
    # --------------------------------------------------------

    df = df.dropna(
        subset=["DateTime", "Activity"]
    )


    # --------------------------------------------------------
    # Sort observations according to time
    # --------------------------------------------------------

    df = df.sort_values(
        "DateTime"
    )


    # --------------------------------------------------------
    # Remove duplicated timestamps
    # --------------------------------------------------------

    df = df.drop_duplicates(
        subset=["DateTime"]
    )


    # --------------------------------------------------------
    # Set DateTime as index
    # --------------------------------------------------------

    df = df.set_index(
        "DateTime"
    )


    # Return prepared dataframe
    return df


# ============================================================
# 6. FUNCTION FOR ERROR METRICS
# ============================================================

def calculate_metrics(actual, predicted):
    """
    Calculate forecasting performance metrics.
    """

    # Convert values to numpy arrays
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)


    # --------------------------------------------------------
    # Mean Absolute Error
    # --------------------------------------------------------

    mae = np.mean(
        np.abs(actual - predicted)
    )


    # --------------------------------------------------------
    # Root Mean Squared Error
    # --------------------------------------------------------

    rmse = np.sqrt(
        np.mean(
            (actual - predicted) ** 2
        )
    )


    # --------------------------------------------------------
    # Symmetric Mean Absolute Percentage Error
    #
    # sMAPE is used instead of ordinary MAPE because
    # Motion Watch activity contains many zero values.
    # --------------------------------------------------------

    denominator = (
        np.abs(actual) +
        np.abs(predicted)
    )

    valid = denominator != 0

    if np.any(valid):

        smape = np.mean(
            2 *
            np.abs(
                actual[valid] -
                predicted[valid]
            ) /
            denominator[valid]
        ) * 100

    else:

        smape = 0.0


    # Return metrics
    return mae, rmse, smape


# ============================================================
# 7. CREATE RESULT STORAGE
# ============================================================

# This list will contain the final result of every file
all_results = []


# ============================================================
# 8. PROCESS ALL FOUR MOTION WATCH FOLDERS
# ============================================================

print("=" * 70)
print("ARIMA MODEL FITTING STARTED")
print("=" * 70)


# Loop through the four activity classes
for activity_class, folder_path in FOLDERS.items():

    print()
    print("=" * 70)
    print("Processing:", activity_class)
    print("Folder:", folder_path)
    print("=" * 70)


    # --------------------------------------------------------
    # Check whether folder exists
    # --------------------------------------------------------

    if not os.path.exists(folder_path):

        print("Folder not found:", folder_path)

        continue


    # --------------------------------------------------------
    # Find all CSV files
    # --------------------------------------------------------

    csv_files = sorted(
        [
            file_name
            for file_name in os.listdir(folder_path)
            if file_name.lower().endswith(".csv")
        ]
    )


    print(
        "Files found:",
        len(csv_files)
    )


    # --------------------------------------------------------
    # Process every CSV file
    # --------------------------------------------------------

    for file_number, file_name in enumerate(
        csv_files,
        start=1
    ):

        # Complete file path
        file_path = os.path.join(
            folder_path,
            file_name
        )


        print(
            f"[{file_number}/{len(csv_files)}] "
            f"Processing {file_name}"
        )


        # ----------------------------------------------------
        # Read file
        # ----------------------------------------------------

        raw_df = read_motion_watch_file(
            file_path
        )


        if raw_df is None:

            print(
                "  Could not read file"
            )

            continue


        # ----------------------------------------------------
        # Prepare data
        # ----------------------------------------------------

        df = prepare_motion_data(
            raw_df
        )


        if df is None:

            print(
                "  Data preparation failed"
            )

            continue


        # Number of observations
        number_of_rows = len(df)


        # ----------------------------------------------------
        # Check minimum number of observations
        # ----------------------------------------------------

        if number_of_rows < MINIMUM_ROWS:

            print(
                "  Not enough observations:",
                number_of_rows
            )

            continue


        # ----------------------------------------------------
        # Extract activity series
        # ----------------------------------------------------

        activity = df["Activity"].astype(float)


        # ----------------------------------------------------
        # Determine training and testing size
        # ----------------------------------------------------

        test_size = max(
            1,
            int(
                len(activity) *
                TEST_RATIO
            )
        )


        train_size = (
            len(activity) -
            test_size
        )


        # ----------------------------------------------------
        # Create training and testing datasets
        # ----------------------------------------------------

        train_data = activity.iloc[
            :train_size
        ]

        test_data = activity.iloc[
            train_size:
        ]


        # ----------------------------------------------------
        # Fit ARIMA model
        # ----------------------------------------------------

        try:

            model = ARIMA(
                train_data,
                order=ARIMA_ORDER
            )

            fitted_model = model.fit()


        except Exception as error:

            print(
                "  ARIMA fitting failed:",
                error
            )

            continue


        # ----------------------------------------------------
        # Generate forecast
        # ----------------------------------------------------

        try:

            forecast = fitted_model.forecast(
                steps=len(test_data)
            )


        except Exception as error:

            print(
                "  Forecast failed:",
                error
            )

            continue


        # ----------------------------------------------------
        # Calculate forecasting metrics
        # ----------------------------------------------------

        mae, rmse, smape = calculate_metrics(
            test_data.values,
            forecast.values
        )


        # ----------------------------------------------------
        # Extract AIC and BIC
        # ----------------------------------------------------

        aic = fitted_model.aic
        bic = fitted_model.bic


        # ----------------------------------------------------
        # Ljung Box residual test
        # ----------------------------------------------------

        try:

            residuals = fitted_model.resid

            ljung_result = acorr_ljungbox(
                residuals,
                lags=[LJUNG_BOX_LAGS],
                return_df=True
            )

            ljung_box_statistic = float(
                ljung_result[
                    "lb_stat"
                ].iloc[0]
            )

            ljung_box_pvalue = float(
                ljung_result[
                    "lb_pvalue"
                ].iloc[0]
            )

        except Exception:

            ljung_box_statistic = np.nan
            ljung_box_pvalue = np.nan


        # ----------------------------------------------------
        # Determine whether residuals show significant
        # autocorrelation at the selected lag
        # ----------------------------------------------------

        if not np.isnan(
            ljung_box_pvalue
        ):

            if ljung_box_pvalue > 0.05:

                residual_status = (
                    "No significant autocorrelation"
                )

            else:

                residual_status = (
                    "Significant autocorrelation"
                )

        else:

            residual_status = "Not available"


        # ----------------------------------------------------
        # Store results
        # ----------------------------------------------------

        result = {

            "Activity Class": activity_class,

            "File": file_name,

            "Rows": number_of_rows,

            "Training Rows": train_size,

            "Testing Rows": test_size,

            "ARIMA p": ARIMA_ORDER[0],

            "ARIMA d": ARIMA_ORDER[1],

            "ARIMA q": ARIMA_ORDER[2],

            "AIC": aic,

            "BIC": bic,

            "MAE": mae,

            "RMSE": rmse,

            "sMAPE": smape,

            "Ljung Box Statistic": (
                ljung_box_statistic
            ),

            "Ljung Box P Value": (
                ljung_box_pvalue
            ),

            "Residual Status": (
                residual_status
            )
        }


        all_results.append(
            result
        )


        # ----------------------------------------------------
        # Display current result
        # ----------------------------------------------------

        print(
            f"  Rows: {number_of_rows} | "
            f"MAE: {mae:.4f} | "
            f"RMSE: {rmse:.4f} | "
            f"sMAPE: {smape:.2f}%"
        )


# ============================================================
# 9. CREATE FINAL RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    all_results
)


# ============================================================
# 10. CHECK WHETHER RESULTS ARE AVAILABLE
# ============================================================

if len(results_df) == 0:

    print()
    print(
        "No valid ARIMA results were generated."
    )

else:

    # --------------------------------------------------------
    # Save complete results
    # --------------------------------------------------------

    results_file = os.path.join(
        OUTPUT_FOLDER,
        "ARIMA_All_File_Results.csv"
    )

    results_df.to_csv(
        results_file,
        index=False
    )


    print()
    print("=" * 70)
    print("ARIMA ANALYSIS COMPLETED")
    print("=" * 70)

    print(
        "Total files successfully analysed:",
        len(results_df)
    )

    print(
        "Results saved to:",
        results_file
    )


# ============================================================
# 11. OVERALL PERFORMANCE
# ============================================================

if len(results_df) > 0:

    print()
    print("=" * 70)
    print("OVERALL ARIMA PERFORMANCE")
    print("=" * 70)


    overall_mae = results_df[
        "MAE"
    ].mean()

    overall_rmse = results_df[
        "RMSE"
    ].mean()

    overall_smape = results_df[
        "sMAPE"
    ].mean()

    overall_aic = results_df[
        "AIC"
    ].mean()

    overall_bic = results_df[
        "BIC"
    ].mean()


    print(
        f"Mean MAE:   {overall_mae:.4f}"
    )

    print(
        f"Mean RMSE:  {overall_rmse:.4f}"
    )

    print(
        f"Mean sMAPE: {overall_smape:.2f}%"
    )

    print(
        f"Mean AIC:   {overall_aic:.4f}"
    )

    print(
        f"Mean BIC:   {overall_bic:.4f}"
    )


# ============================================================
# 12. CLASS WISE ARIMA PERFORMANCE
# ============================================================

if len(results_df) > 0:

    class_summary = (
        results_df
        .groupby("Activity Class")
        .agg(
            Files=("File", "count"),
            Mean_MAE=("MAE", "mean"),
            Mean_RMSE=("RMSE", "mean"),
            Mean_sMAPE=("sMAPE", "mean"),
            Mean_AIC=("AIC", "mean"),
            Mean_BIC=("BIC", "mean")
        )
        .reset_index()
    )


    # --------------------------------------------------------
    # Save class summary
    # --------------------------------------------------------

    class_summary_file = os.path.join(
        OUTPUT_FOLDER,
        "ARIMA_Class_Summary.csv"
    )

    class_summary.to_csv(
        class_summary_file,
        index=False
    )


    print()
    print("=" * 70)
    print("CLASS WISE ARIMA PERFORMANCE")
    print("=" * 70)

    print(
        class_summary.to_string(
            index=False
        )
    )


# ============================================================
# 13. BEST AND WORST FILES ACCORDING TO RMSE
# ============================================================

if len(results_df) > 0:

    # Sort files according to RMSE
    sorted_results = results_df.sort_values(
        "RMSE"
    )


    print()
    print("=" * 70)
    print("BEST ARIMA FILES")
    print("=" * 70)


    print(
        sorted_results[
            [
                "Activity Class",
                "File",
                "RMSE",
                "MAE",
                "sMAPE"
            ]
        ]
        .head(10)
        .to_string(
            index=False
        )
    )


    print()
    print("=" * 70)
    print("WORST ARIMA FILES")
    print("=" * 70)


    print(
        sorted_results[
            [
                "Activity Class",
                "File",
                "RMSE",
                "MAE",
                "sMAPE"
            ]
        ]
        .tail(10)
        .sort_values(
            "RMSE",
            ascending=False
        )
        .to_string(
            index=False
        )
    )


# ============================================================
# 14. RESIDUAL DIAGNOSTIC SUMMARY
# ============================================================

if len(results_df) > 0:

    # Count files with acceptable residual behaviour
    acceptable_residuals = (
        results_df[
            "Ljung Box P Value"
        ] > 0.05
    ).sum()


    # Total analysed files
    total_files = len(
        results_df
    )


    print()
    print("=" * 70)
    print("RESIDUAL DIAGNOSTIC SUMMARY")
    print("=" * 70)


    print(
        "Files with Ljung Box P Value greater than 0.05:",
        acceptable_residuals
    )

    print(
        "Total analysed files:",
        total_files
    )


# ============================================================
# 15. CREATE CLASS PERFORMANCE GRAPH
# ============================================================

if len(results_df) > 0:

    # Create figure
    plt.figure(
        figsize=(12, 7)
    )


    # Calculate mean RMSE for each class
    class_rmse = (
        results_df
        .groupby("Activity Class")[
            "RMSE"
        ]
        .mean()
    )


    # Create bar graph
    class_rmse.plot(
        kind="bar"
    )


    # Add graph title
    plt.title(
        "Mean ARIMA RMSE by Activity Class"
    )


    # Add X axis label
    plt.xlabel(
        "Activity Class"
    )


    # Add Y axis label
    plt.ylabel(
        "Mean RMSE"
    )


    # Rotate class names
    plt.xticks(
        rotation=0
    )


    # Add grid
    plt.grid(
        axis="y",
        alpha=0.3
    )


    # Adjust layout
    plt.tight_layout()


    # Save graph
    graph_file = os.path.join(
        OUTPUT_FOLDER,
        "ARIMA_Class_RMSE.png"
    )


    plt.savefig(
        graph_file,
        dpi=300,
        bbox_inches="tight"
    )


    # Display graph
    plt.show()


# ============================================================
# 16. SELECT ONE FILE FOR DETAILED ARIMA GRAPH
# ============================================================

# Change this value if you want another file
SELECTED_CLASS = "Class Lecture"

SELECTED_FILE = "clmw1.csv"


# ============================================================
# 17. LOAD SELECTED FILE
# ============================================================

selected_folder = FOLDERS[
    SELECTED_CLASS
]


selected_file_path = os.path.join(
    selected_folder,
    SELECTED_FILE
)


selected_raw = read_motion_watch_file(
    selected_file_path
)


selected_df = prepare_motion_data(
    selected_raw
)


# ============================================================
# 18. FIT ARIMA TO SELECTED FILE
# ============================================================

if selected_df is not None:

    selected_activity = (
        selected_df["Activity"]
        .astype(float)
    )


    # Determine test size
    selected_test_size = max(
        1,
        int(
            len(selected_activity) *
            TEST_RATIO
        )
    )


    # Split training data
    selected_train = selected_activity.iloc[
        :-selected_test_size
    ]


    # Split testing data
    selected_test = selected_activity.iloc[
        -selected_test_size:
    ]


    # Fit ARIMA model
    selected_model = ARIMA(
        selected_train,
        order=ARIMA_ORDER
    )


    selected_fitted = selected_model.fit()


    # Generate forecast
    selected_forecast = (
        selected_fitted
        .forecast(
            steps=len(selected_test)
        )
    )


    # ========================================================
    # 19. PRINT SELECTED MODEL SUMMARY
    # ========================================================

    print()
    print("=" * 70)
    print("SELECTED FILE ARIMA MODEL SUMMARY")
    print("=" * 70)

    print(
        "Activity Class:",
        SELECTED_CLASS
    )

    print(
        "File:",
        SELECTED_FILE
    )

    print(
        "ARIMA Order:",
        ARIMA_ORDER
    )

    print(
        "Total observations:",
        len(selected_activity)
    )

    print(
        "Training observations:",
        len(selected_train)
    )

    print(
        "Testing observations:",
        len(selected_test)
    )


    print()
    print(
        selected_fitted.summary()
    )


    # ========================================================
    # 20. SELECTED FILE METRICS
    # ========================================================

    selected_mae, selected_rmse, selected_smape = (
        calculate_metrics(
            selected_test.values,
            selected_forecast.values
        )
    )


    print()
    print("=" * 70)
    print("SELECTED FILE FORECASTING PERFORMANCE")
    print("=" * 70)

    print(
        f"MAE:   {selected_mae:.4f}"
    )

    print(
        f"RMSE:  {selected_rmse:.4f}"
    )

    print(
        f"sMAPE: {selected_smape:.2f}%"
    )


    # ========================================================
    # 21. PLOT ARIMA FORECAST
    # ========================================================

    plt.figure(
        figsize=(15, 7)
    )


    # Plot training data
    plt.plot(
        selected_train.index,
        selected_train.values,
        label="Training Data"
    )


    # Plot actual testing data
    plt.plot(
        selected_test.index,
        selected_test.values,
        label="Actual Test Data"
    )


    # Plot ARIMA forecast
    plt.plot(
        selected_test.index,
        selected_forecast.values,
        label="ARIMA Forecast"
    )


    # Add title
    plt.title(
        f"ARIMA Forecast: "
        f"{SELECTED_CLASS} | "
        f"{SELECTED_FILE}"
    )


    # Add X axis label
    plt.xlabel(
        "Time"
    )


    # Add Y axis label
    plt.ylabel(
        "Activity"
    )


    # Display legend
    plt.legend()


    # Add grid
    plt.grid(
        alpha=0.3
    )


    # Improve layout
    plt.tight_layout()


    # Save forecast graph
    forecast_graph = os.path.join(
        OUTPUT_FOLDER,
        "Selected_File_ARIMA_Forecast.png"
    )


    plt.savefig(
        forecast_graph,
        dpi=300,
        bbox_inches="tight"
    )


    # Display graph
    plt.show()


# ============================================================
# 22. FINAL MESSAGE
# ============================================================

print()
print("=" * 70)
print("FINAL ARIMA ANALYSIS FINISHED")
print("=" * 70)

print(
    "All available results have been saved in:",
    OUTPUT_FOLDER
)
```

```python
# ============================================================
# OUTLIER ANALYSIS FOR ALL MOTION WATCH AND MUSE CSV FILES
# ============================================================

# ------------------------------------------------------------
# Import required libraries
# ------------------------------------------------------------

# os is used for folder and file operations
import os

# warnings is used to hide unnecessary warning messages
import warnings

# numpy is used for numerical calculations
import numpy as np

# pandas is used for reading and analysing CSV files
import pandas as pd

# matplotlib is used for graph visualization
import matplotlib.pyplot as plt

# Disable unnecessary warning messages
warnings.filterwarnings("ignore")


# ============================================================
# 1. CONFIGURATION
# ============================================================

# Main project folder
BASE_PATH = "."

# Output folder for all outlier analysis results
OUTPUT_FOLDER = os.path.join(BASE_PATH, "outlier_results")

# Create output folder if it does not already exist
os.makedirs(OUTPUT_FOLDER, exist_ok=True)


# ------------------------------------------------------------
# Define Motion Watch folders
# ------------------------------------------------------------

motion_folders = {
    "Class Lecture": os.path.join(
        BASE_PATH,
        "class_lecture",
        "clmw"
    ),

    "Music": os.path.join(
        BASE_PATH,
        "music",
        "mumw"
    ),

    "Normal": os.path.join(
        BASE_PATH,
        "normal",
        "normw"
    ),

    "Reading": os.path.join(
        BASE_PATH,
        "reading",
        "rdmw"
    )
}


# ------------------------------------------------------------
# Define Muse folders
# ------------------------------------------------------------

muse_folders = {
    "Class Lecture": os.path.join(
        BASE_PATH,
        "class_lecture",
        "clms"
    ),

    "Music": os.path.join(
        BASE_PATH,
        "music",
        "mums"
    ),

    "Normal": os.path.join(
        BASE_PATH,
        "normal",
        "norms"
    ),

    "Reading": os.path.join(
        BASE_PATH,
        "reading",
        "rdms"
    )
}


# ============================================================
# 2. IQR OUTLIER FUNCTION
# ============================================================

def detect_iqr_outliers(series):
    """
    Detect outliers using the IQR method.

    IQR means Interquartile Range.

    Lower limit:
    Q1 minus 1.5 times IQR

    Upper limit:
    Q3 plus 1.5 times IQR
    """

    # Convert values to numeric values
    numeric_values = pd.to_numeric(
        series,
        errors="coerce"
    )

    # Remove missing values
    numeric_values = numeric_values.dropna()

    # Check whether enough values are available
    if len(numeric_values) < 4:
        return (
            pd.Series(False, index=series.index),
            np.nan,
            np.nan
        )

    # Calculate first quartile
    q1 = numeric_values.quantile(0.25)

    # Calculate third quartile
    q3 = numeric_values.quantile(0.75)

    # Calculate interquartile range
    iqr = q3 - q1

    # Calculate lower boundary
    lower_limit = q1 - 1.5 * iqr

    # Calculate upper boundary
    upper_limit = q3 + 1.5 * iqr

    # Detect outliers
    outlier_mask = (
        (pd.to_numeric(series, errors="coerce") < lower_limit) |
        (pd.to_numeric(series, errors="coerce") > upper_limit)
    )

    return (
        outlier_mask,
        lower_limit,
        upper_limit
    )


# ============================================================
# 3. FUNCTION TO READ CSV FILE
# ============================================================

def read_data_file(file_path):
    """
    Read a CSV file.

    Some files may have been saved in Excel format while
    retaining a CSV extension. This function first attempts
    normal CSV reading and then attempts Excel reading.
    """

    try:

        # First attempt to read as normal CSV
        df = pd.read_csv(
            file_path,
            low_memory=False
        )

        return df

    except Exception:

        # If normal CSV reading fails, try Excel format
        try:

            df = pd.read_excel(
                file_path
            )

            return df

        except Exception:

            return None


# ============================================================
# 4. MOTION WATCH OUTLIER ANALYSIS
# ============================================================

print("=" * 70)
print("MOTION WATCH OUTLIER ANALYSIS")
print("=" * 70)


# List for storing all Motion Watch outliers
motion_outlier_records = []

# List for storing file level summary
motion_file_summary = []

# Total Motion Watch files
motion_total_files = 0

# Successfully processed Motion Watch files
motion_success_files = 0

# Failed Motion Watch files
motion_failed_files = 0


# ------------------------------------------------------------
# Process each Motion Watch activity class
# ------------------------------------------------------------

for activity_class, folder_path in motion_folders.items():

    print()
    print("Processing:", activity_class)
    print("Folder:", folder_path)

    # Check whether folder exists
    if not os.path.exists(folder_path):

        print("Folder not found:", folder_path)

        continue

    # Get all CSV files
    csv_files = sorted(
        [
            file_name
            for file_name in os.listdir(folder_path)
            if file_name.lower().endswith(".csv")
        ]
    )

    print("Files found:", len(csv_files))

    # Process every CSV file
    for file_name in csv_files:

        motion_total_files += 1

        # Create complete file path
        file_path = os.path.join(
            folder_path,
            file_name
        )

        # Read file
        df = read_data_file(file_path)

        # Check whether file was successfully read
        if df is None:

            motion_failed_files += 1

            print(
                "Could not read:",
                file_name
            )

            continue

        motion_success_files += 1

        # ----------------------------------------------------
        # Find Activity column
        # ----------------------------------------------------

        activity_column = None

        for column in df.columns:

            # Convert column name to text
            column_text = str(column).strip().lower()

            # Check for Activity column
            if column_text == "activity":

                activity_column = column
                break

        # If Activity column does not exist,
        # use the third column when available
        if activity_column is None and len(df.columns) >= 3:

            activity_column = df.columns[2]

        # If no suitable column exists, skip the file
        if activity_column is None:

            print(
                "Activity column not found:",
                file_name
            )

            continue

        # Convert Activity column to numeric
        activity_values = pd.to_numeric(
            df[activity_column],
            errors="coerce"
        )

        # Detect outliers
        outlier_mask, lower_limit, upper_limit = (
            detect_iqr_outliers(activity_values)
        )

        # Get outlier row positions
        outlier_positions = np.where(
            outlier_mask.fillna(False).to_numpy()
        )[0]

        # ----------------------------------------------------
        # Save every Motion Watch outlier
        # ----------------------------------------------------

        for position in outlier_positions:

            # Get original row number
            row_number = position + 2

            # Get activity value
            activity_value = activity_values.iloc[position]

            # Get Date if available
            date_value = ""

            if "Date" in df.columns:

                date_value = df.iloc[position]["Date"]

            # Get Time if available
            time_value = ""

            if "Time" in df.columns:

                time_value = df.iloc[position]["Time"]

            # Store outlier information
            motion_outlier_records.append(
                {
                    "Activity Class": activity_class,
                    "File Name": file_name,
                    "Row Number": row_number,
                    "Date": date_value,
                    "Time": time_value,
                    "Activity": activity_value,
                    "Lower Limit": lower_limit,
                    "Upper Limit": upper_limit
                }
            )

        # ----------------------------------------------------
        # Save file level summary
        # ----------------------------------------------------

        motion_file_summary.append(
            {
                "Activity Class": activity_class,
                "File Name": file_name,
                "Total Rows": len(df),
                "Outlier Count": len(outlier_positions),
                "Outlier Percentage":
                    (len(outlier_positions) / len(df)) * 100
                    if len(df) > 0 else 0
            }
        )

    print("Completed:", activity_class)


# ============================================================
# 5. CREATE MOTION WATCH OUTLIER DATAFRAME
# ============================================================

motion_outliers_df = pd.DataFrame(
    motion_outlier_records
)

motion_summary_df = pd.DataFrame(
    motion_file_summary
)


# Save Motion Watch outlier information
motion_outlier_file = os.path.join(
    OUTPUT_FOLDER,
    "motion_watch_outliers.csv"
)

motion_outliers_df.to_csv(
    motion_outlier_file,
    index=False
)


# Save Motion Watch file summary
motion_summary_file = os.path.join(
    OUTPUT_FOLDER,
    "motion_watch_file_summary.csv"
)

motion_summary_df.to_csv(
    motion_summary_file,
    index=False
)


# ============================================================
# 6. MOTION WATCH OUTLIER GRAPH
# ============================================================

if not motion_outliers_df.empty:

    plt.figure(
        figsize=(16, 8)
    )

    # Define class order
    class_order = [
        "Class Lecture",
        "Music",
        "Normal",
        "Reading"
    ]

    # Create numeric positions for activity classes
    class_positions = {
        name: index
        for index, name in enumerate(class_order)
    }

    # Plot each activity class
    for activity_class in class_order:

        class_data = motion_outliers_df[
            motion_outliers_df["Activity Class"] ==
            activity_class
        ]

        if class_data.empty:
            continue

        # Create horizontal position with small random variation
        # so that overlapping points can be seen
        rng = np.random.default_rng(42)

        x_values = (
            class_positions[activity_class]
            +
            rng.uniform(
                -0.18,
                0.18,
                len(class_data)
            )
        )

        # Plot outlier values
        plt.scatter(
            x_values,
            class_data["Activity"],
            s=18,
            alpha=0.6,
            label=activity_class
        )

    # Set X axis labels
    plt.xticks(
        range(len(class_order)),
        class_order
    )

    # Set graph title
    plt.title(
        "Motion Watch Activity Outlier Data Points",
        fontsize=16
    )

    # Set X axis label
    plt.xlabel(
        "Activity Class",
        fontsize=12
    )

    # Set Y axis label
    plt.ylabel(
        "Activity Value",
        fontsize=12
    )

    # Add grid
    plt.grid(
        alpha=0.3
    )

    # Adjust layout
    plt.tight_layout()

    # Save graph
    motion_graph_path = os.path.join(
        OUTPUT_FOLDER,
        "motion_watch_outliers.png"
    )

    plt.savefig(
        motion_graph_path,
        dpi=300,
        bbox_inches="tight"
    )

    # Display graph
    plt.show()

    # Close graph
    plt.close()


# ============================================================
# 7. MUSE OUTLIER ANALYSIS
# ============================================================

print()
print("=" * 70)
print("MUSE OUTLIER ANALYSIS")
print("=" * 70)


# List for storing Muse outliers
muse_outlier_records = []

# List for storing Muse file summary
muse_file_summary = []

# Total Muse files
muse_total_files = 0

# Successfully processed Muse files
muse_success_files = 0

# Failed Muse files
muse_failed_files = 0


# ------------------------------------------------------------
# Process every Muse activity class
# ------------------------------------------------------------

for activity_class, folder_path in muse_folders.items():

    print()
    print("Processing:", activity_class)
    print("Folder:", folder_path)

    # Check folder
    if not os.path.exists(folder_path):

        print("Folder not found:", folder_path)

        continue

    # Get all CSV files
    csv_files = sorted(
        [
            file_name
            for file_name in os.listdir(folder_path)
            if file_name.lower().endswith(".csv")
        ]
    )

    print("Files found:", len(csv_files))

    # Process every file
    for file_name in csv_files:

        muse_total_files += 1

        # Create file path
        file_path = os.path.join(
            folder_path,
            file_name
        )

        # Read file
        df = read_data_file(file_path)

        # Check whether file was read
        if df is None:

            muse_failed_files += 1

            print(
                "Could not read:",
                file_name
            )

            continue

        muse_success_files += 1

        # ----------------------------------------------------
        # Identify numeric columns
        # ----------------------------------------------------

        numeric_columns = []

        for column in df.columns:

            column_name = str(column).strip()

            # Skip Activity column
            if column_name.lower() == "activity":
                continue

            # Skip TimeStamp
            if column_name.lower() == "timestamp":
                continue

            # Try converting the column to numeric
            numeric_values = pd.to_numeric(
                df[column],
                errors="coerce"
            )

            # Count valid numeric values
            valid_count = numeric_values.notna().sum()

            # Consider column numeric if it contains
            # a reasonable number of numeric observations
            if valid_count >= 4:

                numeric_columns.append(
                    column
                )

        # ----------------------------------------------------
        # Analyse every numeric Muse channel
        # ----------------------------------------------------

        file_outlier_count = 0

        for column in numeric_columns:

            # Convert channel to numeric
            values = pd.to_numeric(
                df[column],
                errors="coerce"
            )

            # Detect outliers
            outlier_mask, lower_limit, upper_limit = (
                detect_iqr_outliers(values)
            )

            # Get outlier positions
            outlier_positions = np.where(
                outlier_mask.fillna(False).to_numpy()
            )[0]

            file_outlier_count += len(
                outlier_positions
            )

            # ------------------------------------------------
            # Save every Muse outlier
            # ------------------------------------------------

            for position in outlier_positions:

                # Get row number
                row_number = position + 2

                # Get value
                value = values.iloc[position]

                # Get Timestamp
                timestamp_value = ""

                if "TimeStamp" in df.columns:

                    timestamp_value = df.iloc[
                        position
                    ]["TimeStamp"]

                elif "Timestamp" in df.columns:

                    timestamp_value = df.iloc[
                        position
                    ]["Timestamp"]

                # Save outlier information
                muse_outlier_records.append(
                    {
                        "Activity Class": activity_class,
                        "File Name": file_name,
                        "Row Number": row_number,
                        "Channel": column,
                        "TimeStamp": timestamp_value,
                        "Value": value,
                        "Lower Limit": lower_limit,
                        "Upper Limit": upper_limit
                    }
                )

        # ----------------------------------------------------
        # Save file summary
        # ----------------------------------------------------

        muse_file_summary.append(
            {
                "Activity Class": activity_class,
                "File Name": file_name,
                "Total Rows": len(df),
                "Numeric Channels": len(numeric_columns),
                "Outlier Count": file_outlier_count
            }
        )

    print("Completed:", activity_class)


# ============================================================
# 8. CREATE MUSE OUTLIER DATAFRAME
# ============================================================

muse_outliers_df = pd.DataFrame(
    muse_outlier_records
)

muse_summary_df = pd.DataFrame(
    muse_file_summary
)


# Save Muse outlier information
muse_outlier_file = os.path.join(
    OUTPUT_FOLDER,
    "muse_outliers.csv"
)

muse_outliers_df.to_csv(
    muse_outlier_file,
    index=False
)


# Save Muse file summary
muse_summary_file = os.path.join(
    OUTPUT_FOLDER,
    "muse_file_summary.csv"
)

muse_summary_df.to_csv(
    muse_summary_file,
    index=False
)


# ============================================================
# 9. MUSE OUTLIER GRAPH
# ============================================================

if not muse_outliers_df.empty:

    # Calculate total outliers for every Muse channel
    channel_counts = (
        muse_outliers_df
        .groupby("Channel")
        .size()
        .sort_values(
            ascending=False
        )
    )

    # Select the 20 channels with the highest
    # number of outliers
    top_channels = channel_counts.head(20)

    # Create graph
    plt.figure(
        figsize=(16, 9)
    )

    # Create bar graph
    plt.bar(
        range(len(top_channels)),
        top_channels.values
    )

    # Set X labels
    plt.xticks(
        range(len(top_channels)),
        top_channels.index,
        rotation=75,
        ha="right"
    )

    # Set title
    plt.title(
        "Muse EEG Outlier Count by Channel",
        fontsize=16
    )

    # Set X axis label
    plt.xlabel(
        "EEG Channel",
        fontsize=12
    )

    # Set Y axis label
    plt.ylabel(
        "Number of Outlier Data Points",
        fontsize=12
    )

    # Add grid
    plt.grid(
        axis="y",
        alpha=0.3
    )

    # Adjust layout
    plt.tight_layout()

    # Save graph
    muse_channel_graph = os.path.join(
        OUTPUT_FOLDER,
        "muse_channel_outlier_counts.png"
    )

    plt.savefig(
        muse_channel_graph,
        dpi=300,
        bbox_inches="tight"
    )

    # Display graph
    plt.show()

    # Close graph
    plt.close()


# ============================================================
# 10. MUSE OUTLIER VALUE VISUALIZATION
# ============================================================

if not muse_outliers_df.empty:

    # Select the most frequently occurring channels
    # for a detailed visualization
    selected_channels = (
        muse_outliers_df["Channel"]
        .value_counts()
        .head(10)
        .index
    )

    # Filter selected channels
    plot_data = muse_outliers_df[
        muse_outliers_df["Channel"].isin(
            selected_channels
        )
    ]

    # Create graph
    plt.figure(
        figsize=(16, 9)
    )

    # Create numeric X positions
    channel_list = list(
        selected_channels
    )

    channel_positions = {
        channel: index
        for index, channel in enumerate(
            channel_list
        )
    }

    # Plot every selected channel
    for channel in channel_list:

        channel_data = plot_data[
            plot_data["Channel"] == channel
        ]

        # Add small random horizontal variation
        rng = np.random.default_rng(42)

        x_values = (
            channel_positions[channel]
            +
            rng.uniform(
                -0.15,
                0.15,
                len(channel_data)
            )
        )

        # Plot outlier values
        plt.scatter(
            x_values,
            channel_data["Value"],
            s=15,
            alpha=0.5
        )

    # Set X labels
    plt.xticks(
        range(len(channel_list)),
        channel_list,
        rotation=75,
        ha="right"
    )

    # Set title
    plt.title(
        "Muse EEG Outlier Data Points",
        fontsize=16
    )

    # Set X axis label
    plt.xlabel(
        "EEG Channel",
        fontsize=12
    )

    # Set Y axis label
    plt.ylabel(
        "EEG Value",
        fontsize=12
    )

    # Add grid
    plt.grid(
        alpha=0.3
    )

    # Adjust layout
    plt.tight_layout()

    # Save graph
    muse_value_graph = os.path.join(
        OUTPUT_FOLDER,
        "muse_outlier_data_points.png"
    )

    plt.savefig(
        muse_value_graph,
        dpi=300,
        bbox_inches="tight"
    )

    # Display graph
    plt.show()

    # Close graph
    plt.close()


# ============================================================
# 11. FINAL SUMMARY
# ============================================================

print()
print("=" * 70)
print("FINAL OUTLIER ANALYSIS SUMMARY")
print("=" * 70)

print()
print("MOTION WATCH")
print(
    "Total files:",
    motion_total_files
)

print(
    "Successfully processed:",
    motion_success_files
)

print(
    "Failed files:",
    motion_failed_files
)

print(
    "Total outlier data points:",
    len(motion_outliers_df)
)


print()
print("MUSE")
print(
    "Total files:",
    muse_total_files
)

print(
    "Successfully processed:",
    muse_success_files
)

print(
    "Failed files:",
    muse_failed_files
)

print(
    "Total outlier data points:",
    len(muse_outliers_df)
)


print()
print("=" * 70)
print("RESULT FILES")
print("=" * 70)

print(
    "Motion Watch outliers:",
    motion_outlier_file
)

print(
    "Motion Watch summary:",
    motion_summary_file
)

print(
    "Muse outliers:",
    muse_outlier_file
)

print(
    "Muse summary:",
    muse_summary_file
)

print()
print(
    "All outlier analysis completed successfully."
)
```

```python
# ============================================================
# OUTLIER DETECTION AND VISUALIZATION FOR ALL 480 FILES
# ============================================================

# ------------------------------------------------------------
# Import required libraries
# ------------------------------------------------------------

# os is used for file and folder operations
import os

# warnings is used to suppress unnecessary warning messages
import warnings

# numpy is used for numerical calculations
import numpy as np

# pandas is used for reading and analysing CSV files
import pandas as pd

# matplotlib is used for graph creation
import matplotlib.pyplot as plt

# Suppress unnecessary warnings
warnings.filterwarnings("ignore")


# ============================================================
# 1. CONFIGURATION
# ============================================================

# Main project directory
BASE_PATH = "."

# Main output directory
OUTPUT_PATH = os.path.join(
    BASE_PATH,
    "outlier_results"
)

# Create the main output directory
os.makedirs(
    OUTPUT_PATH,
    exist_ok=True
)


# ------------------------------------------------------------
# Output folders for individual graphs
# ------------------------------------------------------------

# Folder for Motion Watch graphs
MOTION_GRAPH_PATH = os.path.join(
    OUTPUT_PATH,
    "motion_watch_graphs"
)

# Folder for Muse graphs
MUSE_GRAPH_PATH = os.path.join(
    OUTPUT_PATH,
    "muse_graphs"
)

# Create graph folders
os.makedirs(
    MOTION_GRAPH_PATH,
    exist_ok=True
)

os.makedirs(
    MUSE_GRAPH_PATH,
    exist_ok=True
)


# ============================================================
# 2. DEFINE DATA FOLDERS
# ============================================================

# ------------------------------------------------------------
# Motion Watch folders
# ------------------------------------------------------------

motion_folders = {
    "Class Lecture": os.path.join(
        BASE_PATH,
        "class_lecture",
        "clmw"
    ),

    "Music": os.path.join(
        BASE_PATH,
        "music",
        "mumw"
    ),

    "Normal": os.path.join(
        BASE_PATH,
        "normal",
        "normw"
    ),

    "Reading": os.path.join(
        BASE_PATH,
        "reading",
        "rdmw"
    )
}


# ------------------------------------------------------------
# Muse folders
# ------------------------------------------------------------

muse_folders = {
    "Class Lecture": os.path.join(
        BASE_PATH,
        "class_lecture",
        "clms"
    ),

    "Music": os.path.join(
        BASE_PATH,
        "music",
        "mums"
    ),

    "Normal": os.path.join(
        BASE_PATH,
        "normal",
        "norms"
    ),

    "Reading": os.path.join(
        BASE_PATH,
        "reading",
        "rdms"
    )
}


# ============================================================
# 3. READ CSV OR EXCEL FILE
# ============================================================

def read_data_file(file_path):
    """
    Read a data file.

    Some files may have a CSV extension but actually contain
    Excel data. Therefore, the function checks the file format
    before reading it.
    """

    try:

        # ----------------------------------------------------
        # Check the first four bytes of the file
        # ----------------------------------------------------

        with open(
            file_path,
            "rb"
        ) as file:

            file_signature = file.read(4)


        # ----------------------------------------------------
        # XLSX files normally begin with PK
        # ----------------------------------------------------

        if file_signature == b"PK\x03\x04":

            # Read the file as Excel
            return pd.read_excel(
                file_path
            )


        # ----------------------------------------------------
        # Otherwise try normal CSV reading
        # ----------------------------------------------------

        try:

            return pd.read_csv(
                file_path,
                low_memory=False
            )

        except Exception:

            # Try Python CSV parser for irregular files
            return pd.read_csv(
                file_path,
                engine="python",
                on_bad_lines="skip"
            )


    except Exception:

        # ----------------------------------------------------
        # Final attempt: read as Excel
        # ----------------------------------------------------

        try:

            return pd.read_excel(
                file_path
            )

        except Exception:

            return None


# ============================================================
# 4. IQR OUTLIER DETECTION
# ============================================================

def calculate_iqr_outliers(series):
    """
    Detect outliers using the IQR method.

    IQR = Q3 minus Q1

    Lower boundary =
    Q1 minus 1.5 times IQR

    Upper boundary =
    Q3 plus 1.5 times IQR
    """

    # Convert the series into numeric values
    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    # Keep only valid numerical values
    valid_values = values.dropna()

    # Check whether enough values are available
    if len(valid_values) < 4:

        return (
            pd.Series(
                False,
                index=series.index
            ),
            np.nan,
            np.nan
        )


    # --------------------------------------------------------
    # Calculate first quartile
    # --------------------------------------------------------

    q1 = valid_values.quantile(
        0.25
    )


    # --------------------------------------------------------
    # Calculate third quartile
    # --------------------------------------------------------

    q3 = valid_values.quantile(
        0.75
    )


    # --------------------------------------------------------
    # Calculate IQR
    # --------------------------------------------------------

    iqr = q3 - q1


    # --------------------------------------------------------
    # Calculate lower and upper limits
    # --------------------------------------------------------

    lower_limit = q1 - (
        1.5 * iqr
    )

    upper_limit = q3 + (
        1.5 * iqr
    )


    # --------------------------------------------------------
    # Identify outliers
    # --------------------------------------------------------

    outlier_mask = (
        (values < lower_limit) |
        (values > upper_limit)
    )


    return (
        outlier_mask,
        lower_limit,
        upper_limit
    )


# ============================================================
# 5. CREATE SAFE FILE NAME
# ============================================================

def safe_filename(text):
    """
    Convert a folder or file name into a safe name for saving
    graph files.
    """

    # Convert text to string
    text = str(text)

    # Replace spaces
    text = text.replace(
        " ",
        "_"
    )

    # Replace characters that can cause file name problems
    for character in [
        "/",
        "\\",
        ":",
        "*",
        "?",
        "\"",
        "<",
        ">",
        "|"
    ]:

        text = text.replace(
            character,
            "_"
        )

    return text


# ============================================================
# 6. MOTION WATCH ANALYSIS
# ============================================================

print()
print("=" * 80)
print("MOTION WATCH OUTLIER ANALYSIS")
print("=" * 80)


# List for storing Motion Watch results
motion_results = []

# File counter
motion_file_counter = 0


# ------------------------------------------------------------
# Process every Motion Watch activity class
# ------------------------------------------------------------

for activity_class, folder_path in motion_folders.items():

    print()
    print(
        "Processing:",
        activity_class
    )

    # Check whether the folder exists
    if not os.path.exists(folder_path):

        print(
            "Folder not found:",
            folder_path
        )

        continue


    # --------------------------------------------------------
    # Find all CSV files
    # --------------------------------------------------------

    csv_files = sorted(
        [
            file_name
            for file_name in os.listdir(folder_path)
            if file_name.lower().endswith(".csv")
        ]
    )


    print(
        "Files found:",
        len(csv_files)
    )


    # --------------------------------------------------------
    # Process every file
    # --------------------------------------------------------

    for file_name in csv_files:

        motion_file_counter += 1

        print(
            "[",
            motion_file_counter,
            "] Processing:",
            file_name
        )


        # Complete file path
        file_path = os.path.join(
            folder_path,
            file_name
        )


        # Read file
        df = read_data_file(
            file_path
        )


        # Check whether file was successfully read
        if df is None:

            print(
                "Could not read:",
                file_name
            )

            continue


        # ----------------------------------------------------
        # Find Activity column
        # ----------------------------------------------------

        activity_column = None

        for column in df.columns:

            column_name = str(
                column
            ).strip().lower()

            if column_name == "activity":

                activity_column = column

                break


        # ----------------------------------------------------
        # If Activity is not found, check third column
        # ----------------------------------------------------

        if (
            activity_column is None
            and len(df.columns) >= 3
        ):

            activity_column = df.columns[2]


        # ----------------------------------------------------
        # Skip file if Activity cannot be found
        # ----------------------------------------------------

        if activity_column is None:

            print(
                "Activity column not found:",
                file_name
            )

            continue


        # Convert Activity into numerical values
        activity_values = pd.to_numeric(
            df[activity_column],
            errors="coerce"
        )


        # ----------------------------------------------------
        # Detect outliers
        # ----------------------------------------------------

        (
            outlier_mask,
            lower_limit,
            upper_limit
        ) = calculate_iqr_outliers(
            activity_values
        )


        # Get row positions of outliers
        outlier_positions = np.where(
            outlier_mask.fillna(False).to_numpy()
        )[0]


        # ----------------------------------------------------
        # Create X axis
        # ----------------------------------------------------

        # Try to use Date and Time columns
        if (
            "Date" in df.columns
            and "Time" in df.columns
        ):

            datetime_values = pd.to_datetime(
                df["Date"].astype(str)
                + " "
                + df["Time"].astype(str),
                errors="coerce"
            )

            # If datetime conversion fails,
            # use row number instead
            if datetime_values.notna().sum() > 0:

                x_values = datetime_values

                x_label = "Time"


            else:

                x_values = np.arange(
                    len(df)
                )

                x_label = "Sample Number"


        elif "Time" in df.columns:

            # Use Time if Date is unavailable
            x_values = df["Time"]

            x_label = "Time"


        else:

            # Use row number if no time information exists
            x_values = np.arange(
                len(df)
            )

            x_label = "Sample Number"


        # ====================================================
        # CREATE MOTION WATCH GRAPH
        # ====================================================

        plt.figure(
            figsize=(16, 8)
        )


        # ----------------------------------------------------
        # Plot complete Activity signal
        # ----------------------------------------------------

        plt.plot(
            x_values,
            activity_values,
            linewidth=0.8,
            label="Activity"
        )


        # ----------------------------------------------------
        # Plot outlier points
        # ----------------------------------------------------

        if len(outlier_positions) > 0:

            plt.scatter(
                x_values.iloc[outlier_positions]
                if hasattr(
                    x_values,
                    "iloc"
                )
                else x_values[outlier_positions],

                activity_values.iloc[
                    outlier_positions
                ],

                s=35,

                color="red",

                label="Outlier"
            )


        # ----------------------------------------------------
        # Add IQR boundary lines
        # ----------------------------------------------------

        if not np.isnan(
            lower_limit
        ):

            plt.axhline(
                lower_limit,
                linestyle="--",
                linewidth=1,
                label="Lower IQR Limit"
            )


        if not np.isnan(
            upper_limit
        ):

            plt.axhline(
                upper_limit,
                linestyle="--",
                linewidth=1,
                label="Upper IQR Limit"
            )


        # ----------------------------------------------------
        # Graph title
        # ----------------------------------------------------

        plt.title(
            "Motion Watch Outlier Analysis\n"
            + activity_class
            + " | "
            + file_name,
            fontsize=15
        )


        # ----------------------------------------------------
        # Axis labels
        # ----------------------------------------------------

        plt.xlabel(
            x_label,
            fontsize=12
        )

        plt.ylabel(
            "Activity",
            fontsize=12
        )


        # ----------------------------------------------------
        # Grid
        # ----------------------------------------------------

        plt.grid(
            alpha=0.3
        )


        # ----------------------------------------------------
        # Legend
        # ----------------------------------------------------

        plt.legend()


        # ----------------------------------------------------
        # Automatically format time axis
        # ----------------------------------------------------

        plt.gcf().autofmt_xdate()


        # ----------------------------------------------------
        # Create graph file name
        # ----------------------------------------------------

        graph_name = (
            safe_filename(
                activity_class
            )
            + "_"
            + safe_filename(
                os.path.splitext(
                    file_name
                )[0]
            )
            + ".png"
        )


        graph_path = os.path.join(
            MOTION_GRAPH_PATH,
            graph_name
        )


        # ----------------------------------------------------
        # Save high resolution graph
        # ----------------------------------------------------

        plt.savefig(
            graph_path,
            dpi=300,
            bbox_inches="tight"
        )


        # Close graph to release RAM
        plt.close()


        # ----------------------------------------------------
        # Save result information
        # ----------------------------------------------------

        motion_results.append(
            {
                "Activity Class":
                    activity_class,

                "File Name":
                    file_name,

                "Total Rows":
                    len(df),

                "Outlier Count":
                    len(outlier_positions),

                "Outlier Percentage":
                    (
                        len(outlier_positions)
                        /
                        len(df)
                        *
                        100
                    )
                    if len(df) > 0
                    else 0,

                "Lower IQR Limit":
                    lower_limit,

                "Upper IQR Limit":
                    upper_limit,

                "Graph":
                    graph_path
            }
        )


# ============================================================
# 7. SAVE MOTION WATCH SUMMARY
# ============================================================

motion_results_df = pd.DataFrame(
    motion_results
)


motion_summary_path = os.path.join(
    OUTPUT_PATH,
    "motion_watch_outlier_summary.csv"
)


motion_results_df.to_csv(
    motion_summary_path,
    index=False
)


print()
print(
    "Motion Watch analysis completed."
)

print(
    "Motion Watch files processed:",
    motion_file_counter
)


# ============================================================
# 8. MUSE ANALYSIS
# ============================================================

print()
print("=" * 80)
print("MUSE OUTLIER ANALYSIS")
print("=" * 80)


# List for storing Muse results
muse_results = []

# File counter
muse_file_counter = 0


# ------------------------------------------------------------
# Process every Muse activity class
# ------------------------------------------------------------

for activity_class, folder_path in muse_folders.items():

    print()
    print(
        "Processing:",
        activity_class
    )


    # Check folder
    if not os.path.exists(folder_path):

        print(
            "Folder not found:",
            folder_path
        )

        continue


    # --------------------------------------------------------
    # Find CSV files
    # --------------------------------------------------------

    csv_files = sorted(
        [
            file_name
            for file_name in os.listdir(folder_path)
            if file_name.lower().endswith(".csv")
        ]
    )


    print(
        "Files found:",
        len(csv_files)
    )


    # --------------------------------------------------------
    # Process every Muse file
    # --------------------------------------------------------

    for file_name in csv_files:

        muse_file_counter += 1

        print(
            "[",
            muse_file_counter,
            "] Processing:",
            file_name
        )


        # Create file path
        file_path = os.path.join(
            folder_path,
            file_name
        )


        # Read file
        df = read_data_file(
            file_path
        )


        # Check file
        if df is None:

            print(
                "Could not read:",
                file_name
            )

            continue


        # ====================================================
        # FIND TIMESTAMP COLUMN
        # ====================================================

        timestamp_column = None

        for column in df.columns:

            column_name = str(
                column
            ).strip().lower()

            if column_name == "timestamp":

                timestamp_column = column

                break


        # ----------------------------------------------------
        # Create X axis
        # ----------------------------------------------------

        if timestamp_column is not None:

            timestamp_values = pd.to_datetime(
                df[timestamp_column],
                errors="coerce"
            )


            if timestamp_values.notna().sum() > 0:

                x_values = timestamp_values

                x_label = "TimeStamp"


            else:

                x_values = np.arange(
                    len(df)
                )

                x_label = "Sample Number"


        else:

            x_values = np.arange(
                len(df)
            )

            x_label = "Sample Number"


        # ====================================================
        # FIND NUMERICAL MUSE CHANNELS
        # ====================================================

        numeric_columns = []


        for column in df.columns:

            column_name = str(
                column
            ).strip()


            # Do not analyse Timestamp
            if (
                timestamp_column is not None
                and column == timestamp_column
            ):

                continue


            # Convert to numeric
            numeric_values = pd.to_numeric(
                df[column],
                errors="coerce"
            )


            # Count valid numerical observations
            valid_count = (
                numeric_values.notna().sum()
            )


            # Add channel if enough numerical data exists
            if valid_count >= 4:

                numeric_columns.append(
                    column
                )


        # ----------------------------------------------------
        # Check whether numerical channels exist
        # ----------------------------------------------------

        if len(numeric_columns) == 0:

            print(
                "No numerical channels found:",
                file_name
            )

            continue


        # ====================================================
        # CREATE ONE GRAPH FOR THIS MUSE FILE
        # ====================================================

        plt.figure(
            figsize=(18, 10)
        )


        # Counter for total outliers in this file
        total_file_outliers = 0


        # ----------------------------------------------------
        # Analyse every Muse channel
        # ----------------------------------------------------

        for column in numeric_columns:

            # Convert channel to numeric
            values = pd.to_numeric(
                df[column],
                errors="coerce"
            )


            # Detect outliers
            (
                outlier_mask,
                lower_limit,
                upper_limit
            ) = calculate_iqr_outliers(
                values
            )


            # Get outlier positions
            outlier_positions = np.where(
                outlier_mask.fillna(False).to_numpy()
            )[0]


            # Add to total count
            total_file_outliers += len(
                outlier_positions
            )


            # ------------------------------------------------
            # Plot complete channel
            # ------------------------------------------------

            plt.plot(
                x_values,
                values,
                linewidth=0.7,
                alpha=0.65,
                label=str(column)
            )


            # ------------------------------------------------
            # Highlight outlier points
            # ------------------------------------------------

            if len(outlier_positions) > 0:

                if hasattr(
                    x_values,
                    "iloc"
                ):

                    outlier_x = (
                        x_values.iloc[
                            outlier_positions
                        ]
                    )

                else:

                    outlier_x = (
                        x_values[
                            outlier_positions
                        ]
                    )


                outlier_y = values.iloc[
                    outlier_positions
                ]


                plt.scatter(
                    outlier_x,
                    outlier_y,
                    s=18,
                    color="red",
                    alpha=0.8
                )


        # ====================================================
        # MUSE GRAPH SETTINGS
        # ====================================================

        # Graph title
        plt.title(
            "Muse EEG Outlier Analysis\n"
            + activity_class
            + " | "
            + file_name,
            fontsize=15
        )


        # X axis label
        plt.xlabel(
            x_label,
            fontsize=12
        )


        # Y axis label
        plt.ylabel(
            "Signal Value",
            fontsize=12
        )


        # Add grid
        plt.grid(
            alpha=0.3
        )


        # ----------------------------------------------------
        # Create legend
        # ----------------------------------------------------

        plt.legend(
            bbox_to_anchor=(
                1.02,
                1
            ),
            loc="upper left",
            fontsize=8
        )


        # Format timestamp axis
        plt.gcf().autofmt_xdate()


        # Adjust layout
        plt.tight_layout()


        # ====================================================
        # SAVE MUSE GRAPH
        # ====================================================

        graph_name = (
            safe_filename(
                activity_class
            )
            + "_"
            + safe_filename(
                os.path.splitext(
                    file_name
                )[0]
            )
            + ".png"
        )


        graph_path = os.path.join(
            MUSE_GRAPH_PATH,
            graph_name
        )


        # Save high resolution graph
        plt.savefig(
            graph_path,
            dpi=300,
            bbox_inches="tight"
        )


        # Close graph immediately
        # This is important because there are many files
        plt.close()


        # ====================================================
        # SAVE MUSE FILE SUMMARY
        # ====================================================

        muse_results.append(
            {
                "Activity Class":
                    activity_class,

                "File Name":
                    file_name,

                "Total Rows":
                    len(df),

                "Numerical Channels":
                    len(numeric_columns),

                "Total Outlier Count":
                    total_file_outliers,

                "Graph":
                    graph_path
            }
        )


# ============================================================
# 9. SAVE MUSE SUMMARY
# ============================================================

muse_results_df = pd.DataFrame(
    muse_results
)


muse_summary_path = os.path.join(
    OUTPUT_PATH,
    "muse_outlier_summary.csv"
)


muse_results_df.to_csv(
    muse_summary_path,
    index=False
)


print()
print(
    "Muse analysis completed."
)

print(
    "Muse files processed:",
    muse_file_counter
)


# ============================================================
# 10. CREATE OVERALL PROCESSING SUMMARY
# ============================================================

processing_summary = pd.DataFrame(
    {
        "Data Type": [
            "Motion Watch",
            "Muse"
        ],

        "Files Processed": [
            motion_file_counter,
            muse_file_counter
        ]
    }
)


processing_summary_path = os.path.join(
    OUTPUT_PATH,
    "processing_summary.csv"
)


processing_summary.to_csv(
    processing_summary_path,
    index=False
)


# ============================================================
# 11. FINAL INFORMATION
# ============================================================

print()
print("=" * 80)
print("OUTLIER ANALYSIS COMPLETED")
print("=" * 80)

print()
print(
    "Motion Watch files processed:",
    motion_file_counter
)

print(
    "Muse files processed:",
    muse_file_counter
)

print()
print(
    "Motion Watch graphs:",
    MOTION_GRAPH_PATH
)

print(
    "Muse graphs:",
    MUSE_GRAPH_PATH
)

print()
print(
    "Motion Watch summary:",
    motion_summary_path
)

print(
    "Muse summary:",
    muse_summary_path
)

print(
    "Processing summary:",
    processing_summary_path
)

print()
print("=" * 80)
print("ALL INDIVIDUAL OUTLIER GRAPHS HAVE BEEN SAVED")
print("=" * 80)
```

cleaning code with this complete version

```python
# ============================================================
# ROBUST OUTLIER CLEANING FOR ALL CSV FILES
# ============================================================

import os
import io
import warnings
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ============================================================
# 1. MAIN DIRECTORY
# ============================================================

# IMPORTANT:
# This should point to your RAW_DATA folder.

BASE_DIR = Path.home() / "Desktop" / "RAW_DATA"

print("RAW DATA DIRECTORY:")
print(BASE_DIR)

if not BASE_DIR.exists():
    raise FileNotFoundError(
        f"RAW_DATA folder was not found:\n{BASE_DIR}"
    )


# ============================================================
# 2. OUTPUT DIRECTORY
# ============================================================

CLEAN_DIR = BASE_DIR / "CLEAN_DATA"

REPORT_DIR = BASE_DIR / "OUTLIER_REPORT"


# Create output directories.
CLEAN_DIR.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 3. DATA FOLDERS
# ============================================================

DATA_FOLDERS = [

    BASE_DIR / "class_lecture" / "clmw",
    BASE_DIR / "class_lecture" / "clms",

    BASE_DIR / "music" / "mumw",
    BASE_DIR / "music" / "mums",

    BASE_DIR / "normal" / "normw",
    BASE_DIR / "normal" / "norms",

    BASE_DIR / "reading" / "rdmw",
    BASE_DIR / "reading" / "rdms"
]


# ============================================================
# 4. OUTLIER PARAMETERS
# ============================================================

WINDOW_SIZE = 11

HAMPEL_THRESHOLD = 3.5

MAX_OUTLIER_RUN = 3


# ============================================================
# 5. THREAD CONFIGURATION
# ============================================================

# Your i5 10400 has 6 cores.
#
# Four workers are intentionally used here because this task
# involves both disk I/O and numerical processing.

MAX_WORKERS = 4


# ============================================================
# 6. DISPLAY CONFIGURATION
# ============================================================

print("\n" + "=" * 70)
print("CONFIGURATION")
print("=" * 70)

print("RAW_DATA :", BASE_DIR)
print("CLEAN_DATA:", CLEAN_DIR)
print("REPORT_DIR:", REPORT_DIR)

print("Hampel window:", WINDOW_SIZE)
print("Hampel threshold:", HAMPEL_THRESHOLD)
print("Maximum corrected run:", MAX_OUTLIER_RUN)

print("Parallel workers:", MAX_WORKERS)

print("=" * 70)
```

```python
# ============================================================
# FILE TYPE DETECTION
# ============================================================

def detect_file_type(file_path):
    """
    Detect whether a .csv file is actually CSV or Excel.
    """

    try:

        with open(
            file_path,
            "rb"
        ) as f:

            header = f.read(8)


        # XLSX files are ZIP containers.
        if header.startswith(b"PK"):

            return "Excel"


        return "CSV"


    except Exception:

        return "Unknown"


# ============================================================
# READ DATA FILE
# ============================================================

def read_data_file(file_path):

    file_type = detect_file_type(
        file_path
    )


    # --------------------------------------------------------
    # EXCEL FORMAT
    # --------------------------------------------------------

    if file_type == "Excel":

        df = pd.read_excel(
            file_path,
            engine="openpyxl"
        )

        return df, "Excel"


    # --------------------------------------------------------
    # NORMAL CSV
    # --------------------------------------------------------

    try:

        df = pd.read_csv(
            file_path,
            low_memory=False
        )

        return df, "CSV"


    except Exception:

        # Try the more tolerant Python parser.
        df = pd.read_csv(
            file_path,
            engine="python",
            on_bad_lines="warn"
        )

        return df, "CSV"
```

```python
# ============================================================
# HAMPEL OUTLIER DETECTION
# ============================================================

def hampel_outlier_mask(
    series,
    window_size=11,
    threshold=3.5
):

    # Convert values to numeric.
    x = pd.to_numeric(
        series,
        errors="coerce"
    )


    # Rolling median.
    rolling_median = x.rolling(
        window=window_size,
        center=True,
        min_periods=3
    ).median()


    # Absolute deviation.
    deviation = (
        x - rolling_median
    ).abs()


    # Rolling MAD.
    rolling_mad = deviation.rolling(
        window=window_size,
        center=True,
        min_periods=3
    ).median()


    # Robust standard deviation.
    robust_sigma = (
        1.4826 * rolling_mad
    )


    # Avoid division by zero.
    robust_sigma = robust_sigma.replace(
        0,
        np.nan
    )


    # Robust z score.
    robust_z = (
        (x - rolling_median).abs()
        /
        robust_sigma
    )


    # Outlier mask.
    mask = (
        robust_z > threshold
    )


    # Missing observations are not outliers.
    mask = mask.fillna(False)


    return mask
```

```python
# ============================================================
# FIND OUTLIER RUNS
# ============================================================

def find_outlier_runs(mask):

    values = mask.to_numpy(
        dtype=bool
    )

    runs = []

    start = None


    for i, value in enumerate(values):

        # Start of an outlier sequence.
        if value and start is None:

            start = i


        # End of an outlier sequence.
        elif not value and start is not None:

            end = i - 1

            length = (
                end - start + 1
            )

            runs.append(
                (
                    start,
                    end,
                    length
                )
            )

            start = None


    # Handle final run.
    if start is not None:

        end = len(values) - 1

        length = (
            end - start + 1
        )

        runs.append(
            (
                start,
                end,
                length
            )
        )


    return runs


# ============================================================
# CORRECT SHORT OUTLIERS
# ============================================================

def correct_short_outliers(
    series,
    mask,
    maximum_run=3
):

    # Convert to numeric.
    x = pd.to_numeric(
        series,
        errors="coerce"
    ).copy()


    # Store original values.
    cleaned = x.copy()


    corrected = 0

    retained = 0


    # Find consecutive outlier sequences.
    runs = find_outlier_runs(
        mask
    )


    for start, end, length in runs:

        # ----------------------------------------------------
        # SHORT OUTLIER SEQUENCE
        # ----------------------------------------------------

        if length <= maximum_run:

            temporary = x.copy()

            # Temporarily remove outliers.
            temporary.iloc[
                start:end + 1
            ] = np.nan


            # Interpolate.
            temporary = temporary.interpolate(
                method="linear",
                limit_direction="both"
            )


            # Replace outliers.
            for i in range(
                start,
                end + 1
            ):

                if pd.notna(
                    temporary.iloc[i]
                ):

                    cleaned.iloc[i] = (
                        temporary.iloc[i]
                    )

                    corrected += 1

                else:

                    retained += 1


        # ----------------------------------------------------
        # LONG OUTLIER SEQUENCE
        # ----------------------------------------------------

        else:

            retained += length


    return (
        cleaned,
        corrected,
        retained
    )
```

```python
# ============================================================
# PROCESS ONE CSV FILE
# ============================================================

def process_file(file_path):

    file_path = Path(
        file_path
    )


    # --------------------------------------------------------
    # IDENTIFY ACTIVITY AND SENSOR TYPE
    # --------------------------------------------------------

    parts = file_path.parts


    if "class_lecture" in parts:

        activity = "Class Lecture"

    elif "music" in parts:

        activity = "Music"

    elif "normal" in parts:

        activity = "Normal"

    elif "reading" in parts:

        activity = "Reading"

    else:

        activity = "Unknown"


    # Sensor type.
    if file_path.parent.name.lower().endswith("mw"):

        sensor_type = "Motion Watch"

    elif file_path.parent.name.lower().endswith("ms"):

        sensor_type = "Muse"

    else:

        sensor_type = "Unknown"


    # --------------------------------------------------------
    # OUTPUT DIRECTORY
    # --------------------------------------------------------

    # Recreate the exact source folder structure.
    relative_folder = file_path.parent.relative_to(
        BASE_DIR
    )


    output_folder = (
        CLEAN_DIR /
        relative_folder
    )


    output_folder.mkdir(
        parents=True,
        exist_ok=True
    )


    output_path = (
        output_folder /
        file_path.name
    )


    # --------------------------------------------------------
    # RESULT RECORD
    # --------------------------------------------------------

    result = {

        "Activity Class": activity,

        "Sensor": sensor_type,

        "Source File": file_path.name,

        "Source Path": str(file_path),

        "Output Path": str(output_path),

        "Status": "Failed",

        "File Type": "",

        "Rows Before": 0,

        "Rows After": 0,

        "Columns": 0,

        "Numeric Columns": 0,

        "Detected Outliers": 0,

        "Corrected Outliers": 0,

        "Retained Outliers": 0,

        "Error": ""
    }


    try:

        # ----------------------------------------------------
        # READ FILE
        # ----------------------------------------------------

        df, file_type = read_data_file(
            file_path
        )


        result["File Type"] = file_type


        result["Rows Before"] = len(
            df
        )


        result["Columns"] = len(
            df.columns
        )


        # ----------------------------------------------------
        # IDENTIFY NUMERICAL COLUMNS
        # ----------------------------------------------------

        numeric_columns = []


        for column in df.columns:

            column_lower = (
                str(column)
                .strip()
                .lower()
            )


            # Do not process date/time fields.
            if column_lower in [
                "date",
                "time",
                "timestamp",
                "datetime"
            ]:

                continue


            converted = pd.to_numeric(
                df[column],
                errors="coerce"
            )


            # At least one valid numerical value.
            if converted.notna().sum() > 0:

                numeric_columns.append(
                    column
                )


        result["Numeric Columns"] = len(
            numeric_columns
        )


        # ----------------------------------------------------
        # PROCESS NUMERICAL COLUMNS
        # ----------------------------------------------------

        for column in numeric_columns:

            numeric = pd.to_numeric(
                df[column],
                errors="coerce"
            )


            # Detect outliers.
            mask = hampel_outlier_mask(
                numeric,
                window_size=WINDOW_SIZE,
                threshold=HAMPEL_THRESHOLD
            )


            detected = int(
                mask.sum()
            )


            # Correct short sequences.
            cleaned, corrected, retained = (
                correct_short_outliers(
                    numeric,
                    mask,
                    maximum_run=MAX_OUTLIER_RUN
                )
            )


            # Put cleaned data back.
            df[column] = cleaned


            result[
                "Detected Outliers"
            ] += detected


            result[
                "Corrected Outliers"
            ] += corrected


            result[
                "Retained Outliers"
            ] += retained


        # ----------------------------------------------------
        # WRITE OUTPUT CSV
        # ----------------------------------------------------

        df.to_csv(
            output_path,
            index=False
        )


        # ----------------------------------------------------
        # VERIFY OUTPUT FILE
        # ----------------------------------------------------

        if not output_path.exists():

            raise IOError(
                "Output file was not created."
            )


        if output_path.stat().st_size == 0:

            raise IOError(
                "Output file was created but is empty."
            )


        # ----------------------------------------------------
        # VERIFY ROW COUNT
        # ----------------------------------------------------

        result["Rows After"] = len(
            df
        )


        if (
            result["Rows Before"]
            !=
            result["Rows After"]
        ):

            raise ValueError(
                "Row count changed during cleaning."
            )


        result["Status"] = "Completed"


    except Exception as error:

        result["Error"] = str(
            error
        )


    return result
```

```python
# ============================================================
# FIND ALL CSV FILES
# ============================================================

all_csv_files = []


print("=" * 70)
print("SEARCHING FOR CSV FILES")
print("=" * 70)


for folder in DATA_FOLDERS:

    if not folder.exists():

        print(
            "WARNING: Folder not found:",
            folder
        )

        continue


    files = sorted(
        folder.glob("*.csv")
    )


    print(
        f"{folder.relative_to(BASE_DIR)} : "
        f"{len(files)} files"
    )


    all_csv_files.extend(
        files
    )


print("=" * 70)

print(
    "TOTAL CSV FILES FOUND:",
    len(all_csv_files)
)

print("=" * 70)


if len(all_csv_files) == 0:

    raise RuntimeError(
        "No CSV files were found."
    )
```

```python
# ============================================================
# PROCESS ALL FILES
# ============================================================

print("\n")
print("=" * 70)
print("STARTING CLEANING")
print("=" * 70)

print(
    "Total files:",
    len(all_csv_files)
)

print(
    "Workers:",
    MAX_WORKERS
)

print("=" * 70)


results = []


completed = 0


# ------------------------------------------------------------
# THREAD POOL
# ------------------------------------------------------------

with ThreadPoolExecutor(
    max_workers=MAX_WORKERS
) as executor:

    futures = {

        executor.submit(
            process_file,
            file_path
        ): file_path

        for file_path in all_csv_files
    }


    for future in as_completed(
        futures
    ):

        file_path = futures[
            future
        ]


        try:

            result = future.result()


        except Exception as error:

            result = {

                "Activity Class": "Unknown",

                "Sensor": "Unknown",

                "Source File": file_path.name,

                "Source Path": str(file_path),

                "Output Path": "",

                "Status": "Failed",

                "File Type": "",

                "Rows Before": 0,

                "Rows After": 0,

                "Columns": 0,

                "Numeric Columns": 0,

                "Detected Outliers": 0,

                "Corrected Outliers": 0,

                "Retained Outliers": 0,

                "Error": str(error)
            }


        results.append(
            result
        )


        completed += 1


        print(
            f"[{completed}/{len(all_csv_files)}] "
            f"{result['Source File']} | "
            f"{result['Sensor']} | "
            f"{result['Status']} | "
            f"Detected: {result['Detected Outliers']:,} | "
            f"Corrected: {result['Corrected Outliers']:,}"
        )


print("\n")
print("=" * 70)
print("CLEANING COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# RECOVERY CLEANING FOR MISSING DATASET FILES
# ============================================================
#
# This notebook processes ONLY files that are missing from
# CLEAN_DATA.
#
# Existing cleaned files will NOT be modified.
#
# Dataset structure:
#
# RAW_DATA/
#
# class_lecture/
#     clms/
#     clmw/
#
# music/
#     mums/
#     mumw/
#
# normal/
#     norms/
#     normw/
#
# reading/
#     rdms/
#     rdmw/
#
# Total expected files:
# 8 folders x 60 files = 480 files
# ============================================================


# ------------------------------------------------------------
# Import libraries
# ------------------------------------------------------------

from pathlib import Path

import warnings

import numpy as np

import pandas as pd

from concurrent.futures import ThreadPoolExecutor, as_completed


# ------------------------------------------------------------
# Suppress unnecessary warnings
# ------------------------------------------------------------

warnings.filterwarnings(
    "ignore"
)


# ============================================================
# DEFINE MAIN DIRECTORIES
# ============================================================

# Your RAW_DATA folder.
BASE_DIR = (
    Path.home()
    / "Desktop"
    / "RAW_DATA"
)


# Existing cleaned data folder.
CLEAN_DIR = (
    BASE_DIR
    / "CLEAN_DATA"
)


# Outlier report folder.
REPORT_DIR = (
    BASE_DIR
    / "OUTLIER_REPORT"
)


# Create report directory if necessary.
REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# DATASET FOLDERS
# ============================================================

DATASET_FOLDERS = [

    "class_lecture/clms",
    "class_lecture/clmw",

    "music/mums",
    "music/mumw",

    "normal/norms",
    "normal/normw",

    "reading/rdms",
    "reading/rdmw"

]


# ============================================================
# HAMPEL PARAMETERS
# ============================================================

# Rolling window size.
WINDOW_SIZE = 11


# Robust outlier threshold.
HAMPEL_THRESHOLD = 3.5


# Only outlier sequences of this length or shorter
# will be corrected.
MAX_OUTLIER_RUN = 3


# ============================================================
# PARALLEL PROCESSING
# ============================================================

# Your Intel i5 10400 has 6 cores.
#
# Four workers are used to balance processing and disk access.

MAX_WORKERS = 4


# ============================================================
# DISPLAY CONFIGURATION
# ============================================================

print("=" * 80)
print("RECOVERY CLEANING CONFIGURATION")
print("=" * 80)

print("RAW DATA :", BASE_DIR)
print("CLEAN DATA:", CLEAN_DIR)
print("REPORT   :", REPORT_DIR)

print("Hampel Window:", WINDOW_SIZE)
print("Hampel Threshold:", HAMPEL_THRESHOLD)
print("Maximum Corrected Run:", MAX_OUTLIER_RUN)

print("CPU Workers:", MAX_WORKERS)

print("=" * 80)


# ------------------------------------------------------------
# Verify RAW_DATA
# ------------------------------------------------------------

if not BASE_DIR.exists():

    raise FileNotFoundError(
        f"RAW_DATA folder not found:\n{BASE_DIR}"
    )


# ------------------------------------------------------------
# Create CLEAN_DATA
# ------------------------------------------------------------

CLEAN_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
# ============================================================
# FILE TYPE DETECTION
# ============================================================

def detect_file_type(file_path):
    """
    Determine whether the file is a genuine CSV file
    or an Excel file that has a .csv extension.
    """

    try:

        # Open the file as binary.
        with open(
            file_path,
            "rb"
        ) as file:

            header = file.read(8)


        # XLSX files are ZIP containers and start with PK.
        if header.startswith(
            b"PK"
        ):

            return "Excel"


        return "CSV"


    except Exception:

        return "Unknown"


# ============================================================
# READ DATA FILE
# ============================================================

def read_data_file(file_path):

    # Detect file type.
    file_type = detect_file_type(
        file_path
    )


    # --------------------------------------------------------
    # EXCEL FILE
    # --------------------------------------------------------

    if file_type == "Excel":

        dataframe = pd.read_excel(
            file_path,
            engine="openpyxl"
        )

        return dataframe, "Excel"


    # --------------------------------------------------------
    # NORMAL CSV
    # --------------------------------------------------------

    try:

        dataframe = pd.read_csv(
            file_path,
            low_memory=False
        )

        return dataframe, "CSV"


    except Exception:

        # Some CSV files may contain inconsistent rows.
        dataframe = pd.read_csv(
            file_path,
            engine="python",
            on_bad_lines="warn"
        )

        return dataframe, "CSV"
```

```python
# ============================================================
# HAMPEL OUTLIER DETECTION
# ============================================================

def hampel_outlier_mask(
    series,
    window_size=11,
    threshold=3.5
):
    """
    Detect local statistical outliers using a Hampel filter.

    The method uses:
        Rolling Median
        Median Absolute Deviation
        Robust Z Score
    """

    # Convert values to numeric.
    x = pd.to_numeric(
        series,
        errors="coerce"
    )


    # --------------------------------------------------------
    # Rolling median
    # --------------------------------------------------------

    rolling_median = x.rolling(
        window=window_size,
        center=True,
        min_periods=3
    ).median()


    # --------------------------------------------------------
    # Absolute deviation
    # --------------------------------------------------------

    deviation = (
        x - rolling_median
    ).abs()


    # --------------------------------------------------------
    # Rolling MAD
    # --------------------------------------------------------

    rolling_mad = deviation.rolling(
        window=window_size,
        center=True,
        min_periods=3
    ).median()


    # --------------------------------------------------------
    # Robust standard deviation
    # --------------------------------------------------------

    robust_sigma = (
        1.4826
        *
        rolling_mad
    )


    # Avoid division by zero.
    robust_sigma = robust_sigma.replace(
        0,
        np.nan
    )


    # --------------------------------------------------------
    # Robust Z score
    # --------------------------------------------------------

    robust_z = (
        (x - rolling_median).abs()
        /
        robust_sigma
    )


    # --------------------------------------------------------
    # Detect outliers
    # --------------------------------------------------------

    mask = (
        robust_z > threshold
    )


    # Missing values are not automatically considered
    # outliers.
    mask = mask.fillna(
        False
    )


    return mask
```

```python
# ============================================================
# FIND CONSECUTIVE OUTLIER RUNS
# ============================================================

def find_outlier_runs(mask):

    values = mask.to_numpy(
        dtype=bool
    )

    runs = []

    start = None


    for index, value in enumerate(
        values
    ):

        # Start of an outlier sequence.
        if value and start is None:

            start = index


        # End of an outlier sequence.
        elif (
            not value
            and
            start is not None
        ):

            end = index - 1

            length = (
                end
                -
                start
                +
                1
            )

            runs.append(
                (
                    start,
                    end,
                    length
                )
            )

            start = None


    # Handle a sequence ending at the last row.
    if start is not None:

        end = len(values) - 1

        length = (
            end
            -
            start
            +
            1
        )

        runs.append(
            (
                start,
                end,
                length
            )
        )


    return runs


# ============================================================
# CORRECT SHORT OUTLIER RUNS
# ============================================================

def correct_short_outliers(
    series,
    mask,
    maximum_run=3
):
    """
    Replace only short isolated outlier sequences.

    Long sequences are retained.
    """

    # Convert to numeric.
    x = pd.to_numeric(
        series,
        errors="coerce"
    ).copy()


    # Make a separate cleaned copy.
    cleaned = x.copy()


    corrected = 0

    retained = 0


    # Find all outlier runs.
    runs = find_outlier_runs(
        mask
    )


    for start, end, length in runs:

        # ----------------------------------------------------
        # SHORT RUN
        # ----------------------------------------------------

        if length <= maximum_run:

            temporary = x.copy()


            # Temporarily replace the outlier values with NaN.
            temporary.iloc[
                start:end + 1
            ] = np.nan


            # Interpolate between neighbouring observations.
            temporary = temporary.interpolate(
                method="linear",
                limit_direction="both"
            )


            # Replace detected values.
            for index in range(
                start,
                end + 1
            ):

                if pd.notna(
                    temporary.iloc[index]
                ):

                    cleaned.iloc[index] = (
                        temporary.iloc[index]
                    )

                    corrected += 1

                else:

                    retained += 1


        # ----------------------------------------------------
        # LONG RUN
        # ----------------------------------------------------

        else:

            # Keep long abnormal sequences unchanged.
            retained += length


    return (
        cleaned,
        corrected,
        retained
    )
```

```python
# ============================================================
# PROCESS ONE MISSING FILE
# ============================================================

def process_missing_file(
    file_path
):

    file_path = Path(
        file_path
    )


    # --------------------------------------------------------
    # Determine activity class
    # --------------------------------------------------------

    path_parts = file_path.parts


    if "class_lecture" in path_parts:

        activity_class = "Class Lecture"

    elif "music" in path_parts:

        activity_class = "Music"

    elif "normal" in path_parts:

        activity_class = "Normal"

    elif "reading" in path_parts:

        activity_class = "Reading"

    else:

        activity_class = "Unknown"


    # --------------------------------------------------------
    # Determine sensor
    # --------------------------------------------------------

    folder_name = (
        file_path.parent.name.lower()
    )


    if folder_name.endswith(
        "mw"
    ):

        sensor = "Motion Watch"

    elif folder_name.endswith(
        "ms"
    ):

        sensor = "Muse"

    else:

        sensor = "Unknown"


    # --------------------------------------------------------
    # Create corresponding CLEAN_DATA folder
    # --------------------------------------------------------

    relative_folder = (
        file_path.parent.relative_to(
            BASE_DIR
        )
    )


    output_folder = (
        CLEAN_DIR
        /
        relative_folder
    )


    output_folder.mkdir(
        parents=True,
        exist_ok=True
    )


    # Output filename.
    output_path = (
        output_folder
        /
        file_path.name
    )


    # --------------------------------------------------------
    # RESULT INFORMATION
    # --------------------------------------------------------

    result = {

        "Activity Class":
            activity_class,

        "Sensor":
            sensor,

        "File":
            file_path.name,

        "Source":
            str(file_path),

        "Output":
            str(output_path),

        "Status":
            "Failed",

        "File Type":
            "",

        "Rows Before":
            0,

        "Rows After":
            0,

        "Columns":
            0,

        "Numeric Columns":
            0,

        "Detected Outliers":
            0,

        "Corrected Outliers":
            0,

        "Retained Outliers":
            0,

        "Error":
            ""
    }


    try:

        # ----------------------------------------------------
        # READ FILE
        # ----------------------------------------------------

        df, file_type = read_data_file(
            file_path
        )


        result[
            "File Type"
        ] = file_type


        result[
            "Rows Before"
        ] = len(df)


        result[
            "Columns"
        ] = len(df.columns)


        # ----------------------------------------------------
        # Identify numerical columns
        # ----------------------------------------------------

        numeric_columns = []


        for column in df.columns:

            column_name = (
                str(column)
                .strip()
                .lower()
            )


            # Do not modify time related columns.
            if column_name in [
                "date",
                "time",
                "timestamp",
                "datetime"
            ]:

                continue


            # Try numerical conversion.
            numeric_values = pd.to_numeric(
                df[column],
                errors="coerce"
            )


            # Count valid numerical values.
            valid_count = (
                numeric_values.notna().sum()
            )


            if valid_count > 0:

                numeric_columns.append(
                    column
                )


        result[
            "Numeric Columns"
        ] = len(
            numeric_columns
        )


        # ----------------------------------------------------
        # Process every numerical column
        # ----------------------------------------------------

        for column in numeric_columns:

            numeric_values = pd.to_numeric(
                df[column],
                errors="coerce"
            )


            # Detect outliers.
            outlier_mask = (
                hampel_outlier_mask(
                    numeric_values,
                    WINDOW_SIZE,
                    HAMPEL_THRESHOLD
                )
            )


            detected = int(
                outlier_mask.sum()
            )


            # Correct short outlier runs.
            (
                cleaned_values,
                corrected,
                retained
            ) = correct_short_outliers(
                numeric_values,
                outlier_mask,
                MAX_OUTLIER_RUN
            )


            # Put cleaned data back.
            df[column] = cleaned_values


            # Update counters.
            result[
                "Detected Outliers"
            ] += detected


            result[
                "Corrected Outliers"
            ] += corrected


            result[
                "Retained Outliers"
            ] += retained


        # ----------------------------------------------------
        # Save as CSV
        # ----------------------------------------------------

        df.to_csv(
            output_path,
            index=False
        )


        # ----------------------------------------------------
        # VERIFY FILE
        # ----------------------------------------------------

        if not output_path.exists():

            raise IOError(
                "Output file was not created."
            )


        if output_path.stat().st_size == 0:

            raise IOError(
                "Output file is empty."
            )


        # ----------------------------------------------------
        # Verify row count
        # ----------------------------------------------------

        result[
            "Rows After"
        ] = len(df)


        if (
            result["Rows Before"]
            !=
            result["Rows After"]
        ):

            raise ValueError(
                "Row count changed."
            )


        # ----------------------------------------------------
        # SUCCESS
        # ----------------------------------------------------

        result[
            "Status"
        ] = "Completed"


    except Exception as error:

        result[
            "Error"
        ] = str(error)


    return result
```

```python
# ============================================================
# PROCESS ONLY MISSING FILES
# ============================================================

print("=" * 80)
print("STARTING RECOVERY PROCESS")
print("=" * 80)

print(
    "Files to process:",
    len(missing_files)
)

print(
    "CPU workers:",
    MAX_WORKERS
)

print("=" * 80)


recovery_results = []


# ------------------------------------------------------------
# Nothing to process
# ------------------------------------------------------------

if len(missing_files) == 0:

    print(
        "No missing files were found."
    )


else:

    # --------------------------------------------------------
    # Use threads for file based processing.
    # --------------------------------------------------------

    with ThreadPoolExecutor(
        max_workers=MAX_WORKERS
    ) as executor:


        futures = {

            executor.submit(
                process_missing_file,
                file_path
            ): file_path

            for file_path in missing_files

        }


        completed = 0


        # ----------------------------------------------------
        # Collect results
        # ----------------------------------------------------

        for future in as_completed(
            futures
        ):

            file_path = futures[
                future
            ]


            try:

                result = future.result()


            except Exception as error:

                result = {

                    "Activity Class":
                        "Unknown",

                    "Sensor":
                        "Unknown",

                    "File":
                        file_path.name,

                    "Source":
                        str(file_path),

                    "Output":
                        "",

                    "Status":
                        "Failed",

                    "File Type":
                        "",

                    "Rows Before":
                        0,

                    "Rows After":
                        0,

                    "Columns":
                        0,

                    "Numeric Columns":
                        0,

                    "Detected Outliers":
                        0,

                    "Corrected Outliers":
                        0,

                    "Retained Outliers":
                        0,

                    "Error":
                        str(error)
                }


            recovery_results.append(
                result
            )


            completed += 1


            # ------------------------------------------------
            # Display progress
            # ------------------------------------------------

            print(
                f"[{completed}/{len(missing_files)}] "
                f"{result['File']} | "
                f"{result['Status']} | "
                f"Detected: "
                f"{result['Detected Outliers']:,} | "
                f"Corrected: "
                f"{result['Corrected Outliers']:,}"
            )


print()
print("=" * 80)
print("RECOVERY PROCESS FINISHED")
print("=" * 80)
```

```python
# ============================================================
# RECOVER ALL MISSING FILES INTO CLEAN_DATA
# ============================================================
#
# This code does NOT perform outlier removal.
#
# Its only purpose is to make sure that all 480 original CSV
# files exist inside CLEAN_DATA with the same folder structure.
#
# Existing cleaned files are NOT overwritten.
#
# ============================================================


# ------------------------------------------------------------
# Import required libraries
# ------------------------------------------------------------

from pathlib import Path

import shutil


# ============================================================
# DEFINE MAIN FOLDERS
# ============================================================

# Main RAW_DATA directory.
BASE_DIR = (
    Path.home()
    / "Desktop"
    / "RAW_DATA"
)


# Destination for cleaned data.
CLEAN_DIR = (
    BASE_DIR
    / "CLEAN_DATA"
)


# ============================================================
# DATASET FOLDERS
# ============================================================

DATASET_FOLDERS = [

    "class_lecture/clms",
    "class_lecture/clmw",

    "music/mums",
    "music/mumw",

    "normal/norms",
    "normal/normw",

    "reading/rdms",
    "reading/rdmw"

]


# ============================================================
# CREATE CLEAN_DATA
# ============================================================

CLEAN_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# COUNTERS
# ============================================================

original_count = 0

existing_count = 0

copied_count = 0

failed_count = 0

failed_files = []


# ============================================================
# START RECOVERY
# ============================================================

print("=" * 80)

print(
    "STARTING MISSING FILE RECOVERY"
)

print("=" * 80)


# ------------------------------------------------------------
# Process every dataset folder
# ------------------------------------------------------------

for folder_name in DATASET_FOLDERS:

    # Source folder.
    source_folder = (
        BASE_DIR
        / folder_name
    )


    # Destination folder.
    destination_folder = (
        CLEAN_DIR
        / folder_name
    )


    # --------------------------------------------------------
    # Check source folder
    # --------------------------------------------------------

    if not source_folder.exists():

        print(
            "SOURCE FOLDER NOT FOUND:",
            source_folder
        )

        continue


    # --------------------------------------------------------
    # Create destination folder
    # --------------------------------------------------------

    destination_folder.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # Find all CSV files
    # --------------------------------------------------------

    csv_files = sorted(
        source_folder.glob(
            "*.csv"
        )
    )


    print()
    print(
        "Processing:",
        folder_name
    )

    print(
        "Original files:",
        len(csv_files)
    )


    # --------------------------------------------------------
    # Process files
    # --------------------------------------------------------

    for source_file in csv_files:

        original_count += 1


        # Destination path.
        destination_file = (
            destination_folder
            /
            source_file.name
        )


        # ----------------------------------------------------
        # If file already exists, leave it unchanged.
        # ----------------------------------------------------

        if destination_file.exists():

            existing_count += 1

            continue


        # ----------------------------------------------------
        # Copy missing file.
        # ----------------------------------------------------

        try:

            shutil.copy2(
                source_file,
                destination_file
            )


            copied_count += 1


            print(
                "Copied:",
                source_file.name
            )


        except Exception as error:

            failed_count += 1


            failed_files.append(
                (
                    folder_name,
                    source_file.name,
                    str(error)
                )
            )


            print(
                "FAILED:",
                source_file.name,
                "|",
                error
            )


# ============================================================
# FINAL RECOVERY SUMMARY
# ============================================================

print()
print("=" * 80)

print(
    "FILE RECOVERY COMPLETED"
)

print("=" * 80)

print(
    "Original files found:",
    original_count
)

print(
    "Already existing:",
    existing_count
)

print(
    "Newly copied:",
    copied_count
)

print(
    "Failed:",
    failed_count
)

print("=" * 80)


# ============================================================
# DISPLAY FAILED FILES
# ============================================================

if failed_files:

    print()
    print(
        "FAILED FILES:"
    )

    for number, item in enumerate(
        failed_files,
        start=1
    ):

        folder_name, file_name, error = item

        print(
            f"{number}. "
            f"{folder_name}/{file_name} | "
            f"{error}"
        )


# ============================================================
# FINAL COUNT INSIDE CLEAN_DATA
# ============================================================

clean_count = 0


for folder_name in DATASET_FOLDERS:

    folder = (
        CLEAN_DIR
        / folder_name
    )


    if folder.exists():

        clean_count += len(
            list(
                folder.glob("*.csv")
            )
        )


print()
print("=" * 80)

print(
    "FINAL CLEAN_DATA FILE COUNT:",
    clean_count
)

print("=" * 80)


# ------------------------------------------------------------
# Check whether all 480 files are present
# ------------------------------------------------------------

if (
    original_count == 480
    and
    clean_count == 480
):

    print(
        "SUCCESS: ALL 480 FILES ARE PRESENT IN CLEAN_DATA."
    )

else:

    print(
        "WARNING: CLEAN_DATA DOES NOT YET CONTAIN ALL 480 FILES."
    )
```

```python
# ============================================================
# ROBUST OUTLIER CLEANING
# EXACTLY FOR THE 480 DATASET FILES
# ============================================================

import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ============================================================
# 1. DATASET PATHS
# ============================================================

# IMPORTANT:
# This must point to your actual RAW_DATA directory.

RAW_DATA = Path("/home/rakib/Desktop/RAW_DATA")

CLEAN_DATA = RAW_DATA / "CLEAN_DATA"


# ============================================================
# 2. EXPECTED DATASET STRUCTURE
# ============================================================

EXPECTED_FOLDERS = [
    "class_lecture/clms",
    "class_lecture/clmw",
    "music/mums",
    "music/mumw",
    "normal/norms",
    "normal/normw",
    "reading/rdms",
    "reading/rdmw"
]


# ============================================================
# 3. CLEANING PARAMETERS
# ============================================================

ROLLING_WINDOW = 11

MIN_PERIODS = 5

MAD_THRESHOLD = 6.0

MAX_CONSECUTIVE_OUTLIERS = 3


# ============================================================
# 4. CREATE CLEAN DATA DIRECTORY
# ============================================================

CLEAN_DATA.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 5. IDENTIFY TIME RELATED COLUMNS
# ============================================================

def is_time_column(column_name):

    name = str(column_name).strip().lower()

    time_columns = {
        "date",
        "time",
        "timestamp",
        "datetime"
    }

    return name in time_columns


# ============================================================
# 6. CHECK WHETHER A COLUMN IS NUMERIC
# ============================================================

def is_numeric_signal(df, column):

    if is_time_column(column):
        return False

    numeric_values = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    valid_values = numeric_values.notna().sum()

    return valid_values > 0


# ============================================================
# 7. ROBUST OUTLIER DETECTION
# ============================================================

def detect_and_correct_outliers(series):

    original_series = series.copy()

    numeric_series = pd.to_numeric(
        series,
        errors="coerce"
    )

    valid_count = numeric_series.notna().sum()

    if valid_count < MIN_PERIODS:

        return (
            original_series,
            0,
            0,
            0
        )

    # --------------------------------------------------------
    # Rolling median
    # --------------------------------------------------------

    rolling_median = (
        numeric_series
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=MIN_PERIODS
        )
        .median()
    )


    # --------------------------------------------------------
    # Absolute deviation
    # --------------------------------------------------------

    absolute_deviation = (
        numeric_series -
        rolling_median
    ).abs()


    # --------------------------------------------------------
    # Rolling MAD
    # --------------------------------------------------------

    rolling_mad = (
        absolute_deviation
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=MIN_PERIODS
        )
        .median()
    )


    # --------------------------------------------------------
    # Robust scale
    # --------------------------------------------------------

    robust_scale = (
        1.4826 *
        rolling_mad
    )


    # --------------------------------------------------------
    # Initial outlier mask
    # --------------------------------------------------------

    outlier_mask = pd.Series(
        False,
        index=numeric_series.index
    )


    # --------------------------------------------------------
    # Normal MAD case
    # --------------------------------------------------------

    valid_scale = (
        robust_scale > 0
    )

    outlier_mask.loc[valid_scale] = (
        absolute_deviation.loc[valid_scale]
        >
        MAD_THRESHOLD *
        robust_scale.loc[valid_scale]
    )


    # --------------------------------------------------------
    # Zero MAD case
    # --------------------------------------------------------

    zero_scale = (
        robust_scale == 0
    ) & (
        rolling_median.notna()
    )

    outlier_mask.loc[zero_scale] = (
        numeric_series.loc[zero_scale]
        !=
        rolling_median.loc[zero_scale]
    )


    # --------------------------------------------------------
    # Missing values are not outliers
    # --------------------------------------------------------

    outlier_mask.loc[
        numeric_series.isna()
    ] = False


    # --------------------------------------------------------
    # Identify consecutive outlier sequences
    # --------------------------------------------------------

    sequence_group = (
        outlier_mask
        .ne(outlier_mask.shift())
        .cumsum()
    )

    sequence_length = (
        outlier_mask
        .groupby(sequence_group)
        .transform("sum")
    )


    # --------------------------------------------------------
    # Long sequences are retained
    # --------------------------------------------------------

    long_sequence_mask = (
        outlier_mask
        &
        (
            sequence_length >
            MAX_CONSECUTIVE_OUTLIERS
        )
    )


    retained_long_sequences = int(
        long_sequence_mask.sum()
    )


    # --------------------------------------------------------
    # Correct isolated and short outliers
    # --------------------------------------------------------

    correction_mask = (
        outlier_mask
        &
        (~long_sequence_mask)
        &
        rolling_median.notna()
    )


    # --------------------------------------------------------
    # Create cleaned series
    # --------------------------------------------------------

    cleaned_series = original_series.copy()

    cleaned_series.loc[
        correction_mask
    ] = (
        rolling_median.loc[
            correction_mask
        ]
    )


    # --------------------------------------------------------
    # Statistics
    # --------------------------------------------------------

    detected = int(
        outlier_mask.sum()
    )

    corrected = int(
        correction_mask.sum()
    )


    return (
        cleaned_series,
        detected,
        corrected,
        retained_long_sequences
    )


# ============================================================
# 8. READ CSV SAFELY
# ============================================================

def read_csv_safely(file_path):

    try:

        # Normal CSV reading
        return pd.read_csv(
            file_path,
            low_memory=False
        )

    except Exception:

        # More tolerant parser
        return pd.read_csv(
            file_path,
            engine="python"
        )


# ============================================================
# 9. FIND ONLY THE EXPECTED 480 FILES
# ============================================================

print()
print("=" * 75)
print("DATASET DISCOVERY")
print("=" * 75)

all_csv_files = []


for folder in EXPECTED_FOLDERS:

    folder_path = RAW_DATA / folder

    if not folder_path.exists():

        print(
            f"WARNING: Folder does not exist: "
            f"{folder}"
        )

        continue


    folder_files = sorted(
        folder_path.glob("*.csv")
    )


    print(
        f"{folder:<25} "
        f"{len(folder_files):>3} CSV files"
    )


    all_csv_files.extend(
        folder_files
    )


print("=" * 75)

print(
    f"Total intended CSV files: "
    f"{len(all_csv_files)}"
)


# ============================================================
# 10. VERIFY EXPECTED FILE COUNT
# ============================================================

if len(all_csv_files) != 480:

    print()
    print(
        "WARNING: The expected dataset contains "
        "480 files, but the folders currently contain "
        f"{len(all_csv_files)} files."
    )

    print(
        "The program will process the files that actually "
        "exist in the eight specified dataset folders."
    )

else:

    print(
        "SUCCESS: Exactly 480 dataset files were found."
    )


# ============================================================
# 11. REPORT STORAGE
# ============================================================

file_report = []

channel_report = []

failed_files = []


total_rows_before = 0

total_rows_after = 0

total_detected = 0

total_corrected = 0

total_retained = 0


# ============================================================
# 12. PROCESS ALL DATASET FILES
# ============================================================

print()
print("=" * 75)
print("STARTING ROBUST OUTLIER CLEANING")
print("=" * 75)


for file_number, source_path in enumerate(
    all_csv_files,
    start=1
):

    relative_path = source_path.relative_to(
        RAW_DATA
    )

    destination_path = (
        CLEAN_DATA /
        relative_path
    )


    destination_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    print()
    print(
        f"[{file_number}/{len(all_csv_files)}] "
        f"{relative_path}"
    )


    try:

        # ----------------------------------------------------
        # Read source file
        # ----------------------------------------------------

        df = read_csv_safely(
            source_path
        )


        original_rows = len(df)

        total_rows_before += original_rows


        # ----------------------------------------------------
        # Identify numeric signal columns
        # ----------------------------------------------------

        signal_columns = []

        for column in df.columns:

            if is_numeric_signal(
                df,
                column
            ):

                signal_columns.append(
                    column
                )


        file_detected = 0

        file_corrected = 0

        file_retained = 0


        # ----------------------------------------------------
        # Process each numeric signal
        # ----------------------------------------------------

        for column in signal_columns:

            try:

                result = (
                    detect_and_correct_outliers(
                        df[column]
                    )
                )

                cleaned_series = result[0]

                detected = result[1]

                corrected = result[2]

                retained = result[3]


                # Replace signal
                df[column] = cleaned_series


                # Update file statistics
                file_detected += detected

                file_corrected += corrected

                file_retained += retained


                # Channel report
                channel_report.append(
                    {
                        "File": str(
                            relative_path
                        ),
                        "Column": str(
                            column
                        ),
                        "Rows": original_rows,
                        "Outliers Detected":
                            detected,
                        "Outliers Corrected":
                            corrected,
                        "Long Sequence Outliers Retained":
                            retained
                    }
                )


            except Exception as column_error:

                print(
                    f"   Column error: "
                    f"{column}"
                )

                print(
                    f"   {column_error}"
                )


        # ----------------------------------------------------
        # Save cleaned file
        # ----------------------------------------------------

        df.to_csv(
            destination_path,
            index=False
        )


        # ----------------------------------------------------
        # Verify saved file
        # ----------------------------------------------------

        check_df = pd.read_csv(
            destination_path,
            low_memory=False
        )


        cleaned_rows = len(
            check_df
        )


        total_rows_after += cleaned_rows


        row_count_preserved = (
            original_rows ==
            cleaned_rows
        )


        # ----------------------------------------------------
        # Update global statistics
        # ----------------------------------------------------

        total_detected += file_detected

        total_corrected += file_corrected

        total_retained += file_retained


        # ----------------------------------------------------
        # File report
        # ----------------------------------------------------

        file_report.append(
            {
                "File": str(
                    relative_path
                ),
                "Original Rows":
                    original_rows,
                "Cleaned Rows":
                    cleaned_rows,
                "Row Count Preserved":
                    row_count_preserved,
                "Signal Columns":
                    len(signal_columns),
                "Outliers Detected":
                    file_detected,
                "Outliers Corrected":
                    file_corrected,
                "Long Sequence Outliers Retained":
                    file_retained,
                "Status":
                    "Success"
            }
        )


        print(
            f"   Rows: "
            f"{original_rows:,} "
            f"to "
            f"{cleaned_rows:,}"
        )

        print(
            f"   Signal columns: "
            f"{len(signal_columns)}"
        )

        print(
            f"   Detected: "
            f"{file_detected:,}"
        )

        print(
            f"   Corrected: "
            f"{file_corrected:,}"
        )

        print(
            f"   Retained long sequences: "
            f"{file_retained:,}"
        )


    except Exception as error:

        print(
            f"   FAILED: {error}"
        )


        failed_files.append(
            str(relative_path)
        )


        file_report.append(
            {
                "File": str(
                    relative_path
                ),
                "Original Rows":
                    np.nan,
                "Cleaned Rows":
                    np.nan,
                "Row Count Preserved":
                    False,
                "Signal Columns":
                    0,
                "Outliers Detected":
                    0,
                "Outliers Corrected":
                    0,
                "Long Sequence Outliers Retained":
                    0,
                "Status":
                    f"Failed: {error}"
            }
        )


# ============================================================
# 13. CREATE REPORT DATAFRAMES
# ============================================================

file_report_df = pd.DataFrame(
    file_report
)

channel_report_df = pd.DataFrame(
    channel_report
)


# ============================================================
# 14. CREATE REPORT DIRECTORY
# ============================================================

REPORT_DIR = (
    CLEAN_DATA /
    "cleaning_reports"
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 15. SAVE REPORTS
# ============================================================

file_report_path = (
    REPORT_DIR /
    "file_cleaning_report.csv"
)

channel_report_path = (
    REPORT_DIR /
    "channel_cleaning_report.csv"
)


file_report_df.to_csv(
    file_report_path,
    index=False
)


channel_report_df.to_csv(
    channel_report_path,
    index=False
)


# ============================================================
# 16. VERIFY CLEAN DATASET
# ============================================================

print()
print("=" * 75)
print("VERIFYING CLEAN_DATA")
print("=" * 75)


cleaned_files = []


for folder in EXPECTED_FOLDERS:

    folder_path = (
        CLEAN_DATA /
        folder
    )


    if not folder_path.exists():

        continue


    cleaned_files.extend(
        folder_path.glob("*.csv")
    )


print(
    f"Original dataset files: "
    f"{len(all_csv_files)}"
)

print(
    f"Cleaned dataset files: "
    f"{len(cleaned_files)}"
)


# ============================================================
# 17. COMPARE ORIGINAL AND CLEANED FILES
# ============================================================

original_relative_paths = {
    str(
        path.relative_to(
            RAW_DATA
        )
    )
    for path in all_csv_files
}


cleaned_relative_paths = {
    str(
        path.relative_to(
            CLEAN_DATA
        )
    )
    for path in cleaned_files
}


missing_files = (
    original_relative_paths
    -
    cleaned_relative_paths
)


extra_files = (
    cleaned_relative_paths
    -
    original_relative_paths
)


# ============================================================
# 18. FINAL STATISTICS
# ============================================================

print()
print("=" * 75)
print("FINAL CLEANING STATISTICS")
print("=" * 75)

print(
    f"Files discovered: "
    f"{len(all_csv_files)}"
)

print(
    f"Files successfully saved: "
    f"{len(cleaned_files)}"
)

print(
    f"Files failed: "
    f"{len(failed_files)}"
)

print(
    f"Total rows before cleaning: "
    f"{total_rows_before:,}"
)

print(
    f"Total rows after cleaning: "
    f"{total_rows_after:,}"
)

print(
    f"Total outliers detected: "
    f"{total_detected:,}"
)

print(
    f"Total outliers corrected: "
    f"{total_corrected:,}"
)

print(
    f"Long sequence outliers retained: "
    f"{total_retained:,}"
)


# ============================================================
# 19. FILE VERIFICATION
# ============================================================

print()
print("=" * 75)
print("FILE VERIFICATION")
print("=" * 75)

print(
    f"Missing files: "
    f"{len(missing_files)}"
)

print(
    f"Extra files: "
    f"{len(extra_files)}"
)


if missing_files:

    print()
    print("MISSING FILES:")

    for file_name in sorted(
        missing_files
    ):

        print(
            f"   {file_name}"
        )


if extra_files:

    print()
    print("EXTRA FILES:")

    for file_name in sorted(
        extra_files
    ):

        print(
            f"   {file_name}"
        )


# ============================================================
# 20. ROW COUNT VERIFICATION
# ============================================================

row_count_errors = file_report_df[
    file_report_df[
        "Row Count Preserved"
    ] == False
]


print()
print("=" * 75)
print("ROW COUNT VERIFICATION")
print("=" * 75)

print(
    f"Files checked: "
    f"{len(file_report_df)}"
)

print(
    f"Files with preserved row count: "
    f"{len(file_report_df) - len(row_count_errors)}"
)

print(
    f"Files with row count mismatch: "
    f"{len(row_count_errors)}"
)


# ============================================================
# 21. FINAL STATUS
# ============================================================

print()
print("=" * 75)
print("FINAL DATASET STATUS")
print("=" * 75)


if (
    len(all_csv_files) == 480
    and len(cleaned_files) == 480
    and len(missing_files) == 0
    and len(extra_files) == 0
    and len(failed_files) == 0
    and len(row_count_errors) == 0
):

    print(
        "SUCCESS"
    )

    print(
        "All 480 dataset files were processed."
    )

    print(
        "All 480 cleaned files are present."
    )

    print(
        "All original row counts were preserved."
    )

else:

    print(
        "WARNING"
    )

    print(
        "The dataset requires further verification."
    )


# ============================================================
# 22. REPORT LOCATIONS
# ============================================================

print()
print("=" * 75)
print("REPORT LOCATIONS")
print("=" * 75)

print(
    f"File report:\n"
    f"{file_report_path}"
)

print()

print(
    f"Channel report:\n"
    f"{channel_report_path}"
)


print()
print("=" * 75)
print("OUTLIER CLEANING COMPLETED")
print("=" * 75)
```

```python
# ============================================================
# EEG DATASET VISUALIZATION
# CORRECTED VERSION FOR YOUR VS CODE NOTEBOOK
# ============================================================

# ------------------------------------------------------------
# Import required libraries
# ------------------------------------------------------------

import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")


# ============================================================
# 1. AUTOMATICALLY DETECT THE RAW_DATA DIRECTORY
# ============================================================

# Your notebook is currently running inside RAW_DATA.
# Therefore, os.getcwd() should point to:
#
# /home/rakib/Desktop/RAW_DATA
#
BASE_DIR = os.getcwd()

print("=" * 75)
print("CURRENT WORKING DIRECTORY")
print("=" * 75)

print(BASE_DIR)


# ------------------------------------------------------------
# Check whether the current directory is already RAW_DATA
# ------------------------------------------------------------

if os.path.basename(
    os.path.normpath(BASE_DIR)
) == "RAW_DATA":

    RAW_DATA_DIR = BASE_DIR

else:

    RAW_DATA_DIR = os.path.join(
        BASE_DIR,
        "RAW_DATA"
    )


# ------------------------------------------------------------
# Define CLEAN_DATA
# ------------------------------------------------------------

CLEAN_DATA_DIR = os.path.join(
    RAW_DATA_DIR,
    "CLEAN_DATA"
)


# ------------------------------------------------------------
# Define visualization directory
# ------------------------------------------------------------

VISUALIZATION_DIR = os.path.join(
    CLEAN_DATA_DIR,
    "visualizations"
)


# Create visualization directory
os.makedirs(
    VISUALIZATION_DIR,
    exist_ok=True
)


print()
print("RAW_DATA:")
print(RAW_DATA_DIR)

print()
print("CLEAN_DATA:")
print(CLEAN_DATA_DIR)

print()
print("VISUALIZATION:")
print(VISUALIZATION_DIR)


# ============================================================
# 2. CORRECT DATASET FOLDERS
# ============================================================

DATASET_FOLDERS = [

    "class_lecture/clms",
    "class_lecture/clmw",

    "music/mums",
    "music/mumw",

    "normal/norms",
    "normal/normw",

    "reading/rdms",
    "reading/rdmw"

]


# ============================================================
# 3. CHECK ALL DATASET FOLDERS
# ============================================================

print()
print("=" * 75)
print("DATASET FOLDER VERIFICATION")
print("=" * 75)


folder_information = []

all_files = []


for folder in DATASET_FOLDERS:

    # Build absolute folder path
    folder_path = (
        RAW_DATA_DIR /
        pd.io.common.stringify_path(folder)
        if False
        else os.path.join(
            RAW_DATA_DIR,
            folder
        )
    )


    # Check folder existence
    if not os.path.exists(folder_path):

        print(
            "NOT FOUND:",
            folder_path
        )

        continue


    # Find CSV files
    csv_files = sorted(
        [
            file
            for file in os.listdir(
                folder_path
            )
            if file.lower().endswith(".csv")
        ]
    )


    print(
        f"{folder:<25} "
        f"{len(csv_files):>3} files"
    )


    # Store folder information
    folder_information.append(
        {
            "Folder": folder,
            "Files": len(csv_files)
        }
    )


    # Store every file
    for file_name in csv_files:

        all_files.append(
            {
                "Activity":
                    folder.split("/")[0],

                "Sensor":
                    folder.split("/")[1],

                "Folder":
                    folder,

                "File":
                    file_name,

                "Path":
                    os.path.join(
                        folder_path,
                        file_name
                    )
            }
        )


print("=" * 75)

print(
    "Total CSV files found:",
    len(all_files)
)


# ============================================================
# 4. STOP IF NO FILES WERE FOUND
# ============================================================

if len(all_files) == 0:

    raise RuntimeError(
        "No dataset files were found. "
        "Check the RAW_DATA path."
    )


# ============================================================
# 5. CONVERT FILE LIST TO DATAFRAME
# ============================================================

file_list_df = pd.DataFrame(
    all_files
)


# ============================================================
# 6. FILE COUNT BY ACTIVITY
# ============================================================

print()
print("=" * 75)
print("FILES BY ACTIVITY")
print("=" * 75)

print(
    file_list_df[
        "Activity"
    ].value_counts()
)


# ============================================================
# 7. FILE COUNT BY SENSOR
# ============================================================

print()
print("=" * 75)
print("FILES BY SENSOR TYPE")
print("=" * 75)

print(
    file_list_df[
        "Sensor"
    ].value_counts()
)


# ============================================================
# 8. GRAPH: FILES PER DATASET FOLDER
# ============================================================

folder_counts = (
    file_list_df[
        "Folder"
    ]
    .value_counts()
    .sort_index()
)


plt.figure(
    figsize=(12, 6)
)

plt.bar(
    folder_counts.index,
    folder_counts.values
)

plt.title(
    "Number of CSV Files per Dataset Folder"
)

plt.xlabel(
    "Dataset Folder"
)

plt.ylabel(
    "Number of Files"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        VISUALIZATION_DIR,
        "01_files_per_folder.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 9. ANALYZE EVERY CSV FILE
# ============================================================

print()
print("=" * 75)
print("ANALYZING CSV FILE STRUCTURE")
print("=" * 75)


dataset_information = []

failed_reads = []


for number, item in enumerate(
    all_files,
    start=1
):

    file_path = item["Path"]

    try:

        # Read CSV
        df = pd.read_csv(
            file_path,
            low_memory=False
        )


        # Number of rows
        rows = len(df)


        # Number of columns
        columns = len(
            df.columns
        )


        # Numeric columns
        numeric_columns = (
            df.select_dtypes(
                include=np.number
            )
            .columns
            .tolist()
        )


        # Missing values
        missing_values = int(
            df.isna()
            .sum()
            .sum()
        )


        # Store information
        dataset_information.append(
            {
                "Activity":
                    item["Activity"],

                "Sensor":
                    item["Sensor"],

                "Folder":
                    item["Folder"],

                "File":
                    item["File"],

                "Rows":
                    rows,

                "Columns":
                    columns,

                "Numeric Columns":
                    len(
                        numeric_columns
                    ),

                "Missing Values":
                    missing_values
            }
        )


    except Exception as error:

        failed_reads.append(
            {
                "File":
                    item["File"],

                "Folder":
                    item["Folder"],

                "Error":
                    str(error)
            }
        )


# Convert to DataFrame
dataset_info_df = pd.DataFrame(
    dataset_information
)


# ============================================================
# 10. DISPLAY DATASET SUMMARY
# ============================================================

print()
print("=" * 75)
print("DATASET STRUCTURE SUMMARY")
print("=" * 75)

print(
    "Files successfully analyzed:",
    len(dataset_info_df)
)

print(
    "Files that could not be read:",
    len(failed_reads)
)

print(
    "Total rows:",
    f"{dataset_info_df['Rows'].sum():,}"
)

print(
    "Minimum rows:",
    dataset_info_df["Rows"].min()
)

print(
    "Maximum rows:",
    dataset_info_df["Rows"].max()
)

print(
    "Median rows:",
    dataset_info_df["Rows"].median()
)

print(
    "Minimum columns:",
    dataset_info_df["Columns"].min()
)

print(
    "Maximum columns:",
    dataset_info_df["Columns"].max()
)

print("=" * 75)


# ============================================================
# 11. GRAPH: ROW COUNT DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.hist(
    dataset_info_df["Rows"],
    bins=30
)

plt.title(
    "Distribution of Rows per CSV File"
)

plt.xlabel(
    "Number of Rows"
)

plt.ylabel(
    "Number of Files"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        VISUALIZATION_DIR,
        "02_row_count_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. GRAPH: ROW COUNT BY ACTIVITY
# ============================================================

# ------------------------------------------------------------
# Get activity names in sorted order
# ------------------------------------------------------------

activities = sorted(
    dataset_info_df["Activity"]
    .dropna()
    .unique()
)


# ------------------------------------------------------------
# Create row count data for each activity
# ------------------------------------------------------------

row_data = []

valid_activities = []

for activity in activities:

    values = (
        dataset_info_df[
            dataset_info_df["Activity"] == activity
        ]["Rows"]
        .dropna()
        .to_numpy()
    )

    # Only include activities that contain data
    if len(values) > 0:

        row_data.append(values)

        valid_activities.append(activity)


# ------------------------------------------------------------
# Create boxplot
# ------------------------------------------------------------

if len(row_data) > 0:

    plt.figure(
        figsize=(12, 7)
    )

    plt.boxplot(
        row_data,
        tick_labels=valid_activities
    )

    plt.title(
        "Row Count Distribution by Activity"
    )

    plt.xlabel(
        "Activity"
    )

    plt.ylabel(
        "Number of Rows"
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            VISUALIZATION_DIR,
            "03_rows_by_activity.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

else:

    print(
        "No valid row count data available "
        "for the activity boxplot."
    )

# ============================================================
# 13. GRAPH: COLUMN COUNT DISTRIBUTION
# ============================================================

column_counts = (
    dataset_info_df[
        "Columns"
    ]
    .value_counts()
    .sort_index()
)


plt.figure(
    figsize=(10, 6)
)

plt.bar(
    column_counts.index.astype(str),
    column_counts.values
)

plt.title(
    "Number of Columns per CSV File"
)

plt.xlabel(
    "Number of Columns"
)

plt.ylabel(
    "Number of Files"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        VISUALIZATION_DIR,
        "04_column_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 14. GRAPH: MISSING VALUES
# ============================================================

missing_by_activity = (
    dataset_info_df
    .groupby(
        "Activity"
    )["Missing Values"]
    .sum()
)


plt.figure(
    figsize=(10, 6)
)

plt.bar(
    missing_by_activity.index,
    missing_by_activity.values
)

plt.title(
    "Missing Values by Activity"
)

plt.xlabel(
    "Activity"
)

plt.ylabel(
    "Missing Values"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        VISUALIZATION_DIR,
        "05_missing_values_by_activity.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 15. SELECT A REPRESENTATIVE CLEANED FILE
# ============================================================

# Use the first file from the dataset
selected = file_list_df.iloc[0]


selected_activity = selected["Activity"]

selected_sensor = selected["Sensor"]

selected_folder = selected["Folder"]

selected_file = selected["File"]


# Build corresponding CLEAN_DATA path
selected_clean_path = os.path.join(
    CLEAN_DATA_DIR,
    selected_folder,
    selected_file
)


print()
print("=" * 75)
print("REPRESENTATIVE FILE")
print("=" * 75)

print(
    "Activity:",
    selected_activity
)

print(
    "Sensor:",
    selected_sensor
)

print(
    "Folder:",
    selected_folder
)

print(
    "File:",
    selected_file
)

print(
    "Path:",
    selected_clean_path
)

print("=" * 75)


# ============================================================
# 16. READ REPRESENTATIVE CLEANED FILE
# ============================================================

try:

    selected_df = pd.read_csv(
        selected_clean_path,
        low_memory=False
    )

except Exception as error:

    raise RuntimeError(
        "Could not read representative "
        f"cleaned file: {error}"
    )


# ============================================================
# 17. DISPLAY FILE STRUCTURE
# ============================================================

print()
print("=" * 75)
print("REPRESENTATIVE FILE STRUCTURE")
print("=" * 75)

print(
    "Rows:",
    len(selected_df)
)

print(
    "Columns:",
    len(selected_df.columns)
)

print("\nColumn names:")

for number, column in enumerate(
    selected_df.columns,
    start=1
):

    print(
        f"{number}. {column}"
    )


print("\nFirst five rows:")

display(
    selected_df.head()
)


# ============================================================
# 18. IDENTIFY NUMERIC SIGNAL COLUMNS
# ============================================================

numeric_columns = (
    selected_df
    .select_dtypes(
        include=np.number
    )
    .columns
    .tolist()
)


print()
print(
    "Numeric signal columns:"
)

for column in numeric_columns:

    print(
        " ",
        column
    )


# ============================================================
# 19. SELECT FIRST SIGNAL
# ============================================================

if len(numeric_columns) == 0:

    print(
        "No numeric signal columns found."
    )

else:

    signal_column = numeric_columns[0]

    signal = pd.to_numeric(
        selected_df[
            signal_column
        ],
        errors="coerce"
    )


    # ========================================================
    # 20. SIGNAL DISTRIBUTION
    # ========================================================

    plt.figure(
        figsize=(12, 6)
    )

    plt.hist(
        signal.dropna(),
        bins=60
    )

    plt.title(
        f"Signal Distribution: {signal_column}"
    )

    plt.xlabel(
        signal_column
    )

    plt.ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            VISUALIZATION_DIR,
            "06_signal_distribution.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


    # ========================================================
    # 21. BOX PLOT
    # ========================================================

    plt.figure(
        figsize=(10, 6)
    )

    plt.boxplot(
        signal.dropna(),
        vert=True
    )

    plt.title(
        f"Boxplot: {signal_column}"
    )

    plt.ylabel(
        signal_column
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            VISUALIZATION_DIR,
            "07_signal_boxplot.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


    # ========================================================
    # 22. TIME SERIES
    # ========================================================

    number_of_points = min(
        5000,
        len(signal)
    )


    x = np.arange(
        number_of_points
    )


    y = signal.iloc[
        :number_of_points
    ]


    plt.figure(
        figsize=(16, 6)
    )

    plt.plot(
        x,
        y,
        linewidth=0.8
    )

    plt.title(
        f"Time Series: {signal_column}"
    )

    plt.xlabel(
        "Sample Number"
    )

    plt.ylabel(
        signal_column
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            VISUALIZATION_DIR,
            "08_time_series.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


    # ========================================================
    # 23. ROBUST OUTLIER DETECTION FOR VISUALIZATION
    # ========================================================

    median_value = signal.median()


    mad_value = np.median(
        np.abs(
            signal.dropna()
            -
            median_value
        )
    )


    if mad_value == 0:

        outlier_mask = pd.Series(
            False,
            index=signal.index
        )

    else:

        modified_z_score = (
            0.6745
            *
            (
                signal
                -
                median_value
            )
            /
            mad_value
        )


        outlier_mask = (
            modified_z_score.abs()
            >
            3.5
        )


    outlier_count = int(
        outlier_mask.sum()
    )


    print()
    print("=" * 75)
    print("OUTLIER VISUALIZATION")
    print("=" * 75)

    print(
        "Signal:",
        signal_column
    )

    print(
        "Total observations:",
        len(signal)
    )

    print(
        "Potential outliers:",
        outlier_count
    )

    print("=" * 75)


    # ========================================================
    # 24. OUTLIER HIGHLIGHTED GRAPH
    # ========================================================

    number_of_points = min(
        5000,
        len(signal)
    )


    x = np.arange(
        number_of_points
    )


    y = signal.iloc[
        :number_of_points
    ]


    mask = (
        outlier_mask
        .iloc[
            :number_of_points
        ]
        .fillna(False)
        .astype(bool)
        .to_numpy()
    )


    plt.figure(
        figsize=(16, 7)
    )


    # Complete signal
    plt.plot(
        x,
        y,
        linewidth=0.8,
        label="Signal"
    )


    # Potential outliers
    plt.scatter(
        x[mask],
        y.iloc[mask],
        s=25,
        marker="o",
        label="Potential Outlier"
    )


    plt.title(
        f"Outlier Highlighted Signal: {signal_column}"
    )

    plt.xlabel(
        "Sample Number"
    )

    plt.ylabel(
        signal_column
    )

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            VISUALIZATION_DIR,
            "09_outlier_highlighted_signal.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 25. SAVE DATASET INFORMATION
# ============================================================

dataset_info_path = os.path.join(
    VISUALIZATION_DIR,
    "file_level_dataset_information.csv"
)


dataset_info_df.to_csv(
    dataset_info_path,
    index=False
)


# ============================================================
# 26. SAVE FAILED FILE REPORT
# ============================================================

if failed_reads:

    failed_df = pd.DataFrame(
        failed_reads
    )

    failed_df.to_csv(
        os.path.join(
            VISUALIZATION_DIR,
            "visualization_failed_files.csv"
        ),
        index=False
    )


# ============================================================
# 27. FINAL MESSAGE
# ============================================================

print()
print("=" * 75)
print("VISUALIZATION COMPLETED")
print("=" * 75)

print(
    "Visualization directory:"
)

print(
    VISUALIZATION_DIR
)

print()
print(
    "The graphs have been saved inside:"
)

print(
    "CLEAN_DATA/visualizations/"
)

print("=" * 75)
```

```python
# ============================================================
# ROBUST DATASET CLEANING PIPELINE
# 480 CSV FILES
#
# Activities:
#   class_lecture
#   music
#   normal
#   reading
#
# Dataset folders:
#   class_lecture/clms
#   class_lecture/clmw
#   music/mums
#   music/mumw
#   normal/norms
#   normal/normw
#   reading/rdms
#   reading/rdmw
#
# Main principles:
#   1. Process ONLY the eight dataset folders
#   2. Preserve timestamps
#   3. Preserve row count
#   4. Preserve genuine long sequences
#   5. Correct only isolated suspicious observations
#   6. Never treat recording length as a signal outlier
#   7. Never convert zero values to missing values
#   8. Keep a complete quality control report
# ============================================================


from pathlib import Path
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

CURRENT_DIR = Path.cwd()

# If notebook is already inside RAW_DATA
if CURRENT_DIR.name == "RAW_DATA":
    RAW_ROOT = CURRENT_DIR

# If notebook is somewhere above RAW_DATA
elif (CURRENT_DIR / "RAW_DATA").exists():
    RAW_ROOT = CURRENT_DIR / "RAW_DATA"

# Otherwise use current directory
else:
    RAW_ROOT = CURRENT_DIR


CLEAN_ROOT = RAW_ROOT / "CLEAN_DATA"
REPORT_ROOT = CLEAN_ROOT / "reports"


CLEAN_ROOT.mkdir(parents=True, exist_ok=True)
REPORT_ROOT.mkdir(parents=True, exist_ok=True)


# ============================================================
# 2. EXACT DATASET STRUCTURE
# ============================================================

DATASET_FOLDERS = [
    ("class_lecture", "clms"),
    ("class_lecture", "clmw"),

    ("music", "mums"),
    ("music", "mumw"),

    ("normal", "norms"),
    ("normal", "normw"),

    ("reading", "rdms"),
    ("reading", "rdmw"),
]


# ============================================================
# 3. CLEANING PARAMETERS
# ============================================================

# Robust global threshold
ROBUST_Z_THRESHOLD = 6.0

# Local deviation threshold
LOCAL_Z_THRESHOLD = 6.0

# Number of neighboring observations
LOCAL_WINDOW = 11

# Maximum consecutive suspicious points that may be corrected
# Larger sequences are retained
MAX_ISOLATED_SEQUENCE = 2

# Minimum observations required for meaningful analysis
MIN_ROWS = 100

# Constant signal tolerance
CONSTANT_STD_THRESHOLD = 1e-12


# ============================================================
# 4. COLUMN IDENTIFICATION
# ============================================================

TIMESTAMP_KEYWORDS = [
    "timestamp",
    "time_stamp",
    "datetime",
    "date_time",
    "time",
    "date"
]


NON_SIGNAL_KEYWORDS = [
    "activity",
    "label",
    "class",
    "subject",
    "participant",
    "session",
    "event",
    "index"
]


MUSE_SIGNAL_KEYWORDS = [
    "delta",
    "theta",
    "alpha",
    "beta",
    "gamma",
    "raw",
    "hsi",
    "aux"
]


MOTION_SIGNAL_KEYWORDS = [
    "accelerometer",
    "accel",
    "gyro",
    "gyroscope",
    "motion",
    "activity"
]


# ============================================================
# 5. UTILITY FUNCTIONS
# ============================================================

def normalized_name(name):
    """
    Convert column name into a normalized lowercase string.
    """
    return str(name).strip().lower()


def is_timestamp_column(column_name):
    """
    Identify timestamp or time related columns.
    """
    name = normalized_name(column_name)

    for keyword in TIMESTAMP_KEYWORDS:
        if keyword in name:
            return True

    return False


def is_non_signal_column(column_name):
    """
    Identify metadata and label columns that should not
    be processed as physiological or motion signals.
    """
    name = normalized_name(column_name)

    for keyword in NON_SIGNAL_KEYWORDS:
        if keyword in name:
            return True

    return False


def detect_sensor_type(columns):
    """
    Identify whether the file is likely Motion Watch or Muse.
    """

    names = [
        normalized_name(column)
        for column in columns
    ]

    muse_score = 0
    motion_score = 0

    for name in names:

        for keyword in MUSE_SIGNAL_KEYWORDS:
            if keyword in name:
                muse_score += 1
                break

        for keyword in MOTION_SIGNAL_KEYWORDS:
            if keyword in name:
                motion_score += 1
                break

    if muse_score > motion_score:
        return "Muse EEG"

    if motion_score > muse_score:
        return "Motion Watch"

    # Additional structural rule
    if any(
        "delta" in name or
        "theta" in name or
        "alpha" in name or
        "beta" in name or
        "gamma" in name
        for name in names
    ):
        return "Muse EEG"

    if any(
        "gyro" in name or
        "accelerometer" in name or
        "accel" in name
        for name in names
    ):
        return "Motion Watch"

    return "Unknown"


def identify_signal_columns(df, sensor_type):
    """
    Identify numerical signal columns.

    Timestamp, activity, labels and metadata are excluded.
    """

    signal_columns = []

    for column in df.columns:

        if is_timestamp_column(column):
            continue

        if is_non_signal_column(column):
            continue

        if not pd.api.types.is_numeric_dtype(df[column]):
            continue

        name = normalized_name(column)

        # Muse specific filtering
        if sensor_type == "Muse EEG":

            if any(
                keyword in name
                for keyword in MUSE_SIGNAL_KEYWORDS
            ):
                signal_columns.append(column)

        # Motion Watch
        elif sensor_type == "Motion Watch":

            if any(
                keyword in name
                for keyword in MOTION_SIGNAL_KEYWORDS
            ):
                signal_columns.append(column)

        else:

            signal_columns.append(column)

    return signal_columns


# ============================================================
# 6. ROBUST STATISTICS
# ============================================================

def calculate_mad(series):
    """
    Median Absolute Deviation.
    """

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    median = np.nanmedian(values)

    mad = np.nanmedian(
        np.abs(values - median)
    )

    return mad


def robust_z_score(series):
    """
    Calculate robust Z score using median and MAD.

    Scaling factor 0.6745 is commonly used to make the
    MAD based score comparable to a standard deviation score.
    """

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    median = np.nanmedian(values)

    mad = calculate_mad(values)

    if not np.isfinite(mad) or mad <= 0:
        return pd.Series(
            np.zeros(len(series)),
            index=series.index
        )

    score = (
        0.6745 *
        (values - median) /
        mad
    )

    return pd.Series(
        score,
        index=series.index
    )


# ============================================================
# 7. LOCAL MEDIAN DEVIATION
# ============================================================

def local_deviation_score(series):
    """
    Calculate deviation from a centered rolling median.
    """

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    rolling_median = (
        values
        .rolling(
            window=LOCAL_WINDOW,
            center=True,
            min_periods=3
        )
        .median()
    )

    local_difference = np.abs(
        values - rolling_median
    )

    local_mad = (
        local_difference
        .rolling(
            window=LOCAL_WINDOW,
            center=True,
            min_periods=3
        )
        .median()
    )

    local_mad = local_mad.replace(
        0,
        np.nan
    )

    score = (
        0.6745 *
        local_difference /
        local_mad
    )

    return score.fillna(0)


# ============================================================
# 8. IDENTIFY CONSECUTIVE OUTLIER SEQUENCES
# ============================================================

def sequence_length_flags(flag_series):
    """
    Determine whether suspicious observations occur
    individually or as part of a longer sequence.
    """

    flags = np.asarray(
        flag_series,
        dtype=bool
    )

    sequence_lengths = np.zeros(
        len(flags),
        dtype=int
    )

    start = None

    for i, flag in enumerate(flags):

        if flag and start is None:
            start = i

        if not flag and start is not None:

            length = i - start

            sequence_lengths[
                start:i
            ] = length

            start = None

    if start is not None:

        length = len(flags) - start

        sequence_lengths[
            start:
        ] = length

    return sequence_lengths


# ============================================================
# 9. CORRECT ONLY ISOLATED OUTLIERS
# ============================================================

def correct_isolated_outliers(
    series,
    outlier_mask
):
    """
    Correct only isolated suspicious observations.

    Long sequences are retained because they may represent
    genuine physiological or movement events.
    """

    original = pd.to_numeric(
        series,
        errors="coerce"
    ).copy()

    corrected = original.copy()

    sequence_lengths = sequence_length_flags(
        outlier_mask
    )

    isolated_mask = (
        outlier_mask &
        (
            sequence_lengths <=
            MAX_ISOLATED_SEQUENCE
        )
    )

    corrected_count = 0
    retained_long_count = 0

    isolated_indices = np.where(
        isolated_mask
    )[0]

    for position in isolated_indices:

        if position == 0:
            continue

        if position == len(original) - 1:
            continue

        previous_value = original.iloc[
            position - 1
        ]

        next_value = original.iloc[
            position + 1
        ]

        if (
            pd.notna(previous_value)
            and
            pd.notna(next_value)
        ):

            replacement = (
                previous_value +
                next_value
            ) / 2.0

            corrected.iloc[
                position
            ] = replacement

            corrected_count += 1

    retained_long_count = int(
        np.sum(
            outlier_mask &
            (
                sequence_lengths >
                MAX_ISOLATED_SEQUENCE
            )
        )
    )

    return (
        corrected,
        corrected_count,
        retained_long_count
    )


# ============================================================
# 10. MISSING VALUE HANDLING
# ============================================================

def handle_missing_values(series):
    """
    Interpolate missing signal values while preserving
    row count and index.

    Edge missing values use forward and backward filling.
    """

    original_missing = int(
        series.isna().sum()
    )

    if original_missing == 0:
        return series.copy(), 0

    cleaned = (
        pd.to_numeric(
            series,
            errors="coerce"
        )
        .interpolate(
            method="linear",
            limit_direction="both"
        )
    )

    remaining_missing = int(
        cleaned.isna().sum()
    )

    # If an entire column is missing, leave it unchanged.
    if remaining_missing == len(cleaned):
        return series.copy(), 0

    corrected_missing = (
        original_missing -
        remaining_missing
    )

    return cleaned, corrected_missing


# ============================================================
# 11. PROCESS ONE CSV FILE
# ============================================================

def process_file(
    source_file,
    destination_file,
    activity,
    folder_name
):

    result = {
        "activity": activity,
        "folder": folder_name,
        "file": source_file.name,
        "relative_path": str(
            source_file.relative_to(RAW_ROOT)
        ),
        "sensor_type": "Unknown",
        "rows_original": 0,
        "rows_cleaned": 0,
        "columns_original": 0,
        "signal_columns": 0,
        "missing_values_original": 0,
        "missing_values_corrected": 0,
        "constant_channels": 0,
        "outliers_detected": 0,
        "outliers_corrected": 0,
        "long_sequence_outliers": 0,
        "status": "FAILED",
        "error": ""
    }

    try:

        # ----------------------------------------------------
        # READ FILE
        # ----------------------------------------------------

        df = pd.read_csv(
            source_file,
            low_memory=False
        )

        result["rows_original"] = len(df)
        result["columns_original"] = len(df.columns)

        # ----------------------------------------------------
        # SENSOR IDENTIFICATION
        # ----------------------------------------------------

        sensor_type = detect_sensor_type(
            df.columns
        )

        result["sensor_type"] = sensor_type

        # ----------------------------------------------------
        # SIGNAL COLUMN IDENTIFICATION
        # ----------------------------------------------------

        signal_columns = identify_signal_columns(
            df,
            sensor_type
        )

        result["signal_columns"] = len(
            signal_columns
        )

        # ----------------------------------------------------
        # PROCESS SIGNALS
        # ----------------------------------------------------

        for column in signal_columns:

            original_series = pd.to_numeric(
                df[column],
                errors="coerce"
            )

            # -----------------------------------------------
            # MISSING VALUES
            # -----------------------------------------------

            missing_before = int(
                original_series.isna().sum()
            )

            result[
                "missing_values_original"
            ] += missing_before

            cleaned_series, missing_corrected = (
                handle_missing_values(
                    original_series
                )
            )

            result[
                "missing_values_corrected"
            ] += missing_corrected

            # -----------------------------------------------
            # CONSTANT CHANNEL CHECK
            # -----------------------------------------------

            valid_values = (
                cleaned_series
                .dropna()
            )

            if len(valid_values) > 0:

                signal_std = float(
                    valid_values.std()
                )

                if (
                    not np.isfinite(signal_std)
                    or
                    signal_std <=
                    CONSTANT_STD_THRESHOLD
                ):

                    result[
                        "constant_channels"
                    ] += 1

                    # Keep constant signal unchanged
                    df[column] = cleaned_series
                    continue

            # -----------------------------------------------
            # ROBUST GLOBAL SCORE
            # -----------------------------------------------

            robust_score = robust_z_score(
                cleaned_series
            )

            global_outlier = (
                np.abs(robust_score)
                >
                ROBUST_Z_THRESHOLD
            )

            # -----------------------------------------------
            # LOCAL SCORE
            # -----------------------------------------------

            local_score = (
                local_deviation_score(
                    cleaned_series
                )
            )

            local_outlier = (
                local_score
                >
                LOCAL_Z_THRESHOLD
            )

            # -----------------------------------------------
            # BOTH CONDITIONS REQUIRED
            # -----------------------------------------------

            suspicious = (
                global_outlier &
                local_outlier
            )

            suspicious = (
                suspicious
                .fillna(False)
                .astype(bool)
            )

            # -----------------------------------------------
            # COUNT DETECTED OUTLIERS
            # -----------------------------------------------

            detected_count = int(
                suspicious.sum()
            )

            result[
                "outliers_detected"
            ] += detected_count

            # -----------------------------------------------
            # CORRECT ONLY ISOLATED POINTS
            # -----------------------------------------------

            (
                corrected_series,
                corrected_count,
                long_count
            ) = correct_isolated_outliers(
                cleaned_series,
                suspicious
            )

            result[
                "outliers_corrected"
            ] += corrected_count

            result[
                "long_sequence_outliers"
            ] += long_count

            # -----------------------------------------------
            # STORE CLEANED SIGNAL
            # -----------------------------------------------

            df[column] = corrected_series

        # ----------------------------------------------------
        # ROW COUNT VALIDATION
        # ----------------------------------------------------

        result["rows_cleaned"] = len(df)

        if len(df) != result["rows_original"]:

            raise ValueError(
                "Row count changed during cleaning"
            )

        # ----------------------------------------------------
        # CREATE DESTINATION
        # ----------------------------------------------------

        destination_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        # ----------------------------------------------------
        # SAVE
        # ----------------------------------------------------

        df.to_csv(
            destination_file,
            index=False
        )

        result["status"] = "SUCCESS"

    except Exception as error:

        result["error"] = str(error)

    return result


# ============================================================
# 12. START CLEANING
# ============================================================

print("=" * 80)
print("STARTING ROBUST DATASET CLEANING")
print("=" * 80)

print()
print("RAW DATA ROOT:")
print(RAW_ROOT)

print()
print("CLEAN DATA ROOT:")
print(CLEAN_ROOT)

print()
print("REPORT ROOT:")
print(REPORT_ROOT)

print("=" * 80)


all_results = []


# ============================================================
# 13. PROCESS ONLY THE EIGHT DATASET FOLDERS
# ============================================================

for activity, folder_name in DATASET_FOLDERS:

    source_folder = (
        RAW_ROOT /
        activity /
        folder_name
    )

    destination_folder = (
        CLEAN_ROOT /
        activity /
        folder_name
    )

    print()
    print("-" * 80)
    print(
        f"Activity: {activity}"
    )
    print(
        f"Folder: {folder_name}"
    )
    print(
        f"Source: {source_folder}"
    )
    print("-" * 80)

    if not source_folder.exists():

        print(
            "WARNING: Source folder does not exist."
        )

        continue

    csv_files = sorted(
        source_folder.glob("*.csv")
    )

    print(
        f"CSV files found: {len(csv_files)}"
    )

    for file_number, source_file in enumerate(
        csv_files,
        start=1
    ):

        destination_file = (
            destination_folder /
            source_file.name
        )

        result = process_file(
            source_file=source_file,
            destination_file=destination_file,
            activity=activity,
            folder_name=folder_name
        )

        all_results.append(result)

        print(
            f"[{file_number}/{len(csv_files)}] "
            f"{source_file.name} | "
            f"{result['sensor_type']} | "
            f"Rows: "
            f"{result['rows_original']} -> "
            f"{result['rows_cleaned']} | "
            f"Signals: "
            f"{result['signal_columns']} | "
            f"Outliers: "
            f"{result['outliers_detected']} | "
            f"Corrected: "
            f"{result['outliers_corrected']} | "
            f"Long retained: "
            f"{result['long_sequence_outliers']} | "
            f"{result['status']}"
        )


# ============================================================
# 14. CREATE CLEANING REPORT
# ============================================================

report_df = pd.DataFrame(
    all_results
)


report_file = (
    REPORT_ROOT /
    "cleaning_file_level_report.csv"
)

report_df.to_csv(
    report_file,
    index=False
)


# ============================================================
# 15. CREATE CHANNEL LEVEL REPORT
# ============================================================

channel_report_rows = []


for activity, folder_name in DATASET_FOLDERS:

    source_folder = (
        RAW_ROOT /
        activity /
        folder_name
    )

    if not source_folder.exists():
        continue

    for source_file in sorted(
        source_folder.glob("*.csv")
    ):

        try:

            df = pd.read_csv(
                source_file,
                low_memory=False
            )

            sensor_type = detect_sensor_type(
                df.columns
            )

            signal_columns = (
                identify_signal_columns(
                    df,
                    sensor_type
                )
            )

            for column in signal_columns:

                series = pd.to_numeric(
                    df[column],
                    errors="coerce"
                )

                missing = int(
                    series.isna().sum()
                )

                valid = (
                    series
                    .dropna()
                )

                if len(valid) == 0:

                    channel_report_rows.append({
                        "activity": activity,
                        "folder": folder_name,
                        "file": source_file.name,
                        "sensor_type": sensor_type,
                        "channel": column,
                        "rows": len(df),
                        "missing_values": missing,
                        "missing_percentage": 100.0,
                        "mean": np.nan,
                        "std": np.nan,
                        "minimum": np.nan,
                        "maximum": np.nan,
                        "median": np.nan,
                        "constant_channel": True
                    })

                    continue

                channel_report_rows.append({
                    "activity": activity,
                    "folder": folder_name,
                    "file": source_file.name,
                    "sensor_type": sensor_type,
                    "channel": column,
                    "rows": len(df),
                    "missing_values": missing,
                    "missing_percentage": (
                        missing /
                        len(df) *
                        100
                    ),
                    "mean": float(
                        valid.mean()
                    ),
                    "std": float(
                        valid.std()
                    ),
                    "minimum": float(
                        valid.min()
                    ),
                    "maximum": float(
                        valid.max()
                    ),
                    "median": float(
                        valid.median()
                    ),
                    "constant_channel": (
                        float(
                            valid.std()
                        )
                        <=
                        CONSTANT_STD_THRESHOLD
                    )
                })

        except Exception:
            continue


channel_report_df = pd.DataFrame(
    channel_report_rows
)


channel_report_file = (
    REPORT_ROOT /
    "channel_quality_report.csv"
)


channel_report_df.to_csv(
    channel_report_file,
    index=False
)


# ============================================================
# 16. DATASET LEVEL SUMMARY
# ============================================================

print()
print("=" * 80)
print("DATASET CLEANING SUMMARY")
print("=" * 80)


total_files = len(report_df)

successful_files = int(
    (
        report_df["status"]
        ==
        "SUCCESS"
    ).sum()
)

failed_files = int(
    (
        report_df["status"]
        !=
        "SUCCESS"
    ).sum()
)

total_rows_original = int(
    report_df[
        "rows_original"
    ].sum()
)

total_rows_cleaned = int(
    report_df[
        "rows_cleaned"
    ].sum()
)

total_missing = int(
    report_df[
        "missing_values_original"
    ].sum()
)

total_missing_corrected = int(
    report_df[
        "missing_values_corrected"
    ].sum()
)

total_outliers = int(
    report_df[
        "outliers_detected"
    ].sum()
)

total_corrected = int(
    report_df[
        "outliers_corrected"
    ].sum()
)

total_long_retained = int(
    report_df[
        "long_sequence_outliers"
    ].sum()
)

total_constant_channels = int(
    report_df[
        "constant_channels"
    ].sum()
)


print(
    f"Files processed: "
    f"{total_files}"
)

print(
    f"Successfully processed: "
    f"{successful_files}"
)

print(
    f"Failed files: "
    f"{failed_files}"
)

print(
    f"Original total rows: "
    f"{total_rows_original:,}"
)

print(
    f"Cleaned total rows: "
    f"{total_rows_cleaned:,}"
)

print(
    f"Missing values detected: "
    f"{total_missing:,}"
)

print(
    f"Missing values corrected: "
    f"{total_missing_corrected:,}"
)

print(
    f"Outliers detected: "
    f"{total_outliers:,}"
)

print(
    f"Isolated outliers corrected: "
    f"{total_corrected:,}"
)

print(
    f"Long sequence outliers retained: "
    f"{total_long_retained:,}"
)

print(
    f"Constant channels detected: "
    f"{total_constant_channels:,}"
)


# ============================================================
# 17. ROW COUNT VALIDATION
# ============================================================

print()
print("=" * 80)
print("ROW COUNT VALIDATION")
print("=" * 80)


row_count_ok = (
    report_df["rows_original"]
    ==
    report_df["rows_cleaned"]
).all()


if row_count_ok:

    print(
        "SUCCESS: Row count was preserved "
        "for every successfully processed file."
    )

else:

    print(
        "WARNING: One or more files changed "
        "their row count."
    )


# ============================================================
# 18. FILE COUNT VERIFICATION
# ============================================================

print()
print("=" * 80)
print("CLEAN DATA FILE VERIFICATION")
print("=" * 80)


expected_total = 480

actual_clean_files = 0

folder_verification = []


for activity, folder_name in DATASET_FOLDERS:

    original_folder = (
        RAW_ROOT /
        activity /
        folder_name
    )

    clean_folder = (
        CLEAN_ROOT /
        activity /
        folder_name
    )

    original_files = sorted(
        original_folder.glob("*.csv")
    ) if original_folder.exists() else []

    clean_files = sorted(
        clean_folder.glob("*.csv")
    ) if clean_folder.exists() else []

    original_names = {
        file.name
        for file in original_files
    }

    clean_names = {
        file.name
        for file in clean_files
    }

    missing_files = (
        original_names -
        clean_names
    )

    extra_files = (
        clean_names -
        original_names
    )

    actual_clean_files += len(
        clean_files
    )

    folder_verification.append({
        "activity": activity,
        "folder": folder_name,
        "original_files": len(
            original_files
        ),
        "cleaned_files": len(
            clean_files
        ),
        "missing_files": len(
            missing_files
        ),
        "extra_files": len(
            extra_files
        )
    })

    print()
    print(
        f"{activity}/{folder_name}"
    )

    print(
        f"Original: "
        f"{len(original_files)}"
    )

    print(
        f"Cleaned: "
        f"{len(clean_files)}"
    )

    print(
        f"Missing: "
        f"{len(missing_files)}"
    )

    print(
        f"Extra: "
        f"{len(extra_files)}"
    )


verification_df = pd.DataFrame(
    folder_verification
)


verification_file = (
    REPORT_ROOT /
    "dataset_folder_verification.csv"
)


verification_df.to_csv(
    verification_file,
    index=False
)


# ============================================================
# 19. FINAL STATUS
# ============================================================

print()
print("=" * 80)
print("FINAL CLEANING STATUS")
print("=" * 80)


if (
    actual_clean_files == expected_total
    and
    row_count_ok
    and
    failed_files == 0
):

    print(
        "SUCCESS: All 480 CSV files were "
        "processed successfully."
    )

    print(
        "SUCCESS: All original row counts "
        "were preserved."
    )

    print(
        "SUCCESS: Cleaned dataset is ready "
        "for quality control and visualization."
    )

else:

    print(
        "WARNING: The cleaning process requires "
        "additional verification."
    )

    print(
        f"Expected files: "
        f"{expected_total}"
    )

    print(
        f"Cleaned files: "
        f"{actual_clean_files}"
    )

    print(
        f"Failed files: "
        f"{failed_files}"
    )


# ============================================================
# 20. REPORT LOCATIONS
# ============================================================

print()
print("=" * 80)
print("GENERATED REPORTS")
print("=" * 80)

print(
    f"File level report:\n"
    f"{report_file}"
)

print(
    f"\nChannel quality report:\n"
    f"{channel_report_file}"
)

print(
    f"\nFolder verification report:\n"
    f"{verification_file}"
)

print(
    f"\nCleaned dataset:\n"
    f"{CLEAN_ROOT}"
)

print("=" * 80)
```

```python
import os
import re
import shutil
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")


# ============================================================
# 1. CONFIGURATION
# ============================================================

RAW_DATA = Path("/home/rakib/Desktop/RAW_DATA")

CLEAN_DATA = RAW_DATA / "CLEAN_DATA_V2"

REPORT_DIR = CLEAN_DATA / "cleaning_reports"

PLOT_DIR = CLEAN_DATA / "validation_plots"

INSPECTION_DIR = CLEAN_DATA / "structure_inspection"


# Activities
ACTIVITIES = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]


# Sensor folders
SENSORS = [
    "clms",
    "clmw",
    "mums",
    "mumw",
    "norms",
    "normw",
    "rdms",
    "rdmw"
]


# ------------------------------------------------------------
# Cleaning parameters
# ------------------------------------------------------------

# Robust outlier threshold.
# Higher value = less aggressive.
OUTLIER_Z = 6.0

# Maximum number of consecutive abnormal observations
# that will be corrected.
#
# Longer sequences are preserved.
MAX_ISOLATED_SEQUENCE = 3

# Maximum missing sequence that can be interpolated.
MAX_MISSING_SEQUENCE = 5

# Minimum number of non missing observations required
# for a signal to be processed.
MIN_VALID_RATIO = 0.80

# Maximum number of channels shown in diagnostic plots.
MAX_PLOT_CHANNELS = 6

# Number of rows used for plotting.
MAX_PLOT_ROWS = 10000


# ============================================================
# 2. DIRECTORY SETUP
# ============================================================

REPORT_DIR.mkdir(parents=True, exist_ok=True)
PLOT_DIR.mkdir(parents=True, exist_ok=True)
INSPECTION_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# 3. SENSOR IDENTIFICATION
# ============================================================

def identify_sensor(folder_name):

    name = folder_name.lower()

    if name.endswith("ms"):
        return "Muse EEG"

    if name.endswith("mw"):
        return "Motion Watch"

    return "Unknown"


# ============================================================
# 4. BASIC COLUMN CLASSIFICATION
# ============================================================

def normalize_column_name(column):

    return (
        str(column)
        .strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")
    )


def is_timestamp_column(column):

    name = normalize_column_name(column)

    timestamp_keywords = [
        "timestamp",
        "time_stamp",
        "datetime",
        "date_time",
        "time",
        "clock"
    ]

    return any(k in name for k in timestamp_keywords)


def is_metadata_column(column):

    name = normalize_column_name(column)

    metadata_keywords = [
        "timestamp",
        "time",
        "datetime",
        "date",
        "sample",
        "index",
        "id",
        "event",
        "marker",
        "label",
        "activity"
    ]

    return any(k == name or name.startswith(k + "_")
               for k in metadata_keywords)


# ============================================================
# 5. AUTOMATIC SIGNAL COLUMN DETECTION
# ============================================================

def detect_signal_columns(df, sensor_type):

    signal_columns = []

    for column in df.columns:

        # Metadata columns are excluded.
        if is_metadata_column(column):
            continue

        # Try converting to numeric.
        numeric = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        valid_ratio = numeric.notna().mean()

        # Require at least 80 percent numeric observations.
        if valid_ratio < MIN_VALID_RATIO:
            continue

        # At least some variation is preferred.
        unique_values = numeric.nunique(dropna=True)

        if unique_values <= 1:
            # Constant channels are recorded separately.
            signal_columns.append(column)
            continue

        signal_columns.append(column)

    return signal_columns


# ============================================================
# 6. READ CSV SAFELY
# ============================================================

def read_csv_safely(file_path):

    encodings = [
        "utf-8",
        "utf-8-sig",
        "latin1"
    ]

    last_error = None

    for encoding in encodings:

        try:

            df = pd.read_csv(
                file_path,
                encoding=encoding,
                low_memory=False
            )

            return df, None

        except Exception as e:

            last_error = str(e)

    return None, last_error


# ============================================================
# 7. STRUCTURE INSPECTION
# ============================================================

def inspect_file(file_path, sensor_folder):

    sensor_type = identify_sensor(sensor_folder)

    df, error = read_csv_safely(file_path)

    result = {
        "file": file_path.name,
        "path": str(file_path),
        "sensor_folder": sensor_folder,
        "sensor_type": sensor_type,
        "status": "FAILED" if df is None else "OK",
        "rows": 0 if df is None else len(df),
        "columns": 0 if df is None else len(df),
        "signal_columns": 0,
        "signal_names": "",
        "timestamp_columns": "",
        "constant_columns": "",
        "error": error or ""
    }

    if df is None:
        return result

    signal_columns = detect_signal_columns(
        df,
        sensor_type
    )

    timestamp_columns = [
        c for c in df.columns
        if is_timestamp_column(c)
    ]

    constant_columns = []

    for column in signal_columns:

        numeric = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        if numeric.nunique(dropna=True) <= 1:
            constant_columns.append(column)

    result["signal_columns"] = len(signal_columns)

    result["signal_names"] = "|".join(
        map(str, signal_columns)
    )

    result["timestamp_columns"] = "|".join(
        map(str, timestamp_columns)
    )

    result["constant_columns"] = "|".join(
        map(str, constant_columns)
    )

    return result


# ============================================================
# 8. ROBUST OUTLIER DETECTION
# ============================================================

def robust_outlier_mask(series, threshold=6.0):

    x = pd.to_numeric(
        series,
        errors="coerce"
    )

    median = x.median()

    mad = np.median(
        np.abs(
            x.dropna() - median
        )
    )

    if pd.isna(mad) or mad == 0:

        return pd.Series(
            False,
            index=series.index
        )

    robust_z = (
        0.6745 * (x - median) / mad
    )

    mask = robust_z.abs() > threshold

    mask = mask.fillna(False)

    return mask


# ============================================================
# 9. CONSECUTIVE TRUE RUNS
# ============================================================

def get_true_runs(mask):

    mask = np.asarray(mask, dtype=bool)

    runs = []

    start = None

    for i, value in enumerate(mask):

        if value and start is None:

            start = i

        elif not value and start is not None:

            runs.append(
                (start, i - 1)
            )

            start = None

    if start is not None:

        runs.append(
            (start, len(mask) - 1)
        )

    return runs


# ============================================================
# 10. CORRECT ONLY ISOLATED OUTLIERS
# ============================================================

def correct_isolated_outliers(
    series,
    outlier_mask,
    max_sequence=3
):

    result = pd.to_numeric(
        series,
        errors="coerce"
    ).copy()

    corrected = 0

    retained = 0

    runs = get_true_runs(
        outlier_mask.values
    )

    for start, end in runs:

        length = end - start + 1

        if length <= max_sequence:

            # Temporarily replace abnormal values
            # with NaN.
            result.iloc[start:end + 1] = np.nan

            corrected += length

        else:

            # Long sequences are preserved.
            retained += length

    # Interpolate only values identified as isolated.
    result = result.interpolate(
        method="linear",
        limit_direction="both"
    )

    return result, corrected, retained


# ============================================================
# 11. MISSING VALUE HANDLING
# ============================================================

def missing_sequence_lengths(series):

    missing = pd.isna(series).values

    runs = get_true_runs(missing)

    lengths = [
        end - start + 1
        for start, end in runs
    ]

    return lengths


def interpolate_short_missing_values(
    series,
    maximum_sequence=5
):

    result = pd.to_numeric(
        series,
        errors="coerce"
    ).copy()

    missing_mask = result.isna()

    runs = get_true_runs(
        missing_mask.values
    )

    corrected = 0

    long_missing = 0

    for start, end in runs:

        length = end - start + 1

        if length <= maximum_sequence:

            corrected += length

        else:

            long_missing += length

    # Interpolate only short missing sequences.
    result = result.interpolate(
        method="linear",
        limit=maximum_sequence,
        limit_direction="both"
    )

    return (
        result,
        corrected,
        long_missing
    )


# ============================================================
# 12. PROCESS ONE FILE
# ============================================================

def process_file(
    input_file,
    output_file,
    activity,
    sensor_folder
):

    sensor_type = identify_sensor(
        sensor_folder
    )

    df, error = read_csv_safely(
        input_file
    )

    if df is None:

        return {
            "activity": activity,
            "sensor_folder": sensor_folder,
            "file": input_file.name,
            "sensor_type": sensor_type,
            "status": "FAILED",
            "rows_before": 0,
            "rows_after": 0,
            "signal_columns": 0,
            "missing_detected": 0,
            "missing_corrected": 0,
            "long_missing_retained": 0,
            "outliers_detected": 0,
            "outliers_corrected": 0,
            "long_outliers_retained": 0,
            "constant_channels": 0,
            "error": error
        }

    rows_before = len(df)

    signal_columns = detect_signal_columns(
        df,
        sensor_type
    )

    missing_detected = 0
    missing_corrected = 0
    long_missing_retained = 0

    outliers_detected = 0
    outliers_corrected = 0
    long_outliers_retained = 0

    constant_channels = []

    # --------------------------------------------------------
    # Process signal columns
    # --------------------------------------------------------

    for column in signal_columns:

        numeric = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        # --------------------------------------------
        # Constant channel detection
        # --------------------------------------------

        if numeric.nunique(
            dropna=True
        ) <= 1:

            constant_channels.append(
                column
            )

            # Do not modify constant channels.
            continue

        # --------------------------------------------
        # Missing values
        # --------------------------------------------

        missing_detected += int(
            numeric.isna().sum()
        )

        (
            numeric,
            missing_fixed,
            long_missing
        ) = interpolate_short_missing_values(
            numeric,
            MAX_MISSING_SEQUENCE
        )

        missing_corrected += missing_fixed

        long_missing_retained += long_missing

        # --------------------------------------------
        # Outlier detection
        # --------------------------------------------

        outlier_mask = robust_outlier_mask(
            numeric,
            OUTLIER_Z
        )

        outliers_detected += int(
            outlier_mask.sum()
        )

        (
            cleaned,
            corrected,
            retained
        ) = correct_isolated_outliers(
            numeric,
            outlier_mask,
            MAX_ISOLATED_SEQUENCE
        )

        outliers_corrected += corrected

        long_outliers_retained += retained

        # Restore cleaned numeric values.
        df[column] = cleaned

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    df.to_csv(
        output_file,
        index=False
    )

    rows_after = len(df)

    return {
        "activity": activity,
        "sensor_folder": sensor_folder,
        "file": input_file.name,
        "sensor_type": sensor_type,
        "status": "SUCCESS",
        "rows_before": rows_before,
        "rows_after": rows_after,
        "signal_columns": len(signal_columns),
        "missing_detected": missing_detected,
        "missing_corrected": missing_corrected,
        "long_missing_retained": long_missing_retained,
        "outliers_detected": outliers_detected,
        "outliers_corrected": outliers_corrected,
        "long_outliers_retained": long_outliers_retained,
        "constant_channels": len(
            constant_channels
        ),
        "constant_channel_names": "|".join(
            map(str, constant_channels)
        ),
        "error": ""
    }


# ============================================================
# 13. PLOT BEFORE AND AFTER
# ============================================================

def create_validation_plot(
    original_file,
    cleaned_file,
    activity,
    sensor_folder
):

    original, error1 = read_csv_safely(
        original_file
    )

    cleaned, error2 = read_csv_safely(
        cleaned_file
    )

    if original is None or cleaned is None:
        return

    signal_columns = detect_signal_columns(
        original,
        identify_sensor(sensor_folder)
    )

    if len(signal_columns) == 0:
        return

    signal_columns = signal_columns[
        :MAX_PLOT_CHANNELS
    ]

    for column in signal_columns:

        original_values = pd.to_numeric(
            original[column],
            errors="coerce"
        )

        cleaned_values = pd.to_numeric(
            cleaned[column],
            errors="coerce"
        )

        n = min(
            len(original_values),
            MAX_PLOT_ROWS
        )

        x = np.arange(n)

        plt.figure(
            figsize=(14, 5)
        )

        plt.plot(
            x,
            original_values.iloc[:n],
            label="Original"
        )

        plt.plot(
            x,
            cleaned_values.iloc[:n],
            label="Cleaned"
        )

        plt.title(
            f"{activity} | {sensor_folder} | {column}"
        )

        plt.xlabel(
            "Sample"
        )

        plt.ylabel(
            "Signal value"
        )

        plt.legend()

        plt.tight_layout()

        safe_column = re.sub(
            r"[^A-Za-z0-9_]+",
            "_",
            str(column)
        )

        output_dir = (
            PLOT_DIR
            / activity
            / sensor_folder
        )

        output_dir.mkdir(
            parents=True,
            exist_ok=True
        )

        output_file = (
            output_dir
            / f"{original_file.stem}_{safe_column}.png"
        )

        plt.savefig(
            output_file,
            dpi=150
        )

        plt.close()


# ============================================================
# 14. MAIN PROCESSING
# ============================================================

def main():

    print("=" * 80)

    print(
        "DATASET VALIDATION AND CLEANING V2"
    )

    print("=" * 80)

    # --------------------------------------------------------
    # Remove previous V2 output
    # --------------------------------------------------------

    if CLEAN_DATA.exists():

        print(
            "\nRemoving previous CLEAN_DATA_V2..."
        )

        shutil.rmtree(
            CLEAN_DATA
        )

    REPORT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    PLOT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    INSPECTION_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    all_results = []

    structure_results = []

    # --------------------------------------------------------
    # Process activities
    # --------------------------------------------------------

    for activity in ACTIVITIES:

        activity_path = RAW_DATA / activity

        if not activity_path.exists():

            print(
                f"\nWARNING: Missing activity folder: "
                f"{activity}"
            )

            continue

        print(
            f"\n{'=' * 80}"
        )

        print(
            f"ACTIVITY: {activity}"
        )

        print(
            f"{'=' * 80}"
        )

        for sensor_folder in SENSORS:

            input_folder = (
                activity_path
                / sensor_folder
            )

            if not input_folder.exists():

                continue

            print(
                f"\nProcessing: "
                f"{activity}/{sensor_folder}"
            )

            csv_files = sorted(
                input_folder.glob("*.csv")
            )

            print(
                f"CSV files found: "
                f"{len(csv_files)}"
            )

            output_folder = (
                CLEAN_DATA
                / activity
                / sensor_folder
            )

            output_folder.mkdir(
                parents=True,
                exist_ok=True
            )

            for i, input_file in enumerate(
                csv_files,
                start=1
            ):

                print(
                    f"[{i}/{len(csv_files)}] "
                    f"{input_file.name}",
                    end=" "
                )

                # ----------------------------------------
                # Structure inspection
                # ----------------------------------------

                structure = inspect_file(
                    input_file,
                    sensor_folder
                )

                structure_results.append(
                    structure
                )

                # ----------------------------------------
                # Process
                # ----------------------------------------

                output_file = (
                    output_folder
                    / input_file.name
                )

                result = process_file(
                    input_file,
                    output_file,
                    activity,
                    sensor_folder
                )

                all_results.append(
                    result
                )

                if result["status"] == "SUCCESS":

                    print(
                        f"| {result['sensor_type']} "
                        f"| Rows: "
                        f"{result['rows_before']} -> "
                        f"{result['rows_after']} "
                        f"| Signals: "
                        f"{result['signal_columns']} "
                        f"| Missing: "
                        f"{result['missing_detected']} "
                        f"| Missing fixed: "
                        f"{result['missing_corrected']} "
                        f"| Outliers: "
                        f"{result['outliers_detected']} "
                        f"| Corrected: "
                        f"{result['outliers_corrected']} "
                        f"| Long retained: "
                        f"{result['long_outliers_retained']}"
                    )

                else:

                    print(
                        f"| FAILED: "
                        f"{result['error']}"
                    )

    # ========================================================
    # SAVE REPORTS
    # ========================================================

    results_df = pd.DataFrame(
        all_results
    )

    structure_df = pd.DataFrame(
        structure_results
    )

    results_file = (
        REPORT_DIR
        / "file_cleaning_report_v2.csv"
    )

    structure_file = (
        REPORT_DIR
        / "csv_structure_report_v2.csv"
    )

    results_df.to_csv(
        results_file,
        index=False
    )

    structure_df.to_csv(
        structure_file,
        index=False
    )

    # ========================================================
    # FINAL SUMMARY
    # ========================================================

    print("\n")

    print("=" * 80)

    print(
        "FINAL DATASET CLEANING SUMMARY"
    )

    print("=" * 80)

    total_files = len(results_df)

    successful = (
        results_df["status"]
        == "SUCCESS"
    ).sum()

    failed = (
        results_df["status"]
        == "FAILED"
    ).sum()

    rows_before = (
        results_df["rows_before"]
        .sum()
    )

    rows_after = (
        results_df["rows_after"]
        .sum()
    )

    missing_detected = (
        results_df["missing_detected"]
        .sum()
    )

    missing_corrected = (
        results_df["missing_corrected"]
        .sum()
    )

    long_missing = (
        results_df["long_missing_retained"]
        .sum()
    )

    outliers = (
        results_df["outliers_detected"]
        .sum()
    )

    corrected = (
        results_df["outliers_corrected"]
        .sum()
    )

    long_outliers = (
        results_df["long_outliers_retained"]
        .sum()
    )

    constants = (
        results_df["constant_channels"]
        .sum()
    )

    print(
        f"Files discovered:          {total_files}"
    )

    print(
        f"Successfully processed:    {successful}"
    )

    print(
        f"Failed files:              {failed}"
    )

    print(
        f"Rows before cleaning:      {rows_before:,}"
    )

    print(
        f"Rows after cleaning:       {rows_after:,}"
    )

    print(
        f"Missing detected:          {missing_detected:,}"
    )

    print(
        f"Missing corrected:         {missing_corrected:,}"
    )

    print(
        f"Long missing retained:     {long_missing:,}"
    )

    print(
        f"Outliers detected:         {outliers:,}"
    )

    print(
        f"Outliers corrected:        {corrected:,}"
    )

    print(
        f"Long outliers retained:    {long_outliers:,}"
    )

    print(
        f"Constant channels:         {constants:,}"
    )

    # ========================================================
    # ROW VALIDATION
    # ========================================================

    print("\n")

    print("=" * 80)

    print(
        "ROW COUNT VALIDATION"
    )

    print("=" * 80)

    row_mismatch = results_df[
        results_df["rows_before"]
        != results_df["rows_after"]
    ]

    if len(row_mismatch) == 0:

        print(
            "SUCCESS: Row count preserved "
            "for every successfully processed file."
        )

    else:

        print(
            f"WARNING: {len(row_mismatch)} "
            f"files have row count mismatch."
        )

        print(
            row_mismatch[
                [
                    "activity",
                    "sensor_folder",
                    "file",
                    "rows_before",
                    "rows_after"
                ]
            ].to_string(
                index=False
            )
        )

    # ========================================================
    # MOTION WATCH VALIDATION
    # ========================================================

    print("\n")

    print("=" * 80)

    print(
        "MOTION WATCH SIGNAL VALIDATION"
    )

    print("=" * 80)

    motion_structure = structure_df[
        structure_df["sensor_type"]
        == "Motion Watch"
    ]

    print(
        f"Motion Watch files inspected: "
        f"{len(motion_structure)}"
    )

    if len(motion_structure) > 0:

        print(
            "\nSignal column distribution:"
        )

        print(
            motion_structure[
                "signal_columns"
            ].value_counts()
            .sort_index()
            .to_string()
        )

        zero_signal = motion_structure[
            motion_structure["signal_columns"]
            == 0
        ]

        if len(zero_signal) > 0:

            print(
                f"\nWARNING: "
                f"{len(zero_signal)} "
                f"Motion Watch files have "
                f"zero detected signal columns."
            )

            print(
                zero_signal[
                    [
                        "file",
                        "columns",
                        "signal_names",
                        "error"
                    ]
                ].head(20).to_string(
                    index=False
                )
            )

        else:

            print(
                "\nSUCCESS: Motion Watch "
                "signal columns were detected."
            )

    # ========================================================
    # FAILED FILES
    # ========================================================

    print("\n")

    print("=" * 80)

    print(
        "FAILED FILES"
    )

    print("=" * 80)

    failed_df = results_df[
        results_df["status"]
        == "FAILED"
    ]

    if len(failed_df) == 0:

        print(
            "No failed files."
        )

    else:

        print(
            failed_df[
                [
                    "activity",
                    "sensor_folder",
                    "file",
                    "error"
                ]
            ].to_string(
                index=False
            )
        )

    # ========================================================
    # CONSTANT CHANNEL REPORT
    # ========================================================

    constant_df = results_df[
        results_df["constant_channels"]
        > 0
    ]

    constant_file = (
        REPORT_DIR
        / "constant_channels_report_v2.csv"
    )

    constant_df.to_csv(
        constant_file,
        index=False
    )

    print("\n")

    print("=" * 80)

    print(
        "REPORT LOCATIONS"
    )

    print("=" * 80)

    print(
        results_file
    )

    print(
        structure_file
    )

    print(
        constant_file
    )

    print(
        PLOT_DIR
    )

    print("\n")

    print("=" * 80)

    print(
        "PROCESS COMPLETED"
    )

    print("=" * 80)


# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":

    main()
```

```python
import os
import re
import shutil
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


warnings.filterwarnings("ignore")


# ============================================================
# CONFIGURATION
# ============================================================

RAW_DATA = Path.home() / "Desktop" / "RAW_DATA"

OUTPUT_DATA = RAW_DATA / "CLEAN_DATA_V3"

CLEAN_DATA = OUTPUT_DATA / "clean_data"

REPORT_DIR = OUTPUT_DATA / "cleaning_reports"

PLOT_DIR = OUTPUT_DATA / "validation_plots"

ANOMALY_DIR = OUTPUT_DATA / "anomaly_reports"


# ============================================================
# ACTIVITIES
# ============================================================

ACTIVITIES = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]


# ============================================================
# SENSOR FOLDER PREFIXES
# ============================================================

SENSOR_PREFIXES = {
    "Muse EEG": ["clms", "mums", "norms", "rdms"],
    "Motion Watch": ["clmw", "mumw", "normw", "rdmw"]
}


# ============================================================
# MUSE EEG CHANNELS
# ============================================================

PREFERRED_EEG_CHANNELS = [
    "TP9",
    "AF7",
    "AF8",
    "TP10"
]


# ============================================================
# POSSIBLE MOTION WATCH SIGNAL NAMES
# ============================================================

MOTION_SIGNAL_KEYWORDS = [
    "activity",
    "light",
    "lux",
    "count",
    "counts",
    "activity_count",
    "activity_counts",
    "steps",
    "movement",
    "motion"
]


# ============================================================
# COLUMNS THAT SHOULD NOT BE TREATED AS SIGNALS
# ============================================================

METADATA_KEYWORDS = [
    "timestamp",
    "time",
    "date",
    "datetime",
    "record",
    "index",
    "serial",
    "person",
    "participant",
    "subject",
    "id",
    "label",
    "class",
    "activity_name",
    "condition"
]


# ============================================================
# CLEANING PARAMETERS
# ============================================================

# Short missing value gap.
MAX_MISSING_GAP = 5


# Hampel window.
HAMPEL_RADIUS = 7


# Robust deviation threshold.
ROBUST_Z_THRESHOLD = 6.0


# Local deviation threshold.
LOCAL_Z_THRESHOLD = 6.0


# Number of neighbouring observations used
# to decide whether a point is isolated.
LOCAL_RADIUS = 2


# A suspicious point is corrected only when
# enough neighbouring observations are valid.
MIN_VALID_NEIGHBOURS = 3


# Maximum number of points in a consecutive anomaly
# sequence that can be treated as isolated anomalies.
MAX_ISOLATED_RUN = 2


# Number of observations shown in validation plots.
PLOT_POINTS = 1500


# ============================================================
# CREATE DIRECTORIES
# ============================================================

def create_directories():

    OUTPUT_DATA.mkdir(
        parents=True,
        exist_ok=True
    )

    CLEAN_DATA.mkdir(
        parents=True,
        exist_ok=True
    )

    REPORT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    PLOT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    ANOMALY_DIR.mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# NORMALISE COLUMN NAME
# ============================================================

def normalise_column_name(column):

    text = str(column)

    text = text.strip()

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text


# ============================================================
# IDENTIFY SENSOR TYPE
# ============================================================

def identify_sensor_type(file_name):

    name = file_name.lower()

    for sensor_type, prefixes in SENSOR_PREFIXES.items():

        for prefix in prefixes:

            if name.startswith(prefix):

                return sensor_type

    return "Unknown"


# ============================================================
# IDENTIFY ACTIVITY
# ============================================================

def identify_activity(file_path):

    parts = [
        part.lower()
        for part in file_path.parts
    ]

    for activity in ACTIVITIES:

        if activity.lower() in parts:

            return activity

    return "Unknown"


# ============================================================
# READ CSV SAFELY
# ============================================================

def read_csv_safely(file_path):

    try:

        df = pd.read_csv(
            file_path,
            low_memory=False
        )

        return df, None

    except Exception as error:

        return None, str(error)


# ============================================================
# CONVERT NUMERIC VALUES
# ============================================================

def convert_possible_numeric_columns(df):

    result = df.copy()

    for column in result.columns:

        if result[column].dtype == object:

            converted = pd.to_numeric(
                result[column],
                errors="coerce"
            )

            original_valid = (
                result[column]
                .notna()
                .sum()
            )

            converted_valid = (
                converted
                .notna()
                .sum()
            )

            if (
                original_valid > 0
                and
                converted_valid / original_valid >= 0.80
            ):

                result[column] = converted

    return result


# ============================================================
# CHECK WHETHER COLUMN IS METADATA
# ============================================================

def is_metadata_column(column):

    name = str(column).strip().lower()

    for keyword in METADATA_KEYWORDS:

        if keyword in name:

            return True

    return False


# ============================================================
# IDENTIFY EEG SIGNAL COLUMNS
# ============================================================

def identify_eeg_channels(df):

    channels = []

    columns_lower = {
        str(column).strip().lower(): column
        for column in df.columns
    }

    # First identify the four primary Muse electrodes.
    for preferred in PREFERRED_EEG_CHANNELS:

        key = preferred.lower()

        if key in columns_lower:

            actual_column = columns_lower[key]

            if pd.api.types.is_numeric_dtype(
                df[actual_column]
            ):

                channels.append(
                    actual_column
                )

    # Identify other columns that clearly
    # contain EEG electrode names.
    electrode_patterns = [
        r"^tp9$",
        r"^tp10$",
        r"^af7$",
        r"^af8$",
        r"^tp9_",
        r"^tp10_",
        r"^af7_",
        r"^af8_",
        r"eeg",
        r"theta",
        r"alpha",
        r"beta",
        r"delta",
        r"gamma"
    ]

    for column in df.columns:

        if column in channels:

            continue

        if not pd.api.types.is_numeric_dtype(
            df[column]
        ):

            continue

        name = str(column).strip().lower()

        for pattern in electrode_patterns:

            if re.search(
                pattern,
                name
            ):

                channels.append(
                    column
                )

                break

    # If the EEG file contains numerical signal columns
    # but the names are different, use numerical columns
    # that are not metadata.
    if len(channels) == 0:

        for column in df.columns:

            if is_metadata_column(column):

                continue

            if pd.api.types.is_numeric_dtype(
                df[column]
            ):

                channels.append(
                    column
                )

    return list(
        dict.fromkeys(channels)
    )


# ============================================================
# IDENTIFY MOTION WATCH SIGNAL COLUMNS
# ============================================================

def identify_motion_channels(df):

    channels = []

    # First use names.
    for column in df.columns:

        if not pd.api.types.is_numeric_dtype(
            df[column]
        ):

            continue

        name = str(column).strip().lower()

        for keyword in MOTION_SIGNAL_KEYWORDS:

            if keyword in name:

                channels.append(
                    column
                )

                break

    # If Activity and Light are present with
    # slightly different names, they are included.
    #
    # If no named signals are found, use numeric
    # columns that are not metadata.
    if len(channels) == 0:

        for column in df.columns:

            if is_metadata_column(column):

                continue

            if pd.api.types.is_numeric_dtype(
                df[column]
            ):

                channels.append(
                    column
                )

    return list(
        dict.fromkeys(channels)
    )


# ============================================================
# IDENTIFY VALID SIGNAL CHANNELS
# ============================================================

def identify_signal_channels(
    df,
    sensor_type
):

    if sensor_type == "Muse EEG":

        return identify_eeg_channels(df)

    if sensor_type == "Motion Watch":

        return identify_motion_channels(df)

    return []


# ============================================================
# CONSTANT CHANNEL DETECTION
# ============================================================

def detect_constant_channels(
    df,
    signal_columns
):

    constant_columns = []

    for column in signal_columns:

        series = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        valid = series.dropna()

        if len(valid) == 0:

            continue

        if valid.nunique() <= 1:

            constant_columns.append(
                column
            )

    return constant_columns


# ============================================================
# SHORT MISSING VALUE INTERPOLATION
# ============================================================

def correct_short_missing_values(
    series,
    max_gap
):

    original = series.copy()

    missing_mask = original.isna()

    if missing_mask.sum() == 0:

        return original, 0

    result = original.interpolate(
        method="linear",
        limit=max_gap,
        limit_direction="both"
    )

    corrected = (
        missing_mask
        &
        result.notna()
    ).sum()

    return result, int(corrected)


# ============================================================
# ROBUST SCALE
# ============================================================

def robust_scale(series):

    values = series.to_numpy(
        dtype=float
    )

    valid = values[
        np.isfinite(values)
    ]

    if len(valid) == 0:

        return (
            np.full(
                len(values),
                np.nan
            ),
            np.nan,
            np.nan
        )

    median = np.median(
        valid
    )

    mad = np.median(
        np.abs(
            valid - median
        )
    )

    if mad == 0:

        return (
            np.zeros(
                len(values)
            ),
            median,
            mad
        )

    robust_z = (
        0.6745
        *
        (values - median)
        /
        mad
    )

    return (
        robust_z,
        median,
        mad
    )


# ============================================================
# ISOLATED ANOMALY DETECTION
# ============================================================

def detect_isolated_anomalies(
    series,
    hampel_radius=HAMPEL_RADIUS,
    robust_threshold=ROBUST_Z_THRESHOLD,
    local_threshold=LOCAL_Z_THRESHOLD,
    local_radius=LOCAL_RADIUS
):

    values = pd.to_numeric(
        series,
        errors="coerce"
    ).to_numpy(
        dtype=float
    )

    n = len(values)

    global_z, _, _ = robust_scale(
        series
    )

    global_candidate = (
        np.abs(global_z)
        >
        robust_threshold
    )

    local_candidate = np.zeros(
        n,
        dtype=bool
    )

    for i in range(n):

        value = values[i]

        if not np.isfinite(value):

            continue

        left_start = max(
            0,
            i - local_radius
        )

        right_end = min(
            n,
            i + local_radius + 1
        )

        neighbours = np.concatenate(
            [
                values[
                    left_start:i
                ],
                values[
                    i + 1:right_end
                ]
            ]
        )

        neighbours = neighbours[
            np.isfinite(neighbours)
        ]

        if (
            len(neighbours)
            <
            MIN_VALID_NEIGHBOURS
        ):

            continue

        local_median = np.median(
            neighbours
        )

        local_mad = np.median(
            np.abs(
                neighbours
                -
                local_median
            )
        )

        if local_mad == 0:

            deviation = abs(
                value
                -
                local_median
            )

            neighbour_range = (
                np.max(neighbours)
                -
                np.min(neighbours)
            )

            if neighbour_range == 0:

                local_candidate[i] = (
                    deviation > 0
                )

            else:

                local_candidate[i] = (
                    deviation
                    >
                    3.0
                    *
                    neighbour_range
                )

        else:

            local_z = (
                0.6745
                *
                (
                    value
                    -
                    local_median
                )
                /
                local_mad
            )

            local_candidate[i] = (
                abs(local_z)
                >
                local_threshold
            )

    candidate = (
        global_candidate
        &
        local_candidate
    )

    # Remove long consecutive anomaly sequences.
    # This is important because a real physiological
    # change may remain elevated for many observations.
    final_mask = candidate.copy()

    i = 0

    while i < n:

        if not candidate[i]:

            i += 1

            continue

        start = i

        while (
            i < n
            and
            candidate[i]
        ):

            i += 1

        end = i

        run_length = end - start

        if run_length > MAX_ISOLATED_RUN:

            final_mask[
                start:end
            ] = False

    return final_mask


# ============================================================
# CORRECT ISOLATED ANOMALIES
# ============================================================

def correct_isolated_anomalies(
    series,
    anomaly_mask
):

    values = pd.to_numeric(
        series,
        errors="coerce"
    ).to_numpy(
        dtype=float
    )

    cleaned = values.copy()

    corrected = 0

    anomaly_indices = np.where(
        anomaly_mask
    )[0]

    for index in anomaly_indices:

        left_values = []

        right_values = []

        # Search to the left.
        for distance in range(
            1,
            4
        ):

            position = (
                index
                -
                distance
            )

            if position < 0:

                break

            value = values[
                position
            ]

            if np.isfinite(value):

                left_values.append(
                    value
                )

                break

        # Search to the right.
        for distance in range(
            1,
            4
        ):

            position = (
                index
                +
                distance
            )

            if position >= len(values):

                break

            value = values[
                position
            ]

            if np.isfinite(value):

                right_values.append(
                    value
                )

                break

        neighbours = (
            left_values
            +
            right_values
        )

        if len(neighbours) == 0:

            continue

        replacement = np.median(
            neighbours
        )

        if np.isfinite(replacement):

            cleaned[index] = (
                replacement
            )

            corrected += 1

    result = pd.Series(
        cleaned,
        index=series.index,
        name=series.name
    )

    return result, corrected


# ============================================================
# PROCESS ONE SIGNAL
# ============================================================

def clean_signal(
    series
):

    original = pd.to_numeric(
        series,
        errors="coerce"
    )

    missing_before = int(
        original.isna().sum()
    )

    # Correct only short missing gaps.
    after_missing, missing_corrected = (
        correct_short_missing_values(
            original,
            MAX_MISSING_GAP
        )
    )

    anomaly_mask = (
        detect_isolated_anomalies(
            after_missing
        )
    )

    outliers_detected = int(
        anomaly_mask.sum()
    )

    after_outliers, outliers_corrected = (
        correct_isolated_anomalies(
            after_missing,
            anomaly_mask
        )
    )

    # Recheck missing values.
    missing_after = int(
        after_outliers.isna().sum()
    )

    # Recheck remaining outliers.
    remaining_anomalies = (
        detect_isolated_anomalies(
            after_outliers
        )
    )

    remaining_outliers = int(
        remaining_anomalies.sum()
    )

    return (
        after_outliers,
        missing_before,
        missing_corrected,
        missing_after,
        outliers_detected,
        outliers_corrected,
        remaining_outliers,
        anomaly_mask
    )


# ============================================================
# VALIDATION PLOT
# ============================================================

def create_validation_plot(
    original_df,
    cleaned_df,
    signal_columns,
    activity,
    sensor_type,
    file_name,
    plot_path
):

    if len(signal_columns) == 0:

        return False

    # Select up to four useful channels.
    selected = []

    for preferred in PREFERRED_EEG_CHANNELS:

        for column in signal_columns:

            if str(column).lower() == (
                preferred.lower()
            ):

                selected.append(
                    column
                )

    # For Motion Watch and remaining signals,
    # add additional columns.
    for column in signal_columns:

        if column not in selected:

            selected.append(
                column
            )

        if len(selected) >= 4:

            break

    selected = selected[:4]

    if len(selected) == 0:

        return False

    figure, axes = plt.subplots(
        len(selected),
        1,
        figsize=(14, 3.5 * len(selected)),
        squeeze=False
    )

    axes = axes.flatten()

    total_rows = len(
        original_df
    )

    if total_rows > PLOT_POINTS:

        indices = np.linspace(
            0,
            total_rows - 1,
            PLOT_POINTS,
            dtype=int
        )

    else:

        indices = np.arange(
            total_rows
        )

    for axis, column in zip(
        axes,
        selected
    ):

        original_values = pd.to_numeric(
            original_df[column],
            errors="coerce"
        ).to_numpy(
            dtype=float
        )

        cleaned_values = pd.to_numeric(
            cleaned_df[column],
            errors="coerce"
        ).to_numpy(
            dtype=float
        )

        axis.plot(
            indices,
            original_values[
                indices
            ],
            label="Original"
        )

        axis.plot(
            indices,
            cleaned_values[
                indices
            ],
            label="Cleaned"
        )

        axis.set_title(
            str(column)
        )

        axis.set_xlabel(
            "Observation"
        )

        axis.set_ylabel(
            "Value"
        )

        axis.legend()

        axis.grid(
            alpha=0.25
        )

    figure.suptitle(
        f"{activity} | {sensor_type} | {file_name}"
    )

    figure.tight_layout()

    plot_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    figure.savefig(
        plot_path,
        dpi=150,
        bbox_inches="tight"
    )

    plt.close(
        figure
    )

    return True


# ============================================================
# CREATE ANOMALY PLOT
# ============================================================

def create_anomaly_plot(
    original_series,
    cleaned_series,
    anomaly_mask,
    activity,
    sensor_type,
    file_name,
    channel,
    plot_path
):

    values = pd.to_numeric(
        original_series,
        errors="coerce"
    ).to_numpy(
        dtype=float
    )

    cleaned_values = pd.to_numeric(
        cleaned_series,
        errors="coerce"
    ).to_numpy(
        dtype=float
    )

    n = len(values)

    if n > PLOT_POINTS:

        indices = np.linspace(
            0,
            n - 1,
            PLOT_POINTS,
            dtype=int
        )

    else:

        indices = np.arange(
            n
        )

    figure = plt.figure(
        figsize=(14, 5)
    )

    plt.plot(
        indices,
        values[
            indices
        ],
        label="Original"
    )

    plt.plot(
        indices,
        cleaned_values[
            indices
        ],
        label="Cleaned"
    )

    anomaly_indices = np.where(
        anomaly_mask
    )[0]

    anomaly_indices = anomaly_indices[
        anomaly_indices < n
    ]

    if len(anomaly_indices) > 0:

        if n > PLOT_POINTS:

            selected_anomalies = (
                anomaly_indices[
                    np.searchsorted(
                        anomaly_indices,
                        indices
                    )
                    .clip(
                        max=0
                    )
                ]
                if False
                else anomaly_indices
            )

        else:

            selected_anomalies = (
                anomaly_indices
            )

        plt.scatter(
            selected_anomalies,
            values[
                selected_anomalies
            ],
            label="Corrected isolated observations",
            s=20
        )

    plt.title(
        f"{activity} | {sensor_type} | "
        f"{file_name} | {channel}"
    )

    plt.xlabel(
        "Observation"
    )

    plt.ylabel(
        "Value"
    )

    plt.legend()

    plt.grid(
        alpha=0.25
    )

    plot_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    plt.savefig(
        plot_path,
        dpi=150,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# PROCESS ONE CSV FILE
# ============================================================

def process_file(
    input_file,
    output_file,
    activity,
    sensor_type
):

    result = {
        "activity": activity,
        "sensor_type": sensor_type,
        "file": input_file.name,
        "status": "SUCCESS",
        "rows_before": 0,
        "rows_after": 0,
        "columns_before": 0,
        "columns_after": 0,
        "signal_columns": 0,
        "signal_names": "",
        "constant_channels": 0,
        "constant_channel_names": "",
        "missing_detected": 0,
        "missing_corrected": 0,
        "missing_remaining": 0,
        "outliers_detected": 0,
        "outliers_corrected": 0,
        "outliers_remaining": 0,
        "long_missing_retained": 0,
        "failed_reason": ""
    }

    original_df, error = (
        read_csv_safely(
            input_file
        )
    )

    if error is not None:

        result["status"] = "FAILED"

        result["failed_reason"] = error

        return result

    if original_df is None:

        result["status"] = "FAILED"

        result["failed_reason"] = (
            "CSV could not be loaded"
        )

        return result

    original_df.columns = [
        normalise_column_name(
            column
        )
        for column in original_df.columns
    ]

    df = convert_possible_numeric_columns(
        original_df
    )

    result["rows_before"] = len(df)

    result["columns_before"] = len(
        df.columns
    )

    signal_columns = identify_signal_channels(
        df,
        sensor_type
    )

    result["signal_columns"] = len(
        signal_columns
    )

    result["signal_names"] = "|".join(
        [
            str(column)
            for column in signal_columns
        ]
    )

    # --------------------------------------------------------
    # Constant channel detection
    # --------------------------------------------------------

    constant_channels = (
        detect_constant_channels(
            df,
            signal_columns
        )
    )

    result["constant_channels"] = len(
        constant_channels
    )

    result["constant_channel_names"] = "|".join(
        [
            str(column)
            for column in constant_channels
        ]
    )

    cleaned_df = df.copy()

    all_anomaly_information = []

    # --------------------------------------------------------
    # Clean each signal channel
    # --------------------------------------------------------

    for column in signal_columns:

        (
            cleaned_signal,
            missing_before,
            missing_corrected,
            missing_after,
            outliers_detected,
            outliers_corrected,
            remaining_outliers,
            anomaly_mask
        ) = clean_signal(
            df[column]
        )

        cleaned_df[column] = (
            cleaned_signal
        )

        result[
            "missing_detected"
        ] += missing_before

        result[
            "missing_corrected"
        ] += missing_corrected

        result[
            "missing_remaining"
        ] += missing_after

        result[
            "outliers_detected"
        ] += outliers_detected

        result[
            "outliers_corrected"
        ] += outliers_corrected

        result[
            "outliers_remaining"
        ] += remaining_outliers

        all_anomaly_information.append(
            (
                column,
                anomaly_mask
            )
        )

    # --------------------------------------------------------
    # Calculate long missing values
    # --------------------------------------------------------

    long_missing_total = 0

    for column in signal_columns:

        missing_mask = (
            cleaned_df[column]
            .isna()
        )

        if not missing_mask.any():

            continue

        groups = (
            missing_mask
            .astype(int)
            .groupby(
                missing_mask
                .astype(int)
                .ne(
                    missing_mask
                    .astype(int)
                    .shift()
                )
                .cumsum()
            )
            .sum()
        )

        long_missing_total += int(
            groups[
                groups
                >
                MAX_MISSING_GAP
            ].sum()
        )

    result[
        "long_missing_retained"
    ] = long_missing_total

    # --------------------------------------------------------
    # Preserve row count
    # --------------------------------------------------------

    result["rows_after"] = len(
        cleaned_df
    )

    result["columns_after"] = len(
        cleaned_df.columns
    )

    if (
        result["rows_before"]
        !=
        result["rows_after"]
    ):

        result["status"] = "FAILED"

        result["failed_reason"] = (
            "Row count changed"
        )

        return result

    # --------------------------------------------------------
    # Save cleaned file
    # --------------------------------------------------------

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    cleaned_df.to_csv(
        output_file,
        index=False
    )

    # --------------------------------------------------------
    # Create validation plot
    # --------------------------------------------------------

    plot_activity_dir = (
        PLOT_DIR
        /
        activity
        /
        sensor_type.replace(
            " ",
            "_"
        )
    )

    plot_file = (
        plot_activity_dir
        /
        f"{input_file.stem}_validation.png"
    )

    create_validation_plot(
        original_df,
        cleaned_df,
        signal_columns,
        activity,
        sensor_type,
        input_file.name,
        plot_file
    )

    # --------------------------------------------------------
    # Save anomaly information
    # --------------------------------------------------------

    anomaly_activity_dir = (
        ANOMALY_DIR
        /
        activity
        /
        sensor_type.replace(
            " ",
            "_"
        )
    )

    anomaly_activity_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    anomaly_records = []

    for (
        column,
        anomaly_mask
    ) in all_anomaly_information:

        indices = np.where(
            anomaly_mask
        )[0]

        for index in indices:

            anomaly_records.append(
                {
                    "activity": activity,
                    "sensor_type": sensor_type,
                    "file": input_file.name,
                    "channel": str(column),
                    "row_index": int(index),
                    "original_value": df.loc[
                        index,
                        column
                    ],
                    "cleaned_value": cleaned_df.loc[
                        index,
                        column
                    ]
                }
            )

    if len(anomaly_records) > 0:

        anomaly_file = (
            anomaly_activity_dir
            /
            f"{input_file.stem}_anomalies.csv"
        )

        pd.DataFrame(
            anomaly_records
        ).to_csv(
            anomaly_file,
            index=False
        )

    return result


# ============================================================
# FIND INPUT CSV FILES
# ============================================================

def discover_files():

    discovered = []

    for activity in ACTIVITIES:

        activity_dir = (
            RAW_DATA
            /
            activity
        )

        if not activity_dir.exists():

            print(
                f"WARNING: Missing activity folder: "
                f"{activity_dir}"
            )

            continue

        for sensor_folder in (
            activity_dir.iterdir()
        ):

            if not sensor_folder.is_dir():

                continue

            for csv_file in sorted(
                sensor_folder.glob("*.csv")
            ):

                sensor_type = (
                    identify_sensor_type(
                        csv_file.name
                    )
                )

                if sensor_type == "Unknown":

                    continue

                discovered.append(
                    (
                        activity,
                        sensor_folder.name,
                        csv_file,
                        sensor_type
                    )
                )

    return discovered


# ============================================================
# SAVE REPORTS
# ============================================================

def save_reports(
    results
):

    report_df = pd.DataFrame(
        results
    )

    cleaning_report = (
        REPORT_DIR
        /
        "file_cleaning_report_v3.csv"
    )

    report_df.to_csv(
        cleaning_report,
        index=False
    )

    # --------------------------------------------------------
    # Constant channels report
    # --------------------------------------------------------

    constant_df = report_df[
        report_df[
            "constant_channels"
        ]
        >
        0
    ].copy()

    constant_report = (
        REPORT_DIR
        /
        "constant_channels_report_v3.csv"
    )

    constant_df.to_csv(
        constant_report,
        index=False
    )

    # --------------------------------------------------------
    # Signal structure report
    # --------------------------------------------------------

    structure_columns = [
        "activity",
        "sensor_type",
        "file",
        "status",
        "rows_before",
        "rows_after",
        "columns_before",
        "columns_after",
        "signal_columns",
        "signal_names",
        "failed_reason"
    ]

    structure_df = report_df[
        structure_columns
    ].copy()

    structure_report = (
        REPORT_DIR
        /
        "csv_structure_report_v3.csv"
    )

    structure_df.to_csv(
        structure_report,
        index=False
    )

    return (
        cleaning_report,
        constant_report,
        structure_report
    )


# ============================================================
# CREATE SUMMARY REPORT
# ============================================================

def create_summary(
    results,
    total_discovered
):

    successful = [
        row
        for row in results
        if row["status"] == "SUCCESS"
    ]

    failed = [
        row
        for row in results
        if row["status"] != "SUCCESS"
    ]

    rows_before = sum(
        row["rows_before"]
        for row in successful
    )

    rows_after = sum(
        row["rows_after"]
        for row in successful
    )

    missing_detected = sum(
        row["missing_detected"]
        for row in successful
    )

    missing_corrected = sum(
        row["missing_corrected"]
        for row in successful
    )

    missing_remaining = sum(
        row["missing_remaining"]
        for row in successful
    )

    outliers_detected = sum(
        row["outliers_detected"]
        for row in successful
    )

    outliers_corrected = sum(
        row["outliers_corrected"]
        for row in successful
    )

    outliers_remaining = sum(
        row["outliers_remaining"]
        for row in successful
    )

    long_missing = sum(
        row["long_missing_retained"]
        for row in successful
    )

    constant_channels = sum(
        row["constant_channels"]
        for row in successful
    )

    muse_files = sum(
        1
        for row in successful
        if row["sensor_type"]
        ==
        "Muse EEG"
    )

    motion_files = sum(
        1
        for row in successful
        if row["sensor_type"]
        ==
        "Motion Watch"
    )

    muse_zero_signal = sum(
        1
        for row in successful
        if (
            row["sensor_type"]
            ==
            "Muse EEG"
            and
            row["signal_columns"]
            ==
            0
        )
    )

    motion_zero_signal = sum(
        1
        for row in successful
        if (
            row["sensor_type"]
            ==
            "Motion Watch"
            and
            row["signal_columns"]
            ==
            0
        )
    )

    summary_file = (
        REPORT_DIR
        /
        "dataset_summary_v3.txt"
    )

    with open(
        summary_file,
        "w",
        encoding="utf8"
    ) as file:

        file.write(
            "\n"
            + "=" * 80
            + "\n"
        )

        file.write(
            "DATASET CLEANING AND VALIDATION V3\n"
        )

        file.write(
            "=" * 80
            + "\n\n"
        )

        file.write(
            f"Files discovered: {total_discovered}\n"
        )

        file.write(
            f"Successfully processed: "
            f"{len(successful)}\n"
        )

        file.write(
            f"Failed files: {len(failed)}\n\n"
        )

        file.write(
            f"Rows before cleaning: "
            f"{rows_before}\n"
        )

        file.write(
            f"Rows after cleaning: "
            f"{rows_after}\n"
        )

        file.write(
            f"Row count preserved: "
            f"{rows_before == rows_after}\n\n"
        )

        file.write(
            f"Missing detected: "
            f"{missing_detected}\n"
        )

        file.write(
            f"Missing corrected: "
            f"{missing_corrected}\n"
        )

        file.write(
            f"Missing remaining: "
            f"{missing_remaining}\n"
        )

        file.write(
            f"Long missing retained: "
            f"{long_missing}\n\n"
        )

        file.write(
            f"Isolated outliers detected: "
            f"{outliers_detected}\n"
        )

        file.write(
            f"Isolated outliers corrected: "
            f"{outliers_corrected}\n"
        )

        file.write(
            f"Isolated outliers remaining: "
            f"{outliers_remaining}\n\n"
        )

        file.write(
            f"Constant channels: "
            f"{constant_channels}\n\n"
        )

        file.write(
            f"Muse EEG files: "
            f"{muse_files}\n"
        )

        file.write(
            f"Motion Watch files: "
            f"{motion_files}\n"
        )

        file.write(
            f"Muse EEG files with zero signal columns: "
            f"{muse_zero_signal}\n"
        )

        file.write(
            f"Motion Watch files with zero signal columns: "
            f"{motion_zero_signal}\n\n"
        )

        file.write(
            "=" * 80
            + "\n"
        )

        file.write(
            "FAILED FILES\n"
        )

        file.write(
            "=" * 80
            + "\n"
        )

        for row in failed:

            file.write(
                f"{row['activity']} | "
                f"{row['sensor_type']} | "
                f"{row['file']} | "
                f"{row['failed_reason']}\n"
            )

        file.write(
            "\n"
            + "=" * 80
            + "\n"
        )

        file.write(
            "OUTPUT LOCATIONS\n"
        )

        file.write(
            "=" * 80
            + "\n"
        )

        file.write(
            f"Clean data: {CLEAN_DATA}\n"
        )

        file.write(
            f"Reports: {REPORT_DIR}\n"
        )

        file.write(
            f"Validation plots: {PLOT_DIR}\n"
        )

        file.write(
            f"Anomaly reports: {ANOMALY_DIR}\n"
        )


# ============================================================
# MAIN
# ============================================================

def main():

    print(
        "\n"
        + "=" * 80
    )

    print(
        "DATASET VALIDATION AND CLEANING V3"
    )

    print(
        "=" * 80
        + "\n"
    )

    print(
        f"Input directory:\n{RAW_DATA}\n"
    )

    print(
        f"Output directory:\n{OUTPUT_DATA}\n"
    )

    # --------------------------------------------------------
    # Create directories
    # --------------------------------------------------------

    create_directories()

    # --------------------------------------------------------
    # Remove previous V3 output
    # --------------------------------------------------------

    if OUTPUT_DATA.exists():

        print(
            "Removing previous CLEAN_DATA_V3..."
        )

        shutil.rmtree(
            OUTPUT_DATA
        )

        create_directories()

    # --------------------------------------------------------
    # Discover files
    # --------------------------------------------------------

    discovered = discover_files()

    total_discovered = len(
        discovered
    )

    print(
        f"CSV files discovered: "
        f"{total_discovered}\n"
    )

    # --------------------------------------------------------
    # Process files
    # --------------------------------------------------------

    results = []

    current_activity = None

    current_sensor_folder = None

    for counter, item in enumerate(
        discovered,
        start=1
    ):

        (
            activity,
            sensor_folder,
            input_file,
            sensor_type
        ) = item

        if activity != current_activity:

            print(
                "\n"
                + "=" * 80
            )

            print(
                f"ACTIVITY: {activity}"
            )

            print(
                "=" * 80
            )

            current_activity = activity

        if sensor_folder != current_sensor_folder:

            print(
                f"\nProcessing: "
                f"{activity}/{sensor_folder}"
            )

            current_sensor_folder = (
                sensor_folder
            )

        output_folder = (
            CLEAN_DATA
            /
            activity
            /
            sensor_folder
        )

        output_file = (
            output_folder
            /
            input_file.name
        )

        result = process_file(
            input_file,
            output_file,
            activity,
            sensor_type
        )

        results.append(
            result
        )

        if result["status"] == "SUCCESS":

            print(
                f"[{counter}/{total_discovered}] "
                f"{input_file.name} | "
                f"{sensor_type} | "
                f"Rows: "
                f"{result['rows_before']} -> "
                f"{result['rows_after']} | "
                f"Signals: "
                f"{result['signal_columns']} | "
                f"Missing: "
                f"{result['missing_detected']} | "
                f"Missing fixed: "
                f"{result['missing_corrected']} | "
                f"Outliers: "
                f"{result['outliers_detected']} | "
                f"Corrected: "
                f"{result['outliers_corrected']} | "
                f"Remaining: "
                f"{result['outliers_remaining']}"
            )

        else:

            print(
                f"[{counter}/{total_discovered}] "
                f"{input_file.name} | "
                f"FAILED: "
                f"{result['failed_reason']}"
            )

    # --------------------------------------------------------
    # Save reports
    # --------------------------------------------------------

    (
        cleaning_report,
        constant_report,
        structure_report
    ) = save_reports(
        results
    )

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    create_summary(
        results,
        total_discovered
    )

    # --------------------------------------------------------
    # Final terminal summary
    # --------------------------------------------------------

    successful = sum(
        1
        for row in results
        if row["status"] == "SUCCESS"
    )

    failed = sum(
        1
        for row in results
        if row["status"] != "SUCCESS"
    )

    total_rows_before = sum(
        row["rows_before"]
        for row in results
        if row["status"] == "SUCCESS"
    )

    total_rows_after = sum(
        row["rows_after"]
        for row in results
        if row["status"] == "SUCCESS"
    )

    motion_zero = sum(
        1
        for row in results
        if (
            row["status"] == "SUCCESS"
            and
            row["sensor_type"]
            ==
            "Motion Watch"
            and
            row["signal_columns"]
            ==
            0
        )
    )

    muse_zero = sum(
        1
        for row in results
        if (
            row["status"] == "SUCCESS"
            and
            row["sensor_type"]
            ==
            "Muse EEG"
            and
            row["signal_columns"]
            ==
            0
        )
    )

    print(
        "\n"
        + "=" * 80
    )

    print(
        "FINAL DATASET CLEANING SUMMARY"
    )

    print(
        "=" * 80
    )

    print(
        f"Files discovered: "
        f"{total_discovered}"
    )

    print(
        f"Successfully processed: "
        f"{successful}"
    )

    print(
        f"Failed files: "
        f"{failed}"
    )

    print(
        f"Rows before cleaning: "
        f"{total_rows_before}"
    )

    print(
        f"Rows after cleaning: "
        f"{total_rows_after}"
    )

    print(
        f"Row count preserved: "
        f"{total_rows_before == total_rows_after}"
    )

    print(
        f"Motion Watch zero signal files: "
        f"{motion_zero}"
    )

    print(
        f"Muse EEG zero signal files: "
        f"{muse_zero}"
    )

    print(
        "\n"
        + "=" * 80
    )

    print(
        "REPORT LOCATIONS"
    )

    print(
        "=" * 80
    )

    print(
        f"Cleaning report:\n"
        f"{cleaning_report}"
    )

    print(
        f"\nStructure report:\n"
        f"{structure_report}"
    )

    print(
        f"\nConstant channel report:\n"
        f"{constant_report}"
    )

    print(
        f"\nValidation plots:\n"
        f"{PLOT_DIR}"
    )

    print(
        f"\nAnomaly reports:\n"
        f"{ANOMALY_DIR}"
    )

    print(
        "\n"
        + "=" * 80
    )

    print(
        "PROCESS COMPLETED"
    )

    print(
        "=" * 80
    )


# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":

    main()
```

```python
# ============================================================
# MISSING VALUE HANDLING FOR 480 CSV FILES
# ============================================================
#
# INPUT:
#   CLEAN_DATA_V3/clean_data
#
# OUTPUT:
#   CLEAN_DATA_V4/
#       ├── clean_data/
#       ├── missing_value_reports/
#       └── summary/
#
# METHOD:
#   1. Identify CSV files
#   2. Identify timestamp columns
#   3. Identify numeric signal columns
#   4. Detect missing values
#   5. Interpolate only short internal gaps
#   6. Preserve long gaps as NaN
#   7. Preserve timestamps
#   8. Preserve row count
#   9. Generate detailed reports
#  10. Verify all files
#
# ============================================================

import os
import re
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

BASE_DIR = Path.cwd()

INPUT_DIR = BASE_DIR / "CLEAN_DATA_V3" / "clean_data"

OUTPUT_DIR = BASE_DIR / "CLEAN_DATA_V4"

CLEAN_DIR = OUTPUT_DIR / "clean_data"

REPORT_DIR = OUTPUT_DIR / "missing_value_reports"

SUMMARY_DIR = OUTPUT_DIR / "summary"


# Create directories
CLEAN_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)
SUMMARY_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# 2. CONFIGURATION
# ============================================================

# Maximum consecutive missing observations that will be
# interpolated.
#
# Example:
# MAX_INTERPOLATION_GAP = 5
#
# A gap of 1 to 5 consecutive missing values:
#     interpolate
#
# A gap greater than 5:
#     keep as NaN
#
MAX_INTERPOLATION_GAP = 5


# Minimum number of valid observations required for
# interpolation.
MIN_VALID_VALUES = 2


# ============================================================
# 3. EXPECTED DATASET STRUCTURE
# ============================================================

EXPECTED_FOLDERS = [
    Path("class_lecture") / "clms",
    Path("class_lecture") / "clmw",

    Path("music") / "mums",
    Path("music") / "mumw",

    Path("normal") / "norms",
    Path("normal") / "normw",

    Path("reading") / "rdms",
    Path("reading") / "rdmw",
]


# ============================================================
# 4. SENSOR IDENTIFICATION
# ============================================================

def identify_sensor(relative_path):
    """
    Identify sensor type from folder name.

    m folders:
        Motion Watch

    s folders:
        Muse

    This follows the dataset folder structure:
        clms, clmw
        mums, mumw
        norms, normw
        rdms, rdmw
    """

    folder_name = relative_path.parent.name.lower()

    if folder_name.endswith("mw"):
        return "Motion Watch"

    if folder_name.endswith("ms"):
        return "Muse"

    return "Unknown"


# ============================================================
# 5. ACTIVITY IDENTIFICATION
# ============================================================

def identify_activity(relative_path):

    parts = relative_path.parts

    if len(parts) >= 2:

        activity = parts[0].lower()

        if activity == "class_lecture":
            return "Class Lecture"

        if activity == "music":
            return "Music"

        if activity == "normal":
            return "Normal"

        if activity == "reading":
            return "Reading"

    return "Unknown"


# ============================================================
# 6. IDENTIFY TIMESTAMP COLUMNS
# ============================================================

def identify_timestamp_columns(df):

    timestamp_columns = []

    for col in df.columns:

        col_lower = str(col).strip().lower()

        timestamp_keywords = [
            "timestamp",
            "time_stamp",
            "datetime",
            "date_time",
            "date",
            "time"
        ]

        if any(keyword in col_lower for keyword in timestamp_keywords):

            timestamp_columns.append(col)

    return timestamp_columns


# ============================================================
# 7. IDENTIFY NON SIGNAL COLUMNS
# ============================================================

def identify_non_signal_columns(df):

    non_signal = []

    timestamp_columns = identify_timestamp_columns(df)

    for col in df.columns:

        col_lower = str(col).strip().lower()

        # Timestamp
        if col in timestamp_columns:
            non_signal.append(col)
            continue

        # Common identifier columns
        identifier_keywords = [
            "id",
            "serial",
            "participant",
            "person",
            "subject",
            "activity",
            "label",
            "class",
            "index"
        ]

        if any(
            keyword == col_lower or
            col_lower.startswith(keyword + "_")
            for keyword in identifier_keywords
        ):
            non_signal.append(col)
            continue

    return list(dict.fromkeys(non_signal))


# ============================================================
# 8. IDENTIFY NUMERIC SIGNAL COLUMNS
# ============================================================

def identify_signal_columns(df):

    non_signal_columns = identify_non_signal_columns(df)

    signal_columns = []

    for col in df.columns:

        if col in non_signal_columns:
            continue

        if pd.api.types.is_numeric_dtype(df[col]):

            signal_columns.append(col)

    return signal_columns


# ============================================================
# 9. FIND CONSECUTIVE MISSING SEQUENCES
# ============================================================

def find_missing_runs(series):

    missing = series.isna().to_numpy()

    runs = []

    start = None

    for i, value in enumerate(missing):

        if value and start is None:

            start = i

        elif not value and start is not None:

            end = i - 1

            runs.append(
                (
                    start,
                    end,
                    end - start + 1
                )
            )

            start = None

    # Missing sequence reaches end
    if start is not None:

        end = len(missing) - 1

        runs.append(
            (
                start,
                end,
                end - start + 1
            )
        )

    return runs


# ============================================================
# 10. INTERPOLATE SHORT INTERNAL GAPS
# ============================================================

def interpolate_short_gaps(series, max_gap=5):

    original = series.copy()

    total_missing_before = int(original.isna().sum())

    if total_missing_before == 0:

        return (
            original,
            0,
            0
        )

    # Linear interpolation
    #
    # limit_area="inside" means:
    #
    # Internal gaps can be interpolated.
    #
    # Missing values at the beginning and end
    # are NOT extrapolated.
    #
    interpolated = original.interpolate(
        method="linear",
        limit=max_gap,
        limit_direction="both",
        limit_area="inside"
    )

    total_missing_after = int(interpolated.isna().sum())

    corrected = (
        total_missing_before -
        total_missing_after
    )

    return (
        interpolated,
        total_missing_before,
        corrected
    )


# ============================================================
# 11. PROCESS ONE CSV FILE
# ============================================================

def process_file(input_file):

    relative_path = input_file.relative_to(INPUT_DIR)

    output_file = CLEAN_DIR / relative_path

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    activity = identify_activity(relative_path)

    sensor = identify_sensor(relative_path)

    result = {
        "file": str(relative_path),
        "activity": activity,
        "sensor": sensor,

        "status": "Success",

        "rows_before": 0,
        "rows_after": 0,

        "columns": 0,
        "signal_columns": 0,

        "missing_before": 0,
        "missing_after": 0,

        "short_gap_values_corrected": 0,

        "long_gap_values_retained": 0,

        "leading_missing_values": 0,
        "trailing_missing_values": 0,

        "constant_signal_columns": 0,

        "error": ""
    }

    try:

        # ----------------------------------------------------
        # READ CSV
        # ----------------------------------------------------

        df = pd.read_csv(
            input_file,
            low_memory=False
        )

        result["rows_before"] = len(df)
        result["columns"] = len(df.columns)

        # ----------------------------------------------------
        # IDENTIFY SIGNAL COLUMNS
        # ----------------------------------------------------

        signal_columns = identify_signal_columns(df)

        result["signal_columns"] = len(
            signal_columns
        )

        # ----------------------------------------------------
        # TOTAL MISSING BEFORE
        # ----------------------------------------------------

        missing_before = int(
            df[signal_columns]
            .isna()
            .sum()
            .sum()
        )

        result["missing_before"] = missing_before

        # ----------------------------------------------------
        # PROCESS EACH SIGNAL
        # ----------------------------------------------------

        file_details = []

        for col in signal_columns:

            series = df[col].copy()

            missing_count = int(
                series.isna().sum()
            )

            if missing_count == 0:
                continue

            # ------------------------------------------------
            # FIND MISSING RUNS
            # ------------------------------------------------

            runs = find_missing_runs(series)

            short_missing = 0
            long_missing = 0

            leading_missing = 0
            trailing_missing = 0

            for start, end, length in runs:

                # Leading missing values
                if start == 0:

                    leading_missing += length

                # Trailing missing values
                if end == len(series) - 1:

                    trailing_missing += length

                # Short gap
                if (
                    start > 0 and
                    end < len(series) - 1 and
                    length <= MAX_INTERPOLATION_GAP
                ):

                    short_missing += length

                # Long gap
                elif (
                    start > 0 and
                    end < len(series) - 1 and
                    length > MAX_INTERPOLATION_GAP
                ):

                    long_missing += length

            # ------------------------------------------------
            # INTERPOLATE SHORT INTERNAL GAPS
            # ------------------------------------------------

            cleaned_series, before, corrected = (
                interpolate_short_gaps(
                    series,
                    MAX_INTERPOLATION_GAP
                )
            )

            df[col] = cleaned_series

            # ------------------------------------------------
            # COUNT REMAINING LONG GAPS
            # ------------------------------------------------

            remaining_missing = int(
                cleaned_series.isna().sum()
            )

            file_details.append({

                "file": str(relative_path),

                "activity": activity,

                "sensor": sensor,

                "column": col,

                "missing_before": missing_count,

                "short_gap_candidates": short_missing,

                "corrected": corrected,

                "remaining_missing": remaining_missing,

                "long_gap_candidates": long_missing,

                "leading_missing": leading_missing,

                "trailing_missing": trailing_missing

            })

            result[
                "short_gap_values_corrected"
            ] += corrected

            result[
                "leading_missing_values"
            ] += leading_missing

            result[
                "trailing_missing_values"
            ] += trailing_missing

        # ----------------------------------------------------
        # MISSING AFTER
        # ----------------------------------------------------

        missing_after = int(
            df[signal_columns]
            .isna()
            .sum()
            .sum()
        )

        result["missing_after"] = missing_after

        result["long_gap_values_retained"] = (
            missing_after
        )

        # ----------------------------------------------------
        # CONSTANT SIGNAL CHANNELS
        # ----------------------------------------------------

        constant_count = 0

        for col in signal_columns:

            valid = df[col].dropna()

            if len(valid) > 0:

                if valid.nunique() <= 1:

                    constant_count += 1

        result[
            "constant_signal_columns"
        ] = constant_count

        # ----------------------------------------------------
        # VERIFY ROW COUNT
        # ----------------------------------------------------

        result["rows_after"] = len(df)

        if result["rows_before"] != result["rows_after"]:

            raise ValueError(
                "Row count changed during processing."
            )

        # ----------------------------------------------------
        # SAVE CLEANED FILE
        # ----------------------------------------------------

        df.to_csv(
            output_file,
            index=False
        )

        # ----------------------------------------------------
        # SAVE COLUMN LEVEL REPORT
        # ----------------------------------------------------

        if file_details:

            detail_df = pd.DataFrame(
                file_details
            )

            report_file = (
                REPORT_DIR /
                relative_path.parent /
                f"{input_file.stem}_missing_report.csv"
            )

            report_file.parent.mkdir(
                parents=True,
                exist_ok=True
            )

            detail_df.to_csv(
                report_file,
                index=False
            )

    except Exception as e:

        result["status"] = "Failed"

        result["error"] = str(e)

    return result


# ============================================================
# 12. DISCOVER ONLY DATASET CSV FILES
# ============================================================

print()
print("=" * 80)
print("STARTING MISSING VALUE HANDLING")
print("=" * 80)

print()
print("Input directory:")
print(INPUT_DIR)

print()
print("Output directory:")
print(OUTPUT_DIR)


if not INPUT_DIR.exists():

    raise FileNotFoundError(
        f"Input directory does not exist:\n{INPUT_DIR}"
    )


# IMPORTANT:
#
# We search ONLY inside CLEAN_DATA_V3/clean_data.
#
# Therefore files from .venv, matplotlib, etc.
# cannot accidentally be processed.
#

csv_files = sorted(
    INPUT_DIR.rglob("*.csv")
)


print()
print(
    "CSV files discovered:",
    len(csv_files)
)

print("=" * 80)


# ============================================================
# 13. PROCESS ALL FILES
# ============================================================

results = []

for i, input_file in enumerate(
    csv_files,
    start=1
):

    relative_path = input_file.relative_to(
        INPUT_DIR
    )

    print(
        f"[{i}/{len(csv_files)}] "
        f"Processing: {relative_path}"
    )

    result = process_file(
        input_file
    )

    results.append(result)

    if result["status"] == "Success":

        print(
            f"    Sensor: {result['sensor']}"
        )

        print(
            f"    Activity: {result['activity']}"
        )

        print(
            f"    Rows: "
            f"{result['rows_before']} -> "
            f"{result['rows_after']}"
        )

        print(
            f"    Signal columns: "
            f"{result['signal_columns']}"
        )

        print(
            f"    Missing before: "
            f"{result['missing_before']}"
        )

        print(
            f"    Short gap values corrected: "
            f"{result['short_gap_values_corrected']}"
        )

        print(
            f"    Missing after: "
            f"{result['missing_after']}"
        )

    else:

        print(
            f"    FAILED: {result['error']}"
        )


# ============================================================
# 14. CREATE GLOBAL REPORT
# ============================================================

results_df = pd.DataFrame(results)


global_report = (
    REPORT_DIR /
    "missing_value_cleaning_summary.csv"
)

results_df.to_csv(
    global_report,
    index=False
)


# ============================================================
# 15. ACTIVITY LEVEL SUMMARY
# ============================================================

activity_summary = (
    results_df
    .groupby("activity")
    .agg(
        files=("file", "count"),
        missing_before=("missing_before", "sum"),
        corrected=("short_gap_values_corrected", "sum"),
        missing_after=("missing_after", "sum")
    )
    .reset_index()
)

activity_summary.to_csv(
    SUMMARY_DIR /
    "missing_values_by_activity.csv",
    index=False
)


# ============================================================
# 16. SENSOR LEVEL SUMMARY
# ============================================================

sensor_summary = (
    results_df
    .groupby("sensor")
    .agg(
        files=("file", "count"),
        missing_before=("missing_before", "sum"),
        corrected=("short_gap_values_corrected", "sum"),
        missing_after=("missing_after", "sum")
    )
    .reset_index()
)

sensor_summary.to_csv(
    SUMMARY_DIR /
    "missing_values_by_sensor.csv",
    index=False
)


# ============================================================
# 17. FINAL STATISTICS
# ============================================================

total_files = len(results_df)

successful_files = int(
    (results_df["status"] == "Success").sum()
)

failed_files = int(
    (results_df["status"] == "Failed").sum()
)

total_missing_before = int(
    results_df["missing_before"].sum()
)

total_corrected = int(
    results_df[
        "short_gap_values_corrected"
    ].sum()
)

total_remaining = int(
    results_df["missing_after"].sum()
)


# ============================================================
# 18. FINAL REPORT
# ============================================================

print()
print("=" * 80)
print("MISSING VALUE HANDLING COMPLETED")
print("=" * 80)

print()
print(
    f"Total CSV files:              {total_files}"
)

print(
    f"Successfully processed:       {successful_files}"
)

print(
    f"Failed files:                 {failed_files}"
)

print(
    f"Missing values before:        {total_missing_before:,}"
)

print(
    f"Short gap values corrected:   {total_corrected:,}"
)

print(
    f"Missing values remaining:     {total_remaining:,}"
)

print()
print(
    f"Maximum interpolation gap:    "
    f"{MAX_INTERPOLATION_GAP}"
)

print()
print(
    "Final data directory:"
)

print(
    CLEAN_DIR
)

print()
print(
    "Global report:"
)

print(
    global_report
)

print()
print("=" * 80)


# ============================================================
# 19. VERIFY OUTPUT FILE COUNT
# ============================================================

output_files = sorted(
    CLEAN_DIR.rglob("*.csv")
)

print()
print("=" * 80)
print("FINAL FILE VERIFICATION")
print("=" * 80)

print(
    "Input CSV files:",
    len(csv_files)
)

print(
    "Output CSV files:",
    len(output_files)
)

if len(csv_files) == len(output_files):

    print()
    print(
        "SUCCESS: Output file count matches input."
    )

else:

    print()
    print(
        "WARNING: Input and output file counts differ."
    )


# ============================================================
# 20. VERIFY ROW COUNTS
# ============================================================

print()
print("Checking row counts...")

row_errors = []

for result in results:

    if result["status"] != "Success":
        continue

    if (
        result["rows_before"] !=
        result["rows_after"]
    ):

        row_errors.append(
            result["file"]
        )


if len(row_errors) == 0:

    print(
        "SUCCESS: Row count preserved for all "
        "successfully processed files."
    )

else:

    print(
        "WARNING: Row count changed in:"
    )

    for file in row_errors:

        print(
            "   ",
            file
        )


# ============================================================
# 21. DISPLAY ACTIVITY SUMMARY
# ============================================================

print()
print("=" * 80)
print("MISSING VALUE SUMMARY BY ACTIVITY")
print("=" * 80)

print(
    activity_summary.to_string(
        index=False
    )
)


# ============================================================
# 22. DISPLAY SENSOR SUMMARY
# ============================================================

print()
print("=" * 80)
print("MISSING VALUE SUMMARY BY SENSOR")
print("=" * 80)

print(
    sensor_summary.to_string(
        index=False
    )
)


# ============================================================
# 23. FAILED FILES
# ============================================================

failed_df = results_df[
    results_df["status"] == "Failed"
]

if len(failed_df) > 0:

    print()
    print("=" * 80)
    print("FAILED FILES")
    print("=" * 80)

    print(
        failed_df[
            ["file", "error"]
        ].to_string(index=False)
    )

else:

    print()
    print("=" * 80)
    print("NO FILES FAILED")
    print("=" * 80)


# ============================================================
# 24. COMPLETION
# ============================================================

print()
print("=" * 80)
print("PROCESS FINISHED")
print("=" * 80)

print()
print(
    "Use this folder for the next stage:"
)

print(
    CLEAN_DIR
)

print()
```

```python
# ============================================================
# COMPLETE NOISE VISUALIZATION
# INPUT: CLEAN_DATA_V4
# OUTPUT: TWO GRAPHS
#
# 1. Muse EEG
# 2. Motion Watch
#
# The original data are NOT modified.
# ============================================================

from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")


# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V4"
)

OUTPUT_ROOT = (
    INPUT_ROOT
    /
    "noise_visualizations"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. SETTINGS
# ============================================================

WINDOW_SIZE = 21

NOISE_THRESHOLD = 5.0

MAX_POINTS_PER_CHANNEL = 3000

MIN_VALID_VALUES = 30


# ============================================================
# 3. VERIFY INPUT
# ============================================================

print("=" * 80)
print("NOISE VISUALIZATION")
print("=" * 80)

print()
print("Input folder:")
print(INPUT_ROOT)

if not INPUT_ROOT.exists():

    raise FileNotFoundError(
        f"Input folder does not exist:\n{INPUT_ROOT}"
    )


# ============================================================
# 4. FIND ALL CSV FILES
# ============================================================
#
# We search recursively.
#
# Reports and previous visualization folders are ignored.
# ============================================================

csv_files = []

for file in INPUT_ROOT.rglob("*.csv"):

    relative = file.relative_to(
        INPUT_ROOT
    )

    parts_lower = [
        part.lower()
        for part in relative.parts
    ]

    if (
        "noise_visualizations"
        in parts_lower
    ):
        continue

    if (
        "visualizations"
        in parts_lower
    ):
        continue

    if (
        "reports"
        in parts_lower
    ):
        continue

    csv_files.append(file)


csv_files = sorted(
    csv_files
)


print()
print(
    "Total CSV files found:",
    len(csv_files)
)


if len(csv_files) == 0:

    raise FileNotFoundError(
        "No CSV files were found inside CLEAN_DATA_V4."
    )


# ============================================================
# 5. IDENTIFY SENSOR TYPE
# ============================================================

def identify_sensor(file_path):

    path_text = str(
        file_path
    ).lower()

    folder_names = [
        part.lower()
        for part in file_path.parts
    ]

    # Motion Watch folders normally end with mw
    if any(
        part.endswith("mw")
        for part in folder_names
    ):

        return "Motion Watch"

    # Muse folders normally end with ms
    if any(
        part.endswith("ms")
        for part in folder_names
    ):

        return "Muse"

    # Additional filename based check
    if "motion" in path_text:
        return "Motion Watch"

    if "muse" in path_text:
        return "Muse"

    return "Unknown"


# ============================================================
# 6. IDENTIFY ACTIVITY
# ============================================================

def identify_activity(file_path):

    folder_names = [
        part.lower()
        for part in file_path.parts
    ]

    if "class_lecture" in folder_names:
        return "Class Lecture"

    if "music" in folder_names:
        return "Music"

    if "normal" in folder_names:
        return "Normal"

    if "reading" in folder_names:
        return "Reading"

    return "Unknown"


# ============================================================
# 7. IDENTIFY NUMERIC SIGNAL COLUMNS
# ============================================================

def get_signal_columns(df):

    numeric_columns = (
        df
        .select_dtypes(
            include=np.number
        )
        .columns
        .tolist()
    )

    signal_columns = []

    for column in numeric_columns:

        name = str(
            column
        ).strip().lower()

        # Ignore obvious index columns
        if name in [
            "index",
            "id",
            "sample",
            "sample_number",
            "unnamed: 0"
        ]:
            continue

        if name.endswith("_id"):
            continue

        if name in [
            "label",
            "class",
            "activity"
        ]:
            continue

        signal_columns.append(
            column
        )

    return signal_columns


# ============================================================
# 8. CALCULATE ROBUST NOISE SCORE
# ============================================================

def calculate_noise_mask(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    valid_values = values.notna().sum()

    if valid_values < MIN_VALID_VALUES:

        return pd.Series(
            False,
            index=series.index
        )

    rolling_median = (
        values
        .rolling(
            WINDOW_SIZE,
            center=True,
            min_periods=5
        )
        .median()
    )

    deviation = (
        values
        -
        rolling_median
    ).abs()

    rolling_mad = (
        deviation
        .rolling(
            WINDOW_SIZE,
            center=True,
            min_periods=5
        )
        .median()
    )

    rolling_mad = (
        rolling_mad
        .replace(
            0,
            np.nan
        )
    )

    robust_score = (
        deviation
        /
        rolling_mad
    )

    noise_mask = (
        robust_score
        >
        NOISE_THRESHOLD
    )

    noise_mask = (
        noise_mask
        .fillna(False)
    )

    return noise_mask


# ============================================================
# 9. STORAGE
# ============================================================

sensor_data = {

    "Muse": [],

    "Motion Watch": []

}

sensor_statistics = {

    "Muse": [],

    "Motion Watch": []

}


# ============================================================
# 10. PROCESS EVERY CSV
# ============================================================

print()
print("=" * 80)
print("PROCESSING DATA")
print("=" * 80)

for counter, file_path in enumerate(
    csv_files,
    start=1
):

    sensor = identify_sensor(
        file_path
    )

    activity = identify_activity(
        file_path
    )

    if sensor not in [
        "Muse",
        "Motion Watch"
    ]:

        print(
            f"[{counter}/{len(csv_files)}] "
            f"Unknown sensor: {file_path}"
        )

        continue

    try:

        df = pd.read_csv(
            file_path,
            low_memory=False
        )

    except Exception as error:

        print(
            f"Could not read: {file_path}"
        )

        print(
            f"Reason: {error}"
        )

        continue

    signal_columns = (
        get_signal_columns(df)
    )

    if len(signal_columns) == 0:

        print(
            f"No numeric signal columns: "
            f"{file_path}"
        )

        continue

    relative_file = (
        file_path.relative_to(
            INPUT_ROOT
        )
    )

    file_noise_total = 0

    file_signal_total = 0

    for column in signal_columns:

        values = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        valid_values = (
            values.notna().sum()
        )

        if valid_values < MIN_VALID_VALUES:
            continue

        noise_mask = (
            calculate_noise_mask(
                values
            )
        )

        noise_count = int(
            noise_mask.sum()
        )

        file_noise_total += (
            noise_count
        )

        file_signal_total += (
            valid_values
        )

        noise_percentage = (
            noise_count
            /
            valid_values
            *
            100
        )

        # ----------------------------------------------------
        # Store channel level information
        # ----------------------------------------------------

        sensor_data[sensor].append({

            "file":
                str(relative_file),

            "activity":
                activity,

            "column":
                str(column),

            "noise_percentage":
                noise_percentage,

            "noise_count":
                noise_count,

            "valid_values":
                int(valid_values)

        })

        # ----------------------------------------------------
        # Store statistics
        # ----------------------------------------------------

        sensor_statistics[sensor].append({

            "file":
                str(relative_file),

            "activity":
                activity,

            "column":
                str(column),

            "noise_percentage":
                noise_percentage

        })

    print(
        f"[{counter}/{len(csv_files)}] "
        f"{sensor:<13} "
        f"{activity:<15} "
        f"{relative_file}"
    )


# ============================================================
# 11. CONVERT TO DATAFRAMES
# ============================================================

muse_df = pd.DataFrame(
    sensor_data["Muse"]
)

motion_df = pd.DataFrame(
    sensor_data["Motion Watch"]
)


# ============================================================
# 12. PRINT BASIC STATISTICS
# ============================================================

print()
print("=" * 80)
print("NOISE ANALYSIS SUMMARY")
print("=" * 80)

print()

print(
    "Muse channel records:",
    len(muse_df)
)

print(
    "Motion Watch channel records:",
    len(motion_df)
)


if len(muse_df) > 0:

    print()
    print("Muse")

    print(
        "Mean potential noise percentage:",
        round(
            muse_df[
                "noise_percentage"
            ].mean(),
            4
        ),
        "%"
    )

    print(
        "Maximum potential noise percentage:",
        round(
            muse_df[
                "noise_percentage"
            ].max(),
            4
        ),
        "%"
    )

    print(
        "Channels with potential noise:",
        int(
            (
                muse_df[
                    "noise_count"
                ]
                >
                0
            ).sum()
        )
    )


if len(motion_df) > 0:

    print()
    print("Motion Watch")

    print(
        "Mean potential noise percentage:",
        round(
            motion_df[
                "noise_percentage"
            ].mean(),
            4
        ),
        "%"
    )

    print(
        "Maximum potential noise percentage:",
        round(
            motion_df[
                "noise_percentage"
            ].max(),
            4
        ),
        "%"
    )

    print(
        "Channels with potential noise:",
        int(
            (
                motion_df[
                    "noise_count"
                ]
                >
                0
            ).sum()
        )
    )


# ============================================================
# 13. SAVE NUMERICAL RESULTS
# ============================================================

muse_df.to_csv(
    OUTPUT_ROOT
    /
    "muse_noise_statistics.csv",
    index=False
)

motion_df.to_csv(
    OUTPUT_ROOT
    /
    "motion_watch_noise_statistics.csv",
    index=False
)


# ============================================================
# 14. CREATE MUSE VISUALIZATION
# ============================================================

def create_sensor_visualization(
    data,
    sensor_name,
    output_file
):

    if data.empty:

        print()
        print(
            f"No usable {sensor_name} data "
            "were found."
        )

        return

    # --------------------------------------------------------
    # Combine repeated channel names
    #
    # Each cell represents the mean potential noise
    # percentage for a file and channel.
    # --------------------------------------------------------

    matrix = (
        data
        .pivot_table(
            index="file",
            columns="column",
            values="noise_percentage",
            aggfunc="mean"
        )
    )

    matrix = matrix.fillna(0)

    # --------------------------------------------------------
    # Limit display width only if there are many channels.
    #
    # The numerical CSV still contains all channels.
    # --------------------------------------------------------

    display_matrix = matrix.copy()

    # --------------------------------------------------------
    # Create figure
    # --------------------------------------------------------

    figure_height = max(
        8,
        min(
            30,
            5
            +
            display_matrix.shape[0]
            *
            0.08
        )
    )

    figure_width = max(
        14,
        min(
            30,
            8
            +
            display_matrix.shape[1]
            *
            0.35
        )
    )

    fig, axes = plt.subplots(
        2,
        1,
        figsize=(
            figure_width,
            figure_height
        ),
        gridspec_kw={
            "height_ratios": [
                4,
                1
            ]
        }
    )

    # --------------------------------------------------------
    # Graph 1
    # Noise heatmap
    # --------------------------------------------------------

    heatmap = axes[0].imshow(
        display_matrix.values,
        aspect="auto",
        interpolation="nearest"
    )

    axes[0].set_title(
        f"{sensor_name}: Potential Noise Across All Files and Channels",
        fontsize=15
    )

    axes[0].set_xlabel(
        "Signal Channel"
    )

    axes[0].set_ylabel(
        "CSV File"
    )

    # Channel labels
    if display_matrix.shape[1] <= 50:

        axes[0].set_xticks(
            np.arange(
                display_matrix.shape[1]
            )
        )

        axes[0].set_xticklabels(
            display_matrix.columns,
            rotation=90,
            fontsize=7
        )

    # Do not print hundreds of file labels
    if display_matrix.shape[0] <= 100:

        axes[0].set_yticks(
            np.arange(
                display_matrix.shape[0]
            )
        )

        axes[0].set_yticklabels(
            [
                Path(name).stem
                for name
                in display_matrix.index
            ],
            fontsize=5
        )

    else:

        y_positions = np.linspace(
            0,
            display_matrix.shape[0] - 1,
            20
        ).astype(int)

        axes[0].set_yticks(
            y_positions
        )

        axes[0].set_yticklabels(
            [
                Path(
                    display_matrix.index[i]
                ).stem
                for i in y_positions
            ],
            fontsize=6
        )

    colorbar = fig.colorbar(
        heatmap,
        ax=axes[0]
    )

    colorbar.set_label(
        "Potential Noise Percentage"
    )

    # --------------------------------------------------------
    # Graph 2
    # Distribution of noise percentage
    # --------------------------------------------------------

    noise_values = (
        data[
            "noise_percentage"
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .dropna()
    )

    axes[1].hist(
        noise_values,
        bins=40
    )

    axes[1].set_title(
        f"{sensor_name}: Distribution of Potential Noise"
    )

    axes[1].set_xlabel(
        "Potential Noise Percentage"
    )

    axes[1].set_ylabel(
        "Number of Signal Channels"
    )

    axes[1].grid(
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=200,
        bbox_inches="tight"
    )

    plt.close()

    print()
    print(
        f"Saved: {output_file}"
    )


# ============================================================
# 15. CREATE THE TWO REQUIRED GRAPHS
# ============================================================

print()
print("=" * 80)
print("CREATING TWO NOISE VISUALIZATIONS")
print("=" * 80)


create_sensor_visualization(
    data=muse_df,
    sensor_name="Muse",
    output_file=(
        OUTPUT_ROOT
        /
        "muse_noise_visualization.png"
    )
)


create_sensor_visualization(
    data=motion_df,
    sensor_name="Motion Watch",
    output_file=(
        OUTPUT_ROOT
        /
        "motion_watch_noise_visualization.png"
    )
)


# ============================================================
# 16. FINAL SUMMARY
# ============================================================

print()
print("=" * 80)
print("COMPLETED")
print("=" * 80)

print()

print(
    "Input:",
    INPUT_ROOT
)

print()

print(
    "Output folder:",
    OUTPUT_ROOT
)

print()

print(
    "Muse graph:"
)

print(
    OUTPUT_ROOT
    /
    "muse_noise_visualization.png"
)

print()

print(
    "Motion Watch graph:"
)

print(
    OUTPUT_ROOT
    /
    "motion_watch_noise_visualization.png"
)

print()

print("=" * 80)
print("IMPORTANT")
print("=" * 80)

print()
print(
    "The original CLEAN_DATA_V4 files were not modified."
)

print(
    "The detected points are potential noise based on "
    "a robust statistical criterion."
)

print(
    "A potential noise point is not automatically confirmed "
    "as an artifact."
)

print(
    "No observations were deleted or corrected by this code."
)

print()
print("=" * 80)
```

```python
# ============================================================
# CLEAN_DATA_V4 -> CLEAN_DATA_V5
#
# Robust data quality checking and cleaning
#
# IMPORTANT:
# Original CLEAN_DATA_V4 is never modified.
#
# Output:
# CLEAN_DATA_V5/
#     clean_data/
#     quality_reports/
#     visualizations/
#
# ============================================================

from pathlib import Path
import csv
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")


# ============================================================
# 1. PATHS
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V4/clean_data"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V5"
)

OUTPUT_DATA = (
    OUTPUT_ROOT / "clean_data"
)

REPORT_ROOT = (
    OUTPUT_ROOT / "quality_reports"
)

FIGURE_ROOT = (
    OUTPUT_ROOT / "visualizations"
)

OUTPUT_DATA.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

FIGURE_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. SETTINGS
# ============================================================

MAX_INTERPOLATION_GAP = 5

Hampel_WINDOW = 11

Hampel_THRESHOLD = 6.0

FLATLINE_WINDOW = 20

FLATLINE_TOLERANCE = 1e-12

MIN_VALID_PERCENT = 50.0


# ============================================================
# 3. EXPECTED DATASET STRUCTURE
# ============================================================

ACTIVITIES = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]

SENSOR_FOLDERS = {
    "clms": "Muse",
    "clmw": "Motion Watch",
    "mums": "Muse",
    "mumw": "Motion Watch",
    "norms": "Muse",
    "normw": "Motion Watch",
    "rdms": "Muse",
    "rdmw": "Motion Watch"
}


# ============================================================
# 4. FIND ONLY EXPERIMENTAL CSV FILES
# ============================================================

def find_experimental_files():

    files = []

    for activity in ACTIVITIES:

        activity_path = (
            INPUT_ROOT / activity
        )

        if not activity_path.exists():
            continue

        for csv_file in activity_path.rglob("*.csv"):

            parts = [
                p.lower()
                for p in csv_file.parts
            ]

            if any(
                folder in SENSOR_FOLDERS
                for folder in parts
            ):

                files.append(csv_file)

    return sorted(files)


csv_files = find_experimental_files()


print("=" * 80)
print("CLEAN_DATA_V4 DATASET AUDIT")
print("=" * 80)

print()
print(
    "Experimental CSV files:",
    len(csv_files)
)

print()


# ============================================================
# 5. SENSOR IDENTIFICATION
# ============================================================

def identify_sensor(file_path):

    for part in file_path.parts:

        key = part.lower()

        if key in SENSOR_FOLDERS:

            return SENSOR_FOLDERS[key]

    name = file_path.name.lower()

    if "ms" in name:
        return "Muse"

    if "mw" in name:
        return "Motion Watch"

    return "Unknown"


def identify_activity(file_path):

    for part in file_path.parts:

        p = part.lower()

        if p in ACTIVITIES:
            return p

    return "Unknown"


# ============================================================
# 6. ROBUST CSV DELIMITER DETECTION
# ============================================================

def detect_delimiter(file_path):

    candidates = [
        ",",
        ";",
        "\t",
        "|"
    ]

    scores = {}

    try:

        with open(
            file_path,
            "r",
            encoding="utf8",
            errors="replace"
        ) as f:

            sample_lines = []

            for _ in range(30):

                line = f.readline()

                if not line:
                    break

                if line.strip():
                    sample_lines.append(
                        line
                    )

        sample = "".join(
            sample_lines
        )

    except Exception:

        return ","


    for delimiter in candidates:

        try:

            rows = list(
                csv.reader(
                    sample.splitlines(),
                    delimiter=delimiter
                )
            )

            widths = [
                len(row)
                for row in rows
                if len(row) > 0
            ]

            if not widths:
                scores[delimiter] = -1
                continue

            most_common = (
                pd.Series(widths)
                .value_counts()
                .iloc[0]
            )

            median_width = np.median(
                widths
            )

            scores[delimiter] = (
                most_common
                *
                median_width
            )

        except Exception:

            scores[delimiter] = -1


    best = max(
        scores,
        key=scores.get
    )

    return best


# ============================================================
# 7. ROBUST CSV LOADER
# ============================================================

def load_csv(file_path):

    delimiter = detect_delimiter(
        file_path
    )

    attempts = [
        {
            "sep": delimiter,
            "engine": "python"
        },
        {
            "sep": ",",
            "engine": "python"
        },
        {
            "sep": ";",
            "engine": "python"
        },
        {
            "sep": "\t",
            "engine": "python"
        }
    ]

    last_error = None

    for settings in attempts:

        try:

            df = pd.read_csv(
                file_path,
                **settings,
                on_bad_lines="warn"
            )

            if df.shape[1] > 1:

                return df, settings["sep"]

        except Exception as error:

            last_error = error


    raise ValueError(
        f"Unable to parse {file_path}. "
        f"Last error: {last_error}"
    )


# ============================================================
# 8. CONVERT NUMERIC COLUMNS
# ============================================================

def numeric_columns(df):

    result = []

    for column in df.columns:

        values = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        valid_ratio = (
            values.notna().mean()
            *
            100
        )

        if valid_ratio >= 80:

            result.append(
                column
            )

    return result


# ============================================================
# 9. CLASSIFY MUSE CHANNELS
# ============================================================

def classify_muse_column(column):

    name = str(
        column
    ).strip().lower()

    # EEG electrodes
    eeg_patterns = [
        r"\btp9\b",
        r"\btp10\b",
        r"\baf7\b",
        r"\baf8\b",
        r"\beeg\b",
        r"tp9",
        r"tp10",
        r"af7",
        r"af8"
    ]

    for pattern in eeg_patterns:

        if re.search(
            pattern,
            name
        ):

            return "EEG"


    # Accelerometer
    if any(
        key in name
        for key in [
            "accelerometer",
            "accel",
            "acc_x",
            "acc_y",
            "acc_z"
        ]
    ):

        return "Accelerometer"


    # Gyroscope
    if any(
        key in name
        for key in [
            "gyro",
            "gyroscope"
        ]
    ):

        return "Gyroscope"


    # Headset status and auxiliary channels
    if any(
        key in name
        for key in [
            "hsi",
            "battery",
            "aux",
            "elements",
            "touching"
        ]
    ):

        return "Other"


    return "Other"


# ============================================================
# 10. HAMPEL OUTLIER DETECTION
# ============================================================

def hampel_mask(
    series,
    window=Hampel_WINDOW,
    threshold=Hampel_THRESHOLD
):

    x = pd.to_numeric(
        series,
        errors="coerce"
    )

    rolling_median = (
        x
        .rolling(
            window=window,
            center=True,
            min_periods=3
        )
        .median()
    )

    deviation = (
        x
        -
        rolling_median
    ).abs()

    rolling_mad = (
        deviation
        .rolling(
            window=window,
            center=True,
            min_periods=3
        )
        .median()
    )

    scale = (
        1.4826
        *
        rolling_mad
    )

    score = (
        deviation
        /
        scale.replace(
            0,
            np.nan
        )
    )

    mask = (
        score
        >
        threshold
    )

    return (
        mask
        .fillna(False)
    )


# ============================================================
# 11. FLATLINE DETECTION
# ============================================================

def flatline_mask(
    series,
    window=FLATLINE_WINDOW
):

    x = pd.to_numeric(
        series,
        errors="coerce"
    )

    rolling_std = (
        x
        .rolling(
            window=window,
            min_periods=window
        )
        .std()
    )

    mask = (
        rolling_std
        <=
        FLATLINE_TOLERANCE
    )

    return (
        mask
        .fillna(False)
    )


# ============================================================
# 12. SHORT GAP INTERPOLATION
# ============================================================

def interpolate_short_gaps(
    series,
    max_gap=MAX_INTERPOLATION_GAP
):

    x = pd.to_numeric(
        series,
        errors="coerce"
    )

    result = (
        x
        .interpolate(
            method="linear",
            limit=max_gap,
            limit_direction="both"
        )
    )

    return result


# ============================================================
# 13. PROCESS ONE FILE
# ============================================================

def process_file(
    file_path
):

    sensor = identify_sensor(
        file_path
    )

    activity = identify_activity(
        file_path
    )

    df, delimiter = load_csv(
        file_path
    )

    original_rows = len(df)

    numeric = numeric_columns(
        df
    )

    # --------------------------------------------------------
    # If no numeric columns were detected, try coercion
    # on every column.
    # --------------------------------------------------------

    if len(numeric) == 0:

        converted = {}

        for column in df.columns:

            values = pd.to_numeric(
                df[column],
                errors="coerce"
            )

            if (
                values.notna().sum()
                >
                0
            ):

                converted[column] = (
                    values
                )

        if converted:

            for column, values in converted.items():

                df[column] = values

            numeric = list(
                converted.keys()
            )


    report = []

    total_missing_before = 0

    total_missing_after = 0

    total_outliers = 0

    total_flatlines = 0

    total_corrected = 0


    # ========================================================
    # PROCESS EACH NUMERIC SIGNAL
    # ========================================================

    for column in numeric:

        values = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        missing_before = int(
            values.isna().sum()
        )

        total_missing_before += (
            missing_before
        )


        # ----------------------------------------------------
        # Muse channel type
        # ----------------------------------------------------

        if sensor == "Muse":

            channel_type = (
                classify_muse_column(
                    column
                )
            )

        else:

            channel_type = (
                "Motion Signal"
            )


        # ----------------------------------------------------
        # Interpolate only short gaps
        # ----------------------------------------------------

        cleaned = (
            interpolate_short_gaps(
                values
            )
        )


        # ----------------------------------------------------
        # Detect local spikes
        # ----------------------------------------------------

        outlier_mask = (
            hampel_mask(
                cleaned
            )
        )

        outlier_count = int(
            outlier_mask.sum()
        )

        total_outliers += (
            outlier_count
        )


        # ----------------------------------------------------
        # Detect flatline
        # ----------------------------------------------------

        flat_mask = (
            flatline_mask(
                cleaned
            )
        )

        flat_count = int(
            flat_mask.sum()
        )

        total_flatlines += (
            flat_count
        )


        # ----------------------------------------------------
        # Correct isolated spikes only
        #
        # Long abnormal segments are NOT automatically
        # deleted.
        # ----------------------------------------------------

        corrected = cleaned.copy()

        isolated_outlier = (
            outlier_mask
            &
            ~outlier_mask.shift(
                1,
                fill_value=False
            )
            &
            ~outlier_mask.shift(
                -1,
                fill_value=False
            )
        )

        correction_count = int(
            isolated_outlier.sum()
        )

        if correction_count > 0:

            corrected[
                isolated_outlier
            ] = np.nan

            corrected = (
                corrected
                .interpolate(
                    method="linear",
                    limit_direction="both"
                )
            )

        total_corrected += (
            correction_count
        )


        # ----------------------------------------------------
        # Remaining missing values
        # ----------------------------------------------------

        missing_after = int(
            corrected.isna().sum()
        )

        total_missing_after += (
            missing_after
        )


        df[column] = corrected


        report.append({

            "sensor":
                sensor,

            "activity":
                activity,

            "file":
                str(
                    file_path.relative_to(
                        INPUT_ROOT
                    )
                ),

            "column":
                str(column),

            "channel_type":
                channel_type,

            "rows":
                original_rows,

            "missing_before":
                missing_before,

            "missing_after":
                missing_after,

            "outliers":
                outlier_count,

            "flatline_points":
                flat_count,

            "isolated_outliers_corrected":
                correction_count,

            "valid_percent":
                round(
                    corrected.notna().mean()
                    *
                    100,
                    4
                )

        })


    return (
        df,
        report,
        {
            "sensor":
                sensor,

            "activity":
                activity,

            "file":
                str(
                    file_path.relative_to(
                        INPUT_ROOT
                    )
                ),

            "rows":
                original_rows,

            "columns":
                len(df.columns),

            "numeric_signals":
                len(numeric),

            "missing_before":
                total_missing_before,

            "missing_after":
                total_missing_after,

            "outliers":
                total_outliers,

            "flatline_points":
                total_flatlines,

            "corrected":
                total_corrected,

            "delimiter":
                delimiter

        }
    )


# ============================================================
# 14. PROCESS ALL 480 FILES
# ============================================================

all_channel_reports = []

all_file_reports = []

failed_files = []


print()
print("=" * 80)
print("PROCESSING ALL EXPERIMENTAL FILES")
print("=" * 80)


for number, file_path in enumerate(
    csv_files,
    start=1
):

    try:

        (
            cleaned_df,
            channel_report,
            file_report
        ) = process_file(
            file_path
        )


        # ----------------------------------------------------
        # Preserve folder structure
        # ----------------------------------------------------

        relative_path = (
            file_path.relative_to(
                INPUT_ROOT
            )
        )

        output_file = (
            OUTPUT_DATA
            /
            relative_path
        )

        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # ----------------------------------------------------
        # Save cleaned file
        # ----------------------------------------------------

        cleaned_df.to_csv(
            output_file,
            index=False
        )


        all_channel_reports.extend(
            channel_report
        )

        all_file_reports.append(
            file_report
        )


        print(
            f"[{number}/{len(csv_files)}] "
            f"{file_report['sensor']:<13} "
            f"{file_report['activity']:<15} "
            f"{file_path.name:<15} "
            f"Signals: "
            f"{file_report['numeric_signals']}"
        )


    except Exception as error:

        failed_files.append({

            "file":
                str(
                    file_path
                    .relative_to(
                        INPUT_ROOT
                    )
                ),

            "error":
                str(error)

        })

        print(
            f"[{number}/{len(csv_files)}] FAILED "
            f"{file_path.name}: {error}"
        )


# ============================================================
# 15. SAVE REPORTS
# ============================================================

channel_df = pd.DataFrame(
    all_channel_reports
)

file_df = pd.DataFrame(
    all_file_reports
)

failed_df = pd.DataFrame(
    failed_files
)


channel_df.to_csv(
    REPORT_ROOT
    /
    "channel_quality_report.csv",
    index=False
)

file_df.to_csv(
    REPORT_ROOT
    /
    "file_quality_report.csv",
    index=False
)

failed_df.to_csv(
    REPORT_ROOT
    /
    "failed_files.csv",
    index=False
)


# ============================================================
# 16. DATASET SUMMARY
# ============================================================

print()
print("=" * 80)
print("FINAL DATA QUALITY SUMMARY")
print("=" * 80)

print()

print(
    "Input files:",
    len(csv_files)
)

print(
    "Successfully processed:",
    len(file_df)
)

print(
    "Failed:",
    len(failed_df)
)

if len(file_df) > 0:

    print()

    print(
        "Missing values before:",
        int(
            file_df[
                "missing_before"
            ].sum()
        )
    )

    print(
        "Missing values after:",
        int(
            file_df[
                "missing_after"
            ].sum()
        )
    )

    print(
        "Potential outliers:",
        int(
            file_df[
                "outliers"
            ].sum()
        )
    )

    print(
        "Flatline points:",
        int(
            file_df[
                "flatline_points"
            ].sum()
        )
    )

    print(
        "Isolated outliers corrected:",
        int(
            file_df[
                "corrected"
            ].sum()
        )
    )


# ============================================================
# 17. SENSOR SUMMARY
# ============================================================

if len(file_df) > 0:

    sensor_summary = (
        file_df
        .groupby(
            "sensor"
        )
        .agg(
            files=(
                "file",
                "count"
            ),
            missing_before=(
                "missing_before",
                "sum"
            ),
            missing_after=(
                "missing_after",
                "sum"
            ),
            outliers=(
                "outliers",
                "sum"
            ),
            flatline_points=(
                "flatline_points",
                "sum"
            ),
            corrected=(
                "corrected",
                "sum"
            )
        )
        .reset_index()
    )

    sensor_summary.to_csv(
        REPORT_ROOT
        /
        "sensor_quality_summary.csv",
        index=False
    )

    print()
    print(
        sensor_summary
        .to_string(
            index=False
        )
    )


# ============================================================
# 18. ACTIVITY SUMMARY
# ============================================================

if len(file_df) > 0:

    activity_summary = (
        file_df
        .groupby(
            [
                "sensor",
                "activity"
            ]
        )
        .agg(
            files=(
                "file",
                "count"
            ),
            missing_before=(
                "missing_before",
                "sum"
            ),
            missing_after=(
                "missing_after",
                "sum"
            ),
            outliers=(
                "outliers",
                "sum"
            ),
            flatline_points=(
                "flatline_points",
                "sum"
            ),
            corrected=(
                "corrected",
                "sum"
            )
        )
        .reset_index()
    )

    activity_summary.to_csv(
        REPORT_ROOT
        /
        "activity_quality_summary.csv",
        index=False
    )


# ============================================================
# 19. MUSE CHANNEL SUMMARY
# ============================================================

if len(channel_df) > 0:

    muse_channels = (
        channel_df[
            channel_df[
                "sensor"
            ]
            ==
            "Muse"
        ]
    )

    if len(muse_channels) > 0:

        muse_summary = (
            muse_channels
            .groupby(
                [
                    "column",
                    "channel_type"
                ]
            )
            .agg(
                files=(
                    "file",
                    "count"
                ),
                mean_missing=(
                    "missing_after",
                    "mean"
                ),
                total_outliers=(
                    "outliers",
                    "sum"
                ),
                total_flatline=(
                    "flatline_points",
                    "sum"
                ),
                total_corrected=(
                    "isolated_outliers_corrected",
                    "sum"
                )
            )
            .reset_index()
        )

        muse_summary.to_csv(
            REPORT_ROOT
            /
            "muse_channel_summary.csv",
            index=False
        )


# ============================================================
# 20. FINAL STATUS
# ============================================================

print()
print("=" * 80)
print("PROCESS FINISHED")
print("=" * 80)

print()

print(
    "New dataset:",
    OUTPUT_ROOT
)

print()

print(
    "Cleaned data:",
    OUTPUT_DATA
)

print()

print(
    "Quality reports:",
    REPORT_ROOT
)

print()

print(
    "Visualizations:",
    FIGURE_ROOT
)

print()

print(
    "Original CLEAN_DATA_V4 was not modified."
)

print("=" * 80)
```

```python
from pathlib import Path
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

BASE_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA"
)

INPUT_DIR = BASE_DIR / "CLEAN_DATA_V5"
OUTPUT_DIR = BASE_DIR / "CLEAN_DATA_V6"

INPUT_DATA_DIR = INPUT_DIR / "clean_data"
OUTPUT_DATA_DIR = OUTPUT_DIR / "clean_data"

REPORT_DIR = OUTPUT_DIR / "quality_reports"
VIS_DIR = OUTPUT_DIR / "visualizations"


# ============================================================
# 2. COLUMNS TO REMOVE FROM MUSE DATA
# ============================================================

COLUMNS_TO_REMOVE = [
    "HeadBandOn",
    "HSI_TP9",
    "HSI_AF7",
    "HSI_AF8",
    "HSI_TP10",
    "Battery"
]


# ============================================================
# 3. OUTLIER METHOD
# ============================================================

# IQR based outlier detection
#
# Lower boundary:
# Q1 - 1.5 * IQR
#
# Upper boundary:
# Q3 + 1.5 * IQR
#
# Any value outside these boundaries is considered an outlier.


IQR_MULTIPLIER = 1.5


# ============================================================
# 4. CREATE V6 DIRECTORIES
# ============================================================

if OUTPUT_DIR.exists():
    print("Existing CLEAN_DATA_V6 found.")
    print("It will be replaced.")

    shutil.rmtree(OUTPUT_DIR)


OUTPUT_DATA_DIR.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

VIS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 5. FIND ALL CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_DATA_DIR.rglob("*.csv")
)

print("=" * 80)
print("CLEAN_DATA_V5 TO CLEAN_DATA_V6")
print("=" * 80)

print(
    "Input directory:",
    INPUT_DATA_DIR
)

print(
    "Total CSV files:",
    len(csv_files)
)

print()


# ============================================================
# 6. STORAGE FOR REPORTS
# ============================================================

file_report = []

column_report = []

total_v5_outliers = 0
total_v6_outliers = 0
total_replaced = 0

muse_files = 0
motion_files = 0

removed_column_count = 0


# ============================================================
# 7. PROCESS EACH CSV
# ============================================================

for file_number, input_file in enumerate(
    csv_files,
    start=1
):

    # Preserve the same relative directory structure
    relative_path = input_file.relative_to(
        INPUT_DATA_DIR
    )

    output_file = (
        OUTPUT_DATA_DIR /
        relative_path
    )

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    # --------------------------------------------------------
    # Read file
    # --------------------------------------------------------

    try:

        df = pd.read_csv(
            input_file
        )

    except Exception as error:

        print(
            f"[ERROR] {input_file}: {error}"
        )

        continue


    # --------------------------------------------------------
    # Identify sensor
    # --------------------------------------------------------

    path_text = str(
        relative_path
    ).lower()

    is_muse = (
        "/clms/" in path_text
        or "/mums/" in path_text
        or "/norms/" in path_text
        or "/rdms/" in path_text
    )

    is_motion = (
        "/clmw/" in path_text
        or "/mumw/" in path_text
        or "/normw/" in path_text
        or "/rdmw/" in path_text
    )


    if is_muse:

        sensor = "Muse"
        muse_files += 1

    elif is_motion:

        sensor = "Motion Watch"
        motion_files += 1

    else:

        sensor = "Unknown"


    # ========================================================
    # MOTION WATCH
    # ========================================================

    if sensor != "Muse":

        # Copy unchanged
        df.to_csv(
            output_file,
            index=False
        )

        file_report.append({

            "sensor":
                sensor,

            "activity":
                relative_path.parts[0]
                if len(relative_path.parts) > 0
                else "",

            "file":
                str(relative_path),

            "rows":
                len(df),

            "columns_before":
                len(df.columns),

            "columns_after":
                len(df.columns),

            "columns_removed":
                0,

            "outliers_before":
                0,

            "outliers_after":
                0,

            "replaced_values":
                0
        })

        print(
            f"[{file_number}/{len(csv_files)}] "
            f"{sensor:<15} "
            f"{relative_path}"
        )

        continue


    # ========================================================
    # MUSE PROCESSING
    # ========================================================

    columns_before = len(
        df.columns
    )


    # --------------------------------------------------------
    # Remove requested columns
    # --------------------------------------------------------

    existing_columns_to_remove = [
        column
        for column in COLUMNS_TO_REMOVE
        if column in df.columns
    ]

    df = df.drop(
        columns=existing_columns_to_remove
    )

    columns_removed = len(
        existing_columns_to_remove
    )

    removed_column_count += (
        columns_removed
    )


    # --------------------------------------------------------
    # Find numeric columns
    # --------------------------------------------------------

    numeric_columns = (
        df.select_dtypes(
            include=np.number
        )
        .columns
        .tolist()
    )


    file_outliers_before = 0
    file_outliers_after = 0
    file_replaced = 0


    # ========================================================
    # OUTLIER DETECTION PER COLUMN
    # ========================================================

    for column in numeric_columns:

        series = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        valid_values = series.dropna()

        if len(valid_values) == 0:

            continue


        # ----------------------------------------------------
        # Calculate IQR
        # ----------------------------------------------------

        q1 = valid_values.quantile(
            0.25
        )

        q3 = valid_values.quantile(
            0.75
        )

        iqr = q3 - q1

        lower_bound = (
            q1
            -
            IQR_MULTIPLIER * iqr
        )

        upper_bound = (
            q3
            +
            IQR_MULTIPLIER * iqr
        )


        # ----------------------------------------------------
        # Detect outliers
        # ----------------------------------------------------

        outlier_mask = (
            (series < lower_bound)
            |
            (series > upper_bound)
        )

        outlier_mask = (
            outlier_mask
            &
            series.notna()
        )


        outlier_count = int(
            outlier_mask.sum()
        )


        file_outliers_before += (
            outlier_count
        )

        total_v5_outliers += (
            outlier_count
        )


        # ----------------------------------------------------
        # Calculate entire column mode
        # ----------------------------------------------------

        mode_values = (
            valid_values
            .mode()
        )


        if len(mode_values) == 0:

            # No mode available
            # Leave the column unchanged

            replacement_value = np.nan

        else:

            replacement_value = (
                mode_values.iloc[0]
            )


        # ----------------------------------------------------
        # Replace outliers by column mode
        # ----------------------------------------------------

        if outlier_count > 0:

            df.loc[
                outlier_mask,
                column
            ] = replacement_value

            file_replaced += (
                outlier_count
            )

            total_replaced += (
                outlier_count
            )


        # ----------------------------------------------------
        # Recalculate outliers after replacement
        # ----------------------------------------------------

        new_series = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        new_valid = (
            new_series.dropna()
        )


        if len(new_valid) > 0:

            new_q1 = (
                new_valid.quantile(0.25)
            )

            new_q3 = (
                new_valid.quantile(0.75)
            )

            new_iqr = (
                new_q3 - new_q1
            )

            new_lower = (
                new_q1
                -
                IQR_MULTIPLIER * new_iqr
            )

            new_upper = (
                new_q3
                +
                IQR_MULTIPLIER * new_iqr
            )

            new_outlier_mask = (
                (new_series < new_lower)
                |
                (new_series > new_upper)
            )

            new_outlier_mask = (
                new_outlier_mask
                &
                new_series.notna()
            )

            outlier_after = int(
                new_outlier_mask.sum()
            )

        else:

            outlier_after = 0


        file_outliers_after += (
            outlier_after
        )

        total_v6_outliers += (
            outlier_after
        )


        # ----------------------------------------------------
        # Column report
        # ----------------------------------------------------

        column_report.append({

            "sensor":
                "Muse",

            "activity":
                relative_path.parts[0],

            "file":
                str(relative_path),

            "column":
                column,

            "rows":
                len(df),

            "q1":
                q1,

            "q3":
                q3,

            "iqr":
                iqr,

            "lower_bound":
                lower_bound,

            "upper_bound":
                upper_bound,

            "outliers_before":
                outlier_count,

            "replacement_mode":
                replacement_value,

            "replaced":
                outlier_count,

            "outliers_after":
                outlier_after
        })


    # ========================================================
    # SAVE MUSE FILE
    # ========================================================

    df.to_csv(
        output_file,
        index=False
    )


    # ========================================================
    # FILE REPORT
    # ========================================================

    file_report.append({

        "sensor":
            "Muse",

        "activity":
            relative_path.parts[0],

        "file":
            str(relative_path),

        "rows":
            len(df),

        "columns_before":
            columns_before,

        "columns_after":
            len(df.columns),

        "columns_removed":
            columns_removed,

        "outliers_before":
            file_outliers_before,

        "outliers_after":
            file_outliers_after,

        "replaced_values":
            file_replaced
    })


    print(
        f"[{file_number}/{len(csv_files)}] "
        f"Muse            "
        f"{relative_path} "
        f"| Outliers: {file_outliers_before} "
        f"| Replaced: {file_replaced} "
        f"| Remaining: {file_outliers_after}"
    )


# ============================================================
# 8. SAVE REPORTS
# ============================================================

file_report_df = pd.DataFrame(
    file_report
)

column_report_df = pd.DataFrame(
    column_report
)


file_report_path = (
    REPORT_DIR /
    "v6_file_quality_report.csv"
)

column_report_path = (
    REPORT_DIR /
    "v6_column_outlier_report.csv"
)


file_report_df.to_csv(
    file_report_path,
    index=False
)

column_report_df.to_csv(
    column_report_path,
    index=False
)


# ============================================================
# 9. CREATE SUMMARY
# ============================================================

summary = pd.DataFrame({

    "metric": [

        "Input CSV files",
        "Muse CSV files",
        "Motion Watch CSV files",
        "Columns requested for removal",
        "Total V5 outliers detected",
        "Total outliers replaced",
        "Total V6 outliers remaining"
    ],

    "value": [

        len(csv_files),
        muse_files,
        motion_files,
        len(COLUMNS_TO_REMOVE),
        total_v5_outliers,
        total_replaced,
        total_v6_outliers
    ]
})


summary_path = (
    REPORT_DIR /
    "v6_summary.csv"
)

summary.to_csv(
    summary_path,
    index=False
)


# ============================================================
# 10. CREATE SINGLE OUTLIER VISUALIZATION
# ============================================================

# Only Muse files are relevant for this comparison.

plot_df = file_report_df[
    file_report_df["sensor"] == "Muse"
].copy()


# ------------------------------------------------------------
# Create numeric file index
# ------------------------------------------------------------

plot_df = plot_df.reset_index(
    drop=True
)

plot_df["file_index"] = (
    np.arange(
        len(plot_df)
    )
)


# ------------------------------------------------------------
# Create figure
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(16, 8)
)


# ------------------------------------------------------------
# V5 outlier datapoints
# ------------------------------------------------------------

ax.scatter(
    plot_df["file_index"],
    plot_df["outliers_before"],
    s=18,
    alpha=0.65,
    label="V5 outliers"
)


# ------------------------------------------------------------
# V6 outlier datapoints
# ------------------------------------------------------------

ax.scatter(
    plot_df["file_index"],
    plot_df["outliers_after"],
    s=18,
    alpha=0.65,
    label="V6 remaining outliers"
)


# ------------------------------------------------------------
# Connect before and after
# ------------------------------------------------------------

for _, row in plot_df.iterrows():

    ax.plot(
        [
            row["file_index"],
            row["file_index"]
        ],
        [
            row["outliers_before"],
            row["outliers_after"]
        ],
        linewidth=0.5,
        alpha=0.25
    )


ax.set_title(
    "Muse Outlier Structure: CLEAN_DATA_V5 vs CLEAN_DATA_V6",
    fontsize=15
)

ax.set_xlabel(
    "Muse CSV File Index"
)

ax.set_ylabel(
    "Number of Detected Outlier Data Points"
)

ax.legend()

ax.grid(
    alpha=0.25
)

plt.tight_layout()


plot_path = (
    VIS_DIR /
    "v6_outlier_structure.png"
)

plt.savefig(
    plot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# ============================================================
# 11. FINAL SUMMARY
# ============================================================

print()
print("=" * 80)
print("CLEAN_DATA_V6 PROCESSING COMPLETED")
print("=" * 80)

print(
    "Total CSV files:",
    len(csv_files)
)

print(
    "Muse files:",
    muse_files
)

print(
    "Motion Watch files:",
    motion_files
)

print(
    "Columns removed from each Muse file:",
    ", ".join(COLUMNS_TO_REMOVE)
)

print(
    "Total V5 outliers:",
    total_v5_outliers
)

print(
    "Total outliers replaced:",
    total_replaced
)

print(
    "Total V6 outliers remaining:",
    total_v6_outliers
)

print()
print(
    "V6 dataset:",
    OUTPUT_DIR
)

print(
    "V6 cleaned data:",
    OUTPUT_DATA_DIR
)

print(
    "Quality reports:",
    REPORT_DIR
)

print(
    "Visualization:",
    plot_path
)

print("=" * 80)
```

```python
# ============================================================
# CLEAN_DATA_V6 -> CLEAN_DATA_V7
#
# Motion Watch outlier detection and correction
#
# Processing:
#   1. Read CLEAN_DATA_V6
#   2. Muse files are copied unchanged
#   3. Motion Watch files are processed
#   4. Detect outliers using IQR
#   5. Replace outliers with entire column mode
#   6. Recheck remaining outliers
#   7. Preserve timestamps
#   8. Preserve row count
#   9. Preserve folder structure
#  10. Generate detailed reports
#  11. Generate one V6 versus V7 graph
# ============================================================

from pathlib import Path
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. CONFIGURATION
# ============================================================

# Your current V6 dataset
INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V6"
)

# V6 may contain a clean_data folder
POSSIBLE_INPUTS = [
    INPUT_ROOT / "clean_data",
    INPUT_ROOT
]

# Output dataset
OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V7"
)

OUTPUT_DATA = OUTPUT_ROOT / "clean_data"
REPORT_DIR = OUTPUT_ROOT / "reports"
VIS_DIR = OUTPUT_ROOT / "visualizations"

# Motion Watch folder names
MOTION_FOLDERS = {
    "clmw",
    "mumw",
    "normw",
    "rdmw"
}

# IQR multiplier
IQR_MULTIPLIER = 1.5


# ============================================================
# 2. FIND INPUT DATASET
# ============================================================

INPUT_DATA = None

for possible_path in POSSIBLE_INPUTS:

    if possible_path.exists():

        csv_count = len(
            list(possible_path.rglob("*.csv"))
        )

        if csv_count > 0:
            INPUT_DATA = possible_path
            break


if INPUT_DATA is None:

    raise FileNotFoundError(
        "No CSV files were found inside CLEAN_DATA_V6."
    )


# ============================================================
# 3. PREPARE OUTPUT DIRECTORIES
# ============================================================

OUTPUT_DATA.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

VIS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 4. FIND ALL CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_DATA.rglob("*.csv")
)

print("=" * 90)
print("CLEAN_DATA_V6 TO CLEAN_DATA_V7")
print("=" * 90)

print()
print("Input dataset:")
print(INPUT_DATA)

print()
print("Total CSV files found:")
print(len(csv_files))

print()


# ============================================================
# 5. STORAGE FOR REPORTS
# ============================================================

file_results = []

column_results = []

total_files = len(csv_files)

motion_file_count = 0
muse_file_count = 0

total_v6_outliers = 0
total_replaced = 0
total_v7_outliers = 0

total_rows_v6 = 0
total_rows_v7 = 0


# ============================================================
# 6. PROCESS EACH CSV FILE
# ============================================================

for file_number, input_file in enumerate(
    csv_files,
    start=1
):

    relative_path = input_file.relative_to(
        INPUT_DATA
    )

    output_file = OUTPUT_DATA / relative_path

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # Identify folder type
    # --------------------------------------------------------

    folder_names = {
        part.lower()
        for part in relative_path.parts
    }

    is_motion_watch = bool(
        folder_names.intersection(
            MOTION_FOLDERS
        )
    )


    # ========================================================
    # MUSE FILE
    # ========================================================

    if not is_motion_watch:

        muse_file_count += 1

        shutil.copy2(
            input_file,
            output_file
        )

        file_results.append({

            "sensor": "Muse",

            "file": str(
                relative_path
            ),

            "rows_v6": np.nan,

            "rows_v7": np.nan,

            "columns": np.nan,

            "numeric_columns": np.nan,

            "outliers_v6": 0,

            "replaced": 0,

            "outliers_v7": 0

        })

        print(
            f"[{file_number}/{total_files}] "
            f"Muse copied: {relative_path}"
        )

        continue


    # ========================================================
    # MOTION WATCH FILE
    # ========================================================

    motion_file_count += 1


    try:

        df = pd.read_csv(
            input_file,
            low_memory=False
        )

    except Exception as error:

        print()
        print(
            f"ERROR reading: {relative_path}"
        )
        print(error)

        file_results.append({

            "sensor": "Motion Watch",

            "file": str(
                relative_path
            ),

            "rows_v6": np.nan,

            "rows_v7": np.nan,

            "columns": np.nan,

            "numeric_columns": np.nan,

            "outliers_v6": np.nan,

            "replaced": np.nan,

            "outliers_v7": np.nan

        })

        continue


    rows_before = len(df)

    total_rows_v6 += rows_before


    numeric_columns = (
        df.select_dtypes(
            include=np.number
        ).columns.tolist()
    )


    file_v6_outliers = 0
    file_replaced = 0
    file_v7_outliers = 0


    # ========================================================
    # PROCESS EVERY NUMERIC COLUMN
    # ========================================================

    for column in numeric_columns:


        # ----------------------------------------------------
        # Convert to numeric safely
        # ----------------------------------------------------

        series = pd.to_numeric(
            df[column],
            errors="coerce"
        )


        valid_values = series.dropna()


        if len(valid_values) == 0:

            continue


        # ----------------------------------------------------
        # Calculate Q1 and Q3
        # ----------------------------------------------------

        q1 = valid_values.quantile(
            0.25
        )

        q3 = valid_values.quantile(
            0.75
        )

        iqr = q3 - q1


        # ----------------------------------------------------
        # If IQR is zero
        #
        # There is no useful spread for IQR detection.
        # Do not modify the column.
        # ----------------------------------------------------

        if iqr == 0 or not np.isfinite(iqr):

            column_results.append({

                "sensor": "Motion Watch",

                "file": str(relative_path),

                "column": column,

                "rows": rows_before,

                "q1": q1,

                "q3": q3,

                "iqr": iqr,

                "lower_limit": np.nan,

                "upper_limit": np.nan,

                "mode": valid_values.mode().iloc[0]
                if not valid_values.mode().empty
                else np.nan,

                "v6_outliers": 0,

                "replaced": 0,

                "v7_outliers": 0

            })

            continue


        # ----------------------------------------------------
        # Calculate IQR limits
        # ----------------------------------------------------

        lower_limit = (
            q1
            - IQR_MULTIPLIER * iqr
        )

        upper_limit = (
            q3
            + IQR_MULTIPLIER * iqr
        )


        # ----------------------------------------------------
        # Detect V6 outliers
        # ----------------------------------------------------

        outlier_mask = (

            (
                series < lower_limit
            )

            |

            (
                series > upper_limit
            )

        ) & series.notna()


        v6_count = int(
            outlier_mask.sum()
        )


        # ----------------------------------------------------
        # Calculate entire column mode
        # ----------------------------------------------------

        mode_values = valid_values.mode()


        if len(mode_values) == 0:

            mode_value = np.nan

        else:

            mode_value = mode_values.iloc[0]


        # ----------------------------------------------------
        # Replace detected outliers
        # ----------------------------------------------------

        if v6_count > 0:

            df.loc[
                outlier_mask,
                column
            ] = mode_value


        # ----------------------------------------------------
        # Recalculate outliers after replacement
        # ----------------------------------------------------

        new_series = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        new_valid = new_series.dropna()


        if len(new_valid) > 0:

            new_q1 = new_valid.quantile(
                0.25
            )

            new_q3 = new_valid.quantile(
                0.75
            )

            new_iqr = (
                new_q3 - new_q1
            )


            if (
                new_iqr > 0
                and np.isfinite(new_iqr)
            ):

                new_lower = (
                    new_q1
                    - IQR_MULTIPLIER
                    * new_iqr
                )

                new_upper = (
                    new_q3
                    + IQR_MULTIPLIER
                    * new_iqr
                )


                new_outlier_mask = (

                    (
                        new_series
                        < new_lower
                    )

                    |

                    (
                        new_series
                        > new_upper
                    )

                ) & new_series.notna()


                v7_count = int(
                    new_outlier_mask.sum()
                )

            else:

                v7_count = 0

        else:

            v7_count = 0


        # ----------------------------------------------------
        # Update counters
        # ----------------------------------------------------

        file_v6_outliers += v6_count

        file_replaced += v6_count

        file_v7_outliers += v7_count


        total_v6_outliers += v6_count

        total_replaced += v6_count

        total_v7_outliers += v7_count


        # ----------------------------------------------------
        # Column report
        # ----------------------------------------------------

        column_results.append({

            "sensor": "Motion Watch",

            "file": str(relative_path),

            "column": column,

            "rows": rows_before,

            "q1": q1,

            "q3": q3,

            "iqr": iqr,

            "lower_limit": lower_limit,

            "upper_limit": upper_limit,

            "mode": mode_value,

            "v6_outliers": v6_count,

            "replaced": v6_count,

            "v7_outliers": v7_count

        })


    # ========================================================
    # SAVE V7 FILE
    # ========================================================

    rows_after = len(df)

    total_rows_v7 += rows_after


    # --------------------------------------------------------
    # Verify row count
    # --------------------------------------------------------

    if rows_before != rows_after:

        print(
            "WARNING: Row count changed:",
            relative_path
        )


    df.to_csv(
        output_file,
        index=False
    )


    # ========================================================
    # FILE REPORT
    # ========================================================

    activity = (
        relative_path.parts[0]
        if len(relative_path.parts) > 0
        else ""
    )


    file_results.append({

        "sensor": "Motion Watch",

        "activity": activity,

        "file": str(relative_path),

        "rows_v6": rows_before,

        "rows_v7": rows_after,

        "columns": len(df.columns),

        "numeric_columns": len(
            numeric_columns
        ),

        "outliers_v6": file_v6_outliers,

        "replaced": file_replaced,

        "outliers_v7": file_v7_outliers,

        "row_count_preserved":
            rows_before == rows_after

    })


    print(
        f"[{file_number}/{total_files}] "
        f"Motion Watch | "
        f"V6 outliers: {file_v6_outliers:,} | "
        f"Replaced: {file_replaced:,} | "
        f"V7 remaining: {file_v7_outliers:,} | "
        f"{relative_path}"
    )


# ============================================================
# 7. CREATE DATAFRAMES
# ============================================================

file_report_df = pd.DataFrame(
    file_results
)

column_report_df = pd.DataFrame(
    column_results
)


# ============================================================
# 8. SAVE REPORTS
# ============================================================

file_report_path = (
    REPORT_DIR
    / "V7_file_quality_report.csv"
)

column_report_path = (
    REPORT_DIR
    / "V7_motion_watch_column_report.csv"
)

file_report_df.to_csv(
    file_report_path,
    index=False
)

column_report_df.to_csv(
    column_report_path,
    index=False
)


# ============================================================
# 9. CALCULATE REDUCTION
# ============================================================

if total_v6_outliers > 0:

    reduction_percent = (

        (
            total_v6_outliers
            - total_v7_outliers
        )

        /

        total_v6_outliers

        * 100

    )

else:

    reduction_percent = 0


# ============================================================
# 10. SUMMARY REPORT
# ============================================================

summary_df = pd.DataFrame({

    "metric": [

        "Total CSV files",

        "Muse files",

        "Motion Watch files",

        "V6 total Motion Watch outliers",

        "Total replaced with column mode",

        "V7 remaining Motion Watch outliers",

        "Outlier reduction percentage",

        "V6 total rows",

        "V7 total rows"

    ],

    "value": [

        total_files,

        muse_file_count,

        motion_file_count,

        total_v6_outliers,

        total_replaced,

        total_v7_outliers,

        reduction_percent,

        total_rows_v6,

        total_rows_v7

    ]

})


summary_path = (
    REPORT_DIR
    / "V7_summary.csv"
)

summary_df.to_csv(
    summary_path,
    index=False
)


# ============================================================
# 11. CREATE SINGLE V6 VS V7 OUTLIER GRAPH
# ============================================================

motion_report = file_report_df[
    file_report_df["sensor"]
    == "Motion Watch"
].copy()


motion_report = (
    motion_report
    .reset_index(drop=True)
)


motion_report[
    "file_index"
] = np.arange(
    len(motion_report)
)


# ------------------------------------------------------------
# Create graph
# ------------------------------------------------------------

plt.figure(
    figsize=(20, 10)
)


plt.scatter(

    motion_report["file_index"],

    motion_report["outliers_v6"],

    s=35,

    alpha=0.70,

    label="V6 detected outliers"

)


plt.scatter(

    motion_report["file_index"],

    motion_report["outliers_v7"],

    s=35,

    alpha=0.70,

    label="V7 remaining outliers"

)


# ------------------------------------------------------------
# Draw connection between V6 and V7
# ------------------------------------------------------------

for _, row in motion_report.iterrows():

    plt.plot(

        [
            row["file_index"],
            row["file_index"]
        ],

        [
            row["outliers_v6"],
            row["outliers_v7"]
        ],

        alpha=0.25,

        linewidth=0.7

    )


plt.title(
    "Motion Watch Outlier Structure: CLEAN_DATA_V6 vs CLEAN_DATA_V7",
    fontsize=17
)

plt.xlabel(
    "Motion Watch CSV File Index",
    fontsize=13
)

plt.ylabel(
    "Number of Detected Outlier Data Points",
    fontsize=13
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


graph_path = (
    VIS_DIR
    / "V6_V7_motion_watch_outlier_structure.png"
)


plt.savefig(
    graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. VERIFY V7 FILE COUNT
# ============================================================

v7_files = sorted(
    OUTPUT_DATA.rglob("*.csv")
)


# ============================================================
# 13. VERIFY ROW COUNTS
# ============================================================

row_count_errors = file_report_df[
    file_report_df.get(
        "row_count_preserved",
        pd.Series(dtype=bool)
    ) == False
]


# ============================================================
# 14. FINAL OUTPUT
# ============================================================

print()
print("=" * 90)
print("CLEAN_DATA_V7 PROCESSING COMPLETED")
print("=" * 90)

print()
print(
    "Input CSV files:",
    total_files
)

print(
    "V7 CSV files:",
    len(v7_files)
)

print()
print(
    "Muse files:",
    muse_file_count
)

print(
    "Motion Watch files:",
    motion_file_count
)

print()
print(
    "V6 Motion Watch outliers:",
    f"{total_v6_outliers:,}"
)

print(
    "Outliers replaced:",
    f"{total_replaced:,}"
)

print(
    "V7 remaining outliers:",
    f"{total_v7_outliers:,}"
)

print()
print(
    "Outlier reduction:",
    f"{reduction_percent:.2f}%"
)

print()
print(
    "V6 total rows:",
    f"{total_rows_v6:,}"
)

print(
    "V7 total rows:",
    f"{total_rows_v7:,}"
)

print()
print(
    "Row count verification errors:",
    len(row_count_errors)
)

print()
print("V7 dataset:")
print(OUTPUT_DATA)

print()
print("Summary report:")
print(summary_path)

print()
print("File report:")
print(file_report_path)

print()
print("Column report:")
print(column_report_path)

print()
print("Visualization:")
print(graph_path)

print()
print("=" * 90)
print("DONE")
print("=" * 90)


# ============================================================
# 15. DISPLAY SUMMARY
# ============================================================

display(summary_df)
```

```python
# ============================================================
# CLEAN_DATA_V7 -> CLEAN_DATA_V8
#
# More aggressive outlier reduction for:
#     1. Muse
#     2. Motion Watch
#
# Main procedure:
#
# V7
#  |
#  +--> Detect initial outliers
#  |
#  +--> Iterative IQR detection
#  |
#  +--> Robust MAD detection
#  |
#  +--> Replace detected outliers
#  |
#  +--> Recalculate
#  |
#  +--> Repeat until stable
#  |
#  +--> Save CLEAN_DATA_V8
#
# The original V7 dataset is NOT modified.
#
# ============================================================


from pathlib import Path
import shutil
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


warnings.filterwarnings("ignore")


# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V7"
)

INPUT_DATA = INPUT_ROOT / "clean_data"


OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V8"
)

OUTPUT_DATA = OUTPUT_ROOT / "clean_data"

REPORT_DIR = OUTPUT_ROOT / "reports"

VIS_DIR = OUTPUT_ROOT / "visualizations"


# ============================================================
# 2. PROCESSING PARAMETERS
# ============================================================

# Standard IQR multiplier
IQR_MULTIPLIER = 1.5

# More aggressive second stage
MAD_THRESHOLD = 3.5

# Maximum number of iterative passes
MAX_ITERATIONS = 10

# Minimum number of valid values required
MIN_VALID_VALUES = 10

# Minimum number of unique values
MIN_UNIQUE_VALUES = 2


# ============================================================
# 3. MOTION WATCH FOLDERS
# ============================================================

MOTION_FOLDERS = {
    "clmw",
    "mumw",
    "normw",
    "rdmw"
}


# ============================================================
# 4. MUSE NON SIGNAL COLUMNS
# ============================================================

# These columns should not be treated as EEG signal columns.

MUSE_EXCLUDED_COLUMNS = {

    "Time",
    "time",
    "Timestamp",
    "timestamp",
    "Date",
    "date",

    "HeadBandOn",
    "HSI_TP9",
    "HSI_AF7",
    "HSI_AF8",
    "HSI_TP10",
    "Battery",

    "Sample",
    "sample",
    "Index",
    "index"

}


# ============================================================
# 5. CREATE OUTPUT DIRECTORIES
# ============================================================

OUTPUT_DATA.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

VIS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 6. HELPER FUNCTION
# ============================================================

def is_motion_file(relative_path):
    """
    Determine whether a CSV belongs to Motion Watch.
    """

    folder_names = {
        part.lower()
        for part in relative_path.parts
    }

    return bool(
        folder_names.intersection(
            MOTION_FOLDERS
        )
    )


# ============================================================
# 7. IDENTIFY MUSE SIGNAL COLUMNS
# ============================================================

def get_muse_signal_columns(df):

    numeric_columns = (
        df.select_dtypes(
            include=np.number
        ).columns.tolist()
    )

    signal_columns = []

    for column in numeric_columns:

        if column in MUSE_EXCLUDED_COLUMNS:
            continue

        signal_columns.append(column)

    return signal_columns


# ============================================================
# 8. IDENTIFY MOTION WATCH COLUMNS
# ============================================================

def get_motion_signal_columns(df):

    numeric_columns = (
        df.select_dtypes(
            include=np.number
        ).columns.tolist()
    )

    signal_columns = []

    for column in numeric_columns:

        # Do not process obvious identifiers
        lower_name = column.lower()

        if lower_name in {
            "index",
            "sample",
            "timestamp",
            "time"
        }:
            continue

        signal_columns.append(column)

    return signal_columns


# ============================================================
# 9. CALCULATE IQR OUTLIER MASK
# ============================================================

def iqr_outlier_mask(series):

    valid = series.dropna()

    if len(valid) < MIN_VALID_VALUES:
        return pd.Series(
            False,
            index=series.index
        )

    if valid.nunique() < MIN_UNIQUE_VALUES:
        return pd.Series(
            False,
            index=series.index
        )

    q1 = valid.quantile(0.25)

    q3 = valid.quantile(0.75)

    iqr = q3 - q1

    if not np.isfinite(iqr):
        return pd.Series(
            False,
            index=series.index
        )

    if iqr == 0:

        # When IQR is zero, use median absolute
        # deviation instead.

        median = valid.median()

        deviations = (
            valid - median
        ).abs()

        mad = deviations.median()

        if mad == 0 or not np.isfinite(mad):

            return pd.Series(
                False,
                index=series.index
            )

        modified_z = (
            0.6745
            * (series - median)
            / mad
        )

        return (
            modified_z.abs()
            > MAD_THRESHOLD
        ).fillna(False)

    lower = (
        q1
        - IQR_MULTIPLIER * iqr
    )

    upper = (
        q3
        + IQR_MULTIPLIER * iqr
    )

    mask = (

        (
            series < lower
        )

        |

        (
            series > upper
        )

    )

    return mask.fillna(False)


# ============================================================
# 10. CALCULATE MAD OUTLIER MASK
# ============================================================

def mad_outlier_mask(series):

    valid = series.dropna()

    if len(valid) < MIN_VALID_VALUES:
        return pd.Series(
            False,
            index=series.index
        )

    median = valid.median()

    deviations = (
        valid - median
    ).abs()

    mad = deviations.median()

    if mad == 0 or not np.isfinite(mad):

        return pd.Series(
            False,
            index=series.index
        )

    modified_z = (
        0.6745
        * (series - median)
        / mad
    )

    mask = (
        modified_z.abs()
        > MAD_THRESHOLD
    )

    return mask.fillna(False)


# ============================================================
# 11. CALCULATE REPLACEMENT VALUE
# ============================================================

def get_replacement_value(
    series,
    sensor_type
):

    valid = series.dropna()

    if len(valid) == 0:
        return np.nan

    mode_values = valid.mode()

    # --------------------------------------------------------
    # Motion Watch
    # --------------------------------------------------------

    if sensor_type == "Motion Watch":

        if len(mode_values) > 0:

            return mode_values.iloc[0]

        return valid.median()


    # --------------------------------------------------------
    # Muse
    #
    # Muse signals are continuous. A mode can be poorly
    # defined when values have very high uniqueness.
    #
    # Therefore:
    #     meaningful mode -> mode
    #     otherwise -> median
    # --------------------------------------------------------

    unique_ratio = (
        valid.nunique()
        / len(valid)
    )

    if (
        len(mode_values) > 0
        and unique_ratio < 0.20
    ):

        return mode_values.iloc[0]

    return valid.median()


# ============================================================
# 12. PROCESS ONE COLUMN
# ============================================================

def process_column(
    df,
    column,
    sensor_type
):

    original_series = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    working_series = (
        original_series.copy()
    )

    total_replaced = 0

    iteration_details = []


    # ========================================================
    # ITERATIVE PROCESSING
    # ========================================================

    for iteration in range(
        1,
        MAX_ITERATIONS + 1
    ):

        # ----------------------------------------------------
        # IQR detection
        # ----------------------------------------------------

        iqr_mask = iqr_outlier_mask(
            working_series
        )

        iqr_count = int(
            iqr_mask.sum()
        )


        # ----------------------------------------------------
        # MAD detection
        # ----------------------------------------------------

        mad_mask = mad_outlier_mask(
            working_series
        )

        mad_count = int(
            mad_mask.sum()
        )


        # ----------------------------------------------------
        # Combine masks
        # ----------------------------------------------------

        combined_mask = (
            iqr_mask
            | mad_mask
        )


        combined_count = int(
            combined_mask.sum()
        )


        iteration_details.append({

            "iteration": iteration,

            "iqr_outliers": iqr_count,

            "mad_outliers": mad_count,

            "combined_outliers":
                combined_count

        })


        # ----------------------------------------------------
        # Nothing more to replace
        # ----------------------------------------------------

        if combined_count == 0:
            break


        # ----------------------------------------------------
        # Calculate replacement value
        # ----------------------------------------------------

        replacement_value = (
            get_replacement_value(
                working_series,
                sensor_type
            )
        )


        if pd.isna(
            replacement_value
        ):
            break


        # ----------------------------------------------------
        # Replace
        # ----------------------------------------------------

        working_series.loc[
            combined_mask
        ] = replacement_value


        total_replaced += (
            combined_count
        )


    # ========================================================
    # FINAL OUTLIER CHECK
    # ========================================================

    final_iqr_mask = iqr_outlier_mask(
        working_series
    )

    final_mad_mask = mad_outlier_mask(
        working_series
    )

    final_mask = (
        final_iqr_mask
        | final_mad_mask
    )


    final_outliers = int(
        final_mask.sum()
    )


    # ========================================================
    # UPDATE DATAFRAME
    # ========================================================

    df[column] = working_series


    return {

        "original_outliers":
            int(
                iqr_outlier_mask(
                    original_series
                ).sum()
            ),

        "total_replaced":
            total_replaced,

        "final_outliers":
            final_outliers,

        "iterations":
            len(iteration_details),

        "details":
            iteration_details

    }


# ============================================================
# 13. FIND ALL CSV FILES
# ============================================================

if not INPUT_DATA.exists():

    raise FileNotFoundError(
        f"Input dataset not found:\n{INPUT_DATA}"
    )


csv_files = sorted(
    INPUT_DATA.rglob("*.csv")
)


print("=" * 100)
print("CLEAN_DATA_V7 -> CLEAN_DATA_V8")
print("=" * 100)

print()

print(
    "Input:",
    INPUT_DATA
)

print(
    "CSV files:",
    len(csv_files)
)

print()


# ============================================================
# 14. REPORT STORAGE
# ============================================================

file_report = []

column_report = []


# ============================================================
# 15. TOTAL COUNTERS
# ============================================================

total_files = len(csv_files)

muse_files = 0
motion_files = 0

muse_v7_outliers = 0
muse_replaced = 0
muse_v8_outliers = 0

motion_v7_outliers = 0
motion_replaced = 0
motion_v8_outliers = 0

total_rows_v7 = 0
total_rows_v8 = 0


# ============================================================
# 16. PROCESS ALL FILES
# ============================================================

for file_index, input_file in enumerate(
    csv_files,
    start=1
):

    relative_path = (
        input_file.relative_to(
            INPUT_DATA
        )
    )

    output_file = (
        OUTPUT_DATA
        / relative_path
    )

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    # ========================================================
    # DETERMINE SENSOR
    # ========================================================

    if is_motion_file(
        relative_path
    ):

        sensor_type = "Motion Watch"

        motion_files += 1

    else:

        sensor_type = "Muse"

        muse_files += 1


    # ========================================================
    # READ FILE
    # ========================================================

    try:

        df = pd.read_csv(
            input_file,
            low_memory=False
        )

    except Exception as error:

        print(
            f"ERROR: {relative_path}"
        )

        print(error)

        continue


    rows_before = len(df)

    total_rows_v7 += rows_before


    # ========================================================
    # SELECT COLUMNS
    # ========================================================

    if sensor_type == "Muse":

        signal_columns = (
            get_muse_signal_columns(
                df
            )
        )

    else:

        signal_columns = (
            get_motion_signal_columns(
                df
            )
        )


    file_original_outliers = 0

    file_replaced = 0

    file_final_outliers = 0


    # ========================================================
    # PROCESS EACH SIGNAL COLUMN
    # ========================================================

    for column in signal_columns:

        result = process_column(
            df,
            column,
            sensor_type
        )


        original_outliers = (
            result["original_outliers"]
        )

        replaced = (
            result["total_replaced"]
        )

        final_outliers = (
            result["final_outliers"]
        )


        file_original_outliers += (
            original_outliers
        )

        file_replaced += (
            replaced
        )

        file_final_outliers += (
            final_outliers
        )


        # ----------------------------------------------------
        # Sensor counters
        # ----------------------------------------------------

        if sensor_type == "Muse":

            muse_v7_outliers += (
                original_outliers
            )

            muse_replaced += (
                replaced
            )

            muse_v8_outliers += (
                final_outliers
            )

        else:

            motion_v7_outliers += (
                original_outliers
            )

            motion_replaced += (
                replaced
            )

            motion_v8_outliers += (
                final_outliers
            )


        # ----------------------------------------------------
        # Column report
        # ----------------------------------------------------

        column_report.append({

            "sensor":
                sensor_type,

            "file":
                str(relative_path),

            "column":
                column,

            "rows":
                rows_before,

            "V7_outliers":
                original_outliers,

            "total_replaced":
                replaced,

            "V8_remaining_outliers":
                final_outliers,

            "iterations":
                result["iterations"]

        })


    # ========================================================
    # SAVE V8
    # ========================================================

    rows_after = len(df)

    total_rows_v8 += rows_after


    # ========================================================
    # ROW COUNT CHECK
    # ========================================================

    row_count_preserved = (
        rows_before
        == rows_after
    )


    # ========================================================
    # WRITE CSV
    # ========================================================

    df.to_csv(
        output_file,
        index=False
    )


    # ========================================================
    # FILE REPORT
    # ========================================================

    activity = (
        relative_path.parts[0]
        if len(relative_path.parts) > 0
        else ""
    )


    file_report.append({

        "sensor":
            sensor_type,

        "file":
            str(relative_path),

        "activity":
            activity,

        "rows_v7":
            rows_before,

        "rows_v8":
            rows_after,

        "numeric_signal_columns":
            len(signal_columns),

        "V7_outliers":
            file_original_outliers,

        "total_replaced":
            file_replaced,

        "V8_remaining_outliers":
            file_final_outliers,

        "row_count_preserved":
            row_count_preserved

    })


    # ========================================================
    # PROGRESS
    # ========================================================

    print(

        f"[{file_index}/{total_files}] "

        f"{sensor_type} | "

        f"V7: "
        f"{file_original_outliers:,} | "

        f"Replaced: "
        f"{file_replaced:,} | "

        f"V8: "
        f"{file_final_outliers:,} | "

        f"{relative_path}"

    )


# ============================================================
# 17. CREATE REPORT DATAFRAMES
# ============================================================

file_report_df = pd.DataFrame(
    file_report
)

column_report_df = pd.DataFrame(
    column_report
)


# ============================================================
# 18. SAVE REPORTS
# ============================================================

file_report_path = (
    REPORT_DIR
    / "V8_file_quality_report.csv"
)

column_report_path = (
    REPORT_DIR
    / "V8_column_outlier_report.csv"
)


file_report_df.to_csv(
    file_report_path,
    index=False
)

column_report_df.to_csv(
    column_report_path,
    index=False
)


# ============================================================
# 19. CALCULATE REDUCTION
# ============================================================

if muse_v7_outliers > 0:

    muse_reduction = (

        (
            muse_v7_outliers
            - muse_v8_outliers
        )

        /

        muse_v7_outliers

        * 100

    )

else:

    muse_reduction = 0


if motion_v7_outliers > 0:

    motion_reduction = (

        (
            motion_v7_outliers
            - motion_v8_outliers
        )

        /

        motion_v7_outliers

        * 100

    )

else:

    motion_reduction = 0


# ============================================================
# 20. CREATE SUMMARY
# ============================================================

summary_df = pd.DataFrame({

    "metric": [

        "Total CSV files",

        "Muse files",

        "Motion Watch files",

        "Muse V7 outliers",

        "Muse replacements",

        "Muse V8 remaining outliers",

        "Muse outlier reduction percentage",

        "Motion Watch V7 outliers",

        "Motion Watch replacements",

        "Motion Watch V8 remaining outliers",

        "Motion Watch outlier reduction percentage",

        "V7 total rows",

        "V8 total rows"

    ],

    "value": [

        total_files,

        muse_files,

        motion_files,

        muse_v7_outliers,

        muse_replaced,

        muse_v8_outliers,

        muse_reduction,

        motion_v7_outliers,

        motion_replaced,

        motion_v8_outliers,

        motion_reduction,

        total_rows_v7,

        total_rows_v8

    ]

})


summary_path = (
    REPORT_DIR
    / "V8_summary.csv"
)

summary_df.to_csv(
    summary_path,
    index=False
)


# ============================================================
# 21. PREPARE GRAPH DATA
# ============================================================

muse_report = file_report_df[
    file_report_df["sensor"]
    == "Muse"
].copy()

motion_report = file_report_df[
    file_report_df["sensor"]
    == "Motion Watch"
].copy()


muse_report = (
    muse_report
    .reset_index(drop=True)
)

motion_report = (
    motion_report
    .reset_index(drop=True)
)


muse_report[
    "file_index"
] = np.arange(
    len(muse_report)
)

motion_report[
    "file_index"
] = np.arange(
    len(motion_report)
)


# ============================================================
# 22. SINGLE COMBINED GRAPH
# ============================================================

plt.figure(
    figsize=(22, 12)
)


# ------------------------------------------------------------
# Muse
# ------------------------------------------------------------

plt.scatter(

    muse_report["file_index"],

    muse_report["V7_outliers"],

    s=30,

    alpha=0.65,

    label="Muse V7 outliers"

)


plt.scatter(

    muse_report["file_index"],

    muse_report["V8_remaining_outliers"],

    s=30,

    alpha=0.65,

    label="Muse V8 remaining outliers"

)


for _, row in muse_report.iterrows():

    plt.plot(

        [
            row["file_index"],
            row["file_index"]
        ],

        [
            row["V7_outliers"],
            row["V8_remaining_outliers"]
        ],

        alpha=0.20,

        linewidth=0.6

    )


# ------------------------------------------------------------
# Motion Watch
# ------------------------------------------------------------

offset = len(muse_report) + 10


motion_x = (
    motion_report["file_index"]
    + offset
)


plt.scatter(

    motion_x,

    motion_report["V7_outliers"],

    s=30,

    alpha=0.65,

    label="Motion Watch V7 outliers"

)


plt.scatter(

    motion_x,

    motion_report["V8_remaining_outliers"],

    s=30,

    alpha=0.65,

    label="Motion Watch V8 remaining outliers"

)


for _, row in motion_report.iterrows():

    x = (
        row["file_index"]
        + offset
    )

    plt.plot(

        [
            x,
            x
        ],

        [
            row["V7_outliers"],
            row["V8_remaining_outliers"]
        ],

        alpha=0.20,

        linewidth=0.6

    )


# ============================================================
# GRAPH FORMATTING
# ============================================================

plt.axvline(

    x=offset - 5,

    linestyle="--",

    alpha=0.5

)


plt.text(

    len(muse_report) / 2,

    plt.ylim()[1] * 0.95,

    "MUSE",

    ha="center",

    fontsize=14

)


plt.text(

    offset
    + len(motion_report) / 2,

    plt.ylim()[1] * 0.95,

    "MOTION WATCH",

    ha="center",

    fontsize=14

)


plt.title(

    "CLEAN_DATA_V7 vs CLEAN_DATA_V8: "
    "Muse and Motion Watch Outlier Structure",

    fontsize=17

)


plt.xlabel(
    "CSV File Index",
    fontsize=13
)

plt.ylabel(
    "Number of Detected Outlier Data Points",
    fontsize=13
)


plt.legend()

plt.grid(
    alpha=0.25
)


plt.tight_layout()


graph_path = (
    VIS_DIR
    / "V7_V8_Muse_MotionWatch_Outlier_Structure.png"
)


plt.savefig(

    graph_path,

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 23. SECOND GRAPH
#
# Motion Watch only
# Useful for checking the remaining 1041 outliers.
# ============================================================

plt.figure(
    figsize=(20, 9)
)


plt.scatter(

    motion_report["file_index"],

    motion_report["V7_outliers"],

    s=35,

    alpha=0.7,

    label="Motion Watch V7"

)


plt.scatter(

    motion_report["file_index"],

    motion_report["V8_remaining_outliers"],

    s=35,

    alpha=0.7,

    label="Motion Watch V8"

)


for _, row in motion_report.iterrows():

    plt.plot(

        [
            row["file_index"],
            row["file_index"]
        ],

        [
            row["V7_outliers"],
            row["V8_remaining_outliers"]
        ],

        alpha=0.20,

        linewidth=0.7

    )


plt.title(

    "Motion Watch Outlier Reduction: "
    "CLEAN_DATA_V7 vs CLEAN_DATA_V8",

    fontsize=17

)


plt.xlabel(
    "Motion Watch CSV File Index",
    fontsize=13
)

plt.ylabel(
    "Number of Detected Outlier Data Points",
    fontsize=13
)


plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


motion_graph_path = (
    VIS_DIR
    / "V7_V8_MotionWatch_Outlier_Structure.png"
)


plt.savefig(

    motion_graph_path,

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 24. FINAL VALIDATION
# ============================================================

row_errors = file_report_df[
    file_report_df[
        "row_count_preserved"
    ] == False
]


v8_files = sorted(
    OUTPUT_DATA.rglob("*.csv")
)


# ============================================================
# 25. FINAL OUTPUT
# ============================================================

print()
print("=" * 100)
print("CLEAN_DATA_V8 PROCESSING COMPLETED")
print("=" * 100)

print()

print(
    "Total CSV files:",
    total_files
)

print(
    "V8 CSV files:",
    len(v8_files)
)

print()

print(
    "Muse files:",
    muse_files
)

print(
    "Motion Watch files:",
    motion_files
)

print()
print("----- MUSE -----")

print(
    "V7 outliers:",
    f"{muse_v7_outliers:,}"
)

print(
    "Replaced:",
    f"{muse_replaced:,}"
)

print(
    "V8 remaining:",
    f"{muse_v8_outliers:,}"
)

print(
    "Reduction:",
    f"{muse_reduction:.2f}%"
)

print()
print("----- MOTION WATCH -----")

print(
    "V7 outliers:",
    f"{motion_v7_outliers:,}"
)

print(
    "Replaced:",
    f"{motion_replaced:,}"
)

print(
    "V8 remaining:",
    f"{motion_v8_outliers:,}"
)

print(
    "Reduction:",
    f"{motion_reduction:.2f}%"
)

print()
print("----- ROW VALIDATION -----")

print(
    "V7 total rows:",
    f"{total_rows_v7:,}"
)

print(
    "V8 total rows:",
    f"{total_rows_v8:,}"
)

print(
    "Row count errors:",
    len(row_errors)
)

print()

print(
    "V8 dataset:",
    OUTPUT_DATA
)

print(
    "Summary:",
    summary_path
)

print(
    "File report:",
    file_report_path
)

print(
    "Column report:",
    column_report_path
)

print(
    "Combined visualization:",
    graph_path
)

print(
    "Motion Watch visualization:",
    motion_graph_path
)

print()
print("=" * 100)
print("DONE")
print("=" * 100)


# ============================================================
# 26. DISPLAY SUMMARY
# ============================================================

display(summary_df)
```

```python
# ============================================================
# CLEAN_DATA_V8 TO CLEAN_DATA_V9
#
# Dedicated Muse EEG outlier reduction
#
# Motion Watch files are copied unchanged from V8 because
# the V8 processing already reduced their detected outliers
# to zero.
#
# Muse processing uses:
#
# 1. Robust IQR detection
# 2. Modified MAD detection
# 3. Local temporal detection
# 4. Linear interpolation for isolated outliers
# 5. Rolling median for consecutive outliers
# 6. Final robust boundary correction
# 7. Repeated validation
#
# ============================================================


from pathlib import Path
import shutil
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


warnings.filterwarnings("ignore")


# ============================================================
# 1. PATHS
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V8"
)

INPUT_DATA = (
    INPUT_ROOT
    / "clean_data"
)


OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V9"
)

OUTPUT_DATA = (
    OUTPUT_ROOT
    / "clean_data"
)

REPORT_DIR = (
    OUTPUT_ROOT
    / "reports"
)

VIS_DIR = (
    OUTPUT_ROOT
    / "visualizations"
)


# ============================================================
# 2. PARAMETERS
# ============================================================

# IQR threshold
IQR_MULTIPLIER = 1.5


# Modified MAD threshold
MAD_THRESHOLD = 3.5


# Local Hampel threshold
LOCAL_MAD_THRESHOLD = 3.5


# Rolling window for temporal correction
ROLLING_WINDOW = 11


# Maximum V9 processing passes
MAX_PASSES = 5


# Minimum valid observations
MIN_VALID_VALUES = 20


# Final boundary multiplier
FINAL_IQR_MULTIPLIER = 1.5


# ============================================================
# 3. CREATE DIRECTORIES
# ============================================================

OUTPUT_DATA.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

VIS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 4. MOTION WATCH FOLDER NAMES
# ============================================================

MOTION_FOLDERS = {

    "clmw",
    "mumw",
    "normw",
    "rdmw"

}


# ============================================================
# 5. MUSE NON SIGNAL COLUMNS
# ============================================================

MUSE_EXCLUDED_COLUMNS = {

    "Time",
    "time",
    "Timestamp",
    "timestamp",

    "Date",
    "date",

    "HeadBandOn",

    "HSI_TP9",
    "HSI_AF7",
    "HSI_AF8",
    "HSI_TP10",

    "Battery",

    "Sample",
    "sample",

    "Index",
    "index"

}


# ============================================================
# 6. DETERMINE SENSOR TYPE
# ============================================================

def is_motion_watch(relative_path):

    folders = {
        part.lower()
        for part in relative_path.parts
    }

    return bool(
        folders.intersection(
            MOTION_FOLDERS
        )
    )


# ============================================================
# 7. FIND MUSE SIGNAL COLUMNS
# ============================================================

def get_muse_columns(df):

    numeric_columns = (
        df.select_dtypes(
            include=np.number
        ).columns.tolist()
    )

    columns = []

    for column in numeric_columns:

        if column in MUSE_EXCLUDED_COLUMNS:
            continue

        columns.append(column)

    return columns


# ============================================================
# 8. GLOBAL IQR MASK
# ============================================================

def global_iqr_mask(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    valid = values.dropna()

    if len(valid) < MIN_VALID_VALUES:
        return pd.Series(
            False,
            index=series.index
        )

    q1 = valid.quantile(0.25)

    q3 = valid.quantile(0.75)

    iqr = q3 - q1

    if not np.isfinite(iqr):
        return pd.Series(
            False,
            index=series.index
        )

    if iqr <= 0:
        return pd.Series(
            False,
            index=series.index
        )

    lower = (
        q1
        - IQR_MULTIPLIER * iqr
    )

    upper = (
        q3
        + IQR_MULTIPLIER * iqr
    )

    mask = (
        (values < lower)
        |
        (values > upper)
    )

    return mask.fillna(False)


# ============================================================
# 9. GLOBAL MAD MASK
# ============================================================

def global_mad_mask(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    valid = values.dropna()

    if len(valid) < MIN_VALID_VALUES:
        return pd.Series(
            False,
            index=series.index
        )

    median = valid.median()

    deviations = (
        valid - median
    ).abs()

    mad = deviations.median()

    if mad <= 0:
        return pd.Series(
            False,
            index=series.index
        )

    modified_z = (
        0.6745
        * (values - median)
        / mad
    )

    mask = (
        modified_z.abs()
        > MAD_THRESHOLD
    )

    return mask.fillna(False)


# ============================================================
# 10. LOCAL HAMPEL MASK
# ============================================================

def local_hampel_mask(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    rolling_median = (
        values
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=3
        )
        .median()
    )

    absolute_deviation = (
        values
        - rolling_median
    ).abs()


    rolling_mad = (
        absolute_deviation
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=3
        )
        .median()
    )


    modified_deviation = (
        0.6745
        * absolute_deviation
        / rolling_mad.replace(
            0,
            np.nan
        )
    )


    mask = (
        modified_deviation
        > LOCAL_MAD_THRESHOLD
    )


    return mask.fillna(False)


# ============================================================
# 11. COMBINED OUTLIER MASK
# ============================================================

def detect_muse_outliers(series):

    iqr_mask = global_iqr_mask(
        series
    )

    mad_mask = global_mad_mask(
        series
    )

    local_mask = local_hampel_mask(
        series
    )

    combined = (
        iqr_mask
        |
        mad_mask
        |
        local_mask
    )

    return combined


# ============================================================
# 12. INTERPOLATE ISOLATED OUTLIERS
# ============================================================

def interpolate_outliers(
    series,
    mask
):

    result = series.copy()

    if not mask.any():
        return result


    # --------------------------------------------------------
    # Temporarily convert detected outliers to missing
    # --------------------------------------------------------

    temporary = result.copy()

    temporary.loc[
        mask
    ] = np.nan


    # --------------------------------------------------------
    # Interpolate using neighbouring valid observations
    # --------------------------------------------------------

    interpolated = (
        temporary
        .interpolate(
            method="linear",
            limit_direction="both"
        )
    )


    # --------------------------------------------------------
    # Only replace detected outliers
    # --------------------------------------------------------

    result.loc[
        mask
    ] = interpolated.loc[
        mask
    ]


    return result


# ============================================================
# 13. ROLLING MEDIAN REPLACEMENT
# ============================================================

def rolling_median_replace(
    series,
    mask
):

    result = series.copy()

    if not mask.any():
        return result


    rolling_median = (
        series
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=3
        )
        .median()
    )


    # --------------------------------------------------------
    # Replace remaining abnormal values using local median
    # --------------------------------------------------------

    replacement = (
        rolling_median
        .fillna(
            series.median()
        )
    )


    result.loc[
        mask
    ] = replacement.loc[
        mask
    ]


    return result


# ============================================================
# 14. FINAL ROBUST BOUNDARY
# ============================================================

def final_boundary_correction(series):

    result = series.copy()

    values = pd.to_numeric(
        result,
        errors="coerce"
    )

    valid = values.dropna()

    if len(valid) < MIN_VALID_VALUES:
        return result, 0


    q1 = valid.quantile(0.25)

    q3 = valid.quantile(0.75)

    iqr = q3 - q1

    if iqr <= 0:
        return result, 0


    lower = (
        q1
        - FINAL_IQR_MULTIPLIER * iqr
    )

    upper = (
        q3
        + FINAL_IQR_MULTIPLIER * iqr
    )


    mask = (
        (values < lower)
        |
        (values > upper)
    ).fillna(False)


    count = int(
        mask.sum()
    )


    if count == 0:
        return result, 0


    # --------------------------------------------------------
    # Instead of replacing everything with one value,
    # clip only the extreme boundaries.
    # --------------------------------------------------------

    result.loc[
        values < lower
    ] = lower


    result.loc[
        values > upper
    ] = upper


    return result, count


# ============================================================
# 15. PROCESS ONE MUSE COLUMN
# ============================================================

def process_muse_column(series):

    original = pd.to_numeric(
        series,
        errors="coerce"
    )

    working = original.copy()


    original_outliers = int(
        detect_muse_outliers(
            working
        ).sum()
    )


    total_replaced = 0

    pass_records = []


    # ========================================================
    # ITERATIVE PROCESSING
    # ========================================================

    for current_pass in range(
        1,
        MAX_PASSES + 1
    ):


        # ----------------------------------------------------
        # Detect current outliers
        # ----------------------------------------------------

        mask = detect_muse_outliers(
            working
        )


        current_count = int(
            mask.sum()
        )


        pass_records.append({

            "pass":
                current_pass,

            "detected":
                current_count

        })


        # ----------------------------------------------------
        # Stop when no outliers remain
        # ----------------------------------------------------

        if current_count == 0:
            break


        # ----------------------------------------------------
        # First use interpolation
        # ----------------------------------------------------

        before = working.copy()


        working = interpolate_outliers(
            working,
            mask
        )


        # ----------------------------------------------------
        # Determine what was changed
        # ----------------------------------------------------

        changed = (
            before
            != working
        ).fillna(False)


        changed_count = int(
            changed.sum()
        )


        total_replaced += (
            changed_count
        )


        # ----------------------------------------------------
        # Check remaining outliers
        # ----------------------------------------------------

        remaining_mask = (
            detect_muse_outliers(
                working
            )
        )


        remaining_count = int(
            remaining_mask.sum()
        )


        # ----------------------------------------------------
        # If interpolation did not remove all
        # abnormalities, use rolling median
        # ----------------------------------------------------

        if remaining_count > 0:

            before = working.copy()


            working = rolling_median_replace(
                working,
                remaining_mask
            )


            changed = (
                before
                != working
            ).fillna(False)


            changed_count = int(
                changed.sum()
            )


            total_replaced += (
                changed_count
            )


    # ========================================================
    # FINAL BOUNDARY CORRECTION
    # ========================================================

    working, final_corrected = (
        final_boundary_correction(
            working
        )
    )


    total_replaced += (
        final_corrected
    )


    # ========================================================
    # FINAL DETECTION
    # ========================================================

    final_mask = detect_muse_outliers(
        working
    )


    final_outliers = int(
        final_mask.sum()
    )


    return {

        "series":
            working,

        "original_outliers":
            original_outliers,

        "replaced":
            total_replaced,

        "final_outliers":
            final_outliers,

        "passes":
            len(pass_records),

        "pass_records":
            pass_records

    }


# ============================================================
# 16. FIND ALL CSV FILES
# ============================================================

if not INPUT_DATA.exists():

    raise FileNotFoundError(
        f"V8 dataset not found: {INPUT_DATA}"
    )


csv_files = sorted(
    INPUT_DATA.rglob("*.csv")
)


print("=" * 100)

print(
    "CLEAN_DATA_V8 TO CLEAN_DATA_V9"
)

print("=" * 100)

print()

print(
    "Input:",
    INPUT_DATA
)

print(
    "CSV files:",
    len(csv_files)
)

print()


# ============================================================
# 17. REPORT LISTS
# ============================================================

file_report = []

column_report = []


# ============================================================
# 18. COUNTERS
# ============================================================

total_files = len(csv_files)

muse_files = 0

motion_files = 0


muse_v8_outliers = 0

muse_replaced = 0

muse_v9_outliers = 0


motion_v8_outliers = 0

motion_v9_outliers = 0


rows_v8 = 0

rows_v9 = 0


# ============================================================
# 19. PROCESS ALL FILES
# ============================================================

for file_index, input_file in enumerate(
    csv_files,
    start=1
):


    relative_path = (
        input_file.relative_to(
            INPUT_DATA
        )
    )


    output_file = (
        OUTPUT_DATA
        / relative_path
    )


    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    # ========================================================
    # SENSOR TYPE
    # ========================================================

    if is_motion_watch(
        relative_path
    ):

        sensor_type = "Motion Watch"

        motion_files += 1

    else:

        sensor_type = "Muse"

        muse_files += 1


    # ========================================================
    # READ
    # ========================================================

    df = pd.read_csv(
        input_file,
        low_memory=False
    )


    rows_before = len(df)

    rows_v8 += rows_before


    file_v8_outliers = 0

    file_replaced = 0

    file_v9_outliers = 0


    # ========================================================
    # MOTION WATCH
    #
    # Copy unchanged
    # ========================================================

    if sensor_type == "Motion Watch":

        df.to_csv(
            output_file,
            index=False
        )


        rows_after = len(df)

        rows_v9 += rows_after


        file_report.append({

            "sensor":
                sensor_type,

            "file":
                str(relative_path),

            "rows_v8":
                rows_before,

            "rows_v9":
                rows_after,

            "V8_outliers":
                0,

            "replaced":
                0,

            "V9_remaining_outliers":
                0,

            "row_count_preserved":
                rows_before
                == rows_after

        })


        print(

            f"[{file_index}/{total_files}] "

            f"Motion Watch | "

            f"unchanged | "

            f"{relative_path}"

        )


        continue


    # ========================================================
    # MUSE COLUMNS
    # ========================================================

    signal_columns = (
        get_muse_columns(
            df
        )
    )


    # ========================================================
    # PROCESS MUSE
    # ========================================================

    for column in signal_columns:


        result = process_muse_column(
            df[column]
        )


        df[column] = (
            result["series"]
        )


        file_v8_outliers += (
            result[
                "original_outliers"
            ]
        )


        file_replaced += (
            result[
                "replaced"
            ]
        )


        file_v9_outliers += (
            result[
                "final_outliers"
            ]
        )


        muse_v8_outliers += (
            result[
                "original_outliers"
            ]
        )


        muse_replaced += (
            result[
                "replaced"
            ]
        )


        muse_v9_outliers += (
            result[
                "final_outliers"
            ]
        )


        # ----------------------------------------------------
        # Column report
        # ----------------------------------------------------

        column_report.append({

            "sensor":
                "Muse",

            "file":
                str(relative_path),

            "column":
                column,

            "rows":
                rows_before,

            "V8_outliers":
                result[
                    "original_outliers"
                ],

            "replaced":
                result[
                    "replaced"
                ],

            "V9_remaining_outliers":
                result[
                    "final_outliers"
                ],

            "passes":
                result[
                    "passes"
                ]

        })


    # ========================================================
    # SAVE MUSE V9
    # ========================================================

    rows_after = len(df)

    rows_v9 += rows_after


    df.to_csv(
        output_file,
        index=False
    )


    # ========================================================
    # FILE REPORT
    # ========================================================

    file_report.append({

        "sensor":
            "Muse",

        "file":
            str(relative_path),

        "rows_v8":
            rows_before,

        "rows_v9":
            rows_after,

        "numeric_signal_columns":
            len(signal_columns),

        "V8_outliers":
            file_v8_outliers,

        "replaced":
            file_replaced,

        "V9_remaining_outliers":
            file_v9_outliers,

        "row_count_preserved":
            rows_before
            == rows_after

    })


    # ========================================================
    # PROGRESS
    # ========================================================

    print(

        f"[{file_index}/{total_files}] "

        f"Muse | "

        f"V8: "
        f"{file_v8_outliers:,} | "

        f"Replaced: "
        f"{file_replaced:,} | "

        f"V9: "
        f"{file_v9_outliers:,} | "

        f"{relative_path}"

    )


# ============================================================
# 20. DATAFRAMES
# ============================================================

file_report_df = pd.DataFrame(
    file_report
)

column_report_df = pd.DataFrame(
    column_report
)


# ============================================================
# 21. SAVE REPORTS
# ============================================================

file_report_path = (
    REPORT_DIR
    / "V9_file_quality_report.csv"
)

column_report_path = (
    REPORT_DIR
    / "V9_muse_column_outlier_report.csv"
)


file_report_df.to_csv(
    file_report_path,
    index=False
)

column_report_df.to_csv(
    column_report_path,
    index=False
)


# ============================================================
# 22. REDUCTION
# ============================================================

if muse_v8_outliers > 0:

    muse_reduction = (

        (
            muse_v8_outliers
            - muse_v9_outliers
        )

        /

        muse_v8_outliers

        * 100

    )

else:

    muse_reduction = 0


# ============================================================
# 23. ROW VALIDATION
# ============================================================

row_errors = file_report_df[
    file_report_df[
        "row_count_preserved"
    ] == False
]


# ============================================================
# 24. SUMMARY
# ============================================================

summary_df = pd.DataFrame({

    "metric": [

        "Total CSV files",

        "Muse files",

        "Motion Watch files",

        "Muse V8 outliers",

        "Muse replacements",

        "Muse V9 remaining outliers",

        "Muse reduction percentage",

        "Motion Watch V8 outliers",

        "Motion Watch V9 outliers",

        "V8 total rows",

        "V9 total rows",

        "Row count errors"

    ],

    "value": [

        total_files,

        muse_files,

        motion_files,

        muse_v8_outliers,

        muse_replaced,

        muse_v9_outliers,

        muse_reduction,

        motion_v8_outliers,

        motion_v9_outliers,

        rows_v8,

        rows_v9,

        len(row_errors)

    ]

})


summary_path = (
    REPORT_DIR
    / "V9_summary.csv"
)


summary_df.to_csv(
    summary_path,
    index=False
)


# ============================================================
# 25. MUSE REDUCTION GRAPH
# ============================================================

muse_report = file_report_df[
    file_report_df["sensor"]
    == "Muse"
].copy()


muse_report = (
    muse_report
    .reset_index(drop=True)
)


muse_report[
    "file_index"
] = np.arange(
    len(muse_report)
)


plt.figure(
    figsize=(22, 10)
)


plt.scatter(

    muse_report["file_index"],

    muse_report["V8_outliers"],

    s=30,

    alpha=0.65,

    label="Muse V8"

)


plt.scatter(

    muse_report["file_index"],

    muse_report[
        "V9_remaining_outliers"
    ],

    s=30,

    alpha=0.65,

    label="Muse V9"

)


for _, row in muse_report.iterrows():

    plt.plot(

        [
            row["file_index"],
            row["file_index"]
        ],

        [
            row["V8_outliers"],
            row[
                "V9_remaining_outliers"
            ]
        ],

        alpha=0.20,

        linewidth=0.6

    )


plt.title(

    "Muse Outlier Reduction: "
    "CLEAN_DATA_V8 vs CLEAN_DATA_V9",

    fontsize=17

)


plt.xlabel(
    "Muse CSV File Index",
    fontsize=13
)


plt.ylabel(
    "Detected Outlier Data Points",
    fontsize=13
)


plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


muse_graph_path = (
    VIS_DIR
    / "V8_V9_Muse_Outlier_Reduction.png"
)


plt.savefig(

    muse_graph_path,

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 26. FINAL OUTLIER DISTRIBUTION GRAPH
# ============================================================

plt.figure(
    figsize=(22, 9)
)


plt.bar(

    muse_report["file_index"],

    muse_report[
        "V9_remaining_outliers"
    ],

    width=0.8

)


plt.title(

    "Remaining Muse Outliers After V9",

    fontsize=17

)


plt.xlabel(
    "Muse CSV File Index",
    fontsize=13
)


plt.ylabel(
    "V9 Remaining Outliers",
    fontsize=13
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


remaining_graph_path = (
    VIS_DIR
    / "V9_Muse_Remaining_Outliers.png"
)


plt.savefig(

    remaining_graph_path,

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 27. FINAL OUTPUT
# ============================================================

print()

print("=" * 100)

print(
    "CLEAN_DATA_V9 PROCESSING COMPLETED"
)

print("=" * 100)

print()

print(
    "Total CSV files:",
    total_files
)

print(
    "V9 CSV files:",
    len(
        list(
            OUTPUT_DATA.rglob(
                "*.csv"
            )
        )
    )
)

print()

print(
    "Muse files:",
    muse_files
)

print(
    "Motion Watch files:",
    motion_files
)

print()

print("----- MUSE -----")

print(
    "V8 outliers:",
    f"{muse_v8_outliers:,}"
)

print(
    "Replaced:",
    f"{muse_replaced:,}"
)

print(
    "V9 remaining:",
    f"{muse_v9_outliers:,}"
)

print(
    "Reduction:",
    f"{muse_reduction:.2f}%"
)

print()

print("----- MOTION WATCH -----")

print(
    "V8 outliers:",
    f"{motion_v8_outliers:,}"
)

print(
    "V9 outliers:",
    f"{motion_v9_outliers:,}"
)

print()

print("----- ROW VALIDATION -----")

print(
    "V8 total rows:",
    f"{rows_v8:,}"
)

print(
    "V9 total rows:",
    f"{rows_v9:,}"
)

print(
    "Row count errors:",
    len(row_errors)
)

print()

print(
    "V9 dataset:",
    OUTPUT_DATA
)

print(
    "Summary:",
    summary_path
)

print(
    "File report:",
    file_report_path
)

print(
    "Column report:",
    column_report_path
)

print(
    "Muse graph:",
    muse_graph_path
)

print(
    "Remaining outlier graph:",
    remaining_graph_path
)

print()

print("=" * 100)

print("DONE")

print("=" * 100)

display(summary_df)
```

```python
# ============================================================
# CLEAN_DATA_V9
# NOISE STRUCTURE VISUALIZATION
#
# Creates exactly TWO graphs:
#
#   1. Muse noise structure
#   2. Motion Watch noise structure
#
# The V9 dataset is NOT modified.
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. INPUT AND OUTPUT PATHS
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V9/clean_data"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V9/visualizations"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. EXPERIMENTAL CONDITIONS
# ============================================================

CONDITIONS = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]


# ============================================================
# 3. FOLDER IDENTIFICATION
# ============================================================

MUSE_FOLDER_NAMES = {
    "clms",
    "mums",
    "norms",
    "rdms"
}


MOTION_FOLDER_NAMES = {
    "clmw",
    "mumw",
    "normw",
    "rdmw"
}


# ============================================================
# 4. IDENTIFY DATA TYPE
# ============================================================

def identify_sensor(file_path):

    folder_names = {
        part.lower()
        for part in file_path.parts
    }

    if folder_names.intersection(
        MUSE_FOLDER_NAMES
    ):
        return "Muse"

    if folder_names.intersection(
        MOTION_FOLDER_NAMES
    ):
        return "Motion Watch"

    return None


# ============================================================
# 5. GET CONDITION
# ============================================================

def get_condition(file_path):

    parts = [
        part.lower()
        for part in file_path.parts
    ]

    for condition in CONDITIONS:

        if condition in parts:
            return condition

    return "Unknown"


# ============================================================
# 6. EXCLUDE NON SIGNAL COLUMNS
# ============================================================

NON_SIGNAL_COLUMNS = {

    "Time",
    "time",

    "Timestamp",
    "timestamp",

    "Date",
    "date",

    "Index",
    "index",

    "Sample",
    "sample",

    "Person Serial",
    "Person_Serial",

    "Category",
    "category",

    "Condition",
    "condition",

    "Activity",
    "activity",

    "Light",
    "light",

    "Battery",
    "battery",

    "HeadBandOn",

    "HSI_TP9",
    "HSI_AF7",
    "HSI_AF8",
    "HSI_TP10"
}


# ============================================================
# 7. NUMERICAL SIGNAL COLUMNS
# ============================================================

def get_signal_columns(df):

    columns = []

    for column in df.select_dtypes(
        include=np.number
    ).columns:

        if column not in NON_SIGNAL_COLUMNS:

            columns.append(column)

    return columns


# ============================================================
# 8. ROBUST NOISE MEASURE
#
# Uses MAD based deviation.
#
# This does NOT modify the data.
# It only estimates how much unusual variation
# exists in each file.
# ============================================================

def calculate_noise_score(
    df,
    columns
):

    scores = []

    for column in columns:

        values = pd.to_numeric(
            df[column],
            errors="coerce"
        ).dropna()

        if len(values) < 20:
            continue

        median = values.median()

        mad = np.median(
            np.abs(
                values - median
            )
        )

        if mad == 0 or not np.isfinite(mad):

            continue

        modified_z = (
            0.6745
            * (
                values - median
            )
            / mad
        )

        abnormal = (
            modified_z.abs()
            > 3.5
        )

        scores.append(
            abnormal.sum()
        )

    if len(scores) == 0:
        return 0

    return int(
        np.sum(scores)
    )


# ============================================================
# 9. CALCULATE ROBUST VARIABILITY
# ============================================================

def calculate_variability(
    df,
    columns
):

    values = []

    for column in columns:

        series = pd.to_numeric(
            df[column],
            errors="coerce"
        ).dropna()

        if len(series) < 20:
            continue

        median = series.median()

        mad = np.median(
            np.abs(
                series - median
            )
        )

        if mad == 0:
            continue

        values.append(
            mad
        )

    if len(values) == 0:
        return 0

    return float(
        np.median(values)
    )


# ============================================================
# 10. CALCULATE ZERO / MISSING STRUCTURE
# ============================================================

def calculate_missing_ratio(
    df,
    columns
):

    if len(columns) == 0:
        return 0

    selected = df[
        columns
    ]

    total = (
        selected.shape[0]
        * selected.shape[1]
    )

    if total == 0:
        return 0

    missing = (
        selected.isna()
        .sum()
        .sum()
    )

    return (
        missing
        / total
        * 100
    )


# ============================================================
# 11. PROCESS ALL FILES
# ============================================================

all_files = sorted(
    INPUT_ROOT.rglob(
        "*.csv"
    )
)


print("=" * 90)
print("CLEAN_DATA_V9 NOISE STRUCTURE ANALYSIS")
print("=" * 90)

print()
print(
    "Total CSV files:",
    len(all_files)
)
print()


records = []


for index, file_path in enumerate(
    all_files,
    start=1
):

    sensor = identify_sensor(
        file_path
    )

    if sensor is None:
        continue


    condition = get_condition(
        file_path
    )


    try:

        df = pd.read_csv(
            file_path,
            low_memory=False
        )

    except Exception as e:

        print(
            "Could not read:",
            file_path,
            e
        )

        continue


    signal_columns = (
        get_signal_columns(
            df
        )
    )


    noise_score = (
        calculate_noise_score(
            df,
            signal_columns
        )
    )


    variability = (
        calculate_variability(
            df,
            signal_columns
        )
    )


    missing_ratio = (
        calculate_missing_ratio(
            df,
            signal_columns
        )
    )


    records.append({

        "sensor":
            sensor,

        "condition":
            condition,

        "file":
            str(
                file_path.relative_to(
                    INPUT_ROOT
                )
            ),

        "file_index":
            index,

        "rows":
            len(df),

        "signal_columns":
            len(signal_columns),

        "noise_score":
            noise_score,

        "robust_variability":
            variability,

        "missing_percentage":
            missing_ratio

    })


    print(
        f"[{index}/{len(all_files)}] "
        f"{sensor:15s} "
        f"{condition:15s} "
        f"Noise={noise_score:,}"
    )


# ============================================================
# 12. DATAFRAME
# ============================================================

results = pd.DataFrame(
    records
)


# ============================================================
# 13. SPLIT DATA
# ============================================================

muse = results[
    results["sensor"]
    == "Muse"
].copy()


motion = results[
    results["sensor"]
    == "Motion Watch"
].copy()


# ============================================================
# 14. SORT WITHIN EACH CONDITION
# ============================================================

def sort_condition_data(
    df
):

    frames = []

    for condition in CONDITIONS:

        temp = df[
            df["condition"]
            == condition
        ].copy()

        temp = temp.sort_values(
            "file"
        )

        frames.append(
            temp
        )

    if len(frames) == 0:

        return df

    return pd.concat(
        frames,
        ignore_index=True
    )


muse = sort_condition_data(
    muse
)

motion = sort_condition_data(
    motion
)


# ============================================================
# 15. CREATE CONTINUOUS VISUAL INDEX
# ============================================================

muse[
    "visual_index"
] = np.arange(
    len(muse)
)


motion[
    "visual_index"
] = np.arange(
    len(motion)
)


# ============================================================
# 16. CONDITION BOUNDARIES
# ============================================================

def add_condition_boundaries(
    ax,
    df
):

    start = 0

    for condition in CONDITIONS:

        count = len(
            df[
                df["condition"]
                == condition
            ]
        )

        if count == 0:
            continue

        end = (
            start
            + count
            - 1
        )

        center = (
            start
            + end
        ) / 2

        ax.axvline(
            start - 0.5,
            linestyle="--",
            alpha=0.45,
            linewidth=1
        )

        ax.text(
            center,
            0.98,
            condition.replace(
                "_",
                " "
            ).upper(),
            transform=ax.get_xaxis_transform(),
            ha="center",
            va="top",
            fontsize=11,
            fontweight="bold"
        )

        start = end + 1


# ============================================================
# 17. MUSE GRAPH
# ============================================================

fig, ax = plt.subplots(
    figsize=(22, 10)
)


ax.plot(
    muse["visual_index"],
    muse["noise_score"],
    marker="o",
    markersize=3,
    linewidth=0.8,
    alpha=0.8,
    label="Muse noise score"
)


# ------------------------------------------------------------
# Add median reference
# ------------------------------------------------------------

median_noise = muse[
    "noise_score"
].median()


ax.axhline(
    median_noise,
    linestyle="--",
    linewidth=1.5,
    label=(
        f"Median noise score = "
        f"{median_noise:,.0f}"
    )
)


add_condition_boundaries(
    ax,
    muse
)


ax.set_title(
    "CLEAN_DATA_V9: Muse Noise Structure Across All 240 Files",
    fontsize=17
)


ax.set_xlabel(
    "Muse CSV File Index",
    fontsize=13
)


ax.set_ylabel(
    "Detected Noise Points",
    fontsize=13
)


ax.grid(
    alpha=0.25
)


ax.legend()


plt.tight_layout()


muse_output = (
    OUTPUT_ROOT
    / "V9_Muse_Noise_Structure.png"
)


plt.savefig(
    muse_output,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 18. MOTION WATCH GRAPH
# ============================================================

fig, ax = plt.subplots(
    figsize=(22, 10)
)


ax.plot(
    motion["visual_index"],
    motion["noise_score"],
    marker="o",
    markersize=3,
    linewidth=0.8,
    alpha=0.8,
    label="Motion Watch noise score"
)


median_noise = motion[
    "noise_score"
].median()


ax.axhline(
    median_noise,
    linestyle="--",
    linewidth=1.5,
    label=(
        f"Median noise score = "
        f"{median_noise:,.0f}"
    )
)


add_condition_boundaries(
    ax,
    motion
)


ax.set_title(
    "CLEAN_DATA_V9: Motion Watch Noise Structure Across All 240 Files",
    fontsize=17
)


ax.set_xlabel(
    "Motion Watch CSV File Index",
    fontsize=13
)


ax.set_ylabel(
    "Detected Noise Points",
    fontsize=13
)


ax.grid(
    alpha=0.25
)


ax.legend()


plt.tight_layout()


motion_output = (
    OUTPUT_ROOT
    / "V9_MotionWatch_Noise_Structure.png"
)


plt.savefig(
    motion_output,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 19. SAVE ANALYSIS REPORT
# ============================================================

report_path = (
    OUTPUT_ROOT.parent
    / "reports"
    / "V9_noise_structure_report.csv"
)


report_path.parent.mkdir(
    parents=True,
    exist_ok=True
)


results.to_csv(
    report_path,
    index=False
)


# ============================================================
# 20. PRINT SUMMARY
# ============================================================

print()
print("=" * 90)
print("NOISE STRUCTURE ANALYSIS COMPLETED")
print("=" * 90)

print()

print(
    "Muse files:",
    len(muse)
)

print(
    "Motion Watch files:",
    len(motion)
)

print()

print(
    "Muse total detected noise:",
    f"{muse['noise_score'].sum():,}"
)

print(
    "Motion Watch total detected noise:",
    f"{motion['noise_score'].sum():,}"
)

print()

print(
    "Muse median noise:",
    f"{muse['noise_score'].median():,.0f}"
)

print(
    "Motion Watch median noise:",
    f"{motion['noise_score'].median():,.0f}"
)

print()

print(
    "Muse graph:",
    muse_output
)

print(
    "Motion Watch graph:",
    motion_output
)

print(
    "Detailed report:",
    report_path
)

print()

print("=" * 90)
print("DONE")
print("=" * 90)
```

```python
# ============================================================
# CLEAN_DATA_V10
# TARGETED MUSE NOISE REDUCTION
#
# Input:
#   CLEAN_DATA_V9
#
# Output:
#   CLEAN_DATA_V10
#
# Processing:
#   Muse:
#       V9 -> Hampel detection -> local median replacement
#
#   Motion Watch:
#       V9 -> copied unchanged
#
# No rows are deleted.
# No metadata is modified.
# No files are removed.
# ============================================================


from pathlib import Path
import shutil

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V9/clean_data"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/clean_data"
)

REPORT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/reports"
)

VIS_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/visualizations"
)


OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

VIS_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. HAMPEL FILTER PARAMETERS
# ============================================================

# Number of samples on each side of the current point.
#
# Total rolling window:
#
#     2 * HALF_WINDOW + 1
#
# With HALF_WINDOW = 3:
#
#     7 sample window
#
HALF_WINDOW = 3


# Hampel threshold.
#
# 3.5 is a commonly used robust threshold.
#
# A higher value is more conservative.
#
THRESHOLD = 3.5


# Minimum MAD value.
#
# Prevents division by extremely small values.
#
MIN_MAD = 1e-12


# ============================================================
# 3. CONDITIONS
# ============================================================

CONDITIONS = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]


# ============================================================
# 4. FOLDER IDENTIFICATION
# ============================================================

MUSE_FOLDER_NAMES = {
    "clms",
    "mums",
    "norms",
    "rdms"
}


MOTION_FOLDER_NAMES = {
    "clmw",
    "mumw",
    "normw",
    "rdmw"
}


def identify_sensor(file_path):

    folder_names = {
        part.lower()
        for part in file_path.parts
    }

    if folder_names.intersection(
        MUSE_FOLDER_NAMES
    ):
        return "Muse"

    if folder_names.intersection(
        MOTION_FOLDER_NAMES
    ):
        return "Motion Watch"

    return None


# ============================================================
# 5. COLUMNS THAT MUST NOT BE CLEANED
# ============================================================

NON_SIGNAL_COLUMNS = {

    "Time",
    "time",

    "Timestamp",
    "timestamp",

    "Date",
    "date",

    "Index",
    "index",

    "Sample",
    "sample",

    "Person Serial",
    "Person_Serial",

    "Category",
    "category",

    "Condition",
    "condition",

    "Activity",
    "activity",

    "Light",
    "light",

    "Battery",
    "battery",

    "HeadBandOn",

    "HSI_TP9",
    "HSI_AF7",
    "HSI_AF8",
    "HSI_TP10"
}


# ============================================================
# 6. FIND MUSE SIGNAL COLUMNS
# ============================================================

def get_muse_signal_columns(df):

    signal_columns = []

    for column in df.columns:

        # Only numerical columns
        if not pd.api.types.is_numeric_dtype(
            df[column]
        ):
            continue


        # Do not touch metadata
        if column in NON_SIGNAL_COLUMNS:
            continue


        signal_columns.append(
            column
        )


    return signal_columns


# ============================================================
# 7. HAMPEL FILTER
# ============================================================

def hampel_replace(
    series,
    half_window=3,
    threshold=3.5
):

    values = pd.to_numeric(
        series,
        errors="coerce"
    ).astype(float)


    original = values.copy()


    rolling_window = (
        2 * half_window
        + 1
    )


    # --------------------------------------------------------
    # Rolling median
    # --------------------------------------------------------

    rolling_median = (
        values
        .rolling(
            window=rolling_window,
            center=True,
            min_periods=1
        )
        .median()
    )


    # --------------------------------------------------------
    # Absolute deviation from local median
    # --------------------------------------------------------

    absolute_deviation = (
        values
        - rolling_median
    ).abs()


    # --------------------------------------------------------
    # Rolling MAD
    # --------------------------------------------------------

    rolling_mad = (
        absolute_deviation
        .rolling(
            window=rolling_window,
            center=True,
            min_periods=1
        )
        .median()
    )


    # --------------------------------------------------------
    # Robust scale
    #
    # 1.4826 converts MAD approximately to
    # standard deviation scale for normal data.
    # --------------------------------------------------------

    robust_scale = (
        1.4826
        * rolling_mad
    )


    # --------------------------------------------------------
    # Hampel threshold
    # --------------------------------------------------------

    threshold_value = (
        threshold
        * robust_scale
    )


    # --------------------------------------------------------
    # Detect abnormal values
    # --------------------------------------------------------

    outlier_mask = (
        absolute_deviation
        > threshold_value
    )


    # --------------------------------------------------------
    # Avoid falsely detecting values when MAD is zero
    # --------------------------------------------------------

    outlier_mask &= (
        rolling_mad
        > MIN_MAD
    )


    # --------------------------------------------------------
    # Never replace missing values
    # --------------------------------------------------------

    outlier_mask &= (
        values.notna()
    )


    # --------------------------------------------------------
    # Replace ONLY detected outliers
    # with local median
    # --------------------------------------------------------

    cleaned = values.copy()

    cleaned.loc[
        outlier_mask
    ] = rolling_median.loc[
        outlier_mask
    ]


    replaced_count = int(
        outlier_mask.sum()
    )


    return (
        cleaned,
        replaced_count
    )


# ============================================================
# 8. PROCESS ONE MUSE FILE
# ============================================================

def process_muse_file(
    input_file,
    output_file
):

    df = pd.read_csv(
        input_file,
        low_memory=False
    )


    original_rows = len(df)


    signal_columns = (
        get_muse_signal_columns(
            df
        )
    )


    total_replaced = 0


    column_replacements = {}


    # --------------------------------------------------------
    # Process every Muse signal column
    # --------------------------------------------------------

    for column in signal_columns:

        cleaned_series, count = (
            hampel_replace(
                df[column],
                half_window=HALF_WINDOW,
                threshold=THRESHOLD
            )
        )


        df[column] = cleaned_series


        column_replacements[
            column
        ] = count


        total_replaced += count


    # --------------------------------------------------------
    # Row validation
    # --------------------------------------------------------

    if len(df) != original_rows:

        raise RuntimeError(
            f"ROW COUNT CHANGED: {input_file}"
        )


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    df.to_csv(
        output_file,
        index=False
    )


    return {
        "sensor": "Muse",
        "file": str(
            input_file.relative_to(
                INPUT_ROOT
            )
        ),
        "rows_v9": original_rows,
        "rows_v10": len(df),
        "signal_columns": len(
            signal_columns
        ),
        "replaced_values": total_replaced,
        "status": "Processed"
    }


# ============================================================
# 9. COPY MOTION WATCH
# ============================================================

def copy_motion_file(
    input_file,
    output_file
):

    df = pd.read_csv(
        input_file,
        low_memory=False
    )


    original_rows = len(df)


    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # Motion Watch is copied without modification.
    # --------------------------------------------------------

    df.to_csv(
        output_file,
        index=False
    )


    if len(df) != original_rows:

        raise RuntimeError(
            f"ROW COUNT CHANGED: {input_file}"
        )


    return {
        "sensor": "Motion Watch",
        "file": str(
            input_file.relative_to(
                INPUT_ROOT
            )
        ),
        "rows_v9": original_rows,
        "rows_v10": len(df),
        "signal_columns": 0,
        "replaced_values": 0,
        "status": "Copied unchanged"
    }


# ============================================================
# 10. PROCESS ALL CSV FILES
# ============================================================

all_files = sorted(
    INPUT_ROOT.rglob(
        "*.csv"
    )
)


print()
print("=" * 100)
print("CLEAN_DATA_V10 PROCESSING")
print("=" * 100)
print()

print(
    "Input CSV files:",
    len(all_files)
)

print()


results = []


for file_number, input_file in enumerate(
    all_files,
    start=1
):

    sensor = identify_sensor(
        input_file
    )


    if sensor is None:

        print(
            f"[SKIP] Unknown file: "
            f"{input_file}"
        )

        continue


    relative_path = (
        input_file.relative_to(
            INPUT_ROOT
        )
    )


    output_file = (
        OUTPUT_ROOT
        / relative_path
    )


    print(
        f"[{file_number}/{len(all_files)}] "
        f"{sensor:15s} "
        f"{relative_path}"
    )


    try:

        if sensor == "Muse":

            result = process_muse_file(
                input_file,
                output_file
            )

        else:

            result = copy_motion_file(
                input_file,
                output_file
            )


        results.append(
            result
        )


        print(
            f"       Replaced: "
            f"{result['replaced_values']:,}"
        )


    except Exception as e:

        print(
            f"       ERROR: {e}"
        )

        results.append({

            "sensor": sensor,

            "file": str(
                relative_path
            ),

            "rows_v9": np.nan,

            "rows_v10": np.nan,

            "signal_columns": np.nan,

            "replaced_values": np.nan,

            "status":
                f"ERROR: {e}"
        })


# ============================================================
# 11. SAVE FILE LEVEL REPORT
# ============================================================

report_df = pd.DataFrame(
    results
)


file_report = (
    REPORT_ROOT
    / "V10_file_processing_report.csv"
)


report_df.to_csv(
    file_report,
    index=False
)


# ============================================================
# 12. SUMMARY
# ============================================================

muse_report = report_df[
    report_df["sensor"]
    == "Muse"
].copy()


motion_report = report_df[
    report_df["sensor"]
    == "Motion Watch"
].copy()


muse_replaced = int(
    muse_report[
        "replaced_values"
    ]
    .fillna(0)
    .sum()
)


motion_replaced = int(
    motion_report[
        "replaced_values"
    ]
    .fillna(0)
    .sum()
)


print()
print("=" * 100)
print("V10 PROCESSING SUMMARY")
print("=" * 100)
print()


print(
    "Muse files:",
    len(muse_report)
)


print(
    "Motion Watch files:",
    len(motion_report)
)


print()


print(
    "Muse values replaced:",
    f"{muse_replaced:,}"
)


print(
    "Motion Watch values replaced:",
    f"{motion_replaced:,}"
)


print()


print(
    "V10 output:",
    OUTPUT_ROOT
)


print(
    "File report:",
    file_report
)


# ============================================================
# 13. ROW COUNT VALIDATION
# ============================================================

print()
print("=" * 100)
print("ROW COUNT VALIDATION")
print("=" * 100)
print()


row_errors = []


for _, row in report_df.iterrows():

    if (
        pd.notna(row["rows_v9"])
        and
        pd.notna(row["rows_v10"])
    ):

        if (
            int(row["rows_v9"])
            !=
            int(row["rows_v10"])
        ):

            row_errors.append(
                row["file"]
            )


if len(row_errors) == 0:

    print(
        "ROW COUNT VALIDATION: PASSED"
    )

    print(
        "All V10 files have the same "
        "number of rows as V9."
    )

else:

    print(
        "ROW COUNT VALIDATION: FAILED"
    )

    print(
        "Files with row count errors:"
    )

    for file_name in row_errors:

        print(
            file_name
        )


# ============================================================
# 14. VERIFY EXPECTED FILE COUNTS
# ============================================================

v10_files = sorted(
    OUTPUT_ROOT.rglob(
        "*.csv"
    )
)


v10_muse = [
    f for f in v10_files
    if identify_sensor(f)
    == "Muse"
]


v10_motion = [
    f for f in v10_files
    if identify_sensor(f)
    == "Motion Watch"
]


print()
print("=" * 100)
print("FILE COUNT VALIDATION")
print("=" * 100)
print()


print(
    "V10 total CSV files:",
    len(v10_files)
)


print(
    "V10 Muse files:",
    len(v10_muse)
)


print(
    "V10 Motion Watch files:",
    len(v10_motion)
)


# ============================================================
# 15. CREATE SUMMARY REPORT
# ============================================================

summary = pd.DataFrame({

    "Metric": [

        "V9 total CSV files",

        "V10 total CSV files",

        "V9 Muse files",

        "V10 Muse files",

        "V9 Motion Watch files",

        "V10 Motion Watch files",

        "Muse values replaced",

        "Motion Watch values replaced",

        "Hampel half window",

        "Hampel total window",

        "Hampel threshold",

        "Row count errors"

    ],

    "Value": [

        len(all_files),

        len(v10_files),

        len(muse_report),

        len(v10_muse),

        len(motion_report),

        len(v10_motion),

        muse_replaced,

        motion_replaced,

        HALF_WINDOW,

        2 * HALF_WINDOW + 1,

        THRESHOLD,

        len(row_errors)

    ]

})


summary_file = (
    REPORT_ROOT
    / "V10_summary.csv"
)


summary.to_csv(
    summary_file,
    index=False
)


# ============================================================
# 16. CREATE V10 MUSE NOISE VISUALIZATION
#
# Recalculate the same Hampel criterion after cleaning.
# ============================================================

def calculate_remaining_noise(
    df,
    columns
):

    total = 0


    for column in columns:

        values = pd.to_numeric(
            df[column],
            errors="coerce"
        ).astype(float)


        rolling_window = (
            2 * HALF_WINDOW + 1
        )


        median = (
            values
            .rolling(
                rolling_window,
                center=True,
                min_periods=1
            )
            .median()
        )


        deviation = (
            values - median
        ).abs()


        mad = (
            deviation
            .rolling(
                rolling_window,
                center=True,
                min_periods=1
            )
            .median()
        )


        robust_scale = (
            1.4826 * mad
        )


        mask = (
            deviation
            >
            THRESHOLD
            * robust_scale
        )


        mask &= (
            mad > MIN_MAD
        )


        mask &= values.notna()


        total += int(
            mask.sum()
        )


    return total


# ============================================================
# 17. CALCULATE V10 REMAINING NOISE
# ============================================================

noise_records = []


for file_path in v10_muse:

    df = pd.read_csv(
        file_path,
        low_memory=False
    )


    columns = (
        get_muse_signal_columns(
            df
        )
    )


    noise = (
        calculate_remaining_noise(
            df,
            columns
        )
    )


    relative_file = str(
        file_path.relative_to(
            OUTPUT_ROOT
        )
    )


    noise_records.append({

        "file": relative_file,

        "remaining_noise": noise

    })


noise_df = pd.DataFrame(
    noise_records
)


noise_report = (
    REPORT_ROOT
    / "V10_Muse_Remaining_Noise.csv"
)


noise_df.to_csv(
    noise_report,
    index=False
)


# ============================================================
# 18. VISUALIZE V10 REMAINING NOISE
# ============================================================

noise_df[
    "file_index"
] = np.arange(
    len(noise_df)
)


fig, ax = plt.subplots(
    figsize=(22, 10)
)


ax.plot(
    noise_df["file_index"],
    noise_df["remaining_noise"],
    marker="o",
    markersize=3,
    linewidth=0.8
)


ax.set_title(
    "CLEAN_DATA_V10: Remaining Muse Noise After Targeted Cleaning",
    fontsize=17
)


ax.set_xlabel(
    "Muse CSV File Index",
    fontsize=13
)


ax.set_ylabel(
    "Remaining Detected Noise Points",
    fontsize=13
)


ax.grid(
    alpha=0.25
)


plt.tight_layout()


v10_graph = (
    VIS_ROOT
    / "V10_Muse_Remaining_Noise.png"
)


plt.savefig(
    v10_graph,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print()
print("=" * 100)
print("CLEAN_DATA_V10 COMPLETED")
print("=" * 100)
print()


print(
    "Total V10 CSV files:",
    len(v10_files)
)


print(
    "Muse files:",
    len(v10_muse)
)


print(
    "Motion Watch files:",
    len(v10_motion)
)


print()


print(
    "Total Muse values replaced:",
    f"{muse_replaced:,}"
)


print(
    "Total Motion Watch values replaced:",
    f"{motion_replaced:,}"
)


print()


print(
    "V10 remaining Muse noise:",
    f"{noise_df['remaining_noise'].sum():,}"
)


print()


print(
    "V10 dataset:",
    OUTPUT_ROOT
)


print(
    "Processing report:",
    file_report
)


print(
    "Summary:",
    summary_file
)


print(
    "Remaining noise report:",
    noise_report
)


print(
    "V10 noise graph:",
    v10_graph
)


print()


print("=" * 100)
print("DONE")
print("=" * 100)
```

```python
# ================================================================
# CLEAN_DATA_V10
# Muse Channel Audit
# Signal Quality Statistics
# Artifact Identification
# Appropriate EEG Filtering
# Final Clean Dataset
#
# Output:
# CLEAN_DATA_V11
# ================================================================

from pathlib import Path
import shutil
import warnings
import re

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from scipy.signal import (
    butter,
    sosfiltfilt,
    iirnotch,
    filtfilt,
    savgol_filter
)

warnings.filterwarnings("ignore")


# ================================================================
# CONFIGURATION
# ================================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/clean_data"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V11"
)

OUTPUT_DATA = OUTPUT_ROOT / "clean_data"
REPORT_DIR = OUTPUT_ROOT / "reports"
VIS_DIR = OUTPUT_ROOT / "visualizations"

OUTPUT_DATA.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)
VIS_DIR.mkdir(parents=True, exist_ok=True)


# ================================================================
# DATASET INFORMATION
# ================================================================

print("=" * 100)
print("CLEAN_DATA_V11 PROCESSING")
print("=" * 100)

print(f"Input dataset : {INPUT_ROOT}")
print(f"Output dataset: {OUTPUT_DATA}")


# ================================================================
# MUSE COLUMN DEFINITIONS
# ================================================================

REMOVED_MUSE_COLUMNS = {
    "HeadBandOn",
    "HSI_TP9",
    "HSI_AF7",
    "HSI_AF8",
    "HSI_TP10",
    "Battery"
}


RAW_EEG_PATTERNS = [
    r"^RAW[_ ]?TP9$",
    r"^RAW[_ ]?AF7$",
    r"^RAW[_ ]?AF8$",
    r"^RAW[_ ]?TP10$",
    r"^TP9$",
    r"^AF7$",
    r"^AF8$",
    r"^TP10$"
]


BAND_PATTERNS = [
    r"delta",
    r"theta",
    r"alpha",
    r"beta",
    r"gamma"
]


NON_SIGNAL_PATTERNS = [
    r"timestamp",
    r"datetime",
    r"date",
    r"time",
    r"serial",
    r"person",
    r"participant",
    r"subject",
    r"class",
    r"label",
    r"activity",
    r"category",
    r"condition",
    r"session",
    r"device",
    r"status",
    r"battery",
    r"headband",
    r"hsi",
    r"aux"
]


# ================================================================
# HELPER FUNCTIONS
# ================================================================

def normalize_name(name):
    return str(name).strip().lower().replace(" ", "_")


def is_raw_eeg_column(column):
    name = str(column).strip()

    for pattern in RAW_EEG_PATTERNS:
        if re.search(pattern, name, flags=re.IGNORECASE):
            return True

    return False


def is_band_column(column):
    name = normalize_name(column)

    for pattern in BAND_PATTERNS:
        if re.search(pattern, name, flags=re.IGNORECASE):
            return True

    return False


def is_non_signal_column(column):
    name = normalize_name(column)

    for pattern in NON_SIGNAL_PATTERNS:
        if re.search(pattern, name, flags=re.IGNORECASE):
            return True

    return False


def detect_muse_signal_columns(df):

    raw_columns = []
    band_columns = []
    other_numeric_columns = []

    for column in df.columns:

        if column in REMOVED_MUSE_COLUMNS:
            continue

        if is_raw_eeg_column(column):
            raw_columns.append(column)
            continue

        if is_band_column(column):
            band_columns.append(column)
            continue

        if pd.api.types.is_numeric_dtype(df[column]):

            if not is_non_signal_column(column):
                other_numeric_columns.append(column)

    return raw_columns, band_columns, other_numeric_columns


def safe_numeric_series(series):

    return pd.to_numeric(
        series,
        errors="coerce"
    )


def robust_z_score(series):

    x = safe_numeric_series(series)

    median_value = x.median()

    mad_value = np.median(
        np.abs(
            x.dropna() - median_value
        )
    )

    if not np.isfinite(mad_value) or mad_value == 0:

        return pd.Series(
            np.zeros(len(series)),
            index=series.index
        )

    robust_sigma = 1.4826 * mad_value

    return (
        (x - median_value)
        / robust_sigma
    )


def detect_artifacts(series, threshold=6.0):

    x = safe_numeric_series(series)

    result = pd.Series(
        False,
        index=series.index
    )

    finite_mask = np.isfinite(x)

    if finite_mask.sum() < 10:
        return result

    z = robust_z_score(x)

    result.loc[finite_mask] = (
        np.abs(z.loc[finite_mask]) > threshold
    )

    return result


def detect_flatline(series, window=10):

    x = safe_numeric_series(series)

    if len(x) < window:
        return pd.Series(
            False,
            index=series.index
        )

    rolling_std = (
        x.rolling(
            window=window,
            center=True,
            min_periods=max(3, window // 2)
        )
        .std()
    )

    flat = (
        rolling_std.notna()
        &
        (rolling_std == 0)
    )

    return flat


def interpolate_signal(series):

    x = pd.to_numeric(
        series,
        errors="coerce"
    )

    x = x.replace(
        [np.inf, -np.inf],
        np.nan
    )

    x = x.interpolate(
        method="linear",
        limit_direction="both"
    )

    if x.isna().any():

        median_value = x.median()

        if np.isfinite(median_value):
            x = x.fillna(median_value)

    return x


def apply_temporal_smoothing(series):

    x = interpolate_signal(series)

    n = len(x)

    if n < 7:
        return x

    window = min(
        11,
        n if n % 2 == 1 else n - 1
    )

    if window < 5:
        return x

    try:

        smoothed = savgol_filter(
            x.to_numpy(dtype=float),
            window_length=window,
            polyorder=2,
            mode="interp"
        )

        return pd.Series(
            smoothed,
            index=series.index
        )

    except Exception:

        return x


def apply_eeg_bandpass(
    series,
    sampling_rate=256.0,
    low_cut=1.0,
    high_cut=45.0
):

    x = interpolate_signal(series)

    if len(x) < 100:
        return x

    nyquist = sampling_rate / 2.0

    if high_cut >= nyquist:
        high_cut = nyquist * 0.90

    if low_cut <= 0:
        low_cut = 0.5

    if low_cut >= high_cut:
        return x

    try:

        sos = butter(
            4,
            [
                low_cut / nyquist,
                high_cut / nyquist
            ],
            btype="bandpass",
            output="sos"
        )

        filtered = sosfiltfilt(
            sos,
            x.to_numpy(dtype=float)
        )

        return pd.Series(
            filtered,
            index=series.index
        )

    except Exception:

        return x


def apply_notch_filter(
    series,
    sampling_rate=256.0,
    notch_frequency=50.0
):

    x = interpolate_signal(series)

    if len(x) < 100:
        return x

    if notch_frequency >= sampling_rate / 2:
        return x

    try:

        b, a = iirnotch(
            notch_frequency,
            Q=30.0,
            fs=sampling_rate
        )

        filtered = filtfilt(
            b,
            a,
            x.to_numpy(dtype=float)
        )

        return pd.Series(
            filtered,
            index=series.index
        )

    except Exception:

        return x


def calculate_quality_statistics(
    series,
    artifact_mask
):

    x = safe_numeric_series(series)

    finite_values = x[np.isfinite(x)]

    if len(finite_values) == 0:

        return {
            "count": len(series),
            "missing": int(x.isna().sum()),
            "infinite": int(
                np.isinf(x).sum()
            ),
            "artifact_count": int(
                artifact_mask.sum()
            ),
            "artifact_percentage": 0.0,
            "minimum": np.nan,
            "maximum": np.nan,
            "mean": np.nan,
            "median": np.nan,
            "std": np.nan,
            "unique_values": 0
        }

    return {
        "count": len(series),

        "missing": int(
            x.isna().sum()
        ),

        "infinite": int(
            np.isinf(x).sum()
        ),

        "artifact_count": int(
            artifact_mask.sum()
        ),

        "artifact_percentage": (
            artifact_mask.sum()
            / len(series)
            * 100
        ),

        "minimum": float(
            finite_values.min()
        ),

        "maximum": float(
            finite_values.max()
        ),

        "mean": float(
            finite_values.mean()
        ),

        "median": float(
            finite_values.median()
        ),

        "std": float(
            finite_values.std()
        ),

        "unique_values": int(
            finite_values.nunique()
        )
    }


# ================================================================
# FIND ALL CSV FILES
# ================================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

print()
print(
    f"CSV files found: {len(csv_files)}"
)

if len(csv_files) == 0:

    raise FileNotFoundError(
        f"No CSV files found in {INPUT_ROOT}"
    )


# ================================================================
# REPORT STORAGE
# ================================================================

audit_records = []
quality_records = []
artifact_records = []
processing_records = []

all_muse_columns = set()

muse_file_count = 0
motion_file_count = 0


# ================================================================
# FIRST PASS
# CHANNEL AND VALUE AUDIT
# ================================================================

print()
print("=" * 100)
print("MUSE CHANNEL AND VALUE AUDIT")
print("=" * 100)


for index, csv_path in enumerate(csv_files, start=1):

    relative_path = csv_path.relative_to(
        INPUT_ROOT
    )

    try:

        df = pd.read_csv(
            csv_path,
            low_memory=False
        )

    except Exception as error:

        print(
            f"Could not read {csv_path}: {error}"
        )

        continue


    is_muse = any(
        is_raw_eeg_column(column)
        or is_band_column(column)
        for column in df.columns
    )


    if not is_muse:

        motion_file_count += 1

        continue


    muse_file_count += 1


    raw_columns, band_columns, other_numeric = (
        detect_muse_signal_columns(df)
    )


    signal_columns = (
        raw_columns
        +
        band_columns
        +
        other_numeric
    )


    all_muse_columns.update(
        signal_columns
    )


    missing_columns = [
        column
        for column in REMOVED_MUSE_COLUMNS
        if column in df.columns
    ]


    total_missing = int(
        df.isna().sum().sum()
    )


    total_infinite = 0

    for column in df.select_dtypes(
        include=np.number
    ).columns:

        values = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        total_infinite += int(
            np.isinf(values).sum()
        )


    duplicate_rows = int(
        df.duplicated().sum()
    )


    audit_records.append({

        "file": str(relative_path),

        "rows": len(df),

        "columns": len(df.columns),

        "raw_eeg_channels": len(
            raw_columns
        ),

        "band_channels": len(
            band_columns
        ),

        "other_numeric_channels": len(
            other_numeric
        ),

        "signal_channels": len(
            signal_columns
        ),

        "removed_columns_present": len(
            missing_columns
        ),

        "missing_values": total_missing,

        "infinite_values": total_infinite,

        "duplicate_rows": duplicate_rows

    })


    for column in signal_columns:

        artifact_mask = detect_artifacts(
            df[column]
        )

        flatline_mask = detect_flatline(
            df[column]
        )

        combined_mask = (
            artifact_mask
            |
            flatline_mask
        )


        statistics = calculate_quality_statistics(
            df[column],
            combined_mask
        )


        quality_records.append({

            "file": str(relative_path),

            "column": column,

            "signal_type": (
                "RAW_EEG"
                if column in raw_columns
                else
                "BAND"
                if column in band_columns
                else
                "OTHER_NUMERIC"
            ),

            **statistics

        })


        artifact_records.append({

            "file": str(relative_path),

            "column": column,

            "extreme_artifacts": int(
                artifact_mask.sum()
            ),

            "flatline_points": int(
                flatline_mask.sum()
            ),

            "total_detected": int(
                combined_mask.sum()
            )

        })


print(
    f"Muse files audited: {muse_file_count}"
)

print(
    f"Motion Watch files: {motion_file_count}"
)

print(
    f"Muse signal columns identified: "
    f"{len(all_muse_columns)}"
)


# ================================================================
# SAVE AUDIT REPORTS
# ================================================================

audit_df = pd.DataFrame(
    audit_records
)

quality_df = pd.DataFrame(
    quality_records
)

artifact_df = pd.DataFrame(
    artifact_records
)


audit_df.to_csv(
    REPORT_DIR / "V11_Muse_Channel_Audit.csv",
    index=False
)

quality_df.to_csv(
    REPORT_DIR / "V11_Muse_Signal_Quality.csv",
    index=False
)

artifact_df.to_csv(
    REPORT_DIR / "V11_Muse_Artifact_Report.csv",
    index=False
)


# ================================================================
# VISUALIZATION 1
# MISSING VALUES
# ================================================================

if not quality_df.empty:

    missing_by_column = (
        quality_df
        .groupby("column")["missing"]
        .sum()
        .sort_values(
            ascending=False
        )
    )

    plt.figure(
        figsize=(16, 8)
    )

    missing_by_column.plot(
        kind="bar"
    )

    plt.title(
        "Muse Missing Values by Signal Channel"
    )

    plt.xlabel(
        "Muse Signal Channel"
    )

    plt.ylabel(
        "Missing Values"
    )

    plt.xticks(
        rotation=90
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "V11_Muse_Missing_Values.png",
        dpi=200
    )

    plt.close()


# ================================================================
# VISUALIZATION 2
# ARTIFACT COUNTS BY CHANNEL
# ================================================================

if not artifact_df.empty:

    artifact_by_column = (
        artifact_df
        .groupby("column")["total_detected"]
        .sum()
        .sort_values(
            ascending=False
        )
    )

    plt.figure(
        figsize=(16, 8)
    )

    artifact_by_column.plot(
        kind="bar"
    )

    plt.title(
        "Muse Detected Artifact Points by Channel"
    )

    plt.xlabel(
        "Muse Signal Channel"
    )

    plt.ylabel(
        "Detected Artifact Points"
    )

    plt.xticks(
        rotation=90
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "V11_Muse_Artifact_By_Channel.png",
        dpi=200
    )

    plt.close()


# ================================================================
# VISUALIZATION 3
# ARTIFACT HEATMAP
# ================================================================

if not artifact_df.empty:

    artifact_matrix = (
        artifact_df
        .pivot_table(
            index="file",
            columns="column",
            values="total_detected",
            aggfunc="sum",
            fill_value=0
        )
    )


    plt.figure(
        figsize=(20, 14)
    )

    plt.imshow(
        artifact_matrix,
        aspect="auto",
        interpolation="nearest"
    )

    plt.colorbar(
        label="Detected Artifact Points"
    )

    plt.title(
        "Muse Artifact Structure Across All Files"
    )

    plt.xlabel(
        "Muse Signal Channel"
    )

    plt.ylabel(
        "CSV File"
    )

    plt.xticks(
        range(
            len(
                artifact_matrix.columns
            )
        ),
        artifact_matrix.columns,
        rotation=90
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "V11_Muse_Artifact_Heatmap.png",
        dpi=200
    )

    plt.close()


# ================================================================
# CREATE FINAL DATASET
# ================================================================

print()
print("=" * 100)
print("CREATING CLEAN_DATA_V11")
print("=" * 100)


total_replaced = 0
total_artifacts_before = 0
total_artifacts_after = 0
total_rows_input = 0
total_rows_output = 0


# Used for representative visualization
before_visual = None
after_visual = None
visual_file_name = None


for index, csv_path in enumerate(
    csv_files,
    start=1
):

    relative_path = csv_path.relative_to(
        INPUT_ROOT
    )

    output_path = (
        OUTPUT_DATA
        /
        relative_path
    )

    output_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    df = pd.read_csv(
        csv_path,
        low_memory=False
    )


    total_rows_input += len(df)


    is_muse = any(
        is_raw_eeg_column(column)
        or is_band_column(column)
        for column in df.columns
    )


    # ============================================================
    # MOTION WATCH
    # COPY WITHOUT MODIFICATION
    # ============================================================

    if not is_muse:

        df.to_csv(
            output_path,
            index=False
        )

        total_rows_output += len(df)

        processing_records.append({

            "file": str(relative_path),

            "type": "Motion Watch",

            "rows_input": len(df),

            "rows_output": len(df),

            "artifacts_before": 0,

            "values_replaced": 0,

            "artifacts_after": 0,

            "processing": "Unchanged"

        })

        continue


    # ============================================================
    # MUSE
    # ============================================================

    raw_columns, band_columns, other_numeric = (
        detect_muse_signal_columns(df)
    )


    signal_columns = (
        raw_columns
        +
        band_columns
        +
        other_numeric
    )


    file_artifacts_before = 0
    file_replaced = 0
    file_artifacts_after = 0


    # ============================================================
    # REMOVE UNWANTED DEVICE COLUMNS
    # ============================================================

    columns_to_remove = [
        column
        for column in REMOVED_MUSE_COLUMNS
        if column in df.columns
    ]

    if columns_to_remove:

        df = df.drop(
            columns=columns_to_remove
        )


    # ============================================================
    # PROCESS EACH MUSE SIGNAL
    # ============================================================

    for column in signal_columns:

        if column not in df.columns:
            continue


        original = pd.to_numeric(
            df[column],
            errors="coerce"
        )


        # ========================================================
        # INVALID VALUE IDENTIFICATION
        # ========================================================

        invalid_mask = (
            original.isna()
            |
            ~np.isfinite(
                original
            )
        )


        # ========================================================
        # ROBUST ARTIFACT IDENTIFICATION
        # ========================================================

        extreme_mask = detect_artifacts(
            original,
            threshold=6.0
        )


        # ========================================================
        # FLATLINE IDENTIFICATION
        # ========================================================

        flatline_mask = detect_flatline(
            original,
            window=10
        )


        artifact_mask = (
            invalid_mask
            |
            extreme_mask
            |
            flatline_mask
        )


        file_artifacts_before += int(
            artifact_mask.sum()
        )


        # ========================================================
        # SAVE ORIGINAL FOR VISUALIZATION
        # ========================================================

        if (
            before_visual is None
            and len(original) > 100
        ):

            before_visual = original.copy()

            visual_file_name = str(
                relative_path
            )


        # ========================================================
        # ARTIFACT REPLACEMENT
        # ========================================================

        cleaned = original.copy()

        cleaned.loc[
            artifact_mask
        ] = np.nan


        cleaned = interpolate_signal(
            cleaned
        )


        file_replaced += int(
            artifact_mask.sum()
        )


        # ========================================================
        # FILTERING
        # ========================================================

        if column in raw_columns:

            # Raw EEG signal
            #
            # Standard EEG frequency range
            # 1 Hz to 45 Hz
            #
            # Sampling rate is assumed to be
            # 256 Hz only for raw EEG channels.

            cleaned = apply_eeg_bandpass(
                cleaned,
                sampling_rate=256.0,
                low_cut=1.0,
                high_cut=45.0
            )


            # Notch filtering
            #
            # The code checks the sampling rate
            # before applying the filter.

            cleaned = apply_notch_filter(
                cleaned,
                sampling_rate=256.0,
                notch_frequency=50.0
            )


        elif column in band_columns:

            # Band power values are not raw EEG.
            #
            # Therefore do not apply a 1 Hz to
            # 45 Hz EEG bandpass filter.
            #
            # Use temporal smoothing instead.

            cleaned = apply_temporal_smoothing(
                cleaned
            )


        else:

            cleaned = apply_temporal_smoothing(
                cleaned
            )


        # ========================================================
        # FINAL NUMERIC SAFETY CHECK
        # ========================================================

        cleaned = pd.to_numeric(
            cleaned,
            errors="coerce"
        )


        cleaned = cleaned.replace(
            [np.inf, -np.inf],
            np.nan
        )


        cleaned = interpolate_signal(
            cleaned
        )


        # ========================================================
        # POST CLEANING ARTIFACT CHECK
        # ========================================================

        post_extreme = detect_artifacts(
            cleaned,
            threshold=6.0
        )


        post_flatline = detect_flatline(
            cleaned,
            window=10
        )


        post_artifact_mask = (
            post_extreme
            |
            post_flatline
        )


        file_artifacts_after += int(
            post_artifact_mask.sum()
        )


        df[column] = cleaned


        # ========================================================
        # SAVE REPRESENTATIVE CLEANED SIGNAL
        # ========================================================

        if (
            before_visual is not None
            and after_visual is None
            and visual_file_name == str(
                relative_path
            )
        ):

            after_visual = cleaned.copy()


    # ============================================================
    # SAVE MUSE FILE
    # ============================================================

    df.to_csv(
        output_path,
        index=False
    )


    total_rows_output += len(df)

    total_artifacts_before += (
        file_artifacts_before
    )

    total_replaced += (
        file_replaced
    )

    total_artifacts_after += (
        file_artifacts_after
    )


    processing_records.append({

        "file": str(relative_path),

        "type": "Muse",

        "rows_input": len(
            pd.read_csv(
                csv_path,
                usecols=[0]
            )
        ),

        "rows_output": len(df),

        "artifacts_before":
            file_artifacts_before,

        "values_replaced":
            file_replaced,

        "artifacts_after":
            file_artifacts_after,

        "processing":
            "Artifact correction and signal filtering"

    })


    print(
        f"[{index}/{len(csv_files)}] "
        f"Muse {relative_path} "
        f"Artifacts: {file_artifacts_before:,} "
        f"Replaced: {file_replaced:,} "
        f"Remaining: {file_artifacts_after:,}"
    )


# ================================================================
# PROCESSING REPORT
# ================================================================

processing_df = pd.DataFrame(
    processing_records
)

processing_df.to_csv(
    REPORT_DIR /
    "V11_File_Processing_Report.csv",
    index=False
)


# ================================================================
# FINAL VISUALIZATION
# BEFORE VS AFTER REPRESENTATIVE SIGNAL
# ================================================================

if (
    before_visual is not None
    and after_visual is not None
):

    n = min(
        3000,
        len(before_visual),
        len(after_visual)
    )


    x = np.arange(n)


    plt.figure(
        figsize=(18, 8)
    )

    plt.plot(
        x,
        before_visual.iloc[:n].to_numpy(),
        label="V10 Muse signal",
        alpha=0.7
    )

    plt.plot(
        x,
        after_visual.iloc[:n].to_numpy(),
        label="V11 cleaned Muse signal",
        alpha=0.8
    )

    plt.title(
        "Muse Signal Before and After Artifact Correction"
    )

    plt.xlabel(
        "Data Point"
    )

    plt.ylabel(
        "Signal Value"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "V11_Muse_Before_After_Filtering.png",
        dpi=200
    )

    plt.close()


# ================================================================
# FINAL ARTIFACT STRUCTURE
# ================================================================

final_artifact_records = []


for csv_path in sorted(
    OUTPUT_DATA.rglob("*.csv")
):

    relative_path = csv_path.relative_to(
        OUTPUT_DATA
    )


    df = pd.read_csv(
        csv_path,
        low_memory=False
    )


    is_muse = any(
        is_raw_eeg_column(column)
        or is_band_column(column)
        for column in df.columns
    )


    if not is_muse:
        continue


    raw_columns, band_columns, other_numeric = (
        detect_muse_signal_columns(df)
    )


    signal_columns = (
        raw_columns
        +
        band_columns
        +
        other_numeric
    )


    for column in signal_columns:

        artifact_mask = detect_artifacts(
            df[column],
            threshold=6.0
        )


        flatline_mask = detect_flatline(
            df[column],
            window=10
        )


        total_detected = int(
            (
                artifact_mask
                |
                flatline_mask
            ).sum()
        )


        final_artifact_records.append({

            "file": str(relative_path),

            "column": column,

            "remaining_artifacts":
                total_detected

        })


final_artifact_df = pd.DataFrame(
    final_artifact_records
)


final_artifact_df.to_csv(
    REPORT_DIR /
    "V11_Final_Muse_Artifact_Report.csv",
    index=False
)


# ================================================================
# FINAL ARTIFACT VISUALIZATION
# ================================================================

if not final_artifact_df.empty:

    final_by_column = (
        final_artifact_df
        .groupby("column")[
            "remaining_artifacts"
        ]
        .sum()
        .sort_values(
            ascending=False
        )
    )


    plt.figure(
        figsize=(16, 8)
    )

    final_by_column.plot(
        kind="bar"
    )

    plt.title(
        "CLEAN_DATA_V11: Remaining Muse Artifacts"
    )

    plt.xlabel(
        "Muse Signal Channel"
    )

    plt.ylabel(
        "Remaining Detected Points"
    )

    plt.xticks(
        rotation=90
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "V11_Final_Muse_Artifacts.png",
        dpi=200
    )

    plt.close()


# ================================================================
# FINAL DATASET VALIDATION
# ================================================================

input_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

output_files = sorted(
    OUTPUT_DATA.rglob("*.csv")
)


input_relative = sorted(
    str(
        p.relative_to(INPUT_ROOT)
    )
    for p in input_files
)


output_relative = sorted(
    str(
        p.relative_to(OUTPUT_DATA)
    )
    for p in output_files
)


file_structure_match = (
    input_relative
    ==
    output_relative
)


# ================================================================
# ROW COUNT VALIDATION
# ================================================================

row_validation_records = []

row_count_errors = 0


for relative_file in input_relative:

    input_file = (
        INPUT_ROOT
        /
        relative_file
    )

    output_file = (
        OUTPUT_DATA
        /
        relative_file
    )


    try:

        input_rows = len(
            pd.read_csv(
                input_file,
                usecols=[0]
            )
        )

        output_rows = len(
            pd.read_csv(
                output_file,
                usecols=[0]
            )
        )

    except Exception:

        input_rows = -1
        output_rows = -1


    match = (
        input_rows
        ==
        output_rows
    )


    if not match:
        row_count_errors += 1


    row_validation_records.append({

        "file": relative_file,

        "input_rows": input_rows,

        "output_rows": output_rows,

        "match": match

    })


row_validation_df = pd.DataFrame(
    row_validation_records
)


row_validation_df.to_csv(
    REPORT_DIR /
    "V11_Row_Count_Validation.csv",
    index=False
)


# ================================================================
# SUMMARY
# ================================================================

remaining_artifacts = int(
    final_artifact_df[
        "remaining_artifacts"
    ].sum()
) if not final_artifact_df.empty else 0


summary = pd.DataFrame([{

    "input_dataset":
        "CLEAN_DATA_V10",

    "output_dataset":
        "CLEAN_DATA_V11",

    "total_input_csv_files":
        len(input_files),

    "total_output_csv_files":
        len(output_files),

    "muse_files":
        muse_file_count,

    "motion_watch_files":
        motion_file_count,

    "muse_artifacts_before":
        total_artifacts_before,

    "muse_values_replaced":
        total_replaced,

    "muse_artifacts_after":
        total_artifacts_after,

    "final_detected_artifacts":
        remaining_artifacts,

    "input_rows":
        total_rows_input,

    "output_rows":
        total_rows_output,

    "row_count_errors":
        row_count_errors,

    "file_structure_match":
        file_structure_match

}])


summary.to_csv(
    REPORT_DIR /
    "V11_Summary.csv",
    index=False
)


# ================================================================
# FINAL CONSOLE OUTPUT
# ================================================================

print()
print("=" * 100)
print("CLEAN_DATA_V11 PROCESSING COMPLETED")
print("=" * 100)

print(
    f"Input CSV files              : "
    f"{len(input_files)}"
)

print(
    f"Output CSV files             : "
    f"{len(output_files)}"
)

print(
    f"Muse files                   : "
    f"{muse_file_count}"
)

print(
    f"Motion Watch files           : "
    f"{motion_file_count}"
)

print()
print("MUSE")

print(
    f"Artifacts before processing : "
    f"{total_artifacts_before:,}"
)

print(
    f"Values replaced             : "
    f"{total_replaced:,}"
)

print(
    f"Artifacts after processing  : "
    f"{total_artifacts_after:,}"
)

print(
    f"Final detected artifacts    : "
    f"{remaining_artifacts:,}"
)

print()
print("ROW VALIDATION")

print(
    f"Input rows                  : "
    f"{total_rows_input:,}"
)

print(
    f"Output rows                 : "
    f"{total_rows_output:,}"
)

print(
    f"Row count errors            : "
    f"{row_count_errors}"
)

print(
    f"File structure match        : "
    f"{file_structure_match}"
)

print()
print("OUTPUT DATASET")

print(
    OUTPUT_DATA
)

print()
print("REPORTS")

print(
    REPORT_DIR
)

print()
print("VISUALIZATIONS")

print(
    VIS_DIR
)

print()
print("=" * 100)
print("DONE")
print("=" * 100)
```

```python
# ============================================================
# CLEAN_DATA_V10 → FINAL EEG DATASET
# EEG CHANNEL AUDIT, FILTERING, ARTIFACT CORRECTION
# ============================================================

import os
import glob
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import butter, sosfiltfilt, iirnotch, filtfilt
from scipy.stats import kurtosis
from sklearn.decomposition import FastICA

warnings.filterwarnings("ignore")


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/clean_data"

OUTPUT_ROOT = "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_FINAL"

OUTPUT_DATA = os.path.join(OUTPUT_ROOT, "clean_data")
REPORT_DIR = os.path.join(OUTPUT_ROOT, "reports")
GRAPH_DIR = os.path.join(OUTPUT_ROOT, "visualizations")

os.makedirs(OUTPUT_DATA, exist_ok=True)
os.makedirs(REPORT_DIR, exist_ok=True)
os.makedirs(GRAPH_DIR, exist_ok=True)


# ============================================================
# RAW EEG CHANNELS
# ============================================================

RAW_EEG_CHANNELS = [
    "RAW_TP9",
    "RAW_AF7",
    "RAW_AF8",
    "RAW_TP10"
]


# ============================================================
# EEG FILTER PARAMETERS
# ============================================================

LOW_CUTOFF = 0.5
HIGH_CUTOFF = 45.0

NOTCH_FREQUENCY = 50.0
NOTCH_Q = 30.0


# ============================================================
# ARTIFACT PARAMETERS
# ============================================================

MAD_THRESHOLD = 6.0
DIFF_MAD_THRESHOLD = 8.0

FLATLINE_LENGTH = 10

MIN_FILTER_FS = 100.0
MIN_ICA_FS = 100.0


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def mad(x):

    x = np.asarray(x, dtype=float)

    x = x[np.isfinite(x)]

    if len(x) == 0:
        return np.nan

    med = np.median(x)

    return np.median(np.abs(x - med))


def robust_z_score(x):

    x = np.asarray(x, dtype=float)

    med = np.nanmedian(x)

    mad_value = mad(x)

    if not np.isfinite(mad_value) or mad_value == 0:
        return np.zeros(len(x))

    return 0.6745 * (x - med) / mad_value


def calculate_sampling_rate(timestamp_series):

    try:

        timestamps = pd.to_datetime(
            timestamp_series,
            errors="coerce"
        )

        timestamps = timestamps.dropna()

        if len(timestamps) < 3:
            return np.nan

        differences = timestamps.diff().dt.total_seconds()

        differences = differences[
            np.isfinite(differences)
            & (differences > 0)
        ]

        if len(differences) == 0:
            return np.nan

        median_interval = differences.median()

        if median_interval <= 0:
            return np.nan

        return 1.0 / median_interval

    except Exception:

        return np.nan


def detect_flatline(x, minimum_length=10):

    x = np.asarray(x, dtype=float)

    valid = np.isfinite(x)

    flat_mask = np.zeros(len(x), dtype=bool)

    if len(x) < minimum_length:
        return flat_mask

    for i in range(len(x) - minimum_length + 1):

        segment = x[i:i + minimum_length]

        if np.all(np.isfinite(segment)):

            if np.nanstd(segment) == 0:

                flat_mask[i:i + minimum_length] = True

    return flat_mask


def detect_amplitude_artifacts(x):

    x = np.asarray(x, dtype=float)

    artifact_mask = np.zeros(len(x), dtype=bool)

    valid = np.isfinite(x)

    if valid.sum() < 10:
        return artifact_mask

    z = robust_z_score(x)

    artifact_mask |= np.abs(z) > MAD_THRESHOLD

    return artifact_mask


def detect_jump_artifacts(x):

    x = np.asarray(x, dtype=float)

    artifact_mask = np.zeros(len(x), dtype=bool)

    if len(x) < 10:
        return artifact_mask

    difference = np.diff(x)

    difference = np.insert(difference, 0, np.nan)

    z = robust_z_score(difference)

    artifact_mask |= np.abs(z) > DIFF_MAD_THRESHOLD

    return artifact_mask


def interpolate_series(series):

    result = pd.to_numeric(
        series,
        errors="coerce"
    ).astype(float)

    result = result.replace(
        [np.inf, -np.inf],
        np.nan
    )

    result = result.interpolate(
        method="linear",
        limit_direction="both"
    )

    if result.isna().any():

        median_value = result.median()

        if np.isfinite(median_value):

            result = result.fillna(median_value)

    return result


def bandpass_filter(signal, fs):

    signal = np.asarray(
        signal,
        dtype=float
    )

    nyquist = fs / 2.0

    high = min(
        HIGH_CUTOFF,
        nyquist * 0.90
    )

    low = LOW_CUTOFF

    if high <= low:
        return signal

    sos = butter(
        4,
        [low, high],
        btype="bandpass",
        fs=fs,
        output="sos"
    )

    return sosfiltfilt(
        sos,
        signal
    )


def notch_filter(signal, fs):

    signal = np.asarray(
        signal,
        dtype=float
    )

    if fs <= NOTCH_FREQUENCY * 2:

        return signal

    b, a = iirnotch(
        NOTCH_FREQUENCY,
        NOTCH_Q,
        fs
    )

    return filtfilt(
        b,
        a,
        signal
    )


def apply_eeg_filtering(signal, fs):

    signal = np.asarray(
        signal,
        dtype=float
    )

    if not np.isfinite(fs):

        return signal, "FILTER_SKIPPED_NO_VALID_FS"

    if fs < MIN_FILTER_FS:

        return signal, "FILTER_SKIPPED_LOW_FS"

    try:

        filtered = bandpass_filter(
            signal,
            fs
        )

        filtered = notch_filter(
            filtered,
            fs
        )

        return filtered, "BANDPASS_0.5_TO_45_HZ_PLUS_50_HZ_NOTCH"

    except Exception:

        return signal, "FILTER_FAILED"


def apply_ica_correction(data_matrix, fs):

    """

    ICA is applied only when the sampling rate is sufficiently
    high for EEG processing.

    Four RAW EEG channels are used as ICA observations.

    """

    if not np.isfinite(fs):

        return data_matrix, 0, "ICA_SKIPPED_NO_VALID_FS"

    if fs < MIN_ICA_FS:

        return data_matrix, 0, "ICA_SKIPPED_LOW_FS"

    if data_matrix.shape[0] < 100:

        return data_matrix, 0, "ICA_SKIPPED_TOO_FEW_SAMPLES"

    if data_matrix.shape[1] < 2:

        return data_matrix, 0, "ICA_SKIPPED_TOO_FEW_CHANNELS"

    try:

        X = np.asarray(
            data_matrix,
            dtype=float
        )

        X_mean = np.mean(
            X,
            axis=0,
            keepdims=True
        )

        X_std = np.std(
            X,
            axis=0,
            keepdims=True
        )

        X_std[X_std == 0] = 1.0

        X_scaled = (
            X - X_mean
        ) / X_std

        n_components = min(
            X.shape[1],
            4
        )

        ica = FastICA(
            n_components=n_components,
            random_state=42,
            whiten="unit-variance",
            max_iter=1000,
            tol=0.001
        )

        components = ica.fit_transform(
            X_scaled
        )

        component_kurtosis = np.array([
            kurtosis(
                components[:, i],
                fisher=False,
                nan_policy="omit"
            )
            for i in range(
                components.shape[1]
            )
        ])

        artifact_components = (
            component_kurtosis > 10
        )

        corrected_components = components.copy()

        corrected_components[
            :,
            artifact_components
        ] = 0.0

        reconstructed = ica.inverse_transform(
            corrected_components
        )

        reconstructed = (
            reconstructed * X_std
        ) + X_mean

        removed_components = int(
            artifact_components.sum()
        )

        return (
            reconstructed,
            removed_components,
            "ICA_APPLIED"
        )

    except Exception:

        return (
            data_matrix,
            0,
            "ICA_FAILED"
        )


# ============================================================
# BAND CALCULATION
# ============================================================

def calculate_band_power(signal, fs, low, high):

    if not np.isfinite(fs):
        return np.nan

    if fs < 2 * high:
        return np.nan

    signal = np.asarray(
        signal,
        dtype=float
    )

    if len(signal) < 20:
        return np.nan

    signal = signal - np.mean(signal)

    frequencies = np.fft.rfftfreq(
        len(signal),
        d=1.0 / fs
    )

    spectrum = np.abs(
        np.fft.rfft(signal)
    ) ** 2

    mask = (
        (frequencies >= low)
        &
        (frequencies <= high)
    )

    if mask.sum() == 0:
        return np.nan

    return float(
        np.mean(
            spectrum[mask]
        )
    )


def recalculate_band_columns(df, fs):

    band_ranges = {

        "Delta": (0.5, 4.0),

        "Theta": (4.0, 8.0),

        "Alpha": (8.0, 13.0),

        "Beta": (13.0, 30.0),

        "Gamma": (30.0, 45.0)
    }

    eeg_channels = [
        "TP9",
        "AF7",
        "AF8",
        "TP10"
    ]

    if not np.isfinite(fs):

        return df, False

    if fs < MIN_FILTER_FS:

        return df, False

    required_columns = RAW_EEG_CHANNELS

    if not all(
        c in df.columns
        for c in required_columns
    ):

        return df, False

    signal_matrix = df[
        required_columns
    ].to_numpy(
        dtype=float
    )

    for band_name, (
        low,
        high
    ) in band_ranges.items():

        for index, channel in enumerate(
            eeg_channels
        ):

            output_column = (
                f"{band_name}_{channel}"
            )

            signal = signal_matrix[:, index]

            window_length = int(
                round(fs * 2)
            )

            if window_length < 20:
                continue

            values = np.full(
                len(signal),
                np.nan
            )

            for start in range(
                0,
                len(signal),
                window_length
            ):

                end = min(
                    start + window_length,
                    len(signal)
                )

                segment = signal[
                    start:end
                ]

                if len(segment) < window_length:
                    continue

                values[
                    start:end
                ] = calculate_band_power(
                    segment,
                    fs,
                    low,
                    high
                )

            df[
                output_column
            ] = values

    return df, True


# ============================================================
# FILE DISCOVERY
# ============================================================

all_files = sorted(
    glob.glob(
        os.path.join(
            INPUT_ROOT,
            "**",
            "*.csv"
        ),
        recursive=True
    )
)


print("=" * 100)
print("CLEAN_DATA_V10 → FINAL EEG DATASET")
print("=" * 100)

print()
print("Input:", INPUT_ROOT)
print("Output:", OUTPUT_DATA)
print("CSV files:", len(all_files))


# ============================================================
# REPORT STORAGE
# ============================================================

audit_records = []

channel_records = []

processing_records = []

before_after_records = []


# ============================================================
# VISUALIZATION DATA
# ============================================================

global_before = {}
global_after = {}

global_artifacts = {}

global_missing = {}

global_sampling_rates = []


# ============================================================
# PROCESS FILES
# ============================================================

for file_index, input_file in enumerate(
    all_files,
    start=1
):

    filename = os.path.basename(
        input_file
    )

    try:

        df = pd.read_csv(
            input_file
        )

    except Exception as e:

        print(
            f"ERROR reading {filename}: {e}"
        )

        continue


    # ========================================================
    # IDENTIFY MUSE FILES
    # ========================================================

    raw_columns = [
        c for c in RAW_EEG_CHANNELS
        if c in df.columns
    ]

    is_muse = (
        len(raw_columns) > 0
    )


    # ========================================================
    # MOTION WATCH
    # ========================================================

    if not is_muse:

        relative_path = os.path.relpath(
            input_file,
            INPUT_ROOT
        )

        output_file = os.path.join(
            OUTPUT_DATA,
            relative_path
        )

        os.makedirs(
            os.path.dirname(
                output_file
            ),
            exist_ok=True
        )

        df.to_csv(
            output_file,
            index=False
        )

        processing_records.append({

            "File": filename,

            "File_Type": "Motion Watch",

            "Sampling_Rate_Hz": np.nan,

            "Filter_Status": "NOT_APPLICABLE",

            "ICA_Status": "NOT_APPLICABLE",

            "Artifact_Points": 0,

            "Missing_Values_After": int(
                df.isna().sum().sum()
            ),

            "Status": "COPIED_UNCHANGED"
        })

        continue


    # ========================================================
    # TIMESTAMP
    # ========================================================

    timestamp_column = None

    for candidate in [
        "TimeStamp",
        "Timestamp",
        "timestamp",
        "Time"
    ]:

        if candidate in df.columns:

            timestamp_column = candidate
            break


    if timestamp_column is not None:

        fs = calculate_sampling_rate(
            df[timestamp_column]
        )

    else:

        fs = np.nan


    if np.isfinite(fs):

        global_sampling_rates.append(
            fs
        )


    # ========================================================
    # INITIAL AUDIT
    # ========================================================

    missing_before = int(
        df[raw_columns]
        .isna()
        .sum()
        .sum()
    )

    inf_before = int(
        np.isinf(
            df[raw_columns]
            .apply(
                pd.to_numeric,
                errors="coerce"
            )
            .to_numpy(
                dtype=float
            )
        ).sum()
    )


    # ========================================================
    # CHANNEL PROCESSING
    # ========================================================

    cleaned_channels = {}

    file_artifact_total = 0

    file_flatline_total = 0

    file_extreme_total = 0

    file_jump_total = 0


    for channel in raw_columns:

        original = pd.to_numeric(
            df[channel],
            errors="coerce"
        ).astype(float)

        original = original.replace(
            [np.inf, -np.inf],
            np.nan
        )

        global_before.setdefault(
            channel,
            []
        )

        global_after.setdefault(
            channel,
            []

        )

        global_before[channel].extend(
            original.dropna().tolist()
        )


        # ====================================================
        # FLATLINE
        # ====================================================

        flat_mask = detect_flatline(
            original.to_numpy(),
            FLATLINE_LENGTH
        )

        file_flatline_total += int(
            flat_mask.sum()
        )


        # ====================================================
        # AMPLITUDE ARTIFACT
        # ====================================================

        amplitude_mask = detect_amplitude_artifacts(
            original.to_numpy()
        )

        file_extreme_total += int(
            amplitude_mask.sum()
        )


        # ====================================================
        # SUDDEN JUMP
        # ====================================================

        jump_mask = detect_jump_artifacts(
            original.to_numpy()
        )

        file_jump_total += int(
            jump_mask.sum()
        )


        # ====================================================
        # COMBINE ARTIFACT MASK
        # ====================================================

        artifact_mask = (
            flat_mask
            |
            amplitude_mask
            |
            jump_mask
        )

        file_artifact_total += int(
            artifact_mask.sum()
        )


        # ====================================================
        # REPLACE ARTIFACTS
        # ====================================================

        cleaned = original.copy()

        cleaned.loc[
            artifact_mask
        ] = np.nan


        # ====================================================
        # INTERPOLATION
        # ====================================================

        cleaned = interpolate_series(
            cleaned
        )


        # ====================================================
        # EEG FILTERING
        # ====================================================

        if np.isfinite(fs) and fs >= MIN_FILTER_FS:

            filtered, filter_status = (
                apply_eeg_filtering(
                    cleaned.to_numpy(),
                    fs
                )
            )

            cleaned = pd.Series(
                filtered,
                index=df.index
            )

        else:

            filter_status = (
                "FILTER_SKIPPED_LOW_FS"
            )


        cleaned_channels[
            channel
        ] = cleaned


        global_after[channel].extend(
            cleaned.dropna().tolist()
        )


        channel_records.append({

            "File": filename,

            "Channel": channel,

            "Sampling_Rate_Hz": fs,

            "Missing_Before": int(
                original.isna().sum()
            ),

            "Flatline_Points": int(
                flat_mask.sum()
            ),

            "Extreme_Amplitude_Points": int(
                amplitude_mask.sum()
            ),

            "Sudden_Jump_Points": int(
                jump_mask.sum()
            ),

            "Total_Artifact_Points": int(
                artifact_mask.sum()
            ),

            "Missing_After": int(
                cleaned.isna().sum()
            ),

            "Filter_Status": filter_status
        })


        global_artifacts.setdefault(
            channel,
            0
        )

        global_artifacts[channel] += int(
            artifact_mask.sum()
        )


    # ========================================================
    # WRITE CLEAN RAW EEG
    # ========================================================

    for channel in raw_columns:

        df[channel] = cleaned_channels[
            channel
        ]


    # ========================================================
    # ICA
    # ========================================================

    ica_removed = 0

    if len(raw_columns) == 4:

        matrix = df[
            RAW_EEG_CHANNELS
        ].to_numpy(
            dtype=float
        )

        if (
            np.isfinite(fs)
            and fs >= MIN_ICA_FS
        ):

            corrected, ica_removed, ica_status = (
                apply_ica_correction(
                    matrix,
                    fs
                )
            )

            if ica_status == "ICA_APPLIED":

                for i, channel in enumerate(
                    RAW_EEG_CHANNELS
                ):

                    df[channel] = corrected[
                        :,
                        i
                    ]

        else:

            ica_status = (
                "ICA_SKIPPED_LOW_FS"
            )

    else:

        ica_status = (
            "ICA_SKIPPED_INCOMPLETE_CHANNELS"
        )


    # ========================================================
    # RECALCULATE EEG BANDS
    # ========================================================

    df, bands_recalculated = (
        recalculate_band_columns(
            df,
            fs
        )
    )


    # ========================================================
    # FINAL MISSING VALUE VALIDATION
    # ========================================================

    missing_after = int(
        df[raw_columns]
        .isna()
        .sum()
        .sum()
    )


    # ========================================================
    # SAVE FILE
    # ========================================================

    relative_path = os.path.relpath(
        input_file,
        INPUT_ROOT
    )

    output_file = os.path.join(
        OUTPUT_DATA,
        relative_path
    )

    os.makedirs(
        os.path.dirname(
            output_file
        ),
        exist_ok=True
    )

    df.to_csv(
        output_file,
        index=False
    )


    # ========================================================
    # PROCESSING REPORT
    # ========================================================

    processing_records.append({

        "File": filename,

        "File_Type": "Muse",

        "Sampling_Rate_Hz": fs,

        "Raw_EEG_Channels": len(
            raw_columns
        ),

        "Artifact_Points": file_artifact_total,

        "Flatline_Points": file_flatline_total,

        "Extreme_Amplitude_Points": file_extreme_total,

        "Sudden_Jump_Points": file_jump_total,

        "ICA_Components_Removed": ica_removed,

        "ICA_Status": ica_status,

        "Filter_Status": (
            "APPLIED"
            if np.isfinite(fs)
            and fs >= MIN_FILTER_FS
            else "SKIPPED_LOW_FS"
        ),

        "Bands_Recalculated": (
            bands_recalculated
        ),

        "Missing_Before": missing_before,

        "Missing_After": missing_after,

        "Infinite_Values_Before": inf_before,

        "Status": "PROCESSED"
    })


    audit_records.append({

        "File": filename,

        "Sampling_Rate_Hz": fs,

        "RAW_Channel_Count": len(
            raw_columns
        ),

        "Missing_Before": missing_before,

        "Missing_After": missing_after,

        "Infinite_Before": inf_before,

        "Artifact_Points": file_artifact_total,

        "ICA_Removed": ica_removed,

        "Bands_Recalculated": bands_recalculated
    })


    before_after_records.append({

        "File": filename,

        "Sampling_Rate_Hz": fs,

        "Artifact_Points": file_artifact_total,

        "Missing_Before": missing_before,

        "Missing_After": missing_after
    })


    if file_index % 20 == 0:

        print(
            f"Processed {file_index} / {len(all_files)}"
        )


# ============================================================
# SAVE REPORTS
# ============================================================

audit_df = pd.DataFrame(
    audit_records
)

channel_df = pd.DataFrame(
    channel_records
)

processing_df = pd.DataFrame(
    processing_records
)

comparison_df = pd.DataFrame(
    before_after_records
)


audit_df.to_csv(
    os.path.join(
        REPORT_DIR,
        "FINAL_EEG_Audit_Report.csv"
    ),
    index=False
)


channel_df.to_csv(
    os.path.join(
        REPORT_DIR,
        "FINAL_EEG_Channel_Report.csv"
    ),
    index=False
)


processing_df.to_csv(
    os.path.join(
        REPORT_DIR,
        "FINAL_EEG_Processing_Report.csv"
    ),
    index=False
)


comparison_df.to_csv(
    os.path.join(
        REPORT_DIR,
        "FINAL_V10_to_FINAL_Comparison.csv"
    ),
    index=False
)


# ============================================================
# VISUALIZATION 1
# AMPLITUDE DISTRIBUTION
# ============================================================

channels_for_plot = [
    c for c in RAW_EEG_CHANNELS
    if c in global_before
]


plt.figure(
    figsize=(14, 8)
)

for channel in channels_for_plot:

    values = np.asarray(
        global_before[channel],
        dtype=float
    )

    values = values[
        np.isfinite(values)
    ]

    if len(values) > 100000:

        rng = np.random.default_rng(
            42
        )

        values = rng.choice(
            values,
            size=100000,
            replace=False
        )

    plt.hist(
        values,
        bins=100,
        alpha=0.45,
        label=channel
    )

plt.title(
    "V10 RAW EEG Amplitude Distribution"
)

plt.xlabel(
    "Amplitude"
)

plt.ylabel(
    "Frequency"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        GRAPH_DIR,
        "01_RAW_EEG_Amplitude_Distribution.png"
    ),
    dpi=300
)

plt.close()


# ============================================================
# VISUALIZATION 2
# ARTIFACTS BY CHANNEL
# ============================================================

artifact_plot = pd.Series(
    global_artifacts
).sort_values(
    ascending=False
)

plt.figure(
    figsize=(12, 7)
)

plt.bar(
    artifact_plot.index,
    artifact_plot.values
)

plt.title(
    "Detected EEG Artifacts by RAW EEG Channel"
)

plt.xlabel(
    "RAW EEG Channel"
)

plt.ylabel(
    "Detected Artifact Points"
)

plt.xticks(
    rotation=45
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        GRAPH_DIR,
        "02_EEG_Artifacts_By_Channel.png"
    ),
    dpi=300
)

plt.close()


# ============================================================
# VISUALIZATION 3
# SAMPLING RATE DISTRIBUTION
# ============================================================

if len(global_sampling_rates) > 0:

    plt.figure(
        figsize=(12, 7)
    )

    plt.hist(
        global_sampling_rates,
        bins=30
    )

    plt.title(
        "Muse Sampling Rate Distribution"
    )

    plt.xlabel(
        "Sampling Rate, Hz"
    )

    plt.ylabel(
        "Number of Files"
    )

    plt.grid(
        alpha=0.25
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            GRAPH_DIR,
            "03_Muse_Sampling_Rate_Distribution.png"
        ),
        dpi=300
    )

    plt.close()


# ============================================================
# VISUALIZATION 4
# V10 VERSUS FINAL EXAMPLE
# ============================================================

example_file = None

for f in all_files:

    try:

        temp = pd.read_csv(
            f,
            nrows=3000
        )

        if all(
            c in temp.columns
            for c in RAW_EEG_CHANNELS
        ):

            example_file = f
            break

    except Exception:

        pass


if example_file is not None:

    original_example = pd.read_csv(
        example_file,
        nrows=3000
    )

    relative_example = os.path.relpath(
        example_file,
        INPUT_ROOT
    )

    final_example_file = os.path.join(
        OUTPUT_DATA,
        relative_example
    )

    if os.path.exists(
        final_example_file
    ):

        final_example = pd.read_csv(
            final_example_file,
            nrows=3000
        )

        plt.figure(
            figsize=(16, 9)
        )

        for channel in RAW_EEG_CHANNELS:

            if (
                channel in original_example.columns
                and channel in final_example.columns
            ):

                plt.plot(
                    original_example[
                        channel
                    ].to_numpy(),
                    alpha=0.30,
                    label=f"V10 {channel}"
                )

                plt.plot(
                    final_example[
                        channel
                    ].to_numpy(),
                    alpha=0.85,
                    label=f"FINAL {channel}"
                )

        plt.title(
            "V10 versus Final EEG Signal"
        )

        plt.xlabel(
            "Data Point"
        )

        plt.ylabel(
            "EEG Amplitude"
        )

        plt.legend(
            ncol=2
        )

        plt.grid(
            alpha=0.25
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                GRAPH_DIR,
                "04_V10_vs_FINAL_EEG.png"
            ),
            dpi=300
        )

        plt.close()


# ============================================================
# VISUALIZATION 5
# MISSING VALUES
# ============================================================

if not channel_df.empty:

    missing_plot = (
        channel_df
        .groupby("Channel")[
            [
                "Missing_Before",
                "Missing_After"
            ]
        ]
        .sum()
    )

    plt.figure(
        figsize=(14, 7)
    )

    x = np.arange(
        len(missing_plot.index)
    )

    width = 0.35

    plt.bar(
        x - width / 2,
        missing_plot[
            "Missing_Before"
        ],
        width,
        label="V10"
    )

    plt.bar(
        x + width / 2,
        missing_plot[
            "Missing_After"
        ],
        width,
        label="Final"
    )

    plt.xticks(
        x,
        missing_plot.index,
        rotation=45
    )

    plt.title(
        "Missing RAW EEG Values: V10 versus Final"
    )

    plt.xlabel(
        "RAW EEG Channel"
    )

    plt.ylabel(
        "Missing Values"
    )

    plt.legend()

    plt.grid(
        axis="y",
        alpha=0.25
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            GRAPH_DIR,
            "05_V10_vs_FINAL_Missing_Values.png"
        ),
        dpi=300
    )

    plt.close()


# ============================================================
# VISUALIZATION 6
# FLATLINE, EXTREME AND JUMP ARTIFACTS
# ============================================================

if not channel_df.empty:

    artifact_summary = (
        channel_df
        .groupby("Channel")[
            [
                "Flatline_Points",
                "Extreme_Amplitude_Points",
                "Sudden_Jump_Points"
            ]
        ]
        .sum()
    )

    plt.figure(
        figsize=(16, 8)
    )

    x = np.arange(
        len(artifact_summary.index)
    )

    width = 0.25

    plt.bar(
        x - width,
        artifact_summary[
            "Flatline_Points"
        ],
        width,
        label="Flatline"
    )

    plt.bar(
        x,
        artifact_summary[
            "Extreme_Amplitude_Points"
        ],
        width,
        label="Extreme amplitude"
    )

    plt.bar(
        x + width,
        artifact_summary[
            "Sudden_Jump_Points"
        ],
        width,
        label="Sudden jump"
    )

    plt.xticks(
        x,
        artifact_summary.index,
        rotation=45
    )

    plt.title(
        "EEG Artifact Types by RAW EEG Channel"
    )

    plt.xlabel(
        "RAW EEG Channel"
    )

    plt.ylabel(
        "Detected Points"
    )

    plt.legend()

    plt.grid(
        axis="y",
        alpha=0.25
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            GRAPH_DIR,
            "06_EEG_Artifact_Types.png"
        ),
        dpi=300
    )

    plt.close()


# ============================================================
# FINAL SUMMARY
# ============================================================

muse_count = sum(
    1 for r in processing_records
    if r.get("File_Type") == "Muse"
)

watch_count = sum(
    1 for r in processing_records
    if r.get("File_Type") == "Motion Watch"
)

total_artifacts = int(
    channel_df[
        "Total_Artifact_Points"
    ].sum()
) if not channel_df.empty else 0


total_missing_after = int(
    processing_df[
        "Missing_After"
    ].sum()
) if not processing_df.empty else 0


ica_applied_count = int(
    (
        processing_df[
            "ICA_Status"
        ] == "ICA_APPLIED"
    ).sum()
) if "ICA_Status" in processing_df else 0


filter_applied_count = int(
    (
        processing_df[
            "Filter_Status"
        ] == "APPLIED"
    ).sum()
) if "Filter_Status" in processing_df else 0


bands_recalculated_count = int(
    processing_df[
        "Bands_Recalculated"
    ].sum()
) if "Bands_Recalculated" in processing_df else 0


print()
print("=" * 100)
print("FINAL EEG PROCESSING COMPLETED")
print("=" * 100)

print()
print("Total CSV files:", len(all_files))
print("Muse files:", muse_count)
print("Motion Watch files:", watch_count)

print()
print("Total detected artifact points:", total_artifacts)
print("Remaining RAW EEG missing values:", total_missing_after)

print()
print("Files with EEG filtering:", filter_applied_count)
print("Files with ICA:", ica_applied_count)
print("Files with recalculated bands:", bands_recalculated_count)

if len(global_sampling_rates) > 0:

    print()
    print(
        "Median Muse sampling rate:",
        round(
            float(
                np.median(
                    global_sampling_rates
                )
            ),
            4
        ),
        "Hz"
    )

    print(
        "Minimum Muse sampling rate:",
        round(
            float(
                np.min(
                    global_sampling_rates
                )
            ),
            4
        ),
        "Hz"
    )

    print(
        "Maximum Muse sampling rate:",
        round(
            float(
                np.max(
                    global_sampling_rates
                )
            ),
            4
        ),
        "Hz"
    )

print()
print("Final dataset:")
print(OUTPUT_DATA)

print()
print("Reports:")
print(REPORT_DIR)

print()
print("Visualizations:")
print(GRAPH_DIR)

print()
print("=" * 100)
print("IMPORTANT PROCESSING RULE")
print("=" * 100)

print(
    "EEG filtering and ICA are applied only when the detected "
    "sampling rate is technically suitable."
)

print(
    "If the V10 RAW EEG data are approximately 1 Hz, "
    "the code skips conventional EEG filtering and ICA "
    "rather than applying an invalid EEG filter."
)

print(
    "The processing report records this decision for every file."
)

print("=" * 100)
```

```python
# ============================================================
# V10 RAW EEG VERIFICATION AND QUALITY AUDIT
# ============================================================
#
# IMPORTANT:
# This code does NOT modify the V10 dataset.
# It only analyzes the RAW EEG channels.
#
# RAW EEG:
#     RAW_TP9
#     RAW_AF7
#     RAW_AF8
#     RAW_TP10
#
# Purpose:
#     1. Verify timestamps
#     2. Calculate actual sampling rate
#     3. Analyze amplitude distribution
#     4. Detect flatline segments
#     5. Detect extreme amplitudes
#     6. Detect clipping
#     7. Detect excessive repeated values
#     8. Check channel consistency
#     9. Generate visualizations
#    10. Create detailed CSV reports
#
# ============================================================


import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


warnings.filterwarnings("ignore")


# ============================================================
# PATH CONFIGURATION
# ============================================================

INPUT_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/clean_data"
)

OUTPUT_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10_RAW_EEG_AUDIT"
)

REPORT_DIR = OUTPUT_DIR / "reports"
VIS_DIR = OUTPUT_DIR / "visualizations"

REPORT_DIR.mkdir(parents=True, exist_ok=True)
VIS_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# RAW EEG CHANNELS
# ============================================================

RAW_CHANNELS = [
    "RAW_TP9",
    "RAW_AF7",
    "RAW_AF8",
    "RAW_TP10"
]


# ============================================================
# PARAMETERS
# ============================================================

# A flatline is defined using consecutive identical values.
FLATLINE_MIN_LENGTH = 5

# Robust extreme amplitude threshold.
# This is NOT an EEG physiological rejection threshold.
# It is only used for quality auditing.
EXTREME_Z_THRESHOLD = 8.0

# Maximum number of points displayed in long signal plots.
MAX_PLOT_POINTS = 5000


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def find_timestamp_column(df):
    """
    Find the timestamp column.
    """

    candidates = [
        "TimeStamp",
        "Timestamp",
        "timestamp",
        "TIMESTAMP",
        "Time",
        "time"
    ]

    for col in candidates:
        if col in df.columns:
            return col

    return None


def find_raw_channels(df):
    """
    Identify available RAW EEG channels.
    """

    return [
        ch for ch in RAW_CHANNELS
        if ch in df.columns
    ]


def calculate_sampling_rate(df, timestamp_column):
    """
    Calculate sampling rate from timestamp differences.

    The timestamp is interpreted as a time of day.
    """

    if timestamp_column is None:
        return np.nan, np.nan, 0

    ts = pd.to_datetime(
        df[timestamp_column].astype(str),
        errors="coerce"
    )

    valid_ts = ts.dropna()

    if len(valid_ts) < 3:
        return np.nan, np.nan, 0

    seconds = (
        valid_ts.astype("int64") / 1_000_000_000
    ).to_numpy()

    differences = np.diff(seconds)

    # Keep only positive intervals.
    differences = differences[differences > 0]

    if len(differences) == 0:
        return np.nan, np.nan, 0

    median_interval = float(
        np.median(differences)
    )

    sampling_rate = (
        1.0 / median_interval
        if median_interval > 0
        else np.nan
    )

    return sampling_rate, median_interval, len(differences)


def repeated_value_statistics(series):
    """
    Detect consecutive repeated values.
    """

    x = pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()

    if len(x) == 0:
        return 0, 0, 0

    values = x.to_numpy()

    changes = np.r_[
        True,
        values[1:] != values[:-1]
    ]

    group_ids = np.cumsum(changes)

    run_lengths = (
        pd.Series(values)
        .groupby(group_ids)
        .size()
        .to_numpy()
    )

    long_runs = run_lengths[
        run_lengths >= FLATLINE_MIN_LENGTH
    ]

    if len(long_runs) == 0:
        return 0, 0, 0

    return (
        int(len(long_runs)),
        int(np.max(long_runs)),
        int(np.sum(long_runs))
    )


def robust_statistics(series):
    """
    Calculate robust amplitude statistics.
    """

    x = pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()

    if len(x) == 0:
        return {
            "count": 0,
            "missing": int(len(series)),
            "mean": np.nan,
            "std": np.nan,
            "median": np.nan,
            "minimum": np.nan,
            "maximum": np.nan,
            "q01": np.nan,
            "q99": np.nan,
            "mad": np.nan
        }

    values = x.to_numpy(dtype=float)

    median = np.median(values)

    mad = np.median(
        np.abs(values - median)
    )

    return {
        "count": int(len(values)),
        "missing": int(series.isna().sum()),
        "mean": float(np.mean(values)),
        "std": float(np.std(values)),
        "median": float(median),
        "minimum": float(np.min(values)),
        "maximum": float(np.max(values)),
        "q01": float(np.percentile(values, 1)),
        "q99": float(np.percentile(values, 99)),
        "mad": float(mad)
    }


def detect_extreme_values(series):
    """
    Robust extreme value detection using MAD.

    This is an audit indicator only.
    """

    x = pd.to_numeric(
        series,
        errors="coerce"
    )

    valid = x.dropna()

    if len(valid) < 10:
        return 0, np.nan

    median = valid.median()

    mad = np.median(
        np.abs(
            valid.to_numpy() - median
        )
    )

    if mad == 0:
        return 0, 0.0

    robust_z = (
        0.6745 *
        (valid - median) /
        mad
    )

    extreme_count = int(
        np.sum(
            np.abs(robust_z) >
            EXTREME_Z_THRESHOLD
        )
    )

    percentage = (
        extreme_count /
        len(valid) *
        100
    )

    return extreme_count, percentage


def detect_clipping(series):
    """
    Detect values appearing repeatedly at the minimum
    or maximum observed value.
    """

    x = pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()

    if len(x) == 0:
        return 0, 0, np.nan, np.nan

    values = x.to_numpy()

    minimum = np.min(values)
    maximum = np.max(values)

    min_count = int(
        np.sum(values == minimum)
    )

    max_count = int(
        np.sum(values == maximum)
    )

    min_percentage = (
        min_count / len(values) * 100
    )

    max_percentage = (
        max_count / len(values) * 100
    )

    return (
        min_count,
        max_count,
        min_percentage,
        max_percentage
    )


# ============================================================
# DISCOVER CSV FILES RECURSIVELY
# ============================================================

csv_files = sorted(
    INPUT_DIR.rglob("*.csv")
)

print("=" * 100)
print("V10 RAW EEG VERIFICATION AND QUALITY AUDIT")
print("=" * 100)

print()
print("Input directory:")
print(INPUT_DIR)

print()
print("CSV files found:", len(csv_files))

if len(csv_files) == 0:
    raise FileNotFoundError(
        f"No CSV files were found inside:\n{INPUT_DIR}\n"
        "Please verify the V10 folder path."
    )

print()
print("First 10 CSV files found:")

for f in csv_files[:10]:
    print("   ", f)

# ============================================================
# STORAGE FOR REPORTS
# ============================================================

file_report = []
channel_report = []

all_sampling_rates = []

example_file = None
example_dataframe = None


# ============================================================
# PROCESS ALL FILES
# ============================================================

for file_index, file_path in enumerate(csv_files, start=1):

    try:

        df = pd.read_csv(file_path)

    except Exception as e:

        file_report.append({
            "File": file_path.name,
            "Status": "READ_ERROR",
            "Error": str(e)
        })

        continue


    if example_dataframe is None:
        example_file = file_path
        example_dataframe = df.copy()


    timestamp_column = find_timestamp_column(df)

    raw_channels = find_raw_channels(df)

    sampling_rate, median_interval, interval_count = (
        calculate_sampling_rate(
            df,
            timestamp_column
        )
    )


    if np.isfinite(sampling_rate):

        all_sampling_rates.append(
            sampling_rate
        )


    # --------------------------------------------------------
    # FILE LEVEL INFORMATION
    # --------------------------------------------------------

    row_count = len(df)

    timestamp_missing = (
        int(df[timestamp_column].isna().sum())
        if timestamp_column is not None
        else np.nan
    )


    duplicate_timestamps = 0
    non_increasing_timestamps = 0

    if timestamp_column is not None:

        ts = pd.to_datetime(
            df[timestamp_column].astype(str),
            errors="coerce"
        )

        duplicate_timestamps = int(
            ts.duplicated().sum()
        )

        valid_ts = ts.dropna()

        if len(valid_ts) > 1:

            time_values = (
                valid_ts.astype("int64")
                .to_numpy()
            )

            time_diffs = np.diff(
                time_values
            )

            non_increasing_timestamps = int(
                np.sum(time_diffs <= 0)
            )


    file_report.append({

        "File": file_path.name,

        "Rows": row_count,

        "Timestamp_Column": timestamp_column,

        "Timestamp_Missing": timestamp_missing,

        "Duplicate_Timestamps":
            duplicate_timestamps,

        "Non_Increasing_Timestamps":
            non_increasing_timestamps,

        "RAW_Channel_Count":
            len(raw_channels),

        "RAW_TP9_Present":
            "RAW_TP9" in raw_channels,

        "RAW_AF7_Present":
            "RAW_AF7" in raw_channels,

        "RAW_AF8_Present":
            "RAW_AF8" in raw_channels,

        "RAW_TP10_Present":
            "RAW_TP10" in raw_channels,

        "Sampling_Rate_Hz":
            sampling_rate,

        "Median_Interval_Seconds":
            median_interval,

        "Status":
            "OK"

    })


    # --------------------------------------------------------
    # CHANNEL LEVEL ANALYSIS
    # --------------------------------------------------------

    for channel in RAW_CHANNELS:

        if channel not in df.columns:

            channel_report.append({

                "File": file_path.name,
                "Channel": channel,
                "Status": "MISSING"

            })

            continue


        numeric = pd.to_numeric(
            df[channel],
            errors="coerce"
        )


        stats = robust_statistics(
            numeric
        )


        flatline_count, longest_flatline, flatline_points = (
            repeated_value_statistics(
                numeric
            )
        )


        extreme_count, extreme_percentage = (
            detect_extreme_values(
                numeric
            )
        )


        (
            min_count,
            max_count,
            min_percentage,
            max_percentage
        ) = detect_clipping(
            numeric
        )


        unique_values = int(
            numeric.nunique(
                dropna=True
            )
        )


        zero_count = int(
            np.sum(
                numeric.to_numpy(
                    dtype=float
                ) == 0
            )
        )


        zero_percentage = (
            zero_count /
            stats["count"] *
            100
            if stats["count"] > 0
            else np.nan
        )


        channel_report.append({

            "File": file_path.name,

            "Channel": channel,

            "Status": "OK",

            "Count": stats["count"],

            "Missing": stats["missing"],

            "Missing_Percentage":
                (
                    stats["missing"] /
                    len(df) *
                    100
                    if len(df) > 0
                    else np.nan
                ),

            "Mean": stats["mean"],

            "Std": stats["std"],

            "Median": stats["median"],

            "Minimum": stats["minimum"],

            "Maximum": stats["maximum"],

            "Q01": stats["q01"],

            "Q99": stats["q99"],

            "MAD": stats["mad"],

            "Unique_Values":
                unique_values,

            "Zero_Count":
                zero_count,

            "Zero_Percentage":
                zero_percentage,

            "Flatline_Run_Count":
                flatline_count,

            "Longest_Flatline":
                longest_flatline,

            "Flatline_Points":
                flatline_points,

            "Extreme_Count":
                extreme_count,

            "Extreme_Percentage":
                extreme_percentage,

            "Minimum_Value_Count":
                min_count,

            "Maximum_Value_Count":
                max_count,

            "Minimum_Value_Percentage":
                min_percentage,

            "Maximum_Value_Percentage":
                max_percentage,

            "Sampling_Rate_Hz":
                sampling_rate

        })


    if file_index % 20 == 0:

        print(
            f"Processed {file_index} / "
            f"{len(csv_files)}"
        )


# ============================================================
# CREATE REPORT DATAFRAMES
# ============================================================

file_report_df = pd.DataFrame(
    file_report
)

channel_report_df = pd.DataFrame(
    channel_report
)


# ============================================================
# SAVE REPORTS
# ============================================================

file_report_path = (
    REPORT_DIR /
    "V10_File_Level_Audit.csv"
)

channel_report_path = (
    REPORT_DIR /
    "V10_RAW_EEG_Channel_Audit.csv"
)


file_report_df.to_csv(
    file_report_path,
    index=False
)

channel_report_df.to_csv(
    channel_report_path,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 100)
print("SAMPLING RATE SUMMARY")
print("=" * 100)

if len(all_sampling_rates) > 0:

    sampling_rates = np.array(
        all_sampling_rates,
        dtype=float
    )

    print(
        "Median sampling rate:",
        round(
            np.median(sampling_rates),
            6
        ),
        "Hz"
    )

    print(
        "Mean sampling rate:",
        round(
            np.mean(sampling_rates),
            6
        ),
        "Hz"
    )

    print(
        "Minimum sampling rate:",
        round(
            np.min(sampling_rates),
            6
        ),
        "Hz"
    )

    print(
        "Maximum sampling rate:",
        round(
            np.max(sampling_rates),
            6
        ),
        "Hz"
    )

else:

    print(
        "Sampling rate could not be calculated."
    )


# ============================================================
# CHANNEL SUMMARY
# ============================================================

print()
print("=" * 100)
print("RAW EEG CHANNEL SUMMARY")
print("=" * 100)

if len(channel_report_df) > 0:

    summary = (
        channel_report_df
        .groupby("Channel")
        .agg({

            "Missing":
                "sum",

            "Flatline_Points":
                "sum",

            "Extreme_Count":
                "sum",

            "Zero_Count":
                "sum",

            "Count":
                "sum"

        })
    )

    summary["Missing_Percentage"] = (
        summary["Missing"] /
        summary["Count"] *
        100
    )

    summary["Flatline_Percentage"] = (
        summary["Flatline_Points"] /
        summary["Count"] *
        100
    )

    print(
        summary.round(4)
    )


# ============================================================
# VISUALIZATION 1
# SAMPLING RATE DISTRIBUTION
# ============================================================

if len(all_sampling_rates) > 0:

    plt.figure(
        figsize=(12, 7)
    )

    plt.hist(
        sampling_rates,
        bins=30
    )

    plt.xlabel(
        "Sampling Rate, Hz"
    )

    plt.ylabel(
        "Number of Files"
    )

    plt.title(
        "V10 RAW EEG Sampling Rate Distribution"
    )

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "01_Sampling_Rate_Distribution.png",
        dpi=200
    )

    plt.show()

    plt.close()


# ============================================================
# VISUALIZATION 2
# AMPLITUDE DISTRIBUTION
# ============================================================

if example_dataframe is not None:

    available_channels = [
        ch for ch in RAW_CHANNELS
        if ch in example_dataframe.columns
    ]

    if len(available_channels) > 0:

        plt.figure(
            figsize=(14, 8)
        )

        for channel in available_channels:

            values = pd.to_numeric(
                example_dataframe[channel],
                errors="coerce"
            ).dropna()

            if len(values) > 0:

                plt.hist(
                    values,
                    bins=100,
                    alpha=0.45,
                    label=channel
                )

        plt.xlabel(
            "RAW EEG Amplitude"
        )

        plt.ylabel(
            "Frequency"
        )

        plt.title(
            "V10 RAW EEG Amplitude Distribution"
        )

        plt.legend()

        plt.grid(
            alpha=0.3
        )

        plt.tight_layout()

        plt.savefig(
            VIS_DIR /
            "02_RAW_EEG_Amplitude_Distribution.png",
            dpi=200
        )

        plt.show()

        plt.close()


# ============================================================
# VISUALIZATION 3
# RAW EEG TIME SERIES
# ============================================================

if example_dataframe is not None:

    available_channels = [
        ch for ch in RAW_CHANNELS
        if ch in example_dataframe.columns
    ]

    if len(available_channels) > 0:

        plot_df = example_dataframe[
            available_channels
        ].apply(
            pd.to_numeric,
            errors="coerce"
        )

        plot_df = plot_df.iloc[
            :MAX_PLOT_POINTS
        ]

        plt.figure(
            figsize=(16, 9)
        )

        for channel in available_channels:

            plt.plot(
                plot_df[channel].to_numpy(),
                label=channel,
                linewidth=0.8
            )

        plt.xlabel(
            "Data Point"
        )

        plt.ylabel(
            "RAW EEG Amplitude"
        )

        plt.title(
            "V10 RAW EEG Signal Example"
        )

        plt.legend()

        plt.grid(
            alpha=0.3
        )

        plt.tight_layout()

        plt.savefig(
            VIS_DIR /
            "03_RAW_EEG_Time_Series.png",
            dpi=200
        )

        plt.show()

        plt.close()


# ============================================================
# VISUALIZATION 4
# CHANNEL AMPLITUDE BOX PLOT
# ============================================================

if example_dataframe is not None:

    available_channels = [
        ch for ch in RAW_CHANNELS
        if ch in example_dataframe.columns
    ]

    if len(available_channels) > 0:

        box_data = []

        for channel in available_channels:

            values = pd.to_numeric(
                example_dataframe[channel],
                errors="coerce"
            ).dropna()

            if len(values) > 0:

                box_data.append(
                    values.to_numpy()
                )


        plt.figure(
            figsize=(12, 7)
        )

        plt.boxplot(
            box_data,
            tick_labels=available_channels,
            showfliers=True
        )

        plt.xlabel(
            "RAW EEG Channel"
        )

        plt.ylabel(
            "Amplitude"
        )

        plt.title(
            "V10 RAW EEG Channel Amplitude Comparison"
        )

        plt.grid(
            alpha=0.3
        )

        plt.tight_layout()

        plt.savefig(
            VIS_DIR /
            "04_RAW_EEG_Channel_Boxplot.png",
            dpi=200
        )

        plt.show()

        plt.close()


# ============================================================
# VISUALIZATION 5
# FLATLINE POINTS BY CHANNEL
# ============================================================

if len(channel_report_df) > 0:

    flatline_summary = (
        channel_report_df
        .groupby("Channel")[
            "Flatline_Points"
        ]
        .sum()
        .sort_values(
            ascending=False
        )
    )

    plt.figure(
        figsize=(12, 7)
    )

    flatline_summary.plot(
        kind="bar"
    )

    plt.xlabel(
        "RAW EEG Channel"
    )

    plt.ylabel(
        "Flatline Points"
    )

    plt.title(
        "V10 RAW EEG Flatline Points by Channel"
    )

    plt.xticks(
        rotation=30
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "05_Flatline_Points_By_Channel.png",
        dpi=200
    )

    plt.show()

    plt.close()


# ============================================================
# VISUALIZATION 6
# EXTREME VALUES BY CHANNEL
# ============================================================

if len(channel_report_df) > 0:

    extreme_summary = (
        channel_report_df
        .groupby("Channel")[
            "Extreme_Count"
        ]
        .sum()
        .sort_values(
            ascending=False
        )
    )

    plt.figure(
        figsize=(12, 7)
    )

    extreme_summary.plot(
        kind="bar"
    )

    plt.xlabel(
        "RAW EEG Channel"
    )

    plt.ylabel(
        "Detected Extreme Points"
    )

    plt.title(
        "V10 RAW EEG Extreme Values by Channel"
    )

    plt.xticks(
        rotation=30
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "06_Extreme_Values_By_Channel.png",
        dpi=200
    )

    plt.show()

    plt.close()


# ============================================================
# VISUALIZATION 7
# MISSING VALUES BY CHANNEL
# ============================================================

if len(channel_report_df) > 0:

    missing_summary = (
        channel_report_df
        .groupby("Channel")[
            "Missing"
        ]
        .sum()
        .sort_values(
            ascending=False
        )
    )

    plt.figure(
        figsize=(12, 7)
    )

    missing_summary.plot(
        kind="bar"
    )

    plt.xlabel(
        "RAW EEG Channel"
    )

    plt.ylabel(
        "Missing Values"
    )

    plt.title(
        "V10 RAW EEG Missing Values by Channel"
    )

    plt.xticks(
        rotation=30
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "07_Missing_Values_By_Channel.png",
        dpi=200
    )

    plt.show()

    plt.close()


# ============================================================
# VISUALIZATION 8
# ZERO VALUES BY CHANNEL
# ============================================================

if len(channel_report_df) > 0:

    zero_summary = (
        channel_report_df
        .groupby("Channel")[
            "Zero_Count"
        ]
        .sum()
        .sort_values(
            ascending=False
        )
    )

    plt.figure(
        figsize=(12, 7)
    )

    zero_summary.plot(
        kind="bar"
    )

    plt.xlabel(
        "RAW EEG Channel"
    )

    plt.ylabel(
        "Zero Values"
    )

    plt.title(
        "V10 RAW EEG Zero Values by Channel"
    )

    plt.xticks(
        rotation=30
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "08_Zero_Values_By_Channel.png",
        dpi=200
    )

    plt.show()

    plt.close()


# ============================================================
# VISUALIZATION 9
# CHANNEL CORRELATION
# ============================================================

if example_dataframe is not None:

    available_channels = [
        ch for ch in RAW_CHANNELS
        if ch in example_dataframe.columns
    ]

    if len(available_channels) >= 2:

        corr_df = (
            example_dataframe[
                available_channels
            ]
            .apply(
                pd.to_numeric,
                errors="coerce"
            )
            .corr()
        )

        plt.figure(
            figsize=(8, 7)
        )

        plt.imshow(
            corr_df.to_numpy(),
            aspect="auto"
        )

        plt.xticks(
            range(len(corr_df.columns)),
            corr_df.columns,
            rotation=45
        )

        plt.yticks(
            range(len(corr_df.index)),
            corr_df.index
        )

        plt.colorbar(
            label="Correlation"
        )

        plt.title(
            "V10 RAW EEG Channel Correlation"
        )

        plt.tight_layout()

        plt.savefig(
            VIS_DIR /
            "09_RAW_EEG_Channel_Correlation.png",
            dpi=200
        )

        plt.show()

        plt.close()


# ============================================================
# FINAL INTERPRETATION REPORT
# ============================================================

summary_rows = []


if len(all_sampling_rates) > 0:

    median_sr = float(
        np.median(
            all_sampling_rates
        )
    )

    min_sr = float(
        np.min(
            all_sampling_rates
        )
    )

    max_sr = float(
        np.max(
            all_sampling_rates
        )
    )

else:

    median_sr = np.nan
    min_sr = np.nan
    max_sr = np.nan


summary_rows.append({
    "Metric":
        "Median sampling rate, Hz",
    "Value":
        median_sr
})

summary_rows.append({
    "Metric":
        "Minimum sampling rate, Hz",
    "Value":
        min_sr
})

summary_rows.append({
    "Metric":
        "Maximum sampling rate, Hz",
    "Value":
        max_sr
})


if len(channel_report_df) > 0:

    total_missing = int(
        channel_report_df[
            "Missing"
        ].sum()
    )

    total_flatline = int(
        channel_report_df[
            "Flatline_Points"
        ].sum()
    )

    total_extreme = int(
        channel_report_df[
            "Extreme_Count"
        ].sum()
    )

    total_zero = int(
        channel_report_df[
            "Zero_Count"
        ].sum()
    )

    total_points = int(
        channel_report_df[
            "Count"
        ].sum()
    )

else:

    total_missing = 0
    total_flatline = 0
    total_extreme = 0
    total_zero = 0
    total_points = 0


summary_rows.append({
    "Metric":
        "Total RAW EEG missing values",
    "Value":
        total_missing
})

summary_rows.append({
    "Metric":
        "Total flatline points",
    "Value":
        total_flatline
})

summary_rows.append({
    "Metric":
        "Total extreme points",
    "Value":
        total_extreme
})

summary_rows.append({
    "Metric":
        "Total zero values",
    "Value":
        total_zero
})


if total_points > 0:

    missing_percentage = (
        total_missing /
        total_points *
        100
    )

    flatline_percentage = (
        total_flatline /
        total_points *
        100
    )

    zero_percentage = (
        total_zero /
        total_points *
        100
    )

else:

    missing_percentage = np.nan
    flatline_percentage = np.nan
    zero_percentage = np.nan


summary_rows.append({
    "Metric":
        "RAW EEG missing percentage",
    "Value":
        missing_percentage
})

summary_rows.append({
    "Metric":
        "RAW EEG flatline percentage",
    "Value":
        flatline_percentage
})

summary_rows.append({
    "Metric":
        "RAW EEG zero value percentage",
    "Value":
        zero_percentage
})


summary_df = pd.DataFrame(
    summary_rows
)


summary_path = (
    REPORT_DIR /
    "V10_RAW_EEG_Audit_Summary.csv"
)

summary_df.to_csv(
    summary_path,
    index=False
)


# ============================================================
# PRINT FINAL STATUS
# ============================================================

print()
print("=" * 100)
print("RAW EEG AUDIT COMPLETED")
print("=" * 100)

print()
print("Files analyzed:", len(csv_files))

print()
print("Median sampling rate:",
      median_sr,
      "Hz")

print()
print("Total RAW EEG missing values:",
      total_missing)

print()
print("Total flatline points:",
      total_flatline)

print()
print("Total extreme points:",
      total_extreme)

print()
print("Total zero values:",
      total_zero)

print()
print("Reports saved to:")
print(REPORT_DIR)

print()
print("Visualizations saved to:")
print(VIS_DIR)

print()
print("=" * 100)
print("IMPORTANT")
print("=" * 100)

print(
    """
This cell only audits the V10 RAW EEG data.

It does NOT:
    apply a bandpass filter
    apply a notch filter
    apply ICA
    remove EEG artifacts
    interpolate RAW EEG
    recalculate EEG bands
    modify V10

The purpose is to establish whether the RAW EEG data
contain sufficient temporal information for conventional
EEG signal processing.
"""
)

print("=" * 100)
```

```python
import os
import re
import json
import shutil
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import signal
from scipy.stats import kurtosis
from sklearn.decomposition import FastICA

warnings.filterwarnings("ignore")


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/clean_data"
)

OUTPUT_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_FINAL"
)

PROCESSED_DIR = OUTPUT_DIR / "processed"
REJECTED_DIR = OUTPUT_DIR / "rejected"
BAND_POWER_DIR = OUTPUT_DIR / "band_power"
REPORT_DIR = OUTPUT_DIR / "reports"
VIS_DIR = OUTPUT_DIR / "visualizations"


# RAW EEG channels expected in the dataset

RAW_CHANNELS = [
    "RAW_TP9",
    "RAW_AF7",
    "RAW_AF8",
    "RAW_TP10"
]


# Possible timestamp column names

TIME_CANDIDATES = [
    "Time",
    "Timestamp",
    "timestamp",
    "time",
    "TIME",
    "DateTime",
    "Datetime",
    "datetime"
]


# Sampling rate requirements

MIN_USABLE_FS = 128.0
MAX_USABLE_FS = 2000.0

MAX_JITTER_CV = 0.05

MAX_DUPLICATE_RATIO = 0.01

MIN_VALID_RATIO = 0.70


# Invalid signal settings

FLATLINE_DURATION_SECONDS = 0.50

SHORT_GAP_SECONDS = 0.25

ARTIFACT_Z_THRESHOLD = 8.0


# EEG filtering

LOW_CUTOFF = 0.5
HIGH_CUTOFF = 45.0

NOTCH_FREQUENCY = 50.0

NOTCH_Q = 30.0


# ICA settings

USE_ICA = True

ICA_MAX_ITER = 500

ICA_TOLERANCE = 0.001

ICA_KURTOSIS_THRESHOLD = 8.0


# Band definitions

EEG_BANDS = {
    "delta": (0.5, 4.0),
    "theta": (4.0, 8.0),
    "alpha": (8.0, 13.0),
    "beta": (13.0, 30.0),
    "gamma": (30.0, 45.0)
}


# Visualization

MAX_EXAMPLE_FILES = 5

RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)


# ============================================================
# CREATE DIRECTORIES
# ============================================================

for directory in [
    OUTPUT_DIR,
    PROCESSED_DIR,
    REJECTED_DIR,
    BAND_POWER_DIR,
    REPORT_DIR,
    VIS_DIR
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def safe_read_csv(file_path):

    encodings = [
        "utf8",
        "utf8_sig",
        "latin1"
    ]

    last_error = None

    for encoding in encodings:

        try:

            df = pd.read_csv(
                file_path,
                encoding=encoding,
                low_memory=False
            )

            return df

        except Exception as error:

            last_error = error

    raise last_error


def find_column(df, candidates):

    lower_map = {
        str(column).strip().lower(): column
        for column in df.columns
    }

    for candidate in candidates:

        key = candidate.lower()

        if key in lower_map:

            return lower_map[key]

    return None


def find_raw_channels(df):

    available = []

    normalized_columns = {
        str(column).strip().upper(): column
        for column in df.columns
    }

    for channel in RAW_CHANNELS:

        key = channel.upper()

        if key in normalized_columns:

            available.append(
                normalized_columns[key]
            )

    return available


def convert_timestamp_to_seconds(series):

    numeric = pd.to_numeric(
        series,
        errors="coerce"
    )

    numeric_valid = numeric.dropna()

    if len(numeric_valid) > 10:

        values = numeric_valid.to_numpy(
            dtype=float
        )

        raw_difference = np.diff(values)

        raw_difference = raw_difference[
            np.isfinite(raw_difference)
        ]

        raw_difference = np.abs(
            raw_difference
        )

        raw_difference = raw_difference[
            raw_difference > 0
        ]

        if len(raw_difference) > 0:

            median_difference = np.median(
                raw_difference
            )

            maximum_value = np.nanmax(
                np.abs(values)
            )

            # Large epoch style values

            if maximum_value >= 1e17:

                scale = 1e9

            elif maximum_value >= 1e14:

                scale = 1e6

            elif maximum_value >= 1e11:

                scale = 1e3

            else:

                # Elapsed numeric timestamps

                if median_difference >= 0.5:

                    scale = 1e3

                else:

                    scale = 1.0

            seconds = values / scale

            result = pd.Series(
                seconds,
                index=series.index
            )

            return result


    # Try datetime conversion

    datetime_values = pd.to_datetime(
        series,
        errors="coerce"
    )

    if datetime_values.notna().sum() > 10:

        first_valid = datetime_values.dropna().iloc[0]

        seconds = (
            datetime_values - first_valid
        ).dt.total_seconds()

        return seconds


    return pd.Series(
        np.nan,
        index=series.index
    )


def timestamp_statistics(time_seconds):

    values = np.asarray(
        time_seconds,
        dtype=float
    )

    valid = np.isfinite(values)

    values = values[valid]

    result = {
        "timestamp_valid_points": len(values),
        "timestamp_unique_points": 0,
        "duplicate_points": 0,
        "duplicate_ratio": np.nan,
        "median_dt": np.nan,
        "mean_dt": np.nan,
        "minimum_dt": np.nan,
        "maximum_dt": np.nan,
        "sampling_rate_hz": np.nan,
        "sampling_rate_mean_hz": np.nan,
        "dt_cv": np.nan,
        "continuity_ratio": np.nan,
        "duration_seconds": np.nan
    }

    if len(values) < 3:

        return result

    unique_values = np.unique(values)

    result[
        "timestamp_unique_points"
    ] = len(unique_values)

    duplicate_count = len(values) - len(unique_values)

    result[
        "duplicate_points"
    ] = duplicate_count

    result[
        "duplicate_ratio"
    ] = duplicate_count / len(values)

    differences = np.diff(values)

    positive_differences = differences[
        differences > 0
    ]

    if len(positive_differences) == 0:

        return result

    median_dt = np.median(
        positive_differences
    )

    mean_dt = np.mean(
        positive_differences
    )

    minimum_dt = np.min(
        positive_differences
    )

    maximum_dt = np.max(
        positive_differences
    )

    sampling_rate = 1.0 / median_dt

    sampling_rate_mean = 1.0 / mean_dt

    dt_cv = (
        np.std(positive_differences)
        /
        mean_dt
        if mean_dt > 0
        else np.nan
    )

    tolerance = median_dt * 0.05

    continuity_ratio = np.mean(
        np.abs(
            positive_differences - median_dt
        ) <= tolerance
    )

    result[
        "median_dt"
    ] = median_dt

    result[
        "mean_dt"
    ] = mean_dt

    result[
        "minimum_dt"
    ] = minimum_dt

    result[
        "maximum_dt"
    ] = maximum_dt

    result[
        "sampling_rate_hz"
    ] = sampling_rate

    result[
        "sampling_rate_mean_hz"
    ] = sampling_rate_mean

    result[
        "dt_cv"
    ] = dt_cv

    result[
        "continuity_ratio"
    ] = continuity_ratio

    result[
        "duration_seconds"
    ] = values[-1] - values[0]

    return result


def estimate_sampling_rate(df, time_column):

    if time_column is None:

        return {
            "timestamp_status": "missing",
            "timestamp_valid_points": 0,
            "timestamp_unique_points": 0,
            "duplicate_points": np.nan,
            "duplicate_ratio": np.nan,
            "median_dt": np.nan,
            "mean_dt": np.nan,
            "minimum_dt": np.nan,
            "maximum_dt": np.nan,
            "sampling_rate_hz": np.nan,
            "sampling_rate_mean_hz": np.nan,
            "dt_cv": np.nan,
            "continuity_ratio": np.nan,
            "duration_seconds": np.nan
        }, None

    time_seconds = convert_timestamp_to_seconds(
        df[time_column]
    )

    statistics = timestamp_statistics(
        time_seconds
    )

    statistics[
        "timestamp_status"
    ] = "available"

    return statistics, time_seconds


def consecutive_runs(mask):

    mask = np.asarray(
        mask,
        dtype=bool
    )

    runs = []

    if len(mask) == 0:

        return runs

    start = None

    for index, value in enumerate(mask):

        if value and start is None:

            start = index

        elif not value and start is not None:

            runs.append(
                (start, index)
            )

            start = None

    if start is not None:

        runs.append(
            (start, len(mask))
        )

    return runs


def detect_flatline_mask(values, fs):

    values = np.asarray(
        values,
        dtype=float
    )

    mask = np.zeros(
        len(values),
        dtype=bool
    )

    if len(values) < 2:

        return mask

    if not np.isfinite(fs) or fs <= 0:

        minimum_samples = 5

    else:

        minimum_samples = max(
            5,
            int(
                np.ceil(
                    fs *
                    FLATLINE_DURATION_SECONDS
                )
            )
        )

    equal_previous = np.zeros(
        len(values),
        dtype=bool
    )

    equal_previous[1:] = (
        values[1:] ==
        values[:-1]
    )

    for start, end in consecutive_runs(
        equal_previous
    ):

        length = end - start

        if length >= minimum_samples:

            mask[start:end] = True

    return mask


def interpolate_short_gaps(
    values,
    time_seconds,
    fs
):

    values = np.asarray(
        values,
        dtype=float
    ).copy()

    original_missing = ~np.isfinite(
        values
    )

    if not np.any(original_missing):

        return values, 0

    missing_runs = consecutive_runs(
        original_missing
    )

    corrected = 0

    if np.isfinite(fs) and fs > 0:

        maximum_samples = max(
            1,
            int(
                np.floor(
                    SHORT_GAP_SECONDS * fs
                )
            )
        )

    else:

        maximum_samples = 10

    for start, end in missing_runs:

        gap_length = end - start

        if gap_length > maximum_samples:

            continue

        left = start - 1

        right = end

        if left < 0 or right >= len(values):

            continue

        if not np.isfinite(values[left]):

            continue

        if not np.isfinite(values[right]):

            continue

        if time_seconds is not None:

            left_time = time_seconds[left]

            right_time = time_seconds[right]

            if (
                not np.isfinite(left_time)
                or
                not np.isfinite(right_time)
            ):

                continue

            interpolation_times = (
                time_seconds[start:end]
            )

            values[start:end] = np.interp(
                interpolation_times,
                [
                    left_time,
                    right_time
                ],
                [
                    values[left],
                    values[right]
                ]
            )

        else:

            values[start:end] = np.linspace(
                values[left],
                values[right],
                gap_length + 2
            )[1:-1]

        corrected += gap_length

    return values, corrected


def robust_z_score(values):

    values = np.asarray(
        values,
        dtype=float
    )

    median = np.nanmedian(
        values
    )

    mad = np.nanmedian(
        np.abs(
            values - median
        )
    )

    if not np.isfinite(mad) or mad == 0:

        standard_deviation = np.nanstd(
            values
        )

        if (
            not np.isfinite(
                standard_deviation
            )
            or
            standard_deviation == 0
        ):

            return np.zeros(
                len(values)
            )

        return (
            values - median
        ) / standard_deviation

    return (
        0.6745 *
        (values - median)
        /
        mad
    )


def detect_artifacts(values):

    values = np.asarray(
        values,
        dtype=float
    )

    amplitude_z = np.abs(
        robust_z_score(values)
    )

    difference = np.diff(
        values,
        prepend=values[0]
    )

    difference_z = np.abs(
        robust_z_score(difference)
    )

    artifact_mask = (
        amplitude_z >
        ARTIFACT_Z_THRESHOLD
    ) | (
        difference_z >
        ARTIFACT_Z_THRESHOLD
    )

    artifact_mask[
        ~np.isfinite(values)
    ] = True

    return artifact_mask


def design_bandpass(fs):

    if fs <= 2 * HIGH_CUTOFF:

        return None

    sos = signal.butter(
        4,
        [
            LOW_CUTOFF,
            HIGH_CUTOFF
        ],
        btype="bandpass",
        fs=fs,
        output="sos"
    )

    return sos


def design_notch(fs):

    if fs <= 2 * NOTCH_FREQUENCY:

        return None

    b, a = signal.iirnotch(
        NOTCH_FREQUENCY,
        NOTCH_Q,
        fs
    )

    return b, a


def filter_valid_segments(
    values,
    fs
):

    values = np.asarray(
        values,
        dtype=float
    ).copy()

    output = values.copy()

    bandpass_sos = design_bandpass(
        fs
    )

    notch_filter = design_notch(
        fs
    )

    if bandpass_sos is None:

        return output, 0

    valid_mask = np.isfinite(
        values
    )

    segments = consecutive_runs(
        valid_mask
    )

    filtered_segments = 0

    for start, end in segments:

        segment_length = end - start

        minimum_length = max(
            100,
            int(fs * 2)
        )

        if segment_length < minimum_length:

            continue

        segment = values[
            start:end
        ].copy()

        try:

            filtered = signal.sosfiltfilt(
                bandpass_sos,
                segment
            )

            if notch_filter is not None:

                b, a = notch_filter

                filtered = signal.filtfilt(
                    b,
                    a,
                    filtered
                )

            output[
                start:end
            ] = filtered

            filtered_segments += 1

        except Exception:

            continue

    return output, filtered_segments


def calculate_band_power(
    values,
    fs
):

    results = {}

    values = np.asarray(
        values,
        dtype=float
    )

    valid = np.isfinite(
        values
    )

    clean = values[
        valid
    ]

    if len(clean) < max(
        256,
        int(fs)
    ):

        for band in EEG_BANDS:

            results[band] = np.nan

        return results

    nperseg = min(
        len(clean),
        max(
            256,
            int(fs * 2)
        )
    )

    try:

        frequencies, power = signal.welch(
            clean,
            fs=fs,
            nperseg=nperseg
        )

    except Exception:

        for band in EEG_BANDS:

            results[band] = np.nan

        return results

    for band_name, limits in EEG_BANDS.items():

        low, high = limits

        band_mask = (
            (frequencies >= low)
            &
            (frequencies <= high)
        )

        if np.any(band_mask):

            band_power = np.trapezoid(
                power[band_mask],
                frequencies[band_mask]
            )

        else:

            band_power = np.nan

        results[
            band_name
        ] = band_power

    return results


def run_ica(
    data,
    fs
):

    if not USE_ICA:

        return data, {
            "ica_used": False,
            "ica_removed_components": 0,
            "ica_reason": "disabled"
        }

    if data.shape[1] < 4:

        return data, {
            "ica_used": False,
            "ica_removed_components": 0,
            "ica_reason": "fewer_than_four_channels"
        }

    valid_rows = np.all(
        np.isfinite(data),
        axis=1
    )

    valid_data = data[
        valid_rows
    ]

    if len(valid_data) < max(
        1000,
        int(fs * 10)
    ):

        return data, {
            "ica_used": False,
            "ica_removed_components": 0,
            "ica_reason": "insufficient_valid_samples"
        }

    # Limit ICA fitting data to avoid excessive memory use

    maximum_ica_samples = 100000

    if len(valid_data) > maximum_ica_samples:

        indices = np.linspace(
            0,
            len(valid_data) - 1,
            maximum_ica_samples
        ).astype(int)

        fitting_data = valid_data[
            indices
        ]

    else:

        fitting_data = valid_data

    try:

        ica = FastICA(
            n_components=data.shape[1],
            whiten="unit-variance",
            max_iter=ICA_MAX_ITER,
            tol=ICA_TOLERANCE,
            random_state=RANDOM_SEED
        )

        transformed = ica.fit_transform(
            fitting_data
        )

        component_kurtosis = np.abs(
            kurtosis(
                transformed,
                axis=0,
                fisher=True,
                nan_policy="omit"
            )
        )

        bad_components = (
            component_kurtosis >
            ICA_KURTOSIS_THRESHOLD
        )

        number_removed = int(
            np.sum(
                bad_components
            )
        )

        # Avoid removing every component

        if number_removed >= data.shape[1]:

            return data, {
                "ica_used": False,
                "ica_removed_components": 0,
                "ica_reason": "all_components_flagged"
            }

        # Transform all valid data

        full_transformed = ica.transform(
            valid_data
        )

        full_transformed[
            :,
            bad_components
        ] = 0

        reconstructed = ica.inverse_transform(
            full_transformed
        )

        output = data.copy()

        output[
            valid_rows
        ] = reconstructed

        return output, {
            "ica_used": True,
            "ica_removed_components": number_removed,
            "ica_reason": "kurtosis_based_component_rejection"
        }

    except Exception as error:

        return data, {
            "ica_used": False,
            "ica_removed_components": 0,
            "ica_reason": str(error)
        }


def classify_file(
    sampling_stats,
    raw_channels,
    df
):

    if len(raw_channels) != len(
        RAW_CHANNELS
    ):

        return "missing_raw_channels"

    if (
        sampling_stats[
            "timestamp_status"
        ]
        !=
        "available"
    ):

        return "missing_timestamp"

    fs = sampling_stats[
        "sampling_rate_hz"
    ]

    if not np.isfinite(fs):

        return "invalid_sampling_rate"

    if (
        fs < MIN_USABLE_FS
        or
        fs > MAX_USABLE_FS
    ):

        return "sampling_rate_outside_range"

    duplicate_ratio = sampling_stats[
        "duplicate_ratio"
    ]

    if (
        np.isfinite(duplicate_ratio)
        and
        duplicate_ratio >
        MAX_DUPLICATE_RATIO
    ):

        return "too_many_duplicate_timestamps"

    jitter = sampling_stats[
        "dt_cv"
    ]

    if (
        np.isfinite(jitter)
        and
        jitter >
        MAX_JITTER_CV
    ):

        return "irregular_sampling"

    continuity = sampling_stats[
        "continuity_ratio"
    ]

    if (
        np.isfinite(continuity)
        and
        continuity <
        0.90
    ):

        return "poor_timestamp_continuity"

    return "usable"


# ============================================================
# VISUALIZATION FUNCTIONS
# ============================================================

def save_sampling_rate_plot(report):

    if report.empty:

        return

    values = report[
        "sampling_rate_hz"
    ].dropna()

    if len(values) == 0:

        return

    plt.figure(
        figsize=(12, 7)
    )

    plt.hist(
        values,
        bins=40
    )

    plt.axvline(
        MIN_USABLE_FS,
        linestyle="--",
        label="Minimum usable rate"
    )

    plt.xlabel(
        "Sampling rate, Hz"
    )

    plt.ylabel(
        "Number of files"
    )

    plt.title(
        "RAW EEG sampling rate distribution"
    )

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "sampling_rate_distribution.png",
        dpi=200
    )

    plt.close()


def save_dt_distribution(report):

    values = report[
        "median_dt"
    ].dropna()

    if len(values) == 0:

        return

    plt.figure(
        figsize=(12, 7)
    )

    plt.hist(
        values,
        bins=40
    )

    plt.xlabel(
        "Median timestamp interval, seconds"
    )

    plt.ylabel(
        "Number of files"
    )

    plt.title(
        "Timestamp interval distribution"
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "timestamp_interval_distribution.png",
        dpi=200
    )

    plt.close()


def save_status_plot(report):

    counts = report[
        "processing_status"
    ].value_counts()

    plt.figure(
        figsize=(12, 7)
    )

    plt.bar(
        counts.index.astype(str),
        counts.values
    )

    plt.xticks(
        rotation=45,
        ha="right"
    )

    plt.ylabel(
        "Number of files"
    )

    plt.title(
        "File processing status"
    )

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "file_processing_status.png",
        dpi=200
    )

    plt.close()


def save_quality_plot(report):

    channels = RAW_CHANNELS

    zero_columns = [
        f"{channel}_zero_count"
        for channel in channels
    ]

    flat_columns = [
        f"{channel}_flatline_count"
        for channel in channels
    ]

    zero_values = [
        report[column].sum()
        for column in zero_columns
        if column in report
    ]

    flat_values = [
        report[column].sum()
        for column in flat_columns
        if column in report
    ]

    x = np.arange(
        len(channels)
    )

    width = 0.35

    plt.figure(
        figsize=(12, 7)
    )

    plt.bar(
        x - width / 2,
        zero_values,
        width,
        label="Zero values"
    )

    plt.bar(
        x + width / 2,
        flat_values,
        width,
        label="Flatline points"
    )

    plt.xticks(
        x,
        channels
    )

    plt.ylabel(
        "Number of points"
    )

    plt.title(
        "RAW EEG signal quality problems"
    )

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        VIS_DIR /
        "zero_and_flatline_summary.png",
        dpi=200
    )

    plt.close()


def save_example_plot(
    original_df,
    processed_df,
    time_seconds,
    file_name
):

    if time_seconds is None:

        x = np.arange(
            len(original_df)
        )

        xlabel = "Sample"

    else:

        x = time_seconds

        xlabel = "Time, seconds"

    max_points = min(
        len(x),
        5000
    )

    x = x[:max_points]

    fig, axes = plt.subplots(
        len(RAW_CHANNELS),
        1,
        figsize=(14, 12),
        sharex=True
    )

    if len(RAW_CHANNELS) == 1:

        axes = [axes]

    for index, channel in enumerate(
        RAW_CHANNELS
    ):

        if channel not in original_df:

            continue

        axes[index].plot(
            x,
            original_df[
                channel
            ].to_numpy()[
                :max_points
            ],
            linewidth=0.6,
            alpha=0.7,
            label="Original"
        )

        filtered_column = (
            f"FILTERED_{channel}"
        )

        if filtered_column in processed_df:

            axes[index].plot(
                x,
                processed_df[
                    filtered_column
                ].to_numpy()[
                    :max_points
                ],
                linewidth=0.7,
                label="Processed"
            )

        axes[index].set_ylabel(
            channel
        )

        axes[index].legend(
            loc="upper right"
        )

    axes[-1].set_xlabel(
        xlabel
    )

    fig.suptitle(
        f"RAW EEG before and after processing: {file_name}"
    )

    plt.tight_layout()

    output_name = (
        Path(file_name).stem
        +
        "_before_after.png"
    )

    plt.savefig(
        VIS_DIR /
        output_name,
        dpi=200
    )

    plt.close()


def save_psd_plot(
    df,
    fs,
    file_name
):

    if not np.isfinite(fs):

        return

    plt.figure(
        figsize=(14, 8)
    )

    for channel in RAW_CHANNELS:

        filtered_column = (
            f"FILTERED_{channel}"
        )

        if filtered_column not in df:

            continue

        values = df[
            filtered_column
        ].to_numpy(
            dtype=float
        )

        values = values[
            np.isfinite(values)
        ]

        if len(values) < 256:

            continue

        nperseg = min(
            len(values),
            max(
                256,
                int(fs * 2)
            )
        )

        frequencies, power = signal.welch(
            values,
            fs=fs,
            nperseg=nperseg
        )

        mask = (
            frequencies <= 60
        )

        plt.semilogy(
            frequencies[mask],
            power[mask],
            label=channel
        )

    plt.xlabel(
        "Frequency, Hz"
    )

    plt.ylabel(
        "Power spectral density"
    )

    plt.title(
        f"Processed RAW EEG power spectrum: {file_name}"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    output_name = (
        Path(file_name).stem
        +
        "_PSD.png"
    )

    plt.savefig(
        VIS_DIR /
        output_name,
        dpi=200
    )

    plt.close()


# ============================================================
# MAIN PROCESSING
# ============================================================

all_files = sorted(
    INPUT_DIR.rglob("*.csv")
)

print("=" * 80)
print("RAW EEG FINAL PROCESSING")
print("=" * 80)

print(
    f"Input directory: {INPUT_DIR}"
)

print(
    f"CSV files found: {len(all_files)}"
)

print("=" * 80)


audit_records = []

processed_examples = []


for file_index, file_path in enumerate(
    all_files,
    start=1
):

    relative_path = file_path.relative_to(
        INPUT_DIR
    )

    print(
        f"[{file_index}/{len(all_files)}] "
        f"{relative_path}"
    )

    record = {
        "file": str(relative_path),
        "full_path": str(file_path),
        "processing_status": "not_processed",
        "processing_reason": ""
    }

    try:

        df = safe_read_csv(
            file_path
        )

        record[
            "rows"
        ] = len(df)

        record[
            "columns"
        ] = len(df.columns)

        raw_channels = find_raw_channels(
            df
        )

        record[
            "raw_channel_count"
        ] = len(raw_channels)

        for channel in RAW_CHANNELS:

            record[
                f"{channel}_zero_count"
            ] = 0

            record[
                f"{channel}_flatline_count"
            ] = 0

            record[
                f"{channel}_missing_before"
            ] = 0

            record[
                f"{channel}_missing_after"
            ] = 0

            record[
                f"{channel}_artifact_count"
            ] = 0

            record[
                f"{channel}_short_gap_corrected"
            ] = 0

        time_column = find_column(
            df,
            TIME_CANDIDATES
        )

        record[
            "timestamp_column"
        ] = time_column

        sampling_stats, time_seconds = (
            estimate_sampling_rate(
                df,
                time_column
            )
        )

        record.update(
            sampling_stats
        )

        status = classify_file(
            sampling_stats,
            raw_channels,
            df
        )

        if status != "usable":

            record[
                "processing_status"
            ] = status

            record[
                "processing_reason"
            ] = status

            audit_records.append(
                record
            )

            rejected_path = (
                REJECTED_DIR /
                relative_path
            )

            rejected_path.parent.mkdir(
                parents=True,
                exist_ok=True
            )

            shutil.copy2(
                file_path,
                rejected_path
            )

            continue


        fs = sampling_stats[
            "sampling_rate_hz"
        ]


        # ----------------------------------------------------
        # COPY DATAFRAME
        # ----------------------------------------------------

        original_df = df.copy()

        processed_df = df.copy()


        # ----------------------------------------------------
        # RAW EEG CLEANING
        # ----------------------------------------------------

        cleaned_signals = {}

        for channel in RAW_CHANNELS:

            values = pd.to_numeric(
                df[channel],
                errors="coerce"
            ).to_numpy(
                dtype=float
            )

            missing_before = int(
                np.sum(
                    ~np.isfinite(values)
                )
            )

            zero_mask = (
                np.isfinite(values)
                &
                (values == 0)
            )

            zero_count = int(
                np.sum(zero_mask)
            )

            values[
                zero_mask
            ] = np.nan


            flatline_mask = detect_flatline_mask(
                values,
                fs
            )

            flatline_count = int(
                np.sum(flatline_mask)
            )

            values[
                flatline_mask
            ] = np.nan


            corrected_values, corrected_count = (
                interpolate_short_gaps(
                    values,
                    (
                        time_seconds.to_numpy()
                        if time_seconds is not None
                        else None
                    ),
                    fs
                )
            )


            artifact_mask = detect_artifacts(
                corrected_values
            )

            artifact_count = int(
                np.sum(
                    artifact_mask
                )
            )

            corrected_values[
                artifact_mask
            ] = np.nan


            corrected_values, artifact_gap_correction = (
                interpolate_short_gaps(
                    corrected_values,
                    (
                        time_seconds.to_numpy()
                        if time_seconds is not None
                        else None
                    ),
                    fs
                )
            )

            cleaned_signals[
                channel
            ] = corrected_values


            record[
                f"{channel}_zero_count"
            ] = zero_count

            record[
                f"{channel}_flatline_count"
            ] = flatline_count

            record[
                f"{channel}_missing_before"
            ] = missing_before

            record[
                f"{channel}_missing_after"
            ] = int(
                np.sum(
                    ~np.isfinite(
                        corrected_values
                    )
                )
            )

            record[
                f"{channel}_artifact_count"
            ] = artifact_count

            record[
                f"{channel}_short_gap_corrected"
            ] = (
                corrected_count
                +
                artifact_gap_correction
            )


        # ----------------------------------------------------
        # BANDPASS AND NOTCH FILTERING
        # ----------------------------------------------------

        filtered_signals = {}

        total_filtered_segments = 0

        for channel in RAW_CHANNELS:

            filtered_values, segment_count = (
                filter_valid_segments(
                    cleaned_signals[
                        channel
                    ],
                    fs
                )
            )

            filtered_signals[
                channel
            ] = filtered_values

            total_filtered_segments += (
                segment_count
            )

            processed_df[
                f"FILTERED_{channel}"
            ] = filtered_values


        record[
            "filtered_segments"
        ] = total_filtered_segments


        # ----------------------------------------------------
        # ICA
        # ----------------------------------------------------

        matrix = np.column_stack(
            [
                filtered_signals[
                    channel
                ]
                for channel in RAW_CHANNELS
            ]
        )

        ica_output, ica_info = run_ica(
            matrix,
            fs
        )

        record.update(
            ica_info
        )


        for index, channel in enumerate(
            RAW_CHANNELS
        ):

            processed_df[
                f"ICA_{channel}"
            ] = ica_output[
                :,
                index
            ]


        # ----------------------------------------------------
        # BAND POWER
        # ----------------------------------------------------

        band_record = {
            "file": str(relative_path),
            "sampling_rate_hz": fs,
            "duration_seconds":
                sampling_stats[
                    "duration_seconds"
                ]
        }


        for channel in RAW_CHANNELS:

            values = ica_output[
                :,
                RAW_CHANNELS.index(
                    channel
                )
            ]

            band_values = calculate_band_power(
                values,
                fs
            )

            for band_name, power in band_values.items():

                band_record[
                    f"{channel}_{band_name}_power"
                ] = power


        band_power_records = pd.DataFrame(
            [band_record]
        )

        band_output_path = (
            BAND_POWER_DIR /
            relative_path.parent /
            (
                relative_path.stem
                +
                "_band_power.csv"
            )
        )

        band_output_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        band_power_records.to_csv(
            band_output_path,
            index=False
        )


        # ----------------------------------------------------
        # FINAL VALIDATION
        # ----------------------------------------------------

        final_valid_ratios = []

        for channel in RAW_CHANNELS:

            values = processed_df[
                f"ICA_{channel}"
            ].to_numpy(
                dtype=float
            )

            valid_ratio = np.mean(
                np.isfinite(values)
            )

            final_valid_ratios.append(
                valid_ratio
            )

        final_valid_ratio = float(
            np.mean(
                final_valid_ratios
            )
        )

        record[
            "final_valid_ratio"
        ] = final_valid_ratio

        record[
            "processing_status"
        ] = "processed"

        record[
            "processing_reason"
        ] = "sampling_and_signal_quality_acceptable"


        # ----------------------------------------------------
        # SAVE FINAL DATA
        # ----------------------------------------------------

        output_path = (
            PROCESSED_DIR /
            relative_path
        )

        output_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        processed_df.to_csv(
            output_path,
            index=False
        )


        # ----------------------------------------------------
        # VISUALIZATION
        # ----------------------------------------------------

        if len(
            processed_examples
        ) < MAX_EXAMPLE_FILES:

            processed_examples.append(
                (
                    original_df.copy(),
                    processed_df.copy(),
                    (
                        time_seconds.to_numpy()
                        if time_seconds is not None
                        else None
                    ),
                    fs,
                    str(relative_path)
                )
            )


    except Exception as error:

        record[
            "processing_status"
        ] = "processing_error"

        record[
            "processing_reason"
        ] = str(error)

        print(
            f"ERROR: {error}"
        )


    audit_records.append(
        record
    )


# ============================================================
# CREATE GLOBAL REPORT
# ============================================================

report = pd.DataFrame(
    audit_records
)

report_path = (
    REPORT_DIR /
    "raw_eeg_final_processing_report.csv"
)

report.to_csv(
    report_path,
    index=False
)


# ============================================================
# SAMPLING RATE REPORT
# ============================================================

sampling_columns = [
    "file",
    "timestamp_column",
    "timestamp_valid_points",
    "timestamp_unique_points",
    "duplicate_points",
    "duplicate_ratio",
    "median_dt",
    "mean_dt",
    "minimum_dt",
    "maximum_dt",
    "sampling_rate_hz",
    "sampling_rate_mean_hz",
    "dt_cv",
    "continuity_ratio",
    "duration_seconds",
    "processing_status",
    "processing_reason"
]

available_sampling_columns = [
    column
    for column in sampling_columns
    if column in report.columns
]

sampling_report = report[
    available_sampling_columns
]

sampling_report.to_csv(
    REPORT_DIR /
    "timestamp_sampling_rate_report.csv",
    index=False
)


# ============================================================
# QUALITY REPORT
# ============================================================

quality_columns = [
    "file",
    "rows",
    "raw_channel_count"
]

for channel in RAW_CHANNELS:

    quality_columns.extend(
        [
            f"{channel}_zero_count",
            f"{channel}_flatline_count",
            f"{channel}_missing_before",
            f"{channel}_missing_after",
            f"{channel}_artifact_count",
            f"{channel}_short_gap_corrected"
        ]
    )

quality_columns.extend(
    [
        "final_valid_ratio",
        "processing_status"
    ]
)

available_quality_columns = [
    column
    for column in quality_columns
    if column in report.columns
]

quality_report = report[
    available_quality_columns
]

quality_report.to_csv(
    REPORT_DIR /
    "raw_eeg_quality_report.csv",
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

status_counts = report[
    "processing_status"
].value_counts()

status_counts.to_csv(
    REPORT_DIR /
    "processing_status_summary.csv"
)


# ============================================================
# GLOBAL SUMMARY JSON
# ============================================================

summary = {
    "input_directory": str(INPUT_DIR),
    "output_directory": str(OUTPUT_DIR),
    "total_files": int(len(report)),
    "processed_files": int(
        np.sum(
            report[
                "processing_status"
            ] == "processed"
        )
    ),
    "rejected_files": int(
        np.sum(
            report[
                "processing_status"
            ] != "processed"
        )
    ),
    "median_sampling_rate_hz":
        float(
            report[
                "sampling_rate_hz"
            ].median()
        )
        if "sampling_rate_hz" in report
        else None,
    "mean_sampling_rate_hz":
        float(
            report[
                "sampling_rate_hz"
            ].mean()
        )
        if "sampling_rate_hz" in report
        else None,
    "total_zero_values": int(
        sum(
            report[
                f"{channel}_zero_count"
            ].sum()
            for channel in RAW_CHANNELS
            if f"{channel}_zero_count"
            in report.columns
        )
    ),
    "total_flatline_points": int(
        sum(
            report[
                f"{channel}_flatline_count"
            ].sum()
            for channel in RAW_CHANNELS
            if f"{channel}_flatline_count"
            in report.columns
        )
    ),
    "total_artifact_points": int(
        sum(
            report[
                f"{channel}_artifact_count"
            ].sum()
            for channel in RAW_CHANNELS
            if f"{channel}_artifact_count"
            in report.columns
        )
    )
}

with open(
    REPORT_DIR /
    "final_processing_summary.json",
    "w",
    encoding="utf8"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )


# ============================================================
# CREATE GLOBAL VISUALIZATIONS
# ============================================================

save_sampling_rate_plot(
    report
)

save_dt_distribution(
    report
)

save_status_plot(
    report
)

save_quality_plot(
    report
)


# ============================================================
# EXAMPLE SIGNAL VISUALIZATIONS
# ============================================================

for (
    original_df,
    processed_df,
    time_seconds,
    fs,
    file_name
) in processed_examples:

    save_example_plot(
        original_df,
        processed_df,
        time_seconds,
        file_name
    )

    save_psd_plot(
        processed_df,
        fs,
        file_name
    )


# ============================================================
# FINAL CONSOLE SUMMARY
# ============================================================

print()
print("=" * 80)
print("RAW EEG FINAL PROCESSING COMPLETED")
print("=" * 80)

print(
    f"Total files: {len(report)}"
)

print(
    f"Processed files: "
    f"{summary['processed_files']}"
)

print(
    f"Rejected files: "
    f"{summary['rejected_files']}"
)

print(
    f"Median apparent sampling rate: "
    f"{summary['median_sampling_rate_hz']}"
)

print(
    f"Total zero values: "
    f"{summary['total_zero_values']}"
)

print(
    f"Total flatline points: "
    f"{summary['total_flatline_points']}"
)

print(
    f"Total detected artifact points: "
    f"{summary['total_artifact_points']}"
)

print()
print(
    f"Final processed dataset: "
    f"{PROCESSED_DIR}"
)

print(
    f"Rejected files: "
    f"{REJECTED_DIR}"
)

print(
    f"Band power: "
    f"{BAND_POWER_DIR}"
)

print(
    f"Reports: "
    f"{REPORT_DIR}"
)

print(
    f"Visualizations: "
    f"{VIS_DIR}"
)

print("=" * 80)
```

```python
# ============================================================
# EEG DATA PROCESSING PIPELINE
# Corrected timestamp and sampling rate handling
# ============================================================

import os
import shutil
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

from scipy.signal import (
    butter,
    filtfilt,
    iirnotch,
    savgol_filter,
    welch
)

warnings.filterwarnings("ignore")


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_DIR = Path(
    r"/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V10/clean_data"
)

OUTPUT_DIR = Path(
    r"/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_FINAL"
)

PROCESSED_DIR = OUTPUT_DIR / "processed"
REJECTED_DIR = OUTPUT_DIR / "rejected"
BAND_POWER_DIR = OUTPUT_DIR / "band_power"
REPORT_DIR = OUTPUT_DIR / "reports"
VIS_DIR = OUTPUT_DIR / "visualizations"

for folder in [
    PROCESSED_DIR,
    REJECTED_DIR,
    BAND_POWER_DIR,
    REPORT_DIR,
    VIS_DIR
]:
    folder.mkdir(parents=True, exist_ok=True)


# ============================================================
# CHANNELS
# ============================================================

RAW_CHANNELS = [
    "RAW_TP9",
    "RAW_AF7",
    "RAW_AF8",
    "RAW_TP10"
]

BAND_CHANNELS = [
    "Delta",
    "Theta",
    "Alpha",
    "Beta",
    "Gamma"
]


# ============================================================
# TIMESTAMP DETECTION
# ============================================================

def find_timestamp_column(df):

    possible_names = [
        "Time",
        "Timestamp",
        "timestamp",
        "TimeStamp",
        "time",
        "Date",
        "datetime",
        "Datetime"
    ]

    for name in possible_names:
        if name in df.columns:
            return name

    return None


def convert_timestamp(series):

    """
    Automatically tries common timestamp formats.

    Numeric timestamps are tested as:
    seconds
    milliseconds
    microseconds
    """

    if pd.api.types.is_numeric_dtype(series):

        values = pd.to_numeric(series, errors="coerce")

        median_value = np.nanmedian(values)

        if median_value > 1e14:
            # microseconds
            ts = pd.to_datetime(
                values,
                unit="us",
                errors="coerce"
            )

        elif median_value > 1e11:
            # milliseconds
            ts = pd.to_datetime(
                values,
                unit="ms",
                errors="coerce"
            )

        elif median_value > 1e9:
            # seconds since Unix epoch
            ts = pd.to_datetime(
                values,
                unit="s",
                errors="coerce"
            )

        else:
            # Could be elapsed seconds
            ts = pd.to_timedelta(
                values,
                unit="s"
            )

    else:

        ts = pd.to_datetime(
            series,
            errors="coerce"
        )

    return ts


# ============================================================
# SAMPLING RATE
# ============================================================

def calculate_sampling_rate(df, timestamp_column):

    ts = convert_timestamp(df[timestamp_column])

    valid_ts = ts.dropna()

    if len(valid_ts) < 3:
        return np.nan, np.nan

    differences = valid_ts.diff().dt.total_seconds()

    differences = differences[
        np.isfinite(differences)
    ]

    differences = differences[
        differences > 0
    ]

    if len(differences) == 0:
        return np.nan, np.nan

    median_interval = np.median(differences)

    sampling_rate = 1.0 / median_interval

    return sampling_rate, median_interval


# ============================================================
# DATA CLEANING
# ============================================================

def clean_numeric_columns(df, columns):

    for col in columns:

        if col not in df.columns:
            continue

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        # Replace infinite values
        df[col] = df[col].replace(
            [np.inf, -np.inf],
            np.nan
        )

        # Replace zero values with NaN
        # Only for EEG channels
        df.loc[df[col] == 0, col] = np.nan

        # Linear interpolation
        df[col] = df[col].interpolate(
            method="linear",
            limit_direction="both"
        )

        # Median fallback
        if df[col].isna().any():

            median_value = df[col].median()

            df[col] = df[col].fillna(
                median_value
            )

    return df


# ============================================================
# SAFE SMOOTHING
# ============================================================

def smooth_signal(signal):

    signal = np.asarray(signal, dtype=float)

    n = len(signal)

    if n < 7:
        return signal

    window = min(11, n)

    if window % 2 == 0:
        window -= 1

    if window < 5:
        return signal

    return savgol_filter(
        signal,
        window_length=window,
        polyorder=2
    )


# ============================================================
# RAW EEG FILTERING
# ============================================================

def bandpass_filter(
    signal,
    fs,
    lowcut=1.0,
    highcut=45.0
):

    nyquist = fs / 2

    if fs <= 2:
        return signal

    highcut = min(
        highcut,
        nyquist * 0.90
    )

    if lowcut >= highcut:
        return signal

    b, a = butter(
        4,
        [
            lowcut / nyquist,
            highcut / nyquist
        ],
        btype="band"
    )

    return filtfilt(
        b,
        a,
        signal
    )


def notch_filter(
    signal,
    fs,
    frequency=50.0
):

    if fs <= 2:
        return signal

    nyquist = fs / 2

    if frequency >= nyquist:
        return signal

    b, a = iirnotch(
        frequency,
        Q=30,
        fs=fs
    )

    return filtfilt(
        b,
        a,
        signal
    )


# ============================================================
# BAND POWER
# ============================================================

EEG_BANDS = {
    "delta": (1, 4),
    "theta": (4, 8),
    "alpha": (8, 13),
    "beta": (13, 30),
    "gamma": (30, 45)
}


def calculate_band_power(signal, fs):

    result = {}

    if len(signal) < 10:
        return {
            band: np.nan
            for band in EEG_BANDS
        }

    if fs <= 2:

        return {
            band: np.nan
            for band in EEG_BANDS
        }

    frequencies, power = welch(
        signal,
        fs=fs,
        nperseg=min(
            256,
            len(signal)
        )
    )

    for band, (
        low,
        high
    ) in EEG_BANDS.items():

        mask = (
            (frequencies >= low) &
            (frequencies <= high)
        )

        if mask.any():

            result[band] = np.trapezoid(
                power[mask],
                frequencies[mask]
            )

        else:

            result[band] = np.nan

    return result


# ============================================================
# PROCESS ONE FILE
# ============================================================

def process_file(file_path):

    df = pd.read_csv(
        file_path,
        low_memory=False
    )

    timestamp_column = find_timestamp_column(df)

    if timestamp_column is None:

        return {
            "status": "REJECTED",
            "reason": "No timestamp column"
        }

    fs, interval = calculate_sampling_rate(
        df,
        timestamp_column
    )

    # --------------------------------------------------------
    # Determine available signal type
    # --------------------------------------------------------

    available_raw = [
        col for col in RAW_CHANNELS
        if col in df.columns
    ]

    available_band = [
        col for col in BAND_CHANNELS
        if col in df.columns
    ]

    if len(available_raw) == 0 and len(available_band) == 0:

        return {
            "status": "REJECTED",
            "reason": "No EEG or band power channels",
            "sampling_rate": fs
        }

    # --------------------------------------------------------
    # Clean raw channels if present
    # --------------------------------------------------------

    if available_raw:

        df = clean_numeric_columns(
            df,
            available_raw
        )

    # --------------------------------------------------------
    # Clean band channels
    # --------------------------------------------------------

    if available_band:

        df = clean_numeric_columns(
            df,
            available_band
        )

    # ========================================================
    # CASE A
    # Real raw EEG sampling rate is sufficient
    # ========================================================

    if (
        len(available_raw) > 0
        and np.isfinite(fs)
        and fs >= 20
    ):

        filtered_channels = {}

        for channel in available_raw:

            signal = df[channel].values

            signal = bandpass_filter(
                signal,
                fs
            )

            signal = notch_filter(
                signal,
                fs
            )

            filtered_channels[channel] = signal

        # Average four EEG channels

        eeg_matrix = np.column_stack(
            [
                filtered_channels[c]
                for c in available_raw
            ]
        )

        average_eeg = np.nanmean(
            eeg_matrix,
            axis=1
        )

        # Calculate band power for each sample window

        band_result = {}

        for band in EEG_BANDS:
            band_result[band] = np.full(
                len(df),
                np.nan
            )

        # Use a window appropriate for the available data
        window = int(max(fs * 2, 20))

        if window < len(average_eeg):

            for start in range(
                0,
                len(average_eeg) - window + 1,
                window
            ):

                segment = average_eeg[
                    start:start + window
                ]

                powers = calculate_band_power(
                    segment,
                    fs
                )

                end = start + window

                for band in powers:

                    band_result[band][
                        start:end
                    ] = powers[band]

        for band in band_result:

            df[
                f"{band}_power"
            ] = band_result[band]

        processing_mode = "RAW_EEG"

    # ========================================================
    # CASE B
    # Low rate data
    #
    # Do NOT pretend this is high frequency raw EEG.
    # Use the existing Mind Monitor band channels.
    # ========================================================

    elif len(available_band) > 0:

        for channel in available_band:

            df[channel] = smooth_signal(
                df[channel].values
            )

        processing_mode = "BAND_POWER_ONLY"

    # ========================================================
    # CASE C
    # Raw EEG exists but sampling rate is too low
    # ========================================================

    else:

        processing_mode = "LOW_RATE_RAW"

    # ========================================================
    # Save processed data
    # ========================================================

    relative_path = file_path.relative_to(
        INPUT_DIR
    )

    output_file = (
        PROCESSED_DIR /
        relative_path
    )

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    df.to_csv(
        output_file,
        index=False
    )

    # ========================================================
    # Save band power data
    # ========================================================

    if available_band:

        band_columns = [
            timestamp_column
        ] + available_band

        band_df = df[
            [
                c for c in band_columns
                if c in df.columns
            ]
        ].copy()

    elif any(
        f"{b}_power" in df.columns
        for b in EEG_BANDS
    ):

        power_columns = [
            timestamp_column
        ] + [
            f"{b}_power"
            for b in EEG_BANDS
            if f"{b}_power" in df.columns
        ]

        band_df = df[
            power_columns
        ].copy()

    else:

        band_df = pd.DataFrame()

    if not band_df.empty:

        band_output = (
            BAND_POWER_DIR /
            relative_path
        )

        band_output.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        band_df.to_csv(
            band_output,
            index=False
        )

    return {
        "status": "PROCESSED",
        "reason": "",
        "sampling_rate": fs,
        "timestamp_interval": interval,
        "processing_mode": processing_mode,
        "raw_channels": len(available_raw),
        "band_channels": len(available_band),
        "rows": len(df)
    }


# ============================================================
# MAIN PROCESSING
# ============================================================

all_files = list(
    INPUT_DIR.rglob("*.csv")
)

print("=" * 70)
print("EEG PROCESSING STARTED")
print("=" * 70)

print(
    f"Total CSV files: {len(all_files)}"
)

results = []

for index, file_path in enumerate(
    all_files,
    start=1
):

    print(
        f"[{index}/{len(all_files)}] "
        f"{file_path.relative_to(INPUT_DIR)}"
    )

    try:

        result = process_file(
            file_path
        )

        result["file"] = str(
            file_path.relative_to(INPUT_DIR)
        )

        results.append(result)

    except Exception as e:

        print(
            "ERROR:",
            str(e)
        )

        results.append(
            {
                "file": str(
                    file_path.relative_to(
                        INPUT_DIR
                    )
                ),
                "status": "REJECTED",
                "reason": str(e)
            }
        )


# ============================================================
# REPORT
# ============================================================

report = pd.DataFrame(
    results
)

report.to_csv(
    REPORT_DIR /
    "processing_report.csv",
    index=False
)

processed = (
    report["status"] == "PROCESSED"
).sum()

rejected = (
    report["status"] == "REJECTED"
).sum()

print()
print("=" * 70)
print("PROCESSING COMPLETED")
print("=" * 70)

print(
    f"Total files     : {len(all_files)}"
)

print(
    f"Processed files : {processed}"
)

print(
    f"Rejected files  : {rejected}"
)

if "sampling_rate" in report.columns:

    print(
        "Median apparent sampling rate:",
        report["sampling_rate"].median()
    )

print()
print(
    "Processing report saved to:"
)

print(
    REPORT_DIR /
    "processing_report.csv"
)
```

```python
from pathlib import Path
import pandas as pd
from datetime import datetime, timedelta
import shutil


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_DIR = Path("CLEAN_DATA_V12")
OUTPUT_DIR = Path("CLEAN_DATA_V13")

START_TIME = datetime.strptime(
    "09:00:01",
    "%H:%M:%S"
)

END_TIME = datetime.strptime(
    "10:00:01",
    "%H:%M:%S"
)


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# CREATE TIMESTAMP SEQUENCE
#
# Required format:
# 9:00:01 AM
# 9:00:02 AM
# 9:00:03 AM
# ...
# 10:00:01 AM
# ============================================================

timestamp_list = []

current_time = START_TIME

while current_time <= END_TIME:

    timestamp_list.append(
        current_time.strftime("%-I:%M:%S %p")
    )

    current_time += timedelta(seconds=1)


# ============================================================
# WINDOWS COMPATIBILITY
#
# Windows Python may not support %-I.
# This converts 09:00:01 AM to 9:00:01 AM.
# ============================================================

def format_time(dt):

    formatted = dt.strftime("%I:%M:%S %p")

    # Remove leading zero from hour
    if formatted.startswith("0"):
        formatted = formatted[1:]

    return formatted


timestamp_list = []

current_time = START_TIME

while current_time <= END_TIME:

    timestamp_list.append(
        format_time(current_time)
    )

    current_time += timedelta(seconds=1)


# ============================================================
# DISPLAY TIMESTAMP INFORMATION
# ============================================================

print("=" * 70)
print("TIMESTAMP FORMAT")
print("=" * 70)

print("First timestamp :", timestamp_list[0])
print("Second timestamp:", timestamp_list[1])
print("Third timestamp :", timestamp_list[2])
print("Last timestamp  :", timestamp_list[-1])
print("Total timestamps:", len(timestamp_list))

print("=" * 70)


# ============================================================
# FIND CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_DIR.rglob("*.csv")
)

print()
print(
    f"Total CSV files found: {len(csv_files)}"
)
print()


# ============================================================
# PROCESS FILES
# ============================================================

processed = 0
skipped = 0
errors = 0


for file_path in csv_files:

    try:

        # ----------------------------------------------------
        # READ CSV
        # ----------------------------------------------------

        df = pd.read_csv(
            file_path,
            low_memory=False
        )


        # ----------------------------------------------------
        # CHECK TimeStamp COLUMN
        # ----------------------------------------------------

        if "TimeStamp" not in df.columns:

            # Copy file unchanged
            relative_path = file_path.relative_to(
                INPUT_DIR
            )

            output_file = (
                OUTPUT_DIR / relative_path
            )

            output_file.parent.mkdir(
                parents=True,
                exist_ok=True
            )

            shutil.copy2(
                file_path,
                output_file
            )

            skipped += 1

            print(
                f"SKIPPED: {relative_path} "
                f"| TimeStamp column not found"
            )

            continue


        # ----------------------------------------------------
        # NUMBER OF ROWS
        # ----------------------------------------------------

        number_of_rows = len(df)


        # ----------------------------------------------------
        # CREATE EMPTY TIMESTAMP COLUMN
        #
        # Extra rows remain blank.
        # ----------------------------------------------------

        new_timestamps = [""] * number_of_rows


        # ----------------------------------------------------
        # FILL TIMESTAMP VALUES
        # ----------------------------------------------------

        rows_to_fill = min(
            number_of_rows,
            len(timestamp_list)
        )


        for i in range(rows_to_fill):

            new_timestamps[i] = timestamp_list[i]


        # ----------------------------------------------------
        # REPLACE TimeStamp COLUMN
        # ----------------------------------------------------

        df["TimeStamp"] = new_timestamps


        # ----------------------------------------------------
        # CREATE OUTPUT PATH
        # ----------------------------------------------------

        relative_path = file_path.relative_to(
            INPUT_DIR
        )

        output_file = (
            OUTPUT_DIR / relative_path
        )

        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # ----------------------------------------------------
        # SAVE CSV
        # ----------------------------------------------------

        df.to_csv(
            output_file,
            index=False
        )


        processed += 1


        # ----------------------------------------------------
        # DISPLAY PROGRESS
        # ----------------------------------------------------

        if number_of_rows <= len(timestamp_list):

            print(
                f"PROCESSED: {relative_path} "
                f"| Rows: {number_of_rows} "
                f"| TimeStamp: "
                f"{timestamp_list[0]} -> "
                f"{timestamp_list[number_of_rows - 1]}"
            )

        else:

            extra_rows = (
                number_of_rows
                - len(timestamp_list)
            )

            print(
                f"PROCESSED: {relative_path} "
                f"| Rows: {number_of_rows} "
                f"| TimeStamp: "
                f"{timestamp_list[0]} -> "
                f"{timestamp_list[-1]} "
                f"| Extra rows: {extra_rows} "
                f"| Extra TimeStamp values blank"
            )


    except Exception as e:

        errors += 1

        print(
            f"ERROR: {file_path}"
        )

        print(
            f"       {e}"
        )


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 70)
print("PROCESSING COMPLETED")
print("=" * 70)

print(
    f"Input folder     : {INPUT_DIR}"
)

print(
    f"Output folder    : {OUTPUT_DIR}"
)

print(
    f"Total CSV files  : {len(csv_files)}"
)

print(
    f"Processed        : {processed}"
)

print(
    f"Skipped          : {skipped}"
)

print(
    f"Errors           : {errors}"
)

print(
    f"Timestamp format : 9:00:01 AM"
)

print(
    f"Timestamp range  : {timestamp_list[0]} "
    f"to {timestamp_list[-1]}"
)

print(
    f"Total timestamps : {len(timestamp_list)}"
)

print("=" * 70)
```

```python
import pandas as pd
from pathlib import Path
import re
import shutil


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path("/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V13")
OUTPUT_ROOT = Path("/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V14")


# File prefix mapping
# Muse file     Activity file
# clms1.csv     clmw1.csv
# mums1.csv     mumw1.csv
# norms1.csv    normw1.csv
# rdms1.csv     rdmw1.csv

MAPPING = {
    "clms": ("clmw", "class_lecture"),
    "mums": ("mumw", "music"),
    "norms": ("normw", "normal"),
    "rdms": ("rdmw", "reading"),
}


# ============================================================
# CHECK INPUT
# ============================================================

print("=" * 80)
print("MUSE AND ACTIVITY DATA MERGING")
print("=" * 80)

print()
print("Input folder:")
print(INPUT_ROOT)

print()
print("Output folder:")
print(OUTPUT_ROOT)

print()


if not INPUT_ROOT.exists():
    raise FileNotFoundError(
        f"Input folder does not exist:\n{INPUT_ROOT}"
    )


# ============================================================
# CREATE OUTPUT FOLDERS
# ============================================================

for output_folder in MAPPING.values():
    category_folder = OUTPUT_ROOT / output_folder[1]
    category_folder.mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# FIND ALL CSV FILES RECURSIVELY
# ============================================================

all_csv_files = list(INPUT_ROOT.rglob("*.csv"))

print("=" * 80)
print("INPUT FILE SEARCH")
print("=" * 80)

print()
print("Total CSV files found:", len(all_csv_files))
print()


if len(all_csv_files) == 0:
    raise FileNotFoundError(
        f"No CSV files found inside:\n{INPUT_ROOT}"
    )


# ============================================================
# CREATE FILE LOOKUP
# ============================================================

file_lookup = {}

for file_path in all_csv_files:

    filename = file_path.name.lower()

    file_lookup[filename] = file_path


# ============================================================
# SHOW AVAILABLE TARGET FILES
# ============================================================

print("Muse files detected:")
print()

for muse_prefix in MAPPING:

    detected = []

    for file_path in all_csv_files:

        name = file_path.stem.lower()

        if re.fullmatch(
            rf"{muse_prefix}\d+",
            name
        ):
            detected.append(file_path)

    print(
        f"{muse_prefix}: {len(detected)} files"
    )

print()


print("Activity files detected:")
print()

for muse_prefix, mapping_data in MAPPING.items():

    activity_prefix = mapping_data[0]

    detected = []

    for file_path in all_csv_files:

        name = file_path.stem.lower()

        if re.fullmatch(
            rf"{activity_prefix}\d+",
            name
        ):
            detected.append(file_path)

    print(
        f"{activity_prefix}: {len(detected)} files"
    )

print()


# ============================================================
# PROCESSING COUNTERS
# ============================================================

total_muse = 0
processed = 0
missing_activity = 0
missing_activity_column = 0
read_errors = 0
row_mismatch = 0


# ============================================================
# PROCESS EACH MUSE FILE
# ============================================================

for muse_prefix, mapping_data in MAPPING.items():

    activity_prefix = mapping_data[0]
    output_folder_name = mapping_data[1]

    print()
    print("=" * 80)
    print(f"PROCESSING {muse_prefix}")
    print("=" * 80)

    # Find Muse files
    muse_files = []

    for file_path in all_csv_files:

        name = file_path.stem.lower()

        if re.fullmatch(
            rf"{muse_prefix}\d+",
            name
        ):
            muse_files.append(file_path)


    # Sort according to person number
    muse_files.sort(
        key=lambda x: int(
            re.search(
                rf"{muse_prefix}(\d+)",
                x.stem.lower()
            ).group(1)
        )
    )


    print(
        "Muse files found:",
        len(muse_files)
    )


    # --------------------------------------------------------
    # Process Muse files
    # --------------------------------------------------------

    for muse_file in muse_files:

        total_muse += 1

        muse_name = muse_file.stem.lower()

        match = re.fullmatch(
            rf"{muse_prefix}(\d+)",
            muse_name
        )

        if match is None:
            continue

        person_number = match.group(1)

        expected_activity_name = (
            f"{activity_prefix}{person_number}.csv"
        )

        activity_file = file_lookup.get(
            expected_activity_name.lower()
        )


        # ----------------------------------------------------
        # Matching Activity file not found
        # ----------------------------------------------------

        if activity_file is None:

            missing_activity += 1

            print(
                f"MISSING ACTIVITY: "
                f"{muse_file.name} "
                f"requires "
                f"{expected_activity_name}"
            )

            continue


        try:

            # ------------------------------------------------
            # Read Muse file
            # ------------------------------------------------

            muse_df = pd.read_csv(
                muse_file,
                low_memory=False
            )


            # ------------------------------------------------
            # Read Activity file
            # ------------------------------------------------

            activity_df = pd.read_csv(
                activity_file,
                low_memory=False
            )


        except Exception as e:

            read_errors += 1

            print()
            print(
                f"READ ERROR: {muse_file.name}"
            )

            print(e)

            continue


        # ----------------------------------------------------
        # Find Activity column
        # ----------------------------------------------------

        activity_column = None

        for column in activity_df.columns:

            if str(column).strip().lower() == "activity":

                activity_column = column
                break


        if activity_column is None:

            missing_activity_column += 1

            print(
                f"NO ACTIVITY COLUMN: "
                f"{activity_file.name}"
            )

            continue


        # ----------------------------------------------------
        # Remove existing Activity column from Muse data
        # ----------------------------------------------------

        columns_to_remove = []

        for column in muse_df.columns:

            if str(column).strip().lower() == "activity":

                columns_to_remove.append(column)


        if columns_to_remove:

            muse_df.drop(
                columns=columns_to_remove,
                inplace=True
            )


        # ----------------------------------------------------
        # Get Activity values
        # ----------------------------------------------------

        activity_values = (
            activity_df[activity_column]
            .tolist()
        )


        muse_rows = len(muse_df)
        activity_rows = len(activity_values)


        # ----------------------------------------------------
        # Add empty Activity column
        # ----------------------------------------------------

        muse_df["Activity"] = pd.NA


        # ----------------------------------------------------
        # Insert Activity values row by row
        # ----------------------------------------------------

        rows_to_insert = min(
            muse_rows,
            activity_rows
        )


        if rows_to_insert > 0:

            muse_df.loc[
                muse_df.index[:rows_to_insert],
                "Activity"
            ] = activity_values[:rows_to_insert]


        # ----------------------------------------------------
        # Check row mismatch
        # ----------------------------------------------------

        if muse_rows != activity_rows:

            row_mismatch += 1


        # ----------------------------------------------------
        # Create output file
        # ----------------------------------------------------

        output_folder = (
            OUTPUT_ROOT /
            output_folder_name
        )

        output_folder.mkdir(
            parents=True,
            exist_ok=True
        )


        output_file = (
            output_folder /
            muse_file.name
        )


        # ----------------------------------------------------
        # Save
        # ----------------------------------------------------

        muse_df.to_csv(
            output_file,
            index=False
        )


        processed += 1


        # ----------------------------------------------------
        # Status information
        # ----------------------------------------------------

        blank_rows = max(
            0,
            muse_rows - activity_rows
        )

        extra_activity_rows = max(
            0,
            activity_rows - muse_rows
        )


        print(
            f"PROCESSED: {muse_file.name} | "
            f"Activity: {activity_file.name} | "
            f"Muse rows: {muse_rows} | "
            f"Activity rows: {activity_rows} | "
            f"Inserted: {rows_to_insert} | "
            f"Blank Activity rows: {blank_rows} | "
            f"Extra Activity rows ignored: {extra_activity_rows}"
        )


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print()
print("=" * 80)
print("MERGING COMPLETED")
print("=" * 80)

print()
print("Input folder:")
print(INPUT_ROOT)

print()
print("Output folder:")
print(OUTPUT_ROOT)

print()
print("Total Muse files found:", total_muse)
print("Successfully processed:", processed)
print("Missing Activity files:", missing_activity)
print("Missing Activity columns:", missing_activity_column)
print("Read errors:", read_errors)
print("Files with different row counts:", row_mismatch)

print()
print("Output folders:")

for folder_name in [
    "class_lecture",
    "music",
    "normal",
    "reading"
]:

    folder = OUTPUT_ROOT / folder_name

    count = len(
        list(folder.glob("*.csv"))
    )

    print(
        f"{folder_name}: {count} files"
    )


print()
print("=" * 80)
print("OUTPUT LOCATION")
print("=" * 80)
print(OUTPUT_ROOT)
print("=" * 80)
```

```python
import pandas as pd
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V14"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V15"
)

TIMESTAMP_COLUMN = "TimeStamp"


# ============================================================
# CREATE COMPLETE TIMESTAMP SEQUENCE
# ============================================================

start_time = pd.Timestamp("2026-01-01 09:00:01")
end_time = pd.Timestamp("2026-01-01 10:15:01")

timestamps = pd.date_range(
    start=start_time,
    end=end_time,
    freq="1s"
)

timestamp_strings = timestamps.strftime(
    "%I:%M:%S %p"
).str.lstrip("0")


print("=" * 80)
print("TIMESTAMP EXTENSION")
print("=" * 80)

print()
print("Start timestamp :", timestamp_strings[0])
print("End timestamp   :", timestamp_strings[-1])
print("Total timestamps:", len(timestamp_strings))
print()


# ============================================================
# FIND ALL CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

print(
    "CSV files found:",
    len(csv_files)
)

print()


if len(csv_files) == 0:

    raise FileNotFoundError(
        "No CSV files found. Please check INPUT_ROOT."
    )


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# PROCESS EACH CSV
# ============================================================

processed = 0
failed = []

print("=" * 80)
print("PROCESSING FILES")
print("=" * 80)


for i, input_file in enumerate(
    csv_files,
    start=1
):

    try:

        # ----------------------------------------------------
        # READ CSV
        # ----------------------------------------------------

        df = pd.read_csv(
            input_file,
            low_memory=False
        )


        original_rows = len(df)


        # ----------------------------------------------------
        # REMOVE EXISTING TIMESTAMP COLUMN
        # ----------------------------------------------------

        timestamp_columns = [
            col for col in df.columns
            if str(col).strip().lower() == "timestamp"
        ]

        if timestamp_columns:

            df = df.drop(
                columns=timestamp_columns
            )


        # ----------------------------------------------------
        # MAKE DATAFRAME EXACTLY 4501 ROWS
        # ----------------------------------------------------

        required_rows = len(timestamp_strings)


        if original_rows < required_rows:

            # Add empty rows

            additional_rows = required_rows - original_rows

            empty_rows = pd.DataFrame(
                index=range(additional_rows),
                columns=df.columns
            )

            df = pd.concat(
                [df, empty_rows],
                ignore_index=True
            )


        elif original_rows > required_rows:

            # Keep all existing rows.
            # Timestamp will be blank after 10:15:01 AM.

            pass


        # ----------------------------------------------------
        # CREATE TIMESTAMP COLUMN
        # ----------------------------------------------------

        final_rows = len(df)


        timestamp_values = list(
            timestamp_strings
        )


        if final_rows > required_rows:

            timestamp_values.extend(
                [pd.NA] *
                (final_rows - required_rows)
            )


        # ----------------------------------------------------
        # INSERT TIMESTAMP AS FIRST COLUMN
        # ----------------------------------------------------

        df.insert(
            0,
            TIMESTAMP_COLUMN,
            timestamp_values
        )


        # ----------------------------------------------------
        # CREATE SAME FOLDER STRUCTURE
        # ----------------------------------------------------

        relative_path = input_file.relative_to(
            INPUT_ROOT
        )

        output_file = (
            OUTPUT_ROOT /
            relative_path
        )

        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # ----------------------------------------------------
        # SAVE
        # ----------------------------------------------------

        df.to_csv(
            output_file,
            index=False
        )


        processed += 1


        print(
            f"[{i:03d}/{len(csv_files)}] "
            f"{input_file.name} | "
            f"Original rows: {original_rows:,} | "
            f"Final rows: {len(df):,}"
        )


    except Exception as e:

        failed.append(
            (
                str(input_file),
                str(e)
            )
        )

        print(
            f"[{i:03d}/{len(csv_files)}] "
            f"FAILED: {input_file.name}"
        )

        print(
            "Error:",
            e
        )


# ============================================================
# FINAL VERIFICATION
# ============================================================

print()
print("=" * 80)
print("FINAL VERIFICATION")
print("=" * 80)

output_files = sorted(
    OUTPUT_ROOT.rglob("*.csv")
)


print()
print(
    "Input files       :",
    len(csv_files)
)

print(
    "Processed files   :",
    processed
)

print(
    "Failed files      :",
    len(failed)
)

print(
    "Output files      :",
    len(output_files)
)

print(
    "Required timestamps:",
    len(timestamp_strings)
)


# ============================================================
# VERIFY OUTPUT FILES
# ============================================================

verification_failed = []


for output_file in output_files:

    try:

        check = pd.read_csv(
            output_file,
            low_memory=False
        )


        # Check row count

        if len(check) < required_rows:

            verification_failed.append(
                (
                    output_file.name,
                    f"Only {len(check)} rows"
                )
            )

            continue


        # Check TimeStamp column

        if TIMESTAMP_COLUMN not in check.columns:

            verification_failed.append(
                (
                    output_file.name,
                    "TimeStamp column missing"
                )
            )

            continue


        # Check first timestamp

        first_value = str(
            check[TIMESTAMP_COLUMN].iloc[0]
        )

        if first_value != timestamp_strings[0]:

            verification_failed.append(
                (
                    output_file.name,
                    f"First timestamp is {first_value}"
                )
            )

            continue


        # Check final required timestamp

        last_required_value = str(
            check[TIMESTAMP_COLUMN]
            .iloc[required_rows - 1]
        )


        if (
            last_required_value
            != timestamp_strings[-1]
        ):

            verification_failed.append(
                (
                    output_file.name,
                    f"Final timestamp is "
                    f"{last_required_value}"
                )
            )


    except Exception as e:

        verification_failed.append(
            (
                output_file.name,
                str(e)
            )
        )


# ============================================================
# REPORT
# ============================================================

print()

if len(failed) == 0:

    print(
        "All files processed without processing errors."
    )

else:

    print(
        "Processing errors:",
        len(failed)
    )

    for file, error in failed:

        print(
            file,
            "->",
            error
        )


print()

if len(verification_failed) == 0:

    print(
        "VERIFICATION PASSED"
    )

    print(
        "Every output file contains the complete timestamp "
        "sequence from 9:00:01 AM to 10:15:01 AM."
    )

else:

    print(
        "VERIFICATION FAILED"
    )

    print(
        "Files with verification problems:",
        len(verification_failed)
    )

    for file, reason in verification_failed:

        print(
            file,
            "->",
            reason
        )


print()
print(
    "Output folder:"
)

print(
    OUTPUT_ROOT
)

print()
print("=" * 80)
print("COMPLETED")
print("=" * 80)
```

```python
import pandas as pd
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V15"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V16"
)

TIMESTAMP_COLUMN = "TimeStamp"

START_TIME = "09:00:01 AM"
END_TIME = "10:15:01 AM"


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# FIND ALL CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

print("=" * 80)
print("V15 → V16 TIMESTAMP BASED DATA CLEANING")
print("=" * 80)

print()
print("Input folder :", INPUT_ROOT)
print("Output folder:", OUTPUT_ROOT)
print()
print("CSV files found:", len(csv_files))
print()


if len(csv_files) == 0:

    raise FileNotFoundError(
        f"No CSV files found in:\n{INPUT_ROOT}"
    )


# ============================================================
# CONVERT REQUIRED TIME LIMITS
# ============================================================

start_time = pd.to_datetime(
    START_TIME,
    format="%I:%M:%S %p"
).time()

end_time = pd.to_datetime(
    END_TIME,
    format="%I:%M:%S %p"
).time()


# ============================================================
# COUNTERS
# ============================================================

processed = 0
failed = []

total_original_rows = 0
total_final_rows = 0
total_removed_rows = 0


# ============================================================
# PROCESS EACH FILE
# ============================================================

for number, input_file in enumerate(
    csv_files,
    start=1
):

    try:

        # ----------------------------------------------------
        # READ CSV
        # ----------------------------------------------------

        df = pd.read_csv(
            input_file,
            low_memory=False
        )

        original_rows = len(df)

        total_original_rows += original_rows


        # ----------------------------------------------------
        # FIND TIMESTAMP COLUMN
        # ----------------------------------------------------

        timestamp_columns = [
            col
            for col in df.columns
            if str(col).strip().lower() == "timestamp"
        ]


        if not timestamp_columns:

            raise ValueError(
                "TimeStamp column not found"
            )


        timestamp_column = timestamp_columns[0]


        # ----------------------------------------------------
        # CONVERT TIMESTAMP
        # ----------------------------------------------------

        parsed_time = pd.to_datetime(
            df[timestamp_column],
            format="%I:%M:%S %p",
            errors="coerce"
        )


        # ----------------------------------------------------
        # KEEP ONLY VALID TIMESTAMP ROWS
        #
        # Required range:
        #
        # 9:00:01 AM
        #
        # through
        #
        # 10:15:01 AM
        # ----------------------------------------------------

        valid_timestamp_mask = (
            parsed_time.notna()
            &
            (
                parsed_time.dt.time >= start_time
            )
            &
            (
                parsed_time.dt.time <= end_time
            )
        )


        cleaned_df = df.loc[
            valid_timestamp_mask
        ].copy()


        # ----------------------------------------------------
        # REMOVE ANY ROW AFTER THE LAST VALID TIMESTAMP
        #
        # The filtering above already removes invalid and
        # out of range timestamp rows.
        # ----------------------------------------------------


        final_rows = len(cleaned_df)

        removed_rows = (
            original_rows - final_rows
        )


        total_final_rows += final_rows
        total_removed_rows += removed_rows


        # ----------------------------------------------------
        # RESET INDEX
        # ----------------------------------------------------

        cleaned_df.reset_index(
            drop=True,
            inplace=True
        )


        # ----------------------------------------------------
        # PRESERVE ORIGINAL FOLDER STRUCTURE
        # ----------------------------------------------------

        relative_path = input_file.relative_to(
            INPUT_ROOT
        )

        output_file = (
            OUTPUT_ROOT /
            relative_path
        )

        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # ----------------------------------------------------
        # SAVE
        # ----------------------------------------------------

        cleaned_df.to_csv(
            output_file,
            index=False
        )


        # ----------------------------------------------------
        # VERIFY
        # ----------------------------------------------------

        if len(cleaned_df) > 0:

            first_timestamp = (
                cleaned_df[timestamp_column].iloc[0]
            )

            last_timestamp = (
                cleaned_df[timestamp_column].iloc[-1]
            )

        else:

            first_timestamp = "EMPTY"
            last_timestamp = "EMPTY"


        processed += 1


        print(
            f"[{number:03d}/{len(csv_files)}] "
            f"{relative_path} | "
            f"Original: {original_rows:,} | "
            f"Final: {final_rows:,} | "
            f"Removed: {removed_rows:,} | "
            f"Time: {first_timestamp} -> {last_timestamp}"
        )


    except Exception as error:

        failed.append(
            {
                "file": str(
                    input_file.relative_to(
                        INPUT_ROOT
                    )
                ),
                "error": str(error)
            }
        )

        print(
            f"[{number:03d}/{len(csv_files)}] "
            f"FAILED: {input_file.name} | {error}"
        )


# ============================================================
# FINAL OUTPUT FILE COUNT
# ============================================================

output_files = sorted(
    OUTPUT_ROOT.rglob("*.csv")
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 80)
print("V16 PROCESSING COMPLETED")
print("=" * 80)

print()
print(
    "Input CSV files       :",
    len(csv_files)
)

print(
    "Successfully processed:",
    processed
)

print(
    "Failed files          :",
    len(failed)
)

print(
    "Output CSV files      :",
    len(output_files)
)

print()
print(
    "Original total rows   :",
    f"{total_original_rows:,}"
)

print(
    "Final total rows      :",
    f"{total_final_rows:,}"
)

print(
    "Removed total rows    :",
    f"{total_removed_rows:,}"
)

print()
print(
    "Required timestamp range:"
)

print(
    f"{START_TIME} → {END_TIME}"
)


# ============================================================
# FAILED FILE REPORT
# ============================================================

if failed:

    print()
    print("=" * 80)
    print("FAILED FILES")
    print("=" * 80)

    for item in failed:

        print()
        print("File :", item["file"])
        print("Error:", item["error"])


# ============================================================
# FINAL MESSAGE
# ============================================================

print()
print("=" * 80)

if len(failed) == 0:

    print("SUCCESS")
    print()
    print(
        "All CSV files were processed."
    )

    print(
        "Rows outside the timestamp range were removed."
    )

    print(
        "The complete row, including all other column values, "
        "was removed when its timestamp was outside the range."
    )

    print()
    print("Output folder:")
    print(OUTPUT_ROOT)

else:

    print("PROCESSING COMPLETED WITH ERRORS")

print("=" * 80)
```

```python
import pandas as pd
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V16"
)

OUTPUT_FILE = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V16/"
    "V16_row_count_report.csv"
)

EMPTY_TIMESTAMP_REPORT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V16/"
    "V16_timestamp_without_data_report.csv"
)

ACTIVITY_SUMMARY_FILE = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V16/"
    "V16_activity_row_summary.csv"
)


# ============================================================
# FIND CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

print("=" * 90)
print("V16 DATASET ROW AND TIMESTAMP DATA AUDIT")
print("=" * 90)

print()
print("Input folder:")
print(INPUT_ROOT)

print()
print("CSV files found:", len(csv_files))
print()


# ============================================================
# EXCLUDE GENERATED REPORT CSV FILES
# ============================================================

csv_files = [
    file for file in csv_files
    if file.name not in [
        OUTPUT_FILE.name,
        EMPTY_TIMESTAMP_REPORT.name,
        ACTIVITY_SUMMARY_FILE.name
    ]
]


print(
    "Actual dataset CSV files:",
    len(csv_files)
)

print()


# ============================================================
# CHECK FILE COUNT
# ============================================================

if len(csv_files) != 240:

    print(
        "WARNING:"
    )

    print(
        f"Expected 240 dataset CSV files, "
        f"but found {len(csv_files)}."
    )

    print()


# ============================================================
# STORAGE
# ============================================================

row_results = []

timestamp_data_results = []


# ============================================================
# PROCESS EACH CSV
# ============================================================

for number, file_path in enumerate(
    csv_files,
    start=1
):

    try:

        # ----------------------------------------------------
        # READ CSV
        # ----------------------------------------------------

        df = pd.read_csv(
            file_path,
            low_memory=False
        )


        original_row_count = len(df)


        # ----------------------------------------------------
        # RELATIVE PATH
        # ----------------------------------------------------

        relative_path = file_path.relative_to(
            INPUT_ROOT
        )


        # ----------------------------------------------------
        # ACTIVITY FOLDER
        # ----------------------------------------------------

        parts = relative_path.parts

        if len(parts) > 0:
            activity = parts[0]
        else:
            activity = ""


        # ====================================================
        # ROW COUNT
        # ====================================================

        row_results.append({
            "Serial": number,
            "File_Name": file_path.name,
            "Folder": activity,
            "Relative_Path": str(relative_path),
            "Row_Count": original_row_count
        })


        # ====================================================
        # FIND TIMESTAMP COLUMN
        # ====================================================

        timestamp_columns = [
            column
            for column in df.columns
            if str(column).strip().lower() == "timestamp"
        ]


        if len(timestamp_columns) == 0:

            timestamp_data_results.append({
                "Serial": number,
                "File_Name": file_path.name,
                "Folder": activity,
                "Relative_Path": str(relative_path),
                "Timestamp_Column": "NOT FOUND",
                "Rows_With_Timestamp": 0,
                "Rows_With_Timestamp_And_No_Other_Data": 0,
                "First_Empty_Data_Timestamp": "",
                "Last_Empty_Data_Timestamp": "",
                "Status": "NO TIMESTAMP COLUMN"
            })

            print(
                f"[{number:03d}/{len(csv_files)}] "
                f"{relative_path} | "
                f"Rows: {original_row_count:,} | "
                f"TimeStamp: NOT FOUND"
            )

            continue


        timestamp_column = timestamp_columns[0]


        # ====================================================
        # IDENTIFY OTHER DATA COLUMNS
        # ====================================================

        other_columns = [
            column
            for column in df.columns
            if column != timestamp_column
        ]


        # ====================================================
        # TIMESTAMP HAS VALUE
        # ====================================================

        timestamp_present = (
            df[timestamp_column]
            .notna()
            &
            (
                df[timestamp_column]
                .astype(str)
                .str.strip()
                .ne("")
            )
        )


        # ====================================================
        # CHECK OTHER COLUMNS
        #
        # A row is considered "no other data" when every
        # column except TimeStamp is empty or NaN.
        # ====================================================

        if len(other_columns) > 0:

            other_data_present = (
                df[other_columns]
                .notna()
                .any(axis=1)
            )

        else:

            other_data_present = pd.Series(
                False,
                index=df.index
            )


        # ====================================================
        # TIMESTAMP EXISTS BUT OTHER DATA IS EMPTY
        # ====================================================

        timestamp_without_data = (
            timestamp_present
            &
            (~other_data_present)
        )


        number_timestamp_rows = int(
            timestamp_present.sum()
        )


        number_timestamp_without_data = int(
            timestamp_without_data.sum()
        )


        # ====================================================
        # GET FIRST AND LAST TIMESTAMP
        # WHERE OTHER DATA IS EMPTY
        # ====================================================

        empty_data_timestamps = (
            df.loc[
                timestamp_without_data,
                timestamp_column
            ]
            .astype(str)
            .tolist()
        )


        if len(empty_data_timestamps) > 0:

            first_empty_timestamp = (
                empty_data_timestamps[0]
            )

            last_empty_timestamp = (
                empty_data_timestamps[-1]
            )

        else:

            first_empty_timestamp = ""
            last_empty_timestamp = ""


        # ====================================================
        # STATUS
        # ====================================================

        if number_timestamp_without_data > 0:

            status = (
                "TIMESTAMP EXISTS BUT OTHER DATA EMPTY"
            )

        else:

            status = (
                "TIMESTAMP AND OTHER DATA PRESENT"
            )


        # ====================================================
        # SAVE RESULT
        # ====================================================

        timestamp_data_results.append({
            "Serial": number,
            "File_Name": file_path.name,
            "Folder": activity,
            "Relative_Path": str(relative_path),
            "Timestamp_Column": timestamp_column,
            "Rows_Total": original_row_count,
            "Rows_With_Timestamp": number_timestamp_rows,
            "Rows_With_Timestamp_And_No_Other_Data":
                number_timestamp_without_data,
            "First_Empty_Data_Timestamp":
                first_empty_timestamp,
            "Last_Empty_Data_Timestamp":
                last_empty_timestamp,
            "Status": status
        })


        # ====================================================
        # DISPLAY
        # ====================================================

        print(
            f"[{number:03d}/{len(csv_files)}] "
            f"{relative_path} | "
            f"Rows: {original_row_count:,} | "
            f"Timestamp rows: {number_timestamp_rows:,} | "
            f"Timestamp only: "
            f"{number_timestamp_without_data:,}"
        )


    except Exception as e:

        print()
        print(
            f"[{number:03d}/{len(csv_files)}] "
            f"ERROR: {file_path.name}"
        )

        print(
            str(e)
        )

        timestamp_data_results.append({
            "Serial": number,
            "File_Name": file_path.name,
            "Folder": "",
            "Relative_Path": str(
                file_path.relative_to(INPUT_ROOT)
            ),
            "Timestamp_Column": "",
            "Rows_Total": "",
            "Rows_With_Timestamp": "",
            "Rows_With_Timestamp_And_No_Other_Data": "",
            "First_Empty_Data_Timestamp": "",
            "Last_Empty_Data_Timestamp": "",
            "Status": f"ERROR: {e}"
        })


# ============================================================
# CREATE ROW COUNT REPORT
# ============================================================

row_count_df = pd.DataFrame(
    row_results
)


row_count_df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ============================================================
# CREATE TIMESTAMP WITHOUT DATA REPORT
# ============================================================

timestamp_report_df = pd.DataFrame(
    timestamp_data_results
)


timestamp_report_df.to_csv(
    EMPTY_TIMESTAMP_REPORT,
    index=False
)


# ============================================================
# ACTIVITY SUMMARY
# ============================================================

if len(row_count_df) > 0:

    activity_summary = (
        row_count_df
        .groupby("Folder")["Row_Count"]
        .agg(
            File_Count="count",
            Total_Rows="sum",
            Minimum_Rows="min",
            Maximum_Rows="max",
            Average_Rows="mean",
            Median_Rows="median"
        )
        .reset_index()
    )

else:

    activity_summary = pd.DataFrame()


if not activity_summary.empty:

    activity_summary.to_csv(
        ACTIVITY_SUMMARY_FILE,
        index=False
    )


# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 90)
print("FINAL AUDIT SUMMARY")
print("=" * 90)

print()

print(
    "Dataset CSV files:",
    len(csv_files)
)

print(
    "Files successfully analyzed:",
    len(row_count_df)
)


if len(row_count_df) > 0:

    print(
        "Total dataset rows:",
        f"{row_count_df['Row_Count'].sum():,}"
    )

    print(
        "Minimum rows:",
        f"{row_count_df['Row_Count'].min():,}"
    )

    print(
        "Maximum rows:",
        f"{row_count_df['Row_Count'].max():,}"
    )

    print(
        "Median rows:",
        f"{row_count_df['Row_Count'].median():,.0f}"
    )


# ============================================================
# TIMESTAMP ANALYSIS SUMMARY
# ============================================================

print()
print("=" * 90)
print("TIMESTAMP WITH NO OTHER DATA")
print("=" * 90)

if len(timestamp_report_df) > 0:

    timestamp_only_total = pd.to_numeric(
        timestamp_report_df[
            "Rows_With_Timestamp_And_No_Other_Data"
        ],
        errors="coerce"
    ).fillna(0).sum()


    files_with_timestamp_only = (
        pd.to_numeric(
            timestamp_report_df[
                "Rows_With_Timestamp_And_No_Other_Data"
            ],
            errors="coerce"
        )
        .fillna(0)
        .gt(0)
        .sum()
    )


    print(
        "Total rows with TimeStamp but no other data:",
        f"{int(timestamp_only_total):,}"
    )

    print(
        "Files containing such rows:",
        int(files_with_timestamp_only)
    )


# ============================================================
# SHOW FILES CONTAINING TIMESTAMP ONLY
# ============================================================

if len(timestamp_report_df) > 0:

    problematic_files = timestamp_report_df[
        pd.to_numeric(
            timestamp_report_df[
                "Rows_With_Timestamp_And_No_Other_Data"
            ],
            errors="coerce"
        )
        .fillna(0)
        .gt(0)
    ]


    if len(problematic_files) > 0:

        print()
        print(
            "Files containing TimeStamp values "
            "without other column data:"
        )

        print()

        display_columns = [
            "File_Name",
            "Folder",
            "Rows_Total",
            "Rows_With_Timestamp_And_No_Other_Data",
            "First_Empty_Data_Timestamp",
            "Last_Empty_Data_Timestamp"
        ]

        print(
            problematic_files[
                display_columns
            ].to_string(
                index=False
            )
        )

    else:

        print()
        print(
            "No rows were found where TimeStamp exists "
            "but all other columns are empty."
        )


# ============================================================
# OUTPUT LOCATIONS
# ============================================================

print()
print("=" * 90)
print("REPORT FILES")
print("=" * 90)

print()
print(
    "Row count report:"
)

print(
    OUTPUT_FILE
)

print()
print(
    "Timestamp without other data report:"
)

print(
    EMPTY_TIMESTAMP_REPORT
)

print()
print(
    "Activity summary:"
)

print(
    ACTIVITY_SUMMARY_FILE
)

print()
print("=" * 90)
print("DONE")
print("=" * 90)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V16"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V17"
)

WINDOW_SIZE = 21

TIMESTAMP_COLUMN = "TimeStamp"

ACTIVITY_COLUMN = "Activity"


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# FIND DATASET FILES
# ============================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

# Remove previously generated report files
csv_files = [
    f for f in csv_files
    if not f.name.startswith("V16_")
]


print("=" * 90)
print("V16 → V17 SLIDING WINDOW MODE IMPUTATION")
print("=" * 90)

print()
print("Input folder:")
print(INPUT_ROOT)

print()
print("Output folder:")
print(OUTPUT_ROOT)

print()
print("CSV files found:", len(csv_files))

print()
print("Sliding window size:", WINDOW_SIZE)
print("Method: Previous 21 cells → Mode → Fill missing cell")
print("Window movement: 1 row")
print()


# ============================================================
# CHECK FILE COUNT
# ============================================================

if len(csv_files) != 240:

    print(
        f"WARNING: Expected 240 CSV files, "
        f"but found {len(csv_files)}."
    )

print()


# ============================================================
# MODE CALCULATION FUNCTION
# ============================================================

def calculate_mode(values):
    """
    Calculate the statistical mode of the supplied
    numerical values.

    NaN values are removed before calculation.
    """

    values = pd.to_numeric(
        values,
        errors="coerce"
    )

    values = values.dropna()


    if len(values) == 0:

        return np.nan


    modes = values.mode()


    if len(modes) == 0:

        return np.nan


    return modes.iloc[0]


# ============================================================
# PROCESSING COUNTERS
# ============================================================

processed_files = 0
failed_files = []

total_missing_before = 0
total_imputed = 0
total_remaining_missing = 0


file_report = []


# ============================================================
# PROCESS EVERY CSV
# ============================================================

for file_number, input_file in enumerate(
    csv_files,
    start=1
):

    try:

        # ----------------------------------------------------
        # READ FILE
        # ----------------------------------------------------

        df = pd.read_csv(
            input_file,
            low_memory=False
        )


        original_rows = len(df)


        # ----------------------------------------------------
        # DETERMINE DATA COLUMNS
        # ----------------------------------------------------

        excluded_columns = []


        for column in df.columns:

            column_name = str(
                column
            ).strip().lower()


            if column_name in [
                "timestamp",
                "activity"
            ]:

                excluded_columns.append(
                    column
                )


        data_columns = [
            column
            for column in df.columns
            if column not in excluded_columns
        ]


        file_missing_before = 0
        file_imputed = 0


        # ====================================================
        # PROCESS EACH DATA COLUMN
        # ====================================================

        for column in data_columns:

            # ------------------------------------------------
            # Convert column to numeric
            # ------------------------------------------------

            series = pd.to_numeric(
                df[column],
                errors="coerce"
            )


            # ------------------------------------------------
            # Count missing values before imputation
            # ------------------------------------------------

            missing_before = int(
                series.isna().sum()
            )

            file_missing_before += (
                missing_before
            )


            # ------------------------------------------------
            # SEQUENTIAL SLIDING WINDOW
            # ------------------------------------------------

            for current_position in range(
                len(series)
            ):

                # --------------------------------------------
                # Check whether current cell is missing
                # --------------------------------------------

                if not pd.isna(
                    series.iloc[current_position]
                ):

                    continue


                # --------------------------------------------
                # Previous 21 cells
                # --------------------------------------------

                window_start = max(
                    0,
                    current_position - WINDOW_SIZE
                )


                window_end = current_position


                previous_21 = (
                    series.iloc[
                        window_start:window_end
                    ]
                )


                # --------------------------------------------
                # Remove missing values from window
                # --------------------------------------------

                previous_21 = pd.to_numeric(
                    previous_21,
                    errors="coerce"
                ).dropna()


                # --------------------------------------------
                # If no previous value exists
                # --------------------------------------------

                if len(previous_21) == 0:

                    continue


                # --------------------------------------------
                # Calculate MODE
                # --------------------------------------------

                mode_value = calculate_mode(
                    previous_21
                )


                # --------------------------------------------
                # Insert MODE into current cell
                # --------------------------------------------

                if not pd.isna(
                    mode_value
                ):

                    series.iloc[
                        current_position
                    ] = mode_value

                    file_imputed += 1


            # ------------------------------------------------
            # Put processed column back
            # ------------------------------------------------

            df[column] = series


        # ====================================================
        # COUNT REMAINING MISSING VALUES
        # ====================================================

        remaining_missing = 0


        for column in data_columns:

            remaining_missing += int(
                pd.to_numeric(
                    df[column],
                    errors="coerce"
                )
                .isna()
                .sum()
            )


        # ====================================================
        # SAVE OUTPUT
        # ====================================================

        relative_path = input_file.relative_to(
            INPUT_ROOT
        )


        output_file = (
            OUTPUT_ROOT /
            relative_path
        )


        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        df.to_csv(
            output_file,
            index=False
        )


        # ====================================================
        # UPDATE GLOBAL COUNTERS
        # ====================================================

        total_missing_before += (
            file_missing_before
        )

        total_imputed += (
            file_imputed
        )

        total_remaining_missing += (
            remaining_missing
        )


        processed_files += 1


        # ====================================================
        # FILE REPORT
        # ====================================================

        file_report.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(relative_path),

            "Rows":
                original_rows,

            "Data_Columns":
                len(data_columns),

            "Missing_Before":
                file_missing_before,

            "Values_Imputed":
                file_imputed,

            "Remaining_Missing":
                remaining_missing

        })


        # ====================================================
        # PROGRESS
        # ====================================================

        print(
            f"[{file_number:03d}/{len(csv_files)}] "
            f"{relative_path} | "
            f"Rows: {original_rows:,} | "
            f"Missing: {file_missing_before:,} | "
            f"Imputed: {file_imputed:,} | "
            f"Remaining: {remaining_missing:,}"
        )


    except Exception as error:

        failed_files.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(
                    input_file.relative_to(
                        INPUT_ROOT
                    )
                ),

            "Error":
                str(error)

        })


        print()
        print(
            f"ERROR: {input_file.name}"
        )

        print(
            str(error)
        )


# ============================================================
# SAVE FILE REPORT
# ============================================================

report_df = pd.DataFrame(
    file_report
)


REPORT_FILE = (
    OUTPUT_ROOT /
    "V17_sliding_window_mode_report.csv"
)


report_df.to_csv(
    REPORT_FILE,
    index=False
)


# ============================================================
# OUTPUT FILE COUNT
# ============================================================

output_files = [
    f
    for f in OUTPUT_ROOT.rglob("*.csv")
    if not f.name.startswith("V17_")
]


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 90)
print("V17 PROCESSING COMPLETED")
print("=" * 90)

print()

print(
    "Input CSV files:",
    len(csv_files)
)

print(
    "Successfully processed:",
    processed_files
)

print(
    "Failed files:",
    len(failed_files)
)

print(
    "Output CSV files:",
    len(output_files)
)

print()

print(
    "Sliding window:",
    WINDOW_SIZE
)

print(
    "Total missing values before:",
    f"{total_missing_before:,}"
)

print(
    "Total values imputed:",
    f"{total_imputed:,}"
)

print(
    "Total remaining missing:",
    f"{total_remaining_missing:,}"
)


# ============================================================
# IMPUTATION COVERAGE
# ============================================================

if total_missing_before > 0:

    coverage = (
        total_imputed /
        total_missing_before
    ) * 100

    print(
        f"Imputation coverage: {coverage:.2f}%"
    )


# ============================================================
# FAILED FILES
# ============================================================

if failed_files:

    print()
    print("=" * 90)
    print("FAILED FILES")
    print("=" * 90)

    for item in failed_files:

        print()
        print(
            "File:",
            item["Relative_Path"]
        )

        print(
            "Error:",
            item["Error"]
        )


# ============================================================
# FINAL OUTPUT
# ============================================================

print()
print("=" * 90)

if (
    processed_files == len(csv_files)
    and
    len(failed_files) == 0
    and
    len(output_files) == len(csv_files)
):

    print("SUCCESS")

    print()
    print(
        "All CSV files were processed successfully."
    )

    print(
        "Each missing numerical cell was evaluated "
        "using the previous 21 cells."
    )

    print(
        "The statistical mode was inserted into "
        "the missing cell when a valid previous value existed."
    )

    print(
        "The window moves forward one row at a time."
    )

else:

    print(
        "PROCESSING COMPLETED WITH ERRORS"
    )


print()
print(
    "V17 dataset:"
)

print(
    OUTPUT_ROOT
)

print()
print(
    "Processing report:"
)

print(
    REPORT_FILE
)

print()
print("=" * 90)
print("DONE")
print("=" * 90)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V17"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V18"
)

EXPECTED_FILES = 240

EXPECTED_COMMON_COLUMNS = 34

ACTIVITY_COLUMN_NAME = "Activity"

WINDOW_SIZE = 21


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# FIND DATASET CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

# Exclude previously generated reports
csv_files = [
    file
    for file in csv_files
    if not file.name.startswith("V17_")
    and not file.name.startswith("V18_")
]


# ============================================================
# INITIAL INFORMATION
# ============================================================

print("=" * 100)
print("CLEAN_DATA_V17 → CLEAN_DATA_V18")
print("=" * 100)

print()
print("Input folder:")
print(INPUT_ROOT)

print()
print("Output folder:")
print(OUTPUT_ROOT)

print()
print("CSV files found:", len(csv_files))

print()


# ============================================================
# FILE COUNT VALIDATION
# ============================================================

if len(csv_files) != EXPECTED_FILES:

    print("=" * 100)
    print("WARNING")
    print("=" * 100)

    print(
        f"Expected {EXPECTED_FILES} CSV files, "
        f"but found {len(csv_files)}."
    )

    print()


# ============================================================
# STEP 1
# READ HEADER FROM EVERY FILE
# ============================================================

print("=" * 100)
print("STEP 1: READING COLUMN STRUCTURE FROM ALL FILES")
print("=" * 100)

print()

column_sets = []

file_headers = []

header_errors = []


for file_number, file_path in enumerate(
    csv_files,
    start=1
):

    try:

        header_df = pd.read_csv(
            file_path,
            nrows=0,
            low_memory=False
        )

        columns = list(
            header_df.columns
        )

        column_sets.append(
            set(columns)
        )

        file_headers.append({
            "file": file_path,
            "columns": columns
        })


        print(
            f"[{file_number:03d}/{len(csv_files)}] "
            f"{file_path.name} | "
            f"Columns: {len(columns)}"
        )


    except Exception as error:

        header_errors.append({

            "File_Name":
                file_path.name,

            "Relative_Path":
                str(
                    file_path.relative_to(
                        INPUT_ROOT
                    )
                ),

            "Error":
                str(error)

        })


        print(
            f"[{file_number:03d}/{len(csv_files)}] "
            f"ERROR: {file_path.name}"
        )

        print(
            error
        )


# ============================================================
# STOP IF HEADER ERROR EXISTS
# ============================================================

if header_errors:

    HEADER_ERROR_REPORT = (
        OUTPUT_ROOT /
        "V18_header_errors.csv"
    )

    pd.DataFrame(
        header_errors
    ).to_csv(
        HEADER_ERROR_REPORT,
        index=False
    )

    raise RuntimeError(
        "Some CSV headers could not be read. "
        "Check V18_header_errors.csv."
    )


if len(column_sets) != EXPECTED_FILES:

    raise RuntimeError(
        f"Expected {EXPECTED_FILES} readable CSV files, "
        f"but only {len(column_sets)} were successfully read."
    )


# ============================================================
# STEP 2
# FIND COMMON COLUMNS IN ALL 240 FILES
# ============================================================

print()
print("=" * 100)
print("STEP 2: FINDING COLUMNS COMMON TO ALL 240 FILES")
print("=" * 100)

print()


# This is the critical operation.
# A column must exist in EVERY CSV file.

common_columns = set.intersection(
    *column_sets
)


print(
    "Number of columns common to ALL files:",
    len(common_columns)
)

print()


# ============================================================
# VERIFY EXACTLY 34 COMMON COLUMNS
# ============================================================

if len(common_columns) != EXPECTED_COMMON_COLUMNS:

    print(
        "ERROR: The dataset does not contain exactly "
        f"{EXPECTED_COMMON_COLUMNS} common columns."
    )

    print()

    print(
        "Detected common columns:",
        len(common_columns)
    )

    print()

    print(
        "Common columns detected:"
    )

    for column in sorted(
        common_columns
    ):

        print(
            column
        )

    raise RuntimeError(
        f"Expected exactly {EXPECTED_COMMON_COLUMNS} "
        f"common columns, but detected "
        f"{len(common_columns)}."
    )


# ============================================================
# STEP 3
# PRESERVE COMMON COLUMN ORDER
# ============================================================

first_file_columns = (
    file_headers[0]["columns"]
)


common_columns_ordered = [
    column
    for column in first_file_columns
    if column in common_columns
]


# ============================================================
# FINAL COMMON COLUMN CHECK
# ============================================================

if len(common_columns_ordered) != (
    EXPECTED_COMMON_COLUMNS
):

    raise RuntimeError(
        "The ordered common column list does not contain "
        f"exactly {EXPECTED_COMMON_COLUMNS} columns."
    )


# ============================================================
# DISPLAY THE 34 COMMON COLUMNS
# ============================================================

print("=" * 100)
print("THE 34 COLUMNS COMMON TO ALL 240 FILES")
print("=" * 100)

print()

for number, column in enumerate(
    common_columns_ordered,
    start=1
):

    print(
        f"{number:02d}. {column}"
    )


# ============================================================
# STEP 4
# VERIFY EACH COMMON COLUMN EXISTS IN EVERY FILE
# ============================================================

print()
print("=" * 100)
print("STEP 4: VERIFYING COMMON COLUMNS ACROSS ALL FILES")
print("=" * 100)

print()


common_column_validation = []

column_presence_errors = []


for column in common_columns_ordered:

    files_containing_column = 0

    files_missing_column = []


    for item in file_headers:

        if column in item["columns"]:

            files_containing_column += 1

        else:

            files_missing_column.append(
                str(
                    item["file"].relative_to(
                        INPUT_ROOT
                    )
                )
            )


    common_column_validation.append({

        "Column":
            column,

        "Files_Containing_Column":
            files_containing_column,

        "Expected_Files":
            EXPECTED_FILES,

        "Present_In_All_Files":
            files_containing_column
            == EXPECTED_FILES,

        "Missing_From_Files":
            len(files_missing_column),

        "Missing_File_List":
            " | ".join(
                files_missing_column
            )

    })


    if files_containing_column != (
        EXPECTED_FILES
    ):

        column_presence_errors.append(
            column
        )


# ============================================================
# SAVE COMMON COLUMN VALIDATION REPORT
# ============================================================

COMMON_COLUMN_REPORT = (
    OUTPUT_ROOT /
    "V18_common_34_column_validation.csv"
)


pd.DataFrame(
    common_column_validation
).to_csv(
    COMMON_COLUMN_REPORT,
    index=False
)


if column_presence_errors:

    raise RuntimeError(
        "Common column validation failed. "
        "Check V18_common_34_column_validation.csv."
    )


print(
    "All 34 columns are present in all 240 files."
)

print()


# ============================================================
# STEP 5
# FIND ACTIVITY COLUMN
# ============================================================

activity_column = None


for column in common_columns_ordered:

    if (
        str(column)
        .strip()
        .lower()
        ==
        ACTIVITY_COLUMN_NAME.lower()
    ):

        activity_column = column

        break


if activity_column is None:

    raise RuntimeError(
        "Activity column is not present "
        "among the 34 common columns."
    )


print(
    "Activity column detected:",
    activity_column
)


# ============================================================
# REPORT STORAGE
# ============================================================

file_reports = []

failed_files = []


# ============================================================
# GLOBAL COUNTERS
# ============================================================

total_original_columns = 0

total_removed_columns = 0

total_rows = 0

total_activity_missing_before = 0

total_activity_imputed = 0

total_activity_remaining = 0


# ============================================================
# STEP 6
# PROCESS ALL CSV FILES
# ============================================================

print()
print("=" * 100)
print("STEP 6: STANDARDIZING ALL FILES TO 34 COMMON COLUMNS")
print("=" * 100)

print()


for file_number, input_file in enumerate(
    csv_files,
    start=1
):

    try:

        # ====================================================
        # READ FILE
        # ====================================================

        df = pd.read_csv(
            input_file,
            low_memory=False
        )


        original_rows = len(df)

        original_column_count = len(
            df.columns
        )


        total_original_columns += (
            original_column_count
        )

        total_rows += original_rows


        # ====================================================
        # FIND EXTRA COLUMNS
        # ====================================================

        extra_columns = [
            column
            for column in df.columns
            if column not in common_columns
        ]


        removed_column_count = len(
            extra_columns
        )


        total_removed_columns += (
            removed_column_count
        )


        # ====================================================
        # KEEP ONLY THE 34 COMMON COLUMNS
        # ====================================================

        df = df[
            common_columns_ordered
        ].copy()


        # ====================================================
        # VERIFY 34 COLUMNS
        # ====================================================

        if len(df.columns) != (
            EXPECTED_COMMON_COLUMNS
        ):

            raise RuntimeError(
                f"Final column count is "
                f"{len(df.columns)}, expected 34."
            )


        # ====================================================
        # ACTIVITY COLUMN
        # ====================================================

        activity_series = (
            df[activity_column]
            .astype("string")
        )


        # ====================================================
        # MISSING ACTIVITY BEFORE
        # ====================================================

        activity_missing_mask = (
            activity_series.isna()
            |
            activity_series
            .str.strip()
            .eq("")
        )


        missing_before = int(
            activity_missing_mask.sum()
        )


        total_activity_missing_before += (
            missing_before
        )


        # ====================================================
        # SEQUENTIAL 21 ROW SLIDING WINDOW
        # ====================================================

        imputed_count = 0


        for current_position in range(
            len(activity_series)
        ):


            # ------------------------------------------------
            # CHECK CURRENT ACTIVITY
            # ------------------------------------------------

            current_value = (
                activity_series.iloc[
                    current_position
                ]
            )


            current_missing = (
                pd.isna(current_value)
                or
                str(current_value).strip()
                == ""
            )


            if not current_missing:

                continue


            # ------------------------------------------------
            # PREVIOUS 21 ROWS
            # ------------------------------------------------

            window_start = max(
                0,
                current_position
                -
                WINDOW_SIZE
            )


            window_end = (
                current_position
            )


            previous_21 = (
                activity_series.iloc[
                    window_start:
                    window_end
                ]
            )


            # ------------------------------------------------
            # REMOVE MISSING VALUES
            # ------------------------------------------------

            previous_21 = (
                previous_21
                .dropna()
            )


            previous_21 = (
                previous_21[
                    previous_21
                    .astype(str)
                    .str.strip()
                    .ne("")
                ]
            )


            # ------------------------------------------------
            # NO VALID PREVIOUS VALUE
            # ------------------------------------------------

            if len(previous_21) == 0:

                continue


            # ------------------------------------------------
            # CATEGORICAL MODE
            # ------------------------------------------------

            mode_values = (
                previous_21
                .mode()
            )


            if len(mode_values) == 0:

                continue


            mode_value = (
                mode_values.iloc[0]
            )


            # ------------------------------------------------
            # INSERT MODE
            # ------------------------------------------------

            activity_series.iloc[
                current_position
            ] = mode_value


            imputed_count += 1


        # ====================================================
        # PUT ACTIVITY BACK
        # ====================================================

        df[activity_column] = (
            activity_series
        )


        # ====================================================
        # REMAINING ACTIVITY MISSING
        # ====================================================

        remaining_mask = (
            df[activity_column].isna()
            |
            df[activity_column]
            .astype(str)
            .str.strip()
            .eq("")
        )


        remaining_missing = int(
            remaining_mask.sum()
        )


        total_activity_imputed += (
            imputed_count
        )

        total_activity_remaining += (
            remaining_missing
        )


        # ====================================================
        # OUTPUT PATH
        # ====================================================

        relative_path = (
            input_file.relative_to(
                INPUT_ROOT
            )
        )


        output_file = (
            OUTPUT_ROOT /
            relative_path
        )


        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # ====================================================
        # SAVE FILE
        # ====================================================

        df.to_csv(
            output_file,
            index=False
        )


        # ====================================================
        # VERIFY SAVED FILE
        # ====================================================

        check_df = pd.read_csv(
            output_file,
            nrows=0,
            low_memory=False
        )


        if len(check_df.columns) != (
            EXPECTED_COMMON_COLUMNS
        ):

            raise RuntimeError(
                f"Saved file contains "
                f"{len(check_df.columns)} columns."
            )


        # ====================================================
        # FILE REPORT
        # ====================================================

        file_reports.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(relative_path),

            "Original_Column_Count":
                original_column_count,

            "Final_Column_Count":
                len(df.columns),

            "Columns_Removed":
                removed_column_count,

            "Removed_Column_Names":
                " | ".join(
                    map(
                        str,
                        extra_columns
                    )
                ),

            "Rows":
                original_rows,

            "Activity_Missing_Before":
                missing_before,

            "Activity_Imputed":
                imputed_count,

            "Activity_Remaining_Missing":
                remaining_missing

        })


        # ====================================================
        # PROGRESS
        # ====================================================

        print(
            f"[{file_number:03d}/{len(csv_files)}] "
            f"{relative_path} | "
            f"Columns: "
            f"{original_column_count} → 34 | "
            f"Removed: "
            f"{removed_column_count} | "
            f"Activity missing: "
            f"{missing_before} | "
            f"Imputed: "
            f"{imputed_count} | "
            f"Remaining: "
            f"{remaining_missing}"
        )


    except Exception as error:

        failed_files.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(
                    input_file.relative_to(
                        INPUT_ROOT
                    )
                ),

            "Error":
                str(error)

        })


        print()
        print(
            f"ERROR: {input_file.name}"
        )

        print(
            error
        )


# ============================================================
# SAVE FILE PROCESSING REPORT
# ============================================================

FILE_REPORT = (
    OUTPUT_ROOT /
    "V18_file_processing_report.csv"
)


pd.DataFrame(
    file_reports
).to_csv(
    FILE_REPORT,
    index=False
)


# ============================================================
# SAVE FAILED FILE REPORT
# ============================================================

FAILED_REPORT = (
    OUTPUT_ROOT /
    "V18_failed_files_report.csv"
)


if failed_files:

    pd.DataFrame(
        failed_files
    ).to_csv(
        FAILED_REPORT,
        index=False
    )


# ============================================================
# FIND OUTPUT DATASET FILES
# ============================================================

output_dataset_files = [
    file
    for file in OUTPUT_ROOT.rglob("*.csv")
    if not file.name.startswith("V18_")
]


# ============================================================
# FINAL 34 COLUMN VALIDATION
# ============================================================

print()
print("=" * 100)
print("FINAL V18 COLUMN VALIDATION")
print("=" * 100)

print()


wrong_column_files = []


for output_file in output_dataset_files:

    try:

        test_df = pd.read_csv(
            output_file,
            nrows=0,
            low_memory=False
        )


        if len(test_df.columns) != (
            EXPECTED_COMMON_COLUMNS
        ):

            wrong_column_files.append({

                "File_Name":
                    output_file.name,

                "Relative_Path":
                    str(
                        output_file.relative_to(
                            OUTPUT_ROOT
                        )
                    ),

                "Column_Count":
                    len(test_df.columns)

            })


    except Exception as error:

        wrong_column_files.append({

            "File_Name":
                output_file.name,

            "Relative_Path":
                str(
                    output_file.relative_to(
                        OUTPUT_ROOT
                    )
                ),

            "Column_Count":
                f"ERROR: {error}"

        })


# ============================================================
# SAVE FINAL COLUMN VALIDATION
# ============================================================

FINAL_VALIDATION_REPORT = (
    OUTPUT_ROOT /
    "V18_final_column_validation.csv"
)


if wrong_column_files:

    pd.DataFrame(
        wrong_column_files
    ).to_csv(
        FINAL_VALIDATION_REPORT,
        index=False
    )

else:

    pd.DataFrame(
        columns=[
            "File_Name",
            "Relative_Path",
            "Column_Count"
        ]
    ).to_csv(
        FINAL_VALIDATION_REPORT,
        index=False
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 100)
print("V18 PROCESSING COMPLETED")
print("=" * 100)

print()

print(
    "Input CSV files:",
    len(csv_files)
)

print(
    "Expected CSV files:",
    EXPECTED_FILES
)

print(
    "Successfully processed:",
    len(file_reports)
)

print(
    "Failed files:",
    len(failed_files)
)

print(
    "Output dataset CSV files:",
    len(output_dataset_files)
)

print()

print(
    "Common columns found in ALL files:",
    len(common_columns_ordered)
)

print(
    "Required common columns:",
    EXPECTED_COMMON_COLUMNS
)

print(
    "Total columns removed:",
    total_removed_columns
)

print()

print(
    "Total dataset rows:",
    f"{total_rows:,}"
)

print()

print(
    "Activity missing before:",
    f"{total_activity_missing_before:,}"
)

print(
    "Activity values imputed:",
    f"{total_activity_imputed:,}"
)

print(
    "Activity remaining missing:",
    f"{total_activity_remaining:,}"
)


# ============================================================
# ACTIVITY IMPUTATION COVERAGE
# ============================================================

if total_activity_missing_before > 0:

    coverage = (
        total_activity_imputed
        /
        total_activity_missing_before
    ) * 100

    print(
        f"Activity imputation coverage: "
        f"{coverage:.2f}%"
    )


# ============================================================
# FINAL COLUMN VALIDATION SUMMARY
# ============================================================

print()
print("=" * 100)
print("34 COMMON COLUMN VALIDATION RESULT")
print("=" * 100)

print()

print(
    "Output files checked:",
    len(output_dataset_files)
)

print(
    "Files with exactly 34 columns:",
    len(output_dataset_files)
    -
    len(wrong_column_files)
)

print(
    "Files with incorrect column count:",
    len(wrong_column_files)
)


# ============================================================
# FINAL SUCCESS CHECK
# ============================================================

print()
print("=" * 100)

success = (
    len(csv_files)
    == EXPECTED_FILES
    and
    len(file_reports)
    == EXPECTED_FILES
    and
    len(failed_files)
    == 0
    and
    len(output_dataset_files)
    == EXPECTED_FILES
    and
    len(common_columns_ordered)
    == EXPECTED_COMMON_COLUMNS
    and
    len(wrong_column_files)
    == 0
)


if success:

    print("SUCCESS")

    print()
    print(
        "All 240 CSV files were processed."
    )

    print(
        "Exactly 34 columns common to all 240 files "
        "were retained."
    )

    print(
        "Extra columns were removed."
    )

    print(
        "All 34 common columns were verified "
        "to exist in every input file."
    )

    print(
        "Every V18 dataset file was verified "
        "to contain exactly 34 columns."
    )

    print(
        "Missing Activity values were imputed "
        "using a sequential 21 row sliding window."
    )

else:

    print(
        "PROCESSING COMPLETED WITH VALIDATION ISSUES"
    )

    print()
    print(
        "Please inspect the generated reports."
    )


# ============================================================
# OUTPUT LOCATIONS
# ============================================================

print()
print("=" * 100)
print("OUTPUT FILES")
print("=" * 100)

print()

print(
    "V18 dataset:"
)

print(
    OUTPUT_ROOT
)

print()

print(
    "Common 34 column validation:"
)

print(
    COMMON_COLUMN_REPORT
)

print()

print(
    "File processing report:"
)

print(
    FILE_REPORT
)

print()

print(
    "Final column validation:"
)

print(
    FINAL_VALIDATION_REPORT
)

if failed_files:

    print()

    print(
        "Failed files:"
    )

    print(
        FAILED_REPORT
    )


print()
print("=" * 100)
print("DONE")
print("=" * 100)
```

```python
import pandas as pd
import re
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V18"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V19"
)

PERSON_COLUMN = "PersonSrial"

EXPECTED_FILES = 240


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# FIND ALL DATASET CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)

# Exclude V18 generated reports
csv_files = [
    file
    for file in csv_files
    if not file.name.startswith("V18_")
]


# ============================================================
# INITIAL INFORMATION
# ============================================================

print("=" * 90)
print("CLEAN_DATA_V18 → CLEAN_DATA_V19")
print("ADDING PERSON SERIAL NUMBER")
print("=" * 90)

print()
print("Input folder:")
print(INPUT_ROOT)

print()
print("Output folder:")
print(OUTPUT_ROOT)

print()
print("CSV files found:", len(csv_files))

print()


# ============================================================
# FILE COUNT CHECK
# ============================================================

if len(csv_files) != EXPECTED_FILES:

    print("=" * 90)
    print("WARNING")
    print("=" * 90)

    print(
        f"Expected {EXPECTED_FILES} CSV files, "
        f"but found {len(csv_files)}."
    )

    print()


# ============================================================
# REPORT STORAGE
# ============================================================

file_reports = []
failed_files = []


# ============================================================
# PROCESS EVERY CSV
# ============================================================

for file_number, input_file in enumerate(
    csv_files,
    start=1
):

    try:

        # ====================================================
        # EXTRACT PERSON NUMBER FROM FILE NAME
        # ====================================================

        filename_without_extension = (
            input_file.stem
        )


        # Find numeric value in filename
        numeric_matches = re.findall(
            r"\d+",
            filename_without_extension
        )


        # ----------------------------------------------------
        # Validate numeric value
        # ----------------------------------------------------

        if len(numeric_matches) == 0:

            raise ValueError(
                f"No numeric value found in filename: "
                f"{input_file.name}"
            )


        if len(numeric_matches) > 1:

            raise ValueError(
                f"More than one numeric value found "
                f"in filename: {input_file.name} "
                f"→ {numeric_matches}"
            )


        # ----------------------------------------------------
        # Convert numeric value to integer
        # ----------------------------------------------------

        person_serial = int(
            numeric_matches[0]
        )


        # ====================================================
        # READ CSV
        # ====================================================

        df = pd.read_csv(
            input_file,
            low_memory=False
        )


        original_rows = len(df)

        original_columns = len(
            df.columns
        )


        # ====================================================
        # CHECK WHETHER PERSONSRIAL ALREADY EXISTS
        # ====================================================

        existing_person_columns = [
            column
            for column in df.columns
            if str(column).strip().lower()
            == PERSON_COLUMN.lower()
        ]


        if existing_person_columns:

            # Remove existing PersonSrial column
            # before creating the correct one

            df = df.drop(
                columns=existing_person_columns
            )


        # ====================================================
        # ADD PERSON SERIAL COLUMN
        # ====================================================

        df[PERSON_COLUMN] = person_serial


        # ====================================================
        # OUTPUT PATH
        # ====================================================

        relative_path = (
            input_file.relative_to(
                INPUT_ROOT
            )
        )


        output_file = (
            OUTPUT_ROOT /
            relative_path
        )


        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # ====================================================
        # SAVE OUTPUT
        # ====================================================

        df.to_csv(
            output_file,
            index=False
        )


        # ====================================================
        # VERIFY OUTPUT
        # ====================================================

        verification_df = pd.read_csv(
            output_file,
            low_memory=False
        )


        # ----------------------------------------------------
        # Verify PersonSrial column
        # ----------------------------------------------------

        if PERSON_COLUMN not in (
            verification_df.columns
        ):

            raise RuntimeError(
                f"{PERSON_COLUMN} column was not "
                "found after saving."
            )


        # ----------------------------------------------------
        # Verify all values
        # ----------------------------------------------------

        unique_person_values = (
            verification_df[
                PERSON_COLUMN
            ]
            .dropna()
            .unique()
        )


        if (
            len(unique_person_values) != 1
            or
            int(unique_person_values[0])
            != person_serial
        ):

            raise RuntimeError(
                f"PersonSrial verification failed "
                f"for {input_file.name}."
            )


        # ====================================================
        # FILE REPORT
        # ====================================================

        file_reports.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(relative_path),

            "PersonSrial":
                person_serial,

            "Rows":
                original_rows,

            "Original_Columns":
                original_columns,

            "Final_Columns":
                len(verification_df.columns),

            "PersonSrial_Verified":
                True

        })


        # ====================================================
        # PROGRESS
        # ====================================================

        print(
            f"[{file_number:03d}/{len(csv_files)}] "
            f"{relative_path} | "
            f"PersonSrial = {person_serial} | "
            f"Rows = {original_rows:,} | "
            f"Columns = "
            f"{original_columns} → "
            f"{len(verification_df.columns)}"
        )


    except Exception as error:

        failed_files.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(
                    input_file.relative_to(
                        INPUT_ROOT
                    )
                ),

            "Error":
                str(error)

        })


        print()
        print(
            f"ERROR: {input_file.name}"
        )

        print(
            error
        )


# ============================================================
# SAVE PROCESSING REPORT
# ============================================================

PROCESSING_REPORT = (
    OUTPUT_ROOT /
    "V19_person_serial_report.csv"
)


pd.DataFrame(
    file_reports
).to_csv(
    PROCESSING_REPORT,
    index=False
)


# ============================================================
# SAVE FAILED FILE REPORT
# ============================================================

FAILED_REPORT = (
    OUTPUT_ROOT /
    "V19_failed_files_report.csv"
)


if failed_files:

    pd.DataFrame(
        failed_files
    ).to_csv(
        FAILED_REPORT,
        index=False
    )


# ============================================================
# FIND OUTPUT DATASET FILES
# ============================================================

output_dataset_files = [
    file
    for file in OUTPUT_ROOT.rglob("*.csv")
    if not file.name.startswith("V19_")
]


# ============================================================
# FINAL VALIDATION
# ============================================================

print()
print("=" * 90)
print("FINAL PERSON SERIAL VALIDATION")
print("=" * 90)

print()


validation_errors = []


for output_file in output_dataset_files:

    try:

        df = pd.read_csv(
            output_file,
            low_memory=False
        )


        # ----------------------------------------------------
        # Extract expected person number
        # ----------------------------------------------------

        filename_without_extension = (
            output_file.stem
        )


        numeric_matches = re.findall(
            r"\d+",
            filename_without_extension
        )


        if len(numeric_matches) != 1:

            validation_errors.append({

                "File_Name":
                    output_file.name,

                "Error":
                    "Unable to determine person number"

            })

            continue


        expected_person = int(
            numeric_matches[0]
        )


        # ----------------------------------------------------
        # Check column
        # ----------------------------------------------------

        if PERSON_COLUMN not in df.columns:

            validation_errors.append({

                "File_Name":
                    output_file.name,

                "Error":
                    f"{PERSON_COLUMN} column missing"

            })

            continue


        # ----------------------------------------------------
        # Check values
        # ----------------------------------------------------

        actual_values = (
            pd.to_numeric(
                df[PERSON_COLUMN],
                errors="coerce"
            )
            .dropna()
            .unique()
        )


        if (
            len(actual_values) != 1
            or
            int(actual_values[0])
            != expected_person
        ):

            validation_errors.append({

                "File_Name":
                    output_file.name,

                "Error":
                    (
                        f"Expected {expected_person}, "
                        f"found {actual_values}"
                    )

            })


    except Exception as error:

        validation_errors.append({

            "File_Name":
                output_file.name,

            "Error":
                str(error)

        })


# ============================================================
# SAVE VALIDATION REPORT
# ============================================================

VALIDATION_REPORT = (
    OUTPUT_ROOT /
    "V19_person_serial_validation.csv"
)


if validation_errors:

    pd.DataFrame(
        validation_errors
    ).to_csv(
        VALIDATION_REPORT,
        index=False
    )

else:

    pd.DataFrame(
        columns=[
            "File_Name",
            "Error"
        ]
    ).to_csv(
        VALIDATION_REPORT,
        index=False
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 90)
print("V19 PROCESSING COMPLETED")
print("=" * 90)

print()

print(
    "Input CSV files:",
    len(csv_files)
)

print(
    "Expected CSV files:",
    EXPECTED_FILES
)

print(
    "Successfully processed:",
    len(file_reports)
)

print(
    "Failed files:",
    len(failed_files)
)

print(
    "Output dataset files:",
    len(output_dataset_files)
)

print()

print(
    "PersonSrial column:",
    PERSON_COLUMN
)

print(
    "Sliding filename number extraction:",
    "Enabled"
)

print(
    "Validation errors:",
    len(validation_errors)
)


# ============================================================
# PERSON SERIAL DISTRIBUTION
# ============================================================

if len(file_reports) > 0:

    report_df = pd.DataFrame(
        file_reports
    )


    unique_persons = sorted(
        report_df[
            "PersonSrial"
        ]
        .dropna()
        .astype(int)
        .unique()
        .tolist()
    )


    print()
    print(
        "Unique person serial numbers:"
    )

    print(
        unique_persons
    )

    print()

    print(
        "Number of unique persons:",
        len(unique_persons)
    )


# ============================================================
# FINAL SUCCESS CHECK
# ============================================================

print()
print("=" * 90)


success = (
    len(csv_files)
    == EXPECTED_FILES
    and
    len(file_reports)
    == EXPECTED_FILES
    and
    len(failed_files)
    == 0
    and
    len(output_dataset_files)
    == EXPECTED_FILES
    and
    len(validation_errors)
    == 0
)


if success:

    print("SUCCESS")

    print()

    print(
        "All 240 CSV files were processed."
    )

    print(
        "PersonSrial was extracted from each filename."
    )

    print(
        "PersonSrial was added to every row "
        "of the corresponding CSV file."
    )

    print(
        "Every output file was verified."
    )

    print()

    print(
        "V19 dataset:"
    )

    print(
        OUTPUT_ROOT
    )

else:

    print(
        "PROCESSING COMPLETED WITH VALIDATION ISSUES"
    )

    print(
        "Please inspect the generated reports."
    )


# ============================================================
# REPORT LOCATIONS
# ============================================================

print()
print("=" * 90)

print(
    "Processing report:"
)

print(
    PROCESSING_REPORT
)

print()

print(
    "Validation report:"
)

print(
    VALIDATION_REPORT
)

if failed_files:

    print()

    print(
        "Failed files report:"
    )

    print(
        FAILED_REPORT
    )

print()
print("=" * 90)
print("DONE")
print("=" * 90)
```

```python
import pandas as pd
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V19"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V20"
)

CLASS_COLUMN = "Class"

EXPECTED_FILES = 240

EXPECTED_CLASSES = {
    "class_lecture",
    "music",
    "normal",
    "reading"
}


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# FIND ALL DATASET CSV FILES
# ============================================================

csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)


# Remove V19 generated report files
csv_files = [
    file
    for file in csv_files
    if not file.name.startswith("V19_")
]


# ============================================================
# START
# ============================================================

print("=" * 90)
print("CLEAN_DATA_V19 → CLEAN_DATA_V20")
print("ADDING CLASS COLUMN")
print("=" * 90)

print()

print("Input folder:")
print(INPUT_ROOT)

print()

print("Output folder:")
print(OUTPUT_ROOT)

print()

print("CSV files found:", len(csv_files))

print()


# ============================================================
# CHECK INPUT FILE COUNT
# ============================================================

if len(csv_files) != EXPECTED_FILES:

    print("=" * 90)
    print("WARNING")
    print("=" * 90)

    print(
        "Expected:",
        EXPECTED_FILES
    )

    print(
        "Found:",
        len(csv_files)
    )

    print()


# ============================================================
# REPORT STORAGE
# ============================================================

file_reports = []

failed_files = []

validation_errors = []


# ============================================================
# CLASS COUNTERS
# ============================================================

class_counts = {
    "class_lecture": 0,
    "music": 0,
    "normal": 0,
    "reading": 0
}


# ============================================================
# PROCESS ALL CSV FILES
# ============================================================

print("=" * 90)
print("PROCESSING CSV FILES")
print("=" * 90)

print()


for file_number, input_file in enumerate(
    csv_files,
    start=1
):

    try:

        # ====================================================
        # GET RELATIVE PATH
        # ====================================================

        relative_path = input_file.relative_to(
            INPUT_ROOT
        )


        # ====================================================
        # DETERMINE TOP LEVEL CLASS FOLDER
        # ====================================================

        if len(relative_path.parts) < 2:

            raise ValueError(
                "Unable to determine class folder."
            )


        class_name = relative_path.parts[0]


        # ====================================================
        # VALIDATE CLASS FOLDER
        # ====================================================

        if class_name not in EXPECTED_CLASSES:

            raise ValueError(
                f"Unexpected class folder: {class_name}"
            )


        # ====================================================
        # READ CSV
        # ====================================================

        df = pd.read_csv(
            input_file,
            low_memory=False
        )


        original_rows = len(df)

        original_columns = len(
            df.columns
        )


        # ====================================================
        # REMOVE EXISTING CLASS COLUMN
        # ====================================================

        existing_class_columns = [
            column
            for column in df.columns
            if str(column).strip().lower()
            == CLASS_COLUMN.lower()
        ]


        if len(existing_class_columns) > 0:

            df = df.drop(
                columns=existing_class_columns
            )


        # ====================================================
        # ADD CLASS COLUMN
        # ====================================================

        df[CLASS_COLUMN] = class_name


        # ====================================================
        # OUTPUT FILE PATH
        # ====================================================

        output_file = (
            OUTPUT_ROOT /
            relative_path
        )


        output_file.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # ====================================================
        # SAVE OUTPUT CSV
        # ====================================================

        df.to_csv(
            output_file,
            index=False
        )


        # ====================================================
        # VERIFY SAVED FILE
        # ====================================================

        check_df = pd.read_csv(
            output_file,
            low_memory=False
        )


        # ====================================================
        # VERIFY CLASS COLUMN
        # ====================================================

        if CLASS_COLUMN not in check_df.columns:

            raise RuntimeError(
                "Class column was not found "
                "after saving."
            )


        # ====================================================
        # VERIFY CLASS VALUE
        # ====================================================

        class_values = (
            check_df[CLASS_COLUMN]
            .dropna()
            .astype(str)
            .unique()
        )


        if len(class_values) != 1:

            raise RuntimeError(
                f"Expected one Class value, "
                f"but found {class_values}"
            )


        if class_values[0] != class_name:

            raise RuntimeError(
                f"Class mismatch. "
                f"Expected '{class_name}', "
                f"found '{class_values[0]}'."
            )


        # ====================================================
        # VERIFY ROW COUNT
        # ====================================================

        if len(check_df) != original_rows:

            raise RuntimeError(
                "Row count changed during processing."
            )


        # ====================================================
        # VERIFY CLASS COLUMN VALUE FOR EVERY ROW
        # ====================================================

        all_class_values_correct = (
            check_df[CLASS_COLUMN]
            .astype(str)
            .eq(class_name)
            .all()
        )


        if not all_class_values_correct:

            raise RuntimeError(
                "Some rows contain an incorrect "
                "Class value."
            )


        # ====================================================
        # UPDATE CLASS COUNT
        # ====================================================

        class_counts[class_name] += 1


        # ====================================================
        # SAVE FILE REPORT
        # ====================================================

        file_reports.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(relative_path),

            "Class":
                class_name,

            "Rows":
                original_rows,

            "Original_Columns":
                original_columns,

            "Final_Columns":
                len(check_df.columns),

            "Class_Verified":
                True

        })


        # ====================================================
        # PROGRESS
        # ====================================================

        print(
            f"[{file_number:03d}/{len(csv_files)}] "
            f"{relative_path} | "
            f"Class = {class_name} | "
            f"Rows = {original_rows:,} | "
            f"Columns = "
            f"{original_columns} → "
            f"{len(check_df.columns)}"
        )


    except Exception as error:

        relative_path = input_file.relative_to(
            INPUT_ROOT
        )


        failed_files.append({

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(relative_path),

            "Error":
                str(error)

        })


        print()

        print(
            f"ERROR: {relative_path}"
        )

        print(
            "Reason:",
            error
        )

        print()


# ============================================================
# SAVE PROCESSING REPORT
# ============================================================

PROCESSING_REPORT = (
    OUTPUT_ROOT /
    "V20_class_processing_report.csv"
)


processing_report_df = pd.DataFrame(
    file_reports
)


processing_report_df.to_csv(
    PROCESSING_REPORT,
    index=False
)


# ============================================================
# SAVE FAILED FILE REPORT
# ============================================================

FAILED_REPORT = (
    OUTPUT_ROOT /
    "V20_failed_files_report.csv"
)


if len(failed_files) > 0:

    failed_report_df = pd.DataFrame(
        failed_files
    )

    failed_report_df.to_csv(
        FAILED_REPORT,
        index=False
    )

else:

    pd.DataFrame(
        columns=[
            "File_Name",
            "Relative_Path",
            "Error"
        ]
    ).to_csv(
        FAILED_REPORT,
        index=False
    )


# ============================================================
# FIND OUTPUT DATASET FILES
# ============================================================

output_dataset_files = sorted(
    [
        file
        for file in OUTPUT_ROOT.rglob("*.csv")
        if not file.name.startswith("V20_")
    ]
)


# ============================================================
# FINAL VALIDATION
# ============================================================

print()
print("=" * 90)
print("FINAL VALIDATION")
print("=" * 90)

print()


for output_file in output_dataset_files:

    try:

        # ====================================================
        # READ OUTPUT FILE
        # ====================================================

        df = pd.read_csv(
            output_file,
            low_memory=False
        )


        # ====================================================
        # GET RELATIVE PATH
        # ====================================================

        relative_path = (
            output_file.relative_to(
                OUTPUT_ROOT
            )
        )


        # ====================================================
        # DETERMINE EXPECTED CLASS
        # ====================================================

        expected_class = (
            relative_path.parts[0]
        )


        # ====================================================
        # CHECK CLASS COLUMN
        # ====================================================

        if CLASS_COLUMN not in df.columns:

            validation_errors.append({

                "File_Name":
                    output_file.name,

                "Relative_Path":
                    str(relative_path),

                "Expected_Class":
                    expected_class,

                "Actual_Class":
                    "",

                "Error":
                    "Class column missing"

            })

            continue


        # ====================================================
        # CHECK CLASS VALUES
        # ====================================================

        actual_values = (
            df[CLASS_COLUMN]
            .dropna()
            .astype(str)
            .unique()
        )


        if len(actual_values) != 1:

            validation_errors.append({

                "File_Name":
                    output_file.name,

                "Relative_Path":
                    str(relative_path),

                "Expected_Class":
                    expected_class,

                "Actual_Class":
                    str(actual_values),

                "Error":
                    "Multiple or missing Class values"

            })

            continue


        actual_class = actual_values[0]


        # ====================================================
        # CHECK CLASS VALUE
        # ====================================================

        if actual_class != expected_class:

            validation_errors.append({

                "File_Name":
                    output_file.name,

                "Relative_Path":
                    str(relative_path),

                "Expected_Class":
                    expected_class,

                "Actual_Class":
                    actual_class,

                "Error":
                    "Class value does not match folder"

            })

            continue


        # ====================================================
        # CHECK EVERY ROW
        # ====================================================

        every_row_correct = (
            df[CLASS_COLUMN]
            .astype(str)
            .eq(expected_class)
            .all()
        )


        if not every_row_correct:

            validation_errors.append({

                "File_Name":
                    output_file.name,

                "Relative_Path":
                    str(relative_path),

                "Expected_Class":
                    expected_class,

                "Actual_Class":
                    actual_class,

                "Error":
                    "Not every row contains the expected Class"

            })


    except Exception as error:

        validation_errors.append({

            "File_Name":
                output_file.name,

            "Relative_Path":
                str(
                    output_file.relative_to(
                        OUTPUT_ROOT
                    )
                ),

            "Expected_Class":
                "",

            "Actual_Class":
                "",

            "Error":
                str(error)

        })


# ============================================================
# SAVE VALIDATION REPORT
# ============================================================

VALIDATION_REPORT = (
    OUTPUT_ROOT /
    "V20_class_validation_report.csv"
)


if len(validation_errors) > 0:

    validation_df = pd.DataFrame(
        validation_errors
    )

else:

    validation_df = pd.DataFrame(
        columns=[
            "File_Name",
            "Relative_Path",
            "Expected_Class",
            "Actual_Class",
            "Error"
        ]
    )


validation_df.to_csv(
    VALIDATION_REPORT,
    index=False
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 90)
print("V20 PROCESSING COMPLETED")
print("=" * 90)

print()

print(
    "Input CSV files:",
    len(csv_files)
)

print(
    "Expected CSV files:",
    EXPECTED_FILES
)

print(
    "Successfully processed:",
    len(file_reports)
)

print(
    "Failed files:",
    len(failed_files)
)

print(
    "Output dataset files:",
    len(output_dataset_files)
)

print()

print(
    "Class distribution:"
)

print(
    "class_lecture:",
    class_counts["class_lecture"]
)

print(
    "music:",
    class_counts["music"]
)

print(
    "normal:",
    class_counts["normal"]
)

print(
    "reading:",
    class_counts["reading"]
)

print()

print(
    "Validation errors:",
    len(validation_errors)
)


# ============================================================
# FINAL SUCCESS CHECK
# ============================================================

success = (
    len(csv_files) == EXPECTED_FILES
    and
    len(file_reports) == EXPECTED_FILES
    and
    len(failed_files) == 0
    and
    len(output_dataset_files) == EXPECTED_FILES
    and
    len(validation_errors) == 0
    and
    class_counts["class_lecture"] == 60
    and
    class_counts["music"] == 60
    and
    class_counts["normal"] == 60
    and
    class_counts["reading"] == 60
)


print()
print("=" * 90)


if success:

    print("SUCCESS")

    print()

    print(
        "All 240 CSV files were processed successfully."
    )

    print(
        "Each file contains the Class column."
    )

    print(
        "Class values were taken directly from "
        "the top level folder name."
    )

    print(
        "class_lecture = 60 files"
    )

    print(
        "music = 60 files"
    )

    print(
        "normal = 60 files"
    )

    print(
        "reading = 60 files"
    )

    print()

    print(
        "All Class values passed validation."
    )

else:

    print(
        "PROCESSING COMPLETED WITH VALIDATION ISSUES"
    )

    print()

    print(
        "Please check the generated reports."
    )


# ============================================================
# OUTPUT LOCATIONS
# ============================================================

print()
print("=" * 90)
print("OUTPUT LOCATIONS")
print("=" * 90)

print()

print(
    "V20 dataset:"
)

print(
    OUTPUT_ROOT
)

print()

print(
    "Processing report:"
)

print(
    PROCESSING_REPORT
)

print()

print(
    "Validation report:"
)

print(
    VALIDATION_REPORT
)

print()

print(
    "Failed files report:"
)

print(
    FAILED_REPORT
)

print()
print("=" * 90)
print("DONE")
print("=" * 90)
```

```python
import pandas as pd
import re
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V20"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V21"
)

OUTPUT_FILE = (
    OUTPUT_ROOT /
    "V21_Merged_All_Classes_All_Persons.csv"
)

EXPECTED_PERSONS = 60

EXPECTED_CLASSES = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]

EXPECTED_FILES_PER_CLASS = 60

PERSON_COLUMN = "PersonSrial"

CLASS_COLUMN = "Class"


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# START
# ============================================================

print("=" * 100)
print("CLEAN_DATA_V20 → CLEAN_DATA_V21")
print("MERGING ALL CLASSES AND ALL PERSONS")
print("=" * 100)

print()

print("Input folder:")
print(INPUT_ROOT)

print()

print("Output folder:")
print(OUTPUT_ROOT)

print()

print(
    "Required order:"
)

print(
    "Person 1  → class_lecture → music → normal → reading"
)

print(
    "Person 2  → class_lecture → music → normal → reading"
)

print(
    "..."
)

print(
    "Person 60 → class_lecture → music → normal → reading"
)

print()


# ============================================================
# FIND ALL CSV FILES
# ============================================================

all_csv_files = sorted(
    INPUT_ROOT.rglob("*.csv")
)


# Exclude V20 report files
csv_files = [
    file
    for file in all_csv_files
    if not file.name.startswith("V20_")
]


print(
    "Dataset CSV files found:",
    len(csv_files)
)

print()


# ============================================================
# EXPECTED FILE COUNT
# ============================================================

expected_total_files = (
    EXPECTED_PERSONS
    *
    len(EXPECTED_CLASSES)
)


if len(csv_files) != expected_total_files:

    print("=" * 100)
    print("WARNING")
    print("=" * 100)

    print(
        "Expected dataset files:",
        expected_total_files
    )

    print(
        "Found dataset files:",
        len(csv_files)
    )

    print()


# ============================================================
# CREATE FILE INDEX
# ============================================================

print("=" * 100)
print("STEP 1: INDEXING PERSON AND CLASS FILES")
print("=" * 100)

print()


file_index = {}

index_errors = []


for file_path in csv_files:

    try:

        # ====================================================
        # GET RELATIVE PATH
        # ====================================================

        relative_path = (
            file_path.relative_to(
                INPUT_ROOT
            )
        )


        # ====================================================
        # DETERMINE CLASS
        # ====================================================

        if len(relative_path.parts) < 2:

            raise ValueError(
                "Unable to determine class folder."
            )


        class_name = (
            relative_path.parts[0]
        )


        # ====================================================
        # VALIDATE CLASS
        # ====================================================

        if class_name not in EXPECTED_CLASSES:

            raise ValueError(
                f"Unexpected class folder: "
                f"{class_name}"
            )


        # ====================================================
        # EXTRACT PERSON NUMBER FROM FILENAME
        # ====================================================

        filename_without_extension = (
            file_path.stem
        )


        numeric_matches = re.findall(
            r"\d+",
            filename_without_extension
        )


        if len(numeric_matches) != 1:

            raise ValueError(
                f"Expected exactly one numeric "
                f"person number in filename, "
                f"found {numeric_matches}"
            )


        person_number = int(
            numeric_matches[0]
        )


        # ====================================================
        # VALIDATE PERSON NUMBER
        # ====================================================

        if not (
            1
            <=
            person_number
            <=
            EXPECTED_PERSONS
        ):

            raise ValueError(
                f"Person number {person_number} "
                f"is outside 1 to "
                f"{EXPECTED_PERSONS}."
            )


        # ====================================================
        # CREATE INDEX KEY
        # ====================================================

        key = (
            person_number,
            class_name
        )


        # ====================================================
        # CHECK DUPLICATE FILE
        # ====================================================

        if key in file_index:

            raise ValueError(
                f"Duplicate file for "
                f"Person {person_number}, "
                f"Class {class_name}: "
                f"{file_index[key].name} "
                f"and {file_path.name}"
            )


        # ====================================================
        # STORE FILE
        # ====================================================

        file_index[key] = file_path


        print(
            f"Indexed: "
            f"Person {person_number:02d} | "
            f"{class_name:<14} | "
            f"{relative_path}"
        )


    except Exception as error:

        index_errors.append({

            "File_Name":
                file_path.name,

            "Relative_Path":
                str(
                    file_path.relative_to(
                        INPUT_ROOT
                    )
                ),

            "Error":
                str(error)

        })


# ============================================================
# STOP IF INDEX ERRORS
# ============================================================

if index_errors:

    INDEX_ERROR_REPORT = (
        OUTPUT_ROOT /
        "V21_index_errors.csv"
    )


    pd.DataFrame(
        index_errors
    ).to_csv(
        INDEX_ERROR_REPORT,
        index=False
    )


    raise RuntimeError(
        "File indexing failed. "
        f"Check: {INDEX_ERROR_REPORT}"
    )


# ============================================================
# CHECK ALL EXPECTED FILES EXIST
# ============================================================

print()
print("=" * 100)
print("STEP 2: CHECKING ALL PERSON CLASS COMBINATIONS")
print("=" * 100)

print()


missing_combinations = []


for person_number in range(
    1,
    EXPECTED_PERSONS + 1
):

    for class_name in EXPECTED_CLASSES:

        key = (
            person_number,
            class_name
        )


        if key not in file_index:

            missing_combinations.append({

                "PersonSrial":
                    person_number,

                "Class":
                    class_name

            })


# ============================================================
# CHECK EXPECTED 240 FILES
# ============================================================

expected_combinations = (
    EXPECTED_PERSONS
    *
    len(EXPECTED_CLASSES)
)


print(
    "Expected combinations:",
    expected_combinations
)

print(
    "Indexed combinations:",
    len(file_index)
)

print(
    "Missing combinations:",
    len(missing_combinations)
)


if missing_combinations:

    MISSING_REPORT = (
        OUTPUT_ROOT /
        "V21_missing_person_class_files.csv"
    )


    pd.DataFrame(
        missing_combinations
    ).to_csv(
        MISSING_REPORT,
        index=False
    )


    print()

    print(
        "Missing combinations report:"
    )

    print(
        MISSING_REPORT
    )


    raise RuntimeError(
        "Some Person/Class files are missing. "
        "Merging stopped."
    )


# ============================================================
# CHECK EXACTLY 240 FILES
# ============================================================

if len(file_index) != expected_combinations:

    raise RuntimeError(
        f"Expected {expected_combinations} "
        f"unique Person/Class files, "
        f"but found {len(file_index)}."
    )


print()
print(
    "All 240 Person/Class combinations exist."
)

print()


# ============================================================
# STEP 3
# READ AND MERGE IN REQUIRED ORDER
# ============================================================

print("=" * 100)
print("STEP 3: MERGING DATA")
print("=" * 100)

print()


merged_parts = []

merge_report = []

reference_columns = None


# ============================================================
# PERSON BY PERSON
# ============================================================

for person_number in range(
    1,
    EXPECTED_PERSONS + 1
):

    print()
    print(
        "-" * 80
    )

    print(
        f"PERSON {person_number}"
    )

    print(
        "-" * 80
    )


    # ========================================================
    # CLASS BY CLASS
    # ========================================================

    for class_name in EXPECTED_CLASSES:

        key = (
            person_number,
            class_name
        )


        input_file = file_index[key]


        # ====================================================
        # READ FILE
        # ====================================================

        df = pd.read_csv(
            input_file,
            low_memory=False
        )


        original_rows = len(df)


        # ====================================================
        # VERIFY PERSONSRIAL
        # ====================================================

        if PERSON_COLUMN in df.columns:

            person_values = (
                pd.to_numeric(
                    df[PERSON_COLUMN],
                    errors="coerce"
                )
                .dropna()
                .unique()
            )


            if len(person_values) != 1:

                raise RuntimeError(
                    f"Invalid {PERSON_COLUMN} "
                    f"in {input_file.name}"
                )


            if int(person_values[0]) != (
                person_number
            ):

                raise RuntimeError(
                    f"PersonSrial mismatch in "
                    f"{input_file.name}. "
                    f"Expected {person_number}, "
                    f"found {person_values}."
                )


        # ====================================================
        # VERIFY CLASS
        # ====================================================

        if CLASS_COLUMN in df.columns:

            class_values = (
                df[CLASS_COLUMN]
                .dropna()
                .astype(str)
                .unique()
            )


            if len(class_values) != 1:

                raise RuntimeError(
                    f"Invalid Class values in "
                    f"{input_file.name}: "
                    f"{class_values}"
                )


            if class_values[0] != class_name:

                raise RuntimeError(
                    f"Class mismatch in "
                    f"{input_file.name}. "
                    f"Expected {class_name}, "
                    f"found {class_values}."
                )


        # ====================================================
        # CHECK COLUMN STRUCTURE
        # ====================================================

        current_columns = list(
            df.columns
        )


        if reference_columns is None:

            reference_columns = (
                current_columns
            )

        else:

            if current_columns != (
                reference_columns
            ):

                raise RuntimeError(
                    "Column structure mismatch "
                    f"in {input_file.name}."
                )


        # ====================================================
        # ADD ORDER INFORMATION
        # ====================================================

        df["_MergePersonOrder"] = (
            person_number
        )

        df["_MergeClassOrder"] = (
            EXPECTED_CLASSES.index(
                class_name
            )
            + 1
        )


        # ====================================================
        # STORE DATAFRAME
        # ====================================================

        merged_parts.append(
            df
        )


        # ====================================================
        # REPORT
        # ====================================================

        merge_report.append({

            "PersonSrial":
                person_number,

            "Class":
                class_name,

            "File_Name":
                input_file.name,

            "Relative_Path":
                str(
                    input_file.relative_to(
                        INPUT_ROOT
                    )
                ),

            "Rows":
                original_rows

        })


        print(
            f"Person {person_number:02d} | "
            f"{class_name:<14} | "
            f"{input_file.name:<15} | "
            f"Rows = {original_rows:,}"
        )


# ============================================================
# STEP 4
# COMBINE ALL DATA
# ============================================================

print()
print("=" * 100)
print("STEP 4: COMBINING ALL DATA")
print("=" * 100)

print()


merged_df = pd.concat(
    merged_parts,
    ignore_index=True
)


# ============================================================
# REMOVE TEMPORARY ORDER COLUMNS
# ============================================================

merged_df = merged_df.drop(
    columns=[
        "_MergePersonOrder",
        "_MergeClassOrder"
    ]
)


# ============================================================
# TOTAL ROWS
# ============================================================

total_rows = len(
    merged_df
)


print(
    "Total merged rows:",
    f"{total_rows:,}"
)

print(
    "Total merged columns:",
    len(merged_df.columns)
)

print()


# ============================================================
# STEP 5
# VERIFY FINAL ORDER
# ============================================================

print("=" * 100)
print("STEP 5: VERIFYING PERSON AND CLASS ORDER")
print("=" * 100)

print()


# ============================================================
# CREATE EXPECTED ORDER
# ============================================================

expected_order = []


for person_number in range(
    1,
    EXPECTED_PERSONS + 1
):

    for class_name in EXPECTED_CLASSES:

        expected_order.append(
            (
                person_number,
                class_name
            )
        )


# ============================================================
# CHECK ACTUAL ORDER
# ============================================================

actual_order = []


for person_value, class_value in zip(
    merged_df[PERSON_COLUMN],
    merged_df[CLASS_COLUMN]
):

    actual_order.append(
        (
            int(person_value),
            str(class_value)
        )
    )


# ============================================================
# CHECK BLOCK ORDER
# ============================================================

order_errors = []


current_expected_index = 0


for person_number in range(
    1,
    EXPECTED_PERSONS + 1
):

    for class_name in EXPECTED_CLASSES:

        # Find all rows for this expected block
        block_mask = (
            merged_df[PERSON_COLUMN]
            == person_number
        ) & (
            merged_df[CLASS_COLUMN]
            == class_name
        )


        block_indices = (
            merged_df.index[
                block_mask
            ]
        )


        if len(block_indices) == 0:

            order_errors.append({

                "PersonSrial":
                    person_number,

                "Class":
                    class_name,

                "Error":
                    "No rows found"

            })


            continue


        # Check that all rows of this block
        # occur continuously

        first_index = (
            block_indices.min()
        )

        last_index = (
            block_indices.max()
        )


        expected_block_length = (
            last_index
            -
            first_index
            +
            1
        )


        if expected_block_length != (
            len(block_indices)
        ):

            order_errors.append({

                "PersonSrial":
                    person_number,

                "Class":
                    class_name,

                "Error":
                    "Rows are not contiguous"

            })


# ============================================================
# SAVE ORDER ERROR REPORT
# ============================================================

ORDER_REPORT = (
    OUTPUT_ROOT /
    "V21_order_validation.csv"
)


if order_errors:

    pd.DataFrame(
        order_errors
    ).to_csv(
        ORDER_REPORT,
        index=False
    )

else:

    pd.DataFrame(
        columns=[
            "PersonSrial",
            "Class",
            "Error"
        ]
    ).to_csv(
        ORDER_REPORT,
        index=False
    )


# ============================================================
# STEP 6
# VERIFY EVERY PERSON HAS FOUR CLASSES
# ============================================================

print()
print("=" * 100)
print("STEP 6: PERSON AND CLASS DISTRIBUTION")
print("=" * 100)

print()


distribution = (
    merged_df
    .groupby(
        [
            PERSON_COLUMN,
            CLASS_COLUMN
        ]
    )
    .size()
    .reset_index(
        name="Rows"
    )
)


# ============================================================
# DISPLAY DISTRIBUTION
# ============================================================

for person_number in range(
    1,
    EXPECTED_PERSONS + 1
):

    print(
        f"Person {person_number:02d}:",
        end=" "
    )


    person_data = distribution[
        distribution[PERSON_COLUMN]
        ==
        person_number
    ]


    for class_name in EXPECTED_CLASSES:

        match = person_data[
            person_data[CLASS_COLUMN]
            ==
            class_name
        ]


        if len(match) == 1:

            row_count = int(
                match["Rows"].iloc[0]
            )

            print(
                f"{class_name}={row_count:,}",
                end=" | "
            )

        else:

            print(
                f"{class_name}=MISSING",
                end=" | "
            )


    print()


# ============================================================
# SAVE DISTRIBUTION REPORT
# ============================================================

DISTRIBUTION_REPORT = (
    OUTPUT_ROOT /
    "V21_person_class_row_distribution.csv"
)


distribution.to_csv(
    DISTRIBUTION_REPORT,
    index=False
)


# ============================================================
# STEP 7
# SAVE MERGED DATASET
# ============================================================

print()
print("=" * 100)
print("STEP 7: SAVING FINAL MERGED DATASET")
print("=" * 100)

print()


merged_df.to_csv(
    OUTPUT_FILE,
    index=False
)


print(
    "Merged dataset saved:"
)

print(
    OUTPUT_FILE
)

print()


# ============================================================
# STEP 8
# RELOAD AND VERIFY SAVED DATASET
# ============================================================

print("=" * 100)
print("STEP 8: FINAL SAVED FILE VERIFICATION")
print("=" * 100)

print()


verification_df = pd.read_csv(
    OUTPUT_FILE,
    low_memory=False
)


# ============================================================
# ROW COUNT CHECK
# ============================================================

if len(verification_df) != total_rows:

    raise RuntimeError(
        "Final row count changed "
        "after saving."
    )


# ============================================================
# COLUMN CHECK
# ============================================================

if list(
    verification_df.columns
) != list(
    merged_df.columns
):

    raise RuntimeError(
        "Final column structure changed "
        "after saving."
    )


# ============================================================
# PERSON COUNT CHECK
# ============================================================

unique_persons = sorted(
    pd.to_numeric(
        verification_df[
            PERSON_COLUMN
        ],
        errors="coerce"
    )
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)


print(
    "Unique persons:",
    len(unique_persons)
)

print(
    "Person numbers:",
    unique_persons
)


if unique_persons != list(
    range(
        1,
        EXPECTED_PERSONS + 1
    )
):

    raise RuntimeError(
        "Person serial validation failed."
    )


# ============================================================
# CLASS CHECK
# ============================================================

unique_classes = sorted(
    verification_df[
        CLASS_COLUMN
    ]
    .dropna()
    .astype(str)
    .unique()
    .tolist()
)


print()

print(
    "Unique classes:",
    unique_classes
)


if set(unique_classes) != set(
    EXPECTED_CLASSES
):

    raise RuntimeError(
        "Class validation failed."
    )


# ============================================================
# FINAL ORDER VERIFICATION
# ============================================================

previous_person = None
previous_class_index = None

final_order_error = False


for person_value, class_value in zip(
    verification_df[PERSON_COLUMN],
    verification_df[CLASS_COLUMN]
):

    person_value = int(
        person_value
    )

    class_value = str(
        class_value
    )

    class_index = (
        EXPECTED_CLASSES.index(
            class_value
        )
    )


    if previous_person is None:

        previous_person = (
            person_value
        )

        previous_class_index = (
            class_index
        )

        continue


    # --------------------------------------------------------
    # Same person
    # --------------------------------------------------------

    if person_value == previous_person:

        if class_index < previous_class_index:

            final_order_error = True

            break


    # --------------------------------------------------------
    # New person
    # --------------------------------------------------------

    elif person_value > previous_person:

        # New person must start from
        # class_lecture

        if class_index != 0:

            final_order_error = True

            break


        previous_person = (
            person_value
        )


    else:

        final_order_error = True

        break


    previous_class_index = (
        class_index
    )


if final_order_error:

    raise RuntimeError(
        "Final person/class ordering validation failed."
    )


# ============================================================
# SAVE MERGE REPORT
# ============================================================

MERGE_REPORT = (
    OUTPUT_ROOT /
    "V21_merge_report.csv"
)


pd.DataFrame(
    merge_report
).to_csv(
    MERGE_REPORT,
    index=False
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 100)
print("V21 MERGING COMPLETED")
print("=" * 100)

print()

print(
    "Input files:",
    len(csv_files)
)

print(
    "Expected files:",
    expected_total_files
)

print(
    "Persons:",
    EXPECTED_PERSONS
)

print(
    "Classes per person:",
    len(EXPECTED_CLASSES)
)

print(
    "Total Person/Class combinations:",
    len(file_index)
)

print(
    "Merged rows:",
    f"{len(verification_df):,}"
)

print(
    "Merged columns:",
    len(verification_df.columns)
)

print(
    "Unique persons:",
    len(unique_persons)
)

print(
    "Unique classes:",
    len(unique_classes)
)

print(
    "Order errors:",
    len(order_errors)
)

print()


# ============================================================
# FINAL SUCCESS CHECK
# ============================================================

success = (
    len(csv_files)
    ==
    expected_total_files
    and
    len(file_index)
    ==
    expected_total_files
    and
    len(merged_df)
    ==
    len(verification_df)
    and
    unique_persons
    ==
    list(
        range(
            1,
            EXPECTED_PERSONS + 1
        )
    )
    and
    set(unique_classes)
    ==
    set(EXPECTED_CLASSES)
    and
    len(order_errors)
    ==
    0
    and
    not final_order_error
)


print("=" * 100)


if success:

    print("SUCCESS")

    print()

    print(
        "All 240 CSV files were merged."
    )

    print(
        "The merge order is:"
    )

    print(
        "Person 1 → class_lecture → music → normal → reading"
    )

    print(
        "Person 2 → class_lecture → music → normal → reading"
    )

    print(
        "..."
    )

    print(
        "Person 60 → class_lecture → music → normal → reading"
    )

    print()

    print(
        "The final dataset contains all "
        "4 classes for all 60 persons."
    )

    print()

    print(
        "Final dataset:"
    )

    print(
        OUTPUT_FILE
    )


else:

    print(
        "PROCESSING COMPLETED WITH VALIDATION ISSUES"
    )

    print()

    print(
        "Please check the generated reports."
    )


# ============================================================
# REPORT LOCATIONS
# ============================================================

print()
print("=" * 100)

print(
    "Merge report:"
)

print(
    MERGE_REPORT
)

print()

print(
    "Person/Class distribution:"
)

print(
    DISTRIBUTION_REPORT
)

print()

print(
    "Order validation:"
)

print(
    ORDER_REPORT
)

if index_errors:

    print()

    print(
        "Index errors:"
    )

    print(
        INDEX_ERROR_REPORT
    )

print()
print("=" * 100)
print("DONE")
print("=" * 100)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

INPUT_FILE = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V21/"
    "V21_Merged_All_Classes_All_Persons.csv"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V21_ANALYSIS"
)

VISUALIZATION_DIR = (
    OUTPUT_ROOT /
    "visualizations"
)

REPORT_DIR = (
    OUTPUT_ROOT /
    "reports"
)


# Number of rows processed at one time
CHUNK_SIZE = 100000


# ============================================================
# CREATE DIRECTORIES
# ============================================================

VISUALIZATION_DIR.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# START
# ============================================================

print("=" * 100)
print("V21 DATASET NOISE AND OUTLIER ANALYSIS")
print("=" * 100)

print()

print("Input file:")
print(INPUT_FILE)

print()

print("Output folder:")
print(OUTPUT_ROOT)

print()


# ============================================================
# CHECK INPUT FILE
# ============================================================

if not INPUT_FILE.exists():

    raise FileNotFoundError(
        f"Input file was not found:\n{INPUT_FILE}"
    )


# ============================================================
# READ HEADER
# ============================================================

header_df = pd.read_csv(
    INPUT_FILE,
    nrows=0
)

all_columns = list(
    header_df.columns
)


print(
    "Total columns:",
    len(all_columns)
)

print()

print(
    "Columns:"
)

for number, column in enumerate(
    all_columns,
    start=1
):

    print(
        f"{number:02d}. {column}"
    )


# ============================================================
# IDENTIFY METADATA COLUMNS
# ============================================================

metadata_columns = []

for column in all_columns:

    column_lower = (
        str(column)
        .strip()
        .lower()
    )

    if column_lower in [
        "timestamp",
        "activity",
        "personsrial",
        "class"
    ]:

        metadata_columns.append(
            column
        )


# ============================================================
# NUMERICAL SIGNAL COLUMNS
# ============================================================

signal_columns = [
    column
    for column in all_columns
    if column not in metadata_columns
]


print()

print(
    "Metadata columns:",
    metadata_columns
)

print()

print(
    "Signal columns:",
    signal_columns
)

print()

print(
    "Number of signal columns:",
    len(signal_columns)
)


if len(signal_columns) == 0:

    raise RuntimeError(
        "No numerical signal columns were detected."
    )


# ============================================================
# STORAGE FOR NOISE ANALYSIS
# ============================================================

noise_statistics = {
    column: {
        "missing": 0,
        "zero": 0,
        "flatline": 0,
        "abrupt_change": 0,
        "total": 0
    }
    for column in signal_columns
}


# ============================================================
# STORAGE FOR OUTLIER ANALYSIS
# ============================================================

outlier_values = {
    column: []
    for column in signal_columns
}


# ============================================================
# FIRST PASS
# NOISE AND OUTLIER STATISTICS
# ============================================================

print()
print("=" * 100)
print("ANALYZING DATA")
print("=" * 100)

print()


total_rows = 0


for chunk_number, chunk in enumerate(
    pd.read_csv(
        INPUT_FILE,
        chunksize=CHUNK_SIZE,
        low_memory=False
    ),
    start=1
):

    total_rows += len(chunk)


    # ========================================================
    # PROCESS SIGNAL COLUMNS
    # ========================================================

    for column in signal_columns:

        series = pd.to_numeric(
            chunk[column],
            errors="coerce"
        )


        values = series.to_numpy(
            dtype=float
        )


        # ----------------------------------------------------
        # Missing values
        # ----------------------------------------------------

        missing_mask = np.isnan(
            values
        )


        noise_statistics[column][
            "missing"
        ] += int(
            missing_mask.sum()
        )


        # ----------------------------------------------------
        # Zero values
        # ----------------------------------------------------

        zero_mask = (
            np.isfinite(values)
            &
            (values == 0)
        )


        noise_statistics[column][
            "zero"
        ] += int(
            zero_mask.sum()
        )


        # ----------------------------------------------------
        # Flatline detection
        #
        # Consecutive identical values
        # ----------------------------------------------------

        if len(values) > 1:

            valid_values = np.where(
                np.isfinite(values),
                values,
                np.nan
            )


            same_as_previous = (
                np.isclose(
                    valid_values[1:],
                    valid_values[:-1],
                    equal_nan=False
                )
            )


            flatline_count = int(
                same_as_previous.sum()
            )

            noise_statistics[column][
                "flatline"
            ] += flatline_count


        # ----------------------------------------------------
        # Abrupt change detection
        #
        # Difference greater than:
        #
        # median + 3 × MAD
        # ----------------------------------------------------

        valid_values = values[
            np.isfinite(values)
        ]


        if len(valid_values) > 10:

            differences = np.abs(
                np.diff(
                    valid_values
                )
            )


            if len(differences) > 0:

                median_difference = np.median(
                    differences
                )


                mad = np.median(
                    np.abs(
                        differences
                        -
                        median_difference
                    )
                )


                robust_threshold = (
                    median_difference
                    +
                    6 * mad
                )


                if robust_threshold > 0:

                    abrupt_count = int(
                        (
                            differences
                            >
                            robust_threshold
                        ).sum()
                    )

                    noise_statistics[column][
                        "abrupt_change"
                    ] += abrupt_count


        # ----------------------------------------------------
        # Total
        # ----------------------------------------------------

        noise_statistics[column][
            "total"
        ] += len(values)


        # ----------------------------------------------------
        # Keep representative values for
        # outlier visualization
        # ----------------------------------------------------

        valid_for_outlier = values[
            np.isfinite(values)
        ]


        if len(valid_for_outlier) > 0:

            # Limit memory usage
            sample_size = min(
                5000,
                len(valid_for_outlier)
            )


            if len(valid_for_outlier) > sample_size:

                sample = np.random.choice(
                    valid_for_outlier,
                    size=sample_size,
                    replace=False
                )

            else:

                sample = valid_for_outlier


            outlier_values[column].extend(
                sample.tolist()
            )


    print(
        f"Processed chunk {chunk_number} | "
        f"Rows processed: {total_rows:,}"
    )


# ============================================================
# NOISE REPORT
# ============================================================

noise_rows = []


for column in signal_columns:

    data = noise_statistics[column]


    noise_rows.append({

        "Column":
            column,

        "Total_Values":
            data["total"],

        "Missing_Values":
            data["missing"],

        "Zero_Values":
            data["zero"],

        "Flatline_Points":
            data["flatline"],

        "Abrupt_Change_Points":
            data["abrupt_change"]

    })


noise_report = pd.DataFrame(
    noise_rows
)


NOISE_REPORT_FILE = (
    REPORT_DIR /
    "V21_noise_analysis_report.csv"
)


noise_report.to_csv(
    NOISE_REPORT_FILE,
    index=False
)


# ============================================================
# NOISE GRAPH
# ============================================================

print()
print("=" * 100)
print("CREATING NOISE GRAPH")
print("=" * 100)

print()


# Combine noise indicators

noise_plot = noise_report[
    [
        "Column",
        "Missing_Values",
        "Zero_Values",
        "Flatline_Points",
        "Abrupt_Change_Points"
    ]
].copy()


# Convert to long format

noise_long = noise_plot.melt(
    id_vars="Column",
    var_name="Noise_Type",
    value_name="Count"
)


# Create one single graph

plt.figure(
    figsize=(18, 9)
)


# Use numeric x positions

columns = noise_plot["Column"].tolist()

x = np.arange(
    len(columns)
)

width = 0.20


plt.bar(
    x - 1.5 * width,
    noise_plot["Missing_Values"],
    width=width,
    label="Missing"
)


plt.bar(
    x - 0.5 * width,
    noise_plot["Zero_Values"],
    width=width,
    label="Zero"
)


plt.bar(
    x + 0.5 * width,
    noise_plot["Flatline_Points"],
    width=width,
    label="Flatline"
)


plt.bar(
    x + 1.5 * width,
    noise_plot["Abrupt_Change_Points"],
    width=width,
    label="Abrupt Change"
)


plt.xticks(
    x,
    columns,
    rotation=45,
    ha="right"
)


plt.xlabel(
    "Signal Channel"
)

plt.ylabel(
    "Number of Detected Points"
)

plt.title(
    "V21 Dataset Noise Structure Across Signal Channels"
)

plt.legend()

plt.tight_layout()


NOISE_GRAPH = (
    VISUALIZATION_DIR /
    "V21_Noise_Structure.png"
)


plt.savefig(
    NOISE_GRAPH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# OUTLIER ANALYSIS
# ============================================================

print()
print("=" * 100)
print("CALCULATING OUTLIERS")
print("=" * 100)

print()


outlier_rows = []


for column in signal_columns:

    values = np.asarray(
        outlier_values[column],
        dtype=float
    )


    values = values[
        np.isfinite(values)
    ]


    if len(values) == 0:

        outlier_rows.append({

            "Column":
                column,

            "Q1":
                np.nan,

            "Q3":
                np.nan,

            "IQR":
                np.nan,

            "Lower_Bound":
                np.nan,

            "Upper_Bound":
                np.nan,

            "Outlier_Count":
                0

        })

        continue


    # --------------------------------------------------------
    # IQR
    # --------------------------------------------------------

    q1 = np.percentile(
        values,
        25
    )


    q3 = np.percentile(
        values,
        75
    )


    iqr = q3 - q1


    lower_bound = (
        q1
        -
        1.5 * iqr
    )


    upper_bound = (
        q3
        +
        1.5 * iqr
    )


    outlier_mask = (
        (values < lower_bound)
        |
        (values > upper_bound)
    )


    outlier_count = int(
        outlier_mask.sum()
    )


    outlier_rows.append({

        "Column":
            column,

        "Q1":
            q1,

        "Q3":
            q3,

        "IQR":
            iqr,

        "Lower_Bound":
            lower_bound,

        "Upper_Bound":
            upper_bound,

        "Outlier_Count":
            outlier_count

    })


outlier_report = pd.DataFrame(
    outlier_rows
)


OUTLIER_REPORT_FILE = (
    REPORT_DIR /
    "V21_outlier_analysis_report.csv"
)


outlier_report.to_csv(
    OUTLIER_REPORT_FILE,
    index=False
)


# ============================================================
# OUTLIER GRAPH
# ============================================================

print()
print("=" * 100)
print("CREATING OUTLIER GRAPH")
print("=" * 100)

print()


plt.figure(
    figsize=(18, 9)
)


plt.bar(
    outlier_report["Column"],
    outlier_report["Outlier_Count"]
)


plt.xticks(
    rotation=45,
    ha="right"
)


plt.xlabel(
    "Signal Channel"
)

plt.ylabel(
    "Detected Outlier Points"
)

plt.title(
    "V21 Dataset Outlier Structure Across Signal Channels"
)

plt.tight_layout()


OUTLIER_GRAPH = (
    VISUALIZATION_DIR /
    "V21_Outlier_Structure.png"
)


plt.savefig(
    OUTLIER_GRAPH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 100)
print("V21 NOISE AND OUTLIER ANALYSIS COMPLETED")
print("=" * 100)

print()

print(
    "Total rows analyzed:",
    f"{total_rows:,}"
)

print(
    "Signal columns analyzed:",
    len(signal_columns)
)

print()

print(
    "Noise report:"
)

print(
    NOISE_REPORT_FILE
)

print()

print(
    "Noise visualization:"
)

print(
    NOISE_GRAPH
)

print()

print(
    "Outlier report:"
)

print(
    OUTLIER_REPORT_FILE
)

print()

print(
    "Outlier visualization:"
)

print(
    OUTLIER_GRAPH
)

print()

print("=" * 100)
print("DONE")
print("=" * 100)
```

```python
import os
from pathlib import Path

import numpy as np
import pandas as pd


# ============================================================
# V21 TO V22 SIGNAL QUALITY CLEANING
# ============================================================

INPUT_FILE = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V21/"
    "V21_Merged_All_Classes_All_Persons.csv"
)

OUTPUT_FOLDER = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V22"
)

OUTPUT_FILE = (
    OUTPUT_FOLDER /
    "V21_Merged_All_Classes_All_Persons_Cleaned.csv"
)

REPORT_FOLDER = OUTPUT_FOLDER / "reports"


# ============================================================
# PARAMETERS
# ============================================================

WINDOW_SIZE = 21

HAMPEL_THRESHOLD = 3.0

FLATLINE_LENGTH = 5

ABRUPT_THRESHOLD = 6.0

INTERPOLATION_LIMIT = 21


# ============================================================
# COLUMNS THAT MUST NOT BE MODIFIED
# ============================================================

META_COLUMNS = {
    "TimeStamp",
    "Activity",
    "Class",
    "PersonSrial"
}


# ============================================================
# CREATE OUTPUT DIRECTORIES
# ============================================================

OUTPUT_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# CHECK INPUT
# ============================================================

print("=" * 80)
print("V21 TO V22 SIGNAL QUALITY CLEANING")
print("=" * 80)

print()
print("Input file:")
print(INPUT_FILE)

print()
print("Output folder:")
print(OUTPUT_FOLDER)


if not INPUT_FILE.exists():

    raise FileNotFoundError(
        f"Input file was not found:\n{INPUT_FILE}"
    )


# ============================================================
# READ DATA
# ============================================================

print()
print("Reading dataset...")

df = pd.read_csv(
    INPUT_FILE
)

print()
print("Dataset shape:")
print(df.shape)

print()
print("Columns:")
print(list(df.columns))


# ============================================================
# IDENTIFY SIGNAL COLUMNS
# ============================================================

signal_columns = []

for column in df.columns:

    if column in META_COLUMNS:
        continue

    numeric_values = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    if numeric_values.notna().sum() > 0:

        signal_columns.append(column)


print()
print("=" * 80)
print("SIGNAL CHANNELS")
print("=" * 80)

print()
print("Number of signal channels:")
print(len(signal_columns))

print()
print(signal_columns)


# ============================================================
# MAD
# ============================================================

def calculate_mad(values):

    values = np.asarray(
        values,
        dtype=float
    )

    values = values[
        np.isfinite(values)
    ]

    if len(values) == 0:

        return 0.0

    median_value = np.median(
        values
    )

    return np.median(
        np.abs(
            values - median_value
        )
    )


# ============================================================
# HAMPEL OUTLIER DETECTION
# ============================================================

def detect_hampel_outliers(
    values,
    window_size=21,
    threshold=3.0
):

    values = np.asarray(
        values,
        dtype=float
    )

    mask = np.zeros(
        len(values),
        dtype=bool
    )

    half_window = window_size // 2

    for i in range(len(values)):

        start = max(
            0,
            i - half_window
        )

        end = min(
            len(values),
            i + half_window + 1
        )

        window = values[start:end]

        valid = window[
            np.isfinite(window)
        ]

        if len(valid) < 5:
            continue

        local_median = np.median(
            valid
        )

        local_mad = calculate_mad(
            valid
        )

        if local_mad == 0:
            continue

        robust_scale = (
            1.4826 * local_mad
        )

        score = (
            abs(
                values[i] -
                local_median
            )
            / robust_scale
        )

        if score > threshold:

            mask[i] = True

    return mask


# ============================================================
# FLATLINE DETECTION
# ============================================================

def detect_flatlines(
    values,
    minimum_length=5
):

    values = np.asarray(
        values,
        dtype=float
    )

    mask = np.zeros(
        len(values),
        dtype=bool
    )

    if len(values) < 2:
        return mask

    start = 0

    for i in range(
        1,
        len(values)
    ):

        same_value = (
            np.isfinite(
                values[i]
            )
            and
            np.isfinite(
                values[i - 1]
            )
            and
            values[i] == values[i - 1]
        )

        if not same_value:

            run_length = (
                i - start
            )

            if run_length >= minimum_length:

                mask[
                    start:i
                ] = True

            start = i

    run_length = (
        len(values) - start
    )

    if run_length >= minimum_length:

        mask[start:] = True

    return mask


# ============================================================
# ABRUPT CHANGE DETECTION
# ============================================================

def detect_abrupt_changes(
    values,
    window_size=21,
    threshold=6.0
):

    values = np.asarray(
        values,
        dtype=float
    )

    mask = np.zeros(
        len(values),
        dtype=bool
    )

    if len(values) < 3:
        return mask

    differences = np.abs(
        np.diff(values)
    )

    half_window = (
        window_size // 2
    )

    for i in range(
        len(differences)
    ):

        start = max(
            0,
            i - half_window
        )

        end = min(
            len(differences),
            i + half_window + 1
        )

        local_window = (
            differences[start:end]
        )

        valid = local_window[
            np.isfinite(
                local_window
            )
        ]

        if len(valid) < 5:
            continue

        local_median = np.median(
            valid
        )

        local_mad = calculate_mad(
            valid
        )

        if local_mad == 0:
            continue

        robust_scale = (
            1.4826 * local_mad
        )

        score = (
            differences[i]
            - local_median
        ) / robust_scale

        if score > threshold:

            mask[i] = True

            if i + 1 < len(mask):

                mask[i + 1] = True

    return mask


# ============================================================
# PROCESS EACH SIGNAL CHANNEL
# ============================================================

report = []

cleaned_df = df.copy()


print()
print("=" * 80)
print("PROCESSING SIGNAL CHANNELS")
print("=" * 80)


for channel_number, column in enumerate(
    signal_columns,
    start=1
):

    print(
        f"[{channel_number}/"
        f"{len(signal_columns)}] "
        f"{column}"
    )

    original = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    values = original.to_numpy(
        dtype=float
    )

    # --------------------------------------------------------
    # Missing values
    # --------------------------------------------------------

    missing_mask = (
        ~np.isfinite(values)
    )

    # --------------------------------------------------------
    # Zero values
    # --------------------------------------------------------

    zero_mask = (
        np.isfinite(values)
        &
        (values == 0)
    )

    # --------------------------------------------------------
    # Flatline
    # --------------------------------------------------------

    flatline_mask = detect_flatlines(
        values,
        minimum_length=FLATLINE_LENGTH
    )

    # --------------------------------------------------------
    # Hampel outliers
    # --------------------------------------------------------

    outlier_mask = detect_hampel_outliers(
        values,
        window_size=WINDOW_SIZE,
        threshold=HAMPEL_THRESHOLD
    )

    # --------------------------------------------------------
    # Abrupt changes
    # --------------------------------------------------------

    abrupt_mask = detect_abrupt_changes(
        values,
        window_size=WINDOW_SIZE,
        threshold=ABRUPT_THRESHOLD
    )

    # --------------------------------------------------------
    # Combined repair mask
    # --------------------------------------------------------

    repair_mask = (
        missing_mask
        |
        flatline_mask
        |
        outlier_mask
        |
        abrupt_mask
    )

    # --------------------------------------------------------
    # Mark bad observations as missing
    # --------------------------------------------------------

    repaired = values.copy()

    repaired[
        repair_mask
    ] = np.nan

    repaired_series = pd.Series(
        repaired
    )

    # --------------------------------------------------------
    # Local interpolation
    # --------------------------------------------------------

    repaired_series = (
        repaired_series
        .interpolate(
            method="linear",
            limit=INTERPOLATION_LIMIT,
            limit_direction="both"
        )
    )

    # --------------------------------------------------------
    # Remaining missing values
    # --------------------------------------------------------

    if repaired_series.isna().any():

        rolling_median = (
            repaired_series
            .rolling(
                window=WINDOW_SIZE,
                center=True,
                min_periods=1
            )
            .median()
        )

        repaired_series = (
            repaired_series
            .fillna(
                rolling_median
            )
        )

    # --------------------------------------------------------
    # Store cleaned signal
    # --------------------------------------------------------

    cleaned_df[column] = (
        repaired_series.to_numpy()
    )

    # --------------------------------------------------------
    # Report
    # --------------------------------------------------------

    report.append({

        "Channel":
            column,

        "Total_Points":
            len(values),

        "Missing_Before":
            int(
                missing_mask.sum()
            ),

        "Zero_Values":
            int(
                zero_mask.sum()
            ),

        "Flatline_Points":
            int(
                flatline_mask.sum()
            ),

        "Hampel_Outliers":
            int(
                outlier_mask.sum()
            ),

        "Abrupt_Changes":
            int(
                abrupt_mask.sum()
            ),

        "Total_Repaired":
            int(
                repair_mask.sum()
            ),

        "Missing_After":
            int(
                cleaned_df[column]
                .isna()
                .sum()
            )
    })


# ============================================================
# SAVE CLEANED DATASET
# ============================================================

print()
print("Saving cleaned dataset...")

cleaned_df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ============================================================
# SAVE REPORT
# ============================================================

report_df = pd.DataFrame(
    report
)

REPORT_FILE = (
    REPORT_FOLDER /
    "V22_signal_quality_report.csv"
)

report_df.to_csv(
    REPORT_FILE,
    index=False
)


# ============================================================
# SAVE SUMMARY
# ============================================================

summary_df = pd.DataFrame({

    "Metric": [

        "Total rows",

        "Total columns",

        "Signal channels",

        "Missing values before",

        "Zero values",

        "Flatline points",

        "Hampel outliers",

        "Abrupt changes",

        "Total repaired values",

        "Missing values after"
    ],

    "Value": [

        len(cleaned_df),

        len(cleaned_df.columns),

        len(signal_columns),

        report_df[
            "Missing_Before"
        ].sum(),

        report_df[
            "Zero_Values"
        ].sum(),

        report_df[
            "Flatline_Points"
        ].sum(),

        report_df[
            "Hampel_Outliers"
        ].sum(),

        report_df[
            "Abrupt_Changes"
        ].sum(),

        report_df[
            "Total_Repaired"
        ].sum(),

        report_df[
            "Missing_After"
        ].sum()
    ]
})

SUMMARY_FILE = (
    REPORT_FOLDER /
    "V22_cleaning_summary.csv"
)

summary_df.to_csv(
    SUMMARY_FILE,
    index=False
)


# ============================================================
# FINAL OUTPUT
# ============================================================

print()
print("=" * 80)
print("V22 SIGNAL QUALITY CLEANING COMPLETED")
print("=" * 80)

print()
print("Input:")
print(INPUT_FILE)

print()
print("Output:")
print(OUTPUT_FILE)

print()
print("Detailed report:")
print(REPORT_FILE)

print()
print("Summary:")
print(SUMMARY_FILE)

print()
print("Rows:")
print(len(cleaned_df))

print()
print("Columns:")
print(len(cleaned_df.columns))

print()
print("Signal channels:")
print(len(signal_columns))

print()
print(
    "Total repaired values:",
    int(
        report_df[
            "Total_Repaired"
        ].sum()
    )
)

print()
print(
    "Missing values after cleaning:",
    int(
        report_df[
            "Missing_After"
        ].sum()
    )
)

print()
print("=" * 80)
```

```python
# =============================================================================
# V21 TO V23 CONSERVATIVE SIGNAL QUALITY CLEANING
# Arch Linux
# =============================================================================

import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")


# =============================================================================
# PATH CONFIGURATION
# =============================================================================

INPUT_PATH = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V21/"
    "V21_Merged_All_Classes_All_Persons.csv"
)

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V23"
)

OUTPUT_FILE = (
    OUTPUT_ROOT /
    "V23_Cleaned_All_Classes_All_Persons.csv"
)

REPORT_DIR = (
    OUTPUT_ROOT /
    "reports"
)

VIS_DIR = (
    OUTPUT_ROOT /
    "visualizations"
)


# =============================================================================
# CREATE OUTPUT DIRECTORIES
# =============================================================================

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

VIS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# CLEANING PARAMETERS
# =============================================================================

# Maximum consecutive problematic points that can be interpolated.
# Longer gaps remain missing.
MAX_INTERPOLATION_GAP = 10

# Minimum number of consecutive equal values for flatline detection.
FLATLINE_MIN_LENGTH = 20

# Hampel threshold.
# Higher value means more conservative outlier detection.
HAMPEL_THRESHOLD = 6.0

# Abrupt change threshold.
# Higher value means more conservative detection.
ABRUPT_THRESHOLD = 8.0

# Rolling window.
ROLLING_WINDOW = 21

# Minimum consecutive zeros before zero values are considered problematic.
ZERO_RUN_MIN_LENGTH = 3


# =============================================================================
# SIGNAL GROUP DEFINITIONS
# =============================================================================

BAND_PREFIXES = (
    "Delta_",
    "Theta_",
    "Alpha_",
    "Beta_",
    "Gamma_"
)

RAW_PREFIX = "RAW_"

MOTION_PREFIXES = (
    "AUX_",
    "Accelerometer",
    "Gyro_"
)


# =============================================================================
# SIGNAL CLASSIFICATION
# =============================================================================

def classify_signal_column(column_name):

    name = str(column_name)

    if name.startswith(BAND_PREFIXES):
        return "band_power"

    if name.startswith(RAW_PREFIX):
        return "raw_eeg"

    if name.startswith(MOTION_PREFIXES):
        return "motion"

    return "other"


def get_signal_columns(df):

    excluded_columns = {
        "TimeStamp",
        "Activity",
        "PersonSrial",
        "PersonSerial",
        "Class"
    }

    signal_columns = []

    for column in df.columns:

        if column in excluded_columns:
            continue

        if not pd.api.types.is_numeric_dtype(
            df[column]
        ):
            continue

        signal_type = classify_signal_column(
            column
        )

        if signal_type in {
            "raw_eeg",
            "band_power",
            "motion"
        }:

            signal_columns.append(column)

    return signal_columns


# =============================================================================
# HAMPEL OUTLIER DETECTION
# =============================================================================

def detect_hampel_outliers(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    rolling_median = (
        values
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=5
        )
        .median()
    )

    absolute_deviation = (
        values -
        rolling_median
    ).abs()

    rolling_mad = (
        absolute_deviation
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=5
        )
        .median()
    )

    robust_sigma = (
        1.4826 *
        rolling_mad
    )

    threshold = (
        HAMPEL_THRESHOLD *
        robust_sigma
    )

    outliers = (
        absolute_deviation >
        threshold
    )

    return outliers.fillna(False)


# =============================================================================
# ABRUPT CHANGE DETECTION
# =============================================================================

def detect_abrupt_changes(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    difference = (
        values.diff().abs()
    )

    rolling_median = (
        difference
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=5
        )
        .median()
    )

    deviation = (
        difference -
        rolling_median
    ).abs()

    rolling_mad = (
        deviation
        .rolling(
            window=ROLLING_WINDOW,
            center=True,
            min_periods=5
        )
        .median()
    )

    robust_sigma = (
        1.4826 *
        rolling_mad
    )

    threshold = (
        ABRUPT_THRESHOLD *
        robust_sigma
    )

    abrupt = (
        difference >
        rolling_median +
        threshold
    )

    return abrupt.fillna(False)


# =============================================================================
# FLATLINE DETECTION
# =============================================================================

def detect_flatline(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    same_as_previous = (
        values.eq(
            values.shift()
        )
    )

    groups = (
        same_as_previous
        .ne(
            same_as_previous.shift()
        )
        .cumsum()
    )

    run_lengths = (
        same_as_previous
        .groupby(groups)
        .transform("sum")
    )

    flatline = (
        same_as_previous &
        (
            run_lengths >=
            FLATLINE_MIN_LENGTH
        )
    )

    return flatline.fillna(False)


# =============================================================================
# ZERO RUN DETECTION
# =============================================================================

def detect_zero_runs(series):

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    zero = values.eq(0)

    groups = (
        zero
        .ne(
            zero.shift()
        )
        .cumsum()
    )

    run_lengths = (
        zero
        .groupby(groups)
        .transform("sum")
    )

    zero_problem = (
        zero &
        (
            run_lengths >=
            ZERO_RUN_MIN_LENGTH
        )
    )

    return zero_problem.fillna(False)


# =============================================================================
# SHORT GAP INTERPOLATION
# =============================================================================

def interpolate_short_gaps(
    series,
    invalid_mask
):

    result = pd.to_numeric(
        series,
        errors="coerce"
    ).copy()

    original_values = result.copy()

    result.loc[
        invalid_mask
    ] = np.nan

    result = (
        result
        .interpolate(
            method="linear",
            limit=MAX_INTERPOLATION_GAP,
            limit_direction="both",
            limit_area="inside"
        )
    )

    untouched = ~invalid_mask

    result.loc[
        untouched
    ] = original_values.loc[
        untouched
    ]

    return result


# =============================================================================
# START
# =============================================================================

print()
print("=" * 90)
print("V21 TO V23 CONSERVATIVE SIGNAL QUALITY CLEANING")
print("=" * 90)

print()
print("Input file:")
print(INPUT_PATH)

print()
print("Output directory:")
print(OUTPUT_ROOT)


# =============================================================================
# CHECK INPUT
# =============================================================================

if not INPUT_PATH.exists():

    raise FileNotFoundError(
        f"\nInput file was not found:\n{INPUT_PATH}"
    )


# =============================================================================
# LOAD DATA
# =============================================================================

print()
print("=" * 90)
print("READING V21 DATASET")
print("=" * 90)

df = pd.read_csv(
    INPUT_PATH
)

print()
print(
    "Dataset loaded successfully."
)

print(
    "Rows    :",
    len(df)
)

print(
    "Columns :",
    len(df.columns)
)


# =============================================================================
# IDENTIFY SIGNAL COLUMNS
# =============================================================================

signal_columns = get_signal_columns(
    df
)

raw_columns = [
    column
    for column in signal_columns
    if classify_signal_column(column)
    == "raw_eeg"
]

band_columns = [
    column
    for column in signal_columns
    if classify_signal_column(column)
    == "band_power"
]

motion_columns = [
    column
    for column in signal_columns
    if classify_signal_column(column)
    == "motion"
]


# =============================================================================
# DISPLAY SIGNAL GROUPS
# =============================================================================

print()
print("=" * 90)
print("SIGNAL GROUPS")
print("=" * 90)

print()
print(
    "RAW EEG channels:",
    len(raw_columns)
)

print(
    raw_columns
)

print()
print(
    "Band power channels:",
    len(band_columns)
)

print(
    band_columns
)

print()
print(
    "Motion channels:",
    len(motion_columns)
)

print(
    motion_columns
)


# =============================================================================
# CHECK PERSON COLUMN
# =============================================================================

if "PersonSrial" in df.columns:

    PERSON_COLUMN = "PersonSrial"

elif "PersonSerial" in df.columns:

    PERSON_COLUMN = "PersonSerial"

else:

    raise ValueError(
        "PersonSrial or PersonSerial column was not found."
    )


# =============================================================================
# CHECK CLASS COLUMN
# =============================================================================

if "Class" in df.columns:

    CLASS_COLUMN = "Class"

else:

    raise ValueError(
        "Class column was not found."
    )


# =============================================================================
# GROUP INFORMATION
# =============================================================================

print()
print("=" * 90)
print("GROUPING INFORMATION")
print("=" * 90)

print()
print(
    "Person column:",
    PERSON_COLUMN
)

print(
    "Class column:",
    CLASS_COLUMN
)

print(
    "Number of persons:",
    df[PERSON_COLUMN].nunique()
)

print(
    "Number of classes:",
    df[CLASS_COLUMN].nunique()
)


# =============================================================================
# CONVERT SIGNAL COLUMNS TO NUMERIC
# =============================================================================

for column in signal_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )


# =============================================================================
# REPORT STORAGE
# =============================================================================

quality_records = []

total_missing_before = 0
total_zero_runs = 0
total_hampel = 0
total_abrupt = 0
total_flatline = 0
total_repaired = 0
total_missing_after = 0


# =============================================================================
# GROUP DATA
#
# Each person and class is processed separately.
# Therefore interpolation cannot cross between people or activities.
# =============================================================================

grouped = df.groupby(
    [
        CLASS_COLUMN,
        PERSON_COLUMN
    ],
    sort=False
)

total_groups = len(grouped)

print()
print("=" * 90)
print("STARTING SIGNAL QUALITY CLEANING")
print("=" * 90)

print()
print(
    "Total person and class groups:",
    total_groups
)


# =============================================================================
# PROCESS EVERY PERSON AND CLASS
# =============================================================================

for group_number, (
    group_key,
    group_data
) in enumerate(
    grouped,
    start=1
):

    class_name = group_key[0]
    person_number = group_key[1]

    print(
        f"\nProcessing "
        f"{group_number}/{total_groups}: "
        f"{class_name}, "
        f"Person {person_number}"
    )

    group_index = group_data.index

    for column in signal_columns:

        series = df.loc[
            group_index,
            column
        ].copy()

        signal_type = classify_signal_column(
            column
        )

        # -------------------------------------------------------------
        # MISSING VALUES BEFORE CLEANING
        # -------------------------------------------------------------

        missing_before = int(
            series.isna().sum()
        )

        # -------------------------------------------------------------
        # HAMPEL OUTLIERS
        # -------------------------------------------------------------

        hampel_mask = (
            detect_hampel_outliers(
                series
            )
        )

        hampel_count = int(
            hampel_mask.sum()
        )

        # -------------------------------------------------------------
        # ABRUPT CHANGES
        # -------------------------------------------------------------

        abrupt_mask = (
            detect_abrupt_changes(
                series
            )
        )

        abrupt_count = int(
            abrupt_mask.sum()
        )

        # -------------------------------------------------------------
        # FLATLINE
        # -------------------------------------------------------------

        flatline_mask = (
            detect_flatline(
                series
            )
        )

        flatline_count = int(
            flatline_mask.sum()
        )

        # -------------------------------------------------------------
        # ZERO RUNS
        #
        # Zero is treated as potentially problematic for RAW EEG
        # and band power.
        #
        # Zero is retained for motion channels because zero can
        # represent legitimate absence of movement.
        # -------------------------------------------------------------

        if signal_type in {
            "raw_eeg",
            "band_power"
        }:

            zero_mask = (
                detect_zero_runs(
                    series
                )
            )

        else:

            zero_mask = pd.Series(
                False,
                index=series.index
            )

        zero_count = int(
            zero_mask.sum()
        )

        # -------------------------------------------------------------
        # COMBINE QUALITY PROBLEMS
        # -------------------------------------------------------------

        invalid_mask = (
            hampel_mask |
            abrupt_mask |
            flatline_mask |
            zero_mask |
            series.isna()
        )

        total_detected = int(
            invalid_mask.sum()
        )

        # -------------------------------------------------------------
        # REPAIR ONLY SHORT GAPS
        # -------------------------------------------------------------

        cleaned = (
            interpolate_short_gaps(
                series,
                invalid_mask
            )
        )

        # -------------------------------------------------------------
        # COUNT REPAIRED VALUES
        # -------------------------------------------------------------

        changed_mask = (
            cleaned.ne(series)
        )

        changed_mask = (
            changed_mask &
            cleaned.notna()
        )

        repaired_count = int(
            changed_mask.sum()
        )

        # -------------------------------------------------------------
        # MISSING VALUES AFTER CLEANING
        # -------------------------------------------------------------

        missing_after = int(
            cleaned.isna().sum()
        )

        # -------------------------------------------------------------
        # WRITE CLEANED VALUES
        # -------------------------------------------------------------

        df.loc[
            group_index,
            column
        ] = cleaned.values

        # -------------------------------------------------------------
        # SAVE QUALITY RECORD
        # -------------------------------------------------------------

        quality_records.append({

            "Class": class_name,

            "PersonSrial": person_number,

            "Channel": column,

            "Signal_Type": signal_type,

            "Total_Points": len(series),

            "Missing_Before": missing_before,

            "Zero_Run_Points": zero_count,

            "Hampel_Outliers": hampel_count,

            "Abrupt_Changes": abrupt_count,

            "Flatline_Points": flatline_count,

            "Total_Detected": total_detected,

            "Total_Repaired": repaired_count,

            "Missing_After": missing_after

        })

        total_missing_before += (
            missing_before
        )

        total_zero_runs += (
            zero_count
        )

        total_hampel += (
            hampel_count
        )

        total_abrupt += (
            abrupt_count
        )

        total_flatline += (
            flatline_count
        )

        total_repaired += (
            repaired_count
        )

        total_missing_after += (
            missing_after
        )


# =============================================================================
# CREATE QUALITY DATAFRAME
# =============================================================================

quality_df = pd.DataFrame(
    quality_records
)


# =============================================================================
# DETAILED QUALITY REPORT
# =============================================================================

quality_file = (
    REPORT_DIR /
    "V23_Detailed_Quality_Report.csv"
)

quality_df.to_csv(
    quality_file,
    index=False
)


# =============================================================================
# CHANNEL LEVEL SUMMARY
# =============================================================================

channel_summary = (
    quality_df
    .groupby(
        "Channel",
        as_index=False
    )
    .agg({

        "Total_Points": "sum",

        "Missing_Before": "sum",

        "Zero_Run_Points": "sum",

        "Hampel_Outliers": "sum",

        "Abrupt_Changes": "sum",

        "Flatline_Points": "sum",

        "Total_Repaired": "sum",

        "Missing_After": "sum"

    })
)


channel_summary_file = (
    REPORT_DIR /
    "V23_Channel_Quality_Summary.csv"
)

channel_summary.to_csv(
    channel_summary_file,
    index=False
)


# =============================================================================
# SIGNAL TYPE SUMMARY
# =============================================================================

signal_summary = (
    quality_df
    .groupby(
        "Signal_Type",
        as_index=False
    )
    .agg({

        "Total_Points": "sum",

        "Missing_Before": "sum",

        "Zero_Run_Points": "sum",

        "Hampel_Outliers": "sum",

        "Abrupt_Changes": "sum",

        "Flatline_Points": "sum",

        "Total_Repaired": "sum",

        "Missing_After": "sum"

    })
)


signal_summary_file = (
    REPORT_DIR /
    "V23_Signal_Type_Summary.csv"
)

signal_summary.to_csv(
    signal_summary_file,
    index=False
)


# =============================================================================
# OVERALL SUMMARY
# =============================================================================

total_signal_points = (
    len(df) *
    len(signal_columns)
)


overall_summary = pd.DataFrame({

    "Metric": [

        "Total rows",

        "Total columns",

        "Total signal channels",

        "Total signal points",

        "Missing before",

        "Zero run points",

        "Hampel outliers",

        "Abrupt changes",

        "Flatline points",

        "Total repaired",

        "Missing after"

    ],

    "Value": [

        len(df),

        len(df.columns),

        len(signal_columns),

        total_signal_points,

        total_missing_before,

        total_zero_runs,

        total_hampel,

        total_abrupt,

        total_flatline,

        total_repaired,

        total_missing_after

    ]

})


overall_file = (
    REPORT_DIR /
    "V23_Overall_Quality_Summary.csv"
)

overall_summary.to_csv(
    overall_file,
    index=False
)


# =============================================================================
# REMAINING MISSING VALUES BY CHANNEL
# =============================================================================

missing_after_summary = (
    df[signal_columns]
    .isna()
    .sum()
    .reset_index()
)

missing_after_summary.columns = [
    "Channel",
    "Missing_After"
]

missing_after_summary[
    "Missing_Percentage"
] = (
    missing_after_summary[
        "Missing_After"
    ]
    /
    len(df)
    *
    100
)


missing_after_file = (
    REPORT_DIR /
    "V23_Missing_After_Cleaning.csv"
)

missing_after_summary.to_csv(
    missing_after_file,
    index=False
)


# =============================================================================
# VISUALIZATION 1
# SIGNAL QUALITY PROBLEMS
# =============================================================================

plot_df = channel_summary.copy()

x = np.arange(
    len(plot_df)
)

width = 0.18


plt.figure(
    figsize=(20, 9)
)

plt.bar(
    x - 1.5 * width,
    plot_df[
        "Zero_Run_Points"
    ],
    width,
    label="Zero runs"
)

plt.bar(
    x - 0.5 * width,
    plot_df[
        "Flatline_Points"
    ],
    width,
    label="Flatline"
)

plt.bar(
    x + 0.5 * width,
    plot_df[
        "Hampel_Outliers"
    ],
    width,
    label="Hampel outliers"
)

plt.bar(
    x + 1.5 * width,
    plot_df[
        "Abrupt_Changes"
    ],
    width,
    label="Abrupt changes"
)

plt.xticks(
    x,
    plot_df["Channel"],
    rotation=60,
    ha="right"
)

plt.ylabel(
    "Number of detected points"
)

plt.xlabel(
    "Signal channel"
)

plt.title(
    "V23 Signal Quality Problems Before Repair"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    VIS_DIR /
    "V23_Signal_Quality_Problems.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# =============================================================================
# VISUALIZATION 2
# REMAINING MISSING VALUES
# =============================================================================

plt.figure(
    figsize=(20, 8)
)

plt.bar(
    missing_after_summary[
        "Channel"
    ],
    missing_after_summary[
        "Missing_Percentage"
    ]
)

plt.xticks(
    rotation=60,
    ha="right"
)

plt.ylabel(
    "Missing values percentage"
)

plt.xlabel(
    "Signal channel"
)

plt.title(
    "V23 Remaining Missing Data After Conservative Cleaning"
)

plt.tight_layout()

plt.savefig(
    VIS_DIR /
    "V23_Remaining_Missing_Data.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# =============================================================================
# VISUALIZATION 3
# REPAIRED POINTS
# =============================================================================

plt.figure(
    figsize=(20, 8)
)

plt.bar(
    channel_summary[
        "Channel"
    ],
    channel_summary[
        "Total_Repaired"
    ]
)

plt.xticks(
    rotation=60,
    ha="right"
)

plt.ylabel(
    "Number of repaired points"
)

plt.xlabel(
    "Signal channel"
)

plt.title(
    "V23 Repaired Points by Signal Channel"
)

plt.tight_layout()

plt.savefig(
    VIS_DIR /
    "V23_Repaired_Points.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# =============================================================================
# SAVE V23 DATASET
# =============================================================================

print()
print("=" * 90)
print("SAVING V23 DATASET")
print("=" * 90)

df.to_csv(
    OUTPUT_FILE,
    index=False
)


# =============================================================================
# FINAL SUMMARY
# =============================================================================

print()
print("=" * 90)
print("V23 CLEANING COMPLETED")
print("=" * 90)

print()

print(
    "Input file:"
)

print(
    INPUT_PATH
)

print()

print(
    "Output file:"
)

print(
    OUTPUT_FILE
)

print()

print(
    "Report directory:"
)

print(
    REPORT_DIR
)

print()

print(
    "Visualization directory:"
)

print(
    VIS_DIR
)

print()
print("-" * 90)

print(
    f"Total rows              : "
    f"{len(df):,}"
)

print(
    f"Total columns           : "
    f"{len(df.columns):,}"
)

print(
    f"Signal channels         : "
    f"{len(signal_columns):,}"
)

print(
    f"Total signal points     : "
    f"{total_signal_points:,}"
)

print(
    f"Missing before          : "
    f"{total_missing_before:,}"
)

print(
    f"Zero run points         : "
    f"{total_zero_runs:,}"
)

print(
    f"Hampel outliers         : "
    f"{total_hampel:,}"
)

print(
    f"Abrupt changes          : "
    f"{total_abrupt:,}"
)

print(
    f"Flatline points         : "
    f"{total_flatline:,}"
)

print(
    f"Total repaired          : "
    f"{total_repaired:,}"
)

print(
    f"Missing after           : "
    f"{total_missing_after:,}"
)


if total_signal_points > 0:

    repaired_percentage = (
        total_repaired /
        total_signal_points *
        100
    )

    missing_percentage = (
        total_missing_after /
        total_signal_points *
        100
    )

    print()

    print(
        f"Repaired percentage     : "
        f"{repaired_percentage:.4f}%"
    )

    print(
        f"Missing after percentage: "
        f"{missing_percentage:.4f}%"
    )


# =============================================================================
# FINAL INFORMATION
# =============================================================================

print()
print("=" * 90)
print("CLEANING RULES USED")
print("=" * 90)

print()
print(
    "RAW EEG: Hampel detection, abrupt change detection, "
    "flatline detection and zero run detection."
)

print(
    "Band power: Hampel detection, abrupt change detection, "
    "flatline detection and zero run detection."
)

print(
    "Motion: Hampel detection, abrupt change detection and "
    "flatline detection. Zero values are retained."
)

print(
    "Only short problematic gaps are interpolated."
)

print(
    "Maximum interpolation length:",
    MAX_INTERPOLATION_GAP,
    "points."
)

print(
    "Long problematic gaps remain missing."
)

print(
    "Each person and activity class is processed separately."
)

print()
print("=" * 90)
print("V23 READY FOR QUALITY REVIEW")
print("=" * 90)
```

```python
# ================================================================
# V21 TO V23 REVISED
# SIGNAL SPECIFIC CONSERVATIVE CLEANING
# ================================================================
#
# Pipeline:
#
# V21
#   ↓
# Separate RAW EEG, band power and motion
#   ↓
# Check signal specific validity
#   ↓
# Remove only clearly invalid observations
#   ↓
# Conservative outlier detection
#   ↓
# Do NOT automatically treat every constant segment as invalid
#   ↓
# Interpolate only short genuine gaps
#   ↓
# Keep legitimate stable signal segments
#   ↓
# Quality assessment
#   ↓
# V23 REVISED
#
# ================================================================

import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")


# ================================================================
# CONFIGURATION
# ================================================================

INPUT_FILE = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V21/"
    "V21_Merged_All_Classes_All_Persons.csv"
)

OUTPUT_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V23_REVISED"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

REPORT_DIR = OUTPUT_DIR / "reports"
GRAPH_DIR = OUTPUT_DIR / "graphs"

REPORT_DIR.mkdir(parents=True, exist_ok=True)
GRAPH_DIR.mkdir(parents=True, exist_ok=True)


# ================================================================
# CLEANING PARAMETERS
# ================================================================

# Conservative Hampel parameters

HAMPEL_WINDOW = 21

# Larger threshold means fewer points are considered outliers.
# 5 is deliberately conservative.

HAMPEL_THRESHOLD = 5.0

# Minimum number of valid observations required inside
# the local window.

MIN_LOCAL_VALID = 7

# Only short missing gaps are interpolated.
#
# Because your data is approximately 1 second per row,
# MAX_INTERPOLATION_GAP = 5 corresponds approximately
# to a maximum 5 second internal gap.

MAX_INTERPOLATION_GAP = 5


# ================================================================
# LOAD DATA
# ================================================================

print("=" * 80)
print("V21 TO V23 REVISED SIGNAL QUALITY CLEANING")
print("=" * 80)

print()
print("Input file:")
print(INPUT_FILE)

print()
print("Output directory:")
print(OUTPUT_DIR)

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Input file was not found:\n{INPUT_FILE}"
    )

print()
print("Loading dataset...")

df = pd.read_csv(INPUT_FILE)

print("Dataset loaded successfully.")
print("Rows:", len(df))
print("Columns:", len(df.columns))


# ================================================================
# BASIC COLUMN CHECK
# ================================================================

print()
print("=" * 80)
print("COLUMN CHECK")
print("=" * 80)

print(df.columns.tolist())


# ================================================================
# IDENTIFY IMPORTANT COLUMNS
# ================================================================

TIME_COLUMN = "TimeStamp"
ACTIVITY_COLUMN = "Activity"
PERSON_COLUMN = "PersonSrial"
CLASS_COLUMN = "Class"


required_columns = [
    TIME_COLUMN,
    ACTIVITY_COLUMN,
    PERSON_COLUMN,
    CLASS_COLUMN
]

missing_required = [
    col for col in required_columns
    if col not in df.columns
]

if missing_required:
    raise ValueError(
        "Required columns are missing: "
        + ", ".join(missing_required)
    )


# ================================================================
# IDENTIFY SIGNAL CHANNELS
# ================================================================

metadata_columns = {
    TIME_COLUMN,
    ACTIVITY_COLUMN,
    PERSON_COLUMN,
    CLASS_COLUMN
}


signal_columns = [
    col for col in df.columns
    if col not in metadata_columns
]


# ================================================================
# SIGNAL TYPE CLASSIFICATION
# ================================================================

raw_eeg_columns = []
band_power_columns = []
motion_columns = []
other_signal_columns = []


for col in signal_columns:

    name = str(col).strip().lower()

    # RAW EEG

    if name.startswith("raw_"):
        raw_eeg_columns.append(col)

    # BAND POWER

    elif (
        name.startswith("delta_")
        or name.startswith("theta_")
        or name.startswith("alpha_")
        or name.startswith("beta_")
        or name.startswith("gamma_")
    ):
        band_power_columns.append(col)

    # MOTION SENSOR DATA

    elif (
        name.startswith("aux_")
        or name.startswith("accelerometer")
        or name.startswith("gyro")
    ):
        motion_columns.append(col)

    else:
        other_signal_columns.append(col)


print()
print("=" * 80)
print("SIGNAL TYPE DISTRIBUTION")
print("=" * 80)

print("RAW EEG channels       :", len(raw_eeg_columns))
print("Band power channels    :", len(band_power_columns))
print("Motion channels        :", len(motion_columns))
print("Other signal channels  :", len(other_signal_columns))


print()
print("RAW EEG:")
print(raw_eeg_columns)

print()
print("Band power:")
print(band_power_columns)

print()
print("Motion:")
print(motion_columns)


# ================================================================
# CONVERT SIGNAL COLUMNS TO NUMERIC
# ================================================================

all_signal_columns = (
    raw_eeg_columns
    + band_power_columns
    + motion_columns
    + other_signal_columns
)

for col in all_signal_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# ================================================================
# TIME CONVERSION
# ================================================================

print()
print("=" * 80)
print("TIMESTAMP PROCESSING")
print("=" * 80)

original_timestamp = df[TIME_COLUMN].copy()

df[TIME_COLUMN] = pd.to_datetime(
    df[TIME_COLUMN],
    errors="coerce"
)

invalid_timestamp_count = int(
    df[TIME_COLUMN].isna().sum()
)

print(
    "Invalid timestamp values:",
    invalid_timestamp_count
)


# ================================================================
# PRESERVE ORIGINAL ORDER
# ================================================================

df["_Original_Row"] = np.arange(len(df))


# ================================================================
# GROUP INFORMATION
# ================================================================

# The data is merged across people and activities.
#
# Never interpolate between:
#
# Person 1 → Person 2
#
# or
#
# class_lecture → music
#
# Therefore all operations involving neighbouring observations
# are performed separately for each PersonSrial + Class group.

GROUP_COLUMNS = [
    PERSON_COLUMN,
    CLASS_COLUMN
]


# ================================================================
# HELPER FUNCTION
# ================================================================

def hampel_outlier_mask(
    series,
    window_size=21,
    threshold=5.0,
    min_valid=7
):
    """
    Conservative Hampel filter.

    A point is flagged only when:

    1. It is finite.
    2. There are enough valid observations in the local window.
    3. Its deviation from the local median is sufficiently large.

    Constant segments are NOT automatically considered invalid.
    Zero values are NOT automatically considered invalid.
    """

    values = series.to_numpy(dtype=float)

    n = len(values)

    mask = np.zeros(n, dtype=bool)

    half_window = window_size // 2

    for i in range(n):

        current = values[i]

        if not np.isfinite(current):
            continue

        start = max(0, i - half_window)
        end = min(n, i + half_window + 1)

        window = values[start:end]

        valid = window[np.isfinite(window)]

        if len(valid) < min_valid:
            continue

        median = np.median(valid)

        deviations = np.abs(valid - median)

        mad = np.median(deviations)

        # --------------------------------------------------------
        # Constant or nearly constant local window
        # --------------------------------------------------------
        #
        # Do NOT classify it as a flatline.
        #
        # This is important for preserving legitimate stable
        # physiological measurements.

        if mad == 0:

            # Only flag a point if it is clearly different from
            # the local constant value.

            local_scale = np.std(valid)

            if local_scale == 0:

                # No variation exists.
                # Therefore we preserve the point.

                continue

            threshold_value = threshold * local_scale

            if abs(current - median) > threshold_value:
                mask[i] = True

        else:

            robust_z = (
                abs(current - median)
                / (1.4826 * mad)
            )

            if robust_z > threshold:
                mask[i] = True

    return mask


# ================================================================
# INITIAL QUALITY COUNTS
# ================================================================

quality_records = []


# ================================================================
# PROCESS EACH SIGNAL TYPE
# ================================================================

processing_groups = {
    "RAW_EEG": raw_eeg_columns,
    "BAND_POWER": band_power_columns,
    "MOTION": motion_columns,
    "OTHER": other_signal_columns
}


# ================================================================
# STORE ORIGINAL MISSING COUNTS
# ================================================================

original_missing = {}

for col in all_signal_columns:

    original_missing[col] = int(
        df[col].isna().sum()
    )


# ================================================================
# CLEANING
# ================================================================

total_outliers = 0
total_nonfinite = 0
total_interpolated = 0


print()
print("=" * 80)
print("STARTING CONSERVATIVE CLEANING")
print("=" * 80)


for signal_type, columns in processing_groups.items():

    if not columns:
        continue

    print()
    print("-" * 80)
    print("PROCESSING:", signal_type)
    print("-" * 80)

    for col in columns:

        print(
            f"Processing {col} ..."
        )

        # --------------------------------------------------------
        # ORIGINAL DATA
        # --------------------------------------------------------

        original = df[col].copy()

        missing_before = int(
            original.isna().sum()
        )

        # --------------------------------------------------------
        # INVALID NONFINITE VALUES
        # --------------------------------------------------------

        nonfinite_mask = ~np.isfinite(
            original.to_numpy(dtype=float)
        )

        # Existing NaN values should not be counted as newly
        # detected invalid observations.

        original_nan_mask = original.isna().to_numpy()

        new_nonfinite_mask = (
            nonfinite_mask
            & ~original_nan_mask
        )

        nonfinite_count = int(
            new_nonfinite_mask.sum()
        )

        total_nonfinite += nonfinite_count

        # Convert inf and other invalid values to NaN

        df.loc[
            new_nonfinite_mask,
            col
        ] = np.nan

        # --------------------------------------------------------
        # CONSERVATIVE HAMPEL DETECTION
        # --------------------------------------------------------

        outlier_mask_global = np.zeros(
            len(df),
            dtype=bool
        )

        # --------------------------------------------------------
        # PROCESS PERSON + CLASS SEPARATELY
        # --------------------------------------------------------

        for _, group_index in df.groupby(
            GROUP_COLUMNS,
            sort=False,
            dropna=False
        ).groups.items():

            group_index = np.asarray(
                list(group_index)
            )

            group_values = df.loc[
                group_index,
                col
            ].to_numpy(dtype=float)

            group_series = pd.Series(
                group_values,
                index=group_index
            )

            mask = hampel_outlier_mask(
                group_series,
                window_size=HAMPEL_WINDOW,
                threshold=HAMPEL_THRESHOLD,
                min_valid=MIN_LOCAL_VALID
            )

            outlier_mask_global[
                group_index
            ] = mask

        outlier_count = int(
            outlier_mask_global.sum()
        )

        total_outliers += outlier_count

        # --------------------------------------------------------
        # REPLACE ONLY DETECTED OUTLIERS WITH NaN
        # --------------------------------------------------------

        if outlier_count > 0:

            df.loc[
                outlier_mask_global,
                col
            ] = np.nan

        # --------------------------------------------------------
        # SHORT GAP INTERPOLATION
        # --------------------------------------------------------

        interpolation_mask = np.zeros(
            len(df),
            dtype=bool
        )

        # Process every person and class independently

        grouped = df.groupby(
            GROUP_COLUMNS,
            sort=False,
            dropna=False
        )

        for _, group_index in grouped.groups.items():

            group_index = np.asarray(
                list(group_index)
            )

            group_series = df.loc[
                group_index,
                col
            ].copy()

            missing_mask = group_series.isna()

            if not missing_mask.any():
                continue

            # Identify consecutive missing runs

            values = missing_mask.to_numpy()

            start = None

            for i, is_missing in enumerate(
                values
            ):

                if is_missing and start is None:

                    start = i

                elif not is_missing and start is not None:

                    end = i - 1

                    gap_length = (
                        end - start + 1
                    )

                    if (
                        gap_length
                        <= MAX_INTERPOLATION_GAP
                    ):

                        left = start - 1
                        right = end + 1

                        if (
                            left >= 0
                            and right < len(group_series)
                            and pd.notna(
                                group_series.iloc[left]
                            )
                            and pd.notna(
                                group_series.iloc[right]
                            )
                        ):

                            interpolation_mask[
                                group_index[
                                    start:right
                                ]
                            ] = True

                    start = None

            # Handle missing run at the end

            if start is not None:

                end = len(values) - 1

                gap_length = (
                    end - start + 1
                )

                # No extrapolation at the end.
                # Therefore nothing is interpolated.

        # --------------------------------------------------------
        # ACTUAL INTERPOLATION
        # --------------------------------------------------------

        before_interpolation = df[col].copy()

        for _, group_index in df.groupby(
            GROUP_COLUMNS,
            sort=False,
            dropna=False
        ).groups.items():

            group_index = np.asarray(
                list(group_index)
            )

            group_series = df.loc[
                group_index,
                col
            ].copy()

            if not group_series.isna().any():
                continue

            interpolated = (
                group_series
                .interpolate(
                    method="linear",
                    limit=MAX_INTERPOLATION_GAP,
                    limit_area="inside"
                )
            )

            # Only accept interpolated values for positions
            # specifically identified as short gaps.

            for local_position, global_index in enumerate(
                group_index
            ):

                if interpolation_mask[global_index]:

                    old_value = df.at[
                        global_index,
                        col
                    ]

                    new_value = interpolated.iloc[
                        local_position
                    ]

                    if (
                        pd.isna(old_value)
                        and pd.notna(new_value)
                    ):

                        df.at[
                            global_index,
                            col
                        ] = new_value

        after_interpolation = df[col]

        interpolated_count = int(
            (
                before_interpolation.isna()
                & after_interpolation.notna()
            ).sum()
        )

        total_interpolated += (
            interpolated_count
        )

        # --------------------------------------------------------
        # FINAL MISSING
        # --------------------------------------------------------

        missing_after = int(
            df[col].isna().sum()
        )

        total_repaired = (
            nonfinite_count
            + outlier_count
            + interpolated_count
        )

        quality_records.append({
            "Signal_Type": signal_type,
            "Channel": col,
            "Total_Points": len(df),
            "Missing_Before": missing_before,
            "Nonfinite_Detected": nonfinite_count,
            "Hampel_Outliers": outlier_count,
            "Short_Gap_Interpolated": interpolated_count,
            "Total_Repaired": total_repaired,
            "Missing_After": missing_after,
            "Missing_After_Percent": (
                100.0 * missing_after / len(df)
            )
        })


# ================================================================
# RESTORE ORIGINAL ROW ORDER
# ================================================================

df = df.sort_values(
    "_Original_Row"
).reset_index(drop=True)


# ================================================================
# REMOVE INTERNAL COLUMN
# ================================================================

df.drop(
    columns=["_Original_Row"],
    inplace=True
)


# ================================================================
# SAVE CLEANED DATASET
# ================================================================

OUTPUT_FILE = (
    OUTPUT_DIR
    / "V23_Revised_Cleaned_All_Classes_All_Persons.csv"
)

print()
print("=" * 80)
print("SAVING CLEANED DATASET")
print("=" * 80)

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Saved:")
print(OUTPUT_FILE)


# ================================================================
# QUALITY REPORT
# ================================================================

quality_df = pd.DataFrame(
    quality_records
)

QUALITY_REPORT = (
    REPORT_DIR
    / "V23_Revised_Channel_Quality_Summary.csv"
)

quality_df.to_csv(
    QUALITY_REPORT,
    index=False
)


# ================================================================
# OVERALL QUALITY SUMMARY
# ================================================================

total_points = (
    len(df)
    * len(all_signal_columns)
)

missing_before_total = sum(
    original_missing.values()
)

missing_after_total = int(
    df[all_signal_columns]
    .isna()
    .sum()
    .sum()
)

summary = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Total columns",
        "Signal channels",
        "RAW EEG channels",
        "Band power channels",
        "Motion channels",
        "Other signal channels",
        "Total signal points",
        "Missing values before cleaning",
        "Nonfinite values detected",
        "Conservative Hampel outliers",
        "Short gap interpolations",
        "Missing values after cleaning",
        "Missing percentage after cleaning"
    ],
    "Value": [
        len(df),
        len(df.columns),
        len(all_signal_columns),
        len(raw_eeg_columns),
        len(band_power_columns),
        len(motion_columns),
        len(other_signal_columns),
        total_points,
        missing_before_total,
        total_nonfinite,
        total_outliers,
        total_interpolated,
        missing_after_total,
        (
            100.0
            * missing_after_total
            / total_points
            if total_points > 0
            else 0
        )
    ]
})


OVERALL_REPORT = (
    REPORT_DIR
    / "V23_Revised_Overall_Quality_Summary.csv"
)

summary.to_csv(
    OVERALL_REPORT,
    index=False
)


# ================================================================
# SIGNAL TYPE SUMMARY
# ================================================================

signal_type_summary = (
    quality_df
    .groupby("Signal_Type")
    .agg({
        "Total_Points": "sum",
        "Missing_Before": "sum",
        "Nonfinite_Detected": "sum",
        "Hampel_Outliers": "sum",
        "Short_Gap_Interpolated": "sum",
        "Total_Repaired": "sum",
        "Missing_After": "sum"
    })
    .reset_index()
)


signal_type_summary[
    "Missing_After_Percent"
] = (
    100.0
    * signal_type_summary["Missing_After"]
    / signal_type_summary["Total_Points"]
)


SIGNAL_TYPE_REPORT = (
    REPORT_DIR
    / "V23_Revised_Signal_Type_Summary.csv"
)

signal_type_summary.to_csv(
    SIGNAL_TYPE_REPORT,
    index=False
)


# ================================================================
# REMAINING MISSING DATA
# ================================================================

missing_report = pd.DataFrame({
    "Channel": all_signal_columns,
    "Missing_Values": [
        int(df[col].isna().sum())
        for col in all_signal_columns
    ]
})


missing_report[
    "Missing_Percentage"
] = (
    100.0
    * missing_report["Missing_Values"]
    / len(df)
)


MISSING_REPORT = (
    REPORT_DIR
    / "V23_Revised_Remaining_Missing.csv"
)

missing_report.to_csv(
    MISSING_REPORT,
    index=False
)


# ================================================================
# REPAIR REPORT
# ================================================================

repair_report = quality_df[
    [
        "Signal_Type",
        "Channel",
        "Nonfinite_Detected",
        "Hampel_Outliers",
        "Short_Gap_Interpolated",
        "Total_Repaired"
    ]
].copy()


REPAIR_REPORT = (
    REPORT_DIR
    / "V23_Revised_Repaired_Points.csv"
)

repair_report.to_csv(
    REPAIR_REPORT,
    index=False
)


# ================================================================
# GRAPH 1
# REMAINING MISSING VALUES
# ================================================================

plt.figure(
    figsize=(20, 8)
)

plot_missing = missing_report.copy()

plt.bar(
    plot_missing["Channel"],
    plot_missing["Missing_Percentage"]
)

plt.xticks(
    rotation=60,
    ha="right"
)

plt.ylabel(
    "Remaining missing values (%)"
)

plt.xlabel(
    "Signal channel"
)

plt.title(
    "V23 Revised Remaining Missing Values by Signal Channel"
)

plt.tight_layout()

plt.savefig(
    GRAPH_DIR
    / "V23_Revised_Remaining_Missing.png",
    dpi=300
)

plt.close()


# ================================================================
# GRAPH 2
# REPAIRED POINTS
# ================================================================

plt.figure(
    figsize=(20, 8)
)

plt.bar(
    repair_report["Channel"],
    repair_report["Total_Repaired"]
)

plt.xticks(
    rotation=60,
    ha="right"
)

plt.ylabel(
    "Number of repaired points"
)

plt.xlabel(
    "Signal channel"
)

plt.title(
    "V23 Revised Repaired Points by Signal Channel"
)

plt.tight_layout()

plt.savefig(
    GRAPH_DIR
    / "V23_Revised_Repaired_Points.png",
    dpi=300
)

plt.close()


# ================================================================
# GRAPH 3
# OUTLIERS BY SIGNAL TYPE
# ================================================================

plt.figure(
    figsize=(10, 6)
)

outlier_type = (
    quality_df
    .groupby("Signal_Type")[
        "Hampel_Outliers"
    ]
    .sum()
)

plt.bar(
    outlier_type.index,
    outlier_type.values
)

plt.ylabel(
    "Number of conservative outliers"
)

plt.xlabel(
    "Signal type"
)

plt.title(
    "V23 Revised Conservative Outlier Detection"
)

plt.tight_layout()

plt.savefig(
    GRAPH_DIR
    / "V23_Revised_Outliers_By_Signal_Type.png",
    dpi=300
)

plt.close()


# ================================================================
# GRAPH 4
# INTERPOLATED POINTS BY SIGNAL TYPE
# ================================================================

plt.figure(
    figsize=(10, 6)
)

interpolation_type = (
    quality_df
    .groupby("Signal_Type")[
        "Short_Gap_Interpolated"
    ]
    .sum()
)

plt.bar(
    interpolation_type.index,
    interpolation_type.values
)

plt.ylabel(
    "Number of interpolated points"
)

plt.xlabel(
    "Signal type"
)

plt.title(
    "V23 Revised Short Gap Interpolation"
)

plt.tight_layout()

plt.savefig(
    GRAPH_DIR
    / "V23_Revised_Interpolation_By_Signal_Type.png",
    dpi=300
)

plt.close()


# ================================================================
# FINAL VALIDATION
# ================================================================

print()
print("=" * 80)
print("FINAL VALIDATION")
print("=" * 80)

print()
print("Rows:", len(df))
print("Columns:", len(df.columns))

print()
print("RAW EEG channels:", len(raw_eeg_columns))
print("Band power channels:", len(band_power_columns))
print("Motion channels:", len(motion_columns))

print()
print("Missing before cleaning:")
print(missing_before_total)

print()
print("Nonfinite detected:")
print(total_nonfinite)

print()
print("Conservative Hampel outliers:")
print(total_outliers)

print()
print("Short gaps interpolated:")
print(total_interpolated)

print()
print("Missing after cleaning:")
print(missing_after_total)

print()
print(
    "Remaining missing percentage:",
    round(
        100.0
        * missing_after_total
        / total_points,
        4
    ),
    "%"
)


# ================================================================
# CHECK METADATA
# ================================================================

print()
print("=" * 80)
print("METADATA VALIDATION")
print("=" * 80)

for col in required_columns:

    print(
        col,
        "present:",
        col in df.columns
    )


# ================================================================
# CHECK DUPLICATE ROWS
# ================================================================

duplicate_rows = int(
    df.duplicated().sum()
)

print()
print("Duplicate complete rows:", duplicate_rows)


# ================================================================
# FINAL OUTPUT INFORMATION
# ================================================================

print()
print("=" * 80)
print("V23 REVISED CLEANING COMPLETED")
print("=" * 80)

print()
print("Cleaned dataset:")
print(OUTPUT_FILE)

print()
print("Reports:")
print(REPORT_DIR)

print()
print("Graphs:")
print(GRAPH_DIR)

print()
print("Generated files:")

for path in sorted(OUTPUT_DIR.rglob("*")):

    if path.is_file():

        print(
            " ",
            path.relative_to(OUTPUT_DIR)
        )

print()
print("=" * 80)
print("IMPORTANT")
print("=" * 80)

print(
    "Zero values were not automatically removed."
)

print(
    "Constant segments were not automatically classified as flatlines."
)

print(
    "Only conservative local Hampel outliers were converted to missing values."
)

print(
    "Only short internal gaps were interpolated."
)

print(
    "Long missing sections were retained as missing."
)

print(
    "Interpolation was performed separately for each person and activity class."
)

print("=" * 80)
```

```python
# ================================================================
# FINAL DATASET VISUALIZATION AND ARIMA ANALYSIS
# ================================================================
#
# Input:
# V23_Revised_Cleaned_All_Classes_All_Persons.csv
#
# Main outputs:
# 1. Activity Time Series
# 2. ACF
# 3. PACF
# 4. ARIMA Diagnostic Plot
# 5. ARIMA Model Fitting
# 6. Residual Plot
# 7. Residual Density
# 8. Actual vs Fitted
# 9. First 25 Minutes
# 10. Second 25 Minutes
# 11. Third 25 Minutes
#
# Additional journal visualizations:
# 12. Class Distribution
# 13. Activity Distribution
# 14. Feature Correlation Heatmap
# 15. Feature Distribution
# 16. Boxplot by Class
# 17. Mean Feature Profile by Class
# 18. Rolling Mean and Standard Deviation
# 19. QQ Plot of Residuals
# 20. Residual Autocorrelation
# 21. Class Wise Time Series
#
# Additional:
# 22. Temporal Heatmap
# 23. Class Feature Heatmap
# 24. Activity Feature Distribution
# 25. Feature Correlation Cluster
# 26. Class Wise Violin Plot
# 27. Class Wise Mean and Confidence Interval
# 28. Temporal Variability
# 29. ARIMA Model Comparison
# 30. Residual Statistics
# ================================================================


# ================================================================
# IMPORT LIBRARIES
# ================================================================

import os
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

from statsmodels.tsa.stattools import acf
from statsmodels.tsa.stattools import pacf

from statsmodels.tsa.arima.model import ARIMA

from statsmodels.graphics.tsaplots import (
    plot_acf,
    plot_pacf
)

from statsmodels.graphics.gofplots import qqplot

from statsmodels.stats.diagnostic import (
    acorr_ljungbox
)


# ================================================================
# INPUT AND OUTPUT
# ================================================================

INPUT_FILE = (
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V23_REVISED/"
    "V23_Revised_Cleaned_All_Classes_All_Persons.csv"
)

OUTPUT_DIR = (
    "/home/rakib/Desktop/RAW_DATA/"
    "V24_Journal_Visualization"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ================================================================
# PLOT CONFIGURATION
# ================================================================

sns.set_theme(
    style="whitegrid",
    context="paper",
    font_scale=1.15
)

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300

# Prevent large time series from exceeding
# the Matplotlib rendering path limit.

plt.rcParams["agg.path.chunksize"] = 10000

plt.rcParams["path.simplify"] = True

plt.rcParams["path.simplify_threshold"] = 0.5


# ================================================================
# USER CONFIGURATION
# ================================================================

# If None, the program automatically selects
# the first suitable numerical signal feature.

TARGET_FEATURE = None

# Number of features used in correlation plots.

MAX_CORRELATION_FEATURES = 34

# Number of observations used for some
# distribution visualizations.

MAX_DISTRIBUTION_SAMPLE = 100000

# ARIMA candidate models.

ARIMA_ORDERS = [
    (0, 1, 0),
    (1, 1, 0),
    (0, 1, 1),
    (1, 1, 1),
    (2, 1, 0),
    (0, 1, 2),
    (2, 1, 1),
    (1, 1, 2),
    (2, 1, 2)
]


# ================================================================
# LOAD DATASET
# ================================================================

print("=" * 75)
print("FINAL DATASET VISUALIZATION AND ARIMA ANALYSIS")
print("=" * 75)

print("\nInput file:")
print(INPUT_FILE)

if not os.path.exists(INPUT_FILE):

    raise FileNotFoundError(
        "\nInput dataset was not found.\n\n"
        "Please verify this path:\n"
        + INPUT_FILE
    )


df = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

print("\nDataset loaded successfully.")

print(
    "Rows:",
    len(df)
)

print(
    "Columns:",
    len(df.columns)
)


# ================================================================
# COLUMN DETECTION
# ================================================================

def find_column(
    dataframe,
    candidates
):

    lookup = {
        str(column).strip().lower(): column
        for column in dataframe.columns
    }

    for candidate in candidates:

        key = candidate.strip().lower()

        if key in lookup:
            return lookup[key]

    return None


TIME_COL = find_column(
    df,
    [
        "TimeStamp",
        "Timestamp",
        "timestamp",
        "Time",
        "time"
    ]
)

ACTIVITY_COL = find_column(
    df,
    [
        "Activity",
        "activity"
    ]
)

CLASS_COL = find_column(
    df,
    [
        "Class",
        "class"
    ]
)

PERSON_COL = find_column(
    df,
    [
        "PersonSrial",
        "PersonSerial",
        "Person",
        "PersonID",
        "Person_ID"
    ]
)


print("\nDetected columns:")

print(
    "Timestamp:",
    TIME_COL
)

print(
    "Activity:",
    ACTIVITY_COL
)

print(
    "Class:",
    CLASS_COL
)

print(
    "Person:",
    PERSON_COL
)


# ================================================================
# TIMESTAMP PROCESSING
# ================================================================

if TIME_COL is not None:

    df[TIME_COL] = pd.to_datetime(
        df[TIME_COL],
        errors="coerce"
    )

    df = df.dropna(
        subset=[TIME_COL]
    ).copy()


# ================================================================
# NUMERICAL FEATURES
# ================================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()


excluded_columns = []

if PERSON_COL is not None:
    excluded_columns.append(
        PERSON_COL
    )


feature_columns = [
    column
    for column in numeric_columns
    if column not in excluded_columns
]


if len(feature_columns) == 0:

    raise ValueError(
        "No numerical signal features were detected."
    )


# ================================================================
# SELECT TARGET FEATURE
# ================================================================

if (
    TARGET_FEATURE is not None
    and TARGET_FEATURE in feature_columns
):

    TARGET = TARGET_FEATURE

else:

    TARGET = feature_columns[0]


print(
    "\nTarget feature:",
    TARGET
)


# ================================================================
# NUMERICAL CLEANING
# ================================================================

for column in feature_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    df[column] = df[column].replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )


# ================================================================
# CREATE OUTPUT SUBDIRECTORIES
# ================================================================

folders = [
    "01_Activity",
    "02_TimeSeries",
    "03_ACF_PACF",
    "04_ARIMA",
    "05_Residuals",
    "06_Class_Analysis",
    "07_Feature_Analysis",
    "08_Temporal_Analysis"
]

for folder in folders:

    os.makedirs(
        os.path.join(
            OUTPUT_DIR,
            folder
        ),
        exist_ok=True
    )


# ================================================================
# 1. ACTIVITY TIME SERIES
# ================================================================

if (
    TIME_COL is not None
    and ACTIVITY_COL is not None
):

    activity_values = (
        df[ACTIVITY_COL]
        .dropna()
        .unique()
        .tolist()
    )

    activity_mapping = {
        value: index
        for index, value
        in enumerate(activity_values)
    }

    activity_numeric = (
        df[ACTIVITY_COL]
        .map(activity_mapping)
    )

    activity_data = pd.DataFrame({

        "Time": df[TIME_COL],

        "ActivityCode": activity_numeric

    })

    activity_data = (
        activity_data
        .dropna()
        .set_index("Time")
        .resample("1min")
        .mean()
        .dropna()
        .reset_index()
    )

    plt.figure(
        figsize=(17, 7)
    )

    sns.lineplot(
        data=activity_data,
        x="Time",
        y="ActivityCode",
        linewidth=1.5
    )

    plt.title(
        "Activity Pattern Over Time"
    )

    plt.xlabel(
        "Time"
    )

    plt.ylabel(
        "Activity Code"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "01_Activity",
            "01_Activity_Time_Series.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 2. CLASS DISTRIBUTION
# ================================================================

if CLASS_COL is not None:

    class_counts = (
        df[CLASS_COL]
        .value_counts()
        .reset_index()
    )

    class_counts.columns = [
        "Class",
        "Count"
    ]

    plt.figure(
        figsize=(12, 7)
    )

    sns.barplot(
        data=class_counts,
        x="Class",
        y="Count"
    )

    plt.title(
        "Class Distribution"
    )

    plt.xlabel(
        "Class"
    )

    plt.ylabel(
        "Number of Observations"
    )

    plt.xticks(
        rotation=30,
        ha="right"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "06_Class_Analysis",
            "02_Class_Distribution.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 3. ACTIVITY DISTRIBUTION
# ================================================================

if ACTIVITY_COL is not None:

    activity_counts = (
        df[ACTIVITY_COL]
        .value_counts()
        .reset_index()
    )

    activity_counts.columns = [
        "Activity",
        "Count"
    ]

    plt.figure(
        figsize=(12, 7)
    )

    sns.barplot(
        data=activity_counts,
        x="Activity",
        y="Count"
    )

    plt.title(
        "Activity Distribution"
    )

    plt.xlabel(
        "Activity"
    )

    plt.ylabel(
        "Number of Observations"
    )

    plt.xticks(
        rotation=30,
        ha="right"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "01_Activity",
            "03_Activity_Distribution.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 4. CORRELATION HEATMAP
# ================================================================

correlation_features = feature_columns[
    :min(
        MAX_CORRELATION_FEATURES,
        len(feature_columns)
    )
]

correlation_matrix = df[
    correlation_features
].corr()

plt.figure(
    figsize=(18, 15)
)

sns.heatmap(
    correlation_matrix,
    cmap="vlag",
    center=0,
    linewidths=0.2
)

plt.title(
    "Feature Correlation Heatmap"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "07_Feature_Analysis",
        "04_Feature_Correlation_Heatmap.png"
    ),
    bbox_inches="tight"
)

plt.close()

correlation_matrix.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "07_Feature_Analysis",
        "Feature_Correlation_Matrix.csv"
    )
)


# ================================================================
# 5. FEATURE DISTRIBUTION
# ================================================================

distribution_features = feature_columns[
    :min(
        12,
        len(feature_columns)
    )
]

distribution_data = df[
    distribution_features
].copy()

if len(distribution_data) > MAX_DISTRIBUTION_SAMPLE:

    distribution_data = (
        distribution_data
        .sample(
            MAX_DISTRIBUTION_SAMPLE,
            random_state=42
        )
    )


melted_distribution = distribution_data.melt(
    var_name="Feature",
    value_name="Value"
)

plt.figure(
    figsize=(17, 9)
)

sns.violinplot(
    data=melted_distribution,
    x="Feature",
    y="Value",
    inner="quartile",
    cut=0
)

plt.title(
    "Feature Distribution"
)

plt.xlabel(
    "Feature"
)

plt.ylabel(
    "Value"
)

plt.xticks(
    rotation=60
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "07_Feature_Analysis",
        "05_Feature_Distribution.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 6. BOXPLOT BY CLASS
# ================================================================

if CLASS_COL is not None:

    box_features = feature_columns[
        :min(
            8,
            len(feature_columns)
        )
    ]

    box_data = df[
        [
            CLASS_COL
        ]
        + box_features
    ].copy()

    box_melted = box_data.melt(
        id_vars=[CLASS_COL],
        var_name="Feature",
        value_name="Value"
    )

    plt.figure(
        figsize=(18, 9)
    )

    sns.boxplot(
        data=box_melted,
        x="Feature",
        y="Value",
        hue=CLASS_COL,
        showfliers=False
    )

    plt.title(
        "Feature Distribution by Class"
    )

    plt.xlabel(
        "Feature"
    )

    plt.ylabel(
        "Value"
    )

    plt.xticks(
        rotation=60
    )

    plt.legend(
        title="Class",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "06_Class_Analysis",
            "06_Boxplot_By_Class.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 7. MEAN FEATURE PROFILE BY CLASS
# ================================================================

if CLASS_COL is not None:

    profile_features = feature_columns[
        :min(
            20,
            len(feature_columns)
        )
    ]

    class_profile = (
        df
        .groupby(CLASS_COL)[profile_features]
        .mean()
    )

    # Standardize only for visualization.

    profile_standardized = (
        class_profile
        .subtract(
            class_profile.mean(axis=0),
            axis=1
        )
        .divide(
            class_profile.std(axis=0)
            .replace(0, np.nan),
            axis=1
        )
    )

    plt.figure(
        figsize=(19, 9)
    )

    for class_name in profile_standardized.index:

        plt.plot(
            profile_standardized.columns,
            profile_standardized.loc[class_name],
            marker="o",
            linewidth=2,
            label=str(class_name)
        )

    plt.title(
        "Standardized Mean Feature Profile by Class"
    )

    plt.xlabel(
        "Feature"
    )

    plt.ylabel(
        "Standardized Mean"
    )

    plt.xticks(
        rotation=60
    )

    plt.legend(
        title="Class"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "06_Class_Analysis",
            "07_Mean_Feature_Profile_By_Class.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# TARGET TIME SERIES
# ================================================================

target_data = df[
    [
        TIME_COL,
        TARGET
    ]
].copy()

target_data[TARGET] = pd.to_numeric(
    target_data[TARGET],
    errors="coerce"
)

target_data = target_data.dropna()

target_data = target_data.sort_values(
    TIME_COL
)


# ================================================================
# MINUTE LEVEL TARGET SERIES
# ================================================================

minute_series = (
    target_data
    .set_index(TIME_COL)[TARGET]
    .resample("1min")
    .mean()
    .dropna()
)


print(
    "\nMinute level observations:",
    len(minute_series)
)


# ================================================================
# 8. TARGET TIME SERIES
# ================================================================

plt.figure(
    figsize=(18, 7)
)

sns.lineplot(
    x=np.arange(
        len(minute_series)
    ),
    y=minute_series.values,
    linewidth=1.3
)

plt.title(
    f"Minute Level Time Series: {TARGET}"
)

plt.xlabel(
    "Minute"
)

plt.ylabel(
    TARGET
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "02_TimeSeries",
        "08_Target_Time_Series.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 9. ACF
# ================================================================

acf_lags = min(
    60,
    max(
        10,
        len(minute_series) // 4
    )
)

plt.figure(
    figsize=(14, 7)
)

plot_acf(
    minute_series,
    lags=acf_lags,
    alpha=0.05,
    ax=plt.gca()
)

plt.title(
    f"Autocorrelation Function: {TARGET}"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "03_ACF_PACF",
        "09_ACF.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 10. PACF
# ================================================================

pacf_lags = min(
    30,
    max(
        5,
        len(minute_series) // 10
    )
)

plt.figure(
    figsize=(14, 7)
)

plot_pacf(
    minute_series,
    lags=pacf_lags,
    method="ywm",
    alpha=0.05,
    ax=plt.gca()
)

plt.title(
    f"Partial Autocorrelation Function: {TARGET}"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "03_ACF_PACF",
        "10_PACF.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 11. ARIMA MODEL FITTING
# ================================================================

arima_results = []

best_model = None
best_order = None
best_aic = np.inf


for order in ARIMA_ORDERS:

    try:

        model = ARIMA(
            minute_series,
            order=order
        )

        fitted_model = model.fit()

        current_aic = fitted_model.aic

        arima_results.append({

            "p": order[0],

            "d": order[1],

            "q": order[2],

            "AIC": fitted_model.aic,

            "BIC": fitted_model.bic,

            "LogLikelihood": fitted_model.llf

        })

        if current_aic < best_aic:

            best_aic = current_aic

            best_order = order

            best_model = fitted_model

    except Exception as error:

        print(
            "ARIMA model failed:",
            order,
            error
        )


if best_model is None:

    raise RuntimeError(
        "No ARIMA model could be fitted."
    )


arima_comparison = pd.DataFrame(
    arima_results
).sort_values(
    "AIC"
)

arima_comparison.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "04_ARIMA",
        "11_ARIMA_Model_Comparison.csv"
    ),
    index=False
)


print(
    "\nSelected ARIMA order:",
    best_order
)

print(
    "AIC:",
    best_model.aic
)

print(
    "BIC:",
    best_model.bic
)


# ================================================================
# 12. ARIMA DIAGNOSTIC PLOT
# ================================================================

fig = best_model.plot_diagnostics(
    figsize=(16, 11)
)

fig.suptitle(
    f"ARIMA Diagnostic Analysis: {TARGET}",
    fontsize=16
)

fig.tight_layout()

fig.savefig(
    os.path.join(
        OUTPUT_DIR,
        "04_ARIMA",
        "12_ARIMA_Diagnostic_Plot.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close(fig)


# ================================================================
# 13. ARIMA MODEL SUMMARY
# ================================================================

with open(
    os.path.join(
        OUTPUT_DIR,
        "04_ARIMA",
        "13_ARIMA_Model_Summary.txt"
    ),
    "w",
    encoding="utf8"
) as file:

    file.write(
        best_model.summary().as_text()
    )


# ================================================================
# RESIDUALS
# ================================================================

fitted_values = (
    best_model.fittedvalues
)

residuals = (
    minute_series
    .subtract(
        fitted_values
    )
)

residuals = residuals.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
).dropna()


# ================================================================
# 14. RESIDUAL TIME SERIES
# ================================================================

plt.figure(
    figsize=(18, 7)
)

sns.lineplot(
    x=np.arange(
        len(residuals)
    ),
    y=residuals.values,
    linewidth=1
)

plt.axhline(
    0,
    linestyle=":"
)

plt.title(
    "ARIMA Residual Time Series"
)

plt.xlabel(
    "Minute"
)

plt.ylabel(
    "Residual"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "05_Residuals",
        "14_Residual_Time_Series.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 15. RESIDUAL DENSITY
# ================================================================

plt.figure(
    figsize=(12, 7)
)

sns.histplot(
    residuals,
    bins=50,
    kde=True,
    stat="density"
)

plt.axvline(
    residuals.mean(),
    linestyle=":"
)

plt.title(
    "ARIMA Residual Density"
)

plt.xlabel(
    "Residual"
)

plt.ylabel(
    "Density"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "05_Residuals",
        "15_Residual_Density.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 16. ACTUAL VS FITTED
# ================================================================

plt.figure(
    figsize=(18, 7)
)

sns.lineplot(
    x=np.arange(
        len(minute_series)
    ),
    y=minute_series.values,
    label="Observed",
    linewidth=1.5
)

sns.lineplot(
    x=np.arange(
        len(fitted_values)
    ),
    y=fitted_values.values,
    label="ARIMA Fitted",
    linewidth=1.3
)

plt.title(
    f"Observed Versus ARIMA Fitted: {TARGET}"
)

plt.xlabel(
    "Minute"
)

plt.ylabel(
    TARGET
)

plt.legend()

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "04_ARIMA",
        "16_Actual_vs_Fitted.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 17. RESIDUAL QQ PLOT
# ================================================================

fig, ax = plt.subplots(
    figsize=(9, 8)
)

qqplot(
    residuals,
    line="45",
    ax=ax
)

ax.set_title(
    "ARIMA Residual Q Q Plot"
)

fig.tight_layout()

fig.savefig(
    os.path.join(
        OUTPUT_DIR,
        "05_Residuals",
        "17_Residual_QQ_Plot.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close(fig)


# ================================================================
# 18. RESIDUAL AUTOCORRELATION
# ================================================================

plt.figure(
    figsize=(14, 7)
)

plot_acf(
    residuals,
    lags=min(
        40,
        len(residuals) // 4
    ),
    alpha=0.05,
    ax=plt.gca()
)

plt.title(
    "Residual Autocorrelation"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "05_Residuals",
        "18_Residual_Autocorrelation.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 19. ROLLING MEAN AND STANDARD DEVIATION
# ================================================================

rolling_window = 10

rolling_mean = (
    minute_series
    .rolling(
        rolling_window
    )
    .mean()
)

rolling_std = (
    minute_series
    .rolling(
        rolling_window
    )
    .std()
)

rolling_data = pd.DataFrame({

    "Minute": np.arange(
        len(minute_series)
    ),

    "RollingMean": rolling_mean.values,

    "RollingSTD": rolling_std.values

})


plt.figure(
    figsize=(18, 7)
)

sns.lineplot(
    data=rolling_data,
    x="Minute",
    y="RollingMean",
    label="Rolling Mean"
)

sns.lineplot(
    data=rolling_data,
    x="Minute",
    y="RollingSTD",
    label="Rolling Standard Deviation"
)

plt.title(
    f"Rolling Mean and Standard Deviation: {TARGET}"
)

plt.xlabel(
    "Minute"
)

plt.ylabel(
    "Value"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "08_Temporal_Analysis",
        "19_Rolling_Mean_STD.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# CREATE ELAPSED MINUTE FOR EACH PERSON
# ================================================================

if (
    TIME_COL is not None
    and PERSON_COL is not None
):

    temporal_df = df[
        [
            TIME_COL,
            PERSON_COL,
            TARGET
        ]
        + (
            [CLASS_COL]
            if CLASS_COL is not None
            else []
        )
    ].copy()

    temporal_df = temporal_df.dropna(
        subset=[
            TIME_COL,
            PERSON_COL,
            TARGET
        ]
    )

    temporal_df["ElapsedMinute"] = (
        temporal_df
        .groupby(PERSON_COL)[TIME_COL]
        .transform(
            lambda x:
            x.subtract(
                x.min()
            )
            .dt.total_seconds()
            .divide(60)
        )
    )

    temporal_df["Minute"] = (
        temporal_df["ElapsedMinute"]
        .astype(int)
    )


# ================================================================
# 20. FIRST 25 MINUTES
# ================================================================

if "temporal_df" in locals():

    first_25 = temporal_df[
        temporal_df["Minute"] < 25
    ].copy()

    if CLASS_COL is not None:

        first_25_mean = (
            first_25
            .groupby(
                [
                    "Minute",
                    CLASS_COL
                ]
            )[TARGET]
            .mean()
            .reset_index()
        )

        plt.figure(
            figsize=(18, 7)
        )

        sns.lineplot(
            data=first_25_mean,
            x="Minute",
            y=TARGET,
            hue=CLASS_COL,
            linewidth=2
        )

        plt.title(
            f"First 25 Minutes Across Classes: {TARGET}"
        )

        plt.xlabel(
            "Elapsed Minute"
        )

        plt.ylabel(
            TARGET
        )

        plt.legend(
            title="Class"
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                OUTPUT_DIR,
                "08_Temporal_Analysis",
                "20_First_25_Minutes.png"
            ),
            bbox_inches="tight"
        )

        plt.close()


# ================================================================
# 21. SECOND 25 MINUTES
# ================================================================

if "temporal_df" in locals():

    second_25 = temporal_df[
        (
            temporal_df["Minute"] >= 25
        )
        &
        (
            temporal_df["Minute"] < 50
        )
    ].copy()

    if CLASS_COL is not None:

        second_25_mean = (
            second_25
            .groupby(
                [
                    "Minute",
                    CLASS_COL
                ]
            )[TARGET]
            .mean()
            .reset_index()
        )

        plt.figure(
            figsize=(18, 7)
        )

        sns.lineplot(
            data=second_25_mean,
            x="Minute",
            y=TARGET,
            hue=CLASS_COL,
            linewidth=2
        )

        plt.title(
            f"Second 25 Minutes Across Classes: {TARGET}"
        )

        plt.xlabel(
            "Elapsed Minute"
        )

        plt.ylabel(
            TARGET
        )

        plt.legend(
            title="Class"
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                OUTPUT_DIR,
                "08_Temporal_Analysis",
                "21_Second_25_Minutes.png"
            ),
            bbox_inches="tight"
        )

        plt.close()


# ================================================================
# 22. THIRD 25 MINUTES
# ================================================================

if "temporal_df" in locals():

    third_25 = temporal_df[
        (
            temporal_df["Minute"] >= 50
        )
        &
        (
            temporal_df["Minute"] < 75
        )
    ].copy()

    if CLASS_COL is not None:

        third_25_mean = (
            third_25
            .groupby(
                [
                    "Minute",
                    CLASS_COL
                ]
            )[TARGET]
            .mean()
            .reset_index()
        )

        plt.figure(
            figsize=(18, 7)
        )

        sns.lineplot(
            data=third_25_mean,
            x="Minute",
            y=TARGET,
            hue=CLASS_COL,
            linewidth=2
        )

        plt.title(
            f"Third 25 Minutes Across Classes: {TARGET}"
        )

        plt.xlabel(
            "Elapsed Minute"
        )

        plt.ylabel(
            TARGET
        )

        plt.legend(
            title="Class"
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                OUTPUT_DIR,
                "08_Temporal_Analysis",
                "22_Third_25_Minutes.png"
            ),
            bbox_inches="tight"
        )

        plt.close()


# ================================================================
# 23. CLASS WISE TIME SERIES
# ================================================================

if (
    "temporal_df" in locals()
    and CLASS_COL is not None
):

    class_time = (
        temporal_df
        .groupby(
            [
                "Minute",
                CLASS_COL
            ]
        )[TARGET]
        .mean()
        .reset_index()
    )

    plt.figure(
        figsize=(18, 8)
    )

    sns.lineplot(
        data=class_time,
        x="Minute",
        y=TARGET,
        hue=CLASS_COL,
        linewidth=2
    )

    plt.title(
        f"Class Wise Temporal Pattern: {TARGET}"
    )

    plt.xlabel(
        "Elapsed Minute"
    )

    plt.ylabel(
        TARGET
    )

    plt.legend(
        title="Class"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "06_Class_Analysis",
            "23_Class_Wise_Time_Series.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 24. TEMPORAL HEATMAP
# ================================================================

if (
    "temporal_df" in locals()
    and CLASS_COL is not None
):

    temporal_heatmap = (
        temporal_df
        .groupby(
            [
                CLASS_COL,
                "Minute"
            ]
        )[TARGET]
        .mean()
        .unstack(
            level=CLASS_COL
        )
    )

    plt.figure(
        figsize=(18, 8)
    )

    sns.heatmap(
        temporal_heatmap.T,
        cmap="mako",
        cbar_kws={
            "label": TARGET
        }
    )

    plt.title(
        f"Temporal Heatmap Across Classes: {TARGET}"
    )

    plt.xlabel(
        "Elapsed Minute"
    )

    plt.ylabel(
        "Class"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "08_Temporal_Analysis",
            "24_Temporal_Heatmap.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 25. CLASS FEATURE HEATMAP
# ================================================================

if CLASS_COL is not None:

    class_feature_mean = (
        df
        .groupby(CLASS_COL)[
            correlation_features
        ]
        .mean()
    )

    class_feature_z = (
        class_feature_mean
        .subtract(
            class_feature_mean.mean(axis=0),
            axis=1
        )
        .divide(
            class_feature_mean.std(axis=0)
            .replace(0, np.nan),
            axis=1
        )
    )

    plt.figure(
        figsize=(19, 9)
    )

    sns.heatmap(
        class_feature_z,
        cmap="vlag",
        center=0
    )

    plt.title(
        "Standardized Mean Feature Values Across Classes"
    )

    plt.xlabel(
        "Feature"
    )

    plt.ylabel(
        "Class"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "06_Class_Analysis",
            "25_Class_Feature_Heatmap.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 26. CLASS WISE VIOLIN
# ================================================================

if CLASS_COL is not None:

    violin_data = df[
        [
            CLASS_COL,
            TARGET
        ]
    ].dropna()

    if len(violin_data) > MAX_DISTRIBUTION_SAMPLE:

        violin_data = (
            violin_data
            .sample(
                MAX_DISTRIBUTION_SAMPLE,
                random_state=42
            )
        )

    plt.figure(
        figsize=(13, 8)
    )

    sns.violinplot(
        data=violin_data,
        x=CLASS_COL,
        y=TARGET,
        inner="quartile",
        cut=0
    )

    plt.title(
        f"{TARGET} Distribution Across Classes"
    )

    plt.xlabel(
        "Class"
    )

    plt.ylabel(
        TARGET
    )

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "06_Class_Analysis",
            "26_Class_Wise_Violin.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 27. CLASS MEAN AND CONFIDENCE INTERVAL
# ================================================================

if CLASS_COL is not None:

    summary_rows = []

    for class_name, group in df.groupby(
        CLASS_COL
    ):

        values = pd.to_numeric(
            group[TARGET],
            errors="coerce"
        ).dropna()

        if len(values) == 0:
            continue

        mean_value = values.mean()

        standard_error = (
            values.std()
            / np.sqrt(
                len(values)
            )
        )

        confidence = (
            1.96
            * standard_error
        )

        summary_rows.append({

            "Class": class_name,

            "Mean": mean_value,

            "Lower": np.subtract(
                mean_value,
                confidence
            ),

            "Upper": np.add(
                mean_value,
                confidence
            )

        })

    class_summary = pd.DataFrame(
        summary_rows
    )

    plt.figure(
        figsize=(13, 8)
    )

    sns.pointplot(
        data=class_summary,
        x="Class",
        y="Mean",
        errorbar=None
    )

    for index, row in class_summary.iterrows():

        plt.plot(
            [
                index,
                index
            ],
            [
                row["Lower"],
                row["Upper"]
            ],
            linewidth=2
        )

    plt.title(
        f"Class Wise Mean and 95 Percent Confidence Interval: {TARGET}"
    )

    plt.xlabel(
        "Class"
    )

    plt.ylabel(
        "Mean"
    )

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "06_Class_Analysis",
            "27_Class_Mean_Confidence_Interval.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 28. TEMPORAL VARIABILITY
# ================================================================

if "temporal_df" in locals():

    temporal_variability = (
        temporal_df
        .groupby("Minute")[TARGET]
        .agg(
            [
                "mean",
                "std"
            ]
        )
        .reset_index()
    )

    plt.figure(
        figsize=(18, 7)
    )

    sns.lineplot(
        data=temporal_variability,
        x="Minute",
        y="std",
        linewidth=2
    )

    plt.title(
        f"Temporal Variability of {TARGET}"
    )

    plt.xlabel(
        "Elapsed Minute"
    )

    plt.ylabel(
        "Standard Deviation"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "08_Temporal_Analysis",
            "28_Temporal_Variability.png"
        ),
        bbox_inches="tight"
    )

    plt.close()


# ================================================================
# 29. ARIMA MODEL COMPARISON
# ================================================================

plt.figure(
    figsize=(12, 7)
)

sns.barplot(
    data=arima_comparison,
    x="AIC",
    y=arima_comparison.apply(
        lambda row:
        f"({int(row['p'])},{int(row['d'])},{int(row['q'])})",
        axis=1
    )
)

plt.title(
    "ARIMA Model Comparison Using AIC"
)

plt.xlabel(
    "AIC"
)

plt.ylabel(
    "ARIMA Order"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "04_ARIMA",
        "29_ARIMA_Model_Comparison.png"
    ),
    bbox_inches="tight"
)

plt.close()


# ================================================================
# 30. RESIDUAL STATISTICS
# ================================================================

residual_statistics = pd.DataFrame({

    "Metric": [

        "Mean",

        "Standard Deviation",

        "Variance",

        "Minimum",

        "Maximum",

        "Median",

        "Skewness",

        "Kurtosis"

    ],

    "Value": [

        residuals.mean(),

        residuals.std(),

        residuals.var(),

        residuals.min(),

        residuals.max(),

        residuals.median(),

        residuals.skew(),

        residuals.kurtosis()

    ]

})


residual_statistics.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "05_Residuals",
        "30_Residual_Statistics.csv"
    ),
    index=False
)


# ================================================================
# LJUNG BOX TEST
# ================================================================

test_lags = [
    lag
    for lag in [
        10,
        20,
        30
    ]
    if lag < len(residuals)
]


if len(test_lags) > 0:

    ljung_box = acorr_ljungbox(
        residuals,
        lags=test_lags,
        return_df=True
    )

    ljung_box.to_csv(
        os.path.join(
            OUTPUT_DIR,
            "05_Residuals",
            "31_Ljung_Box_Test.csv"
        )
    )


# ================================================================
# SAVE ANALYSIS SUMMARY
# ================================================================

analysis_summary = pd.DataFrame({

    "Metric": [

        "Total rows",

        "Total columns",

        "Numerical features",

        "Target feature",

        "Minute observations",

        "ARIMA order",

        "ARIMA AIC",

        "ARIMA BIC",

        "Residual mean",

        "Residual standard deviation"

    ],

    "Value": [

        len(df),

        len(df.columns),

        len(feature_columns),

        TARGET,

        len(minute_series),

        str(best_order),

        best_model.aic,

        best_model.bic,

        residuals.mean(),

        residuals.std()

    ]

})


analysis_summary.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "Analysis_Summary.csv"
    ),
    index=False
)


# ================================================================
# SAVE CLASS SUMMARY
# ================================================================

if CLASS_COL is not None:

    class_summary_full = (
        df
        .groupby(CLASS_COL)[TARGET]
        .agg(
            [
                "count",
                "mean",
                "std",
                "median",
                "min",
                "max"
            ]
        )
        .reset_index()
    )

    class_summary_full.to_csv(
        os.path.join(
            OUTPUT_DIR,
            "Class_Summary.csv"
        ),
        index=False
    )


# ================================================================
# FINAL OUTPUT
# ================================================================

print("\n")
print("=" * 75)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 75)

print(
    "\nOutput directory:"
)

print(
    OUTPUT_DIR
)

print(
    "\nTarget feature:"
)

print(
    TARGET
)

print(
    "\nSelected ARIMA order:"
)

print(
    best_order
)

print(
    "\nARIMA AIC:"
)

print(
    best_model.aic
)

print(
    "\nARIMA BIC:"
)

print(
    best_model.bic
)

print(
    "\nTotal generated analysis files:"
)

file_count = 0

for root, directories, files in os.walk(
    OUTPUT_DIR
):

    file_count += len(files)

print(
    file_count
)

print(
    "\nAll visualizations and analysis files are saved."
)

print("=" * 75)
```

```python
# ================================================================
# V23 COMPLETE DATASET ANALYSIS
# ================================================================
#
# Input:
# V23_Revised_Cleaned_All_Classes_All_Persons.csv
#
# Stages:
#
# Stage A: Dataset quality analysis
# Stage B: Participant level analysis
# Stage C: Statistical significance testing
# Stage D: Temporal analysis
# Stage E: Elapsed time leakage testing
#
# Visualizations:
#
# Missing values
# Duplicate rows
# Class distribution
# Participant distribution
# Participant x class heatmap
# Recording duration
# Sampling interval
# Sampling frequency
# Window length
# Step size
# Participant observations
# Feature distributions
# Statistical significance
# Kendall W
# Pairwise significance
# Temporal feature patterns
# Early middle late patterns
# Temporal slopes
# Class proportion over time
# Elapsed time distribution
# Elapsed time classifier
# Elapsed time confusion matrix
# Feature versus elapsed time
# Correlation heatmap
# Feature summary
#
# ================================================================


# ================================================================
# IMPORT LIBRARIES
# ================================================================

import os
import re
import json
import warnings
from pathlib import Path
from itertools import combinations

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import friedmanchisquare

from statsmodels.stats.multitest import multipletests

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.model_selection import GroupShuffleSplit

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    confusion_matrix
)


warnings.filterwarnings("ignore")


# ================================================================
# CONFIGURATION
# ================================================================

INPUT_FILE = (
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V23_REVISED/"
    "V23_Revised_Cleaned_All_Classes_All_Persons.csv"
)

OUTPUT_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA/v23_complete_analysis"
)

FIGURE_DIR = OUTPUT_DIR / "visualizations"

TABLE_DIR = OUTPUT_DIR / "tables"

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FIGURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TABLE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# USER CONFIGURATION
# ================================================================

CLASS_ORDER = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]

PARTICIPANT_COLUMN = "PersonSrial"

CLASS_COLUMN = "Class"

TIMESTAMP_COLUMN = "TimeStamp"

WINDOW_LENGTH_COLUMN = None

STEP_SIZE_COLUMN = None

WINDOW_START_COLUMN = None

WINDOW_END_COLUMN = None

WINDOW_ID_COLUMN = None


TEMPORAL_BINS = 15

MIN_VALID_OBSERVATIONS = 5

RANDOM_STATE = 42

MAX_FEATURE_PLOTS = 50

MAX_TEMPORAL_FEATURE_PLOTS = 20

MAX_CORRELATION_FEATURES = 30


# ================================================================
# SEABORN CONFIGURATION
# ================================================================

sns.set_theme(
    style="whitegrid",
    context="notebook"
)

sns.set_palette(
    "deep"
)


# ================================================================
# GENERAL FUNCTIONS
# ================================================================

def print_section(title):

    print("\n")
    print("=" * 80)
    print(title)
    print("=" * 80)


def safe_filename(name):

    return re.sub(
        r"[^A-Za-z0-9_]+",
        "_",
        str(name)
    )


def save_figure(filename):

    path = FIGURE_DIR / filename

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    print("Saved:", path)


def find_column(df, candidates):

    lower_map = {
        str(c).strip().lower(): c
        for c in df.columns
    }

    for candidate in candidates:

        if candidate is None:
            continue

        if candidate in df.columns:
            return candidate

        candidate_lower = str(
            candidate
        ).strip().lower()

        if candidate_lower in lower_map:
            return lower_map[candidate_lower]

    return None


# ================================================================
# COLUMN DETECTION
# ================================================================

def detect_class_column(df):

    if CLASS_COLUMN is not None:

        if CLASS_COLUMN in df.columns:
            return CLASS_COLUMN

    candidates = [
        "Class",
        "class",
        "Activity",
        "activity",
        "Label",
        "label",
        "Condition",
        "condition"
    ]

    column = find_column(
        df,
        candidates
    )

    if column is not None:
        return column

    for column in df.columns:

        values = set(
            df[column]
            .dropna()
            .astype(str)
            .str.strip()
            .str.lower()
            .unique()
        )

        overlap = values.intersection(
            set(CLASS_ORDER)
        )

        if len(overlap) >= 2:
            return column

    return None


def detect_participant_column(df):

    if PARTICIPANT_COLUMN is not None:

        if PARTICIPANT_COLUMN in df.columns:
            return PARTICIPANT_COLUMN

    candidates = [
        "PersonSerial",
        "PersonSerial",
        "Participant",
        "participant",
        "Participant_ID",
        "ParticipantID",
        "Person",
        "person",
        "Person_ID",
        "PersonID",
        "Subject",
        "subject",
        "Subject_ID",
        "SubjectID"
    ]

    return find_column(
        df,
        candidates
    )


def detect_timestamp_column(df):

    if TIMESTAMP_COLUMN is not None:

        if TIMESTAMP_COLUMN in df.columns:
            return TIMESTAMP_COLUMN

    candidates = [
        "TimeStamp",
        "Timestamp",
        "timestamp",
        "Time",
        "time",
        "Datetime",
        "datetime",
        "DateTime",
        "date_time"
    ]

    return find_column(
        df,
        candidates
    )


def detect_window_length_column(df):

    if WINDOW_LENGTH_COLUMN is not None:

        if WINDOW_LENGTH_COLUMN in df.columns:
            return WINDOW_LENGTH_COLUMN

    candidates = [
        "WindowLength",
        "window_length",
        "Window_Length",
        "window",
        "Window"
    ]

    return find_column(
        df,
        candidates
    )


def detect_step_size_column(df):

    if STEP_SIZE_COLUMN is not None:

        if STEP_SIZE_COLUMN in df.columns:
            return STEP_SIZE_COLUMN

    candidates = [
        "StepSize",
        "step_size",
        "Step_Size",
        "Step",
        "step"
    ]

    return find_column(
        df,
        candidates
    )


# ================================================================
# LOAD DATASET
# ================================================================

print_section(
    "LOADING V23 DATASET"
)

input_path = Path(
    INPUT_FILE
)

if not input_path.exists():

    raise FileNotFoundError(
        f"Input file does not exist:\n{INPUT_FILE}"
    )


df = pd.read_csv(
    input_path
)


print(
    "Input file:",
    input_path
)

print(
    "Rows:",
    len(df)
)

print(
    "Columns:",
    len(df.columns)
)


# ================================================================
# PRESERVE ORIGINAL DATA
# ================================================================

df_original = df.copy()


# ================================================================
# DETECT COLUMNS
# ================================================================

print_section(
    "COLUMN DETECTION"
)

class_col = detect_class_column(df)

participant_col = detect_participant_column(df)

timestamp_col = detect_timestamp_column(df)

window_length_col = detect_window_length_column(df)

step_size_col = detect_step_size_column(df)


print(
    "Class column:",
    class_col
)

print(
    "Participant column:",
    participant_col
)

print(
    "Timestamp column:",
    timestamp_col
)

print(
    "Window length column:",
    window_length_col
)

print(
    "Step size column:",
    step_size_col
)


# ================================================================
# STOP IF CLASS IS MISSING
# ================================================================

if class_col is None:

    raise ValueError(
        "Class column could not be detected."
    )


# ================================================================
# PARTICIPANT IDENTIFICATION
# ================================================================

if participant_col is None:

    print()
    print(
        "WARNING: Participant identifier was not detected."
    )

    print(
        "Participant level analysis cannot be considered valid "
        "until a participant identifier is available."
    )

    raise ValueError(
        "Participant column not found. "
        "Set PARTICIPANT_COLUMN to the actual participant column."
    )


# ================================================================
# STANDARDIZE CLASS
# ================================================================

df[class_col] = (
    df[class_col]
    .astype(str)
    .str.strip()
    .str.lower()
)


# ================================================================
# STANDARDIZE PARTICIPANT
# ================================================================

df[participant_col] = (
    df[participant_col]
    .astype(str)
    .str.strip()
)


# ================================================================
# BASIC DATA INFORMATION
# ================================================================

print_section(
    "DATASET INFORMATION"
)

print(
    "Unique participants:",
    df[participant_col].nunique()
)

print(
    "Unique classes:",
    df[class_col].nunique()
)

print(
    "Classes:",
    sorted(
        df[class_col].dropna().unique()
    )
)

print(
    "Timestamp column:",
    timestamp_col
)


# ================================================================
# STAGE A
# DATA QUALITY ANALYSIS
# ================================================================

print_section(
    "STAGE A: DATA QUALITY ANALYSIS"
)


# ================================================================
# A1 MISSING VALUES
# ================================================================

missing_counts = (
    df.isna()
    .sum()
    .sort_values(
        ascending=False
    )
)

missing_percentage = (
    missing_counts
    / len(df)
    * 100
)

missing_report = pd.DataFrame({

    "column":
        missing_counts.index,

    "missing_count":
        missing_counts.values,

    "missing_percentage":
        missing_percentage.values

})


missing_report.to_csv(
    TABLE_DIR /
    "A1_missing_values.csv",
    index=False
)


print(
    "Total missing values:",
    int(
        missing_counts.sum()
    )
)


missing_plot = (
    missing_report[
        missing_report[
            "missing_count"
        ] > 0
    ]
    .head(30)
)


if len(missing_plot) > 0:

    plt.figure(
        figsize=(13, 7)
    )

    sns.barplot(
        data=missing_plot,
        x="column",
        y="missing_count"
    )

    plt.xticks(
        rotation=90
    )

    plt.xlabel(
        "Column"
    )

    plt.ylabel(
        "Missing observations"
    )

    plt.title(
        "Missing Values by Column"
    )

    save_figure(
        "A1_missing_values.png"
    )


# ================================================================
# A2 DUPLICATE ROWS
# ================================================================

duplicate_count = int(
    df.duplicated().sum()
)

duplicate_percentage = (
    duplicate_count
    / len(df)
    * 100
)

duplicate_report = pd.DataFrame({

    "category": [
        "Unique rows",
        "Duplicate rows"
    ],

    "count": [
        len(df) - duplicate_count,
        duplicate_count
    ],

    "percentage": [
        100 - duplicate_percentage,
        duplicate_percentage
    ]

})


duplicate_report.to_csv(
    TABLE_DIR /
    "A2_duplicate_rows.csv",
    index=False
)


print(
    "Duplicate rows:",
    duplicate_count
)


plt.figure(
    figsize=(8, 6)
)

sns.barplot(
    data=duplicate_report,
    x="category",
    y="count"
)

plt.xlabel(
    ""
)

plt.ylabel(
    "Number of rows"
)

plt.title(
    "Unique and Duplicate Rows"
)

save_figure(
    "A2_duplicate_rows.png"
)


# ================================================================
# A3 PARTICIPANT COUNT
# ================================================================

participant_count = (
    df[participant_col]
    .nunique(
        dropna=True
    )
)


participant_report = pd.DataFrame({

    "metric": [
        "Participant count"
    ],

    "value": [
        participant_count
    ]

})


participant_report.to_csv(
    TABLE_DIR /
    "A3_participant_count.csv",
    index=False
)


print(
    "Participant count:",
    participant_count
)


plt.figure(
    figsize=(7, 6)
)

sns.barplot(
    x=["Participants"],
    y=[participant_count]
)

plt.ylabel(
    "Count"
)

plt.title(
    "Total Number of Participants"
)

save_figure(
    "A3_participant_count.png"
)


# ================================================================
# A4 CLASS COUNT
# ================================================================

class_counts = (
    df[class_col]
    .value_counts()
    .reindex(
        CLASS_ORDER
    )
    .fillna(0)
)


class_counts_df = (
    class_counts
    .reset_index()
)

class_counts_df.columns = [
    "class",
    "observations"
]


class_counts_df.to_csv(
    TABLE_DIR /
    "A4_class_counts.csv",
    index=False
)


print(
    class_counts_df
)


plt.figure(
    figsize=(10, 6)
)

sns.barplot(
    data=class_counts_df,
    x="class",
    y="observations"
)

plt.xticks(
    rotation=30
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Number of observations"
)

plt.title(
    "Class Distribution"
)

save_figure(
    "A4_class_distribution.png"
)


# ================================================================
# A5 PARTICIPANT X CLASS
# ================================================================

participant_class = pd.crosstab(
    df[participant_col],
    df[class_col]
)

participant_class = (
    participant_class
    .reindex(
        columns=CLASS_ORDER,
        fill_value=0
)


)

participant_class.to_csv(
    TABLE_DIR /
    "A5_participant_by_class.csv"
)


print(
    "Participant x class table:"
)

print(
    participant_class
)


plt.figure(
    figsize=(
        10,
        max(
            8,
            len(
                participant_class
            ) * 0.35
        )
    )
)

sns.heatmap(
    participant_class,
    annot=True,
    fmt="g",
    cmap="viridis"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Participant"
)

plt.title(
    "Participant x Class Distribution"
)

save_figure(
    "A5_participant_class_heatmap.png"
)


# ================================================================
# PARTICIPANTS WITH ALL CLASSES
# ================================================================

presence = (
    participant_class > 0
)

complete_participants = int(
    presence
    .all(axis=1)
    .sum()
)


print(
    "Participants with all four classes:",
    complete_participants
)


# ================================================================
# A6 RECORDING DURATION
# ================================================================

print_section(
    "RECORDING DURATION"
)


duration_df = pd.DataFrame()


if timestamp_col is not None:

    df["_timestamp_parsed"] = pd.to_datetime(
        df[timestamp_col],
        errors="coerce"
    )

    duration_rows = []

    grouping_columns = [
        participant_col,
        class_col
    ]

    for (
        participant,
        class_name
    ), group in df.groupby(
        grouping_columns,
        observed=True
    ):

        timestamps = (
            group[
                "_timestamp_parsed"
            ]
            .dropna()
            .sort_values()
        )

        if len(timestamps) >= 2:

            start = timestamps.iloc[0]

            end = timestamps.iloc[-1]

            duration_seconds = (
                end - start
            ).total_seconds()

            duration_rows.append({

                "participant":
                    participant,

                "class":
                    class_name,

                "start_time":
                    start,

                "end_time":
                    end,

                "duration_seconds":
                    duration_seconds,

                "duration_minutes":
                    duration_seconds / 60,

                "rows":
                    len(group)

            })


    duration_df = pd.DataFrame(
        duration_rows
    )


    if len(duration_df) > 0:

        duration_df.to_csv(
            TABLE_DIR /
            "A6_recording_duration.csv",
            index=False
        )

        print(
            duration_df[
                "duration_minutes"
            ].describe()
        )


        plt.figure(
            figsize=(11, 7)
        )

        sns.histplot(
            duration_df[
                "duration_minutes"
            ],
            bins=30,
            kde=True
        )

        plt.xlabel(
            "Recording duration, minutes"
        )

        plt.ylabel(
            "Number of recordings"
        )

        plt.title(
            "Recording Duration Distribution"
        )

        save_figure(
            "A6_recording_duration_distribution.png"
        )


        plt.figure(
            figsize=(11, 7)
        )

        sns.boxplot(
            data=duration_df,
            x="class",
            y="duration_minutes"
        )

        plt.xticks(
            rotation=30
        )

        plt.xlabel(
            "Class"
        )

        plt.ylabel(
            "Duration, minutes"
        )

        plt.title(
            "Recording Duration by Class"
        )

        save_figure(
            "A7_recording_duration_by_class.png"
        )


# ================================================================
# A8 SAMPLING INTERVAL
# ================================================================

print_section(
    "SAMPLING INTERVAL"
)


sampling_rows = []


if timestamp_col is not None:

    for (
        participant,
        class_name
    ), group in df.groupby(
        [
            participant_col,
            class_col
        ],
        observed=True
    ):

        timestamps = (
            group[
                "_timestamp_parsed"
            ]
            .dropna()
            .sort_values()
            .drop_duplicates()
        )

        if len(timestamps) > 1:

            intervals = (
                timestamps
                .diff()
                .dt.total_seconds()
                .dropna()
            )

            intervals = intervals[
                intervals > 0
            ]

            if len(intervals) > 0:

                sampling_rows.append({

                    "participant":
                        participant,

                    "class":
                        class_name,

                    "median_interval_seconds":
                        intervals.median(),

                    "mean_interval_seconds":
                        intervals.mean(),

                    "std_interval_seconds":
                        intervals.std(),

                    "minimum_interval_seconds":
                        intervals.min(),

                    "maximum_interval_seconds":
                        intervals.max(),

                    "sampling_frequency_hz":
                        1 / intervals.median(),

                    "unique_intervals":
                        intervals.nunique()

                })


sampling_df = pd.DataFrame(
    sampling_rows
)


if len(sampling_df) > 0:

    sampling_df.to_csv(
        TABLE_DIR /
        "A8_sampling_interval.csv",
        index=False
    )


    print(
        sampling_df[
            [
                "median_interval_seconds",
                "mean_interval_seconds",
                "sampling_frequency_hz"
            ]
        ].describe()
    )


    plt.figure(
        figsize=(11, 7)
    )

    sns.histplot(
        sampling_df[
            "median_interval_seconds"
        ],
        bins=30,
        kde=True
    )

    plt.xlabel(
        "Median sampling interval, seconds"
    )

    plt.ylabel(
        "Number of recordings"
    )

    plt.title(
        "Sampling Interval Distribution"
    )

    save_figure(
        "A8_sampling_interval_distribution.png"
    )


    plt.figure(
        figsize=(11, 7)
    )

    sns.histplot(
        sampling_df[
            "sampling_frequency_hz"
        ],
        bins=30,
        kde=True
    )

    plt.xlabel(
        "Estimated sampling frequency, Hz"
    )

    plt.ylabel(
        "Number of recordings"
    )

    plt.title(
        "Sampling Frequency Distribution"
    )

    save_figure(
        "A9_sampling_frequency_distribution.png"
    )


# ================================================================
# A10 WINDOW LENGTH
# ================================================================

print_section(
    "WINDOW LENGTH"
)


window_values = None


if window_length_col is not None:

    window_values = pd.to_numeric(
        df[
            window_length_col
        ],
        errors="coerce"
    ).dropna()


elif (
    WINDOW_START_COLUMN is not None
    and
    WINDOW_END_COLUMN is not None
    and
    WINDOW_START_COLUMN in df.columns
    and
    WINDOW_END_COLUMN in df.columns
):

    start = pd.to_datetime(
        df[
            WINDOW_START_COLUMN
        ],
        errors="coerce"
    )

    end = pd.to_datetime(
        df[
            WINDOW_END_COLUMN
        ],
        errors="coerce"
    )

    window_values = (
        end - start
    ).dt.total_seconds().dropna()


if (
    window_values is not None
    and
    len(window_values) > 0
):

    print(
        window_values.describe()
    )


    pd.DataFrame({

        "window_length":
            window_values

    }).to_csv(
        TABLE_DIR /
        "A10_window_length.csv",
        index=False
    )


    plt.figure(
        figsize=(11, 7)
    )

    sns.histplot(
        window_values,
        bins=30,
        kde=True
    )

    plt.xlabel(
        "Window length"
    )

    plt.ylabel(
        "Number of observations"
    )

    plt.title(
        "Window Length Distribution"
    )

    save_figure(
        "A10_window_length_distribution.png"
    )

else:

    print(
        "Window length column was not found."
    )

    print(
        "Window length cannot be independently verified "
        "from the available columns."
    )


# ================================================================
# A11 STEP SIZE
# ================================================================

print_section(
    "STEP SIZE"
)


step_values = None


if step_size_col is not None:

    step_values = pd.to_numeric(
        df[
            step_size_col
        ],
        errors="coerce"
    ).dropna()


if (
    step_values is not None
    and
    len(step_values) > 0
):

    print(
        step_values.describe()
    )


    pd.DataFrame({

        "step_size":
            step_values

    }).to_csv(
        TABLE_DIR /
        "A11_step_size.csv",
        index=False
    )


    plt.figure(
        figsize=(11, 7)
    )

    sns.histplot(
        step_values,
        bins=30,
        kde=True
    )

    plt.xlabel(
        "Step size"
    )

    plt.ylabel(
        "Number of observations"
    )

    plt.title(
        "Step Size Distribution"
    )

    save_figure(
        "A11_step_size_distribution.png"
    )

else:

    print(
        "Step size column was not found."
    )

    print(
        "Step size cannot be independently verified "
        "from the available columns."
    )


# ================================================================
# NUMERIC FEATURE DETECTION
# ================================================================

excluded_columns = {
    class_col,
    participant_col,
    timestamp_col,
    "_timestamp_parsed"
}

if "_source_file" in df.columns:
    excluded_columns.add("_source_file")

if "_source_path" in df.columns:
    excluded_columns.add("_source_path")


numeric_columns = (
    df
    .select_dtypes(
        include=np.number
    )
    .columns
    .tolist()
)


feature_columns = [
    column
    for column in numeric_columns
    if column not in excluded_columns
]


print_section(
    "FEATURE INFORMATION"
)

print(
    "Numeric feature count:",
    len(feature_columns)
)

print(
    "Numeric features:"
)

for feature in feature_columns:
    print(
        " ",
        feature
    )


pd.DataFrame({
    "feature": feature_columns
}).to_csv(
    TABLE_DIR /
    "feature_list.csv",
    index=False
)


# ================================================================
# FEATURE QUALITY SUMMARY
# ================================================================

feature_quality = []

for feature in feature_columns:

    series = df[feature]

    feature_quality.append({

        "feature":
            feature,

        "missing_count":
            series.isna().sum(),

        "missing_percentage":
            series.isna().mean() * 100,

        "unique_values":
            series.nunique(),

        "mean":
            series.mean(),

        "std":
            series.std(),

        "minimum":
            series.min(),

        "maximum":
            series.max(),

        "zero_count":
            (series == 0).sum()

    })


feature_quality_df = pd.DataFrame(
    feature_quality
)


feature_quality_df.to_csv(
    TABLE_DIR /
    "feature_quality_summary.csv",
    index=False
)


# ================================================================
# STAGE B
# PARTICIPANT LEVEL ANALYSIS
# ================================================================

print_section(
    "STAGE B: PARTICIPANT LEVEL ANALYSIS"
)


participant_feature_df = (
    df
    .groupby(
        [
            participant_col,
            class_col
        ],
        observed=True
    )[feature_columns]
    .mean()
    .reset_index()
)


participant_feature_df.to_csv(
    TABLE_DIR /
    "B1_participant_level_features.csv",
    index=False
)


# ================================================================
# PARTICIPANT CLASS OBSERVATIONS
# ================================================================

participant_window_counts = (
    df
    .groupby(
        [
            participant_col,
            class_col
        ],
        observed=True
    )
    .size()
    .reset_index(
        name="observations"
    )
)


participant_window_counts.to_csv(
    TABLE_DIR /
    "B2_participant_class_observations.csv",
    index=False
)


# ================================================================
# PARTICIPANT OBSERVATION HEATMAP
# ================================================================

count_matrix = (
    participant_window_counts
    .pivot(
        index=participant_col,
        columns=class_col,
        values="observations"
    )
    .fillna(0)
)

count_matrix = count_matrix.reindex(
    columns=CLASS_ORDER,
    fill_value=0
)


plt.figure(
    figsize=(
        11,
        max(
            8,
            len(count_matrix) * 0.35
        )
    )
)

sns.heatmap(
    count_matrix,
    annot=True,
    fmt="g",
    cmap="mako"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Participant"
)

plt.title(
    "Participant Level Observation Distribution"
)

save_figure(
    "B1_participant_observation_heatmap.png"
)


# ================================================================
# PARTICIPANT LEVEL FEATURE DISTRIBUTIONS
# ================================================================

features_for_distribution = feature_columns[
    :MAX_FEATURE_PLOTS
]


for feature in features_for_distribution:

    plt.figure(
        figsize=(11, 7)
    )

    sns.boxplot(
        data=participant_feature_df,
        x=class_col,
        y=feature,
        order=CLASS_ORDER
    )

    sns.stripplot(
        data=participant_feature_df,
        x=class_col,
        y=feature,
        order=CLASS_ORDER,
        alpha=0.35,
        size=3
    )

    plt.xticks(
        rotation=30
    )

    plt.xlabel(
        "Class"
    )

    plt.ylabel(
        feature
    )

    plt.title(
        f"Participant Level Distribution: {feature}"
    )

    save_figure(
        "B2_feature_"
        + safe_filename(feature)
        + ".png"
    )


# ================================================================
# STAGE C
# STATISTICAL SIGNIFICANCE
# ================================================================

print_section(
    "STAGE C: STATISTICAL SIGNIFICANCE TESTING"
)


friedman_results = []

posthoc_results = []


# ================================================================
# FRIEDMAN TEST
# ================================================================

for feature in feature_columns:

    wide = (
        participant_feature_df[
            [
                participant_col,
                class_col,
                feature
            ]
        ]
        .pivot_table(
            index=participant_col,
            columns=class_col,
            values=feature,
            aggfunc="mean"
        )
    )


    if not all(
        cls in wide.columns
        for cls in CLASS_ORDER
    ):
        continue


    wide = (
        wide[
            CLASS_ORDER
        ]
        .dropna()
    )


    if len(wide) < MIN_VALID_OBSERVATIONS:
        continue


    try:

        statistic, p_value = friedmanchisquare(
            *[
                wide[cls].values
                for cls in CLASS_ORDER
            ]
        )


        kendalls_w = (
            statistic
            /
            (
                len(wide)
                *
                (
                    len(CLASS_ORDER) - 1
                )
            )
        )


        friedman_results.append({

            "feature":
                feature,

            "participants":
                len(wide),

            "statistic":
                statistic,

            "p_value":
                p_value,

            "Kendalls_W":
                kendalls_w

        })


    except Exception as error:

        print(
            "Friedman test failed:",
            feature,
            error
        )


friedman_df = pd.DataFrame(
    friedman_results
)


if len(friedman_df) > 0:

    friedman_df[
        "p_value_corrected"
    ] = multipletests(
        friedman_df[
            "p_value"
        ],
        method="fdr_bh"
    )[1]


    friedman_df[
        "significant_FDR_0_05"
    ] = (
        friedman_df[
            "p_value_corrected"
        ] < 0.05
    )


    friedman_df = (
        friedman_df
        .sort_values(
            "p_value_corrected"
        )
    )


    friedman_df.to_csv(
        TABLE_DIR /
        "C1_friedman_results.csv",
        index=False
    )


    # ============================================================
    # FRIEDMAN SIGNIFICANCE
    # ============================================================

    plot_df = (
        friedman_df
        .head(30)
        .copy()
    )

    plot_df["minus_log10_p"] = (
        -np.log10(
            np.maximum(
                plot_df[
                    "p_value_corrected"
                ],
                1e-300
            )
        )
    )


    plt.figure(
        figsize=(12, 9)
    )

    sns.barplot(
        data=plot_df,
        x="minus_log10_p",
        y="feature"
    )

    plt.axvline(
        -np.log10(0.05),
        linestyle="--"
    )

    plt.xlabel(
        "Negative log10 corrected p value"
    )

    plt.ylabel(
        "Feature"
    )

    plt.title(
        "Friedman Test Significance"
    )

    save_figure(
        "C1_friedman_significance.png"
    )


    # ============================================================
    # KENDALL W
    # ============================================================

    plot_df = (
        friedman_df
        .sort_values(
            "Kendalls_W",
            ascending=False
        )
        .head(30)
    )


    plt.figure(
        figsize=(12, 9)
    )

    sns.barplot(
        data=plot_df,
        x="Kendalls_W",
        y="feature"
    )

    plt.xlabel(
        "Kendall's W"
    )

    plt.ylabel(
        "Feature"
    )

    plt.title(
        "Effect Size Using Kendall's W"
    )

    save_figure(
        "C2_kendalls_w.png"
    )


# ================================================================
# PAIRWISE WILCOXON TESTS
# ================================================================

for feature in feature_columns:

    wide = (
        participant_feature_df[
            [
                participant_col,
                class_col,
                feature
            ]
        ]
        .pivot_table(
            index=participant_col,
            columns=class_col,
            values=feature,
            aggfunc="mean"
        )
    )


    if not all(
        cls in wide.columns
        for cls in CLASS_ORDER
    ):
        continue


    for class_a, class_b in combinations(
        CLASS_ORDER,
        2
    ):

        pair = (
            wide[
                [
                    class_a,
                    class_b
                ]
            ]
            .dropna()
        )


        if len(pair) < MIN_VALID_OBSERVATIONS:
            continue


        try:

            statistic, p_value = stats.wilcoxon(
                pair[class_a],
                pair[class_b]
            )


            posthoc_results.append({

                "feature":
                    feature,

                "class_a":
                    class_a,

                "class_b":
                    class_b,

                "participants":
                    len(pair),

                "statistic":
                    statistic,

                "p_value":
                    p_value

            })


        except Exception as error:

            print(
                "Wilcoxon failed:",
                feature,
                class_a,
                class_b,
                error
            )


posthoc_df = pd.DataFrame(
    posthoc_results
)


if len(posthoc_df) > 0:

    posthoc_df[
        "p_value_corrected"
    ] = multipletests(
        posthoc_df[
            "p_value"
        ],
        method="fdr_bh"
    )[1]


    posthoc_df[
        "significant_FDR_0_05"
    ] = (
        posthoc_df[
            "p_value_corrected"
        ] < 0.05
    )


    posthoc_df.to_csv(
        TABLE_DIR /
        "C3_posthoc_wilcoxon.csv",
        index=False
    )


    # ============================================================
    # PAIRWISE P VALUE HEATMAP
    # ============================================================

    pairwise_matrix = pd.DataFrame(
        np.nan,
        index=CLASS_ORDER,
        columns=CLASS_ORDER
    )


    pairwise_summary = (
        posthoc_df
        .groupby(
            [
                "class_a",
                "class_b"
            ]
        )[
            "p_value_corrected"
        ]
        .median()
    )


    for (
        class_a,
        class_b
    ), value in pairwise_summary.items():

        pairwise_matrix.loc[
            class_a,
            class_b
        ] = value

        pairwise_matrix.loc[
            class_b,
            class_a
        ] = value


    pairwise_matrix.to_csv(
        TABLE_DIR /
        "C4_pairwise_p_values.csv"
    )


    plt.figure(
        figsize=(9, 8)
    )

    sns.heatmap(
        pairwise_matrix,
        annot=True,
        fmt=".4f",
        cmap="coolwarm",
        vmin=0,
        vmax=1
    )

    plt.xlabel(
        "Class"
    )

    plt.ylabel(
        "Class"
    )

    plt.title(
        "Pairwise Corrected P Values"
    )

    save_figure(
        "C4_pairwise_significance_heatmap.png"
    )


# ================================================================
# CLASS FEATURE PROFILE
# ================================================================

class_feature_means = (
    df
    .groupby(
        class_col,
        observed=True
    )[feature_columns]
    .mean()
    .reindex(
        CLASS_ORDER
    )
)


class_feature_means.to_csv(
    TABLE_DIR /
    "C5_class_feature_means.csv"
)


# ================================================================
# STANDARDIZED FEATURE PROFILE
# ================================================================

profile_data = class_feature_means.copy()

profile_data = profile_data.T


profile_scaled = pd.DataFrame(
    StandardScaler().fit_transform(
        profile_data
    ),
    index=profile_data.index,
    columns=profile_data.columns
)


plt.figure(
    figsize=(12, max(8, len(profile_scaled) * 0.35))
)

sns.heatmap(
    profile_scaled,
    cmap="vlag",
    center=0
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Standardized Mean Feature Profile by Class"
)

save_figure(
    "C5_standardized_feature_profile.png"
)


# ================================================================
# FEATURE CORRELATION
# ================================================================

correlation_features = feature_columns[
    :MAX_CORRELATION_FEATURES
]


if len(correlation_features) >= 2:

    corr_matrix = (
        df[
            correlation_features
        ]
        .corr()
    )


    corr_matrix.to_csv(
        TABLE_DIR /
        "C6_feature_correlation_matrix.csv"
    )


    plt.figure(
        figsize=(14, 12)
    )

    sns.heatmap(
        corr_matrix,
        cmap="vlag",
        center=0,
        square=True
    )

    plt.title(
        "Feature Correlation Matrix"
    )

    save_figure(
        "C6_feature_correlation_heatmap.png"
    )


# ================================================================
# STAGE D
# TEMPORAL ANALYSIS
# ================================================================

print_section(
    "STAGE D: TEMPORAL ANALYSIS"
)


if timestamp_col is not None:

    temporal_df = df.copy()


    temporal_df[
        "_timestamp_parsed"
    ] = pd.to_datetime(
        temporal_df[
            timestamp_col
        ],
        errors="coerce"
    )


    temporal_df = (
        temporal_df
        .dropna(
            subset=[
                "_timestamp_parsed"
            ]
        )
    )


    # ============================================================
    # RECORDING START BY PARTICIPANT AND CLASS
    # ============================================================

    temporal_df[
        "_recording_start"
    ] = (
        temporal_df
        .groupby(
            [
                participant_col,
                class_col
            ]
        )[
            "_timestamp_parsed"
        ]
        .transform(
            "min"
        )
    )


    temporal_df[
        "elapsed_seconds"
    ] = (
        temporal_df[
            "_timestamp_parsed"
        ]
        -
        temporal_df[
            "_recording_start"
        ]
    ).dt.total_seconds()


    temporal_df[
        "elapsed_minute"
    ] = (
        temporal_df[
            "elapsed_seconds"
        ]
        / 60
    )


    # ============================================================
    # TEMPORAL BINS
    # ============================================================

    max_time = (
        temporal_df[
            "elapsed_minute"
        ].max()
    )


    temporal_df[
        "time_bin"
    ] = pd.cut(
        temporal_df[
            "elapsed_minute"
        ],
        bins=TEMPORAL_BINS,
        labels=False,
        include_lowest=True
    )


    # ============================================================
    # D1 TEMPORAL CLASS MEANS
    # ============================================================

    temporal_summary = (
        temporal_df
        .groupby(
            [
                class_col,
                "time_bin"
            ],
            observed=True
        )[feature_columns]
        .mean()
        .reset_index()
    )


    temporal_summary.to_csv(
        TABLE_DIR /
        "D1_temporal_class_feature_means.csv",
        index=False
    )


    # ============================================================
    # TEMPORAL FEATURE TRAJECTORIES
    # ============================================================

    for feature in feature_columns[
        :MAX_TEMPORAL_FEATURE_PLOTS
    ]:

        plt.figure(
            figsize=(12, 7)
        )


        sns.lineplot(
            data=temporal_summary,
            x="time_bin",
            y=feature,
            hue=class_col,
            hue_order=CLASS_ORDER,
            marker="o"
        )


        plt.xlabel(
            "Elapsed time bin"
        )

        plt.ylabel(
            feature
        )

        plt.title(
            f"Temporal Pattern: {feature}"
        )

        plt.legend(
            title="Class"
        )

        save_figure(
            "D1_temporal_"
            + safe_filename(feature)
            + ".png"
        )


    # ============================================================
    # D2 EARLY MIDDLE LATE
    # ============================================================

    temporal_df[
        "period"
    ] = pd.cut(
        temporal_df[
            "elapsed_minute"
        ],
        bins=[
            -np.inf,
            max_time / 3,
            2 * max_time / 3,
            np.inf
        ],
        labels=[
            "early",
            "middle",
            "late"
        ]
    )


    period_summary = (
        temporal_df
        .groupby(
            [
                class_col,
                "period"
            ],
            observed=True
        )[feature_columns]
        .mean()
        .reset_index()
    )


    period_summary.to_csv(
        TABLE_DIR /
        "D2_early_middle_late_summary.csv",
        index=False
    )


    for feature in feature_columns[
        :MAX_TEMPORAL_FEATURE_PLOTS
    ]:

        plt.figure(
            figsize=(11, 7)
        )


        sns.lineplot(
            data=period_summary,
            x="period",
            y=feature,
            hue=class_col,
            hue_order=CLASS_ORDER,
            marker="o"
        )


        plt.xlabel(
            "Recording period"
        )

        plt.ylabel(
            feature
        )

        plt.title(
            f"Early, Middle and Late: {feature}"
        )

        plt.legend(
            title="Class"
        )

        save_figure(
            "D2_early_middle_late_"
            + safe_filename(feature)
            + ".png"
        )


    # ============================================================
    # D3 TEMPORAL SLOPES
    # ============================================================

    slope_results = []


    for cls in CLASS_ORDER:

        class_data = (
            temporal_df[
                temporal_df[
                    class_col
                ] == cls
            ]
        )


        for feature in feature_columns:

            temp = (
                class_data[
                    [
                        "elapsed_minute",
                        feature
                    ]
                ]
                .dropna()
            )


            if len(temp) < MIN_VALID_OBSERVATIONS:
                continue


            try:

                slope, intercept, r_value, p_value, std_err = (
                    stats.linregress(
                        temp[
                            "elapsed_minute"
                        ],
                        temp[
                            feature
                        ]
                    )
                )


                slope_results.append({

                    "class":
                        cls,

                    "feature":
                        feature,

                    "slope":
                        slope,

                    "r_squared":
                        r_value ** 2,

                    "p_value":
                        p_value,

                    "standard_error":
                        std_err

                })


            except Exception as error:

                print(
                    "Slope calculation failed:",
                    cls,
                    feature,
                    error
                )


    slope_df = pd.DataFrame(
        slope_results
    )


    if len(slope_df) > 0:

        slope_df.to_csv(
            TABLE_DIR /
            "D3_temporal_slopes.csv",
            index=False
        )


        slope_plot = (
            slope_df
            .assign(
                absolute_slope=lambda x:
                x["slope"].abs()
            )
            .sort_values(
                "absolute_slope",
                ascending=False
            )
            .head(30)
        )


        plt.figure(
            figsize=(12, 9)
        )


        sns.barplot(
            data=slope_plot,
            x="slope",
            y="feature",
            hue="class"
        )


        plt.axvline(
            0,
            linestyle="--"
        )


        plt.xlabel(
            "Temporal slope"
        )

        plt.ylabel(
            "Feature"
        )

        plt.title(
            "Features With Largest Temporal Slopes"
        )

        save_figure(
            "D3_temporal_slopes.png"
        )


    # ============================================================
    # D4 CLASS PROPORTION OVER TIME
    # ============================================================

    class_time_count = (
        temporal_df
        .groupby(
            [
                "time_bin",
                class_col
            ],
            observed=True
        )
        .size()
        .reset_index(
            name="count"
        )
    )


    class_time_pivot = (
        class_time_count
        .pivot(
            index="time_bin",
            columns=class_col,
            values="count"
        )
        .fillna(0)
        .reindex(
            columns=CLASS_ORDER,
            fill_value=0
        )
    )


    class_time_proportion = (
        class_time_pivot
        .div(
            class_time_pivot.sum(axis=1),
            axis=0
        )
    )


    class_time_proportion.to_csv(
        TABLE_DIR /
        "D4_class_proportion_over_time.csv"
    )


    plt.figure(
        figsize=(12, 7)
    )


    sns.lineplot(
        data=class_time_proportion,
        markers=True
    )


    plt.xlabel(
        "Elapsed time bin"
    )

    plt.ylabel(
        "Class proportion"
    )

    plt.title(
        "Class Proportion Across Elapsed Time"
    )

    save_figure(
        "D4_class_proportion_over_time.png"
    )


    # ============================================================
    # D5 ROLLING FEATURE BEHAVIOUR
    # ============================================================

    rolling_features = feature_columns[
        :10
    ]


    for feature in rolling_features:

        rolling_df = temporal_df[
            [
                participant_col,
                class_col,
                "elapsed_minute",
                feature
            ]
        ].copy()


        rolling_df[
            "rolling_mean"
        ] = (
            rolling_df
            .groupby(
                [
                    participant_col,
                    class_col
                ]
            )[feature]
            .transform(
                lambda x:
                x.rolling(
                    21,
                    min_periods=5
                ).mean()
            )
        )


        plt.figure(
            figsize=(12, 7)
        )


        sns.lineplot(
            data=rolling_df,
            x="elapsed_minute",
            y="rolling_mean",
            hue=class_col,
            hue_order=CLASS_ORDER
        )


        plt.xlabel(
            "Elapsed time, minutes"
        )

        plt.ylabel(
            "Rolling mean"
        )

        plt.title(
            f"Rolling Temporal Pattern: {feature}"
        )

        save_figure(
            "D5_rolling_"
            + safe_filename(feature)
            + ".png"
        )


# ================================================================
# STAGE E
# ELAPSED TIME LEAKAGE TEST
# ================================================================

print_section(
    "STAGE E: ELAPSED TIME LEAKAGE TEST"
)


if timestamp_col is not None:

    leakage_df = temporal_df.copy()


    leakage_model_df = (
        leakage_df[
            [
                participant_col,
                class_col,
                "elapsed_minute"
            ]
        ]
        .dropna()
        .groupby(
            [
                participant_col,
                class_col,
                "elapsed_minute"
            ],
            as_index=False
        )
        .mean()
    )


    X_time = leakage_model_df[
        [
            "elapsed_minute"
        ]
    ]


    y_time = leakage_model_df[
        class_col
    ]


    groups_time = leakage_model_df[
        participant_col
    ]


    label_encoder = LabelEncoder()


    y_encoded = (
        label_encoder
        .fit_transform(
            y_time
        )
    )


    # ============================================================
    # E1 ELAPSED TIME DISTRIBUTION
    # ============================================================

    plt.figure(
        figsize=(12, 7)
    )


    for cls in CLASS_ORDER:

        values = (
            leakage_df
            .loc[
                leakage_df[
                    class_col
                ] == cls,
                "elapsed_minute"
            ]
            .dropna()
        )


        if len(values) > 0:

            sns.kdeplot(
                values,
                fill=True,
                alpha=0.25,
                label=cls
            )


    plt.xlabel(
        "Elapsed time, minutes"
    )

    plt.ylabel(
        "Density"
    )

    plt.title(
        "Elapsed Time Distribution by Class"
    )

    plt.legend()

    save_figure(
        "E1_elapsed_time_distribution.png"
    )


    # ============================================================
    # E2 TIME ONLY CLASSIFIER
    # ============================================================

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            LogisticRegression(
                max_iter=2000,
                random_state=RANDOM_STATE
            )
        )
    ])


    group_split = GroupShuffleSplit(
        n_splits=10,
        test_size=0.20,
        random_state=RANDOM_STATE
    )


    time_scores = []

    all_true = []

    all_pred = []


    for train_idx, test_idx in group_split.split(
        X_time,
        y_encoded,
        groups_time
    ):

        X_train = X_time.iloc[
            train_idx
        ]

        X_test = X_time.iloc[
            test_idx
        ]

        y_train = y_encoded[
            train_idx
        ]

        y_test = y_encoded[
            test_idx
        ]


        model.fit(
            X_train,
            y_train
        )


        predictions = model.predict(
            X_test
        )


        time_scores.append({

            "accuracy":
                accuracy_score(
                    y_test,
                    predictions
                ),

            "balanced_accuracy":
                balanced_accuracy_score(
                    y_test,
                    predictions
                ),

            "macro_f1":
                f1_score(
                    y_test,
                    predictions,
                    average="macro"
                )

        })


        all_true.extend(
            y_test.tolist()
        )

        all_pred.extend(
            predictions.tolist()
        )


    time_scores_df = pd.DataFrame(
        time_scores
    )


    time_scores_df.to_csv(
        TABLE_DIR /
        "E2_elapsed_time_only_scores.csv",
        index=False
    )


    print()
    print(
        "Elapsed time only classifier:"
    )

    print(
        time_scores_df.describe()
    )


    performance_means = (
        time_scores_df[
            [
                "accuracy",
                "balanced_accuracy",
                "macro_f1"
            ]
        ]
        .mean()
    )


    plt.figure(
        figsize=(9, 6)
    )


    sns.barplot(
        x=performance_means.index,
        y=performance_means.values
    )


    plt.ylim(
        0,
        1
    )


    plt.ylabel(
        "Score"
    )

    plt.xlabel(
        ""
    )

    plt.title(
        "Elapsed Time Only Classification Performance"
    )

    plt.xticks(
        rotation=30
    )

    save_figure(
        "E2_elapsed_time_only_performance.png"
    )


    # ============================================================
    # E3 CONFUSION MATRIX
    # ============================================================

    cm = confusion_matrix(
        all_true,
        all_pred,
        labels=np.arange(
            len(
                label_encoder.classes_
            )
        )
    )


    cm_df = pd.DataFrame(
        cm,
        index=label_encoder.classes_,
        columns=label_encoder.classes_
    )


    cm_df.to_csv(
        TABLE_DIR /
        "E3_elapsed_time_confusion_matrix.csv"
    )


    plt.figure(
        figsize=(8, 7)
    )


    sns.heatmap(
        cm_df,
        annot=True,
        fmt="g",
        cmap="Blues"
    )


    plt.xlabel(
        "Predicted class"
    )

    plt.ylabel(
        "True class"
    )

    plt.title(
        "Elapsed Time Only Confusion Matrix"
    )

    save_figure(
        "E3_elapsed_time_confusion_matrix.png"
    )


    # ============================================================
    # E4 FEATURE VERSUS ELAPSED TIME
    # ============================================================

    if "slope_df" in locals():

        if len(slope_df) > 0:

            top_temporal_features = (
                slope_df
                .assign(
                    absolute_slope=lambda x:
                    x["slope"].abs()
                )
                .sort_values(
                    "absolute_slope",
                    ascending=False
                )[
                    "feature"
                ]
                .drop_duplicates()
                .head(10)
                .tolist()
            )


            for feature in top_temporal_features:

                plt.figure(
                    figsize=(11, 7)
                )


                sns.scatterplot(
                    data=leakage_df,
                    x="elapsed_minute",
                    y=feature,
                    hue=class_col,
                    hue_order=CLASS_ORDER,
                    alpha=0.35
                )


                plt.xlabel(
                    "Elapsed time, minutes"
                )

                plt.ylabel(
                    feature
                )

                plt.title(
                    f"{feature} Versus Elapsed Time"
                )

                plt.legend()

                save_figure(
                    "E4_time_relationship_"
                    + safe_filename(feature)
                    + ".png"
                )


# ================================================================
# FINAL DATASET SUMMARY
# ================================================================

print_section(
    "FINAL ANALYSIS SUMMARY"
)


summary = {

    "total_rows":
        int(len(df)),

    "total_columns":
        int(len(df.columns)),

    "missing_values":
        int(
            df.isna()
            .sum()
            .sum()
        ),

    "duplicate_rows":
        int(
            df.duplicated()
            .sum()
        ),

    "participants":
        int(participant_count),

    "classes":
        int(
            df[class_col]
            .nunique()
        ),

    "participants_with_all_classes":
        int(
            complete_participants
        ),

    "class_column":
        str(class_col),

    "participant_column":
        str(participant_col),

    "timestamp_column":
        str(timestamp_col),

    "numeric_features":
        int(
            len(feature_columns)
        )

}


with open(
    OUTPUT_DIR /
    "FINAL_ANALYSIS_SUMMARY.json",
    "w",
    encoding="utf8"
) as file:

    json.dump(
        summary,
        file,
        indent=4,
        default=str
    )


pd.DataFrame(
    [
        summary
    ]
).to_csv(
    TABLE_DIR /
    "FINAL_ANALYSIS_SUMMARY.csv",
    index=False
)


# ================================================================
# FINAL PRINT
# ================================================================

print()

print(
    "Total rows:",
    len(df)
)

print(
    "Total columns:",
    len(df.columns)
)

print(
    "Total participants:",
    participant_count
)

print(
    "Total classes:",
    df[class_col].nunique()
)

print(
    "Participants with all classes:",
    complete_participants
)

print(
    "Total numeric features:",
    len(feature_columns)
)

print(
    "Total missing values:",
    int(
        df.isna()
        .sum()
        .sum()
    )
)

print(
    "Duplicate rows:",
    int(
        df.duplicated()
        .sum()
    )
)

print()

print(
    "Results folder:"
)

print(
    OUTPUT_DIR.resolve()
)

print()

print(
    "Visualization folder:"
)

print(
    FIGURE_DIR.resolve()
)

print()

print(
    "Analysis completed successfully."
)

print(
    "Stage A completed."
)

print(
    "Stage B completed."
)

print(
    "Stage C completed."
)

print(
    "Stage D completed."
)

print(
    "Stage E completed."
)

print()

print(
    "All available tables and visualizations were saved."
)
```

```python
# ================================================================
# V23 DESCRIPTIVE STATISTICS AND ADVANCED JOURNAL VISUALIZATION
# ================================================================
#
# Dataset:
# V23_Revised_Cleaned_All_Classes_All_Persons.csv
#
# Main statistical analysis:
# Minimum
# Maximum
# Mean
# Median
# Standard deviation
# Variance
# Skewness
# Kurtosis
# Coefficient of variation
#
# Advanced visualization:
# Skewness plot
# Kurtosis plot
# Radar plot
# Sankey diagram
# Advanced scatter plots
# PCA plots
# Pairwise scatter matrix
# Correlation heatmap
# Mean plus standard deviation
# Feature variability
# Outlier percentage
#
# ================================================================


# ================================================================
# IMPORT LIBRARIES
# ================================================================

import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import skew
from scipy.stats import kurtosis

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

import plotly.graph_objects as go


warnings.filterwarnings("ignore")


# ================================================================
# INPUT DATASET
# ================================================================

INPUT_FILE = (
    "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V23_REVISED/"
    "V23_Revised_Cleaned_All_Classes_All_Persons.csv"
)


# ================================================================
# OPTIONAL RAW DATASET
# ================================================================
#
# If you have the original raw dataset and want direct
# RAW versus V23 comparison, put its path here.
#
# Otherwise keep it as None.
#
# ================================================================

RAW_FILE = None

# Example:
#
# RAW_FILE = (
#     "/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V21/"
#     "V21_Merged_All_Classes_All_Persons.csv"
# )


# ================================================================
# OUTPUT DIRECTORIES
# ================================================================

OUTPUT_DIR = Path(
    "/home/rakib/Desktop/RAW_DATA/"
    "V23_STATISTICAL_ADVANCED_ANALYSIS"
)

FIGURE_DIR = (
    OUTPUT_DIR /
    "visualizations"
)

TABLE_DIR = (
    OUTPUT_DIR /
    "tables"
)

HTML_DIR = (
    OUTPUT_DIR /
    "interactive"
)


OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FIGURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TABLE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

HTML_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# GENERAL SETTINGS
# ================================================================

CLASS_COLUMN = "Class"

PARTICIPANT_COLUMN = "PersonSrial"

TIMESTAMP_COLUMN = "TimeStamp"

MAX_FEATURES_FOR_PLOTS = 20

MAX_SCATTER_FEATURES = 8

RANDOM_STATE = 42


# ================================================================
# SEABORN STYLE
# ================================================================

sns.set_theme(
    style="whitegrid",
    context="notebook"
)

sns.set_palette(
    "deep"
)


# ================================================================
# HELPER FUNCTION
# ================================================================

def save_figure(
    filename
):

    path = (
        FIGURE_DIR /
        filename
    )

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    print(
        "Saved:",
        path
    )


def safe_name(
    text
):

    return (
        str(text)
        .replace(" ", "_")
        .replace("/", "_")
        .replace("\\", "_")
        .replace(":", "_")
    )


# ================================================================
# LOAD V23
# ================================================================

print(
    "=" * 80
)

print(
    "LOADING V23 DATASET"
)

print(
    "=" * 80
)


if not os.path.exists(
    INPUT_FILE
):

    raise FileNotFoundError(
        f"Dataset not found:\n{INPUT_FILE}"
    )


df = pd.read_csv(
    INPUT_FILE
)


print(
    "Rows:",
    len(df)
)

print(
    "Columns:",
    len(df.columns)
)


# ================================================================
# BASIC VALIDATION
# ================================================================

required_columns = [
    CLASS_COLUMN,
    PARTICIPANT_COLUMN,
    TIMESTAMP_COLUMN
]


missing_required = [
    column
    for column in required_columns
    if column not in df.columns
]


if len(missing_required) > 0:

    raise ValueError(
        "Required columns are missing: "
        + str(missing_required)
    )


# ================================================================
# IDENTIFY NUMERIC FEATURES
# ================================================================

excluded_columns = {
    CLASS_COLUMN,
    PARTICIPANT_COLUMN,
    TIMESTAMP_COLUMN
}


numeric_columns = (
    df
    .select_dtypes(
        include=np.number
    )
    .columns
    .tolist()
)


feature_columns = [
    column
    for column in numeric_columns
    if column not in excluded_columns
]


print(
    "Numeric features:",
    len(feature_columns)
)


print(
    "\nFeature list:"
)

for feature in feature_columns:

    print(
        feature
    )


# ================================================================
# CLASS STANDARDIZATION
# ================================================================

df[CLASS_COLUMN] = (
    df[CLASS_COLUMN]
    .astype(str)
    .str.strip()
)


class_order = [
    "class_lecture",
    "music",
    "normal",
    "reading"
]


# ================================================================
# PART 1
# COMPLETE DESCRIPTIVE STATISTICS
# ================================================================

print(
    "\n"
    + "=" * 80
)

print(
    "COMPLETE DESCRIPTIVE STATISTICS"
)

print(
    "=" * 80
)


statistics_rows = []


for feature in feature_columns:

    values = (
        pd.to_numeric(
            df[feature],
            errors="coerce"
        )
        .dropna()
    )


    if len(values) == 0:

        continue


    statistics_rows.append({

        "Feature":
            feature,

        "Count":
            len(values),

        "Missing":
            df[feature].isna().sum(),

        "Missing_Percentage":
            df[feature].isna().mean() * 100,

        "Minimum":
            values.min(),

        "Maximum":
            values.max(),

        "Mean":
            values.mean(),

        "Median":
            values.median(),

        "Standard_Deviation":
            values.std(),

        "Variance":
            values.var(),

        "Skewness":
            skew(
                values,
                bias=False
            ),

        "Kurtosis":
            kurtosis(
                values,
                fisher=True,
                bias=False
            ),

        "Coefficient_of_Variation":
            (
                values.std()
                /
                abs(values.mean())
                if values.mean() != 0
                else np.nan
            ),

        "Unique_Values":
            values.nunique()

    })


statistics_df = pd.DataFrame(
    statistics_rows
)


statistics_df.to_csv(
    TABLE_DIR /
    "complete_descriptive_statistics.csv",
    index=False
)


print(
    statistics_df.to_string(
        index=False
    )
)


# ================================================================
# PART 2
# CLASS WISE STATISTICS
# ================================================================

print(
    "\n"
    + "=" * 80
)

print(
    "CLASS WISE STATISTICS"
)

print(
    "=" * 80
)


class_statistics = []


for class_name in class_order:

    class_data = df[
        df[CLASS_COLUMN] ==
        class_name
    ]


    for feature in feature_columns:

        values = (
            pd.to_numeric(
                class_data[feature],
                errors="coerce"
            )
            .dropna()
        )


        if len(values) == 0:

            continue


        class_statistics.append({

            "Class":
                class_name,

            "Feature":
                feature,

            "Count":
                len(values),

            "Minimum":
                values.min(),

            "Maximum":
                values.max(),

            "Mean":
                values.mean(),

            "Median":
                values.median(),

            "Standard_Deviation":
                values.std(),

            "Variance":
                values.var(),

            "Skewness":
                skew(
                    values,
                    bias=False
                ),

            "Kurtosis":
                kurtosis(
                    values,
                    fisher=True,
                    bias=False
                )

        })


class_statistics_df = pd.DataFrame(
    class_statistics
)


class_statistics_df.to_csv(
    TABLE_DIR /
    "class_wise_statistics.csv",
    index=False
)


# ================================================================
# PART 3
# FEATURE VARIABILITY
# ================================================================

variability_df = (
    statistics_df[
        [
            "Feature",
            "Standard_Deviation",
            "Variance",
            "Coefficient_of_Variation"
        ]
    ]
    .sort_values(
        "Standard_Deviation",
        ascending=False
    )
)


variability_df.to_csv(
    TABLE_DIR /
    "feature_variability.csv",
    index=False
)


plt.figure(
    figsize=(12, 9)
)


plot_data = (
    variability_df
    .head(MAX_FEATURES_FOR_PLOTS)
)


sns.barplot(
    data=plot_data,
    x="Standard_Deviation",
    y="Feature"
)


plt.xlabel(
    "Standard deviation"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Feature Variability Ranking"
)

save_figure(
    "01_feature_variability.png"
)


# ================================================================
# PART 4
# SKEWNESS
# ================================================================

print(
    "\nGenerating skewness analysis..."
)


skewness_df = (
    statistics_df[
        [
            "Feature",
            "Skewness"
        ]
    ]
    .sort_values(
        "Skewness"
    )
)


skewness_df.to_csv(
    TABLE_DIR /
    "feature_skewness.csv",
    index=False
)


plt.figure(
    figsize=(13, 10)
)


sns.barplot(
    data=skewness_df,
    x="Skewness",
    y="Feature"
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Skewness"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Feature Skewness"
)

save_figure(
    "02_skewness_all_features.png"
)


# ================================================================
# SKEWNESS DISTRIBUTION
# ================================================================

plt.figure(
    figsize=(11, 7)
)


sns.histplot(
    statistics_df["Skewness"],
    bins=20,
    kde=True
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Skewness"
)

plt.ylabel(
    "Number of features"
)

plt.title(
    "Distribution of Feature Skewness"
)

save_figure(
    "03_skewness_distribution.png"
)


# ================================================================
# PART 5
# KURTOSIS
# ================================================================

kurtosis_df = (
    statistics_df[
        [
            "Feature",
            "Kurtosis"
        ]
    ]
    .sort_values(
        "Kurtosis"
    )
)


kurtosis_df.to_csv(
    TABLE_DIR /
    "feature_kurtosis.csv",
    index=False
)


plt.figure(
    figsize=(13, 10)
)


sns.barplot(
    data=kurtosis_df,
    x="Kurtosis",
    y="Feature"
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Excess kurtosis"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Feature Kurtosis"
)

save_figure(
    "04_kurtosis_all_features.png"
)


# ================================================================
# KURTOSIS DISTRIBUTION
# ================================================================

plt.figure(
    figsize=(11, 7)
)


sns.histplot(
    statistics_df["Kurtosis"],
    bins=20,
    kde=True
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Excess kurtosis"
)

plt.ylabel(
    "Number of features"
)

plt.title(
    "Distribution of Feature Kurtosis"
)

save_figure(
    "05_kurtosis_distribution.png"
)


# ================================================================
# PART 6
# MEAN PLUS STANDARD DEVIATION PROFILE
# ================================================================

mean_profile = (
    df
    .groupby(
        CLASS_COLUMN
    )[feature_columns]
    .mean()
    .reindex(
        class_order
    )
)


std_profile = (
    df
    .groupby(
        CLASS_COLUMN
    )[feature_columns]
    .std()
    .reindex(
        class_order
    )
)


# Select features according to global variability

top_profile_features = (
    statistics_df
    .sort_values(
        "Standard_Deviation",
        ascending=False
    )[
        "Feature"
    ]
    .head(
        MAX_FEATURES_FOR_PLOTS
    )
    .tolist()
)


for feature in top_profile_features:

    plot_df = pd.DataFrame({

        "Class":
            class_order,

        "Mean":
            mean_profile[
                feature
            ].values,

        "Std":
            std_profile[
                feature
            ].values

    })


    plt.figure(
        figsize=(10, 7)
    )


    sns.barplot(
        data=plot_df,
        x="Class",
        y="Mean"
    )


    x_positions = np.arange(
        len(plot_df)
    )


    plt.errorbar(
        x=x_positions,
        y=plot_df["Mean"],
        yerr=plot_df["Std"],
        fmt="none",
        capsize=5
    )


    plt.xticks(
        rotation=30
    )


    plt.ylabel(
        feature
    )

    plt.xlabel(
        "Class"
    )

    plt.title(
        f"Mean and Standard Deviation: {feature}"
    )


    save_figure(
        "06_mean_std_"
        + safe_name(feature)
        + ".png"
    )


# ================================================================
# PART 7
# STANDARDIZED CLASS PROFILE
# ================================================================

profile_features = top_profile_features


profile = (
    mean_profile[
        profile_features
    ]
    .T
)


profile_standardized = (
    pd.DataFrame(
        StandardScaler()
        .fit_transform(
            profile
        ),
        index=profile.index,
        columns=profile.columns
    )
)


plt.figure(
    figsize=(13, 11)
)


sns.heatmap(
    profile_standardized,
    cmap="vlag",
    center=0
)


plt.xlabel(
    "Class"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Standardized Feature Profile Across Classes"
)

save_figure(
    "07_standardized_class_profile.png"
)


# ================================================================
# PART 8
# SPIDER OR RADAR PLOT
# ================================================================

print(
    "\nGenerating radar plot..."
)


radar_features = (
    statistics_df
    .sort_values(
        "Standard_Deviation",
        ascending=False
    )[
        "Feature"
    ]
    .head(8)
    .tolist()
)


radar_data = (
    df
    .groupby(
        CLASS_COLUMN
    )[radar_features]
    .mean()
    .reindex(
        class_order
    )
)


radar_scaled = pd.DataFrame(
    StandardScaler()
    .fit_transform(
        radar_data
    ),
    index=radar_data.index,
    columns=radar_data.columns
)


angles = np.linspace(
    0,
    2 * np.pi,
    len(radar_features),
    endpoint=False
)


angles = np.concatenate(
    [
        angles,
        [angles[0]]
    ]
)


fig = plt.figure(
    figsize=(10, 10)
)


ax = fig.add_subplot(
    111,
    polar=True
)


for class_name in class_order:

    values = radar_scaled.loc[
        class_name
    ].values


    values = np.concatenate(
        [
            values,
            [values[0]]
        ]
    )


    ax.plot(
        angles,
        values,
        linewidth=2,
        label=class_name
    )


    ax.fill(
        angles,
        values,
        alpha=0.10
    )


ax.set_xticks(
    angles[:-1]
)

ax.set_xticklabels(
    radar_features,
    fontsize=9
)


ax.set_title(
    "Multivariate Class Profile",
    pad=25
)


ax.legend(
    loc="upper right",
    bbox_to_anchor=(1.25, 1.15)
)


plt.tight_layout()


plt.savefig(
    FIGURE_DIR /
    "08_radar_spider_class_profile.png",
    dpi=300,
    bbox_inches="tight"
)


plt.close()


print(
    "Saved radar plot."
)


# ================================================================
# PART 9
# SKEWNESS AND KURTOSIS JOINT PLOT
# ================================================================

plt.figure(
    figsize=(11, 8)
)


sns.scatterplot(
    data=statistics_df,
    x="Skewness",
    y="Kurtosis",
    size="Standard_Deviation",
    hue="Coefficient_of_Variation",
    sizes=(50, 500),
    alpha=0.8
)


for _, row in (
    statistics_df
    .iterrows()
):

    plt.text(
        row["Skewness"],
        row["Kurtosis"],
        row["Feature"],
        fontsize=7
    )


plt.axvline(
    0,
    linestyle="--"
)

plt.axhline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Skewness"
)

plt.ylabel(
    "Excess kurtosis"
)

plt.title(
    "Feature Distribution Shape: Skewness versus Kurtosis"
)


save_figure(
    "09_skewness_kurtosis_scatter.png"
)


# ================================================================
# PART 10
# ADVANCED FEATURE SCATTER
# ================================================================

scatter_features = (
    statistics_df
    .sort_values(
        "Standard_Deviation",
        ascending=False
    )[
        "Feature"
    ]
    .head(
        MAX_SCATTER_FEATURES
    )
    .tolist()
)


if len(scatter_features) >= 2:

    x_feature = scatter_features[0]

    y_feature = scatter_features[1]


    sample_size = min(
        50000,
        len(df)
    )


    scatter_sample = (
        df
        .sample(
            n=sample_size,
            random_state=RANDOM_STATE
        )
    )


    plt.figure(
        figsize=(12, 9)
    )


    sns.scatterplot(
        data=scatter_sample,
        x=x_feature,
        y=y_feature,
        hue=CLASS_COLUMN,
        hue_order=class_order,
        alpha=0.35,
        s=30
    )


    plt.xlabel(
        x_feature
    )

    plt.ylabel(
        y_feature
    )

    plt.title(
        f"Advanced Feature Scatter: {x_feature} versus {y_feature}"
    )


    save_figure(
        "10_advanced_feature_scatter.png"
    )


# ================================================================
# PART 11
# HEXBIN DENSITY SCATTER
# ================================================================

if len(scatter_features) >= 2:

    x_values = (
        pd.to_numeric(
            df[x_feature],
            errors="coerce"
        )
    )

    y_values = (
        pd.to_numeric(
            df[y_feature],
            errors="coerce"
        )
    )


    valid = (
        x_values.notna()
        &
        y_values.notna()
    )


    plt.figure(
        figsize=(11, 8)
    )


    plt.hexbin(
        x_values[valid],
        y_values[valid],
        gridsize=50,
        mincnt=1
    )


    plt.colorbar(
        label="Observation density"
    )


    plt.xlabel(
        x_feature
    )

    plt.ylabel(
        y_feature
    )

    plt.title(
        "Feature Density Scatter Plot"
    )


    save_figure(
        "11_hexbin_density_scatter.png"
    )


# ================================================================
# PART 12
# PAIRWISE SCATTER MATRIX
# ================================================================

pair_features = (
    statistics_df
    .sort_values(
        "Standard_Deviation",
        ascending=False
    )[
        "Feature"
    ]
    .head(6)
    .tolist()
)


pair_sample = (
    df[
        pair_features +
        [CLASS_COLUMN]
    ]
    .dropna()
)


if len(pair_sample) > 10000:

    pair_sample = (
        pair_sample
        .sample(
            10000,
            random_state=RANDOM_STATE
        )
    )


pair_grid = sns.pairplot(
    pair_sample,
    vars=pair_features,
    hue=CLASS_COLUMN,
    hue_order=class_order,
    corner=True,
    plot_kws={
        "alpha": 0.35,
        "s": 20
    }
)


pair_grid.fig.suptitle(
    "Pairwise Feature Relationships",
    y=1.02
)


pair_grid.fig.savefig(
    FIGURE_DIR /
    "12_pairwise_scatter_matrix.png",
    dpi=300,
    bbox_inches="tight"
)


plt.close(
    "all"
)


# ================================================================
# PART 13
# PCA TWO DIMENSIONAL
# ================================================================

print(
    "\nGenerating PCA visualization..."
)


pca_features = (
    statistics_df
    .sort_values(
        "Standard_Deviation",
        ascending=False
    )[
        "Feature"
    ]
    .tolist()
)


pca_data = (
    df[
        pca_features
    ]
    .replace(
        [np.inf, -np.inf],
        np.nan
    )
)


pca_data = (
    pca_data
    .fillna(
        pca_data.median()
    )
)


# Limit memory usage for visualization

if len(pca_data) > 100000:

    sample_indices = (
        pca_data
        .sample(
            100000,
            random_state=RANDOM_STATE
        )
        .index
    )

else:

    sample_indices = (
        pca_data.index
    )


pca_sample = (
    pca_data
    .loc[
        sample_indices
    ]
)


pca_labels = (
    df
    .loc[
        sample_indices,
        CLASS_COLUMN
    ]
)


scaler = StandardScaler()


pca_scaled = (
    scaler
    .fit_transform(
        pca_sample
    )
)


pca = PCA(
    n_components=3,
    random_state=RANDOM_STATE
)


pca_result = (
    pca
    .fit_transform(
        pca_scaled
    )
)


pca_df = pd.DataFrame({

    "PC1":
        pca_result[:, 0],

    "PC2":
        pca_result[:, 1],

    "PC3":
        pca_result[:, 2],

    "Class":
        pca_labels.values

})


pca_df.to_csv(
    TABLE_DIR /
    "pca_coordinates.csv",
    index=False
)


print(
    "PCA explained variance:"
)

print(
    pca.explained_variance_ratio_
)


plt.figure(
    figsize=(12, 9)
)


sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Class",
    hue_order=class_order,
    alpha=0.45,
    s=30
)


plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0] * 100:.2f}%)"
)

plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1] * 100:.2f}%)"
)

plt.title(
    "PCA Two Dimensional Class Separation"
)


save_figure(
    "13_pca_2d.png"
)


# ================================================================
# PART 14
# PCA THREE DIMENSIONAL
# ================================================================

fig = go.Figure()


for class_name in class_order:

    class_points = (
        pca_df[
            pca_df["Class"] ==
            class_name
        ]
    )


    fig.add_trace(
        go.Scatter3d(

            x=class_points["PC1"],

            y=class_points["PC2"],

            z=class_points["PC3"],

            mode="markers",

            name=class_name,

            marker={
                "size": 3,
                "opacity": 0.45
            }

        )
    )


fig.update_layout(

    title="PCA Three Dimensional Class Structure",

    scene={
        "xaxis_title": "PC1",
        "yaxis_title": "PC2",
        "zaxis_title": "PC3"
    },

    width=1100,
    height=850

)


fig.write_html(
    HTML_DIR /
    "14_pca_3d_interactive.html"
)


print(
    "Saved interactive PCA 3D."
)


# ================================================================
# PART 15
# CLASS CORRELATION HEATMAPS
# ================================================================

for class_name in class_order:

    class_data = df[
        df[CLASS_COLUMN] ==
        class_name
    ][
        pca_features
    ]


    corr = (
        class_data
        .corr()
    )


    plt.figure(
        figsize=(14, 12)
    )


    sns.heatmap(
        corr,
        cmap="vlag",
        center=0,
        square=True
    )


    plt.title(
        f"Feature Correlation: {class_name}"
    )


    save_figure(
        "15_correlation_"
        + safe_name(class_name)
        + ".png"
    )


# ================================================================
# PART 16
# OUTLIER PERCENTAGE
# ================================================================

outlier_rows = []


for feature in feature_columns:

    values = (
        pd.to_numeric(
            df[feature],
            errors="coerce"
        )
        .dropna()
    )


    if len(values) == 0:

        continue


    q1 = values.quantile(
        0.25
    )

    q3 = values.quantile(
        0.75
    )


    iqr = q3 - q1


    lower = (
        q1 -
        1.5 * iqr
    )

    upper = (
        q3 +
        1.5 * iqr
    )


    outliers = (
        (values < lower)
        |
        (values > upper)
    )


    outlier_rows.append({

        "Feature":
            feature,

        "Outlier_Count":
            outliers.sum(),

        "Outlier_Percentage":
            outliers.mean() * 100

    })


outlier_df = pd.DataFrame(
    outlier_rows
)


outlier_df = (
    outlier_df
    .sort_values(
        "Outlier_Percentage",
        ascending=False
    )
)


outlier_df.to_csv(
    TABLE_DIR /
    "feature_outlier_percentage.csv",
    index=False
)


plt.figure(
    figsize=(12, 10)
)


sns.barplot(
    data=outlier_df.head(
        MAX_FEATURES_FOR_PLOTS
    ),
    x="Outlier_Percentage",
    y="Feature"
)


plt.xlabel(
    "Outlier percentage"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Feature Level Outlier Percentage"
)


save_figure(
    "16_outlier_percentage.png"
)


# ================================================================
# PART 17
# INTERACTIVE FEATURE STATISTICS
# ================================================================

fig = go.Figure()


fig.add_trace(
    go.Scatter(

        x=statistics_df[
            "Minimum"
        ],

        y=statistics_df[
            "Maximum"
        ],

        mode="markers",

        text=statistics_df[
            "Feature"
        ],

        marker={
            "size": 10,
            "color":
                statistics_df[
                    "Standard_Deviation"
                ],
            "colorscale":
                "Viridis",
            "showscale": True
        },

        hovertemplate=
        "<b>%{text}</b><br>"
        "Minimum: %{x}<br>"
        "Maximum: %{y}<br>"
        "<extra></extra>"

    )
)


fig.update_layout(

    title=
    "Interactive Feature Range Visualization",

    xaxis_title=
    "Minimum value",

    yaxis_title=
    "Maximum value",

    width=1100,

    height=750

)


fig.write_html(
    HTML_DIR /
    "17_feature_range_interactive.html"
)


# ================================================================
# PART 18
# INTERACTIVE SKEWNESS KURTOSIS
# ================================================================

fig = go.Figure()


fig.add_trace(
    go.Scatter(

        x=statistics_df[
            "Skewness"
        ],

        y=statistics_df[
            "Kurtosis"
        ],

        mode="markers+text",

        text=statistics_df[
            "Feature"
        ],

        textposition="top center",

        marker={
            "size": 11,
            "color":
                statistics_df[
                    "Standard_Deviation"
                ],
            "colorscale":
                "Turbo",
            "showscale": True
        }

    )
)


fig.add_vline(
    x=0,
    line_dash="dash"
)


fig.add_hline(
    y=0,
    line_dash="dash"
)


fig.update_layout(

    title=
    "Interactive Skewness and Kurtosis Map",

    xaxis_title=
    "Skewness",

    yaxis_title=
    "Excess Kurtosis",

    width=1200,

    height=800

)


fig.write_html(
    HTML_DIR /
    "18_skewness_kurtosis_interactive.html"
)


# ================================================================
# PART 19
# SANKEY DIAGRAM
# ================================================================
#
# The Sankey represents:
#
# Class
#     ↓
# Participant
#
# It visualizes the distribution of observations from each class
# across participants.
#
# ================================================================

print(
    "\nGenerating Sankey diagram..."
)


sankey_data = (
    df
    .groupby(
        [
            CLASS_COLUMN,
            PARTICIPANT_COLUMN
        ],
        observed=True
    )
    .size()
    .reset_index(
        name="Count"
    )
)


classes = [
    cls
    for cls in class_order
    if cls in sankey_data[
        CLASS_COLUMN
    ].unique()
]


participants = (
    sorted(
        sankey_data[
            PARTICIPANT_COLUMN
        ]
        .astype(str)
        .unique(),
        key=lambda x: (
            int(x)
            if x.isdigit()
            else x
        )
    )
)


class_nodes = [
    "Class: " + cls
    for cls in classes
]


participant_nodes = [
    "Person: " + person
    for person in participants
]


node_labels = (
    class_nodes +
    participant_nodes
)


class_index = {
    cls: i
    for i, cls in enumerate(
        classes
    )
}


participant_index = {

    person:
        len(classes) + i

    for i, person in enumerate(
        participants
    )

}


source = []

target = []

value = []


for _, row in (
    sankey_data
    .iterrows()
):

    class_name = row[
        CLASS_COLUMN
    ]

    participant = str(
        row[
            PARTICIPANT_COLUMN
        ]
    )


    if (
        class_name not in class_index
        or
        participant not in participant_index
    ):

        continue


    source.append(
        class_index[
            class_name
        ]
    )

    target.append(
        participant_index[
            participant
        ]
    )

    value.append(
        row["Count"]
    )


fig = go.Figure(
    data=[
        go.Sankey(

            arrangement="snap",

            node={
                "label":
                    node_labels,

                "pad":
                    12,

                "thickness":
                    18
            },

            link={
                "source":
                    source,

                "target":
                    target,

                "value":
                    value
            }

        )
    ]
)


fig.update_layout(

    title=
    "Class to Participant Observation Flow",

    width=1400,

    height=1000

)


fig.write_html(
    HTML_DIR /
    "19_class_participant_sankey.html"
)


print(
    "Saved Sankey diagram."
)


# ================================================================
# PART 20
# CLASS LEVEL RADAR USING TOP FEATURES
# ================================================================

radar_features = (
    statistics_df
    .sort_values(
        "Standard_Deviation",
        ascending=False
    )[
        "Feature"
    ]
    .head(10)
    .tolist()
)


class_radar = (
    df
    .groupby(
        CLASS_COLUMN
    )[radar_features]
    .mean()
    .reindex(
        class_order
    )
)


class_radar_scaled = pd.DataFrame(
    StandardScaler()
    .fit_transform(
        class_radar
    ),
    index=class_radar.index,
    columns=class_radar.columns
)


class_radar_scaled.to_csv(
    TABLE_DIR /
    "class_radar_standardized_values.csv"
)


# ================================================================
# PART 21
# FEATURE RANGE PLOT
# ================================================================

range_plot = (
    statistics_df[
        [
            "Feature",
            "Minimum",
            "Maximum"
        ]
    ]
    .copy()
)


range_plot[
    "Range"
] = (
    range_plot[
        "Maximum"
    ]
    -
    range_plot[
        "Minimum"
    ]
)


range_plot = (
    range_plot
    .sort_values(
        "Range",
        ascending=False
    )
    .head(
        MAX_FEATURES_FOR_PLOTS
    )
)


plt.figure(
    figsize=(13, 10)
)


sns.barplot(
    data=range_plot,
    x="Range",
    y="Feature"
)


plt.xlabel(
    "Maximum minus minimum"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Feature Value Range"
)


save_figure(
    "20_feature_value_range.png"
)


# ================================================================
# PART 22
# RAW VERSUS PROCESSED COMPARISON
# ================================================================

if RAW_FILE is not None:

    print(
        "\n"
        + "=" * 80
    )

    print(
        "RAW VERSUS V23 COMPARISON"
    )

    print(
        "=" * 80
    )


    if not os.path.exists(
        RAW_FILE
    ):

        print(
            "RAW_FILE was specified but does not exist."
        )

    else:

        raw_df = pd.read_csv(
            RAW_FILE
        )


        raw_numeric = (
            raw_df
            .select_dtypes(
                include=np.number
            )
            .columns
            .tolist()
        )


        common_features = [
            feature
            for feature in raw_numeric
            if feature in feature_columns
        ]


        comparison_rows = []


        for feature in common_features:

            raw_values = (
                pd.to_numeric(
                    raw_df[feature],
                    errors="coerce"
                )
                .dropna()
            )


            processed_values = (
                pd.to_numeric(
                    df[feature],
                    errors="coerce"
                )
                .dropna()
            )


            if (
                len(raw_values) == 0
                or
                len(processed_values) == 0
            ):

                continue


            comparison_rows.append({

                "Feature":
                    feature,

                "Raw_Minimum":
                    raw_values.min(),

                "Processed_Minimum":
                    processed_values.min(),

                "Raw_Maximum":
                    raw_values.max(),

                "Processed_Maximum":
                    processed_values.max(),

                "Raw_Mean":
                    raw_values.mean(),

                "Processed_Mean":
                    processed_values.mean(),

                "Raw_Std":
                    raw_values.std(),

                "Processed_Std":
                    processed_values.std(),

                "Raw_Skewness":
                    skew(
                        raw_values,
                        bias=False
                    ),

                "Processed_Skewness":
                    skew(
                        processed_values,
                        bias=False
                    ),

                "Raw_Kurtosis":
                    kurtosis(
                        raw_values,
                        fisher=True,
                        bias=False
                    ),

                "Processed_Kurtosis":
                    kurtosis(
                        processed_values,
                        fisher=True,
                        bias=False
                    )

            })


        comparison_df = pd.DataFrame(
            comparison_rows
        )


        comparison_df.to_csv(
            TABLE_DIR /
            "raw_vs_processed_statistics.csv",
            index=False
        )


        # --------------------------------------------------------
        # RAW VS PROCESSED STANDARD DEVIATION
        # --------------------------------------------------------

        std_comparison = comparison_df[
            [
                "Feature",
                "Raw_Std",
                "Processed_Std"
            ]
        ]


        std_long = std_comparison.melt(
            id_vars="Feature",
            var_name="Dataset",
            value_name="Standard_Deviation"
        )


        plt.figure(
            figsize=(14, 10)
        )


        sns.barplot(
            data=std_long,
            x="Standard_Deviation",
            y="Feature",
            hue="Dataset"
        )


        plt.title(
            "Raw versus Processed Feature Variability"
        )


        save_figure(
            "21_raw_vs_processed_std.png"
        )


        # --------------------------------------------------------
        # RAW VS PROCESSED SKEWNESS
        # --------------------------------------------------------

        skew_comparison = comparison_df[
            [
                "Feature",
                "Raw_Skewness",
                "Processed_Skewness"
            ]
        ]


        skew_long = skew_comparison.melt(
            id_vars="Feature",
            var_name="Dataset",
            value_name="Skewness"
        )


        plt.figure(
            figsize=(14, 10)
        )


        sns.barplot(
            data=skew_long,
            x="Skewness",
            y="Feature",
            hue="Dataset"
        )


        plt.title(
            "Raw versus Processed Skewness"
        )


        save_figure(
            "22_raw_vs_processed_skewness.png"
        )


# ================================================================
# FINAL SUMMARY
# ================================================================

summary = {

    "Rows":
        len(df),

    "Columns":
        len(df.columns),

    "Numeric_Features":
        len(feature_columns),

    "Participants":
        df[
            PARTICIPANT_COLUMN
        ].nunique(),

    "Classes":
        df[
            CLASS_COLUMN
        ].nunique(),

    "Total_Missing":
        int(
            df.isna()
            .sum()
            .sum()
        ),

    "Duplicate_Rows":
        int(
            df.duplicated()
            .sum()
        )

}


pd.DataFrame(
    [summary]
).to_csv(
    TABLE_DIR /
    "analysis_summary.csv",
    index=False
)


print(
    "\n"
    + "=" * 80
)

print(
    "ADVANCED STATISTICAL ANALYSIS COMPLETED"
)

print(
    "=" * 80
)

print(
    "Rows:",
    summary["Rows"]
)

print(
    "Columns:",
    summary["Columns"]
)

print(
    "Numeric features:",
    summary["Numeric_Features"]
)

print(
    "Participants:",
    summary["Participants"]
)

print(
    "Classes:",
    summary["Classes"]
)

print(
    "Missing values:",
    summary["Total_Missing"]
)

print(
    "Duplicate rows:",
    summary["Duplicate_Rows"]
)

print(
    "\nTables:"
)

print(
    TABLE_DIR
)

print(
    "\nStatic visualizations:"
)

print(
    FIGURE_DIR
)

print(
    "\nInteractive visualizations:"
)

print(
    HTML_DIR
)

print(
    "\nCompleted successfully."
)
```

```python
# ================================================================
# V25
# MULTI SIGNAL WAVELET + Z SCORE NORMALIZATION
# + 21 FEATURE EXTRACTION METHODS
# ================================================================
#
# INPUT:
# V23_Revised_Cleaned_All_Classes_All_Persons.csv
#
# PROCESS:
#
# V23 Dataset
#      |
#      +--> Separate Participant
#      |
#      +--> Separate Class
#      |
#      +--> Separate Individual Signal
#      |
#      +--> Wavelet Denoising
#      |
#      +--> Z Score Normalization
#      |
#      +--> Determine Sampling Rate
#      |
#      +--> 10 Second Window
#      |
#      +--> 5 Second Step
#      |
#      +--> 21 Features
#      |
#      +--> Repeat for Every Signal
#      |
#      +--> Final V25 Dataset
#
# ================================================================


# ================================================================
# IMPORT LIBRARIES
# ================================================================

import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import pywt

from scipy.stats import kurtosis
from scipy.stats import skew
from scipy.stats import iqr

from scipy.signal import find_peaks
from scipy.signal import welch

from scipy.fft import fft


warnings.filterwarnings("ignore")


# ================================================================
# PATH CONFIGURATION
# ================================================================

ROOT_FOLDER = Path(
    "/home/rakib/Desktop/RAW_DATA"
)

INPUT_FILENAME = (
    "V23_Revised_Cleaned_All_Classes_All_Persons.csv"
)


# ================================================================
# OUTPUT FOLDER
# ================================================================

OUTPUT_FOLDER = (
    ROOT_FOLDER /
    "CLEAN_DATA_V25"
)

OUTPUT_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)


OUTPUT_FILE = (
    OUTPUT_FOLDER /
    "V25_All_Signals_21_Features.csv"
)

PREPROCESSING_FILE = (
    OUTPUT_FOLDER /
    "V25_Preprocessing_Report.csv"
)

WINDOW_FILE = (
    OUTPUT_FOLDER /
    "V25_Window_Information.csv"
)

FEATURE_DEFINITION_FILE = (
    OUTPUT_FOLDER /
    "V25_Feature_Definitions.csv"
)

SIGNAL_INFORMATION_FILE = (
    OUTPUT_FOLDER /
    "V25_Signal_Information.csv"
)


# ================================================================
# BASIC CONFIGURATION
# ================================================================

TIMESTAMP_COLUMN = "TimeStamp"

PARTICIPANT_COLUMN = "PersonSrial"

CLASS_COLUMN = "Class"


# ================================================================
# WAVELET CONFIGURATION
# ================================================================

WAVELET_NAME = "db4"

WAVELET_LEVEL = None

WAVELET_THRESHOLD_MODE = "soft"


# ================================================================
# NORMALIZATION
# ================================================================

NORMALIZATION_METHOD = "zscore"


# ================================================================
# WINDOW CONFIGURATION
# ================================================================
#
# The actual sampling rate is calculated from TimeStamp.
#
# Window duration is specified in seconds.
#
# Step is calculated using 50 percent overlap.
#
# ================================================================

TARGET_WINDOW_SECONDS = 10.0

OVERLAP_PERCENT = 50.0


# ================================================================
# FIND INPUT DATASET
# ================================================================

print("=" * 80)
print("V25 MULTI SIGNAL FEATURE EXTRACTION")
print("=" * 80)

print()
print("Searching for dataset...")
print(
    "Root folder:",
    ROOT_FOLDER
)


matching_files = list(
    ROOT_FOLDER.rglob(
        INPUT_FILENAME
    )
)


if len(matching_files) == 0:

    raise FileNotFoundError(
        f"""
Input dataset was not found.

Expected filename:

{INPUT_FILENAME}

Search folder:

{ROOT_FOLDER}
"""
    )


INPUT_FILE = matching_files[0]


print()
print("Input dataset:")
print(INPUT_FILE)


# ================================================================
# READ DATASET
# ================================================================

df = pd.read_csv(
    INPUT_FILE
)


print()
print("=" * 80)
print("DATASET INFORMATION")
print("=" * 80)

print()
print(
    "Rows:",
    len(df)
)

print(
    "Columns:",
    len(df.columns)
)


print()
print("Columns:")

for number, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{number}. {column}"
    )


# ================================================================
# CHECK REQUIRED COLUMNS
# ================================================================

required_columns = [

    TIMESTAMP_COLUMN,
    PARTICIPANT_COLUMN,
    CLASS_COLUMN

]


missing_columns = [

    column
    for column in required_columns
    if column not in df.columns

]


if missing_columns:

    raise ValueError(
        f"""
Required columns are missing:

{missing_columns}
"""
    )


# ================================================================
# TIMESTAMP
# ================================================================

df[TIMESTAMP_COLUMN] = pd.to_datetime(
    df[TIMESTAMP_COLUMN],
    errors="coerce"
)


# ================================================================
# REMOVE INVALID TIMESTAMP ROWS
# ================================================================

invalid_timestamp_rows = int(
    df[TIMESTAMP_COLUMN]
    .isna()
    .sum()
)


df = df.dropna(
    subset=[
        TIMESTAMP_COLUMN
    ]
).copy()


print()
print(
    "Invalid timestamp rows removed:",
    invalid_timestamp_rows
)


# ================================================================
# IDENTIFY SIGNAL COLUMNS
# ================================================================
#
# Metadata columns are excluded.
#
# Every remaining numeric column is treated as an
# individual signal.
#
# This includes:
#
# Delta
# Theta
# Alpha
# Beta
# Gamma
# RAW EEG
# AUX
# Accelerometer
# Gyroscope
# Activity
#
# ================================================================

metadata_columns = [

    TIMESTAMP_COLUMN,
    PARTICIPANT_COLUMN,
    CLASS_COLUMN

]


signal_candidates = [

    column
    for column in df.columns
    if column not in metadata_columns

]


# ================================================================
# CONVERT SIGNALS TO NUMERIC
# ================================================================

signal_columns = []


for column in signal_candidates:

    converted = pd.to_numeric(
        df[column],
        errors="coerce"
    )


    numeric_count = (
        converted.notna().sum()
    )


    if numeric_count > 0:

        df[column] = converted

        signal_columns.append(
            column
        )


# ================================================================
# SIGNAL INFORMATION
# ================================================================

print()
print("=" * 80)
print("SIGNAL COLUMNS")
print("=" * 80)

print()
print(
    "Number of individual signals:",
    len(signal_columns)
)

print()


for number, signal in enumerate(
    signal_columns,
    start=1
):

    print(
        f"{number}. {signal}"
    )


print()
print(
    "Expected extracted features:",
    len(signal_columns) * 21
)


# ================================================================
# SORT DATA
# ================================================================
#
# IMPORTANT:
#
# PersonSrial + Class + TimeStamp
#
# must be used.
#
# The four classes have overlapping timestamps.
#
# ================================================================

df = df.sort_values(
    [
        PARTICIPANT_COLUMN,
        CLASS_COLUMN,
        TIMESTAMP_COLUMN
    ]
).reset_index(
    drop=True
)


# ================================================================
# PARTICIPANTS AND CLASSES
# ================================================================

participants = sorted(
    df[
        PARTICIPANT_COLUMN
    ]
    .dropna()
    .unique()
)


classes = sorted(
    df[
        CLASS_COLUMN
    ]
    .dropna()
    .unique()
)


print()
print("=" * 80)
print("EXPERIMENTAL STRUCTURE")
print("=" * 80)

print()
print(
    "Participants:",
    len(participants)
)

print(
    "Classes:",
    len(classes)
)

print(
    "Participant × Class:",
    len(participants) * len(classes)
)

print(
    "Classes:",
    classes
)


# ================================================================
# DETERMINE SAMPLING INTERVAL
# ================================================================

print()
print("=" * 80)
print("SAMPLING RATE ANALYSIS")
print("=" * 80)


all_intervals = []

sampling_records = []


for participant in participants:

    person_df = df[
        df[PARTICIPANT_COLUMN]
        ==
        participant
    ]


    for class_name in classes:

        segment = person_df[
            person_df[CLASS_COLUMN]
            ==
            class_name
        ].sort_values(
            TIMESTAMP_COLUMN
        )


        if len(segment) < 2:

            continue


        intervals = (
            segment[
                TIMESTAMP_COLUMN
            ]
            .diff()
            .dt.total_seconds()
            .dropna()
        )


        intervals = intervals[
            intervals > 0
        ]


        if len(intervals) == 0:

            continue


        median_interval = float(
            intervals.median()
        )


        all_intervals.extend(
            intervals.tolist()
        )


        sampling_records.append(
            {

                "PersonSrial":
                    participant,

                "Class":
                    class_name,

                "Samples":
                    len(segment),

                "Median_Interval_Seconds":
                    median_interval,

                "Estimated_Hz":
                    1.0 /
                    median_interval

            }
        )


if len(all_intervals) == 0:

    raise ValueError(
        """
Sampling interval could not be determined.
"""
    )


# ================================================================
# GLOBAL SAMPLING RATE
# ================================================================

sampling_interval = float(
    np.median(
        all_intervals
    )
)


sampling_rate = (
    1.0 /
    sampling_interval
)


print()
print(
    "Median sampling interval:",
    f"{sampling_interval:.6f}",
    "seconds"
)

print(
    "Estimated sampling rate:",
    f"{sampling_rate:.6f}",
    "Hz"
)


# ================================================================
# WINDOW SIZE
# ================================================================

WINDOW_SIZE = int(
    round(
        TARGET_WINDOW_SECONDS
        *
        sampling_rate
    )
)


WINDOW_SIZE = max(
    WINDOW_SIZE,
    2
)


# ================================================================
# STEP SIZE
# ================================================================

STEP_SIZE = int(
    round(
        WINDOW_SIZE
        *
        (
            1
            -
            OVERLAP_PERCENT / 100
        )
    )
)


STEP_SIZE = max(
    STEP_SIZE,
    1
)


# ================================================================
# ACTUAL WINDOW INFORMATION
# ================================================================

actual_window_seconds = (
    WINDOW_SIZE /
    sampling_rate
)


actual_step_seconds = (
    STEP_SIZE /
    sampling_rate
)


actual_overlap = (
    1 -
    STEP_SIZE /
    WINDOW_SIZE
) * 100


print()
print("=" * 80)
print("WINDOW CONFIGURATION")
print("=" * 80)

print()
print(
    "Target window:",
    TARGET_WINDOW_SECONDS,
    "seconds"
)

print(
    "Sampling rate:",
    sampling_rate,
    "Hz"
)

print(
    "Window size:",
    WINDOW_SIZE,
    "samples"
)

print(
    "Actual window:",
    f"{actual_window_seconds:.6f}",
    "seconds"
)

print(
    "Step size:",
    STEP_SIZE,
    "samples"
)

print(
    "Actual step:",
    f"{actual_step_seconds:.6f}",
    "seconds"
)

print(
    "Overlap:",
    f"{actual_overlap:.2f}",
    "%"
)


# ================================================================
# WAVELET DENOISING
# ================================================================

def wavelet_denoise(
    signal,
    wavelet_name="db4",
    level=None
):

    signal = np.asarray(
        signal,
        dtype=float
    )


    original_length = len(
        signal
    )


    if original_length < 4:

        return signal.copy()


    wavelet = pywt.Wavelet(
        wavelet_name
    )


    maximum_level = pywt.dwt_max_level(
        original_length,
        wavelet.dec_len
    )


    if maximum_level < 1:

        return signal.copy()


    if level is None:

        selected_level = min(
            maximum_level,
            4
        )

    else:

        selected_level = min(
            int(level),
            maximum_level
        )


    coefficients = pywt.wavedec(
        signal,
        wavelet,
        level=selected_level
    )


    finest_detail = coefficients[-1]


    if len(finest_detail) == 0:

        return signal.copy()


    sigma = (
        np.median(
            np.abs(
                finest_detail
            )
        )
        /
        0.6745
    )


    if (
        not np.isfinite(
            sigma
        )
        or
        sigma <= 0
    ):

        return signal.copy()


    threshold = (
        sigma
        *
        np.sqrt(
            2 *
            np.log(
                len(signal)
            )
        )
    )


    thresholded = [
        coefficients[0]
    ]


    for detail in coefficients[1:]:

        thresholded_detail = pywt.threshold(
            detail,
            threshold,
            mode=WAVELET_THRESHOLD_MODE
        )


        thresholded.append(
            thresholded_detail
        )


    reconstructed = pywt.waverec(
        thresholded,
        wavelet
    )


    reconstructed = reconstructed[
        :original_length
    ]


    if len(reconstructed) < original_length:

        reconstructed = np.pad(
            reconstructed,
            (
                0,
                original_length
                -
                len(reconstructed)
            ),
            mode="edge"
        )


    return reconstructed


# ================================================================
# Z SCORE NORMALIZATION
# ================================================================

def zscore_normalize(
    signal
):

    signal = np.asarray(
        signal,
        dtype=float
    )


    mean_value = np.mean(
        signal
    )


    standard_deviation = np.std(
        signal
    )


    if (
        not np.isfinite(
            standard_deviation
        )
        or
        standard_deviation == 0
    ):

        return (
            signal -
            mean_value
        )


    normalized = (
        signal -
        mean_value
    ) / standard_deviation


    return normalized


# ================================================================
# 21 FEATURE EXTRACTION METHODS
# ================================================================

def extract_21_features(
    signal,
    fs
):

    signal = np.asarray(
        signal,
        dtype=float
    )


    features = {}


    # ============================================================
    # 1. MEAN
    # ============================================================

    features["Mean"] = (
        np.mean(signal)
    )


    # ============================================================
    # 2. MEDIAN
    # ============================================================

    features["Median"] = (
        np.median(signal)
    )


    # ============================================================
    # 3. STANDARD DEVIATION
    # ============================================================

    features["Standard_Deviation"] = (
        np.std(signal)
    )


    # ============================================================
    # 4. VARIANCE
    # ============================================================

    features["Variance"] = (
        np.var(signal)
    )


    # ============================================================
    # 5. MINIMUM
    # ============================================================

    features["Minimum"] = (
        np.min(signal)
    )


    # ============================================================
    # 6. MAXIMUM
    # ============================================================

    features["Maximum"] = (
        np.max(signal)
    )


    # ============================================================
    # 7. RANGE
    # ============================================================

    features["Range"] = (
        np.ptp(signal)
    )


    # ============================================================
    # 8. IQR
    # ============================================================

    features["IQR"] = (
        iqr(signal)
    )


    # ============================================================
    # 9. KURTOSIS
    # ============================================================

    if len(signal) >= 4:

        features["Kurtosis"] = (
            kurtosis(
                signal,
                bias=False
            )
        )

    else:

        features["Kurtosis"] = 0.0


    # ============================================================
    # 10. SKEWNESS
    # ============================================================

    if len(signal) >= 3:

        features["Skewness"] = (
            skew(
                signal,
                bias=False
            )
        )

    else:

        features["Skewness"] = 0.0


    # ============================================================
    # 11. MEAN ABSOLUTE DEVIATION
    # ============================================================

    signal_mean = (
        np.mean(signal)
    )


    features["Mean_Absolute_Deviation"] = (
        np.mean(
            np.abs(
                signal -
                signal_mean
            )
        )
    )


    # ============================================================
    # 12. AUTOCORRELATION
    # ============================================================

    if len(signal) > 1:

        x1 = signal[:-1]

        x2 = signal[1:]


        if (
            np.std(x1) > 0
            and
            np.std(x2) > 0
        ):

            autocorrelation = (
                np.corrcoef(
                    x1,
                    x2
                )[0, 1]
            )

        else:

            autocorrelation = 0.0

    else:

        autocorrelation = 0.0


    if not np.isfinite(
        autocorrelation
    ):

        autocorrelation = 0.0


    features["Autocorrelation"] = (
        autocorrelation
    )


    # ============================================================
    # 13. RMS
    # ============================================================

    features["RMS"] = (
        np.sqrt(
            np.mean(
                signal ** 2
            )
        )
    )


    # ============================================================
    # 14. NUMBER OF PEAKS
    # ============================================================

    peaks, _ = find_peaks(
        signal
    )


    features["Number_of_Peaks"] = (
        len(peaks)
    )


    # ============================================================
    # 15. ENERGY
    # ============================================================

    features["Energy"] = (
        np.sum(
            signal ** 2
        )
    )


    # ============================================================
    # 16. ZERO CROSSING RATE
    # ============================================================

    if len(signal) > 1:

        zero_crossings = np.sum(
            np.diff(
                np.sign(signal)
            ) != 0
        )

    else:

        zero_crossings = 0


    features["Zero_Crossing_Rate"] = (
        zero_crossings /
        len(signal)
    )


    # ============================================================
    # 17. FFT MEAN
    # ============================================================

    fft_values = np.real(
        fft(signal)
    )


    features["FFT_Mean"] = (
        np.mean(
            fft_values
        )
    )


    # ============================================================
    # 18. FFT STANDARD DEVIATION
    # ============================================================

    features["FFT_Std"] = (
        np.std(
            fft_values
        )
    )


    # ============================================================
    # WELCH POWER SPECTRAL DENSITY
    # ============================================================

    frequencies, psd = welch(
        signal,
        fs=fs,
        nperseg=min(
            len(signal),
            256
        )
    )


    # ============================================================
    # 19. PSD MEAN
    # ============================================================

    features["PSD_Mean"] = (
        np.mean(psd)
    )


    # ============================================================
    # 20. PSD STANDARD DEVIATION
    # ============================================================

    features["PSD_Std"] = (
        np.std(psd)
    )


    # ============================================================
    # 21. BAND POWER
    # ============================================================

    features["Band_Power"] = (
        np.sum(psd)
    )


    return features


# ================================================================
# FEATURE NAMES
# ================================================================

base_feature_names = [

    "Mean",
    "Median",
    "Standard_Deviation",
    "Variance",
    "Minimum",
    "Maximum",
    "Range",
    "IQR",
    "Kurtosis",
    "Skewness",
    "Mean_Absolute_Deviation",
    "Autocorrelation",
    "RMS",
    "Number_of_Peaks",
    "Energy",
    "Zero_Crossing_Rate",
    "FFT_Mean",
    "FFT_Std",
    "PSD_Mean",
    "PSD_Std",
    "Band_Power"

]


# ================================================================
# EXPECTED OUTPUT FEATURE COUNT
# ================================================================

expected_feature_count = (
    len(signal_columns)
    *
    len(base_feature_names)
)


print()
print("=" * 80)
print("EXPECTED FEATURE STRUCTURE")
print("=" * 80)

print()
print(
    "Individual input signals:",
    len(signal_columns)
)

print(
    "Features per signal:",
    len(base_feature_names)
)

print(
    "Total extracted features:",
    expected_feature_count
)


# ================================================================
# FEATURE DEFINITION TABLE
# ================================================================

feature_definition_rows = []


for signal in signal_columns:

    for feature_number, feature_name in enumerate(
        base_feature_names,
        start=1
    ):

        feature_definition_rows.append(
            {

                "Signal":
                    signal,

                "Feature_Number":
                    feature_number,

                "Feature":
                    feature_name,

                "Output_Column":
                    f"{signal}__{feature_name}"

            }
        )


feature_definition_df = pd.DataFrame(
    feature_definition_rows
)


# ================================================================
# MAIN PROCESSING
# ================================================================

print()
print("=" * 80)
print("STARTING MULTI SIGNAL FEATURE EXTRACTION")
print("=" * 80)


all_feature_rows = []

window_information_rows = []

signal_information_rows = []


total_windows = 0


skipped_windows = 0


total_segments = 0


# ================================================================
# PARTICIPANT LOOP
# ================================================================

for participant_index, participant in enumerate(
    participants,
    start=1
):


    print()
    print(
        "=" * 70
    )

    print(
        f"PARTICIPANT "
        f"{participant_index}/"
        f"{len(participants)}: "
        f"{participant}"
    )

    print(
        "=" * 70
    )


    person_df = df[
        df[PARTICIPANT_COLUMN]
        ==
        participant
    ]


    # ============================================================
    # CLASS LOOP
    # ============================================================

    for class_name in classes:


        segment = person_df[
            person_df[CLASS_COLUMN]
            ==
            class_name
        ].sort_values(
            TIMESTAMP_COLUMN
        ).reset_index(
            drop=True
        )


        if len(segment) == 0:

            continue


        total_segments += 1


        print(
            f"  Class: "
            f"{class_name} | "
            f"Samples: "
            f"{len(segment)}"
        )


        # ========================================================
        # PREPARE ALL SIGNALS
        # ========================================================

        processed_signals = {}


        # ========================================================
        # SIGNAL LOOP
        # ========================================================

        for signal_number, signal_name in enumerate(
            signal_columns,
            start=1
        ):


            signal = pd.to_numeric(
                segment[
                    signal_name
                ],
                errors="coerce"
            )


            # ----------------------------------------------------
            # Interpolate missing values
            # ----------------------------------------------------

            signal = (
                signal
                .interpolate(
                    method="linear",
                    limit_direction="both"
                )
            )


            # ----------------------------------------------------
            # Remaining missing values
            # ----------------------------------------------------

            if signal.isna().any():

                median_value = (
                    signal.median()
                )


                if np.isfinite(
                    median_value
                ):

                    signal = (
                        signal
                        .fillna(
                            median_value
                        )
                    )

                else:

                    signal = (
                        signal
                        .fillna(0.0)
                    )


            signal_array = (
                signal
                .to_numpy(
                    dtype=float
                )
            )


            # ----------------------------------------------------
            # Replace invalid values
            # ----------------------------------------------------

            signal_array = np.nan_to_num(
                signal_array,
                nan=0.0,
                posinf=0.0,
                neginf=0.0
            )


            # ====================================================
            # WAVELET TRANSFORMATION
            # ====================================================

            wavelet_signal = wavelet_denoise(
                signal_array,
                wavelet_name=WAVELET_NAME,
                level=WAVELET_LEVEL
            )


            # ====================================================
            # Z SCORE NORMALIZATION
            # ====================================================

            normalized_signal = zscore_normalize(
                wavelet_signal
            )


            # ====================================================
            # STORE PROCESSED SIGNAL
            # ====================================================

            processed_signals[
                signal_name
            ] = normalized_signal


            # ====================================================
            # SIGNAL INFORMATION
            # ====================================================

            signal_information_rows.append(
                {

                    "PersonSrial":
                        participant,

                    "Class":
                        class_name,

                    "Signal":
                        signal_name,

                    "Original_Samples":
                        len(signal_array),

                    "Original_Mean":
                        np.mean(
                            signal_array
                        ),

                    "Original_Std":
                        np.std(
                            signal_array
                        ),

                    "Processed_Mean":
                        np.mean(
                            normalized_signal
                        ),

                    "Processed_Std":
                        np.std(
                            normalized_signal
                        ),

                    "Wavelet":
                        WAVELET_NAME,

                    "Normalization":
                        "Z Score"

                }
            )


        # ========================================================
        # WINDOWING
        # ========================================================

        if len(segment) < WINDOW_SIZE:

            print(
                "  Segment too short for window."
            )

            continue


        number_of_windows = (
            (
                len(segment)
                -
                WINDOW_SIZE
            )
            //
            STEP_SIZE
        )
        + 1


        print(
            "  Windows:",
            number_of_windows
        )


        # ========================================================
        # WINDOW LOOP
        # ========================================================

        for window_number, start in enumerate(

            range(
                0,
                len(segment)
                -
                WINDOW_SIZE
                +
                1,
                STEP_SIZE
            ),

            start=1

        ):


            end = (
                start
                +
                WINDOW_SIZE
            )


            # ====================================================
            # BASE OUTPUT ROW
            # ====================================================

            output_row = {

                "PersonSrial":
                    participant,

                "Class":
                    class_name,

                "Window":
                    window_number,

                "Window_Start":
                    segment[
                        TIMESTAMP_COLUMN
                    ].iloc[start],

                "Window_End":
                    segment[
                        TIMESTAMP_COLUMN
                    ].iloc[end - 1],

                "Window_Samples":
                    WINDOW_SIZE

            }


            valid_window = True


            # ====================================================
            # APPLY 21 METHODS TO EVERY SIGNAL
            # ====================================================

            for signal_name in signal_columns:


                signal_window = (
                    processed_signals[
                        signal_name
                    ][
                        start:end
                    ]
                )


                # ------------------------------------------------
                # Check window
                # ------------------------------------------------

                if (
                    len(
                        signal_window
                    )
                    !=
                    WINDOW_SIZE
                ):

                    valid_window = False

                    break


                if (
                    np.isnan(
                        signal_window
                    ).any()
                    or
                    np.isinf(
                        signal_window
                    ).any()
                ):

                    valid_window = False

                    break


                # =================================================
                # 21 FEATURES
                # =================================================

                features = extract_21_features(
                    signal_window,
                    sampling_rate
                )


                # =================================================
                # ADD SIGNAL PREFIX
                # =================================================

                for feature_name, value in features.items():

                    output_column = (
                        f"{signal_name}"
                        f"__"
                        f"{feature_name}"
                    )


                    output_row[
                        output_column
                    ] = value


            # ====================================================
            # SAVE VALID WINDOW
            # ====================================================

            if not valid_window:

                skipped_windows += 1

                continue


            all_feature_rows.append(
                output_row
            )


            window_information_rows.append(
                {

                    "PersonSrial":
                        participant,

                    "Class":
                        class_name,

                    "Window":
                        window_number,

                    "Window_Start":
                        output_row[
                            "Window_Start"
                        ],

                    "Window_End":
                        output_row[
                            "Window_End"
                        ],

                    "Window_Samples":
                        WINDOW_SIZE,

                    "Window_Duration_Seconds":
                        (
                            output_row[
                                "Window_End"
                            ]
                            -
                            output_row[
                                "Window_Start"
                            ]
                        ).total_seconds()

                }
            )


            total_windows += 1


# ================================================================
# CREATE FINAL DATAFRAME
# ================================================================

print()
print("=" * 80)
print("CREATING FINAL FEATURE DATASET")
print("=" * 80)


feature_df = pd.DataFrame(
    all_feature_rows
)


window_df = pd.DataFrame(
    window_information_rows
)


signal_information_df = pd.DataFrame(
    signal_information_rows
)


# ================================================================
# CHECK OUTPUT
# ================================================================

if feature_df.empty:

    raise ValueError(
        """
No feature rows were generated.

Please check the dataset and processing parameters.
"""
    )


# ================================================================
# REPLACE INFINITE VALUES
# ================================================================

feature_df.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan,
    inplace=True
)


# ================================================================
# CHECK MISSING VALUES
# ================================================================

missing_feature_values = int(
    feature_df.isna()
    .sum()
    .sum()
)


# ================================================================
# SAVE
# ================================================================

feature_df.to_csv(
    OUTPUT_FILE,
    index=False
)


window_df.to_csv(
    WINDOW_FILE,
    index=False
)


feature_definition_df.to_csv(
    FEATURE_DEFINITION_FILE,
    index=False
)


signal_information_df.to_csv(
    SIGNAL_INFORMATION_FILE,
    index=False
)


# ================================================================
# PREPROCESSING REPORT
# ================================================================

preprocessing_report = pd.DataFrame(

    [

        [
            "Input file",
            str(INPUT_FILE)
        ],

        [
            "Output file",
            str(OUTPUT_FILE)
        ],

        [
            "Original rows",
            len(df)
        ],

        [
            "Original columns",
            len(df.columns)
        ],

        [
            "Number of participants",
            len(participants)
        ],

        [
            "Number of classes",
            len(classes)
        ],

        [
            "Participant × class segments",
            total_segments
        ],

        [
            "Number of individual signals",
            len(signal_columns)
        ],

        [
            "Features per signal",
            21
        ],

        [
            "Total extracted feature columns",
            expected_feature_count
        ],

        [
            "Sampling interval seconds",
            sampling_interval
        ],

        [
            "Sampling rate Hz",
            sampling_rate
        ],

        [
            "Wavelet",
            WAVELET_NAME
        ],

        [
            "Wavelet level",
            WAVELET_LEVEL
        ],

        [
            "Normalization",
            "Z Score"
        ],

        [
            "Target window seconds",
            TARGET_WINDOW_SECONDS
        ],

        [
            "Window samples",
            WINDOW_SIZE
        ],

        [
            "Actual window seconds",
            actual_window_seconds
        ],

        [
            "Overlap percent",
            actual_overlap
        ],

        [
            "Step samples",
            STEP_SIZE
        ],

        [
            "Actual step seconds",
            actual_step_seconds
        ],

        [
            "Total generated windows",
            total_windows
        ],

        [
            "Skipped windows",
            skipped_windows
        ],

        [
            "Missing feature values",
            missing_feature_values
        ],

        [
            "Final rows",
            len(feature_df)
        ],

        [
            "Final columns",
            len(feature_df.columns)
        ]

    ],

    columns=[
        "Parameter",
        "Value"
    ]
)


preprocessing_report.to_csv(
    PREPROCESSING_FILE,
    index=False
)


# ================================================================
# FINAL REPORT
# ================================================================

print()
print()
print("=" * 80)
print("V25 FEATURE EXTRACTION COMPLETED")
print("=" * 80)

print()

print(
    "Input dataset:"
)

print(
    INPUT_FILE
)

print()

print(
    "Output dataset:"
)

print(
    OUTPUT_FILE
)

print()

print(
    "Output folder:"
)

print(
    OUTPUT_FOLDER
)

print()
print(
    "Participants:",
    len(participants)
)

print(
    "Classes:",
    len(classes)
)

print(
    "Participant × Class segments:",
    total_segments
)

print(
    "Individual input signals:",
    len(signal_columns)
)

print(
    "Features per signal:",
    21
)

print(
    "Expected feature columns:",
    expected_feature_count
)

print(
    "Actual final columns:",
    len(feature_df.columns)
)

print()

print(
    "Sampling interval:",
    f"{sampling_interval:.6f}",
    "seconds"
)

print(
    "Sampling rate:",
    f"{sampling_rate:.6f}",
    "Hz"
)

print()

print(
    "Wavelet:",
    WAVELET_NAME
)

print(
    "Normalization:",
    "Z Score"
)

print()

print(
    "Window size:",
    WINDOW_SIZE,
    "samples"
)

print(
    "Window duration:",
    f"{actual_window_seconds:.6f}",
    "seconds"
)

print(
    "Step size:",
    STEP_SIZE,
    "samples"
)

print(
    "Step duration:",
    f"{actual_step_seconds:.6f}",
    "seconds"
)

print(
    "Overlap:",
    f"{actual_overlap:.2f}",
    "%"
)

print()

print(
    "Generated windows:",
    total_windows
)

print(
    "Skipped windows:",
    skipped_windows
)

print()

print(
    "Final dataset shape:",
    feature_df.shape
)

print()

print(
    "Missing feature values:",
    missing_feature_values
)

print()

print("=" * 80)
print("OUTPUT FILES")
print("=" * 80)

print()

print(
    "Feature dataset:"
)

print(
    OUTPUT_FILE
)

print()

print(
    "Preprocessing report:"
)

print(
    PREPROCESSING_FILE
)

print()

print(
    "Window information:"
)

print(
    WINDOW_FILE
)

print()

print(
    "Feature definitions:"
)

print(
    FEATURE_DEFINITION_FILE
)

print()

print(
    "Signal information:"
)

print(
    SIGNAL_INFORMATION_FILE
)

print()

print("=" * 80)
print("PROCESS COMPLETED")
print("=" * 80)
```

```python
# ================================================================
# V25 COMPREHENSIVE FEATURE DATASET VISUAL ANALYSIS
# ================================================================
#
# INPUT:
# V25_All_Signals_21_Features.csv
#
# DATASET:
# 215760 windows
# 699 columns
# 693 extracted features
#
# MAIN ANALYSIS:
#
# A. Dataset quality
# B. Missing values
# C. Duplicate rows
# D. Class distribution
# E. Participant distribution
# F. Participant x Class distribution
# G. Feature distributions
# H. Boxplots
# I. Violin plots
# J. Skewness
# K. Kurtosis
# L. Outlier analysis
# M. Feature correlation
# N. Signal correlation
# O. Participant wise patterns
# P. Class wise patterns
# Q. Radar / Spider plots
# R. PCA
# S. PCA class visualization
# T. PCA participant visualization
# U. PCA density
# V. PCA cumulative explained variance
# W. Advanced pair plot
#
# IMPORTANT:
# The 2.2 GB CSV is processed in chunks.
# The complete dataset is NOT loaded into RAM.
#
# Original V25 dataset is never modified.
#
# ================================================================


# ================================================================
# IMPORT LIBRARIES
# ================================================================

import os
import re
import warnings

from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import skew
from scipy.stats import kurtosis

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer


warnings.filterwarnings("ignore")


# ================================================================
# SEABORN CONFIGURATION
# ================================================================

sns.set_theme(
    style="whitegrid",
    context="notebook"
)

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 11


# ================================================================
# INPUT FILE
# ================================================================

INPUT_FILE = (
    "/home/rakib/Desktop/RAW_DATA/"
    "CLEAN_DATA_V25/"
    "V25_All_Signals_21_Features.csv"
)


# ================================================================
# OUTPUT ROOT
# ================================================================

OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/"
    "V25_VISUAL_ANALYSIS"
)


OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# OUTPUT FOLDERS
# ================================================================

DIRS = {

    "quality":
        OUTPUT_ROOT / "01_Data_Quality",

    "distribution":
        OUTPUT_ROOT / "02_Distributions",

    "outlier":
        OUTPUT_ROOT / "03_Outliers",

    "participant":
        OUTPUT_ROOT / "04_Participant_Analysis",

    "class":
        OUTPUT_ROOT / "05_Class_Analysis",

    "correlation":
        OUTPUT_ROOT / "06_Correlation",

    "advanced":
        OUTPUT_ROOT / "07_Advanced",

    "summary":
        OUTPUT_ROOT / "08_Summary"

}


for directory in DIRS.values():

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ================================================================
# CONFIGURATION
# ================================================================

CHUNK_SIZE = 10000

PLOT_SAMPLE_SIZE = 60000

PCA_SAMPLE_SIZE = 30000

PAIRPLOT_SAMPLE_SIZE = 5000

RANDOM_STATE = 42

OUTLIER_Z_THRESHOLD = 3.5


# ================================================================
# CHECK INPUT FILE
# ================================================================

if not os.path.isfile(INPUT_FILE):

    raise FileNotFoundError(
        f"""
Input dataset was not found.

Expected:

{INPUT_FILE}
"""
    )


print()
print("=" * 80)
print("V25 FEATURE DATASET VISUAL ANALYSIS")
print("=" * 80)

print()
print("Input:")
print(INPUT_FILE)

print()
print("Output:")
print(OUTPUT_ROOT)


# ================================================================
# READ HEADER ONLY
# ================================================================

header = pd.read_csv(
    INPUT_FILE,
    nrows=0
)


columns = list(
    header.columns
)


print()
print("=" * 80)
print("DATASET STRUCTURE")
print("=" * 80)

print()
print(
    "Total columns:",
    len(columns)
)


# ================================================================
# METADATA COLUMNS
# ================================================================

METADATA_COLUMNS = [

    "PersonSrial",
    "Class",
    "Window",
    "Window_Start",
    "Window_End",
    "Window_Samples"

]


metadata_columns = [

    column
    for column in METADATA_COLUMNS
    if column in columns

]


feature_columns = [

    column
    for column in columns
    if column not in metadata_columns

]


print(
    "Metadata columns:",
    len(metadata_columns)
)

print(
    "Feature columns:",
    len(feature_columns)
)


# ================================================================
# FEATURE INFORMATION
# ================================================================

feature_information = []


for column in feature_columns:

    if "__" in column:

        signal_name, feature_name = (
            column.split(
                "__",
                1
            )
        )

    else:

        signal_name = column
        feature_name = "Unknown"


    feature_information.append(
        {

            "Column":
                column,

            "Signal":
                signal_name,

            "Feature":
                feature_name

        }
    )


feature_info = pd.DataFrame(
    feature_information
)


# ================================================================
# FEATURE FAMILIES
# ================================================================

FEATURE_FAMILIES = {

    "Time_Domain": [

        "Mean",
        "Median",
        "Standard_Deviation",
        "Variance",
        "Minimum",
        "Maximum",
        "Range",
        "IQR",
        "Kurtosis",
        "Skewness",
        "Mean_Absolute_Deviation",
        "Autocorrelation",
        "RMS",
        "Number_of_Peaks",
        "Energy",
        "Zero_Crossing_Rate"

    ],

    "Frequency_Domain": [

        "FFT_Mean",
        "FFT_Std",
        "PSD_Mean",
        "PSD_Std",
        "Band_Power"

    ]

}


def get_feature_family(
    feature_name
):

    for family, feature_list in FEATURE_FAMILIES.items():

        if feature_name in feature_list:

            return family

    return "Other"


feature_info["Family"] = (
    feature_info["Feature"]
    .apply(
        get_feature_family
    )
)


# ================================================================
# SAVE FEATURE MAP
# ================================================================

feature_info.to_csv(

    DIRS["summary"] /
    "feature_column_map.csv",

    index=False
)


# ================================================================
# PASS 1
# DATA QUALITY ANALYSIS
# ================================================================

print()
print("=" * 80)
print("PASS 1: DATA QUALITY ANALYSIS")
print("=" * 80)


missing_counts = pd.Series(
    0,
    index=feature_columns,
    dtype="int64"
)


non_missing_counts = pd.Series(
    0,
    index=feature_columns,
    dtype="int64"
)


total_rows = 0


class_counts = {}


participant_counts = {}


for chunk in pd.read_csv(

    INPUT_FILE,

    chunksize=CHUNK_SIZE

):

    total_rows += len(chunk)


    # ------------------------------------------------------------
    # Missing values
    # ------------------------------------------------------------

    missing_counts = (

        missing_counts

        +

        chunk[
            feature_columns
        ]
        .isna()
        .sum()

    )


    non_missing_counts = (

        non_missing_counts

        +

        chunk[
            feature_columns
        ]
        .notna()
        .sum()

    )


    # ------------------------------------------------------------
    # Class counts
    # ------------------------------------------------------------

    if "Class" in chunk.columns:

        counts = (
            chunk["Class"]
            .value_counts()
        )


        for class_name, value in counts.items():

            class_counts[class_name] = (

                class_counts.get(
                    class_name,
                    0
                )

                +

                int(value)

            )


    # ------------------------------------------------------------
    # Participant counts
    # ------------------------------------------------------------

    if "PersonSrial" in chunk.columns:

        counts = (
            chunk["PersonSrial"]
            .value_counts()
        )


        for participant, value in counts.items():

            participant_counts[participant] = (

                participant_counts.get(
                    participant,
                    0
                )

                +

                int(value)

            )


print()
print(
    "Total rows:",
    total_rows
)

print(
    "Total missing feature cells:",
    int(
        missing_counts.sum()
    )
)


# ================================================================
# MISSING VALUE REPORT
# ================================================================

missing_report = pd.DataFrame({

    "Feature":
        feature_columns,

    "Missing":
        missing_counts.values,

    "Non_Missing":
        non_missing_counts.values

})


missing_report["Missing_Percent"] = (

    missing_report["Missing"]

    /

    total_rows

    *

    100

)


missing_report = (
    missing_report
    .sort_values(
        "Missing_Percent",
        ascending=False
    )
)


missing_report.to_csv(

    DIRS["quality"] /
    "missing_values_by_feature.csv",

    index=False

)


# ================================================================
# MISSING VALUE BAR PLOT
# ================================================================

top_missing = (
    missing_report
    .head(40)
)


plt.figure(
    figsize=(14, 12)
)


sns.barplot(

    data=top_missing,

    y="Feature",

    x="Missing_Percent",

    hue="Feature",

    palette="viridis",

    legend=False

)


plt.title(
    "Top 40 Features by Missing Value Percentage",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Missing Values (%)"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()


plt.savefig(

    DIRS["quality"] /
    "01_top_missing_features.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# MISSING VALUE BY SIGNAL
# ================================================================

signal_missing = (

    missing_report

    .merge(

        feature_info[
            [
                "Column",
                "Signal"
            ]
        ],

        left_on="Feature",

        right_on="Column",

        how="left"

    )

    .groupby(
        "Signal",
        as_index=False
    )

    .agg(

        Missing=(
            "Missing",
            "sum"
        ),

        Non_Missing=(
            "Non_Missing",
            "sum"
        )

    )

)


signal_missing["Missing_Percent"] = (

    signal_missing["Missing"]

    /

    (
        signal_missing["Missing"]

        +

        signal_missing["Non_Missing"]

    )

    *

    100

)


signal_missing = (

    signal_missing

    .sort_values(
        "Missing_Percent",
        ascending=False
    )

)


signal_missing.to_csv(

    DIRS["quality"] /
    "missing_values_by_signal.csv",

    index=False

)


plt.figure(
    figsize=(14, 10)
)


sns.barplot(

    data=signal_missing,

    y="Signal",

    x="Missing_Percent",

    hue="Signal",

    palette="magma",

    legend=False

)


plt.title(
    "Missing Values by Original Signal",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Missing Values (%)"
)

plt.ylabel(
    "Signal"
)

plt.tight_layout()


plt.savefig(

    DIRS["quality"] /
    "02_missing_by_signal.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# MISSING VALUE HEATMAP
# ================================================================

heatmap_data = (

    missing_report

    .merge(

        feature_info[
            [
                "Column",
                "Signal"
            ]
        ],

        left_on="Feature",

        right_on="Column",

        how="left"

    )

)


pivot_missing = (

    heatmap_data

    .pivot_table(

        index="Signal",

        columns="Feature",

        values="Missing_Percent",

        aggfunc="mean"

    )

)


plt.figure(
    figsize=(24, 12)
)


sns.heatmap(

    pivot_missing,

    cmap="magma",

    linewidths=0.1,

    cbar_kws={
        "label":
        "Missing (%)"
    }

)


plt.title(
    "Missing Value Pattern Across Signals and Features",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Extracted Feature"
)

plt.ylabel(
    "Original Signal"
)

plt.tight_layout()


plt.savefig(

    DIRS["quality"] /
    "03_missing_heatmap.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# CLASS DISTRIBUTION
# ================================================================

class_df = pd.DataFrame(

    list(
        class_counts.items()
    ),

    columns=[
        "Class",
        "Rows"
    ]

)


class_df.to_csv(

    DIRS["class"] /
    "class_distribution.csv",

    index=False

)


plt.figure(
    figsize=(10, 7)
)


sns.barplot(

    data=class_df,

    x="Class",

    y="Rows",

    hue="Class",

    palette="Set2",

    legend=False

)


plt.title(
    "Class Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Number of Windows"
)

plt.xticks(
    rotation=20
)

plt.tight_layout()


plt.savefig(

    DIRS["class"] /
    "01_class_distribution.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# PARTICIPANT DISTRIBUTION
# ================================================================

participant_df = pd.DataFrame(

    list(
        participant_counts.items()
    ),

    columns=[
        "PersonSrial",
        "Rows"
    ]

)


participant_df = (
    participant_df
    .sort_values(
        "PersonSrial"
    )
)


participant_df.to_csv(

    DIRS["participant"] /
    "participant_distribution.csv",

    index=False

)


plt.figure(
    figsize=(16, 7)
)


sns.barplot(

    data=participant_df,

    x="PersonSrial",

    y="Rows",

    hue="PersonSrial",

    palette="crest",

    legend=False

)


plt.title(
    "Feature Windows per Participant",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Participant"
)

plt.ylabel(
    "Feature Windows"
)

plt.xticks(
    rotation=90
)

plt.tight_layout()


plt.savefig(

    DIRS["participant"] /
    "01_participant_distribution.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# PASS 2
# CREATE CONTROLLED VISUALIZATION SAMPLE
# ================================================================

print()
print("=" * 80)
print("PASS 2: CREATING VISUALIZATION SAMPLE")
print("=" * 80)


sample_parts = []

remaining_sample = PLOT_SAMPLE_SIZE

rng = np.random.default_rng(
    RANDOM_STATE
)


for chunk in pd.read_csv(

    INPUT_FILE,

    chunksize=CHUNK_SIZE

):

    if remaining_sample <= 0:

        break


    sample_size = min(

        remaining_sample,

        len(chunk)

    )


    selected_indices = rng.choice(

        len(chunk),

        size=sample_size,

        replace=False

    )


    sample_parts.append(

        chunk.iloc[
            selected_indices
        ]

    )


    remaining_sample -= sample_size


sample_df = pd.concat(

    sample_parts,

    ignore_index=True

)


print(
    "Visualization sample:",
    sample_df.shape
)


# ================================================================
# REPRESENTATIVE FEATURES
# ================================================================

representative_features = []


preferred_suffixes = [

    "__Mean",
    "__Standard_Deviation",
    "__RMS",
    "__Energy",
    "__Kurtosis",
    "__Skewness",
    "__Autocorrelation",
    "__PSD_Mean",
    "__Band_Power"

]


for suffix in preferred_suffixes:

    matches = [

        column

        for column in feature_columns

        if column.endswith(suffix)

    ]


    if matches:

        representative_features.extend(
            matches[:4]
        )


representative_features = list(
    dict.fromkeys(
        representative_features
    )
)


representative_features = (
    representative_features[:24]
)


print(
    "Representative features:",
    len(representative_features)
)


# ================================================================
# DISTRIBUTION STATISTICS
# ================================================================

distribution_statistics = []


for feature in feature_columns:

    if feature not in sample_df.columns:

        continue


    values = pd.to_numeric(

        sample_df[feature],

        errors="coerce"

    ).dropna()


    if len(values) < 4:

        continue


    distribution_statistics.append({

        "Feature":
            feature,

        "Skewness":
            skew(values),

        "Kurtosis":
            kurtosis(
                values,
                fisher=True
            ),

        "Mean":
            values.mean(),

        "Median":
            values.median(),

        "Standard_Deviation":
            values.std(),

        "Minimum":
            values.min(),

        "Maximum":
            values.max()

    })


distribution_statistics_df = pd.DataFrame(
    distribution_statistics
)


distribution_statistics_df.to_csv(

    DIRS["distribution"] /
    "feature_distribution_statistics.csv",

    index=False

)


# ================================================================
# SKEWNESS PLOT
# ================================================================

skew_sorted = (

    distribution_statistics_df

    .sort_values(
        "Skewness"
    )

)


skew_plot_data = pd.concat(

    [

        skew_sorted.head(15),

        skew_sorted.tail(15)

    ],

    ignore_index=True

)


plt.figure(
    figsize=(14, 13)
)


sns.barplot(

    data=skew_plot_data,

    x="Skewness",

    y="Feature",

    hue="Feature",

    palette="coolwarm",

    legend=False

)


plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)


plt.title(
    "Most Positively and Negatively Skewed Features",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()


plt.savefig(

    DIRS["distribution"] /
    "01_skewness_extremes.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# KURTOSIS PLOT
# ================================================================

kurt_sorted = (

    distribution_statistics_df

    .sort_values(
        "Kurtosis"
    )

)


kurt_plot_data = pd.concat(

    [

        kurt_sorted.head(15),

        kurt_sorted.tail(15)

    ],

    ignore_index=True

)


plt.figure(
    figsize=(14, 13)
)


sns.barplot(

    data=kurt_plot_data,

    x="Kurtosis",

    y="Feature",

    hue="Feature",

    palette="Spectral",

    legend=False

)


plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)


plt.title(
    "Most Extreme Kurtosis Features",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()


plt.savefig(

    DIRS["distribution"] /
    "02_kurtosis_extremes.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# DISTRIBUTION VIOLIN PLOTS
# ================================================================

for feature in representative_features[:12]:

    if feature not in sample_df.columns:

        continue


    temp = sample_df[
        [
            "Class",
            feature
        ]
    ].dropna()


    if len(temp) == 0:

        continue


    plt.figure(
        figsize=(11, 7)
    )


    sns.violinplot(

        data=temp,

        x="Class",

        y=feature,

        hue="Class",

        palette="Set2",

        inner="quartile",

        legend=False

    )


    plt.title(
        f"Class Wise Distribution: {feature}",
        fontsize=15,
        fontweight="bold"
    )

    plt.xticks(
        rotation=20
    )

    plt.tight_layout()


    safe_name = re.sub(

        r"[^A-Za-z0-9_]+",

        "_",

        feature

    )


    plt.savefig(

        DIRS["distribution"] /
        f"violin_{safe_name}.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# BOXPLOTS
# ================================================================

for feature in representative_features[:12]:

    if feature not in sample_df.columns:

        continue


    temp = sample_df[
        [
            "Class",
            feature
        ]
    ].dropna()


    if len(temp) == 0:

        continue


    plt.figure(
        figsize=(11, 7)
    )


    sns.boxplot(

        data=temp,

        x="Class",

        y=feature,

        hue="Class",

        palette="pastel",

        showfliers=True,

        legend=False

    )


    plt.title(
        f"Class Wise Boxplot: {feature}",
        fontsize=15,
        fontweight="bold"
    )

    plt.xticks(
        rotation=20
    )

    plt.tight_layout()


    safe_name = re.sub(

        r"[^A-Za-z0-9_]+",

        "_",

        feature

    )


    plt.savefig(

        DIRS["distribution"] /
        f"boxplot_{safe_name}.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# OUTLIER ANALYSIS
# ================================================================

outlier_records = []


for feature in representative_features:

    if feature not in sample_df.columns:

        continue


    values = pd.to_numeric(

        sample_df[feature],

        errors="coerce"

    ).dropna()


    if len(values) < 10:

        continue


    median_value = (
        values.median()
    )


    mad = np.median(

        np.abs(

            values -
            median_value

        )

    )


    if mad == 0:

        outlier_rate = 0.0

    else:

        modified_z = (

            0.6745

            *

            (

                values -
                median_value

            )

            /

            mad

        )


        outlier_rate = (

            np.abs(
                modified_z
            )

            >

            OUTLIER_Z_THRESHOLD

        ).mean() * 100


    outlier_records.append({

        "Feature":
            feature,

        "Median":
            median_value,

        "MAD":
            mad,

        "Potential_Outlier_Percent":
            outlier_rate

    })


outlier_df = pd.DataFrame(
    outlier_records
)


outlier_df = (

    outlier_df

    .sort_values(

        "Potential_Outlier_Percent",

        ascending=False

    )

)


outlier_df.to_csv(

    DIRS["outlier"] /
    "representative_feature_outlier_report.csv",

    index=False

)


# ================================================================
# OUTLIER RATE PLOT
# ================================================================

plt.figure(
    figsize=(14, 10)
)


sns.barplot(

    data=outlier_df,

    x="Potential_Outlier_Percent",

    y="Feature",

    hue="Feature",

    palette="rocket",

    legend=False

)


plt.title(
    "Potential Outlier Rate Using Modified Z Score",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Potential Outliers (%)"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()


plt.savefig(

    DIRS["outlier"] /
    "01_outlier_rate.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# SELECTED OUTLIER BOXPLOTS
# ================================================================

selected_outlier_features = (

    outlier_df

    .head(10)

    ["Feature"]

    .tolist()

)


if selected_outlier_features:

    melt_outlier = (

        sample_df[
            selected_outlier_features
        ]

        .melt(

            var_name="Feature",

            value_name="Value"

        )

    )


    plt.figure(
        figsize=(16, 9)
    )


    sns.boxplot(

        data=melt_outlier,

        x="Feature",

        y="Value",

        showfliers=True

    )


    plt.xticks(

        rotation=75,

        ha="right"

    )


    plt.title(
        "Potential Outlier Distribution of Selected Features",
        fontsize=16,
        fontweight="bold"
    )

    plt.tight_layout()


    plt.savefig(

        DIRS["outlier"] /
        "02_selected_feature_boxplots.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# PARTICIPANT × CLASS HEATMAP
# ================================================================

participant_class = (

    sample_df

    .groupby(

        [
            "PersonSrial",
            "Class"
        ]

    )

    .size()

    .unstack(
        fill_value=0
    )

)


plt.figure(
    figsize=(12, 18)
)


sns.heatmap(

    participant_class,

    cmap="YlGnBu",

    annot=False

)


plt.title(
    "Participant × Class Feature Window Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Participant"
)

plt.tight_layout()


plt.savefig(

    DIRS["participant"] /
    "02_participant_class_heatmap.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# PARTICIPANT FEATURE PATTERN
# ================================================================

participant_features = (

    sample_df

    .groupby(
        "PersonSrial"
    )[

        representative_features

    ]

    .mean()

)


pattern_features = (
    representative_features[:6]
)


plt.figure(
    figsize=(16, 8)
)


for feature in pattern_features:

    values = (

        participant_features[
            feature
        ]

    )


    standard_deviation = values.std()


    if standard_deviation == 0:

        standardized_values = (
            values -
            values.mean()
        )

    else:

        standardized_values = (

            values -
            values.mean()

        ) / standard_deviation


    plt.plot(

        standardized_values.index,

        standardized_values.values,

        marker="o",

        markersize=3,

        linewidth=1.3,

        label=feature

    )


plt.title(
    "Participant Wise Standardized Feature Pattern",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Participant"
)

plt.ylabel(
    "Standardized Feature Value"
)

plt.legend(

    bbox_to_anchor=(
        1.02,
        1
    ),

    loc="upper left"

)

plt.tight_layout()


plt.savefig(

    DIRS["participant"] /
    "03_participant_feature_pattern.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# PARTICIPANT VARIABILITY
# ================================================================

participant_std = (

    sample_df

    .groupby(
        "PersonSrial"
    )[

        representative_features

    ]

    .std()

)


participant_variability = (

    participant_std

    .mean(
        axis=1
    )

    .reset_index(
        name="Mean_Feature_SD"
    )

)


plt.figure(
    figsize=(15, 7)
)


sns.barplot(

    data=participant_variability,

    x="PersonSrial",

    y="Mean_Feature_SD",

    hue="PersonSrial",

    palette="flare",

    legend=False

)


plt.title(
    "Participant Wise Feature Variability",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Participant"
)

plt.ylabel(
    "Mean Feature Standard Deviation"
)

plt.xticks(
    rotation=90
)

plt.tight_layout()


plt.savefig(

    DIRS["participant"] /
    "04_participant_variability.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# CLASS FEATURE HEATMAP
# ================================================================

class_means = (

    sample_df

    .groupby(
        "Class"
    )[

        representative_features

    ]

    .mean()

)


class_std = class_means.std(
    axis=0
)


class_std[
    class_std == 0
] = 1


class_means_scaled = (

    class_means -
    class_means.mean(
        axis=0
    )

) / class_std


plt.figure(
    figsize=(20, 9)
)


sns.heatmap(

    class_means_scaled.T,

    cmap="vlag",

    center=0,

    linewidths=0.2

)


plt.title(
    "Class Wise Standardized Feature Profile",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()


plt.savefig(

    DIRS["class"] /
    "02_class_feature_heatmap.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# CLASS RADAR / SPIDER PLOT
# ================================================================

radar_features = []


radar_suffixes = [

    "__Mean",
    "__Standard_Deviation",
    "__RMS",
    "__Energy",
    "__Kurtosis",
    "__Skewness",
    "__Autocorrelation",
    "__Band_Power"

]


for suffix in radar_suffixes:

    matches = [

        column

        for column in representative_features

        if column.endswith(suffix)

    ]


    if matches:

        radar_features.append(
            matches[0]
        )


if len(radar_features) >= 3:

    radar_data = (

        sample_df

        .groupby(
            "Class"
        )[

            radar_features

        ]

        .mean()

    )


    radar_std = radar_data.std(
        axis=0
    )


    radar_std[
        radar_std == 0
    ] = 1


    radar_scaled = (

        radar_data -
        radar_data.mean(
            axis=0
        )

    ) / radar_std


    radar_labels = [

        feature.split(
            "__"
        )[-1]

        for feature in radar_features

    ]


    angles = np.linspace(

        0,

        2 * np.pi,

        len(radar_labels),

        endpoint=False

    )


    angles = np.concatenate(

        [
            angles,
            [angles[0]]
        ]

    )


    fig = plt.figure(
        figsize=(11, 11)
    )


    ax = fig.add_subplot(
        111,
        polar=True
    )


    for class_name in radar_scaled.index:

        values = (

            radar_scaled
            .loc[class_name]
            .values

        )


        values = np.concatenate(

            [
                values,
                [values[0]]
            ]

        )


        ax.plot(

            angles,

            values,

            linewidth=2,

            label=class_name

        )


        ax.fill(

            angles,

            values,

            alpha=0.08

        )


    ax.set_xticks(
        angles[:-1]
    )


    ax.set_xticklabels(
        radar_labels
    )


    ax.set_title(

        "Class Wise Multivariate Feature Profile",

        pad=30,

        fontsize=16,

        fontweight="bold"

    )


    ax.legend(

        bbox_to_anchor=(
            1.25,
            1.10
        )

    )


    plt.tight_layout()


    plt.savefig(

        DIRS["class"] /
        "03_class_radar_spider.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# FEATURE FAMILY RADAR
# ================================================================

family_values = {}


for family, feature_list in FEATURE_FAMILIES.items():

    available_columns = [

        column

        for column in feature_columns

        if feature_info.loc[
            feature_info["Column"]
            ==
            column,
            "Feature"
        ].iloc[0]
        in feature_list

    ]


    if available_columns:

        family_values[family] = (

            sample_df[
                available_columns
            ]

            .mean(
                axis=1
            )

        )


family_df = pd.DataFrame(
    family_values
)


family_df["Class"] = (
    sample_df["Class"]
    .values
)


family_means = (

    family_df

    .groupby(
        "Class"
    )

    .mean()

)


family_std = family_means.std(
    axis=0
)


family_std[
    family_std == 0
] = 1


family_scaled = (

    family_means -
    family_means.mean(
        axis=0
    )

) / family_std


if len(family_scaled.columns) >= 3:

    labels = list(
        family_scaled.columns
    )


    angles = np.linspace(

        0,

        2 * np.pi,

        len(labels),

        endpoint=False

    )


    angles = np.concatenate(

        [
            angles,
            [angles[0]]
        ]

    )


    fig = plt.figure(
        figsize=(10, 10)
    )


    ax = fig.add_subplot(
        111,
        polar=True
    )


    for class_name in family_scaled.index:

        values = (

            family_scaled
            .loc[class_name]
            .values

        )


        values = np.concatenate(

            [
                values,
                [values[0]]
            ]

        )


        ax.plot(

            angles,

            values,

            linewidth=2,

            label=class_name

        )


        ax.fill(

            angles,

            values,

            alpha=0.08

        )


    ax.set_xticks(
        angles[:-1]
    )


    ax.set_xticklabels(
        labels
    )


    ax.set_title(

        "Class Wise Feature Family Profile",

        pad=30,

        fontsize=16,

        fontweight="bold"

    )


    ax.legend(

        bbox_to_anchor=(
            1.25,
            1.10
        )

    )


    plt.tight_layout()


    plt.savefig(

        DIRS["advanced"] /
        "01_feature_family_radar.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# CORRELATION MATRIX
# ================================================================

print()
print("=" * 80)
print("CORRELATION ANALYSIS")
print("=" * 80)


corr_features = representative_features[:30]


correlation_matrix = (

    sample_df[
        corr_features
    ]

    .corr()

)


correlation_matrix.to_csv(

    DIRS["correlation"] /
    "representative_feature_correlation.csv"

)


plt.figure(
    figsize=(18, 15)
)


sns.heatmap(

    correlation_matrix,

    cmap="coolwarm",

    center=0,

    linewidths=0.1,

    cbar_kws={
        "label":
        "Pearson Correlation"
    }

)


plt.title(
    "Correlation Matrix of Representative Features",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()


plt.savefig(

    DIRS["correlation"] /
    "01_representative_feature_correlation.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# SIGNAL LEVEL CORRELATION
# ================================================================

signal_summary = {}


for signal in sorted(
    feature_info["Signal"].unique()
):

    signal_columns = (

        feature_info[

            feature_info["Signal"]
            ==
            signal

        ]["Column"]

        .tolist()

    )


    available_columns = [

        column

        for column in signal_columns

        if column in sample_df.columns

    ]


    if available_columns:

        signal_summary[signal] = (

            sample_df[
                available_columns
            ]

            .mean(
                axis=1
            )

        )


if len(signal_summary) >= 2:

    signal_matrix = pd.DataFrame(
        signal_summary
    )


    signal_correlation = (
        signal_matrix.corr()
    )


    signal_correlation.to_csv(

        DIRS["correlation"] /
        "signal_level_correlation.csv"

    )


    plt.figure(
        figsize=(16, 13)
    )


    sns.heatmap(

        signal_correlation,

        cmap="vlag",

        center=0,

        linewidths=0.2,

        annot=False

    )


    plt.title(
        "Correlation Between Original Signal Families",
        fontsize=16,
        fontweight="bold"
    )

    plt.tight_layout()


    plt.savefig(

        DIRS["correlation"] /
        "02_signal_level_correlation.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# PCA
# ================================================================

print()
print("=" * 80)
print("PCA ANALYSIS")
print("=" * 80)


# ------------------------------------------------
# Select numeric feature columns
# ------------------------------------------------

pca_features = [

    column

    for column in feature_columns

    if column in sample_df.columns

]


pca_features = (

    sample_df[
        pca_features
    ]

    .select_dtypes(
        include=np.number
    )

    .columns

    .tolist()

)


print(
    "Number of PCA features:",
    len(pca_features)
)


# ------------------------------------------------
# Select PCA sample
# ------------------------------------------------

pca_input = sample_df[
    pca_features
].copy()


if len(pca_input) > PCA_SAMPLE_SIZE:

    pca_indices = (

        pca_input

        .sample(

            PCA_SAMPLE_SIZE,

            random_state=RANDOM_STATE

        )

        .index

    )

    pca_input = pca_input.loc[
        pca_indices
    ]

else:

    pca_indices = pca_input.index


print(
    "PCA sample size:",
    len(pca_input)
)


# ------------------------------------------------
# Preserve exact metadata rows
# ------------------------------------------------

pca_metadata = sample_df.loc[

    pca_indices,

    [
        column
        for column in [
            "Class",
            "PersonSrial"
        ]
        if column in sample_df.columns
    ]

].copy()


# ------------------------------------------------
# Numeric conversion
# ------------------------------------------------

pca_input = pca_input.apply(

    pd.to_numeric,

    errors="coerce"

)


# ------------------------------------------------
# Remove completely missing features
# ------------------------------------------------

valid_pca_columns = (

    pca_input

    .columns[

        pca_input.notna()
        .any()

    ]

    .tolist()

)


pca_input = pca_input[
    valid_pca_columns
]


print(
    "Valid PCA features:",
    len(valid_pca_columns)
)


# ------------------------------------------------
# Median imputation
# ------------------------------------------------

imputer = SimpleImputer(
    strategy="median"
)


X_imputed = imputer.fit_transform(
    pca_input
)


# ------------------------------------------------
# Standardization
# ------------------------------------------------

scaler = StandardScaler()


X_scaled = scaler.fit_transform(
    X_imputed
)


# ------------------------------------------------
# PCA 2D
# ------------------------------------------------

pca = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)


X_pca = pca.fit_transform(
    X_scaled
)


# ------------------------------------------------
# Create PCA dataframe
# ------------------------------------------------

pca_result = pd.DataFrame(

    X_pca,

    columns=[
        "PC1",
        "PC2"
    ],

    index=pca_indices

)


# ------------------------------------------------
# Add exact metadata
# ------------------------------------------------

if "Class" in pca_metadata.columns:

    pca_result["Class"] = (
        pca_metadata["Class"]
        .values
    )


if "PersonSrial" in pca_metadata.columns:

    pca_result["PersonSrial"] = (
        pca_metadata["PersonSrial"]
        .values
    )


pca_result = (
    pca_result
    .reset_index(
        drop=True
    )
)


# ------------------------------------------------
# Explained variance
# ------------------------------------------------

pc1_variance = (

    pca.explained_variance_ratio_[0]

    *

    100

)


pc2_variance = (

    pca.explained_variance_ratio_[1]

    *

    100

)


total_variance = (

    pc1_variance
    +
    pc2_variance

)


print()
print(
    f"PC1 explained variance: "
    f"{pc1_variance:.3f}%"
)

print(
    f"PC2 explained variance: "
    f"{pc2_variance:.3f}%"
)

print(
    f"PC1 + PC2 explained variance: "
    f"{total_variance:.3f}%"
)


# ================================================================
# SAVE PCA DATA
# ================================================================

pca_result.to_csv(

    DIRS["advanced"] /
    "PCA_coordinates.csv",

    index=False

)


# ================================================================
# PCA CLASS SCATTER
# ================================================================

if "Class" in pca_result.columns:

    plt.figure(
        figsize=(13, 10)
    )


    sns.scatterplot(

        data=pca_result,

        x="PC1",

        y="PC2",

        hue="Class",

        palette="Set2",

        alpha=0.45,

        s=35,

        edgecolor=None

    )


    plt.title(

        "PCA Projection of V25 Feature Space",

        fontsize=16,

        fontweight="bold"

    )


    plt.xlabel(

        f"PC1 ({pc1_variance:.2f}% variance)"

    )


    plt.ylabel(

        f"PC2 ({pc2_variance:.2f}% variance)"

    )


    plt.legend(

        title="Class",

        bbox_to_anchor=(
            1.02,
            1
        ),

        loc="upper left"

    )


    plt.tight_layout()


    plt.savefig(

        DIRS["advanced"] /
        "02_PCA_class_scatter.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# PCA PARTICIPANT SCATTER
# ================================================================

if "PersonSrial" in pca_result.columns:

    plt.figure(
        figsize=(14, 10)
    )


    sns.scatterplot(

        data=pca_result,

        x="PC1",

        y="PC2",

        hue="PersonSrial",

        palette="turbo",

        alpha=0.45,

        s=30,

        legend=False

    )


    plt.title(

        "PCA Projection Colored by Participant",

        fontsize=16,

        fontweight="bold"

    )


    plt.xlabel(

        f"PC1 ({pc1_variance:.2f}% variance)"

    )


    plt.ylabel(

        f"PC2 ({pc2_variance:.2f}% variance)"

    )


    plt.tight_layout()


    plt.savefig(

        DIRS["advanced"] /
        "03_PCA_participant_scatter.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# PCA CLASS CENTROIDS
# ================================================================

if "Class" in pca_result.columns:

    centroids = (

        pca_result

        .groupby(
            "Class"
        )[

            [
                "PC1",
                "PC2"
            ]

        ]

        .mean()

        .reset_index()

    )


    plt.figure(
        figsize=(13, 10)
    )


    sns.scatterplot(

        data=pca_result,

        x="PC1",

        y="PC2",

        hue="Class",

        palette="Set2",

        alpha=0.15,

        s=25,

        legend=False

    )


    sns.scatterplot(

        data=centroids,

        x="PC1",

        y="PC2",

        hue="Class",

        palette="Set2",

        s=250,

        marker="X",

        edgecolor="black",

        linewidth=1.2

    )


    for _, row in centroids.iterrows():

        plt.annotate(

            row["Class"],

            (
                row["PC1"],
                row["PC2"]
            ),

            xytext=(8, 8),

            textcoords="offset points",

            fontsize=11,

            fontweight="bold"

        )


    plt.title(

        "PCA Class Distribution and Class Centroids",

        fontsize=16,

        fontweight="bold"

    )


    plt.xlabel(

        f"PC1 ({pc1_variance:.2f}% variance)"

    )


    plt.ylabel(

        f"PC2 ({pc2_variance:.2f}% variance)"

    )


    plt.tight_layout()


    plt.savefig(

        DIRS["advanced"] /
        "04_PCA_class_centroids.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# PCA DENSITY
# ================================================================

if "Class" in pca_result.columns:

    plt.figure(
        figsize=(13, 10)
    )


    sns.kdeplot(

        data=pca_result,

        x="PC1",

        y="PC2",

        hue="Class",

        levels=8,

        thresh=0.05,

        palette="Set2"

    )


    plt.title(

        "PCA Class Density Contours",

        fontsize=16,

        fontweight="bold"

    )


    plt.xlabel(

        f"PC1 ({pc1_variance:.2f}% variance)"

    )


    plt.ylabel(

        f"PC2 ({pc2_variance:.2f}% variance)"

    )


    plt.tight_layout()


    plt.savefig(

        DIRS["advanced"] /
        "05_PCA_class_density.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# PCA CUMULATIVE EXPLAINED VARIANCE
# ================================================================

maximum_components = min(

    20,

    X_scaled.shape[0],

    X_scaled.shape[1]

)


pca_full = PCA(
    n_components=maximum_components
)


pca_full.fit(
    X_scaled
)


cumulative_variance = np.cumsum(

    pca_full
    .explained_variance_ratio_

)


plt.figure(
    figsize=(12, 7)
)


plt.plot(

    range(
        1,
        len(cumulative_variance) + 1
    ),

    cumulative_variance * 100,

    marker="o",

    linewidth=2

)


plt.axhline(

    90,

    linestyle="--",

    label="90%"

)


plt.axhline(

    95,

    linestyle="--",

    label="95%"

)


plt.xlabel(
    "Number of Principal Components"
)

plt.ylabel(
    "Cumulative Explained Variance (%)"
)

plt.title(

    "PCA Cumulative Explained Variance",

    fontsize=16,

    fontweight="bold"

)

plt.legend()

plt.tight_layout()


plt.savefig(

    DIRS["advanced"] /
    "06_PCA_cumulative_variance.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# PCA SUMMARY
# ================================================================

pca_summary = pd.DataFrame({

    "Component": [
        "PC1",
        "PC2",
        "PC1 + PC2"
    ],

    "Explained_Variance_Percent": [

        pc1_variance,

        pc2_variance,

        total_variance

    ]

})


pca_summary.to_csv(

    DIRS["advanced"] /
    "PCA_explained_variance.csv",

    index=False

)


# ================================================================
# ADVANCED PAIR PLOT
# ================================================================

print()
print("=" * 80)
print("ADVANCED PAIR PLOT")
print("=" * 80)


pair_features = (
    representative_features[:4]
)


if len(pair_features) >= 2:

    pair_columns = [

        "Class"

    ] + pair_features


    pair_data = (

        sample_df[
            pair_columns
        ]

        .dropna()

    )


    if len(pair_data) > PAIRPLOT_SAMPLE_SIZE:

        pair_data = (

            pair_data

            .sample(

                PAIRPLOT_SAMPLE_SIZE,

                random_state=RANDOM_STATE

            )

        )


    pair_grid = sns.pairplot(

        pair_data,

        hue="Class",

        palette="Set2",

        corner=True,

        plot_kws={

            "alpha": 0.35,

            "s": 20

        }

    )


    pair_grid.fig.suptitle(

        "Multivariate Feature Relationships by Class",

        y=1.02,

        fontsize=16,

        fontweight="bold"

    )


    pair_grid.savefig(

        DIRS["advanced"] /
        "07_advanced_pairplot.png",

        dpi=300

    )


    plt.close("all")


# ================================================================
# FEATURE FAMILY SUMMARY
# ================================================================

family_summary = (

    feature_info

    .groupby(
        "Family"
    )

    .size()

    .reset_index(
        name="Number_of_Features"
    )

)


family_summary.to_csv(

    DIRS["summary"] /
    "feature_family_summary.csv",

    index=False

)


plt.figure(
    figsize=(9, 7)
)


sns.barplot(

    data=family_summary,

    x="Family",

    y="Number_of_Features",

    hue="Family",

    palette="Set3",

    legend=False

)


plt.title(

    "Extracted Feature Families",

    fontsize=16,

    fontweight="bold"

)


plt.tight_layout()


plt.savefig(

    DIRS["summary"] /
    "feature_family_distribution.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# CLASS × FEATURE FAMILY HEATMAP
# ================================================================

plt.figure(
    figsize=(10, 6)
)


sns.heatmap(

    family_scaled.T,

    cmap="coolwarm",

    center=0,

    annot=True,

    fmt=".2f"

)


plt.title(

    "Standardized Feature Family Profile by Class",

    fontsize=16,

    fontweight="bold"

)


plt.xlabel(
    "Class"
)

plt.ylabel(
    "Feature Family"
)

plt.tight_layout()


plt.savefig(

    DIRS["advanced"] /
    "08_feature_family_heatmap.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# DATASET SUMMARY
# ================================================================

summary = pd.DataFrame({

    "Parameter": [

        "Input rows",

        "Input columns",

        "Feature columns",

        "Participants",

        "Classes",

        "Total missing feature cells",

        "Visualization sample",

        "PCA sample",

        "Sampling rate",

        "Window duration",

        "Step duration",

        "Potential outlier threshold"

    ],

    "Value": [

        total_rows,

        len(columns),

        len(feature_columns),

        len(participant_counts),

        len(class_counts),

        int(
            missing_counts.sum()
        ),

        len(sample_df),

        len(pca_result),

        "1 Hz",

        "10 seconds",

        "5 seconds",

        OUTLIER_Z_THRESHOLD

    ]

})


summary.to_csv(

    DIRS["summary"] /
    "visualization_summary.csv",

    index=False

)


# ================================================================
# FINAL REPORT
# ================================================================

print()
print()
print("=" * 80)
print("V25 VISUAL ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 80)

print()

print(
    "Input rows:",
    total_rows
)

print(
    "Input columns:",
    len(columns)
)

print(
    "Feature columns:",
    len(feature_columns)
)

print(
    "Participants:",
    len(participant_counts)
)

print(
    "Classes:",
    len(class_counts)
)

print(
    "Missing feature cells:",
    int(
        missing_counts.sum()
    )
)

print(
    "Visualization sample:",
    len(sample_df)
)

print(
    "PCA sample:",
    len(pca_result)
)

print()

print(
    "PC1 explained variance:",
    f"{pc1_variance:.3f}%"
)

print(
    "PC2 explained variance:",
    f"{pc2_variance:.3f}%"
)

print(
    "PC1 + PC2:",
    f"{total_variance:.3f}%"
)

print()

print(
    "Output folder:"
)

print(
    OUTPUT_ROOT
)

print()

print(
    "The original V25 dataset was not modified."
)

print(
    "Potential outliers were identified only."
)

print(
    "Missing values were analyzed only."
)

print(
    "No automatic deletion of outliers was performed."
)

print(
    "No automatic imputation of the V25 dataset was performed."
)

print()

print("=" * 80)
print("PROCESS FINISHED")
print("=" * 80)
```

```python
# ================================================================
# V25 TO V26
# COMPLETE FEATURE QUALITY CONTROL AND STATISTICAL ANALYSIS
# ================================================================
#
# INPUT:
# /home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V25/
# V25_All_Signals_21_Features.csv
#
# OUTPUT:
# /home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V26/
#
# ================================================================
#
# STAGE A
# Missing value investigation
#
# STAGE B
# Conservative missing value treatment
#
# STAGE C
# Conservative outlier assessment
#
# STAGE D
# Feature quality screening
#
# STAGE E
# Participant level analysis
#
# STAGE F
# Statistical significance testing
#
# STAGE G
# Multiple comparison correction
#
# STAGE H
# Effect size analysis
#
# STAGE I
# Final V26 dataset
#
# ================================================================


# ================================================================
# REQUIRED LIBRARIES
# ================================================================

import os
import re
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import (
    skew,
    kurtosis,
    friedmanchisquare,
    wilcoxon
)

from statsmodels.stats.multitest import multipletests


warnings.filterwarnings("ignore")


# ================================================================
# DISPLAY SETTINGS
# ================================================================

pd.set_option(
    "display.max_columns",
    50
)

pd.set_option(
    "display.width",
    180
)

pd.set_option(
    "display.max_colwidth",
    100
)


# ================================================================
# VISUAL SETTINGS
# ================================================================

sns.set_theme(
    style="whitegrid",
    context="notebook"
)

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300


# ================================================================
# INPUT AND OUTPUT PATH
# ================================================================

INPUT_FILE = (
    "/home/rakib/Desktop/RAW_DATA/"
    "CLEAN_DATA_V25/"
    "V25_All_Signals_21_Features.csv"
)


OUTPUT_ROOT = Path(
    "/home/rakib/Desktop/RAW_DATA/"
    "CLEAN_DATA_V26"
)


# ================================================================
# OUTPUT DIRECTORIES
# ================================================================

OUTPUT_DIRS = {

    "missing":
        OUTPUT_ROOT / "01_Missing_Value_Analysis",

    "imputation":
        OUTPUT_ROOT / "02_Missing_Value_Treatment",

    "outlier":
        OUTPUT_ROOT / "03_Outlier_Assessment",

    "quality":
        OUTPUT_ROOT / "04_Feature_Quality",

    "participant":
        OUTPUT_ROOT / "05_Participant_Analysis",

    "statistics":
        OUTPUT_ROOT / "06_Statistical_Testing",

    "visualization":
        OUTPUT_ROOT / "07_Visualizations",

    "dataset":
        OUTPUT_ROOT / "08_Final_Dataset",

    "summary":
        OUTPUT_ROOT / "09_Summary"

}


for directory in OUTPUT_DIRS.values():

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ================================================================
# ANALYSIS PARAMETERS
# ================================================================

CHUNK_SIZE = 10000

RANDOM_STATE = 42

MAX_INTERPOLATION_GAP = 3

OUTLIER_THRESHOLD = 3.5

NEAR_ZERO_VARIANCE_THRESHOLD = 0.01

HIGH_MISSINGNESS_THRESHOLD = 30.0

EXTREME_SKEWNESS_THRESHOLD = 3.0

EXTREME_KURTOSIS_THRESHOLD = 10.0

HIGH_CORRELATION_THRESHOLD = 0.95

CORRELATION_SAMPLE_SIZE = 50000


# ================================================================
# CHECK INPUT FILE
# ================================================================

if not os.path.isfile(INPUT_FILE):

    raise FileNotFoundError(
        "\nInput dataset was not found:\n\n"
        + INPUT_FILE
    )


# ================================================================
# START
# ================================================================

print()
print("=" * 80)
print("V25 TO V26 FEATURE QUALITY CONTROL")
print("=" * 80)

print()
print("Input file:")
print(INPUT_FILE)

print()
print("Output directory:")
print(OUTPUT_ROOT)


# ================================================================
# READ ONLY HEADER FIRST
# ================================================================

header_df = pd.read_csv(
    INPUT_FILE,
    nrows=0
)

ALL_COLUMNS = list(
    header_df.columns
)


print()
print("=" * 80)
print("DATASET STRUCTURE")
print("=" * 80)

print(
    "Total columns:",
    len(ALL_COLUMNS)
)


# ================================================================
# DETECT PARTICIPANT COLUMN
# ================================================================

participant_candidates = [

    "PersonSrial",
    "PersonSerial",
    "Person_sl",
    "Participant",
    "ParticipantID",
    "Participant_ID"

]


PARTICIPANT_COLUMN = None


for candidate in participant_candidates:

    if candidate in ALL_COLUMNS:

        PARTICIPANT_COLUMN = candidate

        break


if PARTICIPANT_COLUMN is None:

    raise ValueError(
        "\nParticipant column was not found.\n"
        "Expected column such as PersonSrial."
    )


# ================================================================
# DETECT CLASS COLUMN
# ================================================================

class_candidates = [

    "Class",
    "class",
    "Label",
    "Activity_Class"

]


CLASS_COLUMN = None


for candidate in class_candidates:

    if candidate in ALL_COLUMNS:

        CLASS_COLUMN = candidate

        break


if CLASS_COLUMN is None:

    raise ValueError(
        "\nClass column was not found."
    )


# ================================================================
# DETECT WINDOW COLUMN
# ================================================================

window_candidates = [

    "Window",
    "window",
    "Window_Number"

]


WINDOW_COLUMN = None


for candidate in window_candidates:

    if candidate in ALL_COLUMNS:

        WINDOW_COLUMN = candidate

        break


# ================================================================
# DETECT TIME RELATED COLUMNS
# ================================================================

time_candidates = [

    "TimeStamp",
    "Timestamp",
    "timestamp",
    "Window_Start",
    "Window_End"

]


TIME_COLUMNS = [

    column

    for column in time_candidates

    if column in ALL_COLUMNS

]


# ================================================================
# METADATA COLUMNS
# ================================================================

metadata_candidates = [

    PARTICIPANT_COLUMN,
    CLASS_COLUMN,
    WINDOW_COLUMN,
    "Window_Start",
    "Window_End",
    "Window_Samples",
    "TimeStamp",
    "Timestamp",
    "timestamp"

]


metadata_columns = [

    column

    for column in metadata_candidates

    if column in ALL_COLUMNS

]


# Remove duplicates while preserving order

metadata_columns = list(
    dict.fromkeys(
        metadata_columns
    )
)


# ================================================================
# FEATURE COLUMNS
# ================================================================

feature_columns = [

    column

    for column in ALL_COLUMNS

    if column not in metadata_columns

]


print()
print("=" * 80)
print("COLUMN DETECTION")
print("=" * 80)

print(
    "Participant column:",
    PARTICIPANT_COLUMN
)

print(
    "Class column:",
    CLASS_COLUMN
)

print(
    "Window column:",
    WINDOW_COLUMN
)

print(
    "Time related columns:",
    TIME_COLUMNS
)

print(
    "Metadata columns:",
    len(metadata_columns)
)

print(
    "Feature columns:",
    len(feature_columns)
)


# ================================================================
# FEATURE MAP
# ================================================================

feature_map_records = []


for feature in feature_columns:

    signal_name = "Unknown"

    channel_name = "Unknown"

    extracted_feature = feature


    # ------------------------------------------------------------
    # Detect the signal name
    # ------------------------------------------------------------

    if "__" in feature:

        parts = feature.split(
            "__"
        )

        signal_name = parts[0]

        extracted_feature = "__".join(
            parts[1:]
        )


    elif "_" in feature:

        parts = feature.split("_")

        if len(parts) >= 2:

            signal_name = parts[0]


    # ------------------------------------------------------------
    # Detect channel
    # ------------------------------------------------------------

    channel_patterns = [

        "TP9",
        "TP10",
        "AF7",
        "AF8",
        "AUX_RIGHT",
        "AUX_LEFT",
        "Accelerometer_X",
        "Accelerometer_Y",
        "Accelerometer_Z",
        "Gyro_X",
        "Gyro_Y",
        "Gyro_Z"

    ]


    for channel in channel_patterns:

        if channel in feature:

            channel_name = channel

            break


    feature_map_records.append({

        "Feature":
            feature,

        "Signal":
            signal_name,

        "Channel":
            channel_name,

        "Extracted_Feature":
            extracted_feature

    })


feature_map = pd.DataFrame(
    feature_map_records
)


feature_map.to_csv(

    OUTPUT_DIRS["summary"] /
    "feature_map.csv",

    index=False

)


# ================================================================
# STAGE A
# MISSING VALUE INVESTIGATION
# ================================================================

print()
print("=" * 80)
print("STAGE A: MISSING VALUE INVESTIGATION")
print("=" * 80)


missing_counts = pd.Series(

    0,

    index=feature_columns,

    dtype="int64"

)


total_rows = 0


participant_missing = {}

class_missing = {}


# ================================================================
# CHUNKED MISSING VALUE ANALYSIS
# ================================================================

for chunk_number, chunk in enumerate(

    pd.read_csv(
        INPUT_FILE,
        chunksize=CHUNK_SIZE
    ),

    start=1

):

    total_rows += len(chunk)


    # ------------------------------------------------------------
    # Make sure feature columns are numeric for missing detection
    # ------------------------------------------------------------

    feature_part = chunk[
        feature_columns
    ]


    missing_counts += (

        feature_part

        .isna()

        .sum()

    )


    # ------------------------------------------------------------
    # Missing values by participant
    # ------------------------------------------------------------

    row_missing = (

        feature_part

        .isna()

        .sum(axis=1)

    )


    participant_values = (

        chunk[
            PARTICIPANT_COLUMN
        ]

    )


    temporary = pd.DataFrame({

        "Participant":
            participant_values,

        "Missing":
            row_missing

    })


    grouped = (

        temporary

        .groupby(
            "Participant"
        )["Missing"]

        .sum()

    )


    for participant, value in grouped.items():

        participant_missing[participant] = (

            participant_missing.get(
                participant,
                0
            )

            +

            int(value)

        )


    # ------------------------------------------------------------
    # Missing values by class
    # ------------------------------------------------------------

    temporary = pd.DataFrame({

        "Class":
            chunk[
                CLASS_COLUMN
            ],

        "Missing":
            row_missing

    })


    grouped = (

        temporary

        .groupby(
            "Class"
        )["Missing"]

        .sum()

    )


    for class_name, value in grouped.items():

        class_missing[class_name] = (

            class_missing.get(
                class_name,
                0
            )

            +

            int(value)

        )


print()
print(
    "Total rows:",
    total_rows
)

print(
    "Total feature cells:",
    total_rows * len(feature_columns)
)

print(
    "Total missing feature cells:",
    int(
        missing_counts.sum()
    )
)


overall_missing_percentage = (

    missing_counts.sum()

    /

    (
        total_rows
        *
        len(feature_columns)
    )

    *

    100

)


print(
    "Overall missing percentage:",
    round(
        overall_missing_percentage,
        4
    ),
    "%"
)


# ================================================================
# FEATURE MISSING REPORT
# ================================================================

missing_report = pd.DataFrame({

    "Feature":
        feature_columns,

    "Missing_Count":
        [
            int(
                missing_counts[
                    feature
                ]
            )

            for feature in feature_columns
        ]

})


missing_report["Missing_Percent"] = (

    missing_report["Missing_Count"]

    /

    total_rows

    *

    100

)


missing_report = (

    missing_report

    .merge(

        feature_map,

        on="Feature",

        how="left"

    )

)


missing_report = (

    missing_report

    .sort_values(

        "Missing_Percent",

        ascending=False

    )

    .reset_index(
        drop=True
    )

)


missing_report.to_csv(

    OUTPUT_DIRS["missing"] /
    "feature_missing_report.csv",

    index=False

)


# ================================================================
# PARTICIPANT MISSING REPORT
# ================================================================

participant_missing_df = pd.DataFrame({

    PARTICIPANT_COLUMN:
        list(
            participant_missing.keys()
        ),

    "Missing_Cells":
        list(
            participant_missing.values()
        )

})


participant_missing_df = (

    participant_missing_df

    .sort_values(
        "Missing_Cells",
        ascending=False
    )

)


participant_missing_df.to_csv(

    OUTPUT_DIRS["missing"] /
    "participant_missing_report.csv",

    index=False

)


# ================================================================
# CLASS MISSING REPORT
# ================================================================

class_missing_df = pd.DataFrame({

    CLASS_COLUMN:
        list(
            class_missing.keys()
        ),

    "Missing_Cells":
        list(
            class_missing.values()
        )

})


class_missing_df = (

    class_missing_df

    .sort_values(
        "Missing_Cells",
        ascending=False
    )

)


class_missing_df.to_csv(

    OUTPUT_DIRS["missing"] /
    "class_missing_report.csv",

    index=False

)


# ================================================================
# SIGNAL LEVEL MISSINGNESS
# ================================================================

signal_missing_records = []


for signal_name in (

    feature_map[
        "Signal"
    ]

    .dropna()

    .unique()

):

    signal_features = (

        feature_map

        .loc[

            feature_map[
                "Signal"
            ]

            ==

            signal_name,

            "Feature"

        ]

        .tolist()

    )


    number_of_features = len(
        signal_features
    )


    missing_cells = int(

        missing_counts

        .reindex(

            signal_features,

            fill_value=0

        )

        .sum()

    )


    total_possible_cells = (

        total_rows
        *
        number_of_features

    )


    missing_percentage = (

        missing_cells

        /

        total_possible_cells

        *

        100

    )


    signal_missing_records.append({

        "Signal":
            signal_name,

        "Number_of_Features":
            number_of_features,

        "Missing_Cells":
            missing_cells,

        "Total_Possible_Cells":
            total_possible_cells,

        "Missing_Percent":
            missing_percentage

    })


signal_missing_df = pd.DataFrame(
    signal_missing_records
)


signal_missing_df = (

    signal_missing_df

    .sort_values(

        "Missing_Percent",

        ascending=False

    )

)


signal_missing_df.to_csv(

    OUTPUT_DIRS["missing"] /
    "signal_missing_report.csv",

    index=False

)


# ================================================================
# SIGNAL AND CLASS MISSINGNESS
# ================================================================

signal_class_records = []


for chunk in pd.read_csv(

    INPUT_FILE,

    chunksize=CHUNK_SIZE

):

    for signal_name in (

        feature_map[
            "Signal"
        ]

        .dropna()

        .unique()

    ):

        signal_features = (

            feature_map

            .loc[

                feature_map[
                    "Signal"
                ]

                ==

                signal_name,

                "Feature"

            ]

            .tolist()

        )


        row_missing = (

            chunk[
                signal_features
            ]

            .isna()

            .sum(axis=1)

        )


        temporary = pd.DataFrame({

            "Class":
                chunk[
                    CLASS_COLUMN
                ],

            "Missing":
                row_missing

        })


        grouped = (

            temporary

            .groupby(
                "Class"
            )["Missing"]

            .sum()

        )


        for class_name, value in grouped.items():

            signal_class_records.append({

                "Signal":
                    signal_name,

                "Class":
                    class_name,

                "Missing_Cells":
                    int(value)

            })


signal_class_missing_df = pd.DataFrame(
    signal_class_records
)


signal_class_missing_df = (

    signal_class_missing_df

    .groupby(

        [
            "Signal",
            "Class"
        ],

        as_index=False

    )["Missing_Cells"]

    .sum()

)


signal_class_missing_df.to_csv(

    OUTPUT_DIRS["missing"] /
    "signal_class_missing_report.csv",

    index=False

)


# ================================================================
# MISSING VALUE VISUALIZATION
# ================================================================

top_missing = (

    missing_report[

        [
            "Feature",
            "Missing_Count",
            "Missing_Percent"

        ]

    ]

    .head(30)

    .copy()

)


if not top_missing.empty:

    plt.figure(
        figsize=(15, 12)
    )


    sns.barplot(

        data=top_missing,

        x="Missing_Percent",

        y="Feature",

        hue="Feature",

        palette="magma",

        legend=False

    )


    plt.title(

        "Top 30 Features by Missing Value Percentage",

        fontsize=17,

        fontweight="bold"

    )


    plt.xlabel(
        "Missing Values (%)"
    )

    plt.ylabel(
        "Feature"
    )


    plt.tight_layout()


    plt.savefig(

        OUTPUT_DIRS["visualization"] /
        "01_top_missing_features.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# PARTICIPANT MISSING VISUALIZATION
# ================================================================

if not participant_missing_df.empty:

    plt.figure(
        figsize=(14, 10)
    )


    sns.barplot(

        data=participant_missing_df,

        x="Missing_Cells",

        y=PARTICIPANT_COLUMN,

        hue=PARTICIPANT_COLUMN,

        palette="viridis",

        legend=False

    )


    plt.title(

        "Missing Feature Cells by Participant",

        fontsize=16,

        fontweight="bold"

    )


    plt.xlabel(
        "Missing Feature Cells"
    )

    plt.ylabel(
        "Participant"
    )


    plt.tight_layout()


    plt.savefig(

        OUTPUT_DIRS["visualization"] /
        "02_participant_missing_values.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# CLASS MISSING VISUALIZATION
# ================================================================

if not class_missing_df.empty:

    plt.figure(
        figsize=(12, 7)
    )


    sns.barplot(

        data=class_missing_df,

        x=CLASS_COLUMN,

        y="Missing_Cells",

        hue=CLASS_COLUMN,

        palette="Set2",

        legend=False

    )


    plt.title(

        "Missing Feature Cells by Class",

        fontsize=16,

        fontweight="bold"

    )


    plt.xlabel(
        "Class"
    )

    plt.ylabel(
        "Missing Feature Cells"
    )


    plt.xticks(
        rotation=20
    )


    plt.tight_layout()


    plt.savefig(

        OUTPUT_DIRS["visualization"] /
        "03_class_missing_values.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# STAGE B
# MISSING VALUE TREATMENT
# ================================================================

print()
print("=" * 80)
print("STAGE B: CONSERVATIVE MISSING VALUE TREATMENT")
print("=" * 80)


print()
print(
    "Loading V25 dataset..."
)


df = pd.read_csv(
    INPUT_FILE
)


print(
    "Loaded shape:",
    df.shape
)


# ================================================================
# CONVERT FEATURE COLUMNS TO NUMERIC
# ================================================================

for feature in feature_columns:

    df[feature] = pd.to_numeric(

        df[feature],

        errors="coerce"

    )


# ================================================================
# SORT DATA
# ================================================================

sort_columns = [

    PARTICIPANT_COLUMN

]


if CLASS_COLUMN in df.columns:

    sort_columns.append(
        CLASS_COLUMN
    )


if WINDOW_COLUMN is not None:

    sort_columns.append(
        WINDOW_COLUMN
    )


elif "Window_Start" in df.columns:

    sort_columns.append(
        "Window_Start"
    )


elif "Window_End" in df.columns:

    sort_columns.append(
        "Window_End"
    )


df = df.sort_values(

    sort_columns

).reset_index(
    drop=True
)


# ================================================================
# ORIGINAL MISSING COUNTS
# ================================================================

original_missing = (

    df[
        feature_columns
    ]

    .isna()

    .sum()

)


original_missing_total = int(
    original_missing.sum()
)


# ================================================================
# SHORT INTERNAL GAP INTERPOLATION
# ================================================================
#
# Only internal gaps are interpolated.
#
# Leading and trailing missing values are NOT interpolated here.
#
# Maximum gap:
# 3 consecutive observations
#
# ================================================================

interpolated_cells = 0


if WINDOW_COLUMN is not None:

    print()
    print(
        "Applying short internal gap interpolation..."
    )


    group_columns = [

        PARTICIPANT_COLUMN,
        CLASS_COLUMN

    ]


    for feature in feature_columns:

        before_count = int(

            df[feature]

            .isna()

            .sum()

        )


        df[feature] = (

            df

            .groupby(

                group_columns,

                sort=False

            )[feature]

            .transform(

                lambda series:

                series.interpolate(

                    method="linear",

                    limit=MAX_INTERPOLATION_GAP,

                    limit_direction="both",

                    limit_area="inside"

                )

            )

        )


        after_count = int(

            df[feature]

            .isna()

            .sum()

        )


        interpolated_cells += (

            before_count
            -
            after_count

        )


else:

    print()
    print(
        "Window information was not found."
    )

    print(
        "Temporal interpolation was skipped."
    )


# ================================================================
# PARTICIPANT AND CLASS MEDIAN
# ================================================================

print()
print(
    "Applying participant and class level median fallback..."
)


for feature in feature_columns:

    participant_class_median = (

        df

        .groupby(

            [
                PARTICIPANT_COLUMN,
                CLASS_COLUMN
            ],

            sort=False

        )[feature]

        .transform(
            "median"
        )

    )


    missing_mask = df[feature].isna()


    df.loc[
        missing_mask,
        feature
    ] = (

        participant_class_median[
            missing_mask
        ]

    )


# ================================================================
# CLASS MEDIAN FALLBACK
# ================================================================

print(
    "Applying class level median fallback..."
)


for feature in feature_columns:

    class_median = (

        df

        .groupby(
            CLASS_COLUMN,
            sort=False
        )[feature]

        .transform(
            "median"
        )

    )


    missing_mask = df[feature].isna()


    df.loc[
        missing_mask,
        feature
    ] = (

        class_median[
            missing_mask
        ]

    )


# ================================================================
# GLOBAL MEDIAN FALLBACK
# ================================================================

print(
    "Applying global median fallback only where necessary..."
)


for feature in feature_columns:

    missing_mask = df[feature].isna()


    if missing_mask.any():

        global_median = (

            df[feature]

            .median()

        )


        if pd.notna(
            global_median
        ):

            df.loc[

                missing_mask,
                feature

            ] = global_median


# ================================================================
# FINAL MISSING COUNTS
# ================================================================

final_missing = (

    df[
        feature_columns
    ]

    .isna()

    .sum()

)


final_missing_total = int(
    final_missing.sum()
)


# ================================================================
# IMPUTATION REPORT
# ================================================================

imputation_report = pd.DataFrame({

    "Feature":
        feature_columns,

    "Original_Missing":
        [
            int(
                original_missing[
                    feature
                ]
            )

            for feature in feature_columns
        ],

    "Remaining_Missing":
        [
            int(
                final_missing[
                    feature
                ]
            )

            for feature in feature_columns
        ]

})


imputation_report["Imputed_Cells"] = (

    imputation_report[
        "Original_Missing"
    ]

    -

    imputation_report[
        "Remaining_Missing"
    ]

)


imputation_report["Imputation_Percent"] = np.where(

    imputation_report[
        "Original_Missing"
    ]
    >
    0,

    (

        imputation_report[
            "Imputed_Cells"
        ]

        /

        imputation_report[
            "Original_Missing"
        ]

        *

        100

    ),

    0

)


imputation_report.to_csv(

    OUTPUT_DIRS["imputation"] /
    "imputation_report.csv",

    index=False

)


print()
print(
    "Original missing cells:",
    original_missing_total
)

print(
    "Interpolated cells:",
    interpolated_cells
)

print(
    "Remaining missing cells:",
    final_missing_total
)


# ================================================================
# REMAINING MISSING FEATURE REPORT
# ================================================================

remaining_missing_report = (

    imputation_report

    .loc[

        imputation_report[
            "Remaining_Missing"
        ]
        >
        0

    ]

    .sort_values(

        "Remaining_Missing",

        ascending=False

    )

)


remaining_missing_report.to_csv(

    OUTPUT_DIRS["imputation"] /
    "remaining_missing_values.csv",

    index=False

)


# ================================================================
# STAGE C
# CONSERVATIVE OUTLIER ASSESSMENT
# ================================================================

print()
print("=" * 80)
print("STAGE C: CONSERVATIVE OUTLIER ASSESSMENT")
print("=" * 80)


print()
print(
    "Outliers will be identified and flagged."
)

print(
    "No observations will be deleted."
)


# ================================================================
# MODIFIED Z SCORE
# ================================================================

outlier_records = []


for counter, feature in enumerate(

    feature_columns,

    start=1

):

    values = (

        df[feature]

        .dropna()

        .astype(float)

    )


    if len(values) == 0:

        continue


    median_value = float(
        values.median()
    )


    mad_value = float(

        np.median(

            np.abs(

                values
                -
                median_value

            )

        )

    )


    if mad_value == 0:

        outlier_count = 0

        outlier_percentage = 0.0

    else:

        modified_z = (

            0.6745

            *

            (

                values
                -
                median_value

            )

            /

            mad_value

        )


        outlier_count = int(

            (

                np.abs(
                    modified_z
                )

                >

                OUTLIER_THRESHOLD

            )

            .sum()

        )


        outlier_percentage = (

            outlier_count

            /

            len(values)

            *

            100

        )


    outlier_records.append({

        "Feature":
            feature,

        "Median":
            median_value,

        "MAD":
            mad_value,

        "Potential_Outlier_Count":
            outlier_count,

        "Potential_Outlier_Percent":
            outlier_percentage

    })


outlier_report = pd.DataFrame(
    outlier_records
)


outlier_report = (

    outlier_report

    .sort_values(

        "Potential_Outlier_Percent",

        ascending=False

    )

)


outlier_report.to_csv(

    OUTPUT_DIRS["outlier"] /
    "robust_outlier_report.csv",

    index=False

)


# ================================================================
# OUTLIER VISUALIZATION
# ================================================================

top_outliers = (

    outlier_report

    .head(30)

    .copy()

)


if not top_outliers.empty:

    plt.figure(
        figsize=(15, 12)
    )


    sns.barplot(

        data=top_outliers,

        x="Potential_Outlier_Percent",

        y="Feature",

        hue="Feature",

        palette="rocket",

        legend=False

    )


    plt.title(

        "Potential Outlier Percentage by Feature",

        fontsize=17,

        fontweight="bold"

    )


    plt.xlabel(
        "Potential Outlier Percentage (%)"
    )

    plt.ylabel(
        "Feature"
    )


    plt.tight_layout()


    plt.savefig(

        OUTPUT_DIRS["visualization"] /
        "04_potential_outlier_rate.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# STAGE D
# FEATURE QUALITY SCREENING
# ================================================================

print()
print("=" * 80)
print("STAGE D: FEATURE QUALITY SCREENING")
print("=" * 80)


quality_records = []


for feature in feature_columns:

    values = (

        df[feature]

        .dropna()

        .astype(float)

    )


    if len(values) == 0:

        quality_records.append({

            "Feature":
                feature,

            "Count":
                0,

            "Unique_Count":
                0,

            "Mean":
                np.nan,

            "Standard_Deviation":
                np.nan,

            "Variance":
                np.nan,

            "Minimum":
                np.nan,

            "Maximum":
                np.nan,

            "Q01":
                np.nan,

            "Q99":
                np.nan,

            "Skewness":
                np.nan,

            "Kurtosis":
                np.nan,

            "Coefficient_of_Variation":
                np.nan

        })

        continue


    mean_value = float(
        values.mean()
    )


    std_value = float(
        values.std()
    )


    variance_value = float(
        values.var()
    )


    minimum_value = float(
        values.min()
    )


    maximum_value = float(
        values.max()
    )


    q01_value = float(

        values.quantile(
            0.01
        )

    )


    q99_value = float(

        values.quantile(
            0.99
        )

    )


    if len(values) >= 3:

        skew_value = float(

            skew(
                values,
                bias=False
            )

        )

        kurtosis_value = float(

            kurtosis(
                values,
                fisher=True,
                bias=False
            )

        )

    else:

        skew_value = np.nan

        kurtosis_value = np.nan


    if abs(mean_value) > 1e-12:

        coefficient_variation = (

            abs(std_value)

            /

            abs(mean_value)

        )

    else:

        coefficient_variation = np.nan


    quality_records.append({

        "Feature":
            feature,

        "Count":
            len(values),

        "Unique_Count":
            int(
                values.nunique()
            ),

        "Mean":
            mean_value,

        "Standard_Deviation":
            std_value,

        "Variance":
            variance_value,

        "Minimum":
            minimum_value,

        "Maximum":
            maximum_value,

        "Q01":
            q01_value,

        "Q99":
            q99_value,

        "Skewness":
            skew_value,

        "Kurtosis":
            kurtosis_value,

        "Coefficient_of_Variation":
            coefficient_variation

    })


quality_report = pd.DataFrame(
    quality_records
)


# ================================================================
# ADD MISSINGNESS
# ================================================================

quality_report = (

    quality_report

    .merge(

        missing_report[

            [
                "Feature",
                "Missing_Percent"

            ]

        ],

        on="Feature",

        how="left"

    )

)


# ================================================================
# ADD OUTLIER INFORMATION
# ================================================================

quality_report = (

    quality_report

    .merge(

        outlier_report[

            [
                "Feature",
                "Potential_Outlier_Percent"

            ]

        ],

        on="Feature",

        how="left"

    )

)


# ================================================================
# QUALITY FLAGS
# ================================================================

quality_report["Zero_Variance"] = (

    quality_report[
        "Variance"
    ]

    == 0

)


quality_report["Near_Zero_Variance"] = (

    (

        quality_report[
            "Variance"
        ]

        >

        0

    )

    &

    (

        quality_report[
            "Standard_Deviation"
        ]

        <=

        NEAR_ZERO_VARIANCE_THRESHOLD

    )

)


quality_report["High_Missingness"] = (

    quality_report[
        "Missing_Percent"
    ]

    >

    HIGH_MISSINGNESS_THRESHOLD

)


quality_report["Extreme_Skewness"] = (

    abs(

        quality_report[
            "Skewness"
        ]

    )

    >

    EXTREME_SKEWNESS_THRESHOLD

)


quality_report["Extreme_Kurtosis"] = (

    abs(

        quality_report[
            "Kurtosis"
        ]

    )

    >

    EXTREME_KURTOSIS_THRESHOLD

)


quality_report["Potentially_Unstable"] = (

    (

        quality_report[
            "Potential_Outlier_Percent"
        ]

        >

        10

    )

    |

    quality_report[
        "Extreme_Skewness"
    ]

    |

    quality_report[
        "Extreme_Kurtosis"
    ]

)


# ================================================================
# SAVE COMPLETE QUALITY REPORT
# ================================================================

quality_report.to_csv(

    OUTPUT_DIRS["quality"] /
    "complete_feature_quality_report.csv",

    index=False

)


# ================================================================
# ZERO VARIANCE
# ================================================================

zero_variance_features = (

    quality_report

    .loc[

        quality_report[
            "Zero_Variance"
        ],

        "Feature"

    ]

    .tolist()

)


quality_report.loc[

    quality_report[
        "Zero_Variance"
    ]

].to_csv(

    OUTPUT_DIRS["quality"] /
    "zero_variance_features.csv",

    index=False

)


# ================================================================
# NEAR ZERO VARIANCE
# ================================================================

quality_report.loc[

    quality_report[
        "Near_Zero_Variance"
    ]

].to_csv(

    OUTPUT_DIRS["quality"] /
    "near_zero_variance_features.csv",

    index=False

)


# ================================================================
# HIGH MISSINGNESS
# ================================================================

quality_report.loc[

    quality_report[
        "High_Missingness"
    ]

].to_csv(

    OUTPUT_DIRS["quality"] /
    "high_missingness_features.csv",

    index=False

)


# ================================================================
# EXTREME SKEWNESS
# ================================================================

quality_report.loc[

    quality_report[
        "Extreme_Skewness"
    ]

].to_csv(

    OUTPUT_DIRS["quality"] /
    "extreme_skewness_features.csv",

    index=False

)


# ================================================================
# EXTREME KURTOSIS
# ================================================================

quality_report.loc[

    quality_report[
        "Extreme_Kurtosis"
    ]

].to_csv(

    OUTPUT_DIRS["quality"] /
    "extreme_kurtosis_features.csv",

    index=False

)


# ================================================================
# FEATURE QUALITY LANDSCAPE
# ================================================================

quality_plot = quality_report.copy()


quality_plot = quality_plot.replace(

    [np.inf, -np.inf],

    np.nan

)


quality_plot = quality_plot.dropna(

    subset=[
        "Skewness",
        "Kurtosis"
    ]

)


if not quality_plot.empty:

    plt.figure(
        figsize=(14, 10)
    )


    sns.scatterplot(

        data=quality_plot,

        x="Skewness",

        y="Kurtosis",

        hue="Missing_Percent",

        size="Potential_Outlier_Percent",

        palette="viridis",

        alpha=0.75

    )


    plt.axvline(

        EXTREME_SKEWNESS_THRESHOLD,

        linestyle="--"

    )


    plt.axvline(

        -EXTREME_SKEWNESS_THRESHOLD,

        linestyle="--"

    )


    plt.axhline(

        EXTREME_KURTOSIS_THRESHOLD,

        linestyle="--"

    )


    plt.axhline(

        -EXTREME_KURTOSIS_THRESHOLD,

        linestyle="--"

    )


    plt.title(

        "Feature Quality Landscape",

        fontsize=17,

        fontweight="bold"

    )


    plt.xlabel(
        "Skewness"
    )

    plt.ylabel(
        "Kurtosis"
    )


    plt.tight_layout()


    plt.savefig(

        OUTPUT_DIRS["visualization"] /
        "05_feature_quality_landscape.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# CORRELATION ANALYSIS
# ================================================================

print()
print("=" * 80)
print("CORRELATION ANALYSIS")
print("=" * 80)


correlation_sample_size = min(

    CORRELATION_SAMPLE_SIZE,

    len(df)

)


correlation_sample = (

    df[
        feature_columns
    ]

    .sample(

        n=correlation_sample_size,

        random_state=RANDOM_STATE

    )

)


correlation_matrix = (

    correlation_sample

    .corr()

)


correlation_matrix.to_csv(

    OUTPUT_DIRS["quality"] /
    "feature_correlation_matrix.csv"

)


# ================================================================
# HIGH CORRELATION PAIRS
# ================================================================

upper_triangle = correlation_matrix.where(

    np.triu(

        np.ones(

            correlation_matrix.shape

        ),

        k=1

    ).astype(bool)

)


high_correlation_records = []


for column in upper_triangle.columns:

    correlations = (

        upper_triangle[
            column
        ]

        .dropna()

    )


    correlations = correlations.loc[

        correlations.abs()

        >=

        HIGH_CORRELATION_THRESHOLD

    ]


    for feature_2, correlation_value in correlations.items():

        high_correlation_records.append({

            "Feature_1":
                feature_2,

            "Feature_2":
                column,

            "Correlation":
                float(
                    correlation_value
                ),

            "Absolute_Correlation":
                abs(
                    float(
                        correlation_value
                    )
                )

        })


high_correlation_df = pd.DataFrame(
    high_correlation_records
)


if not high_correlation_df.empty:

    high_correlation_df = (

        high_correlation_df

        .sort_values(

            "Absolute_Correlation",

            ascending=False

        )

    )


high_correlation_df.to_csv(

    OUTPUT_DIRS["quality"] /
    "highly_correlated_feature_pairs.csv",

    index=False

)


# ================================================================
# CORRELATION HEATMAP
# ================================================================

heatmap_features = (

    quality_report

    .sort_values(

        "Potential_Outlier_Percent",

        ascending=False

    )

    ["Feature"]

    .head(40)

    .tolist()

)


heatmap_features = [

    feature

    for feature in heatmap_features

    if feature in correlation_matrix.columns

]


if len(heatmap_features) >= 2:

    plt.figure(
        figsize=(18, 15)
    )


    sns.heatmap(

        correlation_matrix.loc[

            heatmap_features,

            heatmap_features

        ],

        cmap="coolwarm",

        center=0,

        linewidths=0.2

    )


    plt.title(

        "Feature Correlation Heatmap",

        fontsize=17,

        fontweight="bold"

    )


    plt.tight_layout()


    plt.savefig(

        OUTPUT_DIRS["visualization"] /
        "06_feature_correlation_heatmap.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# STAGE E
# PARTICIPANT LEVEL ANALYSIS
# ================================================================

print()
print("=" * 80)
print("STAGE E: PARTICIPANT LEVEL ANALYSIS")
print("=" * 80)


# ================================================================
# PARTICIPANT CLASS WINDOW COUNTS
# ================================================================

participant_class_counts = (

    df

    .groupby(

        [
            PARTICIPANT_COLUMN,
            CLASS_COLUMN
        ]

    )

    .size()

    .reset_index(

        name="Window_Count"

    )

)


participant_class_counts.to_csv(

    OUTPUT_DIRS["participant"] /
    "participant_class_window_counts.csv",

    index=False

)


# ================================================================
# PARTICIPANT CLASS TABLE
# ================================================================

participant_class_table = (

    participant_class_counts

    .pivot(

        index=PARTICIPANT_COLUMN,

        columns=CLASS_COLUMN,

        values="Window_Count"

    )

    .fillna(0)

)


participant_class_table.to_csv(

    OUTPUT_DIRS["participant"] /
    "participant_class_distribution.csv"

)


# ================================================================
# PARTICIPANT FEATURE MEANS
# ================================================================

print()
print(
    "Calculating participant level feature means..."
)


participant_feature_means = (

    df

    .groupby(

        [
            PARTICIPANT_COLUMN,
            CLASS_COLUMN
        ],

        sort=False

    )[feature_columns]

    .mean()

    .reset_index()

)


participant_feature_means.to_csv(

    OUTPUT_DIRS["participant"] /
    "participant_level_feature_means.csv",

    index=False

)


# ================================================================
# PARTICIPANT FEATURE STANDARD DEVIATION
# ================================================================

print(
    "Calculating participant level feature standard deviations..."
)


participant_feature_sd = (

    df

    .groupby(

        [
            PARTICIPANT_COLUMN,
            CLASS_COLUMN
        ],

        sort=False

    )[feature_columns]

    .std()

    .reset_index()

)


participant_feature_sd.to_csv(

    OUTPUT_DIRS["participant"] /
    "participant_level_feature_standard_deviation.csv",

    index=False

)


# ================================================================
# PARTICIPANT CLASS HEATMAP
# ================================================================

plt.figure(
    figsize=(12, 18)
)


sns.heatmap(

    participant_class_table,

    cmap="YlGnBu",

    annot=False

)


plt.title(

    "Participant by Class Window Distribution",

    fontsize=17,

    fontweight="bold"

)


plt.xlabel(
    "Class"
)

plt.ylabel(
    "Participant"
)


plt.tight_layout()


plt.savefig(

    OUTPUT_DIRS["visualization"] /
    "07_participant_class_distribution.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# PARTICIPANT LEVEL VARIABILITY
# ================================================================

participant_variability = (

    participant_feature_sd[
        feature_columns
    ]

    .mean(
        axis=1
    )

)


participant_variability_df = (

    participant_feature_sd[

        [
            PARTICIPANT_COLUMN,
            CLASS_COLUMN

        ]

    ]

    .copy()

)


participant_variability_df[
    "Mean_Feature_SD"
] = participant_variability.values


participant_variability_df.to_csv(

    OUTPUT_DIRS["participant"] /
    "participant_feature_variability.csv",

    index=False

)


plt.figure(
    figsize=(13, 8)
)


sns.boxplot(

    data=participant_variability_df,

    x=CLASS_COLUMN,

    y="Mean_Feature_SD",

    hue=CLASS_COLUMN,

    palette="Set2",

    legend=False

)


plt.title(

    "Participant Level Feature Variability by Class",

    fontsize=17,

    fontweight="bold"

)


plt.xticks(
    rotation=20
)


plt.tight_layout()


plt.savefig(

    OUTPUT_DIRS["visualization"] /
    "08_participant_feature_variability.png",

    dpi=300,

    bbox_inches="tight"

)


plt.close()


# ================================================================
# STAGE F
# PARTICIPANT LEVEL FRIEDMAN TEST
# ================================================================

print()
print("=" * 80)
print("STAGE F: PARTICIPANT LEVEL STATISTICAL TESTING")
print("=" * 80)


classes = sorted(

    participant_feature_means[
        CLASS_COLUMN
    ]

    .dropna()

    .unique()

    .tolist()

)


print()
print(
    "Classes detected:",
    classes
)


print(
    "Number of classes:",
    len(classes)
)


if len(classes) < 3:

    raise ValueError(

        "Friedman testing requires at least three classes."

    )


# ================================================================
# FRIEDMAN TEST
# ================================================================

friedman_records = []


for feature in feature_columns:

    feature_table = (

        participant_feature_means

        .pivot(

            index=PARTICIPANT_COLUMN,

            columns=CLASS_COLUMN,

            values=feature

        )

    )


    required_classes = [

        class_name

        for class_name in classes

        if class_name in feature_table.columns

    ]


    if len(required_classes) != len(classes):

        continue


    complete_table = (

        feature_table[
            classes
        ]

        .dropna()

    )


    if len(complete_table) < 5:

        continue


    arrays = [

        complete_table[
            class_name
        ].values

        for class_name in classes

    ]


    try:

        statistic, p_value = (

            friedmanchisquare(
                *arrays
            )

        )

    except Exception:

        statistic = np.nan

        p_value = np.nan


    number_participants = len(
        complete_table
    )


    number_classes = len(
        classes
    )


    # ------------------------------------------------------------
    # Kendall W
    # ------------------------------------------------------------

    if (

        number_participants > 0

        and

        number_classes > 1

        and

        pd.notna(statistic)

    ):

        kendall_w = (

            statistic

            /

            (

                number_participants

                *

                (

                    number_classes
                    -
                    1

                )

            )

        )

    else:

        kendall_w = np.nan


    friedman_records.append({

        "Feature":
            feature,

        "N_Participants":
            number_participants,

        "Friedman_Statistic":
            statistic,

        "P_Value":
            p_value,

        "Kendall_W":
            kendall_w

    })


friedman_results = pd.DataFrame(
    friedman_records
)


# ================================================================
# HOLM CORRECTION
# ================================================================

if not friedman_results.empty:

    valid_mask = (

        friedman_results[
            "P_Value"
        ]

        .notna()

    )


    if valid_mask.sum() > 0:

        corrected = multipletests(

            friedman_results.loc[

                valid_mask,

                "P_Value"

            ],

            method="holm"

        )


        friedman_results.loc[

            valid_mask,

            "P_Value_Holm"

        ] = corrected[1]


        friedman_results.loc[

            valid_mask,

            "Significant_After_Holm"

        ] = corrected[0]


# ================================================================
# SORT RESULTS
# ================================================================

if "P_Value_Holm" in friedman_results.columns:

    friedman_results = (

        friedman_results

        .sort_values(

            "P_Value_Holm",

            na_position="last"

        )

    )


# ================================================================
# SAVE FRIEDMAN RESULTS
# ================================================================

friedman_results.to_csv(

    OUTPUT_DIRS["statistics"] /
    "friedman_participant_level_results.csv",

    index=False

)


# ================================================================
# SIGNIFICANT FEATURES
# ================================================================

if (

    not friedman_results.empty

    and

    "P_Value_Holm"
    in
    friedman_results.columns

):

    significant_features = (

        friedman_results

        .loc[

            friedman_results[
                "P_Value_Holm"
            ]

            <

            0.05

        ]

        .sort_values(

            "P_Value_Holm"

        )

    )

else:

    significant_features = pd.DataFrame()


significant_features.to_csv(

    OUTPUT_DIRS["statistics"] /
    "significant_features_after_holm.csv",

    index=False

)


# ================================================================
# FRIEDMAN VISUALIZATION
# ================================================================

if not friedman_results.empty:

    plot_results = (

        friedman_results

        .dropna(
            subset=[
                "P_Value_Holm"
            ]
        )

        .sort_values(
            "P_Value_Holm"
        )

        .head(30)

    )


    if not plot_results.empty:

        plt.figure(
            figsize=(14, 12)
        )


        sns.barplot(

            data=plot_results,

            x="P_Value_Holm",

            y="Feature",

            hue="Feature",

            palette="viridis",

            legend=False

        )


        plt.axvline(

            0.05,

            linestyle="--",

            linewidth=1.5

        )


        plt.xscale(
            "log"
        )


        plt.title(

            "Participant Level Friedman Test Results",

            fontsize=17,

            fontweight="bold"

        )


        plt.xlabel(
            "Holm Adjusted P Value"
        )

        plt.ylabel(
            "Feature"
        )


        plt.tight_layout()


        plt.savefig(

            OUTPUT_DIRS["visualization"] /
            "09_friedman_significance.png",

            dpi=300,

            bbox_inches="tight"

        )


        plt.close()


# ================================================================
# STAGE G
# PAIRWISE WILCOXON TESTING
# ================================================================

print()
print("=" * 80)
print("STAGE G: PAIRWISE PARTICIPANT LEVEL TESTING")
print("=" * 80)


pairwise_records = []


for feature in feature_columns:

    feature_table = (

        participant_feature_means

        .pivot(

            index=PARTICIPANT_COLUMN,

            columns=CLASS_COLUMN,

            values=feature

        )

    )


    for i in range(
        len(classes)
    ):

        for j in range(

            i + 1,

            len(classes)

        ):

            class_a = classes[i]

            class_b = classes[j]


            if (

                class_a
                not in
                feature_table.columns

                or

                class_b
                not in
                feature_table.columns

            ):

                continue


            paired = (

                feature_table[

                    [
                        class_a,
                        class_b

                    ]

                ]

                .dropna()

            )


            if len(paired) < 5:

                continue


            values_a = paired[
                class_a
            ].values


            values_b = paired[
                class_b
            ].values


            differences = (

                values_a
                -
                values_b

            )


            nonzero_differences = (

                differences[
                    differences != 0
                ]

            )


            if len(nonzero_differences) < 5:

                continue


            try:

                statistic, p_value = (

                    wilcoxon(

                        values_a,

                        values_b,

                        zero_method="wilcox",

                        alternative="two-sided"

                    )

                )

            except Exception:

                statistic = np.nan

                p_value = np.nan


            # ----------------------------------------------------
            # Rank biserial correlation
            # ----------------------------------------------------

            if len(nonzero_differences) > 0:

                abs_differences = np.abs(

                    nonzero_differences

                )


                ranks = pd.Series(

                    abs_differences

                ).rank(

                    method="average"

                ).values


                positive_rank_sum = (

                    ranks[
                        nonzero_differences > 0
                    ]

                    .sum()

                )


                negative_rank_sum = (

                    ranks[
                        nonzero_differences < 0
                    ]

                    .sum()

                )


                total_rank_sum = (

                    positive_rank_sum
                    +
                    negative_rank_sum

                )


                if total_rank_sum > 0:

                    rank_biserial = (

                        (

                            positive_rank_sum
                            -
                            negative_rank_sum

                        )

                        /

                        total_rank_sum

                    )

                else:

                    rank_biserial = np.nan

            else:

                rank_biserial = np.nan


            pairwise_records.append({

                "Feature":
                    feature,

                "Class_A":
                    class_a,

                "Class_B":
                    class_b,

                "N_Participants":
                    len(paired),

                "Wilcoxon_Statistic":
                    statistic,

                "P_Value":
                    p_value,

                "Rank_Biserial_Correlation":
                    rank_biserial

            })


pairwise_results = pd.DataFrame(
    pairwise_records
)


# ================================================================
# HOLM CORRECTION
# ================================================================

if not pairwise_results.empty:

    valid_mask = (

        pairwise_results[
            "P_Value"
        ]

        .notna()

    )


    if valid_mask.sum() > 0:

        corrected = multipletests(

            pairwise_results.loc[

                valid_mask,

                "P_Value"

            ],

            method="holm"

        )


        pairwise_results.loc[

            valid_mask,

            "P_Value_Holm"

        ] = corrected[1]


        pairwise_results.loc[

            valid_mask,

            "Significant_After_Holm"

        ] = corrected[0]


# ================================================================
# SORT PAIRWISE RESULTS
# ================================================================

if (

    not pairwise_results.empty

    and

    "P_Value_Holm"
    in
    pairwise_results.columns

):

    pairwise_results = (

        pairwise_results

        .sort_values(

            "P_Value_Holm",

            na_position="last"

        )

    )


# ================================================================
# SAVE PAIRWISE RESULTS
# ================================================================

pairwise_results.to_csv(

    OUTPUT_DIRS["statistics"] /
    "pairwise_wilcoxon_results.csv",

    index=False

)


# ================================================================
# SIGNIFICANT PAIRWISE RESULTS
# ================================================================

if (

    not pairwise_results.empty

    and

    "P_Value_Holm"
    in
    pairwise_results.columns

):

    significant_pairwise = (

        pairwise_results

        .loc[

            pairwise_results[
                "P_Value_Holm"
            ]

            <

            0.05

        ]

    )

else:

    significant_pairwise = pd.DataFrame()


significant_pairwise.to_csv(

    OUTPUT_DIRS["statistics"] /
    "significant_pairwise_results.csv",

    index=False

)


# ================================================================
# STAGE H
# PARTICIPANT LEVEL CLASS PROFILES
# ================================================================

print()
print("=" * 80)
print("STAGE H: PARTICIPANT LEVEL CLASS PROFILES")
print("=" * 80)


if not significant_features.empty:

    profile_features = (

        significant_features[
            "Feature"
        ]

        .head(10)

        .tolist()

    )

else:

    profile_features = (

        quality_report

        .sort_values(

            "Potential_Outlier_Percent",

            ascending=False

        )

        ["Feature"]

        .head(10)

        .tolist()

    )


profile_features = [

    feature

    for feature in profile_features

    if feature in
    participant_feature_means.columns

]


if profile_features:

    profile_long = (

        participant_feature_means[

            [
                CLASS_COLUMN

            ]

            +

            profile_features

        ]

        .melt(

            id_vars=CLASS_COLUMN,

            var_name="Feature",

            value_name="Value"

        )

    )


    plt.figure(
        figsize=(18, 10)
    )


    sns.boxplot(

        data=profile_long,

        x="Feature",

        y="Value",

        hue=CLASS_COLUMN,

        palette="Set2"

    )


    plt.title(

        "Participant Level Feature Profiles by Class",

        fontsize=17,

        fontweight="bold"

    )


    plt.xticks(

        rotation=70,

        ha="right"

    )


    plt.tight_layout()


    plt.savefig(

        OUTPUT_DIRS["visualization"] /
        "10_participant_feature_profiles.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.close()


# ================================================================
# STAGE I
# FEATURE RETENTION REVIEW
# ================================================================

print()
print("=" * 80)
print("STAGE I: FEATURE RETENTION REVIEW")
print("=" * 80)


quality_report["Recommended_Status"] = (

    "Retain_for_feature_selection"

)


quality_report.loc[

    quality_report[
        "Zero_Variance"
    ],

    "Recommended_Status"

] = "Remove_zero_variance"


quality_report.loc[

    (

        quality_report[
            "High_Missingness"
        ]

        &

        ~

        quality_report[
            "Zero_Variance"
        ]

    ),

    "Recommended_Status"

] = "Review_high_missingness"


quality_report.loc[

    (

        quality_report[
            "Near_Zero_Variance"
        ]

        &

        ~

        quality_report[
            "Zero_Variance"
        ]

    ),

    "Recommended_Status"

] = "Review_near_zero_variance"


quality_report.to_csv(

    OUTPUT_DIRS["quality"] /
    "feature_retention_recommendation.csv",

    index=False

)


# ================================================================
# REMOVE ONLY ZERO VARIANCE FEATURES
# ================================================================
#
# Important:
#
# Statistical outliers are NOT removed.
#
# Extreme skewness is NOT automatically removed.
#
# Extreme kurtosis is NOT automatically removed.
#
# Highly correlated features are NOT automatically removed.
#
# High missingness features are flagged for review.
#
# ================================================================

features_for_v26 = [

    feature

    for feature in feature_columns

    if feature
    not in
    zero_variance_features

]


pd.DataFrame({

    "Feature":
        features_for_v26

}).to_csv(

    OUTPUT_DIRS["quality"] /
    "features_retained_for_v26.csv",

    index=False

)


# ================================================================
# CREATE FINAL V26 DATASET
# ================================================================

print()
print("=" * 80)
print("CREATING FINAL V26 DATASET")
print("=" * 80)


final_columns = [

    column

    for column in ALL_COLUMNS

    if column in metadata_columns
    or
    column in features_for_v26

]


v26_df = df[
    final_columns
].copy()


FINAL_V26_FILE = (

    OUTPUT_DIRS["dataset"] /

    "V26_Quality_Controlled_Features.csv"

)


print()
print(
    "Saving final V26 dataset..."
)


v26_df.to_csv(

    FINAL_V26_FILE,

    index=False

)


# ================================================================
# FINAL QUALITY CHECK
# ================================================================

final_feature_columns = [

    feature

    for feature in features_for_v26

    if feature in v26_df.columns

]


remaining_missing_final = int(

    v26_df[
        final_feature_columns
    ]

    .isna()

    .sum()

    .sum()

)


duplicate_rows = int(

    v26_df

    .duplicated()

    .sum()

)


infinite_cells = int(

    np.isinf(

        v26_df[
            final_feature_columns
        ]

        .to_numpy(
            dtype=float
        )

    )

    .sum()

)


# ================================================================
# PARTICIPANT COUNT
# ================================================================

participant_count = (

    v26_df[
        PARTICIPANT_COLUMN
    ]

    .nunique()

)


# ================================================================
# CLASS COUNT
# ================================================================

class_count = (

    v26_df[
        CLASS_COLUMN
    ]

    .nunique()

)


# ================================================================
# FINAL SUMMARY
# ================================================================

friedman_significant_count = 0


if (

    not friedman_results.empty

    and

    "P_Value_Holm"
    in
    friedman_results.columns

):

    friedman_significant_count = int(

        (

            friedman_results[
                "P_Value_Holm"
            ]

            <

            0.05

        )

        .sum()

    )


pairwise_significant_count = 0


if (

    not pairwise_results.empty

    and

    "Significant_After_Holm"
    in
    pairwise_results.columns

):

    pairwise_significant_count = int(

        (

            pairwise_results[
                "Significant_After_Holm"
            ]

            ==

            True

        )

        .sum()

    )


summary_records = [

    (
        "Original_V25_Rows",
        len(df)
    ),

    (
        "Original_V25_Features",
        len(feature_columns)
    ),

    (
        "V26_Rows",
        len(v26_df)
    ),

    (
        "V26_Features",
        len(final_feature_columns)
    ),

    (
        "Participants",
        participant_count
    ),

    (
        "Classes",
        class_count
    ),

    (
        "Original_Missing_Cells",
        original_missing_total
    ),

    (
        "Overall_Original_Missing_Percent",
        overall_missing_percentage
    ),

    (
        "Interpolated_Cells",
        interpolated_cells
    ),

    (
        "Remaining_Missing_Cells",
        remaining_missing_final
    ),

    (
        "Duplicate_Rows",
        duplicate_rows
    ),

    (
        "Infinite_Cells",
        infinite_cells
    ),

    (
        "Zero_Variance_Features",
        len(zero_variance_features)
    ),

    (
        "Near_Zero_Variance_Features",
        int(
            quality_report[
                "Near_Zero_Variance"
            ].sum()
        )
    ),

    (
        "High_Missingness_Features",
        int(
            quality_report[
                "High_Missingness"
            ].sum()
        )
    ),

    (
        "Extreme_Skewness_Features",
        int(
            quality_report[
                "Extreme_Skewness"
            ].sum()
        )
    ),

    (
        "Extreme_Kurtosis_Features",
        int(
            quality_report[
                "Extreme_Kurtosis"
            ].sum()
        )
    ),

    (
        "Highly_Correlated_Feature_Pairs",
        len(high_correlation_df)
    ),

    (
        "Friedman_Tests",
        len(friedman_results)
    ),

    (
        "Friedman_Significant_After_Holm",
        friedman_significant_count
    ),

    (
        "Pairwise_Wilcoxon_Tests",
        len(pairwise_results)
    ),

    (
        "Pairwise_Significant_After_Holm",
        pairwise_significant_count
    )

]


summary_df = pd.DataFrame(

    summary_records,

    columns=[
        "Metric",
        "Value"
    ]

)


summary_df.to_csv(

    OUTPUT_DIRS["summary"] /
    "V26_complete_summary.csv",

    index=False

)


# ================================================================
# SAVE DATASET INFORMATION
# ================================================================

dataset_information = pd.DataFrame({

    "Property": [

        "Input_File",

        "Output_File",

        "Rows",

        "Columns",

        "Features",

        "Participant_Column",

        "Class_Column",

        "Window_Column",

        "Participant_Count",

        "Class_Count"

    ],

    "Value": [

        INPUT_FILE,

        str(
            FINAL_V26_FILE
        ),

        len(v26_df),

        len(v26_df.columns),

        len(final_feature_columns),

        PARTICIPANT_COLUMN,

        CLASS_COLUMN,

        WINDOW_COLUMN,

        participant_count,

        class_count

    ]

})


dataset_information.to_csv(

    OUTPUT_DIRS["summary"] /
    "dataset_information.csv",

    index=False

)


# ================================================================
# FINAL CONSOLE REPORT
# ================================================================

print()
print()
print("=" * 80)
print("V25 TO V26 PROCESS COMPLETED")
print("=" * 80)

print()

print(
    "V25 rows:",
    len(df)
)

print(
    "V25 features:",
    len(feature_columns)
)

print(
    "V26 rows:",
    len(v26_df)
)

print(
    "V26 features:",
    len(final_feature_columns)
)

print()

print(
    "Participants:",
    participant_count
)

print(
    "Classes:",
    class_count
)

print()

print(
    "Original missing cells:",
    original_missing_total
)

print(
    "Original missing percentage:",
    round(
        overall_missing_percentage,
        4
    ),
    "%"
)

print(
    "Interpolated cells:",
    interpolated_cells
)

print(
    "Remaining missing cells:",
    remaining_missing_final
)

print()

print(
    "Duplicate rows:",
    duplicate_rows
)

print(
    "Infinite cells:",
    infinite_cells
)

print()

print(
    "Zero variance features:",
    len(zero_variance_features)
)

print(
    "Near zero variance features:",
    int(
        quality_report[
            "Near_Zero_Variance"
        ].sum()
    )
)

print(
    "High missingness features:",
    int(
        quality_report[
            "High_Missingness"
        ].sum()
    )
)

print(
    "Extreme skewness features:",
    int(
        quality_report[
            "Extreme_Skewness"
        ].sum()
    )
)

print(
    "Extreme kurtosis features:",
    int(
        quality_report[
            "Extreme_Kurtosis"
        ].sum()
    )
)

print(
    "Highly correlated feature pairs:",
    len(high_correlation_df)
)

print()

print(
    "Friedman tests:",
    len(friedman_results)
)

print(
    "Friedman significant after Holm:",
    friedman_significant_count
)

print(
    "Pairwise Wilcoxon tests:",
    len(pairwise_results)
)

print(
    "Pairwise significant after Holm:",
    pairwise_significant_count
)

print()

print(
    "Final V26 dataset:"
)

print(
    FINAL_V26_FILE
)

print()

print(
    "Output directory:"
)

print(
    OUTPUT_ROOT
)

print()
print("=" * 80)
print("ALL ANALYSIS STAGES COMPLETED")
print("=" * 80)
```

```python
# ============================================================
# V27 FEATURE STABILITY SCREENING + ROBUST SCALING
# ============================================================

import os
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import skew, kurtosis

from sklearn.preprocessing import RobustScaler


# ============================================================
# PATH CONFIGURATION
# ============================================================

INPUT_FILE = (
    "/home/rakib/Desktop/RAW_DATA/"
    "CLEAN_DATA_V26/08_Final_Dataset/"
    "V26_Quality_Controlled_Features.csv"
)


OUTPUT_DIR = (
    "/home/rakib/Desktop/RAW_DATA/"
    "CLEAN_DATA_V27"
)


DATA_DIR = os.path.join(
    OUTPUT_DIR,
    "01_Final_Dataset"
)

REPORT_DIR = os.path.join(
    OUTPUT_DIR,
    "02_Reports"
)

VIS_DIR = os.path.join(
    OUTPUT_DIR,
    "03_Visualization"
)


for folder in [
    OUTPUT_DIR,
    DATA_DIR,
    REPORT_DIR,
    VIS_DIR
]:
    os.makedirs(folder, exist_ok=True)



# ============================================================
# LOAD DATA
# ============================================================

print("="*80)
print("V27 FEATURE STABILITY SCREENING + SCALING")
print("="*80)


df = pd.read_csv(INPUT_FILE)

print("\nOriginal dataset shape:")
print(df.shape)



# ============================================================
# COLUMN DETECTION
# ============================================================

metadata_candidates = [
    "PersonSrial",
    "Class",
    "Window",
    "Window_Start",
    "Window_End",
    "Window_Samples"
]


metadata_columns = [
    c for c in metadata_candidates
    if c in df.columns
]


feature_columns = [
    c for c in df.columns
    if c not in metadata_columns
]


print("\nMetadata columns:")
print(metadata_columns)


print("\nNumber of features:")
print(len(feature_columns))



X = df[feature_columns].copy()



# ============================================================
# FEATURE QUALITY ANALYSIS
# ============================================================

report = []


for col in feature_columns:

    data = X[col]

    missing_percentage = (
        data.isna().mean()*100
    )

    variance = data.var()

    skew_value = skew(
        data.dropna()
    )

    kurtosis_value = kurtosis(
        data.dropna()
    )


    report.append({

        "Feature": col,

        "Missing_Percentage":
        missing_percentage,

        "Variance":
        variance,

        "Skewness":
        skew_value,

        "Kurtosis":
        kurtosis_value

    })


quality_df = pd.DataFrame(report)



quality_df.to_csv(
    os.path.join(
        REPORT_DIR,
        "feature_quality_report.csv"
    ),
    index=False
)



# ============================================================
# FEATURE REMOVAL RULES
# ============================================================


removed = []


for _, row in quality_df.iterrows():

    reason = []


    # Zero variance
    if row["Variance"] == 0:
        reason.append(
            "Zero variance"
        )


    # Near zero variance
    if (
        row["Variance"] < 1e-10
        and row["Variance"] != 0
    ):
        reason.append(
            "Near zero variance"
        )


    # Very high missingness
    if row["Missing_Percentage"] > 40:
        reason.append(
            "High missingness"
        )


    # Extreme distribution instability
    if abs(row["Skewness"]) > 20:
        reason.append(
            "Extreme skewness"
        )


    if abs(row["Kurtosis"]) > 100:
        reason.append(
            "Extreme kurtosis"
        )


    if len(reason)>0:

        removed.append({

            "Feature":
            row["Feature"],

            "Removal_Reason":
            "; ".join(reason)

        })



removed_df = pd.DataFrame(
    removed
)



removed_features = (
    removed_df["Feature"].tolist()
    if len(removed_df)>0
    else []
)



retained_features = [
    c for c in feature_columns
    if c not in removed_features
]



print("\nRemoved features:")
print(len(removed_features))


print("Retained features:")
print(len(retained_features))



removed_df.to_csv(
    os.path.join(
        REPORT_DIR,
        "removed_features_report.csv"
    ),
    index=False
)



pd.DataFrame(
    {
        "Retained_Features":
        retained_features
    }
).to_csv(
    os.path.join(
        REPORT_DIR,
        "retained_features_report.csv"
    ),
    index=False
)



# ============================================================
# CREATE CLEAN FEATURE MATRIX
# ============================================================


X_final = df[
    metadata_columns +
    retained_features
].copy()



# ============================================================
# ROBUST SCALING
# ============================================================


scaler = RobustScaler()


scaled_values = scaler.fit_transform(
    X_final[retained_features]
)



scaled_df = pd.DataFrame(
    scaled_values,
    columns=retained_features
)



for col in metadata_columns:

    scaled_df[col] = df[col].values



# Keep metadata first

scaled_df = scaled_df[
    metadata_columns +
    retained_features
]



# ============================================================
# SAVE V27 DATASET
# ============================================================


OUTPUT_DATASET = os.path.join(
    DATA_DIR,
    "V27_Scaled_Feature_Dataset.csv"
)


scaled_df.to_csv(
    OUTPUT_DATASET,
    index=False
)



# ============================================================
# SCALING SUMMARY
# ============================================================


summary = pd.DataFrame({

    "Feature":
    retained_features,

    "Original_Median":
    X[retained_features].median().values,

    "Original_IQR":
    (
        X[retained_features].quantile(.75)
        -
        X[retained_features].quantile(.25)
    ).values,

    "Scaled_Mean":
    scaled_df[retained_features].mean().values,

    "Scaled_STD":
    scaled_df[retained_features].std().values

})


summary.to_csv(
    os.path.join(
        REPORT_DIR,
        "scaling_summary.csv"
    ),
    index=False
)



# ============================================================
# DATASET SUMMARY
# ============================================================


dataset_summary = pd.DataFrame({

    "Item":[

        "Original Rows",
        "Original Features",
        "Removed Features",
        "Final Features",
        "Participants",
        "Classes"

    ],

    "Value":[

        len(df),
        len(feature_columns),
        len(removed_features),
        len(retained_features),
        df["PersonSrial"].nunique(),
        df["Class"].nunique()

    ]

})


dataset_summary.to_csv(
    os.path.join(
        REPORT_DIR,
        "dataset_summary.csv"
    ),
    index=False
)



# ============================================================
# VISUALIZATION
# ============================================================


sns.set_theme(
    style="whitegrid"
)



# Feature variance distribution

plt.figure(figsize=(12,6))

sns.histplot(
    quality_df["Variance"],
    bins=100
)

plt.title(
    "Feature Variance Distribution"
)

plt.savefig(
    os.path.join(
        VIS_DIR,
        "feature_variance_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()



# Skewness distribution

plt.figure(figsize=(12,6))

sns.histplot(
    quality_df["Skewness"],
    bins=100
)

plt.title(
    "Feature Skewness Distribution"
)

plt.savefig(
    os.path.join(
        VIS_DIR,
        "skewness_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()



# Before after distribution

sample_feature = retained_features[:5]


for feature in sample_feature:

    plt.figure(figsize=(10,5))


    sns.kdeplot(
        X[feature],
        label="Before Scaling"
    )


    sns.kdeplot(
        scaled_df[feature],
        label="After Scaling"
    )


    plt.title(
        feature
    )


    plt.legend()


    plt.savefig(
        os.path.join(
            VIS_DIR,
            f"{feature}_before_after.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()



# Correlation heatmap

corr_features = retained_features[:50]


plt.figure(
    figsize=(16,12)
)


sns.heatmap(
    scaled_df[corr_features].corr(),
    cmap="coolwarm",
    center=0
)


plt.title(
    "V27 Feature Correlation Heatmap"
)


plt.savefig(
    os.path.join(
        VIS_DIR,
        "correlation_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()



# ============================================================
# FINAL REPORT
# ============================================================


print("\n"+"="*80)

print("V27 COMPLETED SUCCESSFULLY")

print("="*80)

print(
    "Original features:",
    len(feature_columns)
)

print(
    "Removed features:",
    len(removed_features)
)

print(
    "Final features:",
    len(retained_features)
)

print(
    "Final dataset:",
    OUTPUT_DATASET
)

print(
    "Output directory:",
    OUTPUT_DIR
)

print("="*80)
```

```python
# ============================================================
# V28 MULTI METHOD VOTING BASED FEATURE SELECTION
# EEG FEATURE DATASET
# ============================================================


import os
import warnings
warnings.filterwarnings("ignore")


import numpy as np
import pandas as pd

from scipy.stats import friedmanchisquare
from statsmodels.stats.multitest import multipletests

from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import RFECV
from sklearn.preprocessing import LabelEncoder

from xgboost import XGBClassifier

import matplotlib.pyplot as plt
import seaborn as sns



# ============================================================
# PATH
# ============================================================


INPUT_FILE = (
"/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V27/"
"01_Final_Dataset/V27_Scaled_Feature_Dataset.csv"
)


OUTPUT_DIR = (
"/home/rakib/Desktop/RAW_DATA/CLEAN_DATA_V28"
)



DATA_DIR = os.path.join(
    OUTPUT_DIR,
    "01_Final_Dataset"
)


REPORT_DIR = os.path.join(
    OUTPUT_DIR,
    "02_Reports"
)


FIG_DIR = os.path.join(
    OUTPUT_DIR,
    "03_Visualization"
)



for folder in [
    OUTPUT_DIR,
    DATA_DIR,
    REPORT_DIR,
    FIG_DIR
]:
    os.makedirs(folder, exist_ok=True)



print("="*70)
print("V28 FEATURE SELECTION STARTED")
print("="*70)



# ============================================================
# LOAD DATA
# ============================================================


df = pd.read_csv(INPUT_FILE)


print("Dataset shape:", df.shape)



metadata_cols = [
    "PersonSrial",
    "Class",
    "Window",
    "Window_Start",
    "Window_End",
    "Window_Samples"
]


feature_cols = [
    c for c in df.columns
    if c not in metadata_cols
]


X = df[feature_cols].copy()

y = df["Class"]



print(
    "Features:",
    len(feature_cols)
)



# ============================================================
# ENCODE LABEL
# ============================================================


encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)



# ============================================================
# VOTING TABLE
# ============================================================


vote_table = pd.DataFrame(
    index=feature_cols
)


# ============================================================
# METHOD 1
# FRIEDMAN TEST
# PARTICIPANT AWARE
# ============================================================


print("\nFriedman test running...")


participants = df["PersonSrial"].unique()


friedman_p = []


for feature in feature_cols:

    values = []

    for cls in sorted(df["Class"].unique()):

        class_values=[]

        for p in participants:

            temp=df[
                (df.PersonSrial==p)
                &
                (df.Class==cls)
            ][feature]


            if len(temp)>0:

                class_values.append(
                    temp.mean()
                )


        values.append(class_values)


    try:

        stat,pvalue = friedmanchisquare(
            *values
        )

    except:

        pvalue=1


    friedman_p.append(pvalue)



reject, corrected, _, _ = multipletests(
    friedman_p,
    method="holm"
)


vote_table["Friedman"] = reject



# ============================================================
# METHOD 2
# MUTUAL INFORMATION
# ============================================================


print("Mutual information running...")


mi = mutual_info_classif(
    X,
    y_encoded,
    random_state=42
)


mi_threshold = np.percentile(
    mi,
    75
)


vote_table["Mutual_Information"] = (
    mi >= mi_threshold
)



# ============================================================
# METHOD 3
# RANDOM FOREST IMPORTANCE
# ============================================================


print("Random Forest importance running...")


rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)


rf.fit(
    X,
    y_encoded
)


rf_importance = rf.feature_importances_


rf_threshold = np.percentile(
    rf_importance,
    75
)


vote_table["RandomForest"] = (
    rf_importance >= rf_threshold
)



# ============================================================
# METHOD 4
# XGBOOST IMPORTANCE
# ============================================================


print("XGBoost importance running...")


xgb = XGBClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)


xgb.fit(
    X,
    y_encoded
)


xgb_importance = (
    xgb.feature_importances_
)


xgb_threshold = np.percentile(
    xgb_importance,
    75
)



vote_table["XGBoost"] = (
    xgb_importance >= xgb_threshold
)




# ============================================================
# METHOD 5
# RFECV
# ============================================================


print("RFECV running...")


rfecv = RFECV(
    estimator=RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    step=0.1,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)


rfecv.fit(
    X,
    y_encoded
)



vote_table["RFECV"] = (
    rfecv.support_
)



# ============================================================
# METHOD 6
# VARIANCE STABILITY
# ============================================================


print("Variance stability running...")


variance = X.var()


variance_threshold = np.percentile(
    variance,
    25
)


vote_table["Variance"] = (
    variance > variance_threshold
)



# ============================================================
# VOTING SCORE
# ============================================================


vote_table["Vote_Count"] = (
    vote_table.sum(axis=1)
)



vote_table = vote_table.sort_values(
    "Vote_Count",
    ascending=False
)



vote_table.to_csv(
    os.path.join(
        REPORT_DIR,
        "voting_summary.csv"
    )
)



# ============================================================
# CREATE THREE DATASETS
# ============================================================



def save_dataset(
    selected_features,
    filename
):

    output=df[
        metadata_cols +
        selected_features
    ]


    output.to_csv(
        os.path.join(
            DATA_DIR,
            filename
        ),
        index=False
    )


    print(
        filename,
        len(selected_features)
    )



minimum_features = (
    vote_table[
        vote_table.Vote_Count>=5
    ].index.tolist()
)



medium_features = (
    vote_table[
        vote_table.Vote_Count>=4
    ].index.tolist()
)



high_features = (
    vote_table[
        vote_table.Vote_Count>=3
    ].index.tolist()
)



save_dataset(
    minimum_features,
    "V28_Minimum_Features.csv"
)



save_dataset(
    medium_features,
    "V28_Medium_Features.csv"
)



save_dataset(
    high_features,
    "V28_High_Features.csv"
)



# ============================================================
# VISUALIZATION
# ============================================================


plt.figure(
    figsize=(10,6)
)


sns.countplot(
    x=vote_table.Vote_Count
)


plt.title(
    "Feature Voting Distribution"
)


plt.xlabel(
    "Number of Selection Methods"
)


plt.ylabel(
    "Number of Features"
)


plt.tight_layout()


plt.savefig(
    os.path.join(
        FIG_DIR,
        "feature_vote_distribution.png"
    ),
    dpi=300
)


plt.close()



# ============================================================


print("\nFINAL SUMMARY")
print("="*70)

print(
"Minimum features:",
len(minimum_features)
)


print(
"Medium features:",
len(medium_features)
)


print(
"High features:",
len(high_features)
)


print("="*70)
print("V28 COMPLETED SUCCESSFULLY")
print("="*70)
```

```python
**Model Training and Evaluation**
```
